```
import math
import mathutils
import os
import sys
import glob
import bpy
from bpy_extras import anim_utils
from bpy_extras.object_utils import world_to_camera_view
from bpy.app.handlers import persistent
from mathutils import Vector

# ==============================================================================
# CONFIGURATION
# ==============================================================================
FOOTSTEP_SOUND_PATH = "/Users/dobbinp1/Downloads/stomp.wav"
GROUND_THRESHOLD = 0.3  # Threshold for grounded character footstep detection


def find_foot_bone(pose_bones, side="Left"):
    target_keywords = [
        f"mixamorig:{side}foot",
        f"{side.lower()}foot",
        f"foot.{side[0].lower()}",
        f"foot_{side[0].lower()}"
    ]
    for bone in pose_bones:
        b_name = bone.name.lower()
        if any(kw in b_name for kw in target_keywords):
            return bone
    return None


def generate_footstep_audio_strips(scene):
    """
    Evaluates character footfalls across all frame ranges and places sound strips
    into the VSE timeline so that audio encodes directly during video rendering.
    """
    if not os.path.exists(FOOTSTEP_SOUND_PATH):
        print(f"[AUDIO ERROR] File not found at: {FOOTSTEP_SOUND_PATH}")
        return

    # Ensure Sequence Editor exists
    if not scene.sequence_editor:
        scene.sequence_editor_create()

    seq_editor = scene.sequence_editor

    # Clear previous footstep audio strips (use .strips_all and .strips in Blender 4.0+)
    for seq in list(seq_editor.strips_all):
        if seq.name.startswith("Footstep_Audio"):
            seq_editor.strips.remove(seq)

    rig = next((o for o in scene.objects if o.type == 'ARMATURE'), None)
    if not rig:
        print("[AUDIO WARNING] No armature found to evaluate footsteps.")
        return

    depsgraph = bpy.context.evaluated_depsgraph_get()
    turn_start = scene.get("turn_start_frame", scene.frame_end)

    left_was_down = False
    right_was_down = False

    footstep_channel = 1
    strip_counter = 1

    # Scan frame sequence and place audio strips at exact impact moments
    for frame in range(scene.frame_start, turn_start):
        scene.frame_set(frame)
        depsgraph.update()

        eval_rig = rig.evaluated_get(depsgraph)

        left_bone = find_foot_bone(eval_rig.pose.bones, "Left")
        right_bone = find_foot_bone(eval_rig.pose.bones, "Right")

        if not left_bone or not right_bone:
            continue

        l_z = (eval_rig.matrix_world @ left_bone.matrix).translation.z
        r_z = (eval_rig.matrix_world @ right_bone.matrix).translation.z

        l_is_down = l_z <= GROUND_THRESHOLD
        r_is_down = r_z <= GROUND_THRESHOLD

        if l_is_down and not left_was_down:
            seq_editor.strips.new_sound(
                name=f"Footstep_Audio_{strip_counter}",
                filepath=FOOTSTEP_SOUND_PATH,
                channel=footstep_channel,
                frame_start=frame
            )
            strip_counter += 1

        if r_is_down and not right_was_down:
            seq_editor.strips.new_sound(
                name=f"Footstep_Audio_{strip_counter}",
                filepath=FOOTSTEP_SOUND_PATH,
                channel=footstep_channel,
                frame_start=frame
            )
            strip_counter += 1

        left_was_down = l_is_down
        right_was_down = r_is_down

    scene.frame_set(scene.frame_start)
    print(f"[AUDIO SUCCESS] Generated {strip_counter - 1} footstep audio strips in VSE.")


def configure_render_audio_settings(scene):
    """
    Configures FFmpeg video and audio encoding defaults for movie exports.
    """
    # In Blender, movie formats are configured via image_settings.file_format
    scene.render.image_settings.file_format = 'JPEG'

    # Configure container and audio codecs
    scene.render.ffmpeg.format = 'MPEG4'
    scene.render.ffmpeg.codec = 'H264'
    scene.render.ffmpeg.audio_codec = 'AAC'
    scene.render.ffmpeg.audio_bitrate = 192


# ==============================================================================
# CLEANUP & HELPERS
# ==============================================================================

def cleanup_previous_run():
    target_names = [
        "Master_NLA_Control",
        "Scroll_GN_Object",
        "Scroll_Start_Pin",
        "Scroll_End_Pin",
        "Scroll_Master_Container",
        "HDRI_Ground_Plane",
        "Ground_Mesh_Object",
        "Grass_Ground_Plane",
        "Grass_Blade_Instance",
        "Camera_Focus_Target",
        "Camera_Anchor",
        "Vertical_Reel_Camera"
    ]
    for obj in list(bpy.data.objects):
        if obj.name in target_names or (obj.name.startswith("mixamorig") and obj.type not in {'ARMATURE', 'MESH'}):
            if obj.animation_data:
                obj.animation_data_clear()
            bpy.data.objects.remove(obj, do_unlink=True)

    for action in list(bpy.data.actions):
        if action.users == 0 or action.use_fake_user:
            action.use_fake_user = False
            bpy.data.actions.remove(action)

    for ng in list(bpy.data.node_groups):
        if ng.name in ["Scroll_Unfurl_GN", "Scroll_Pure_GN", "HDRI_Ground_GN", "Grass_Ground_GN", "Grass_Scatter_Group"]:
            bpy.data.node_groups.remove(ng)

    for coll_name in ["Scroll_Master_Container", "Grass_Artifacts_Collection"]:
        if coll_name in bpy.data.collections:
            coll = bpy.data.collections[coll_name]
            for obj in list(coll.objects):
                bpy.data.objects.remove(obj, do_unlink=True)
            bpy.data.collections.remove(coll)

    bpy.context.view_layer.update()


def get_slot_fcurves(action, anim_target_object):
    fcurves = []
    if not action:
        return fcurves

    anim_data = getattr(anim_target_object, "animation_data", None)
    if anim_data and getattr(anim_data, "action_slot", None):
        slot = anim_data.action_slot
        channelbag = anim_utils.action_get_channelbag_for_slot(action, slot)
        if channelbag and hasattr(channelbag, "fcurves"):
            return list(channelbag.fcurves)

    if hasattr(action, "slots"):
        for slot in action.slots:
            for cb in getattr(slot, "channelbags", []):
                fcurves.extend(getattr(cb, "fcurves", []))
        if fcurves:
            return fcurves

    if hasattr(action, "fcurves"):
        return list(action.fcurves)

    return fcurves


def get_object_fcurves(obj):
    if not obj or not obj.animation_data or not obj.animation_data.action:
        return []
    return get_slot_fcurves(obj.animation_data.action, obj)


def detect_natural_forward_axis(action, rig_object):
    fcurves = get_slot_fcurves(action, rig_object)
    max_delta = 0.0
    best_axis = 1
    net_delta = 0.0

    start_frame, end_frame = action.frame_range

    for fc in fcurves:
        if fc.data_path.endswith("location") and fc.array_index in (0, 1, 2):
            delta = fc.evaluate(end_frame) - fc.evaluate(start_frame)
            if abs(delta) > abs(max_delta):
                max_delta = delta
                best_axis = fc.array_index
                net_delta = delta

    direction_sign = 1.0 if net_delta >= 0 else -1.0
    return best_axis, direction_sign


def find_hand_vertex_groups(target_mesh):
    r_name, l_name = "mixamorig:RightHand", "mixamorig:LeftHand"
    if target_mesh and hasattr(target_mesh, "vertex_groups"):
        for vg in target_mesh.vertex_groups:
            vg_lower = vg.name.lower()
            if any(k in vg_lower for k in ["righthand", "hand_r", "hand.r"]):
                r_name = vg.name
            elif any(k in vg_lower for k in ["lefthand", "hand_l", "hand.l"]):
                l_name = vg.name
    return r_name, l_name


def create_gn_named_attr_node(nodes):
    try:
        return nodes.new("GeometryNodeNamedAttribute")
    except Exception:
        return nodes.new("GeometryNodeInputNamedAttribute")

# ==============================================================================
# RIG-PARENTED CAMERA RIG SETUP
# ==============================================================================

def setup_modern_reel_camera(target_rig, target_mesh, turn_start, turn_end_frame, unfurl_end_frame):
    scene = bpy.context.scene
    scene.render.resolution_x = 1080
    scene.render.resolution_y = 1920
    scene.render.resolution_percentage = 100

    focus_target = bpy.data.objects.get("Camera_Focus_Target") or bpy.data.objects.new("Camera_Focus_Target", None)
    if focus_target.name not in scene.collection.objects:
        scene.collection.objects.link(focus_target)

    cam_data = bpy.data.cameras.get("Vertical_Reel_Cam_Data") or bpy.data.cameras.new("Vertical_Reel_Cam_Data")
    cam_data.lens = 45.0
    cam_data.sensor_fit = 'VERTICAL'

    cam_obj = bpy.data.objects.get("Vertical_Reel_Camera") or bpy.data.objects.new("Vertical_Reel_Camera", cam_data)
    if cam_obj.name not in scene.collection.objects:
        scene.collection.objects.link(cam_obj)

    scene.camera = cam_obj
    cam_obj.parent = None

    cam_obj.constraints.clear()
    track = cam_obj.constraints.new(type='TRACK_TO')
    track.target = focus_target
    track.track_axis = 'TRACK_NEGATIVE_Z'
    track.up_axis = 'UP_Y'

    if cam_obj.animation_data:
        cam_obj.animation_data_clear()
    if focus_target.animation_data:
        focus_target.animation_data_clear()

    max_radius = 0.0
    sample_frames = list(range(1, unfurl_end_frame + 1, 5))
    if unfurl_end_frame not in sample_frames:
        sample_frames.append(unfurl_end_frame)

    depsgraph = bpy.context.evaluated_depsgraph_get()

    for f in sample_frames:
        scene.frame_set(f)
        depsgraph.update()

        eval_mesh_obj = target_mesh.evaluated_get(depsgraph)
        me = eval_mesh_obj.to_mesh()
        matrix = eval_mesh_obj.matrix_world

        world_verts = [matrix @ v.co for v in me.vertices]
        eval_mesh_obj.to_mesh_clear()

        if not world_verts:
            continue

        min_x = min(v.x for v in world_verts)
        max_x = max(v.x for v in world_verts)
        min_y = min(v.y for v in world_verts)
        max_y = max(v.y for v in world_verts)
        min_z = min(v.z for v in world_verts)
        max_z = max(v.z for v in world_verts)

        center = Vector(((min_x + max_x) * 0.5, (min_y + max_y) * 0.5, (min_z + max_z) * 0.5))

        for v in world_verts:
            dist = (v - center).length
            if dist > max_radius:
                max_radius = dist

    padded_radius = max_radius * 1.15

    sensor_height = cam_data.sensor_height
    focal_length = cam_data.lens

    fov_v = 2.0 * math.atan((sensor_height * 0.5) / focal_length)
    aspect_ratio = scene.render.resolution_x / scene.render.resolution_y
    fov_h = 2.0 * math.atan(math.tan(fov_v * 0.5) * aspect_ratio)

    min_fov = min(fov_v, fov_h)
    required_dist = padded_radius / math.tan(min_fov * 0.5)

    root_bone_name = "mixamorig:Hips" if "mixamorig:Hips" in target_rig.pose.bones else target_rig.pose.bones[0].name

    for f in range(1, unfurl_end_frame + 1):
        scene.frame_set(f)
        depsgraph.update()

        eval_mesh_obj = target_mesh.evaluated_get(depsgraph)
        bbox_world = [eval_mesh_obj.matrix_world @ Vector(corner) for corner in eval_mesh_obj.bound_box]
        min_z = min(v.z for v in bbox_world)
        max_z = max(v.z for v in bbox_world)
        target_mid_z = min_z + ((max_z - min_z) * 0.50)

        eval_rig = target_rig.evaluated_get(depsgraph)
        pbone = eval_rig.pose.bones[root_bone_name]
        bone_world_matrix = eval_rig.matrix_world @ pbone.matrix

        char_ground_loc = Vector((bone_world_matrix.translation.x, bone_world_matrix.translation.y, min_z))

        focus_target.location = Vector((char_ground_loc.x, char_ground_loc.y, target_mid_z))
        focus_target.keyframe_insert(data_path="location", frame=f)

        rig_matrix = eval_rig.matrix_world
        forward_vec = (rig_matrix.to_3x3() @ Vector((0, 1, 0))).xy.normalized()
        if forward_vec.length == 0:
            forward_vec = Vector((0, -1))

        base_angle = math.atan2(forward_vec.y, forward_vec.x)

        if f <= turn_start:
            angle_offset = math.radians(-45.0)
        elif f >= turn_end_frame:
            angle_offset = 0.0
        else:
            t = (f - turn_start) / (turn_end_frame - turn_start)
            angle_offset = math.radians(-45.0) * (1.0 - t)

        cam_angle = base_angle + angle_offset

        cam_x = char_ground_loc.x - (required_dist * math.cos(cam_angle))
        cam_y = char_ground_loc.y - (required_dist * math.sin(cam_angle))

        cam_obj.location = Vector((cam_x, cam_y, target_mid_z))
        cam_obj.keyframe_insert(data_path="location", frame=f)

    for obj in [cam_obj, focus_target]:
        if obj and obj.animation_data and obj.animation_data.action:
            action = obj.animation_data.action
            fcurves = getattr(action, "fcurves", None)
            if fcurves is None and hasattr(action, "bindings"):
                fcurves = [fc for b in action.bindings for fc in b.fcurves if hasattr(b, "fcurves")]
            if fcurves:
                for fc in fcurves:
                    for kp in fc.keyframe_points:
                        kp.interpolation = 'LINEAR'

    for obj in [target_rig, target_mesh]:
        if obj and obj.animation_data and obj.animation_data.nla_tracks:
            for track in obj.animation_data.nla_tracks:
                for strip in track.strips:
                    strip.extrapolation = 'HOLD'

    scene.frame_set(1)
    depsgraph.update()

    return cam_obj, focus_target


def frame_scroll_image(cam_obj, focus_target, target_mesh, image_path, scroll_width=0.35, zoom_duration=45):
    scene = bpy.context.scene
    unfurl_end_frame = scene.frame_end
    zoom_end_frame = unfurl_end_frame + zoom_duration

    scene.frame_set(unfurl_end_frame)
    bpy.context.view_layer.update()

    img_aspect = 1.0
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        if img.size[1] > 0:
            img_aspect = img.size[0] / img.size[1]

    tile_height = scroll_width * img_aspect
    cam_data = cam_obj.data
    aspect_ratio = scene.render.resolution_x / scene.render.resolution_y
    sensor_width = cam_data.sensor_height * aspect_ratio

    fov_h_half = math.atan((sensor_width / 2.0) / cam_data.lens)
    target_framing_distance = (tile_height / 2.0) / math.tan(fov_h_half)

    mesh_matrix = target_mesh.matrix_world
    depsgraph = bpy.context.evaluated_depsgraph_get()
    eval_mesh = target_mesh.evaluated_get(depsgraph)

    r_hand_name, l_hand_name = find_hand_vertex_groups(target_mesh)
    vg_r = eval_mesh.vertex_groups.get(r_hand_name)
    vg_l = eval_mesh.vertex_groups.get(l_hand_name)

    r_verts = [mesh_matrix @ v.co for v in eval_mesh.data.vertices if any(g.group == vg_r.index for g in v.groups)] if vg_r else []
    l_verts = [mesh_matrix @ v.co for v in eval_mesh.data.vertices if any(g.group == vg_l.index for g in v.groups)] if vg_l else []

    target_height = target_mesh.dimensions.z if target_mesh.dimensions.z > 0 else 1.8
    torso_fallback = mesh_matrix.translation + Vector((0, 0, target_height * 0.55))

    r_hand_pos = sum(r_verts, Vector((0, 0, 0))) / len(r_verts) if r_verts else torso_fallback
    l_hand_pos = sum(l_verts, Vector((0, 0, 0))) / len(l_verts) if l_verts else torso_fallback

    scroll_midpoint = (r_hand_pos + l_hand_pos) * 0.5

    cam_pos = cam_obj.matrix_world.translation.copy()
    cam_dir = cam_pos - scroll_midpoint
    cam_dir.z = 0.0

    scroll_normal = cam_dir.normalized() if cam_dir.length > 0.001 else Vector((0, -1, 0))

    tile_center = scroll_midpoint + (scroll_normal * 0.10)
    end_pos = tile_center + (scroll_normal * target_framing_distance)

    cam_obj.parent = None
    focus_target.parent = None

    current_world_focus = focus_target.matrix_world.translation.copy()
    focus_target.matrix_world.translation = current_world_focus

    focus_target.location = current_world_focus
    focus_target.keyframe_insert(data_path="location", frame=unfurl_end_frame)
    focus_target.location = tile_center
    focus_target.keyframe_insert(data_path="location", frame=zoom_end_frame)

    cam_obj.location = cam_pos
    cam_obj.keyframe_insert(data_path="location", frame=unfurl_end_frame)
    cam_obj.location = end_pos
    cam_obj.keyframe_insert(data_path="location", frame=zoom_end_frame)

    for obj in [cam_obj, focus_target]:
        for fc in get_object_fcurves(obj):
            for kp in fc.keyframe_points:
                if kp.co.x >= unfurl_end_frame:
                    kp.interpolation = 'BEZIER'
                    kp.handle_left_type = 'AUTO_CLAMPED'
                    kp.handle_right_type = 'AUTO_CLAMPED'

    scene.frame_end = zoom_end_frame
    return zoom_end_frame

# ==============================================================================
# GROUND & SKY DOME SETUP (PBR ENHANCED)
# ==============================================================================

def ensure_world_sphere_size(cam_obj, ground_obj, safety_multiplier=2.5):
    if not cam_obj or not ground_obj:
        return

    scene = bpy.context.scene
    max_cam_dist = 0.0

    for frame in range(scene.frame_start, scene.frame_end + 1, 5):
        scene.frame_set(frame)
        bpy.context.view_layer.update()
        max_cam_dist = max(max_cam_dist, cam_obj.matrix_world.translation.length)

    world_radius = max(25.0, max_cam_dist * safety_multiplier)

    mod = ground_obj.modifiers.get("Grass_Scatter_GN")
    if mod and mod.node_group:
        for node in mod.node_group.nodes:
            if node.bl_idname == "GeometryNodeMeshIcoSphere":
                node.inputs["Radius"].default_value = world_radius

    if ground_obj.type == 'MESH':
        bpy.context.view_layer.objects.active = ground_obj
        bpy.ops.object.mode_set(mode='OBJECT')
        ground_scale = world_radius / 12.5
        ground_obj.scale = (ground_scale, ground_scale, 1.0)
        bpy.ops.object.transform_apply(location=False, rotation=False, scale=True)

    cam_obj.data.clip_end = max(cam_obj.data.clip_end, world_radius * 2.5)
    scene.frame_set(1)


def find_grass_maps(grass_dir):

    if not os.path.exists(grass_dir):
        return {}

    extensions = ['*.jpg', '*.jpeg', '*.png', '*.tif', '*.tiff']
    matches = []
    for ext in extensions:
        matches.extend(glob.glob(os.path.join(grass_dir, ext)))
        matches.extend(glob.glob(os.path.join(grass_dir, "**", ext), recursive=True))

    maps = {}
    for m in matches:
        fn = os.path.basename(m).lower()
        if any(k in fn for k in ["color", "diff", "albedo", "col", "basecolor"]) and "color" not in maps:
            maps["color"] = m
        elif "normalgl" in fn or ("normal" in fn and "dx" not in fn) and "normal" not in maps:
            maps["normal"] = m
        elif "roughness" in fn or "rough" in fn and "roughness" not in maps:
            maps["roughness"] = m
        elif any(k in fn for k in ["disp", "height", "displacement"]) and "disp" not in maps:
            maps["disp"] = m
        elif any(k in fn for k in ["ao", "ambientocclusion", "occlusion"]) and "ao" not in maps:
            maps["ao"] = m

    return maps


def create_ground_material(grass_dir=None):
    mat = bpy.data.materials.new(name="Ground_Soil_Material")
    mat.use_nodes = True
    nodes, links = mat.node_tree.nodes, mat.node_tree.links
    nodes.clear()

    out_node = nodes.new(type="ShaderNodeOutputMaterial")
    bsdf = nodes.new(type="ShaderNodeBsdfPrincipled")

    pbr_maps = find_grass_maps(grass_dir) if grass_dir else {}

    tex_coord = nodes.new(type="ShaderNodeTexCoord")
    mapping = nodes.new(type="ShaderNodeMapping")
    mapping.inputs["Scale"].default_value = (20.0, 20.0, 1.0)
    links.new(tex_coord.outputs["UV"], mapping.inputs["Vector"])

    # 1. Base Color & Ambient Occlusion (AO)
    if "color" in pbr_maps:
        tex_color = nodes.new(type="ShaderNodeTexImage")
        tex_color.image = bpy.data.images.load(pbr_maps["color"])
        links.new(mapping.outputs["Vector"], tex_color.inputs["Vector"])

        if "ao" in pbr_maps:
            tex_ao = nodes.new(type="ShaderNodeTexImage")
            tex_ao.image = bpy.data.images.load(pbr_maps["ao"])
            tex_ao.image.colorspace_settings.name = 'Non-Color'
            links.new(mapping.outputs["Vector"], tex_ao.inputs["Vector"])

            mix_ao = nodes.new(type="ShaderNodeMix")
            mix_ao.data_type = 'RGBA'
            mix_ao.blend_type = 'MULTIPLY'
            mix_ao.inputs["Factor"].default_value = 0.7

            links.new(tex_color.outputs["Color"], mix_ao.inputs[6])
            links.new(tex_ao.outputs["Color"], mix_ao.inputs[7])
            links.new(mix_ao.outputs[2], bsdf.inputs["Base Color"])
        else:
            links.new(tex_color.outputs["Color"], bsdf.inputs["Base Color"])
    else:
        noise = nodes.new("ShaderNodeTexNoise")
        noise.inputs["Scale"].default_value = 18.0
        noise.inputs["Detail"].default_value = 6.0

        color_ramp = nodes.new("ShaderNodeValToRGB")
        color_ramp.color_ramp.elements[0].color = (0.04, 0.025, 0.01, 1.0)
        color_ramp.color_ramp.elements[1].color = (0.05, 0.07, 0.02, 1.0)

        links.new(noise.outputs["Factor"], color_ramp.inputs["Factor"])
        links.new(color_ramp.outputs["Color"], bsdf.inputs["Base Color"])

    # 2. Roughness Map
    if "roughness" in pbr_maps:
        tex_rough = nodes.new(type="ShaderNodeTexImage")
        tex_rough.image = bpy.data.images.load(pbr_maps["roughness"])
        tex_rough.image.colorspace_settings.name = 'Non-Color'
        links.new(mapping.outputs["Vector"], tex_rough.inputs["Vector"])
        links.new(tex_rough.outputs["Color"], bsdf.inputs["Roughness"])
    else:
        bsdf.inputs["Roughness"].default_value = 0.95

    # 3. Normal Map
    if "normal" in pbr_maps:
        tex_norm = nodes.new(type="ShaderNodeTexImage")
        tex_norm.image = bpy.data.images.load(pbr_maps["normal"])
        tex_norm.image.colorspace_settings.name = 'Non-Color'
        links.new(mapping.outputs["Vector"], tex_norm.inputs["Vector"])

        norm_map = nodes.new(type="ShaderNodeNormalMap")
        norm_map.inputs["Strength"].default_value = 1.8
        links.new(tex_norm.outputs["Color"], norm_map.inputs["Color"])
        links.new(norm_map.outputs["Normal"], bsdf.inputs["Normal"])

    # 4. Bump / Displacement Map Fallback
    if "disp" in pbr_maps and "normal" not in pbr_maps:
        tex_disp = nodes.new(type="ShaderNodeTexImage")
        tex_disp.image = bpy.data.images.load(pbr_maps["disp"])
        tex_disp.image.colorspace_settings.name = 'Non-Color'
        links.new(mapping.outputs["Vector"], tex_disp.inputs["Vector"])

        bump_node = nodes.new(type="ShaderNodeBump")
        bump_node.inputs["Strength"].default_value = 0.5
        links.new(tex_disp.outputs["Color"], bump_node.inputs["Height"])
        links.new(bump_node.outputs["Normal"], bsdf.inputs["Normal"])

    links.new(bsdf.outputs["BSDF"], out_node.inputs["Surface"])
    return mat


def create_sky_material(sky_hdri_path):
    mat = bpy.data.materials.new(name="Sky_Dome_Material")
    mat.use_nodes = True
    nodes, links = mat.node_tree.nodes, mat.node_tree.links
    nodes.clear()

    out_node = nodes.new("ShaderNodeOutputMaterial")
    emission = nodes.new("ShaderNodeEmission")
    emission.inputs["Strength"].default_value = 1.0

    transparent = nodes.new("ShaderNodeBsdfTransparent")
    mix_shader = nodes.new("ShaderNodeMixShader")
    light_path = nodes.new("ShaderNodeLightPath")

    links.new(light_path.outputs["Is Camera Ray"], mix_shader.inputs["Fac"])
    links.new(transparent.outputs["BSDF"], mix_shader.inputs[1])
    links.new(emission.outputs["Emission"], mix_shader.inputs[2])
    links.new(mix_shader.outputs["Shader"], out_node.inputs["Surface"])

    if os.path.exists(sky_hdri_path):
        tex_env = nodes.new("ShaderNodeTexEnvironment")
        tex_env.image = bpy.data.images.load(sky_hdri_path)
        links.new(tex_env.outputs["Color"], emission.inputs["Color"])

    return mat


def setup_grass_ground_geometry_nodes(grass_dir, sky_hdri_path, ground_z_offset=0.0):
    bpy.ops.mesh.primitive_plane_add(size=60, location=(0, 0, ground_z_offset))
    ground_obj = bpy.context.active_object
    ground_obj.name = "Grass_Ground_Plane"

    ground_mat = create_ground_material(grass_dir=grass_dir)
    sky_mat = create_sky_material(sky_hdri_path)
    ground_obj.data.materials.append(ground_mat)
    ground_obj.data.materials.append(sky_mat)

    gn_mod = ground_obj.modifiers.new(name="Grass_Scatter_GN", type='NODES')
    gn_group = bpy.data.node_groups.new("Grass_Scatter_Group", type='GeometryNodeTree')
    gn_mod.node_group = gn_group

    nodes = gn_group.nodes
    links = gn_group.links
    nodes.clear()

    group_in = nodes.new("NodeGroupInput")
    gn_group.interface.new_socket("Geometry", in_out='INPUT', socket_type='NodeSocketGeometry')
    group_out = nodes.new("NodeGroupOutput")
    gn_group.interface.new_socket("Geometry", in_out='OUTPUT', socket_type='NodeSocketGeometry')

    sky_sphere = nodes.new("GeometryNodeMeshIcoSphere")
    sky_sphere.inputs["Radius"].default_value = 12.5
    sky_sphere.inputs["Subdivisions"].default_value = 4

    flip_faces = nodes.new("GeometryNodeFlipFaces")
    links.new(sky_sphere.outputs["Mesh"], flip_faces.inputs["Mesh"])

    transform_sky = nodes.new("GeometryNodeTransform")
    transform_sky.inputs["Translation"].default_value = (0.0, 0.0, ground_z_offset)
    links.new(flip_faces.outputs["Mesh"], transform_sky.inputs["Geometry"])

    set_sky_mat = nodes.new("GeometryNodeSetMaterial")
    set_sky_mat.inputs["Material"].default_value = sky_mat
    links.new(transform_sky.outputs["Geometry"], set_sky_mat.inputs["Geometry"])

    set_ground_mat = nodes.new("GeometryNodeSetMaterial")
    set_ground_mat.inputs["Material"].default_value = ground_mat
    links.new(group_in.outputs["Geometry"], set_ground_mat.inputs["Geometry"])

    subdivide = nodes.new("GeometryNodeSubdivideMesh")
    subdivide.inputs["Level"].default_value = 5
    links.new(group_in.outputs["Geometry"], subdivide.inputs["Mesh"])

    noise_tex = nodes.new("ShaderNodeTexNoise")
    noise_tex.inputs["Scale"].default_value = 0.05
    noise_tex.inputs["Detail"].default_value = 2.0

    vector_math = nodes.new("ShaderNodeVectorMath")
    vector_math.operation = "SCALE"
    vector_math.inputs["Scale"].default_value = 0.8
    links.new(noise_tex.outputs["Color"], vector_math.inputs["Vector"])

    set_pos = nodes.new("GeometryNodeSetPosition")
    links.new(subdivide.outputs["Mesh"], set_pos.inputs["Geometry"])
    links.new(vector_math.outputs["Vector"], set_pos.inputs["Offset"])

    set_ground_mat = nodes.new("GeometryNodeSetMaterial")
    set_ground_mat.inputs["Material"].default_value = ground_mat
    links.new(set_pos.outputs["Geometry"], set_ground_mat.inputs["Geometry"])

    join_geo = nodes.new("GeometryNodeJoinGeometry")
    links.new(set_ground_mat.outputs["Geometry"], join_geo.inputs["Geometry"])
    links.new(set_sky_mat.outputs["Geometry"], join_geo.inputs["Geometry"])

    links.new(join_geo.outputs["Geometry"], group_out.inputs["Geometry"])

    return ground_obj

# ==============================================================================
# SCROLL MATERIAL & GEOMETRY NODES SETUP
# ==============================================================================

def create_image_scroll_material(image_path):
    mat_name = "Scroll_Texture_Mat"
    if mat := bpy.data.materials.get(mat_name):
        bpy.data.materials.remove(mat)

    mat = bpy.data.materials.new(name=mat_name)
    mat.use_nodes = True
    nodes, links = mat.node_tree.nodes, mat.node_tree.links
    nodes.clear()

    output_node = nodes.new("ShaderNodeOutputMaterial")
    bsdf = nodes.new("ShaderNodeBsdfPrincipled")
    bsdf.inputs["Roughness"].default_value = 0.5

    tex_img = nodes.new("ShaderNodeTexImage")
    if os.path.exists(image_path):
        tex_img.image = bpy.data.images.load(image_path)
        tex_img.extension = 'REPEAT'

    attr_node = nodes.new("ShaderNodeAttribute")
    attr_node.attribute_name = "UVMap"

    links.new(attr_node.outputs["Vector"], tex_img.inputs["Vector"])
    links.new(tex_img.outputs["Color"], bsdf.inputs["Base Color"])
    links.new(bsdf.outputs["BSDF"], output_node.inputs["Surface"])

    return mat, tex_img.image if os.path.exists(image_path) else None


def create_pure_geometry_node_scroll(image_path=""):
    gn_group = bpy.data.node_groups.new("Scroll_Pure_GN", type="GeometryNodeTree")

    gn_group.interface.new_socket(name="Geometry", in_out="OUTPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Character Mesh", in_out="INPUT", socket_type="NodeSocketObject")

    sock_r_vg = gn_group.interface.new_socket(name="Right Hand Group", in_out="INPUT", socket_type="NodeSocketString")
    sock_r_vg.default_value = "mixamorig:RightHand"

    sock_l_vg = gn_group.interface.new_socket(name="Left Hand Group", in_out="INPUT", socket_type="NodeSocketString")
    sock_l_vg.default_value = "mixamorig:LeftHand"

    sock_offset = gn_group.interface.new_socket(name="Forward Offset", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_offset.default_value = 0.60

    sock_unfurl = gn_group.interface.new_socket(name="Unfurl", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_unfurl.default_value = 0.0

    sock_width = gn_group.interface.new_socket(name="Scroll Width", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_width.default_value = 0.35

    nodes, links = gn_group.nodes, gn_group.links
    in_node, out_node = nodes.new("NodeGroupInput"), nodes.new("NodeGroupOutput")

    char_info = nodes.new("GeometryNodeObjectInfo")
    char_info.transform_space = "RELATIVE"
    links.new(in_node.outputs["Character Mesh"], char_info.inputs["Object"])

    pos_node = nodes.new("GeometryNodeInputPosition")

    named_r = create_gn_named_attr_node(nodes)
    named_r.data_type = "FLOAT"
    named_r.inputs["Name"].default_value = "mixamorig:RightHand"
    links.new(in_node.outputs["Right Hand Group"], named_r.inputs["Name"])

    attr_r_out = named_r.outputs.get("Attribute") or named_r.outputs[0]

    cmp_r = nodes.new("FunctionNodeCompare")
    cmp_r.data_type = "FLOAT"
    cmp_r.operation = "GREATER_THAN"
    cmp_r.inputs[1].default_value = 0.001
    links.new(attr_r_out, cmp_r.inputs[0])

    stat_r = nodes.new("GeometryNodeAttributeStatistic")
    stat_r.data_type = "FLOAT_VECTOR"
    links.new(char_info.outputs["Geometry"], stat_r.inputs["Geometry"])
    links.new(cmp_r.outputs["Result"], stat_r.inputs["Selection"])
    links.new(pos_node.outputs["Position"], stat_r.inputs["Attribute"])

    named_l = create_gn_named_attr_node(nodes)
    named_l.data_type = "FLOAT"
    named_l.inputs["Name"].default_value = "mixamorig:LeftHand"
    links.new(in_node.outputs["Left Hand Group"], named_l.inputs["Name"])

    attr_l_out = named_l.outputs.get("Attribute") or named_l.outputs[0]

    cmp_l = nodes.new("FunctionNodeCompare")
    cmp_l.data_type = "FLOAT"
    cmp_l.operation = "GREATER_THAN"
    cmp_l.inputs[1].default_value = 0.001
    links.new(attr_l_out, cmp_l.inputs[0])

    stat_l = nodes.new("GeometryNodeAttributeStatistic")
    stat_l.data_type = "FLOAT_VECTOR"
    links.new(char_info.outputs["Geometry"], stat_l.inputs["Geometry"])
    links.new(cmp_l.outputs["Result"], stat_l.inputs["Selection"])
    links.new(pos_node.outputs["Position"], stat_l.inputs["Attribute"])

    neg_offset = nodes.new("ShaderNodeMath")
    neg_offset.operation = "MULTIPLY"
    neg_offset.inputs[1].default_value = -1.0
    links.new(in_node.outputs["Forward Offset"], neg_offset.inputs[0])

    forward_vec = nodes.new("ShaderNodeCombineXYZ")
    forward_vec.inputs["X"].default_value = 0.0
    links.new(neg_offset.outputs["Value"], forward_vec.inputs["Y"])
    forward_vec.inputs["Z"].default_value = 0.0

    start_pos = nodes.new("ShaderNodeVectorMath")
    start_pos.operation = "ADD"
    links.new(stat_r.outputs["Mean"], start_pos.inputs[0])
    links.new(forward_vec.outputs["Vector"], start_pos.inputs[1])

    target_pos = nodes.new("ShaderNodeVectorMath")
    target_pos.operation = "ADD"
    links.new(stat_l.outputs["Mean"], target_pos.inputs[0])
    links.new(forward_vec.outputs["Vector"], target_pos.inputs[1])

    delta_vec = nodes.new("ShaderNodeVectorMath")
    delta_vec.operation = "SUBTRACT"
    links.new(target_pos.outputs["Vector"], delta_vec.inputs[0])
    links.new(start_pos.outputs["Vector"], delta_vec.inputs[1])

    dist_node = nodes.new("ShaderNodeVectorMath")
    dist_node.operation = "DISTANCE"
    links.new(start_pos.outputs["Vector"], dist_node.inputs[0])

    scale_vec = nodes.new("ShaderNodeVectorMath")
    scale_vec.operation = "SCALE"
    links.new(delta_vec.outputs["Vector"], scale_vec.inputs["Vector"])
    links.new(in_node.outputs["Unfurl"], scale_vec.inputs["Scale"])

    end_pos = nodes.new("ShaderNodeVectorMath")
    end_pos.operation = "ADD"
    links.new(start_pos.outputs["Vector"], end_pos.inputs[0])
    links.new(scale_vec.outputs["Vector"], end_pos.inputs[1])
    links.new(end_pos.outputs["Vector"], dist_node.inputs[1])

    curve_line = nodes.new("GeometryNodeCurvePrimitiveLine")
    links.new(start_pos.outputs["Vector"], curve_line.inputs["Start"])
    links.new(end_pos.outputs["Vector"], curve_line.inputs["End"])

    set_tilt = nodes.new("GeometryNodeSetCurveTilt")
    set_tilt.inputs["Tilt"].default_value = 1.5708
    links.new(curve_line.outputs["Curve"], set_tilt.inputs["Curve"])

    spline_param_main = nodes.new("GeometryNodeSplineParameter")
    store_u = nodes.new("GeometryNodeStoreNamedAttribute")
    store_u.inputs["Name"].default_value = "U_Factor"
    store_u.data_type = 'FLOAT'
    store_u.domain = 'POINT'
    links.new(set_tilt.outputs["Curve"], store_u.inputs["Geometry"])

    spline_factor_out = spline_param_main.outputs.get("Factor") or spline_param_main.outputs[0]
    links.new(spline_factor_out, store_u.inputs["Value"])

    half_width_neg = nodes.new("ShaderNodeMath")
    half_width_neg.operation = "MULTIPLY"
    half_width_neg.inputs[1].default_value = -0.5
    links.new(in_node.outputs["Scroll Width"], half_width_neg.inputs[0])

    half_width_pos = nodes.new("ShaderNodeMath")
    half_width_pos.operation = "MULTIPLY"
    half_width_pos.inputs[1].default_value = 0.5
    links.new(in_node.outputs["Scroll Width"], half_width_pos.inputs[0])

    combine_start = nodes.new("ShaderNodeCombineXYZ")
    links.new(half_width_neg.outputs["Value"], combine_start.inputs["X"])

    combine_end = nodes.new("ShaderNodeCombineXYZ")
    links.new(half_width_pos.outputs["Value"], combine_end.inputs["X"])

    profile_line = nodes.new("GeometryNodeCurvePrimitiveLine")
    links.new(combine_start.outputs["Vector"], profile_line.inputs["Start"])
    links.new(combine_end.outputs["Vector"], profile_line.inputs["End"])

    spline_param_prof = nodes.new("GeometryNodeSplineParameter")
    store_v = nodes.new("GeometryNodeStoreNamedAttribute")
    store_v.inputs["Name"].default_value = "V_Factor"
    store_v.data_type = 'FLOAT'
    store_v.domain = 'POINT'
    links.new(profile_line.outputs["Curve"], store_v.inputs["Geometry"])

    prof_factor_out = spline_param_prof.outputs.get("Factor") or spline_param_prof.outputs[0]
    links.new(prof_factor_out, store_v.inputs["Value"])

    curve_to_mesh = nodes.new("GeometryNodeCurveToMesh")
    links.new(store_u.outputs["Geometry"], curve_to_mesh.inputs["Curve"])
    links.new(store_v.outputs["Geometry"], curve_to_mesh.inputs["Profile Curve"])

    img_aspect = 1.0
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        if img.size[1] > 0:
            img_aspect = img.size[0] / img.size[1]

    h_times_aspect = nodes.new("ShaderNodeMath")
    h_times_aspect.operation = "MULTIPLY"
    h_times_aspect.inputs[1].default_value = img_aspect
    links.new(in_node.outputs["Scroll Width"], h_times_aspect.inputs[0])

    tile_count = nodes.new("ShaderNodeMath")
    tile_count.operation = "DIVIDE"
    links.new(dist_node.outputs["Value"], tile_count.inputs[0])
    links.new(h_times_aspect.outputs["Value"], tile_count.inputs[1])

    read_u = create_gn_named_attr_node(nodes)
    read_u.data_type = 'FLOAT'
    read_u.inputs["Name"].default_value = "U_Factor"
    read_u_out = read_u.outputs.get("Attribute") or read_u.outputs[0]

    scaled_u = nodes.new("ShaderNodeMath")
    scaled_u.operation = "MULTIPLY"
    links.new(read_u_out, scaled_u.inputs[0])
    links.new(tile_count.outputs["Value"], scaled_u.inputs[1])

    read_v = create_gn_named_attr_node(nodes)
    read_v.data_type = 'FLOAT'
    read_v.inputs["Name"].default_value = "V_Factor"
    read_v_out = read_v.outputs.get("Attribute") or read_v.outputs[0]

    flip_v = nodes.new("ShaderNodeMath")
    flip_v.operation = "SUBTRACT"
    flip_v.inputs[0].default_value = 1.0
    links.new(read_v_out, flip_v.inputs[1])

    combine_uv = nodes.new("ShaderNodeCombineXYZ")
    links.new(scaled_u.outputs["Value"], combine_uv.inputs["X"])
    links.new(flip_v.outputs["Value"], combine_uv.inputs["Y"])

    store_uv = nodes.new("GeometryNodeStoreNamedAttribute")
    store_uv.inputs["Name"].default_value = "UVMap"
    store_uv.data_type = 'FLOAT2'
    store_uv.domain = 'CORNER'
    links.new(curve_to_mesh.outputs["Mesh"], store_uv.inputs["Geometry"])
    links.new(combine_uv.outputs["Vector"], store_uv.inputs["Value"])

    set_material = nodes.new("GeometryNodeSetMaterial")
    scroll_mat, _ = create_image_scroll_material(image_path)
    set_material.inputs["Material"].default_value = scroll_mat
    links.new(store_uv.outputs["Geometry"], set_material.inputs["Geometry"])

    links.new(set_material.outputs["Geometry"], out_node.inputs["Geometry"])

    return gn_group, scroll_mat


def setup_geometry_node_scroll(target_mesh, unfurl_start, unfurl_end, image_path=""):
    mesh_data = bpy.data.meshes.new("Scroll_Procedural_Mesh")
    scroll_obj = bpy.data.objects.new("Scroll_GN_Object", mesh_data)
    bpy.context.scene.collection.objects.link(scroll_obj)

    mod = scroll_obj.modifiers.new("Scroll_GN", type="NODES")
    gn_tree, scroll_mat = create_pure_geometry_node_scroll(image_path=image_path)
    mod.node_group = gn_tree

    scroll_obj.data.materials.append(scroll_mat)

    r_hand_name, l_hand_name = find_hand_vertex_groups(target_mesh)

    unfurl_id = None
    for item in gn_tree.interface.items_tree:
        if getattr(item, "in_out", None) == "INPUT":
            if item.name == "Character Mesh":
                mod[item.identifier] = target_mesh
            elif item.name == "Right Hand Group":
                mod[item.identifier] = r_hand_name
            elif item.name == "Left Hand Group":
                mod[item.identifier] = l_hand_name
            elif item.name == "Forward Offset":
                mod[item.identifier] = 0.60
            elif item.name == "Scroll Width":
                mod[item.identifier] = 0.35
            elif item.name == "Unfurl":
                unfurl_id = item.identifier

    scroll_obj.hide_viewport = True
    scroll_obj.hide_render = True
    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=1)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=1)

    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=unfurl_start - 1)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=unfurl_start - 1)

    scroll_obj.hide_viewport = False
    scroll_obj.hide_render = False
    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=unfurl_start)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=unfurl_start)

    for fc in get_object_fcurves(scroll_obj):
        if fc.data_path in ["hide_viewport", "hide_render"]:
            for kp in fc.keyframe_points:
                kp.interpolation = 'CONSTANT'

    if unfurl_id:
        mod[unfurl_id] = 0.001
        mod.keyframe_insert(data_path=f'["{unfurl_id}"]', frame=unfurl_start)

        mod[unfurl_id] = 1.0
        mod.keyframe_insert(data_path=f'["{unfurl_id}"]', frame=unfurl_end)

    return scroll_obj

# ==============================================================================
# MAIN CONTROLLER
# ==============================================================================

def import_and_extract_action(fbx_path, action_name):
    existing_objects = set(bpy.data.objects)
    bpy.ops.import_scene.fbx(filepath=fbx_path)
    bpy.context.view_layer.update()

    imported_objects = set(bpy.data.objects) - existing_objects
    extracted_action = None
    for obj in imported_objects:
        if obj.animation_data and obj.animation_data.action:
            extracted_action = obj.animation_data.action
            extracted_action.name = action_name
            extracted_action.use_fake_user = True
            break
    return extracted_action, imported_objects


def purge_imported_objects(objects_to_remove):
    for obj in objects_to_remove:
        if obj.name in bpy.data.objects:
            if obj.animation_data:
                obj.animation_data_clear()
            for coll in list(obj.users_collection):
                coll.objects.unlink(obj)
    bpy.context.view_layer.update()


def setup_world_hdri_lighting(sky_hdri_path, strength=2.5):
    scene = bpy.context.scene
    if not scene.world:
        scene.world = bpy.data.worlds.new("World")

    scene.world.use_nodes = True
    nodes = scene.world.node_tree.nodes
    links = scene.world.node_tree.links
    nodes.clear()

    out_node = nodes.new("ShaderNodeOutputWorld")
    bg_node = nodes.new("ShaderNodeBackground")
    bg_node.inputs["Strength"].default_value = strength

    links.new(bg_node.outputs["Background"], out_node.inputs["Surface"])

    if os.path.exists(sky_hdri_path):
        tex_env = nodes.new("ShaderNodeTexEnvironment")
        tex_env.image = bpy.data.images.load(sky_hdri_path)
        links.new(tex_env.outputs["Color"], bg_node.inputs["Color"])


def main():
    cleanup_previous_run()

    walk_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Mutant Walking.fbx"
    turn_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Right Turn 90.fbx"
    unfurl_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Mutant Idle.fbx"

    sky_hdri_path = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/horn-koppe_spring_4k.exr"
    output_blend = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/output_scene.blend"

    image_path = "/Users/dobbinp1/Downloads/sea_PanoramaIMGP5963.jpg"
    grass_dir = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Grass004_4K-JPG"

    scene = bpy.context.scene
    scene.view_settings.exposure = 1.0
    scene.view_settings.view_transform = 'AgX'
    scene.view_settings.look = 'AgX - Medium High Contrast'

    setup_world_hdri_lighting(sky_hdri_path, strength=1.0)

    ground_obj = setup_grass_ground_geometry_nodes(grass_dir, sky_hdri_path, ground_z_offset=-0.35)

    CYCLES = 4
    CROSSFADE = 15
    TARGET_DISTANCE = 4.0

    walk_action, walk_objs = import_and_extract_action(walk_fbx, "walk")
    target_rig = next((obj for obj in walk_objs if obj.type == "ARMATURE"), None)
    target_mesh = next((obj for obj in walk_objs if obj.type == "MESH"), None)

    if not target_rig or not target_mesh:
        raise RuntimeError("Armature or Character Mesh not found in walk FBX.")

    target_rig.location.z = 0.0

    travel_axis_idx, motion_sign = detect_natural_forward_axis(walk_action, target_rig)
    initial_heading = math.radians(90.0) if travel_axis_idx == 0 else 0.0
    target_rig.rotation_euler[2] = initial_heading

    action_len = max(30, int(walk_action.frame_range[1] - walk_action.frame_range[0]))
    anim_data = target_rig.animation_data_create()

    walk_track = anim_data.nla_tracks.new()
    walk_track.name = "Walk_Track"

    current_frame = 1
    for i in range(CYCLES):
        strip = walk_track.strips.new(name=f"Walk_{i+1}", start=int(current_frame), action=walk_action)
        strip.blend_type = "REPLACE"
        current_frame += action_len

    walk_end_frame = int(current_frame)

    turn_action, turn_objs = import_and_extract_action(turn_fbx, "rightturn")
    purge_imported_objects(turn_objs)

    turn_track = anim_data.nla_tracks.new()
    turn_track.name = "RightTurn_Track"

    turn_start = int(walk_end_frame - CROSSFADE)
    turn_strip = turn_track.strips.new(name="RightTurn", start=turn_start, action=turn_action)
    turn_strip.blend_type = "REPLACE"
    turn_strip.blend_in = CROSSFADE
    turn_strip.use_auto_blend = True
    turn_strip.extrapolation = "NOTHING"
    turn_end_frame = int(turn_strip.frame_end)

    bpy.context.scene["turn_start_frame"] = turn_start

    unfurl_action, unfurl_objs = import_and_extract_action(unfurl_fbx, "unfurl")
    purge_imported_objects(unfurl_objs)

    unfurl_track = anim_data.nla_tracks.new()
    unfurl_track.name = "Unfurl_Track"

    unfurl_start = int(turn_end_frame - CROSSFADE)
    unfurl_strip = unfurl_track.strips.new(name="Unfurl", start=unfurl_start, action=unfurl_action)
    unfurl_strip.blend_type = "REPLACE"
    unfurl_strip.blend_in = CROSSFADE
    unfurl_strip.use_auto_blend = True
    unfurl_strip.extrapolation = "HOLD"
    unfurl_end_frame = int(unfurl_strip.frame_end)

    root_action = bpy.data.actions.new(name="World_Root_Motion")
    target_rig.animation_data.action = root_action

    if hasattr(root_action, "fcurves") and hasattr(root_action.fcurves, "new"):
        loc_fcurve = root_action.fcurves.new(data_path="location", index=travel_axis_idx)
        rot_fcurve = root_action.fcurves.new(data_path="rotation_euler", index=2)
        z_fcurve = root_action.fcurves.new(data_path="location", index=2)
    else:
        loc_fcurve = root_action.fcurve_ensure_for_datablock(datablock=target_rig, data_path="location", index=travel_axis_idx)
        rot_fcurve = root_action.fcurve_ensure_for_datablock(datablock=target_rig, data_path="rotation_euler", index=2)
        z_fcurve = root_action.fcurve_ensure_for_datablock(datablock=target_rig, data_path="location", index=2)

    z_fcurve.keyframe_points.insert(frame=1.0, value=0.0)
    z_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=0.0)

    final_target_value = TARGET_DISTANCE * motion_sign

    k_loc1 = loc_fcurve.keyframe_points.insert(frame=1.0, value=0.0)
    k_loc2 = loc_fcurve.keyframe_points.insert(frame=float(turn_start), value=final_target_value)
    k_loc3 = loc_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=final_target_value)

    k_loc1.interpolation = "LINEAR"
    k_loc2.interpolation = "CONSTANT"
    k_loc3.interpolation = "CONSTANT"

    target_heading = initial_heading - math.radians(90.0)

    k_rot1 = rot_fcurve.keyframe_points.insert(frame=1.0, value=initial_heading)
    k_rot2 = rot_fcurve.keyframe_points.insert(frame=float(turn_start), value=initial_heading)
    k_rot3 = rot_fcurve.keyframe_points.insert(frame=float(turn_end_frame), value=target_heading)
    k_rot4 = rot_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=target_heading)

    k_rot1.interpolation = "CONSTANT"
    k_rot2.interpolation = "BEZIER"
    k_rot3.interpolation = "CONSTANT"
    k_rot4.interpolation = "CONSTANT"

    root_track = anim_data.nla_tracks.new()
    root_track.name = "World_Root_Track"
    root_strip = root_track.strips.new(name="Root_Motion_Strip", start=1, action=root_action)
    root_strip.blend_type = "REPLACE"
    root_strip.extrapolation = "HOLD"

    anim_data.action = None

    setup_geometry_node_scroll(
        target_mesh=target_mesh,
        unfurl_start=unfurl_start,
        unfurl_end=unfurl_end_frame,
        image_path=image_path,
    )

    cam_obj, focus_target = setup_modern_reel_camera(
        target_rig=target_rig,
        target_mesh=target_mesh,
        turn_start=turn_start,
        turn_end_frame=turn_end_frame,
        unfurl_end_frame=unfurl_end_frame
    )

    bpy.context.scene.frame_start = 1
    bpy.context.scene.frame_end = unfurl_end_frame

    frame_scroll_image(
        cam_obj=cam_obj,
        focus_target=focus_target,
        target_mesh=target_mesh,
        image_path=image_path,
        scroll_width=0.35,
        zoom_duration=45
    )

    ensure_world_sphere_size(cam_obj, ground_obj, safety_multiplier=2.5)

    # --------------------------------------------------------------------------
    # VIDEO AUDIO ENCODING SETUP
    # --------------------------------------------------------------------------
    # 1. Generate VSE strips based on footfall bone positions
    generate_footstep_audio_strips(scene)

    # 2. Configure output render settings for MP4/AAC video encoding
    configure_render_audio_settings(scene)

    bpy.ops.wm.save_as_mainfile(filepath=output_blend)
    print(f"\n[SUCCESS] Saved scene to: {output_blend}")


if __name__ == "__main__":
    main()
```

```
import math
import mathutils
import os
import sys
import glob
import bpy
import aud
from bpy_extras import anim_utils
from bpy_extras.object_utils import world_to_camera_view
from bpy.app.handlers import persistent
from mathutils import Vector

# ==============================================================================
# CONFIGURATION
# ==============================================================================
FOOTSTEP_SOUND_PATH = "/Users/dobbinp1/Downloads/stomp.wav"
GROUND_THRESHOLD = 0.3  # Threshold for grounded character footstep detection

# Setup Audio Device
if "_audio_store" not in sys.modules:
    import types
    sys.modules["_audio_store"] = types.ModuleType("_audio_store")

audio_store = sys.modules["_audio_store"]
if not hasattr(audio_store, "device"):
    audio_store.device = aud.Device()

AUDIO_DEVICE = audio_store.device
SOUND_FACTORY = None
AUDIO_HANDLES = []

if os.path.exists(FOOTSTEP_SOUND_PATH):
    SOUND_FACTORY = aud.Sound(FOOTSTEP_SOUND_PATH).cache()
    print("[AUDIO] Loaded sound file successfully.")
else:
    print(f"[AUDIO ERROR] File not found at: {FOOTSTEP_SOUND_PATH}")

_left_foot_was_down = False
_right_foot_was_down = False
_last_frame = -1


def play_footstep():
    global AUDIO_HANDLES
    if SOUND_FACTORY:
        handle = AUDIO_DEVICE.play(SOUND_FACTORY)
        AUDIO_HANDLES.append(handle)
        if len(AUDIO_HANDLES) > 10:
            AUDIO_HANDLES.pop(0)


def find_foot_bone(pose_bones, side="Left"):
    target_keywords = [
        f"mixamorig:{side}foot",
        f"{side.lower()}foot",
        f"foot.{side[0].lower()}",
        f"foot_{side[0].lower()}"
    ]
    for bone in pose_bones:
        b_name = bone.name.lower()
        if any(kw in b_name for kw in target_keywords):
            return bone
    return None


@persistent
def footstep_frame_handler(scene):
    global _left_foot_was_down, _right_foot_was_down, _last_frame

    try:
        current_frame = scene.frame_current
        turn_start = scene.get("turn_start_frame", None)

        if turn_start is not None and current_frame >= turn_start:
            _left_foot_was_down = False
            _right_foot_was_down = False
            return

        if current_frame <= _last_frame or abs(current_frame - _last_frame) > 10:
            _left_foot_was_down = False
            _right_foot_was_down = False
        _last_frame = current_frame

        rig = next((o for o in scene.objects if o.type == 'ARMATURE'), None)
        if not rig:
            return

        depsgraph = bpy.context.evaluated_depsgraph_get()
        eval_rig = rig.evaluated_get(depsgraph)

        left_bone = find_foot_bone(eval_rig.pose.bones, "Left")
        right_bone = find_foot_bone(eval_rig.pose.bones, "Right")

        if not left_bone or not right_bone:
            return

        l_z = (eval_rig.matrix_world @ left_bone.matrix).translation.z
        r_z = (eval_rig.matrix_world @ right_bone.matrix).translation.z

        l_is_down = l_z <= GROUND_THRESHOLD
        r_is_down = r_z <= GROUND_THRESHOLD

        if l_is_down and not _left_foot_was_down:
            play_footstep()

        if r_is_down and not _right_foot_was_down:
            play_footstep()

        _left_foot_was_down = l_is_down
        _right_foot_was_down = r_is_down

    except Exception:
        import traceback
        traceback.print_exc()


def register_audio_handler():
    for h in list(bpy.app.handlers.frame_change_post):
        if getattr(h, "__name__", "") == "footstep_frame_handler":
            bpy.app.handlers.frame_change_post.remove(h)

    bpy.app.handlers.frame_change_post.append(footstep_frame_handler)


register_audio_handler()

# ==============================================================================
# CLEANUP & HELPERS
# ==============================================================================

def cleanup_previous_run():
    target_names = [
        "Master_NLA_Control",
        "Scroll_GN_Object",
        "Scroll_Start_Pin",
        "Scroll_End_Pin",
        "Scroll_Master_Container",
        "HDRI_Ground_Plane",
        "Ground_Mesh_Object",
        "Grass_Ground_Plane",
        "Grass_Blade_Instance",
        "Camera_Focus_Target",
        "Camera_Anchor",
        "Vertical_Reel_Camera"
    ]
    for obj in list(bpy.data.objects):
        if obj.name in target_names or (obj.name.startswith("mixamorig") and obj.type not in {'ARMATURE', 'MESH'}):
            if obj.animation_data:
                obj.animation_data_clear()
            bpy.data.objects.remove(obj, do_unlink=True)

    for action in list(bpy.data.actions):
        if action.users == 0 or action.use_fake_user:
            action.use_fake_user = False
            bpy.data.actions.remove(action)

    for ng in list(bpy.data.node_groups):
        if ng.name in ["Scroll_Unfurl_GN", "Scroll_Pure_GN", "HDRI_Ground_GN", "Grass_Ground_GN", "Grass_Scatter_Group"]:
            bpy.data.node_groups.remove(ng)

    for coll_name in ["Scroll_Master_Container", "Grass_Artifacts_Collection"]:
        if coll_name in bpy.data.collections:
            coll = bpy.data.collections[coll_name]
            for obj in list(coll.objects):
                bpy.data.objects.remove(obj, do_unlink=True)
            bpy.data.collections.remove(coll)

    bpy.context.view_layer.update()


def get_slot_fcurves(action, anim_target_object):
    fcurves = []
    if not action:
        return fcurves

    anim_data = getattr(anim_target_object, "animation_data", None)
    if anim_data and getattr(anim_data, "action_slot", None):
        slot = anim_data.action_slot
        channelbag = anim_utils.action_get_channelbag_for_slot(action, slot)
        if channelbag and hasattr(channelbag, "fcurves"):
            return list(channelbag.fcurves)

    if hasattr(action, "slots"):
        for slot in action.slots:
            for cb in getattr(slot, "channelbags", []):
                fcurves.extend(getattr(cb, "fcurves", []))
        if fcurves:
            return fcurves

    if hasattr(action, "fcurves"):
        return list(action.fcurves)

    return fcurves


def get_object_fcurves(obj):
    if not obj or not obj.animation_data or not obj.animation_data.action:
        return []
    return get_slot_fcurves(obj.animation_data.action, obj)


def detect_natural_forward_axis(action, rig_object):
    fcurves = get_slot_fcurves(action, rig_object)
    max_delta = 0.0
    best_axis = 1
    net_delta = 0.0

    start_frame, end_frame = action.frame_range

    for fc in fcurves:
        if fc.data_path.endswith("location") and fc.array_index in (0, 1, 2):
            delta = fc.evaluate(end_frame) - fc.evaluate(start_frame)
            if abs(delta) > abs(max_delta):
                max_delta = delta
                best_axis = fc.array_index
                net_delta = delta

    direction_sign = 1.0 if net_delta >= 0 else -1.0
    return best_axis, direction_sign


def find_hand_vertex_groups(target_mesh):
    r_name, l_name = "mixamorig:RightHand", "mixamorig:LeftHand"
    if target_mesh and hasattr(target_mesh, "vertex_groups"):
        for vg in target_mesh.vertex_groups:
            vg_lower = vg.name.lower()
            if any(k in vg_lower for k in ["righthand", "hand_r", "hand.r"]):
                r_name = vg.name
            elif any(k in vg_lower for k in ["lefthand", "hand_l", "hand.l"]):
                l_name = vg.name
    return r_name, l_name


def create_gn_named_attr_node(nodes):
    try:
        return nodes.new("GeometryNodeNamedAttribute")
    except Exception:
        return nodes.new("GeometryNodeInputNamedAttribute")

# ==============================================================================
# RIG-PARENTED CAMERA RIG SETUP
# ==============================================================================

def setup_modern_reel_camera(target_rig, target_mesh, turn_start, turn_end_frame, unfurl_end_frame):
    scene = bpy.context.scene
    scene.render.resolution_x = 1080
    scene.render.resolution_y = 1920
    scene.render.resolution_percentage = 100

    focus_target = bpy.data.objects.get("Camera_Focus_Target") or bpy.data.objects.new("Camera_Focus_Target", None)
    if focus_target.name not in scene.collection.objects:
        scene.collection.objects.link(focus_target)

    cam_data = bpy.data.cameras.get("Vertical_Reel_Cam_Data") or bpy.data.cameras.new("Vertical_Reel_Cam_Data")
    cam_data.lens = 45.0
    cam_data.sensor_fit = 'VERTICAL'

    cam_obj = bpy.data.objects.get("Vertical_Reel_Camera") or bpy.data.objects.new("Vertical_Reel_Camera", cam_data)
    if cam_obj.name not in scene.collection.objects:
        scene.collection.objects.link(cam_obj)

    scene.camera = cam_obj
    cam_obj.parent = None

    cam_obj.constraints.clear()
    track = cam_obj.constraints.new(type='TRACK_TO')
    track.target = focus_target
    track.track_axis = 'TRACK_NEGATIVE_Z'
    track.up_axis = 'UP_Y'

    if cam_obj.animation_data:
        cam_obj.animation_data_clear()
    if focus_target.animation_data:
        focus_target.animation_data_clear()

    max_radius = 0.0
    sample_frames = list(range(1, unfurl_end_frame + 1, 5))
    if unfurl_end_frame not in sample_frames:
        sample_frames.append(unfurl_end_frame)

    depsgraph = bpy.context.evaluated_depsgraph_get()

    for f in sample_frames:
        scene.frame_set(f)
        depsgraph.update()

        eval_mesh_obj = target_mesh.evaluated_get(depsgraph)
        me = eval_mesh_obj.to_mesh()
        matrix = eval_mesh_obj.matrix_world

        world_verts = [matrix @ v.co for v in me.vertices]
        eval_mesh_obj.to_mesh_clear()

        if not world_verts:
            continue

        min_x = min(v.x for v in world_verts)
        max_x = max(v.x for v in world_verts)
        min_y = min(v.y for v in world_verts)
        max_y = max(v.y for v in world_verts)
        min_z = min(v.z for v in world_verts)
        max_z = max(v.z for v in world_verts)

        center = Vector(((min_x + max_x) * 0.5, (min_y + max_y) * 0.5, (min_z + max_z) * 0.5))

        for v in world_verts:
            dist = (v - center).length
            if dist > max_radius:
                max_radius = dist

    padded_radius = max_radius * 1.15

    sensor_height = cam_data.sensor_height
    focal_length = cam_data.lens

    fov_v = 2.0 * math.atan((sensor_height * 0.5) / focal_length)
    aspect_ratio = scene.render.resolution_x / scene.render.resolution_y
    fov_h = 2.0 * math.atan(math.tan(fov_v * 0.5) * aspect_ratio)

    min_fov = min(fov_v, fov_h)
    required_dist = padded_radius / math.tan(min_fov * 0.5)

    root_bone_name = "mixamorig:Hips" if "mixamorig:Hips" in target_rig.pose.bones else target_rig.pose.bones[0].name

    for f in range(1, unfurl_end_frame + 1):
        scene.frame_set(f)
        depsgraph.update()

        eval_mesh_obj = target_mesh.evaluated_get(depsgraph)
        bbox_world = [eval_mesh_obj.matrix_world @ Vector(corner) for corner in eval_mesh_obj.bound_box]
        min_z = min(v.z for v in bbox_world)
        max_z = max(v.z for v in bbox_world)
        target_mid_z = min_z + ((max_z - min_z) * 0.50)

        eval_rig = target_rig.evaluated_get(depsgraph)
        pbone = eval_rig.pose.bones[root_bone_name]
        bone_world_matrix = eval_rig.matrix_world @ pbone.matrix

        char_ground_loc = Vector((bone_world_matrix.translation.x, bone_world_matrix.translation.y, min_z))

        focus_target.location = Vector((char_ground_loc.x, char_ground_loc.y, target_mid_z))
        focus_target.keyframe_insert(data_path="location", frame=f)

        rig_matrix = eval_rig.matrix_world
        forward_vec = (rig_matrix.to_3x3() @ Vector((0, 1, 0))).xy.normalized()
        if forward_vec.length == 0:
            forward_vec = Vector((0, -1))

        base_angle = math.atan2(forward_vec.y, forward_vec.x)

        if f <= turn_start:
            angle_offset = math.radians(-45.0)
        elif f >= turn_end_frame:
            angle_offset = 0.0
        else:
            t = (f - turn_start) / (turn_end_frame - turn_start)
            angle_offset = math.radians(-45.0) * (1.0 - t)

        cam_angle = base_angle + angle_offset

        cam_x = char_ground_loc.x - (required_dist * math.cos(cam_angle))
        cam_y = char_ground_loc.y - (required_dist * math.sin(cam_angle))

        cam_obj.location = Vector((cam_x, cam_y, target_mid_z))
        cam_obj.keyframe_insert(data_path="location", frame=f)

    for obj in [cam_obj, focus_target]:
        if obj and obj.animation_data and obj.animation_data.action:
            action = obj.animation_data.action
            fcurves = getattr(action, "fcurves", None)
            if fcurves is None and hasattr(action, "bindings"):
                fcurves = [fc for b in action.bindings for fc in b.fcurves if hasattr(b, "fcurves")]
            if fcurves:
                for fc in fcurves:
                    for kp in fc.keyframe_points:
                        kp.interpolation = 'LINEAR'

    for obj in [target_rig, target_mesh]:
        if obj and obj.animation_data and obj.animation_data.nla_tracks:
            for track in obj.animation_data.nla_tracks:
                for strip in track.strips:
                    strip.extrapolation = 'HOLD'

    scene.frame_set(1)
    depsgraph.update()

    return cam_obj, focus_target


def frame_scroll_image(cam_obj, focus_target, target_mesh, image_path, scroll_width=0.35, zoom_duration=45):
    scene = bpy.context.scene
    unfurl_end_frame = scene.frame_end
    zoom_end_frame = unfurl_end_frame + zoom_duration

    scene.frame_set(unfurl_end_frame)
    bpy.context.view_layer.update()

    img_aspect = 1.0
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        if img.size[1] > 0:
            img_aspect = img.size[0] / img.size[1]

    tile_height = scroll_width * img_aspect
    cam_data = cam_obj.data
    aspect_ratio = scene.render.resolution_x / scene.render.resolution_y
    sensor_width = cam_data.sensor_height * aspect_ratio

    fov_h_half = math.atan((sensor_width / 2.0) / cam_data.lens)
    target_framing_distance = (tile_height / 2.0) / math.tan(fov_h_half)

    mesh_matrix = target_mesh.matrix_world
    depsgraph = bpy.context.evaluated_depsgraph_get()
    eval_mesh = target_mesh.evaluated_get(depsgraph)

    r_hand_name, l_hand_name = find_hand_vertex_groups(target_mesh)
    vg_r = eval_mesh.vertex_groups.get(r_hand_name)
    vg_l = eval_mesh.vertex_groups.get(l_hand_name)

    r_verts = [mesh_matrix @ v.co for v in eval_mesh.data.vertices if any(g.group == vg_r.index for g in v.groups)] if vg_r else []
    l_verts = [mesh_matrix @ v.co for v in eval_mesh.data.vertices if any(g.group == vg_l.index for g in v.groups)] if vg_l else []

    target_height = target_mesh.dimensions.z if target_mesh.dimensions.z > 0 else 1.8
    torso_fallback = mesh_matrix.translation + Vector((0, 0, target_height * 0.55))

    r_hand_pos = sum(r_verts, Vector((0, 0, 0))) / len(r_verts) if r_verts else torso_fallback
    l_hand_pos = sum(l_verts, Vector((0, 0, 0))) / len(l_verts) if l_verts else torso_fallback

    scroll_midpoint = (r_hand_pos + l_hand_pos) * 0.5

    cam_pos = cam_obj.matrix_world.translation.copy()
    cam_dir = cam_pos - scroll_midpoint
    cam_dir.z = 0.0

    scroll_normal = cam_dir.normalized() if cam_dir.length > 0.001 else Vector((0, -1, 0))

    tile_center = scroll_midpoint + (scroll_normal * 0.10)
    end_pos = tile_center + (scroll_normal * target_framing_distance)

    cam_obj.parent = None
    focus_target.parent = None

    current_world_focus = focus_target.matrix_world.translation.copy()
    focus_target.matrix_world.translation = current_world_focus

    focus_target.location = current_world_focus
    focus_target.keyframe_insert(data_path="location", frame=unfurl_end_frame)
    focus_target.location = tile_center
    focus_target.keyframe_insert(data_path="location", frame=zoom_end_frame)

    cam_obj.location = cam_pos
    cam_obj.keyframe_insert(data_path="location", frame=unfurl_end_frame)
    cam_obj.location = end_pos
    cam_obj.keyframe_insert(data_path="location", frame=zoom_end_frame)

    for obj in [cam_obj, focus_target]:
        for fc in get_object_fcurves(obj):
            for kp in fc.keyframe_points:
                if kp.co.x >= unfurl_end_frame:
                    kp.interpolation = 'BEZIER'
                    kp.handle_left_type = 'AUTO_CLAMPED'
                    kp.handle_right_type = 'AUTO_CLAMPED'

    scene.frame_end = zoom_end_frame
    return zoom_end_frame

# ==============================================================================
# GROUND & SKY DOME SETUP (PBR ENHANCED)
# ==============================================================================

def ensure_world_sphere_size(cam_obj, ground_obj, safety_multiplier=2.5):
    if not cam_obj or not ground_obj:
        return

    scene = bpy.context.scene
    max_cam_dist = 0.0

    for frame in range(scene.frame_start, scene.frame_end + 1, 5):
        scene.frame_set(frame)
        bpy.context.view_layer.update()
        max_cam_dist = max(max_cam_dist, cam_obj.matrix_world.translation.length)

    world_radius = max(25.0, max_cam_dist * safety_multiplier)

    mod = ground_obj.modifiers.get("Grass_Scatter_GN")
    if mod and mod.node_group:
        for node in mod.node_group.nodes:
            if node.bl_idname == "GeometryNodeMeshIcoSphere":
                node.inputs["Radius"].default_value = world_radius

    if ground_obj.type == 'MESH':
        bpy.context.view_layer.objects.active = ground_obj
        bpy.ops.object.mode_set(mode='OBJECT')
        ground_scale = world_radius / 12.5
        ground_obj.scale = (ground_scale, ground_scale, 1.0)
        bpy.ops.object.transform_apply(location=False, rotation=False, scale=True)

    cam_obj.data.clip_end = max(cam_obj.data.clip_end, world_radius * 2.5)
    scene.frame_set(1)


def find_grass_maps(grass_dir):
    """Scans grass directory and returns mapped PBR texture paths."""
    if not os.path.exists(grass_dir):
        return {}

    extensions = ['*.jpg', '*.jpeg', '*.png', '*.tif', '*.tiff']
    matches = []
    for ext in extensions:
        matches.extend(glob.glob(os.path.join(grass_dir, ext)))
        matches.extend(glob.glob(os.path.join(grass_dir, "**", ext), recursive=True))

    maps = {}
    for m in matches:
        fn = os.path.basename(m).lower()
        if any(k in fn for k in ["color", "diff", "albedo", "col", "basecolor"]) and "color" not in maps:
            maps["color"] = m
        elif "normalgl" in fn or ("normal" in fn and "dx" not in fn) and "normal" not in maps:
            maps["normal"] = m
        elif "roughness" in fn or "rough" in fn and "roughness" not in maps:
            maps["roughness"] = m
        elif any(k in fn for k in ["disp", "height", "displacement"]) and "disp" not in maps:
            maps["disp"] = m
        elif any(k in fn for k in ["ao", "ambientocclusion", "occlusion"]) and "ao" not in maps:
            maps["ao"] = m

    return maps


def create_ground_material(grass_dir=None):
    mat = bpy.data.materials.new(name="Ground_Soil_Material")
    mat.use_nodes = True
    nodes, links = mat.node_tree.nodes, mat.node_tree.links
    nodes.clear()

    out_node = nodes.new(type="ShaderNodeOutputMaterial")
    bsdf = nodes.new(type="ShaderNodeBsdfPrincipled")

    pbr_maps = find_grass_maps(grass_dir) if grass_dir else {}

    tex_coord = nodes.new(type="ShaderNodeTexCoord")
    mapping = nodes.new(type="ShaderNodeMapping")
    mapping.inputs["Scale"].default_value = (20.0, 20.0, 1.0)
    links.new(tex_coord.outputs["UV"], mapping.inputs["Vector"])

    # 1. Base Color & Ambient Occlusion (AO)
    if "color" in pbr_maps:
        tex_color = nodes.new(type="ShaderNodeTexImage")
        tex_color.image = bpy.data.images.load(pbr_maps["color"])
        links.new(mapping.outputs["Vector"], tex_color.inputs["Vector"])

        if "ao" in pbr_maps:
            tex_ao = nodes.new(type="ShaderNodeTexImage")
            tex_ao.image = bpy.data.images.load(pbr_maps["ao"])
            tex_ao.image.colorspace_settings.name = 'Non-Color'
            links.new(mapping.outputs["Vector"], tex_ao.inputs["Vector"])

            mix_ao = nodes.new(type="ShaderNodeMix")
            mix_ao.data_type = 'RGBA'
            mix_ao.blend_type = 'MULTIPLY'
            mix_ao.inputs["Factor"].default_value = 0.7  # Control AO shadow depth

            links.new(tex_color.outputs["Color"], mix_ao.inputs[6])
            links.new(tex_ao.outputs["Color"], mix_ao.inputs[7])
            links.new(mix_ao.outputs[2], bsdf.inputs["Base Color"])
        else:
            links.new(tex_color.outputs["Color"], bsdf.inputs["Base Color"])
    else:
        # Fallback Procedural Pattern
        noise = nodes.new("ShaderNodeTexNoise")
        noise.inputs["Scale"].default_value = 18.0
        noise.inputs["Detail"].default_value = 6.0

        color_ramp = nodes.new("ShaderNodeValToRGB")
        color_ramp.color_ramp.elements[0].color = (0.04, 0.025, 0.01, 1.0)
        color_ramp.color_ramp.elements[1].color = (0.05, 0.07, 0.02, 1.0)

        links.new(noise.outputs["Factor"], color_ramp.inputs["Factor"])
        links.new(color_ramp.outputs["Color"], bsdf.inputs["Base Color"])

    # 2. Roughness Map
    if "roughness" in pbr_maps:
        tex_rough = nodes.new(type="ShaderNodeTexImage")
        tex_rough.image = bpy.data.images.load(pbr_maps["roughness"])
        tex_rough.image.colorspace_settings.name = 'Non-Color'
        links.new(mapping.outputs["Vector"], tex_rough.inputs["Vector"])
        links.new(tex_rough.outputs["Color"], bsdf.inputs["Roughness"])
    else:
        bsdf.inputs["Roughness"].default_value = 0.95

    # 3. Normal Map
    if "normal" in pbr_maps:
        tex_norm = nodes.new(type="ShaderNodeTexImage")
        tex_norm.image = bpy.data.images.load(pbr_maps["normal"])
        tex_norm.image.colorspace_settings.name = 'Non-Color'
        links.new(mapping.outputs["Vector"], tex_norm.inputs["Vector"])

        norm_map = nodes.new(type="ShaderNodeNormalMap")
        norm_map.inputs["Strength"].default_value = 1.8  # Enhance 3D blade relief
        links.new(tex_norm.outputs["Color"], norm_map.inputs["Color"])
        links.new(norm_map.outputs["Normal"], bsdf.inputs["Normal"])

    # 4. Bump / Displacement Map Fallback
    if "disp" in pbr_maps and "normal" not in pbr_maps:
        tex_disp = nodes.new(type="ShaderNodeTexImage")
        tex_disp.image = bpy.data.images.load(pbr_maps["disp"])
        tex_disp.image.colorspace_settings.name = 'Non-Color'
        links.new(mapping.outputs["Vector"], tex_disp.inputs["Vector"])

        bump_node = nodes.new(type="ShaderNodeBump")
        bump_node.inputs["Strength"].default_value = 0.5
        links.new(tex_disp.outputs["Color"], bump_node.inputs["Height"])
        links.new(bump_node.outputs["Normal"], bsdf.inputs["Normal"])

    links.new(bsdf.outputs["BSDF"], out_node.inputs["Surface"])
    return mat


def create_sky_material(sky_hdri_path):
    mat = bpy.data.materials.new(name="Sky_Dome_Material")
    mat.use_nodes = True
    nodes, links = mat.node_tree.nodes, mat.node_tree.links
    nodes.clear()

    out_node = nodes.new("ShaderNodeOutputMaterial")
    emission = nodes.new("ShaderNodeEmission")
    emission.inputs["Strength"].default_value = 1.0

    transparent = nodes.new("ShaderNodeBsdfTransparent")
    mix_shader = nodes.new("ShaderNodeMixShader")
    light_path = nodes.new("ShaderNodeLightPath")

    links.new(light_path.outputs["Is Camera Ray"], mix_shader.inputs["Fac"])
    links.new(transparent.outputs["BSDF"], mix_shader.inputs[1])
    links.new(emission.outputs["Emission"], mix_shader.inputs[2])
    links.new(mix_shader.outputs["Shader"], out_node.inputs["Surface"])

    if os.path.exists(sky_hdri_path):
        tex_env = nodes.new("ShaderNodeTexEnvironment")
        tex_env.image = bpy.data.images.load(sky_hdri_path)
        links.new(tex_env.outputs["Color"], emission.inputs["Color"])

    return mat


def setup_grass_ground_geometry_nodes(grass_dir, sky_hdri_path, ground_z_offset=0.0):
    bpy.ops.mesh.primitive_plane_add(size=60, location=(0, 0, ground_z_offset))
    ground_obj = bpy.context.active_object
    ground_obj.name = "Grass_Ground_Plane"

    ground_mat = create_ground_material(grass_dir=grass_dir)
    sky_mat = create_sky_material(sky_hdri_path)
    ground_obj.data.materials.append(ground_mat)
    ground_obj.data.materials.append(sky_mat)

    gn_mod = ground_obj.modifiers.new(name="Grass_Scatter_GN", type='NODES')
    gn_group = bpy.data.node_groups.new("Grass_Scatter_Group", type='GeometryNodeTree')
    gn_mod.node_group = gn_group

    nodes = gn_group.nodes
    links = gn_group.links
    nodes.clear()

    group_in = nodes.new("NodeGroupInput")
    gn_group.interface.new_socket("Geometry", in_out='INPUT', socket_type='NodeSocketGeometry')
    group_out = nodes.new("NodeGroupOutput")
    gn_group.interface.new_socket("Geometry", in_out='OUTPUT', socket_type='NodeSocketGeometry')

    sky_sphere = nodes.new("GeometryNodeMeshIcoSphere")
    sky_sphere.inputs["Radius"].default_value = 12.5
    sky_sphere.inputs["Subdivisions"].default_value = 4

    flip_faces = nodes.new("GeometryNodeFlipFaces")
    links.new(sky_sphere.outputs["Mesh"], flip_faces.inputs["Mesh"])

    transform_sky = nodes.new("GeometryNodeTransform")
    transform_sky.inputs["Translation"].default_value = (0.0, 0.0, ground_z_offset)
    links.new(flip_faces.outputs["Mesh"], transform_sky.inputs["Geometry"])

    set_sky_mat = nodes.new("GeometryNodeSetMaterial")
    set_sky_mat.inputs["Material"].default_value = sky_mat
    links.new(transform_sky.outputs["Geometry"], set_sky_mat.inputs["Geometry"])

    set_ground_mat = nodes.new("GeometryNodeSetMaterial")
    set_ground_mat.inputs["Material"].default_value = ground_mat
    links.new(group_in.outputs["Geometry"], set_ground_mat.inputs["Geometry"])



    # 1. Subdivide the ground mesh so it has enough vertices to deform
    subdivide = nodes.new("GeometryNodeSubdivideMesh")
    subdivide.inputs["Level"].default_value = 5  # Increase for higher vertex density
    links.new(group_in.outputs["Geometry"], subdivide.inputs["Mesh"])

    # 2. Generate Noise for terrain height variation
    noise_tex = nodes.new("ShaderNodeTexNoise")
    noise_tex.inputs["Scale"].default_value = 0.05  # Lower scale = larger rolling hills
    noise_tex.inputs["Detail"].default_value = 2.0

    # 3. Scale the displacement height
    vector_math = nodes.new("ShaderNodeVectorMath")
    vector_math.operation = "SCALE"
    vector_math.inputs["Scale"].default_value = 0.8  # Max elevation displacement in Blender units
    links.new(noise_tex.outputs["Color"], vector_math.inputs["Vector"])

    # 4. Apply displacement to the vertex positions
    set_pos = nodes.new("GeometryNodeSetPosition")
    links.new(subdivide.outputs["Mesh"], set_pos.inputs["Geometry"])
    links.new(vector_math.outputs["Vector"], set_pos.inputs["Offset"])

    # 5. Assign material to displaced ground
    set_ground_mat = nodes.new("GeometryNodeSetMaterial")
    set_ground_mat.inputs["Material"].default_value = ground_mat
    links.new(set_pos.outputs["Geometry"], set_ground_mat.inputs["Geometry"])



    join_geo = nodes.new("GeometryNodeJoinGeometry")
    links.new(set_ground_mat.outputs["Geometry"], join_geo.inputs["Geometry"])
    links.new(set_sky_mat.outputs["Geometry"], join_geo.inputs["Geometry"])

    links.new(join_geo.outputs["Geometry"], group_out.inputs["Geometry"])

    return ground_obj

# ==============================================================================
# SCROLL MATERIAL & GEOMETRY NODES SETUP
# ==============================================================================

def create_image_scroll_material(image_path):
    mat_name = "Scroll_Texture_Mat"
    if mat := bpy.data.materials.get(mat_name):
        bpy.data.materials.remove(mat)

    mat = bpy.data.materials.new(name=mat_name)
    mat.use_nodes = True
    nodes, links = mat.node_tree.nodes, mat.node_tree.links
    nodes.clear()

    output_node = nodes.new("ShaderNodeOutputMaterial")
    bsdf = nodes.new("ShaderNodeBsdfPrincipled")
    bsdf.inputs["Roughness"].default_value = 0.5

    tex_img = nodes.new("ShaderNodeTexImage")
    if os.path.exists(image_path):
        tex_img.image = bpy.data.images.load(image_path)
        tex_img.extension = 'REPEAT'

    attr_node = nodes.new("ShaderNodeAttribute")
    attr_node.attribute_name = "UVMap"

    links.new(attr_node.outputs["Vector"], tex_img.inputs["Vector"])
    links.new(tex_img.outputs["Color"], bsdf.inputs["Base Color"])
    links.new(bsdf.outputs["BSDF"], output_node.inputs["Surface"])

    return mat, tex_img.image if os.path.exists(image_path) else None


def create_pure_geometry_node_scroll(image_path=""):
    gn_group = bpy.data.node_groups.new("Scroll_Pure_GN", type="GeometryNodeTree")

    gn_group.interface.new_socket(name="Geometry", in_out="OUTPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Character Mesh", in_out="INPUT", socket_type="NodeSocketObject")

    sock_r_vg = gn_group.interface.new_socket(name="Right Hand Group", in_out="INPUT", socket_type="NodeSocketString")
    sock_r_vg.default_value = "mixamorig:RightHand"

    sock_l_vg = gn_group.interface.new_socket(name="Left Hand Group", in_out="INPUT", socket_type="NodeSocketString")
    sock_l_vg.default_value = "mixamorig:LeftHand"

    sock_offset = gn_group.interface.new_socket(name="Forward Offset", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_offset.default_value = 0.60

    sock_unfurl = gn_group.interface.new_socket(name="Unfurl", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_unfurl.default_value = 0.0

    sock_width = gn_group.interface.new_socket(name="Scroll Width", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_width.default_value = 0.35

    nodes, links = gn_group.nodes, gn_group.links
    in_node, out_node = nodes.new("NodeGroupInput"), nodes.new("NodeGroupOutput")

    char_info = nodes.new("GeometryNodeObjectInfo")
    char_info.transform_space = "RELATIVE"
    links.new(in_node.outputs["Character Mesh"], char_info.inputs["Object"])

    pos_node = nodes.new("GeometryNodeInputPosition")

    named_r = create_gn_named_attr_node(nodes)
    named_r.data_type = "FLOAT"
    named_r.inputs["Name"].default_value = "mixamorig:RightHand"
    links.new(in_node.outputs["Right Hand Group"], named_r.inputs["Name"])

    attr_r_out = named_r.outputs.get("Attribute") or named_r.outputs[0]

    cmp_r = nodes.new("FunctionNodeCompare")
    cmp_r.data_type = "FLOAT"
    cmp_r.operation = "GREATER_THAN"
    cmp_r.inputs[1].default_value = 0.001
    links.new(attr_r_out, cmp_r.inputs[0])

    stat_r = nodes.new("GeometryNodeAttributeStatistic")
    stat_r.data_type = "FLOAT_VECTOR"
    links.new(char_info.outputs["Geometry"], stat_r.inputs["Geometry"])
    links.new(cmp_r.outputs["Result"], stat_r.inputs["Selection"])
    links.new(pos_node.outputs["Position"], stat_r.inputs["Attribute"])

    named_l = create_gn_named_attr_node(nodes)
    named_l.data_type = "FLOAT"
    named_l.inputs["Name"].default_value = "mixamorig:LeftHand"
    links.new(in_node.outputs["Left Hand Group"], named_l.inputs["Name"])

    attr_l_out = named_l.outputs.get("Attribute") or named_l.outputs[0]

    cmp_l = nodes.new("FunctionNodeCompare")
    cmp_l.data_type = "FLOAT"
    cmp_l.operation = "GREATER_THAN"
    cmp_l.inputs[1].default_value = 0.001
    links.new(attr_l_out, cmp_l.inputs[0])

    stat_l = nodes.new("GeometryNodeAttributeStatistic")
    stat_l.data_type = "FLOAT_VECTOR"
    links.new(char_info.outputs["Geometry"], stat_l.inputs["Geometry"])
    links.new(cmp_l.outputs["Result"], stat_l.inputs["Selection"])
    links.new(pos_node.outputs["Position"], stat_l.inputs["Attribute"])

    neg_offset = nodes.new("ShaderNodeMath")
    neg_offset.operation = "MULTIPLY"
    neg_offset.inputs[1].default_value = -1.0
    links.new(in_node.outputs["Forward Offset"], neg_offset.inputs[0])

    forward_vec = nodes.new("ShaderNodeCombineXYZ")
    forward_vec.inputs["X"].default_value = 0.0
    links.new(neg_offset.outputs["Value"], forward_vec.inputs["Y"])
    forward_vec.inputs["Z"].default_value = 0.0

    start_pos = nodes.new("ShaderNodeVectorMath")
    start_pos.operation = "ADD"
    links.new(stat_r.outputs["Mean"], start_pos.inputs[0])
    links.new(forward_vec.outputs["Vector"], start_pos.inputs[1])

    target_pos = nodes.new("ShaderNodeVectorMath")
    target_pos.operation = "ADD"
    links.new(stat_l.outputs["Mean"], target_pos.inputs[0])
    links.new(forward_vec.outputs["Vector"], target_pos.inputs[1])

    delta_vec = nodes.new("ShaderNodeVectorMath")
    delta_vec.operation = "SUBTRACT"
    links.new(target_pos.outputs["Vector"], delta_vec.inputs[0])
    links.new(start_pos.outputs["Vector"], delta_vec.inputs[1])

    dist_node = nodes.new("ShaderNodeVectorMath")
    dist_node.operation = "DISTANCE"
    links.new(start_pos.outputs["Vector"], dist_node.inputs[0])

    scale_vec = nodes.new("ShaderNodeVectorMath")
    scale_vec.operation = "SCALE"
    links.new(delta_vec.outputs["Vector"], scale_vec.inputs["Vector"])
    links.new(in_node.outputs["Unfurl"], scale_vec.inputs["Scale"])

    end_pos = nodes.new("ShaderNodeVectorMath")
    end_pos.operation = "ADD"
    links.new(start_pos.outputs["Vector"], end_pos.inputs[0])
    links.new(scale_vec.outputs["Vector"], end_pos.inputs[1])
    links.new(end_pos.outputs["Vector"], dist_node.inputs[1])

    curve_line = nodes.new("GeometryNodeCurvePrimitiveLine")
    links.new(start_pos.outputs["Vector"], curve_line.inputs["Start"])
    links.new(end_pos.outputs["Vector"], curve_line.inputs["End"])

    set_tilt = nodes.new("GeometryNodeSetCurveTilt")
    set_tilt.inputs["Tilt"].default_value = 1.5708
    links.new(curve_line.outputs["Curve"], set_tilt.inputs["Curve"])

    spline_param_main = nodes.new("GeometryNodeSplineParameter")
    store_u = nodes.new("GeometryNodeStoreNamedAttribute")
    store_u.inputs["Name"].default_value = "U_Factor"
    store_u.data_type = 'FLOAT'
    store_u.domain = 'POINT'
    links.new(set_tilt.outputs["Curve"], store_u.inputs["Geometry"])

    spline_factor_out = spline_param_main.outputs.get("Factor") or spline_param_main.outputs[0]
    links.new(spline_factor_out, store_u.inputs["Value"])

    half_width_neg = nodes.new("ShaderNodeMath")
    half_width_neg.operation = "MULTIPLY"
    half_width_neg.inputs[1].default_value = -0.5
    links.new(in_node.outputs["Scroll Width"], half_width_neg.inputs[0])

    half_width_pos = nodes.new("ShaderNodeMath")
    half_width_pos.operation = "MULTIPLY"
    half_width_pos.inputs[1].default_value = 0.5
    links.new(in_node.outputs["Scroll Width"], half_width_pos.inputs[0])

    combine_start = nodes.new("ShaderNodeCombineXYZ")
    links.new(half_width_neg.outputs["Value"], combine_start.inputs["X"])

    combine_end = nodes.new("ShaderNodeCombineXYZ")
    links.new(half_width_pos.outputs["Value"], combine_end.inputs["X"])

    profile_line = nodes.new("GeometryNodeCurvePrimitiveLine")
    links.new(combine_start.outputs["Vector"], profile_line.inputs["Start"])
    links.new(combine_end.outputs["Vector"], profile_line.inputs["End"])

    spline_param_prof = nodes.new("GeometryNodeSplineParameter")
    store_v = nodes.new("GeometryNodeStoreNamedAttribute")
    store_v.inputs["Name"].default_value = "V_Factor"
    store_v.data_type = 'FLOAT'
    store_v.domain = 'POINT'
    links.new(profile_line.outputs["Curve"], store_v.inputs["Geometry"])

    prof_factor_out = spline_param_prof.outputs.get("Factor") or spline_param_prof.outputs[0]
    links.new(prof_factor_out, store_v.inputs["Value"])

    curve_to_mesh = nodes.new("GeometryNodeCurveToMesh")
    links.new(store_u.outputs["Geometry"], curve_to_mesh.inputs["Curve"])
    links.new(store_v.outputs["Geometry"], curve_to_mesh.inputs["Profile Curve"])

    img_aspect = 1.0
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        if img.size[1] > 0:
            img_aspect = img.size[0] / img.size[1]

    h_times_aspect = nodes.new("ShaderNodeMath")
    h_times_aspect.operation = "MULTIPLY"
    h_times_aspect.inputs[1].default_value = img_aspect
    links.new(in_node.outputs["Scroll Width"], h_times_aspect.inputs[0])

    tile_count = nodes.new("ShaderNodeMath")
    tile_count.operation = "DIVIDE"
    links.new(dist_node.outputs["Value"], tile_count.inputs[0])
    links.new(h_times_aspect.outputs["Value"], tile_count.inputs[1])

    read_u = create_gn_named_attr_node(nodes)
    read_u.data_type = 'FLOAT'
    read_u.inputs["Name"].default_value = "U_Factor"
    read_u_out = read_u.outputs.get("Attribute") or read_u.outputs[0]

    scaled_u = nodes.new("ShaderNodeMath")
    scaled_u.operation = "MULTIPLY"
    links.new(read_u_out, scaled_u.inputs[0])
    links.new(tile_count.outputs["Value"], scaled_u.inputs[1])

    read_v = create_gn_named_attr_node(nodes)
    read_v.data_type = 'FLOAT'
    read_v.inputs["Name"].default_value = "V_Factor"
    read_v_out = read_v.outputs.get("Attribute") or read_v.outputs[0]

    flip_v = nodes.new("ShaderNodeMath")
    flip_v.operation = "SUBTRACT"
    flip_v.inputs[0].default_value = 1.0
    links.new(read_v_out, flip_v.inputs[1])

    combine_uv = nodes.new("ShaderNodeCombineXYZ")
    links.new(scaled_u.outputs["Value"], combine_uv.inputs["X"])
    links.new(flip_v.outputs["Value"], combine_uv.inputs["Y"])

    store_uv = nodes.new("GeometryNodeStoreNamedAttribute")
    store_uv.inputs["Name"].default_value = "UVMap"
    store_uv.data_type = 'FLOAT2'
    store_uv.domain = 'CORNER'
    links.new(curve_to_mesh.outputs["Mesh"], store_uv.inputs["Geometry"])
    links.new(combine_uv.outputs["Vector"], store_uv.inputs["Value"])

    set_material = nodes.new("GeometryNodeSetMaterial")
    scroll_mat, _ = create_image_scroll_material(image_path)
    set_material.inputs["Material"].default_value = scroll_mat
    links.new(store_uv.outputs["Geometry"], set_material.inputs["Geometry"])

    links.new(set_material.outputs["Geometry"], out_node.inputs["Geometry"])

    return gn_group, scroll_mat


def setup_geometry_node_scroll(target_mesh, unfurl_start, unfurl_end, image_path=""):
    mesh_data = bpy.data.meshes.new("Scroll_Procedural_Mesh")
    scroll_obj = bpy.data.objects.new("Scroll_GN_Object", mesh_data)
    bpy.context.scene.collection.objects.link(scroll_obj)

    mod = scroll_obj.modifiers.new("Scroll_GN", type="NODES")
    gn_tree, scroll_mat = create_pure_geometry_node_scroll(image_path=image_path)
    mod.node_group = gn_tree

    scroll_obj.data.materials.append(scroll_mat)

    r_hand_name, l_hand_name = find_hand_vertex_groups(target_mesh)

    unfurl_id = None
    for item in gn_tree.interface.items_tree:
        if getattr(item, "in_out", None) == "INPUT":
            if item.name == "Character Mesh":
                mod[item.identifier] = target_mesh
            elif item.name == "Right Hand Group":
                mod[item.identifier] = r_hand_name
            elif item.name == "Left Hand Group":
                mod[item.identifier] = l_hand_name
            elif item.name == "Forward Offset":
                mod[item.identifier] = 0.60
            elif item.name == "Scroll Width":
                mod[item.identifier] = 0.35
            elif item.name == "Unfurl":
                unfurl_id = item.identifier

    scroll_obj.hide_viewport = True
    scroll_obj.hide_render = True
    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=1)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=1)

    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=unfurl_start - 1)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=unfurl_start - 1)

    scroll_obj.hide_viewport = False
    scroll_obj.hide_render = False
    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=unfurl_start)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=unfurl_start)

    for fc in get_object_fcurves(scroll_obj):
        if fc.data_path in ["hide_viewport", "hide_render"]:
            for kp in fc.keyframe_points:
                kp.interpolation = 'CONSTANT'

    if unfurl_id:
        mod[unfurl_id] = 0.001
        mod.keyframe_insert(data_path=f'["{unfurl_id}"]', frame=unfurl_start)

        mod[unfurl_id] = 1.0
        mod.keyframe_insert(data_path=f'["{unfurl_id}"]', frame=unfurl_end)

    return scroll_obj

# ==============================================================================
# MAIN CONTROLLER
# ==============================================================================

def import_and_extract_action(fbx_path, action_name):
    existing_objects = set(bpy.data.objects)
    bpy.ops.import_scene.fbx(filepath=fbx_path)
    bpy.context.view_layer.update()

    imported_objects = set(bpy.data.objects) - existing_objects
    extracted_action = None
    for obj in imported_objects:
        if obj.animation_data and obj.animation_data.action:
            extracted_action = obj.animation_data.action
            extracted_action.name = action_name
            extracted_action.use_fake_user = True
            break
    return extracted_action, imported_objects


def purge_imported_objects(objects_to_remove):
    for obj in objects_to_remove:
        if obj.name in bpy.data.objects:
            if obj.animation_data:
                obj.animation_data_clear()
            for coll in list(obj.users_collection):
                coll.objects.unlink(obj)
    bpy.context.view_layer.update()


def setup_world_hdri_lighting(sky_hdri_path, strength=2.5):
    scene = bpy.context.scene
    if not scene.world:
        scene.world = bpy.data.worlds.new("World")

    scene.world.use_nodes = True
    nodes = scene.world.node_tree.nodes
    links = scene.world.node_tree.links
    nodes.clear()

    out_node = nodes.new("ShaderNodeOutputWorld")
    bg_node = nodes.new("ShaderNodeBackground")
    bg_node.inputs["Strength"].default_value = strength

    links.new(bg_node.outputs["Background"], out_node.inputs["Surface"])

    if os.path.exists(sky_hdri_path):
        tex_env = nodes.new("ShaderNodeTexEnvironment")
        tex_env.image = bpy.data.images.load(sky_hdri_path)
        links.new(tex_env.outputs["Color"], bg_node.inputs["Color"])


def main():
    cleanup_previous_run()

    walk_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Mutant Walking.fbx"
    turn_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Right Turn 90.fbx"
    unfurl_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Mutant Idle.fbx"

    sky_hdri_path = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/horn-koppe_spring_4k.exr"
    output_blend = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/output_scene.blend"

    image_path = "/Users/dobbinp1/Downloads/sea_PanoramaIMGP5963.jpg"
    grass_dir = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Grass004_4K-JPG"

    scene = bpy.context.scene
    scene.view_settings.exposure = 1.0
    scene.view_settings.view_transform = 'AgX'
    scene.view_settings.look = 'AgX - Medium High Contrast'

    setup_world_hdri_lighting(sky_hdri_path, strength=1.0)

    ground_obj = setup_grass_ground_geometry_nodes(grass_dir, sky_hdri_path, ground_z_offset=-0.35)

    CYCLES = 4
    CROSSFADE = 15
    TARGET_DISTANCE = 4.0

    walk_action, walk_objs = import_and_extract_action(walk_fbx, "walk")
    target_rig = next((obj for obj in walk_objs if obj.type == "ARMATURE"), None)
    target_mesh = next((obj for obj in walk_objs if obj.type == "MESH"), None)

    if not target_rig or not target_mesh:
        raise RuntimeError("Armature or Character Mesh not found in walk FBX.")

    target_rig.location.z = 0.0

    travel_axis_idx, motion_sign = detect_natural_forward_axis(walk_action, target_rig)
    initial_heading = math.radians(90.0) if travel_axis_idx == 0 else 0.0
    target_rig.rotation_euler[2] = initial_heading

    action_len = max(30, int(walk_action.frame_range[1] - walk_action.frame_range[0]))
    anim_data = target_rig.animation_data_create()

    walk_track = anim_data.nla_tracks.new()
    walk_track.name = "Walk_Track"

    current_frame = 1
    for i in range(CYCLES):
        strip = walk_track.strips.new(name=f"Walk_{i+1}", start=int(current_frame), action=walk_action)
        strip.blend_type = "REPLACE"
        current_frame += action_len

    walk_end_frame = int(current_frame)

    turn_action, turn_objs = import_and_extract_action(turn_fbx, "rightturn")
    purge_imported_objects(turn_objs)

    turn_track = anim_data.nla_tracks.new()
    turn_track.name = "RightTurn_Track"

    turn_start = int(walk_end_frame - CROSSFADE)
    turn_strip = turn_track.strips.new(name="RightTurn", start=turn_start, action=turn_action)
    turn_strip.blend_type = "REPLACE"
    turn_strip.blend_in = CROSSFADE
    turn_strip.use_auto_blend = True
    turn_strip.extrapolation = "NOTHING"
    turn_end_frame = int(turn_strip.frame_end)

    bpy.context.scene["turn_start_frame"] = turn_start

    unfurl_action, unfurl_objs = import_and_extract_action(unfurl_fbx, "unfurl")
    purge_imported_objects(unfurl_objs)

    unfurl_track = anim_data.nla_tracks.new()
    unfurl_track.name = "Unfurl_Track"

    unfurl_start = int(turn_end_frame - CROSSFADE)
    unfurl_strip = unfurl_track.strips.new(name="Unfurl", start=unfurl_start, action=unfurl_action)
    unfurl_strip.blend_type = "REPLACE"
    unfurl_strip.blend_in = CROSSFADE
    unfurl_strip.use_auto_blend = True
    unfurl_strip.extrapolation = "HOLD"
    unfurl_end_frame = int(unfurl_strip.frame_end)

    root_action = bpy.data.actions.new(name="World_Root_Motion")
    target_rig.animation_data.action = root_action

    if hasattr(root_action, "fcurves") and hasattr(root_action.fcurves, "new"):
        loc_fcurve = root_action.fcurves.new(data_path="location", index=travel_axis_idx)
        rot_fcurve = root_action.fcurves.new(data_path="rotation_euler", index=2)
        z_fcurve = root_action.fcurves.new(data_path="location", index=2)
    else:
        loc_fcurve = root_action.fcurve_ensure_for_datablock(datablock=target_rig, data_path="location", index=travel_axis_idx)
        rot_fcurve = root_action.fcurve_ensure_for_datablock(datablock=target_rig, data_path="rotation_euler", index=2)
        z_fcurve = root_action.fcurve_ensure_for_datablock(datablock=target_rig, data_path="location", index=2)

    z_fcurve.keyframe_points.insert(frame=1.0, value=0.0)
    z_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=0.0)

    final_target_value = TARGET_DISTANCE * motion_sign

    k_loc1 = loc_fcurve.keyframe_points.insert(frame=1.0, value=0.0)
    k_loc2 = loc_fcurve.keyframe_points.insert(frame=float(turn_start), value=final_target_value)
    k_loc3 = loc_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=final_target_value)

    k_loc1.interpolation = "LINEAR"
    k_loc2.interpolation = "CONSTANT"
    k_loc3.interpolation = "CONSTANT"

    target_heading = initial_heading - math.radians(90.0)

    k_rot1 = rot_fcurve.keyframe_points.insert(frame=1.0, value=initial_heading)
    k_rot2 = rot_fcurve.keyframe_points.insert(frame=float(turn_start), value=initial_heading)
    k_rot3 = rot_fcurve.keyframe_points.insert(frame=float(turn_end_frame), value=target_heading)
    k_rot4 = rot_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=target_heading)

    k_rot1.interpolation = "CONSTANT"
    k_rot2.interpolation = "BEZIER"
    k_rot3.interpolation = "CONSTANT"
    k_rot4.interpolation = "CONSTANT"

    root_track = anim_data.nla_tracks.new()
    root_track.name = "World_Root_Track"
    root_strip = root_track.strips.new(name="Root_Motion_Strip", start=1, action=root_action)
    root_strip.blend_type = "REPLACE"
    root_strip.extrapolation = "HOLD"

    anim_data.action = None

    setup_geometry_node_scroll(
        target_mesh=target_mesh,
        unfurl_start=unfurl_start,
        unfurl_end=unfurl_end_frame,
        image_path=image_path,
    )

    cam_obj, focus_target = setup_modern_reel_camera(
        target_rig=target_rig,
        target_mesh=target_mesh,
        turn_start=turn_start,
        turn_end_frame=turn_end_frame,
        unfurl_end_frame=unfurl_end_frame
    )

    bpy.context.scene.frame_start = 1
    bpy.context.scene.frame_end = unfurl_end_frame

    frame_scroll_image(
        cam_obj=cam_obj,
        focus_target=focus_target,
        target_mesh=target_mesh,
        image_path=image_path,
        scroll_width=0.35,
        zoom_duration=45
    )

    ensure_world_sphere_size(cam_obj, ground_obj, safety_multiplier=2.5)
    register_audio_handler()

    bpy.ops.wm.save_as_mainfile(filepath=output_blend)
    print(f"\n[SUCCESS] Saved scene to: {output_blend}")


if __name__ == "__main__":
    main()
```

```
import math
import mathutils
import os
import sys
import glob
import bpy
import aud
from bpy_extras import anim_utils
from bpy_extras.object_utils import world_to_camera_view
from bpy.app.handlers import persistent
from mathutils import Vector

# ==============================================================================
# CONFIGURATION
# ==============================================================================
FOOTSTEP_SOUND_PATH = "/Users/dobbinp1/Downloads/stomp.wav"
GROUND_THRESHOLD = 0.3 # Adjusted threshold for grounded character

# Setup Audio Device
if "_audio_store" not in sys.modules:
    import types
    sys.modules["_audio_store"] = types.ModuleType("_audio_store")

audio_store = sys.modules["_audio_store"]
if not hasattr(audio_store, "device"):
    audio_store.device = aud.Device()

AUDIO_DEVICE = audio_store.device
SOUND_FACTORY = None
AUDIO_HANDLES = []

if os.path.exists(FOOTSTEP_SOUND_PATH):
    SOUND_FACTORY = aud.Sound(FOOTSTEP_SOUND_PATH).cache()
    print("[AUDIO] Loaded sound file successfully.")
else:
    print(f"[AUDIO ERROR] File not found at: {FOOTSTEP_SOUND_PATH}")

_left_foot_was_down = False
_right_foot_was_down = False
_last_frame = -1


def play_footstep():
    global AUDIO_HANDLES
    if SOUND_FACTORY:
        handle = AUDIO_DEVICE.play(SOUND_FACTORY)
        AUDIO_HANDLES.append(handle)
        if len(AUDIO_HANDLES) > 10:
            AUDIO_HANDLES.pop(0)


def find_foot_bone(pose_bones, side="Left"):
    target_keywords = [
        f"mixamorig:{side}foot",
        f"{side.lower()}foot",
        f"foot.{side[0].lower()}",
        f"foot_{side[0].lower()}"
    ]
    for bone in pose_bones:
        b_name = bone.name.lower()
        if any(kw in b_name for kw in target_keywords):
            return bone
    return None


@persistent
def footstep_frame_handler(scene):
    global _left_foot_was_down, _right_foot_was_down, _last_frame

    try:
        current_frame = scene.frame_current
        turn_start = scene.get("turn_start_frame", None)

        if turn_start is not None and current_frame >= turn_start:
            _left_foot_was_down = False
            _right_foot_was_down = False
            return

        if current_frame <= _last_frame or abs(current_frame - _last_frame) > 10:
            _left_foot_was_down = False
            _right_foot_was_down = False
        _last_frame = current_frame

        rig = next((o for o in scene.objects if o.type == 'ARMATURE'), None)
        if not rig:
            return

        depsgraph = bpy.context.evaluated_depsgraph_get()
        eval_rig = rig.evaluated_get(depsgraph)

        left_bone = find_foot_bone(eval_rig.pose.bones, "Left")
        right_bone = find_foot_bone(eval_rig.pose.bones, "Right")

        if not left_bone or not right_bone:
            return

        l_z = (eval_rig.matrix_world @ left_bone.matrix).translation.z
        r_z = (eval_rig.matrix_world @ right_bone.matrix).translation.z

        l_is_down = l_z <= GROUND_THRESHOLD
        r_is_down = r_z <= GROUND_THRESHOLD

        if l_is_down and not _left_foot_was_down:
            play_footstep()

        if r_is_down and not _right_foot_was_down:
            play_footstep()

        _left_foot_was_down = l_is_down
        _right_foot_was_down = r_is_down

    except Exception:
        import traceback
        traceback.print_exc()


def register_audio_handler():
    for h in list(bpy.app.handlers.frame_change_post):
        if getattr(h, "__name__", "") == "footstep_frame_handler":
            bpy.app.handlers.frame_change_post.remove(h)

    bpy.app.handlers.frame_change_post.append(footstep_frame_handler)


register_audio_handler()

# ==============================================================================
# CLEANUP & HELPERS
# ==============================================================================

def cleanup_previous_run():
    target_names = [
        "Master_NLA_Control",
        "Scroll_GN_Object",
        "Scroll_Start_Pin",
        "Scroll_End_Pin",
        "Scroll_Master_Container",
        "HDRI_Ground_Plane",
        "Ground_Mesh_Object",
        "Grass_Ground_Plane",
        "Grass_Blade_Instance",
        "Camera_Focus_Target",
        "Camera_Anchor",
        "Vertical_Reel_Camera"
    ]
    for obj in list(bpy.data.objects):
        if obj.name in target_names or (obj.name.startswith("mixamorig") and obj.type not in {'ARMATURE', 'MESH'}):
            if obj.animation_data:
                obj.animation_data_clear()
            bpy.data.objects.remove(obj, do_unlink=True)

    for action in list(bpy.data.actions):
        if action.users == 0 or action.use_fake_user:
            action.use_fake_user = False
            bpy.data.actions.remove(action)

    for ng in list(bpy.data.node_groups):
        if ng.name in ["Scroll_Unfurl_GN", "Scroll_Pure_GN", "HDRI_Ground_GN", "Grass_Ground_GN", "Grass_Scatter_Group"]:
            bpy.data.node_groups.remove(ng)

    for coll_name in ["Scroll_Master_Container", "Grass_Artifacts_Collection"]:
        if coll_name in bpy.data.collections:
            coll = bpy.data.collections[coll_name]
            for obj in list(coll.objects):
                bpy.data.objects.remove(obj, do_unlink=True)
            bpy.data.collections.remove(coll)

    bpy.context.view_layer.update()


def get_slot_fcurves(action, anim_target_object):
    fcurves = []
    if not action:
        return fcurves

    anim_data = getattr(anim_target_object, "animation_data", None)
    if anim_data and getattr(anim_data, "action_slot", None):
        slot = anim_data.action_slot
        channelbag = anim_utils.action_get_channelbag_for_slot(action, slot)
        if channelbag and hasattr(channelbag, "fcurves"):
            return list(channelbag.fcurves)

    if hasattr(action, "slots"):
        for slot in action.slots:
            for cb in getattr(slot, "channelbags", []):
                fcurves.extend(getattr(cb, "fcurves", []))
        if fcurves:
            return fcurves

    if hasattr(action, "fcurves"):
        return list(action.fcurves)

    return fcurves


def get_object_fcurves(obj):
    if not obj or not obj.animation_data or not obj.animation_data.action:
        return []
    return get_slot_fcurves(obj.animation_data.action, obj)


def detect_natural_forward_axis(action, rig_object):
    fcurves = get_slot_fcurves(action, rig_object)
    max_delta = 0.0
    best_axis = 1
    net_delta = 0.0

    start_frame, end_frame = action.frame_range

    for fc in fcurves:
        if fc.data_path.endswith("location") and fc.array_index in (0, 1, 2):
            delta = fc.evaluate(end_frame) - fc.evaluate(start_frame)
            if abs(delta) > abs(max_delta):
                max_delta = delta
                best_axis = fc.array_index
                net_delta = delta

    direction_sign = 1.0 if net_delta >= 0 else -1.0
    return best_axis, direction_sign


def find_hand_vertex_groups(target_mesh):
    r_name, l_name = "mixamorig:RightHand", "mixamorig:LeftHand"
    if target_mesh and hasattr(target_mesh, "vertex_groups"):
        for vg in target_mesh.vertex_groups:
            vg_lower = vg.name.lower()
            if any(k in vg_lower for k in ["righthand", "hand_r", "hand.r"]):
                r_name = vg.name
            elif any(k in vg_lower for k in ["lefthand", "hand_l", "hand.l"]):
                l_name = vg.name
    return r_name, l_name


def create_gn_named_attr_node(nodes):
    try:
        return nodes.new("GeometryNodeNamedAttribute")
    except Exception:
        return nodes.new("GeometryNodeInputNamedAttribute")

# ==============================================================================
# RIG-PARENTED CAMERA RIG SETUP
# ==============================================================================

def setup_modern_reel_camera(target_rig, target_mesh, turn_start, turn_end_frame, unfurl_end_frame):
    scene = bpy.context.scene
    scene.render.resolution_x = 1080
    scene.render.resolution_y = 1920
    scene.render.resolution_percentage = 100

    focus_target = bpy.data.objects.get("Camera_Focus_Target") or bpy.data.objects.new("Camera_Focus_Target", None)
    if focus_target.name not in scene.collection.objects:
        scene.collection.objects.link(focus_target)

    cam_data = bpy.data.cameras.get("Vertical_Reel_Cam_Data") or bpy.data.cameras.new("Vertical_Reel_Cam_Data")
    cam_data.lens = 45.0
    cam_data.sensor_fit = 'VERTICAL'

    cam_obj = bpy.data.objects.get("Vertical_Reel_Camera") or bpy.data.objects.new("Vertical_Reel_Camera", cam_data)
    if cam_obj.name not in scene.collection.objects:
        scene.collection.objects.link(cam_obj)

    scene.camera = cam_obj
    cam_obj.parent = None

    cam_obj.constraints.clear()
    track = cam_obj.constraints.new(type='TRACK_TO')
    track.target = focus_target
    track.track_axis = 'TRACK_NEGATIVE_Z'
    track.up_axis = 'UP_Y'

    if cam_obj.animation_data:
        cam_obj.animation_data_clear()
    if focus_target.animation_data:
        focus_target.animation_data_clear()

    max_radius = 0.0
    sample_frames = list(range(1, unfurl_end_frame + 1, 5))
    if unfurl_end_frame not in sample_frames:
        sample_frames.append(unfurl_end_frame)

    depsgraph = bpy.context.evaluated_depsgraph_get()

    for f in sample_frames:
        scene.frame_set(f)
        depsgraph.update()

        eval_mesh_obj = target_mesh.evaluated_get(depsgraph)
        me = eval_mesh_obj.to_mesh()
        matrix = eval_mesh_obj.matrix_world

        world_verts = [matrix @ v.co for v in me.vertices]
        eval_mesh_obj.to_mesh_clear()

        if not world_verts:
            continue

        min_x = min(v.x for v in world_verts)
        max_x = max(v.x for v in world_verts)
        min_y = min(v.y for v in world_verts)
        max_y = max(v.y for v in world_verts)
        min_z = min(v.z for v in world_verts)
        max_z = max(v.z for v in world_verts)

        center = Vector(((min_x + max_x) * 0.5, (min_y + max_y) * 0.5, (min_z + max_z) * 0.5))

        for v in world_verts:
            dist = (v - center).length
            if dist > max_radius:
                max_radius = dist

    padded_radius = max_radius * 1.15

    sensor_height = cam_data.sensor_height
    focal_length = cam_data.lens

    fov_v = 2.0 * math.atan((sensor_height * 0.5) / focal_length)
    aspect_ratio = scene.render.resolution_x / scene.render.resolution_y
    fov_h = 2.0 * math.atan(math.tan(fov_v * 0.5) * aspect_ratio)

    min_fov = min(fov_v, fov_h)
    required_dist = padded_radius / math.tan(min_fov * 0.5)

    root_bone_name = "mixamorig:Hips" if "mixamorig:Hips" in target_rig.pose.bones else target_rig.pose.bones[0].name

    for f in range(1, unfurl_end_frame + 1):
        scene.frame_set(f)
        depsgraph.update()

        eval_mesh_obj = target_mesh.evaluated_get(depsgraph)
        bbox_world = [eval_mesh_obj.matrix_world @ Vector(corner) for corner in eval_mesh_obj.bound_box]
        min_z = min(v.z for v in bbox_world)
        max_z = max(v.z for v in bbox_world)
        target_mid_z = min_z + ((max_z - min_z) * 0.50)

        eval_rig = target_rig.evaluated_get(depsgraph)
        pbone = eval_rig.pose.bones[root_bone_name]
        bone_world_matrix = eval_rig.matrix_world @ pbone.matrix

        char_ground_loc = Vector((bone_world_matrix.translation.x, bone_world_matrix.translation.y, min_z))

        focus_target.location = Vector((char_ground_loc.x, char_ground_loc.y, target_mid_z))
        focus_target.keyframe_insert(data_path="location", frame=f)

        rig_matrix = eval_rig.matrix_world
        forward_vec = (rig_matrix.to_3x3() @ Vector((0, 1, 0))).xy.normalized()
        if forward_vec.length == 0:
            forward_vec = Vector((0, -1))

        base_angle = math.atan2(forward_vec.y, forward_vec.x)

        if f <= turn_start:
            angle_offset = math.radians(-45.0)
        elif f >= turn_end_frame:
            angle_offset = 0.0
        else:
            t = (f - turn_start) / (turn_end_frame - turn_start)
            angle_offset = math.radians(-45.0) * (1.0 - t)

        cam_angle = base_angle + angle_offset

        cam_x = char_ground_loc.x - (required_dist * math.cos(cam_angle))
        cam_y = char_ground_loc.y - (required_dist * math.sin(cam_angle))

        cam_obj.location = Vector((cam_x, cam_y, target_mid_z))
        cam_obj.keyframe_insert(data_path="location", frame=f)

    for obj in [cam_obj, focus_target]:
        if obj and obj.animation_data and obj.animation_data.action:
            action = obj.animation_data.action
            fcurves = getattr(action, "fcurves", None)
            if fcurves is None and hasattr(action, "bindings"):
                fcurves = [fc for b in action.bindings for fc in b.fcurves if hasattr(b, "fcurves")]
            if fcurves:
                for fc in fcurves:
                    for kp in fc.keyframe_points:
                        kp.interpolation = 'LINEAR'

    for obj in [target_rig, target_mesh]:
        if obj and obj.animation_data and obj.animation_data.nla_tracks:
            for track in obj.animation_data.nla_tracks:
                for strip in track.strips:
                    strip.extrapolation = 'HOLD'

    scene.frame_set(1)
    depsgraph.update()

    return cam_obj, focus_target


def frame_scroll_image(cam_obj, focus_target, target_mesh, image_path, scroll_width=0.35, zoom_duration=45):
    scene = bpy.context.scene
    unfurl_end_frame = scene.frame_end
    zoom_end_frame = unfurl_end_frame + zoom_duration

    scene.frame_set(unfurl_end_frame)
    bpy.context.view_layer.update()

    img_aspect = 1.0
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        if img.size[1] > 0:
            img_aspect = img.size[0] / img.size[1]

    tile_height = scroll_width * img_aspect
    cam_data = cam_obj.data
    aspect_ratio = scene.render.resolution_x / scene.render.resolution_y
    sensor_width = cam_data.sensor_height * aspect_ratio

    fov_h_half = math.atan((sensor_width / 2.0) / cam_data.lens)
    target_framing_distance = (tile_height / 2.0) / math.tan(fov_h_half)

    mesh_matrix = target_mesh.matrix_world
    depsgraph = bpy.context.evaluated_depsgraph_get()
    eval_mesh = target_mesh.evaluated_get(depsgraph)

    r_hand_name, l_hand_name = find_hand_vertex_groups(target_mesh)
    vg_r = eval_mesh.vertex_groups.get(r_hand_name)
    vg_l = eval_mesh.vertex_groups.get(l_hand_name)

    r_verts = [mesh_matrix @ v.co for v in eval_mesh.data.vertices if any(g.group == vg_r.index for g in v.groups)] if vg_r else []
    l_verts = [mesh_matrix @ v.co for v in eval_mesh.data.vertices if any(g.group == vg_l.index for g in v.groups)] if vg_l else []

    target_height = target_mesh.dimensions.z if target_mesh.dimensions.z > 0 else 1.8
    torso_fallback = mesh_matrix.translation + Vector((0, 0, target_height * 0.55))

    r_hand_pos = sum(r_verts, Vector((0, 0, 0))) / len(r_verts) if r_verts else torso_fallback
    l_hand_pos = sum(l_verts, Vector((0, 0, 0))) / len(l_verts) if l_verts else torso_fallback

    scroll_midpoint = (r_hand_pos + l_hand_pos) * 0.5

    cam_pos = cam_obj.matrix_world.translation.copy()
    cam_dir = cam_pos - scroll_midpoint
    cam_dir.z = 0.0

    scroll_normal = cam_dir.normalized() if cam_dir.length > 0.001 else Vector((0, -1, 0))

    tile_center = scroll_midpoint + (scroll_normal * 0.10)
    end_pos = tile_center + (scroll_normal * target_framing_distance)

    cam_obj.parent = None
    focus_target.parent = None

    current_world_focus = focus_target.matrix_world.translation.copy()
    focus_target.matrix_world.translation = current_world_focus

    focus_target.location = current_world_focus
    focus_target.keyframe_insert(data_path="location", frame=unfurl_end_frame)
    focus_target.location = tile_center
    focus_target.keyframe_insert(data_path="location", frame=zoom_end_frame)

    cam_obj.location = cam_pos
    cam_obj.keyframe_insert(data_path="location", frame=unfurl_end_frame)
    cam_obj.location = end_pos
    cam_obj.keyframe_insert(data_path="location", frame=zoom_end_frame)

    for obj in [cam_obj, focus_target]:
        for fc in get_object_fcurves(obj):
            for kp in fc.keyframe_points:
                if kp.co.x >= unfurl_end_frame:
                    kp.interpolation = 'BEZIER'
                    kp.handle_left_type = 'AUTO_CLAMPED'
                    kp.handle_right_type = 'AUTO_CLAMPED'

    scene.frame_end = zoom_end_frame
    return zoom_end_frame

# ==============================================================================
# GROUND & SKY DOME SETUP
# ==============================================================================

def ensure_world_sphere_size(cam_obj, ground_obj, safety_multiplier=2.5):
    if not cam_obj or not ground_obj:
        return

    scene = bpy.context.scene
    max_cam_dist = 0.0

    for frame in range(scene.frame_start, scene.frame_end + 1, 5):
        scene.frame_set(frame)
        bpy.context.view_layer.update()
        max_cam_dist = max(max_cam_dist, cam_obj.matrix_world.translation.length)

    world_radius = max(25.0, max_cam_dist * safety_multiplier)

    mod = ground_obj.modifiers.get("Grass_Scatter_GN")
    if mod and mod.node_group:
        for node in mod.node_group.nodes:
            if node.bl_idname == "GeometryNodeMeshIcoSphere":
                node.inputs["Radius"].default_value = world_radius

    if ground_obj.type == 'MESH':
        bpy.context.view_layer.objects.active = ground_obj
        bpy.ops.object.mode_set(mode='OBJECT')
        ground_scale = world_radius / 12.5
        ground_obj.scale = (ground_scale, ground_scale, 1.0)
        bpy.ops.object.transform_apply(location=False, rotation=False, scale=True)

    cam_obj.data.clip_end = max(cam_obj.data.clip_end, world_radius * 2.5)
    scene.frame_set(1)


def find_grass_texture(grass_dir):
    if not os.path.exists(grass_dir):
        return None

    extensions = ['*.jpg', '*.jpeg', '*.png', '*.tif', '*.tiff']
    matches = []
    for ext in extensions:
        matches.extend(glob.glob(os.path.join(grass_dir, ext)))
        matches.extend(glob.glob(os.path.join(grass_dir, "**", ext), recursive=True))

    for m in matches:
        filename = os.path.basename(m).lower()
        if any(k in filename for k in ["color", "diff", "albedo", "col", "basecolor"]):
            return m

    return matches[0] if matches else None


def create_ground_material(grass_dir=None):
    mat = bpy.data.materials.new(name="Ground_Soil_Material")
    mat.use_nodes = True
    nodes, links = mat.node_tree.nodes, mat.node_tree.links
    nodes.clear()

    out_node = nodes.new(type="ShaderNodeOutputMaterial")
    bsdf = nodes.new(type="ShaderNodeBsdfPrincipled")

    tex_file = find_grass_texture(grass_dir) if grass_dir else None

    if tex_file and os.path.exists(tex_file):
        tex_node = nodes.new(type="ShaderNodeTexImage")
        tex_node.image = bpy.data.images.load(tex_file)

        tex_coord = nodes.new(type="ShaderNodeTexCoord")
        mapping = nodes.new(type="ShaderNodeMapping")
        mapping.inputs["Scale"].default_value = (20.0, 20.0, 1.0)

        links.new(tex_coord.outputs["UV"], mapping.inputs["Vector"])
        links.new(mapping.outputs["Vector"], tex_node.inputs["Vector"])
        links.new(tex_node.outputs["Color"], bsdf.inputs["Base Color"])
    else:
        noise = nodes.new("ShaderNodeTexNoise")
        noise.inputs["Scale"].default_value = 18.0
        noise.inputs["Detail"].default_value = 6.0

        color_ramp = nodes.new("ShaderNodeValToRGB")
        color_ramp.color_ramp.elements[0].color = (0.04, 0.025, 0.01, 1.0)
        color_ramp.color_ramp.elements[1].color = (0.05, 0.07, 0.02, 1.0)

        links.new(noise.outputs["Factor"], color_ramp.inputs["Factor"])
        links.new(color_ramp.outputs["Color"], bsdf.inputs["Base Color"])

    bsdf.inputs["Roughness"].default_value = 0.95
    links.new(bsdf.outputs["BSDF"], out_node.inputs["Surface"])
    return mat


def create_sky_material(sky_hdri_path):
    mat = bpy.data.materials.new(name="Sky_Dome_Material")
    mat.use_nodes = True
    nodes, links = mat.node_tree.nodes, mat.node_tree.links
    nodes.clear()

    out_node = nodes.new("ShaderNodeOutputMaterial")
    emission = nodes.new("ShaderNodeEmission")
    emission.inputs["Strength"].default_value = 1.0

    transparent = nodes.new("ShaderNodeBsdfTransparent")
    mix_shader = nodes.new("ShaderNodeMixShader")
    light_path = nodes.new("ShaderNodeLightPath")

    links.new(light_path.outputs["Is Camera Ray"], mix_shader.inputs["Fac"])
    links.new(transparent.outputs["BSDF"], mix_shader.inputs[1])
    links.new(emission.outputs["Emission"], mix_shader.inputs[2])
    links.new(mix_shader.outputs["Shader"], out_node.inputs["Surface"])

    if os.path.exists(sky_hdri_path):
        tex_env = nodes.new("ShaderNodeTexEnvironment")
        tex_env.image = bpy.data.images.load(sky_hdri_path)
        links.new(tex_env.outputs["Color"], emission.inputs["Color"])

    return mat


def setup_grass_ground_geometry_nodes(grass_dir, sky_hdri_path, ground_z_offset=0.0):
    bpy.ops.mesh.primitive_plane_add(size=60, location=(0, 0, ground_z_offset))
    ground_obj = bpy.context.active_object
    ground_obj.name = "Grass_Ground_Plane"

    ground_mat = create_ground_material(grass_dir=grass_dir)
    sky_mat = create_sky_material(sky_hdri_path)
    ground_obj.data.materials.append(ground_mat)
    ground_obj.data.materials.append(sky_mat)

    gn_mod = ground_obj.modifiers.new(name="Grass_Scatter_GN", type='NODES')
    gn_group = bpy.data.node_groups.new("Grass_Scatter_Group", type='GeometryNodeTree')
    gn_mod.node_group = gn_group

    nodes = gn_group.nodes
    links = gn_group.links
    nodes.clear()

    group_in = nodes.new("NodeGroupInput")
    gn_group.interface.new_socket("Geometry", in_out='INPUT', socket_type='NodeSocketGeometry')
    group_out = nodes.new("NodeGroupOutput")
    gn_group.interface.new_socket("Geometry", in_out='OUTPUT', socket_type='NodeSocketGeometry')

    sky_sphere = nodes.new("GeometryNodeMeshIcoSphere")
    sky_sphere.inputs["Radius"].default_value = 12.5
    sky_sphere.inputs["Subdivisions"].default_value = 4

    flip_faces = nodes.new("GeometryNodeFlipFaces")
    links.new(sky_sphere.outputs["Mesh"], flip_faces.inputs["Mesh"])

    transform_sky = nodes.new("GeometryNodeTransform")
    transform_sky.inputs["Translation"].default_value = (0.0, 0.0, ground_z_offset)
    links.new(flip_faces.outputs["Mesh"], transform_sky.inputs["Geometry"])

    set_sky_mat = nodes.new("GeometryNodeSetMaterial")
    set_sky_mat.inputs["Material"].default_value = sky_mat
    links.new(transform_sky.outputs["Geometry"], set_sky_mat.inputs["Geometry"])

    set_ground_mat = nodes.new("GeometryNodeSetMaterial")
    set_ground_mat.inputs["Material"].default_value = ground_mat
    links.new(group_in.outputs["Geometry"], set_ground_mat.inputs["Geometry"])

    join_geo = nodes.new("GeometryNodeJoinGeometry")
    links.new(set_ground_mat.outputs["Geometry"], join_geo.inputs["Geometry"])
    links.new(set_sky_mat.outputs["Geometry"], join_geo.inputs["Geometry"])

    links.new(join_geo.outputs["Geometry"], group_out.inputs["Geometry"])

    return ground_obj

# ==============================================================================
# SCROLL MATERIAL & GEOMETRY NODES SETUP
# ==============================================================================

def create_image_scroll_material(image_path):
    mat_name = "Scroll_Texture_Mat"
    if mat := bpy.data.materials.get(mat_name):
        bpy.data.materials.remove(mat)

    mat = bpy.data.materials.new(name=mat_name)
    mat.use_nodes = True
    nodes, links = mat.node_tree.nodes, mat.node_tree.links
    nodes.clear()

    output_node = nodes.new("ShaderNodeOutputMaterial")
    bsdf = nodes.new("ShaderNodeBsdfPrincipled")
    bsdf.inputs["Roughness"].default_value = 0.5

    tex_img = nodes.new("ShaderNodeTexImage")
    if os.path.exists(image_path):
        tex_img.image = bpy.data.images.load(image_path)
        tex_img.extension = 'REPEAT'

    attr_node = nodes.new("ShaderNodeAttribute")
    attr_node.attribute_name = "UVMap"

    links.new(attr_node.outputs["Vector"], tex_img.inputs["Vector"])
    links.new(tex_img.outputs["Color"], bsdf.inputs["Base Color"])
    links.new(bsdf.outputs["BSDF"], output_node.inputs["Surface"])

    return mat, tex_img.image if os.path.exists(image_path) else None


def create_pure_geometry_node_scroll(image_path=""):
    gn_group = bpy.data.node_groups.new("Scroll_Pure_GN", type="GeometryNodeTree")

    gn_group.interface.new_socket(name="Geometry", in_out="OUTPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Character Mesh", in_out="INPUT", socket_type="NodeSocketObject")

    sock_r_vg = gn_group.interface.new_socket(name="Right Hand Group", in_out="INPUT", socket_type="NodeSocketString")
    sock_r_vg.default_value = "mixamorig:RightHand"

    sock_l_vg = gn_group.interface.new_socket(name="Left Hand Group", in_out="INPUT", socket_type="NodeSocketString")
    sock_l_vg.default_value = "mixamorig:LeftHand"

    sock_offset = gn_group.interface.new_socket(name="Forward Offset", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_offset.default_value = 0.60

    sock_unfurl = gn_group.interface.new_socket(name="Unfurl", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_unfurl.default_value = 0.0

    sock_width = gn_group.interface.new_socket(name="Scroll Width", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_width.default_value = 0.35

    nodes, links = gn_group.nodes, gn_group.links
    in_node, out_node = nodes.new("NodeGroupInput"), nodes.new("NodeGroupOutput")

    char_info = nodes.new("GeometryNodeObjectInfo")
    char_info.transform_space = "RELATIVE"
    links.new(in_node.outputs["Character Mesh"], char_info.inputs["Object"])

    pos_node = nodes.new("GeometryNodeInputPosition")

    named_r = create_gn_named_attr_node(nodes)
    named_r.data_type = "FLOAT"
    named_r.inputs["Name"].default_value = "mixamorig:RightHand"
    links.new(in_node.outputs["Right Hand Group"], named_r.inputs["Name"])

    attr_r_out = named_r.outputs.get("Attribute") or named_r.outputs[0]

    cmp_r = nodes.new("FunctionNodeCompare")
    cmp_r.data_type = "FLOAT"
    cmp_r.operation = "GREATER_THAN"
    cmp_r.inputs[1].default_value = 0.001
    links.new(attr_r_out, cmp_r.inputs[0])

    stat_r = nodes.new("GeometryNodeAttributeStatistic")
    stat_r.data_type = "FLOAT_VECTOR"
    links.new(char_info.outputs["Geometry"], stat_r.inputs["Geometry"])
    links.new(cmp_r.outputs["Result"], stat_r.inputs["Selection"])
    links.new(pos_node.outputs["Position"], stat_r.inputs["Attribute"])

    named_l = create_gn_named_attr_node(nodes)
    named_l.data_type = "FLOAT"
    named_l.inputs["Name"].default_value = "mixamorig:LeftHand"
    links.new(in_node.outputs["Left Hand Group"], named_l.inputs["Name"])

    attr_l_out = named_l.outputs.get("Attribute") or named_l.outputs[0]

    cmp_l = nodes.new("FunctionNodeCompare")
    cmp_l.data_type = "FLOAT"
    cmp_l.operation = "GREATER_THAN"
    cmp_l.inputs[1].default_value = 0.001
    links.new(attr_l_out, cmp_l.inputs[0])

    stat_l = nodes.new("GeometryNodeAttributeStatistic")
    stat_l.data_type = "FLOAT_VECTOR"
    links.new(char_info.outputs["Geometry"], stat_l.inputs["Geometry"])
    links.new(cmp_l.outputs["Result"], stat_l.inputs["Selection"])
    links.new(pos_node.outputs["Position"], stat_l.inputs["Attribute"])

    neg_offset = nodes.new("ShaderNodeMath")
    neg_offset.operation = "MULTIPLY"
    neg_offset.inputs[1].default_value = -1.0
    links.new(in_node.outputs["Forward Offset"], neg_offset.inputs[0])

    forward_vec = nodes.new("ShaderNodeCombineXYZ")
    forward_vec.inputs["X"].default_value = 0.0
    links.new(neg_offset.outputs["Value"], forward_vec.inputs["Y"])
    forward_vec.inputs["Z"].default_value = 0.0

    start_pos = nodes.new("ShaderNodeVectorMath")
    start_pos.operation = "ADD"
    links.new(stat_r.outputs["Mean"], start_pos.inputs[0])
    links.new(forward_vec.outputs["Vector"], start_pos.inputs[1])

    target_pos = nodes.new("ShaderNodeVectorMath")
    target_pos.operation = "ADD"
    links.new(stat_l.outputs["Mean"], target_pos.inputs[0])
    links.new(forward_vec.outputs["Vector"], target_pos.inputs[1])

    delta_vec = nodes.new("ShaderNodeVectorMath")
    delta_vec.operation = "SUBTRACT"
    links.new(target_pos.outputs["Vector"], delta_vec.inputs[0])
    links.new(start_pos.outputs["Vector"], delta_vec.inputs[1])

    dist_node = nodes.new("ShaderNodeVectorMath")
    dist_node.operation = "DISTANCE"
    links.new(start_pos.outputs["Vector"], dist_node.inputs[0])

    scale_vec = nodes.new("ShaderNodeVectorMath")
    scale_vec.operation = "SCALE"
    links.new(delta_vec.outputs["Vector"], scale_vec.inputs["Vector"])
    links.new(in_node.outputs["Unfurl"], scale_vec.inputs["Scale"])

    end_pos = nodes.new("ShaderNodeVectorMath")
    end_pos.operation = "ADD"
    links.new(start_pos.outputs["Vector"], end_pos.inputs[0])
    links.new(scale_vec.outputs["Vector"], end_pos.inputs[1])
    links.new(end_pos.outputs["Vector"], dist_node.inputs[1])

    curve_line = nodes.new("GeometryNodeCurvePrimitiveLine")
    links.new(start_pos.outputs["Vector"], curve_line.inputs["Start"])
    links.new(end_pos.outputs["Vector"], curve_line.inputs["End"])

    set_tilt = nodes.new("GeometryNodeSetCurveTilt")
    set_tilt.inputs["Tilt"].default_value = 1.5708
    links.new(curve_line.outputs["Curve"], set_tilt.inputs["Curve"])

    spline_param_main = nodes.new("GeometryNodeSplineParameter")
    store_u = nodes.new("GeometryNodeStoreNamedAttribute")
    store_u.inputs["Name"].default_value = "U_Factor"
    store_u.data_type = 'FLOAT'
    store_u.domain = 'POINT'
    links.new(set_tilt.outputs["Curve"], store_u.inputs["Geometry"])

    spline_factor_out = spline_param_main.outputs.get("Factor") or spline_param_main.outputs[0]
    links.new(spline_factor_out, store_u.inputs["Value"])

    half_width_neg = nodes.new("ShaderNodeMath")
    half_width_neg.operation = "MULTIPLY"
    half_width_neg.inputs[1].default_value = -0.5
    links.new(in_node.outputs["Scroll Width"], half_width_neg.inputs[0])

    half_width_pos = nodes.new("ShaderNodeMath")
    half_width_pos.operation = "MULTIPLY"
    half_width_pos.inputs[1].default_value = 0.5
    links.new(in_node.outputs["Scroll Width"], half_width_pos.inputs[0])

    combine_start = nodes.new("ShaderNodeCombineXYZ")
    links.new(half_width_neg.outputs["Value"], combine_start.inputs["X"])

    combine_end = nodes.new("ShaderNodeCombineXYZ")
    links.new(half_width_pos.outputs["Value"], combine_end.inputs["X"])

    profile_line = nodes.new("GeometryNodeCurvePrimitiveLine")
    links.new(combine_start.outputs["Vector"], profile_line.inputs["Start"])
    links.new(combine_end.outputs["Vector"], profile_line.inputs["End"])

    spline_param_prof = nodes.new("GeometryNodeSplineParameter")
    store_v = nodes.new("GeometryNodeStoreNamedAttribute")
    store_v.inputs["Name"].default_value = "V_Factor"
    store_v.data_type = 'FLOAT'
    store_v.domain = 'POINT'
    links.new(profile_line.outputs["Curve"], store_v.inputs["Geometry"])

    prof_factor_out = spline_param_prof.outputs.get("Factor") or spline_param_prof.outputs[0]
    links.new(prof_factor_out, store_v.inputs["Value"])

    curve_to_mesh = nodes.new("GeometryNodeCurveToMesh")
    links.new(store_u.outputs["Geometry"], curve_to_mesh.inputs["Curve"])
    links.new(store_v.outputs["Geometry"], curve_to_mesh.inputs["Profile Curve"])

    img_aspect = 1.0
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        if img.size[1] > 0:
            img_aspect = img.size[0] / img.size[1]

    h_times_aspect = nodes.new("ShaderNodeMath")
    h_times_aspect.operation = "MULTIPLY"
    h_times_aspect.inputs[1].default_value = img_aspect
    links.new(in_node.outputs["Scroll Width"], h_times_aspect.inputs[0])

    tile_count = nodes.new("ShaderNodeMath")
    tile_count.operation = "DIVIDE"
    links.new(dist_node.outputs["Value"], tile_count.inputs[0])
    links.new(h_times_aspect.outputs["Value"], tile_count.inputs[1])

    read_u = create_gn_named_attr_node(nodes)
    read_u.data_type = 'FLOAT'
    read_u.inputs["Name"].default_value = "U_Factor"
    read_u_out = read_u.outputs.get("Attribute") or read_u.outputs[0]

    scaled_u = nodes.new("ShaderNodeMath")
    scaled_u.operation = "MULTIPLY"
    links.new(read_u_out, scaled_u.inputs[0])
    links.new(tile_count.outputs["Value"], scaled_u.inputs[1])

    read_v = create_gn_named_attr_node(nodes)
    read_v.data_type = 'FLOAT'
    read_v.inputs["Name"].default_value = "V_Factor"
    read_v_out = read_v.outputs.get("Attribute") or read_v.outputs[0]

    flip_v = nodes.new("ShaderNodeMath")
    flip_v.operation = "SUBTRACT"
    flip_v.inputs[0].default_value = 1.0
    links.new(read_v_out, flip_v.inputs[1])

    combine_uv = nodes.new("ShaderNodeCombineXYZ")
    links.new(scaled_u.outputs["Value"], combine_uv.inputs["X"])
    links.new(flip_v.outputs["Value"], combine_uv.inputs["Y"])

    store_uv = nodes.new("GeometryNodeStoreNamedAttribute")
    store_uv.inputs["Name"].default_value = "UVMap"
    store_uv.data_type = 'FLOAT2'
    store_uv.domain = 'CORNER'
    links.new(curve_to_mesh.outputs["Mesh"], store_uv.inputs["Geometry"])
    links.new(combine_uv.outputs["Vector"], store_uv.inputs["Value"])

    set_material = nodes.new("GeometryNodeSetMaterial")
    scroll_mat, _ = create_image_scroll_material(image_path)
    set_material.inputs["Material"].default_value = scroll_mat
    links.new(store_uv.outputs["Geometry"], set_material.inputs["Geometry"])

    links.new(set_material.outputs["Geometry"], out_node.inputs["Geometry"])

    return gn_group, scroll_mat


def setup_geometry_node_scroll(target_mesh, unfurl_start, unfurl_end, image_path=""):
    mesh_data = bpy.data.meshes.new("Scroll_Procedural_Mesh")
    scroll_obj = bpy.data.objects.new("Scroll_GN_Object", mesh_data)
    bpy.context.scene.collection.objects.link(scroll_obj)

    mod = scroll_obj.modifiers.new("Scroll_GN", type="NODES")
    gn_tree, scroll_mat = create_pure_geometry_node_scroll(image_path=image_path)
    mod.node_group = gn_tree

    scroll_obj.data.materials.append(scroll_mat)

    r_hand_name, l_hand_name = find_hand_vertex_groups(target_mesh)

    unfurl_id = None
    for item in gn_tree.interface.items_tree:
        if getattr(item, "in_out", None) == "INPUT":
            if item.name == "Character Mesh":
                mod[item.identifier] = target_mesh
            elif item.name == "Right Hand Group":
                mod[item.identifier] = r_hand_name
            elif item.name == "Left Hand Group":
                mod[item.identifier] = l_hand_name
            elif item.name == "Forward Offset":
                mod[item.identifier] = 0.60
            elif item.name == "Scroll Width":
                mod[item.identifier] = 0.35
            elif item.name == "Unfurl":
                unfurl_id = item.identifier

    scroll_obj.hide_viewport = True
    scroll_obj.hide_render = True
    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=1)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=1)

    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=unfurl_start - 1)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=unfurl_start - 1)

    scroll_obj.hide_viewport = False
    scroll_obj.hide_render = False
    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=unfurl_start)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=unfurl_start)

    for fc in get_object_fcurves(scroll_obj):
        if fc.data_path in ["hide_viewport", "hide_render"]:
            for kp in fc.keyframe_points:
                kp.interpolation = 'CONSTANT'

    if unfurl_id:
        mod[unfurl_id] = 0.001
        mod.keyframe_insert(data_path=f'["{unfurl_id}"]', frame=unfurl_start)

        mod[unfurl_id] = 1.0
        mod.keyframe_insert(data_path=f'["{unfurl_id}"]', frame=unfurl_end)

    return scroll_obj

# ==============================================================================
# MAIN CONTROLLER
# ==============================================================================

def import_and_extract_action(fbx_path, action_name):
    existing_objects = set(bpy.data.objects)
    bpy.ops.import_scene.fbx(filepath=fbx_path)
    bpy.context.view_layer.update()

    imported_objects = set(bpy.data.objects) - existing_objects
    extracted_action = None
    for obj in imported_objects:
        if obj.animation_data and obj.animation_data.action:
            extracted_action = obj.animation_data.action
            extracted_action.name = action_name
            extracted_action.use_fake_user = True
            break
    return extracted_action, imported_objects


def purge_imported_objects(objects_to_remove):
    for obj in objects_to_remove:
        if obj.name in bpy.data.objects:
            if obj.animation_data:
                obj.animation_data_clear()
            for coll in list(obj.users_collection):
                coll.objects.unlink(obj)
    bpy.context.view_layer.update()


def setup_world_hdri_lighting(sky_hdri_path, strength=2.5):
    scene = bpy.context.scene
    if not scene.world:
        scene.world = bpy.data.worlds.new("World")

    scene.world.use_nodes = True
    nodes = scene.world.node_tree.nodes
    links = scene.world.node_tree.links
    nodes.clear()

    out_node = nodes.new("ShaderNodeOutputWorld")
    bg_node = nodes.new("ShaderNodeBackground")
    bg_node.inputs["Strength"].default_value = strength

    links.new(bg_node.outputs["Background"], out_node.inputs["Surface"])

    if os.path.exists(sky_hdri_path):
        tex_env = nodes.new("ShaderNodeTexEnvironment")
        tex_env.image = bpy.data.images.load(sky_hdri_path)
        links.new(tex_env.outputs["Color"], bg_node.inputs["Color"])


def main():
    cleanup_previous_run()

    walk_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Mutant Walking.fbx"
    turn_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Right Turn 90.fbx"
    unfurl_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Mutant Idle.fbx"

    sky_hdri_path = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/horn-koppe_spring_4k.exr"
    output_blend = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/output_scene.blend"

    image_path = "/Users/dobbinp1/Downloads/sea_PanoramaIMGP5963.jpg"
    grass_dir = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Grass004_4K-JPG"

    scene = bpy.context.scene
    scene.view_settings.exposure = 1.0
    scene.view_settings.view_transform = 'AgX'
    scene.view_settings.look = 'AgX - Medium High Contrast'

    setup_world_hdri_lighting(sky_hdri_path, strength=1.0)

    # Set ground Z offset to 0.0 to match world ground level
    ground_obj = setup_grass_ground_geometry_nodes(grass_dir, sky_hdri_path, ground_z_offset=0.0)

    CYCLES = 4
    CROSSFADE = 15
    TARGET_DISTANCE = 4.0

    walk_action, walk_objs = import_and_extract_action(walk_fbx, "walk")
    target_rig = next((obj for obj in walk_objs if obj.type == "ARMATURE"), None)
    target_mesh = next((obj for obj in walk_objs if obj.type == "MESH"), None)

    if not target_rig or not target_mesh:
        raise RuntimeError("Armature or Character Mesh not found in walk FBX.")

    # Explicitly snap armature base to ground Z = 0.0
    target_rig.location.z = 0.0

    travel_axis_idx, motion_sign = detect_natural_forward_axis(walk_action, target_rig)
    initial_heading = math.radians(90.0) if travel_axis_idx == 0 else 0.0
    target_rig.rotation_euler[2] = initial_heading

    action_len = max(30, int(walk_action.frame_range[1] - walk_action.frame_range[0]))
    anim_data = target_rig.animation_data_create()

    walk_track = anim_data.nla_tracks.new()
    walk_track.name = "Walk_Track"

    current_frame = 1
    for i in range(CYCLES):
        strip = walk_track.strips.new(name=f"Walk_{i+1}", start=int(current_frame), action=walk_action)
        strip.blend_type = "REPLACE"
        current_frame += action_len

    walk_end_frame = int(current_frame)

    turn_action, turn_objs = import_and_extract_action(turn_fbx, "rightturn")
    purge_imported_objects(turn_objs)

    turn_track = anim_data.nla_tracks.new()
    turn_track.name = "RightTurn_Track"

    turn_start = int(walk_end_frame - CROSSFADE)
    turn_strip = turn_track.strips.new(name="RightTurn", start=turn_start, action=turn_action)
    turn_strip.blend_type = "REPLACE"
    turn_strip.blend_in = CROSSFADE
    turn_strip.use_auto_blend = True
    turn_strip.extrapolation = "NOTHING"
    turn_end_frame = int(turn_strip.frame_end)

    bpy.context.scene["turn_start_frame"] = turn_start

    unfurl_action, unfurl_objs = import_and_extract_action(unfurl_fbx, "unfurl")
    purge_imported_objects(unfurl_objs)

    unfurl_track = anim_data.nla_tracks.new()
    unfurl_track.name = "Unfurl_Track"

    unfurl_start = int(turn_end_frame - CROSSFADE)
    unfurl_strip = unfurl_track.strips.new(name="Unfurl", start=unfurl_start, action=unfurl_action)
    unfurl_strip.blend_type = "REPLACE"
    unfurl_strip.blend_in = CROSSFADE
    unfurl_strip.use_auto_blend = True
    unfurl_strip.extrapolation = "HOLD"
    unfurl_end_frame = int(unfurl_strip.frame_end)

    root_action = bpy.data.actions.new(name="World_Root_Motion")
    target_rig.animation_data.action = root_action

    if hasattr(root_action, "fcurves") and hasattr(root_action.fcurves, "new"):
        loc_fcurve = root_action.fcurves.new(data_path="location", index=travel_axis_idx)
        rot_fcurve = root_action.fcurves.new(data_path="rotation_euler", index=2)
        z_fcurve = root_action.fcurves.new(data_path="location", index=2)
    else:
        loc_fcurve = root_action.fcurve_ensure_for_datablock(datablock=target_rig, data_path="location", index=travel_axis_idx)
        rot_fcurve = root_action.fcurve_ensure_for_datablock(datablock=target_rig, data_path="rotation_euler", index=2)
        z_fcurve = root_action.fcurve_ensure_for_datablock(datablock=target_rig, data_path="location", index=2)

    # Keyframe Z axis to force 0.0 throughout the animation sequence
    z_fcurve.keyframe_points.insert(frame=1.0, value=0.0)
    z_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=0.0)

    final_target_value = TARGET_DISTANCE * motion_sign

    k_loc1 = loc_fcurve.keyframe_points.insert(frame=1.0, value=0.0)
    k_loc2 = loc_fcurve.keyframe_points.insert(frame=float(turn_start), value=final_target_value)
    k_loc3 = loc_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=final_target_value)

    k_loc1.interpolation = "LINEAR"
    k_loc2.interpolation = "CONSTANT"
    k_loc3.interpolation = "CONSTANT"

    target_heading = initial_heading - math.radians(90.0)

    k_rot1 = rot_fcurve.keyframe_points.insert(frame=1.0, value=initial_heading)
    k_rot2 = rot_fcurve.keyframe_points.insert(frame=float(turn_start), value=initial_heading)
    k_rot3 = rot_fcurve.keyframe_points.insert(frame=float(turn_end_frame), value=target_heading)
    k_rot4 = rot_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=target_heading)

    k_rot1.interpolation = "CONSTANT"
    k_rot2.interpolation = "BEZIER"
    k_rot3.interpolation = "CONSTANT"
    k_rot4.interpolation = "CONSTANT"

    root_track = anim_data.nla_tracks.new()
    root_track.name = "World_Root_Track"
    root_strip = root_track.strips.new(name="Root_Motion_Strip", start=1, action=root_action)
    root_strip.blend_type = "REPLACE"
    root_strip.extrapolation = "HOLD"

    anim_data.action = None

    setup_geometry_node_scroll(
        target_mesh=target_mesh,
        unfurl_start=unfurl_start,
        unfurl_end=unfurl_end_frame,
        image_path=image_path,
    )

    cam_obj, focus_target = setup_modern_reel_camera(
        target_rig=target_rig,
        target_mesh=target_mesh,
        turn_start=turn_start,
        turn_end_frame=turn_end_frame,
        unfurl_end_frame=unfurl_end_frame
    )

    bpy.context.scene.frame_start = 1
    bpy.context.scene.frame_end = unfurl_end_frame

    frame_scroll_image(
        cam_obj=cam_obj,
        focus_target=focus_target,
        target_mesh=target_mesh,
        image_path=image_path,
        scroll_width=0.35,
        zoom_duration=45
    )

    ensure_world_sphere_size(cam_obj, ground_obj, safety_multiplier=2.5)
    register_audio_handler()

    bpy.ops.wm.save_as_mainfile(filepath=output_blend)
    print(f"\n[SUCCESS] Saved scene to: {output_blend}")


if __name__ == "__main__":
    main()
```

```
import math
import mathutils
import os
import sys
import glob
import bpy
import aud
from bpy_extras import anim_utils
from bpy_extras.object_utils import world_to_camera_view
from bpy.app.handlers import persistent
from mathutils import Vector

# ==============================================================================
# CONFIGURATION
# ==============================================================================
FOOTSTEP_SOUND_PATH = "/Users/dobbinp1/Downloads/stomp.wav"
GROUND_THRESHOLD = 0.24

# Setup Audio Device
if "_audio_store" not in sys.modules:
    import types
    sys.modules["_audio_store"] = types.ModuleType("_audio_store")

audio_store = sys.modules["_audio_store"]
if not hasattr(audio_store, "device"):
    audio_store.device = aud.Device()

AUDIO_DEVICE = audio_store.device
SOUND_FACTORY = None
AUDIO_HANDLES = []

if os.path.exists(FOOTSTEP_SOUND_PATH):
    SOUND_FACTORY = aud.Sound(FOOTSTEP_SOUND_PATH).cache()
    print("[AUDIO] Loaded sound file successfully.")
else:
    print(f"[AUDIO ERROR] File not found at: {FOOTSTEP_SOUND_PATH}")

_left_foot_was_down = False
_right_foot_was_down = False
_last_frame = -1


def play_footstep():
    global AUDIO_HANDLES
    if SOUND_FACTORY:
        handle = AUDIO_DEVICE.play(SOUND_FACTORY)
        AUDIO_HANDLES.append(handle)
        if len(AUDIO_HANDLES) > 10:
            AUDIO_HANDLES.pop(0)


def find_foot_bone(pose_bones, side="Left"):
    target_keywords = [
        f"mixamorig:{side}foot",
        f"{side.lower()}foot",
        f"foot.{side[0].lower()}",
        f"foot_{side[0].lower()}"
    ]
    for bone in pose_bones:
        b_name = bone.name.lower()
        if any(kw in b_name for kw in target_keywords):
            return bone
    return None


@persistent
def footstep_frame_handler(scene):
    global _left_foot_was_down, _right_foot_was_down, _last_frame

    try:
        current_frame = scene.frame_current
        turn_start = scene.get("turn_start_frame", None)

        if turn_start is not None and current_frame >= turn_start:
            _left_foot_was_down = False
            _right_foot_was_down = False
            return

        if current_frame <= _last_frame or abs(current_frame - _last_frame) > 10:
            _left_foot_was_down = False
            _right_foot_was_down = False
        _last_frame = current_frame

        rig = next((o for o in scene.objects if o.type == 'ARMATURE'), None)
        if not rig:
            return

        depsgraph = bpy.context.evaluated_depsgraph_get()
        eval_rig = rig.evaluated_get(depsgraph)

        left_bone = find_foot_bone(eval_rig.pose.bones, "Left")
        right_bone = find_foot_bone(eval_rig.pose.bones, "Right")

        if not left_bone or not right_bone:
            return

        l_z = (eval_rig.matrix_world @ left_bone.matrix).translation.z
        r_z = (eval_rig.matrix_world @ right_bone.matrix).translation.z

        l_is_down = l_z <= GROUND_THRESHOLD
        r_is_down = r_z <= GROUND_THRESHOLD

        if l_is_down and not _left_foot_was_down:
            play_footstep()

        if r_is_down and not _right_foot_was_down:
            play_footstep()

        _left_foot_was_down = l_is_down
        _right_foot_was_down = r_is_down

    except Exception:
        import traceback
        traceback.print_exc()


def register_audio_handler():
    for h in list(bpy.app.handlers.frame_change_post):
        if getattr(h, "__name__", "") == "footstep_frame_handler":
            bpy.app.handlers.frame_change_post.remove(h)

    bpy.app.handlers.frame_change_post.append(footstep_frame_handler)


register_audio_handler()

# ==============================================================================
# CLEANUP & HELPERS
# ==============================================================================

def cleanup_previous_run():
    target_names = [
        "Master_NLA_Control",
        "Scroll_GN_Object",
        "Scroll_Start_Pin",
        "Scroll_End_Pin",
        "Scroll_Master_Container",
        "HDRI_Ground_Plane",
        "Ground_Mesh_Object",
        "Grass_Ground_Plane",
        "Grass_Blade_Instance",
        "Camera_Focus_Target",
        "Camera_Anchor",
        "Vertical_Reel_Camera"
    ]
    for obj in list(bpy.data.objects):
        if obj.name in target_names or (obj.name.startswith("mixamorig") and obj.type not in {'ARMATURE', 'MESH'}):
            if obj.animation_data:
                obj.animation_data_clear()
            bpy.data.objects.remove(obj, do_unlink=True)

    for action in list(bpy.data.actions):
        if action.users == 0 or action.use_fake_user:
            action.use_fake_user = False
            bpy.data.actions.remove(action)

    for ng in list(bpy.data.node_groups):
        if ng.name in ["Scroll_Unfurl_GN", "Scroll_Pure_GN", "HDRI_Ground_GN", "Grass_Ground_GN", "Grass_Scatter_Group"]:
            bpy.data.node_groups.remove(ng)

    for coll_name in ["Scroll_Master_Container", "Grass_Artifacts_Collection"]:
        if coll_name in bpy.data.collections:
            coll = bpy.data.collections[coll_name]
            for obj in list(coll.objects):
                bpy.data.objects.remove(obj, do_unlink=True)
            bpy.data.collections.remove(coll)

    bpy.context.view_layer.update()


def get_slot_fcurves(action, anim_target_object):
    fcurves = []
    if not action:
        return fcurves

    anim_data = getattr(anim_target_object, "animation_data", None)
    if anim_data and getattr(anim_data, "action_slot", None):
        slot = anim_data.action_slot
        channelbag = anim_utils.action_get_channelbag_for_slot(action, slot)
        if channelbag and hasattr(channelbag, "fcurves"):
            return list(channelbag.fcurves)

    if hasattr(action, "slots"):
        for slot in action.slots:
            for cb in getattr(slot, "channelbags", []):
                fcurves.extend(getattr(cb, "fcurves", []))
        if fcurves:
            return fcurves

    if hasattr(action, "fcurves"):
        return list(action.fcurves)

    return fcurves


def get_object_fcurves(obj):
    if not obj or not obj.animation_data or not obj.animation_data.action:
        return []
    return get_slot_fcurves(obj.animation_data.action, obj)


def detect_natural_forward_axis(action, rig_object):
    fcurves = get_slot_fcurves(action, rig_object)
    max_delta = 0.0
    best_axis = 1
    net_delta = 0.0

    start_frame, end_frame = action.frame_range

    for fc in fcurves:
        if fc.data_path.endswith("location") and fc.array_index in (0, 1, 2):
            delta = fc.evaluate(end_frame) - fc.evaluate(start_frame)
            if abs(delta) > abs(max_delta):
                max_delta = delta
                best_axis = fc.array_index
                net_delta = delta

    direction_sign = 1.0 if net_delta >= 0 else -1.0
    return best_axis, direction_sign


def find_hand_vertex_groups(target_mesh):
    r_name, l_name = "mixamorig:RightHand", "mixamorig:LeftHand"
    if target_mesh and hasattr(target_mesh, "vertex_groups"):
        for vg in target_mesh.vertex_groups:
            vg_lower = vg.name.lower()
            if any(k in vg_lower for k in ["righthand", "hand_r", "hand.r"]):
                r_name = vg.name
            elif any(k in vg_lower for k in ["lefthand", "hand_l", "hand.l"]):
                l_name = vg.name
    return r_name, l_name

# ==============================================================================
# RIG-PARENTED CAMERA RIG SETUP (BEST PRACTICE)
# ==============================================================================

def setup_modern_reel_camera(target_rig, target_mesh, turn_start, turn_end_frame, unfurl_end_frame):
    scene = bpy.context.scene
    scene.render.resolution_x = 1080
    scene.render.resolution_y = 1920
    scene.render.resolution_percentage = 100

    # 1. Setup Camera and Focus Target
    focus_target = bpy.data.objects.get("Camera_Focus_Target") or bpy.data.objects.new("Camera_Focus_Target", None)
    if focus_target.name not in scene.collection.objects:
        scene.collection.objects.link(focus_target)

    cam_data = bpy.data.cameras.get("Vertical_Reel_Cam_Data") or bpy.data.cameras.new("Vertical_Reel_Cam_Data")
    cam_data.lens = 45.0
    cam_data.sensor_fit = 'VERTICAL'

    cam_obj = bpy.data.objects.get("Vertical_Reel_Camera") or bpy.data.objects.new("Vertical_Reel_Camera", cam_data)
    if cam_obj.name not in scene.collection.objects:
        scene.collection.objects.link(cam_obj)

    scene.camera = cam_obj
    cam_obj.parent = None

    cam_obj.constraints.clear()
    track = cam_obj.constraints.new(type='TRACK_TO')
    track.target = focus_target
    track.track_axis = 'TRACK_NEGATIVE_Z'
    track.up_axis = 'UP_Y'

    if cam_obj.animation_data:
        cam_obj.animation_data_clear()
    if focus_target.animation_data:
        focus_target.animation_data_clear()

    # 2. Calculate Maximum Deformed Extent Across Sample Frames
    max_radius = 0.0
    sample_frames = list(range(1, unfurl_end_frame + 1, 5))
    if unfurl_end_frame not in sample_frames:
        sample_frames.append(unfurl_end_frame)

    depsgraph = bpy.context.evaluated_depsgraph_get()

    for f in sample_frames:
        scene.frame_set(f)
        depsgraph.update()

        # Evaluate the mesh WITH armature modifiers active
        eval_mesh_obj = target_mesh.evaluated_get(depsgraph)
        me = eval_mesh_obj.to_mesh()  # Gets actual deformed geometry
        matrix = eval_mesh_obj.matrix_world

        # Find world-space vertices for radius calculation
        world_verts = [matrix @ v.co for v in me.vertices]
        eval_mesh_obj.to_mesh_clear()  # Free memory

        if not world_verts:
            continue

        # Find center of world-space bounds
        min_x = min(v.x for v in world_verts)
        max_x = max(v.x for v in world_verts)
        min_y = min(v.y for v in world_verts)
        max_y = max(v.y for v in world_verts)
        min_z = min(v.z for v in world_verts)
        max_z = max(v.z for v in world_verts)

        center = Vector(((min_x + max_x) * 0.5, (min_y + max_y) * 0.5, (min_z + max_z) * 0.5))

        # Radius to furthest vertex from frame center
        for v in world_verts:
            dist = (v - center).length
            if dist > max_radius:
                max_radius = dist

    # Safety margin (15% padding)
    padded_radius = max_radius * 1.15

    # 3. Trigonometric Distance for Planar FOV (Vertical 9:16 Aspect Ratio)
    sensor_width = cam_data.sensor_width
    sensor_height = cam_data.sensor_height
    focal_length = cam_data.lens

    fov_v = 2.0 * math.atan((sensor_height * 0.5) / focal_length)
    aspect_ratio = scene.render.resolution_x / scene.render.resolution_y
    fov_h = 2.0 * math.atan(math.tan(fov_v * 0.5) * aspect_ratio)

    # Tighter FOV in 9:16 portrait mode is Horizontal (fov_h)
    min_fov = min(fov_v, fov_h)

    # Correct perspective camera distance formula
    required_dist = padded_radius / math.tan(min_fov * 0.5)

    root_bone_name = "mixamorig:Hips" if "mixamorig:Hips" in target_rig.pose.bones else target_rig.pose.bones[0].name

    # 4. Fast Keyframe Loop (Uses Bounding Box Corners for Speed)
    for f in range(1, unfurl_end_frame + 1):
        scene.frame_set(f)
        depsgraph.update()

        eval_mesh_obj = target_mesh.evaluated_get(depsgraph)

        # Fast Z-bound calculation using 8 corners of evaluated bounding box
        bbox_world = [eval_mesh_obj.matrix_world @ Vector(corner) for corner in eval_mesh_obj.bound_box]
        min_z = min(v.z for v in bbox_world)
        max_z = max(v.z for v in bbox_world)
        target_mid_z = min_z + ((max_z - min_z) * 0.50)

        eval_rig = target_rig.evaluated_get(depsgraph)
        pbone = eval_rig.pose.bones[root_bone_name]
        bone_world_matrix = eval_rig.matrix_world @ pbone.matrix

        char_ground_loc = Vector((bone_world_matrix.translation.x, bone_world_matrix.translation.y, min_z))

        focus_target.location = Vector((char_ground_loc.x, char_ground_loc.y, target_mid_z))
        focus_target.keyframe_insert(data_path="location", frame=f)

        rig_matrix = eval_rig.matrix_world
        forward_vec = (rig_matrix.to_3x3() @ Vector((0, 1, 0))).xy.normalized()
        if forward_vec.length == 0:
            forward_vec = Vector((0, -1))

        base_angle = math.atan2(forward_vec.y, forward_vec.x)

        if f <= turn_start:
            angle_offset = math.radians(-45.0)
        elif f >= turn_end_frame:
            angle_offset = 0.0
        else:
            t = (f - turn_start) / (turn_end_frame - turn_start)
            angle_offset = math.radians(-45.0) * (1.0 - t)

        cam_angle = base_angle + angle_offset

        cam_x = char_ground_loc.x - (required_dist * math.cos(cam_angle))
        cam_y = char_ground_loc.y - (required_dist * math.sin(cam_angle))

        cam_obj.location = Vector((cam_x, cam_y, target_mid_z))
        cam_obj.keyframe_insert(data_path="location", frame=f)

    # 5. Set Linear Interpolation across Blender 3.x and 4.x+
    for obj in [cam_obj, focus_target]:
        if obj and obj.animation_data and obj.animation_data.action:
            action = obj.animation_data.action
            fcurves = getattr(action, "fcurves", None)
            if fcurves is None and hasattr(action, "bindings"):
                fcurves = [fc for b in action.bindings for fc in b.fcurves if hasattr(b, "fcurves")]
            if fcurves:
                for fc in fcurves:
                    for kp in fc.keyframe_points:
                        kp.interpolation = 'LINEAR'

    # 6. Set NLA Extrapolation to HOLD so character remains still post frame 225
    for obj in [target_rig, target_mesh]:
        if obj and obj.animation_data and obj.animation_data.nla_tracks:
            for track in obj.animation_data.nla_tracks:
                for strip in track.strips:
                    strip.extrapolation = 'HOLD'

    scene.frame_set(1)
    depsgraph.update()

    return cam_obj, focus_target


def frame_scroll_image(cam_obj, focus_target, target_mesh, image_path, scroll_width=0.35, zoom_duration=45):
    scene = bpy.context.scene
    unfurl_end_frame = scene.frame_end
    zoom_end_frame = unfurl_end_frame + zoom_duration

    scene.frame_set(unfurl_end_frame)
    bpy.context.view_layer.update()

    img_aspect = 1.0
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        if img.size[1] > 0:
            img_aspect = img.size[0] / img.size[1]

    tile_height = scroll_width * img_aspect
    cam_data = cam_obj.data
    aspect_ratio = scene.render.resolution_x / scene.render.resolution_y
    sensor_width = cam_data.sensor_height * aspect_ratio

    fov_h_half = math.atan((sensor_width / 2.0) / cam_data.lens)
    target_framing_distance = (tile_height / 2.0) / math.tan(fov_h_half)

    mesh_matrix = target_mesh.matrix_world
    depsgraph = bpy.context.evaluated_depsgraph_get()
    eval_mesh = target_mesh.evaluated_get(depsgraph)

    r_hand_name, l_hand_name = find_hand_vertex_groups(target_mesh)
    vg_r = eval_mesh.vertex_groups.get(r_hand_name)
    vg_l = eval_mesh.vertex_groups.get(l_hand_name)

    r_verts = [mesh_matrix @ v.co for v in eval_mesh.data.vertices if any(g.group == vg_r.index for g in v.groups)] if vg_r else []
    l_verts = [mesh_matrix @ v.co for v in eval_mesh.data.vertices if any(g.group == vg_l.index for g in v.groups)] if vg_l else []

    target_height = target_mesh.dimensions.z if target_mesh.dimensions.z > 0 else 1.8
    torso_fallback = mesh_matrix.translation + Vector((0, 0, target_height * 0.55))

    r_hand_pos = sum(r_verts, Vector((0, 0, 0))) / len(r_verts) if r_verts else torso_fallback
    l_hand_pos = sum(l_verts, Vector((0, 0, 0))) / len(l_verts) if l_verts else torso_fallback

    scroll_midpoint = (r_hand_pos + l_hand_pos) * 0.5

    cam_pos = cam_obj.matrix_world.translation.copy()
    cam_dir = cam_pos - scroll_midpoint
    cam_dir.z = 0.0

    scroll_normal = cam_dir.normalized() if cam_dir.length > 0.001 else Vector((0, -1, 0))

    tile_center = scroll_midpoint + (scroll_normal * 0.10)
    end_pos = tile_center + (scroll_normal * target_framing_distance)

    # Detach to free world space for precise final zoom pass
    cam_obj.parent = None
    focus_target.parent = None

    current_world_focus = focus_target.matrix_world.translation.copy()
    focus_target.matrix_world.translation = current_world_focus

    focus_target.location = current_world_focus
    focus_target.keyframe_insert(data_path="location", frame=unfurl_end_frame)
    focus_target.location = tile_center
    focus_target.keyframe_insert(data_path="location", frame=zoom_end_frame)

    cam_obj.location = cam_pos
    cam_obj.keyframe_insert(data_path="location", frame=unfurl_end_frame)
    cam_obj.location = end_pos
    cam_obj.keyframe_insert(data_path="location", frame=zoom_end_frame)

    for obj in [cam_obj, focus_target]:
        for fc in get_object_fcurves(obj):
            for kp in fc.keyframe_points:
                if kp.co.x >= unfurl_end_frame:
                    kp.interpolation = 'BEZIER'
                    kp.handle_left_type = 'AUTO_CLAMPED'
                    kp.handle_right_type = 'AUTO_CLAMPED'

    scene.frame_end = zoom_end_frame
    return zoom_end_frame

# ==============================================================================
# GEOMETRY NODES GROUND, GRASS & SKY DOME SETUP
# ==============================================================================

def ensure_world_sphere_size(cam_obj, ground_obj, safety_multiplier=2.5):
    if not cam_obj or not ground_obj:
        return

    scene = bpy.context.scene
    max_cam_dist = 0.0

    for frame in range(scene.frame_start, scene.frame_end + 1, 5):
        scene.frame_set(frame)
        bpy.context.view_layer.update()
        max_cam_dist = max(max_cam_dist, cam_obj.matrix_world.translation.length)

    world_radius = max(25.0, max_cam_dist * safety_multiplier)

    mod = ground_obj.modifiers.get("Grass_Scatter_GN")
    if mod and mod.node_group:
        for node in mod.node_group.nodes:
            if node.bl_idname == "GeometryNodeMeshIcoSphere":
                node.inputs["Radius"].default_value = world_radius

    if ground_obj.type == 'MESH':
        bpy.context.view_layer.objects.active = ground_obj
        bpy.ops.object.mode_set(mode='OBJECT')
        ground_scale = world_radius / 12.5
        ground_obj.scale = (ground_scale, ground_scale, 1.0)
        bpy.ops.object.transform_apply(location=False, rotation=False, scale=True)

    cam_obj.data.clip_end = max(cam_obj.data.clip_end, world_radius * 2.5)
    scene.frame_set(1)


def create_ground_material(ground_image_path=None):
    """Procedural soil material to eliminate flat green ground blobs."""
    mat = bpy.data.materials.new(name="Ground_Soil_Material")
    mat.use_nodes = True
    nodes, links = mat.node_tree.nodes, mat.node_tree.links
    nodes.clear()

    out_node = nodes.new(type="ShaderNodeOutputMaterial")
    bsdf = nodes.new(type="ShaderNodeBsdfPrincipled")  # Correct bl_idname

    if ground_image_path and os.path.exists(ground_image_path):
        tex_node = nodes.new(type="ShaderNodeTexImage")
        tex_node.image = bpy.data.images.load(ground_image_path)
        links.new(tex_node.outputs["Color"], bsdf.inputs["Base Color"])
    else:
        # Noise-driven mixture of dark earth soil and patchy olive green
        noise = nodes.new("ShaderNodeTexNoise")
        noise.inputs["Scale"].default_value = 18.0
        noise.inputs["Detail"].default_value = 6.0

        color_ramp = nodes.new("ShaderNodeValToRGB")
        # Color 0: Dark Earth Soil Brown
        color_ramp.color_ramp.elements[0].color = (0.04, 0.025, 0.01, 1.0)
        # Color 1: Patchy Olive Green
        color_ramp.color_ramp.elements[1].color = (0.05, 0.07, 0.02, 1.0)

        links.new(noise.outputs["Factor"], color_ramp.inputs["Factor"])
        links.new(color_ramp.outputs["Color"], bsdf.inputs["Base Color"])

    bsdf.inputs["Roughness"].default_value = 0.95
    links.new(bsdf.outputs["BSDF"], out_node.inputs["Surface"])
    return mat


def create_grass_instance_material():
    """Natural grass material with randomized per-instance color and translucency."""
    mat = bpy.data.materials.new(name="Grass_Blade_Material")
    mat.use_nodes = True
    nodes, links = mat.node_tree.nodes, mat.node_tree.links
    nodes.clear()

    out_node = nodes.new("ShaderNodeOutputMaterial")
    bsdf = nodes.new(type="ShaderNodeBsdfPrincipled")

    obj_info = nodes.new("ShaderNodeObjectInfo")
    color_ramp = nodes.new("ShaderNodeValToRGB")

    # Tone variations across grass instances
    color_ramp.color_ramp.elements[0].color = (0.02, 0.10, 0.01, 1.0)  # Deep Green
    color_ramp.color_ramp.elements[1].color = (0.08, 0.16, 0.02, 1.0)  # Bright Spring Green

    # Add dry yellow tip accent
    elem_dry = color_ramp.color_ramp.elements.new(0.85)
    elem_dry.color = (0.22, 0.18, 0.03, 1.0)

    links.new(obj_info.outputs["Random"], color_ramp.inputs["Factor"])
    links.new(color_ramp.outputs["Color"], bsdf.inputs["Base Color"])

    if "Subsurface Weight" in bsdf.inputs:
        # Blender 4.0+ API
        bsdf.inputs["Subsurface Weight"].default_value = 0.3
        # Subsurface Color was removed in 4.0; tint comes from Base Color / Subsurface Radius
        if "Subsurface Color" in bsdf.inputs:
            bsdf.inputs["Subsurface Color"].default_value = (0.1, 0.35, 0.02, 1.0)
    else:
        # Blender 3.x legacy API fallback
        if "Subsurface" in bsdf.inputs:
            bsdf.inputs["Subsurface"].default_value = 0.3
        if "Subsurface Color" in bsdf.inputs:
            bsdf.inputs["Subsurface Color"].default_value = (0.1, 0.35, 0.02, 1.0)

    bsdf.inputs["Roughness"].default_value = 0.45
    links.new(bsdf.outputs["BSDF"], out_node.inputs["Surface"])
    return mat


def create_sky_material(sky_hdri_path):
    mat = bpy.data.materials.new(name="Sky_Dome_Material")
    mat.use_nodes = True
    nodes, links = mat.node_tree.nodes, mat.node_tree.links
    nodes.clear()

    out_node = nodes.new("ShaderNodeOutputMaterial")
    emission = nodes.new("ShaderNodeEmission")
    emission.inputs["Strength"].default_value = 1.0

    transparent = nodes.new("ShaderNodeBsdfTransparent")
    mix_shader = nodes.new("ShaderNodeMixShader")
    light_path = nodes.new("ShaderNodeLightPath")

    # Is Camera Ray -> Fac (Camera sees Emission, all other rays see Transparent)
    links.new(light_path.outputs["Is Camera Ray"], mix_shader.inputs["Fac"])
    links.new(transparent.outputs["BSDF"], mix_shader.inputs[1])
    links.new(emission.outputs["Emission"], mix_shader.inputs[2])
    links.new(mix_shader.outputs["Shader"], out_node.inputs["Surface"])

    if os.path.exists(sky_hdri_path):
        tex_env = nodes.new("ShaderNodeTexEnvironment")
        tex_env.image = bpy.data.images.load(sky_hdri_path)
        links.new(tex_env.outputs["Color"], emission.inputs["Color"])

    return mat


def setup_grass_ground_geometry_nodes(grass_dir, sky_hdri_path, ground_z_offset=-0.2):
    """Creates ground mesh, instances grass with natural density/rotations, and sets up sky dome."""
    # Create ground plane
    bpy.ops.mesh.primitive_plane_add(size=60, location=(0, 0, ground_z_offset))
    ground_obj = bpy.context.active_object
    ground_obj.name = "Grass_Ground_Plane"

    # Assign Ground Base Material & Sky Material
    ground_mat = create_ground_material()
    sky_mat = create_sky_material(sky_hdri_path)
    ground_obj.data.materials.append(ground_mat)
    ground_obj.data.materials.append(sky_mat)

    # Assign Grass Material
    grass_mat = create_grass_instance_material()

    # Create grass instance object
    bpy.ops.mesh.primitive_cone_add(radius1=0.05, depth=0.4, location=(0, 0, -10))
    grass_instance_obj = bpy.context.active_object
    grass_instance_obj.name = "Grass_Blade_Instance"
    grass_instance_obj.data.materials.append(grass_mat)

    # Setup Modifier & Geometry Node Tree
    gn_mod = ground_obj.modifiers.new(name="Grass_Scatter_GN", type='NODES')
    gn_group = bpy.data.node_groups.new("Grass_Scatter_Group", type='GeometryNodeTree')
    gn_mod.node_group = gn_group

    nodes = gn_group.nodes
    links = gn_group.links
    nodes.clear()

    # Add IO
    group_in = nodes.new("NodeGroupInput")
    gn_group.interface.new_socket("Geometry", in_out='INPUT', socket_type='NodeSocketGeometry')
    group_out = nodes.new("NodeGroupOutput")
    gn_group.interface.new_socket("Geometry", in_out='OUTPUT', socket_type='NodeSocketGeometry')

    # Distribute Points (Poisson Disk)
    distribute = nodes.new("GeometryNodeDistributePointsOnFaces")
    distribute.distribute_method = 'POISSON'

    # Safe Socket Access via socket name search
    if "Distance Min" in distribute.inputs:
        distribute.inputs["Distance Min"].default_value = 0.08
    if "Density Factor" in distribute.inputs:
        distribute.inputs["Density Factor"].default_value = 150.0
    elif "Density" in distribute.inputs:
        distribute.inputs["Density"].default_value = 150.0

    # Instance Grass
    instance_node = nodes.new("GeometryNodeInstanceOnPoints")
    obj_info = nodes.new("GeometryNodeObjectInfo")
    obj_info.inputs["Object"].default_value = grass_instance_obj

    # Random Rotation (Z-axis 0 to 2pi)
    rand_rot = nodes.new("FunctionNodeRandomValue")
    rand_rot.data_type = 'FLOAT_VECTOR'

    rand_rot.inputs["Min"].default_value = (0.0, 0.0, 0.0)
    rand_rot.inputs["Max"].default_value = (0.1, 0.1, 2.0 * math.pi)

    # Random Scale (0.7 to 1.2 across axes)
    rand_scale = nodes.new("FunctionNodeRandomValue")
    rand_scale.data_type = 'FLOAT_VECTOR'
    rand_scale.inputs["Min"].default_value = (0.7, 0.7, 0.6)
    rand_scale.inputs["Max"].default_value = (1.2, 1.2, 1.3)

    # Dynamically resolve valid output sockets across Blender versions
    rot_out = rand_rot.outputs.get("Value") or rand_rot.outputs.get("Vector") or rand_rot.outputs[0]
    scale_out = rand_scale.outputs.get("Value") or rand_scale.outputs.get("Vector") or rand_scale.outputs[0]

    links.new(rot_out, instance_node.inputs["Rotation"])
    links.new(scale_out, instance_node.inputs["Scale"])
    links.new(group_in.outputs["Geometry"], distribute.inputs["Mesh"])
    links.new(distribute.outputs["Points"], instance_node.inputs["Points"])
    links.new(obj_info.outputs["Geometry"], instance_node.inputs["Instance"])

    # Sky Dome Mesh
    sky_sphere = nodes.new("GeometryNodeMeshIcoSphere")
    sky_sphere.inputs["Radius"].default_value = 12.5
    sky_sphere.inputs["Subdivisions"].default_value = 4

    flip_faces = nodes.new("GeometryNodeFlipFaces")
    links.new(sky_sphere.outputs["Mesh"], flip_faces.inputs["Mesh"])

    transform_sky = nodes.new("GeometryNodeTransform")
    transform_sky.inputs["Translation"].default_value = (0.0, 0.0, ground_z_offset)
    links.new(flip_faces.outputs["Mesh"], transform_sky.inputs["Geometry"])

    set_sky_mat = nodes.new("GeometryNodeSetMaterial")
    set_sky_mat.inputs["Material"].default_value = sky_mat
    links.new(transform_sky.outputs["Geometry"], set_sky_mat.inputs["Geometry"])

    # Join Ground, Grass Instances, and Sky Dome
    join_geo = nodes.new("GeometryNodeJoinGeometry")
    links.new(group_in.outputs["Geometry"], join_geo.inputs["Geometry"])
    links.new(instance_node.outputs["Instances"], join_geo.inputs["Geometry"])
    links.new(set_sky_mat.outputs["Geometry"], join_geo.inputs["Geometry"])

    links.new(join_geo.outputs["Geometry"], group_out.inputs["Geometry"])

    return ground_obj

# ==============================================================================
# SCROLL MATERIAL & GEOMETRY NODES SETUP
# ==============================================================================

def create_image_scroll_material(image_path):
    mat_name = "Scroll_Texture_Mat"
    if mat := bpy.data.materials.get(mat_name):
        bpy.data.materials.remove(mat)

    mat = bpy.data.materials.new(name=mat_name)
    mat.use_nodes = True
    nodes, links = mat.node_tree.nodes, mat.node_tree.links
    nodes.clear()

    output_node = nodes.new("ShaderNodeOutputMaterial")
    bsdf = nodes.new("ShaderNodeBsdfPrincipled")
    bsdf.inputs["Roughness"].default_value = 0.5

    tex_img = nodes.new("ShaderNodeTexImage")
    if os.path.exists(image_path):
        tex_img.image = bpy.data.images.load(image_path)
        tex_img.extension = 'REPEAT'

    uv_map_node = nodes.new("ShaderNodeUVMap")
    uv_map_node.uv_map = "UVMap"

    links.new(uv_map_node.outputs["UV"], tex_img.inputs["Vector"])
    links.new(tex_img.outputs["Color"], bsdf.inputs["Base Color"])
    links.new(bsdf.outputs["BSDF"], output_node.inputs["Surface"])

    return mat, tex_img.image if os.path.exists(image_path) else None


def create_pure_geometry_node_scroll(image_path=""):
    gn_group = bpy.data.node_groups.new("Scroll_Pure_GN", type="GeometryNodeTree")

    gn_group.interface.new_socket(name="Geometry", in_out="OUTPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Character Mesh", in_out="INPUT", socket_type="NodeSocketObject")

    sock_r_vg = gn_group.interface.new_socket(name="Right Hand Group", in_out="INPUT", socket_type="NodeSocketString")
    sock_r_vg.default_value = "mixamorig:RightHand"

    sock_l_vg = gn_group.interface.new_socket(name="Left Hand Group", in_out="INPUT", socket_type="NodeSocketString")
    sock_l_vg.default_value = "mixamorig:LeftHand"

    sock_offset = gn_group.interface.new_socket(name="Forward Offset", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_offset.default_value = 0.60

    sock_unfurl = gn_group.interface.new_socket(name="Unfurl", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_unfurl.default_value = 0.0

    sock_width = gn_group.interface.new_socket(name="Scroll Width", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_width.default_value = 0.35

    nodes, links = gn_group.nodes, gn_group.links
    in_node, out_node = nodes.new("NodeGroupInput"), nodes.new("NodeGroupOutput")

    char_info = nodes.new("GeometryNodeObjectInfo")
    char_info.transform_space = "RELATIVE"
    links.new(in_node.outputs["Character Mesh"], char_info.inputs["Object"])

    pos_node = nodes.new("GeometryNodeInputPosition")

    named_r = nodes.new("GeometryNodeInputNamedAttribute")
    named_r.data_type = "FLOAT"
    links.new(in_node.outputs["Right Hand Group"], named_r.inputs["Name"])

    cmp_r = nodes.new("FunctionNodeCompare")
    cmp_r.data_type = "FLOAT"
    cmp_r.operation = "GREATER_THAN"
    cmp_r.inputs[1].default_value = 0.001
    links.new(named_r.outputs["Attribute"], cmp_r.inputs[0])

    stat_r = nodes.new("GeometryNodeAttributeStatistic")
    stat_r.data_type = "FLOAT_VECTOR"
    links.new(char_info.outputs["Geometry"], stat_r.inputs["Geometry"])
    links.new(cmp_r.outputs["Result"], stat_r.inputs["Selection"])
    links.new(pos_node.outputs["Position"], stat_r.inputs["Attribute"])

    named_l = nodes.new("GeometryNodeInputNamedAttribute")
    named_l.data_type = "FLOAT"
    links.new(in_node.outputs["Left Hand Group"], named_l.inputs["Name"])

    cmp_l = nodes.new("FunctionNodeCompare")
    cmp_l.data_type = "FLOAT"
    cmp_l.operation = "GREATER_THAN"
    cmp_l.inputs[1].default_value = 0.001
    links.new(named_l.outputs["Attribute"], cmp_l.inputs[0])

    stat_l = nodes.new("GeometryNodeAttributeStatistic")
    stat_l.data_type = "FLOAT_VECTOR"
    links.new(char_info.outputs["Geometry"], stat_l.inputs["Geometry"])
    links.new(cmp_l.outputs["Result"], stat_l.inputs["Selection"])
    links.new(pos_node.outputs["Position"], stat_l.inputs["Attribute"])

    neg_offset = nodes.new("ShaderNodeMath")
    neg_offset.operation = "MULTIPLY"
    neg_offset.inputs[1].default_value = -1.0
    links.new(in_node.outputs["Forward Offset"], neg_offset.inputs[0])

    forward_vec = nodes.new("ShaderNodeCombineXYZ")
    forward_vec.inputs["X"].default_value = 0.0
    links.new(neg_offset.outputs["Value"], forward_vec.inputs["Y"])
    forward_vec.inputs["Z"].default_value = 0.0

    start_pos = nodes.new("ShaderNodeVectorMath")
    start_pos.operation = "ADD"
    links.new(stat_r.outputs["Mean"], start_pos.inputs[0])
    links.new(forward_vec.outputs["Vector"], start_pos.inputs[1])

    target_pos = nodes.new("ShaderNodeVectorMath")
    target_pos.operation = "ADD"
    links.new(stat_l.outputs["Mean"], target_pos.inputs[0])
    links.new(forward_vec.outputs["Vector"], target_pos.inputs[1])

    delta_vec = nodes.new("ShaderNodeVectorMath")
    delta_vec.operation = "SUBTRACT"
    links.new(target_pos.outputs["Vector"], delta_vec.inputs[0])
    links.new(start_pos.outputs["Vector"], delta_vec.inputs[1])

    dist_node = nodes.new("ShaderNodeVectorMath")
    dist_node.operation = "DISTANCE"
    links.new(start_pos.outputs["Vector"], dist_node.inputs[0])

    scale_vec = nodes.new("ShaderNodeVectorMath")
    scale_vec.operation = "SCALE"
    links.new(delta_vec.outputs["Vector"], scale_vec.inputs["Vector"])
    links.new(in_node.outputs["Unfurl"], scale_vec.inputs["Scale"])

    end_pos = nodes.new("ShaderNodeVectorMath")
    end_pos.operation = "ADD"
    links.new(start_pos.outputs["Vector"], end_pos.inputs[0])
    links.new(scale_vec.outputs["Vector"], end_pos.inputs[1])
    links.new(end_pos.outputs["Vector"], dist_node.inputs[1])

    curve_line = nodes.new("GeometryNodeCurvePrimitiveLine")
    links.new(start_pos.outputs["Vector"], curve_line.inputs["Start"])
    links.new(end_pos.outputs["Vector"], curve_line.inputs["End"])

    set_tilt = nodes.new("GeometryNodeSetCurveTilt")
    set_tilt.inputs["Tilt"].default_value = 1.5708
    links.new(curve_line.outputs["Curve"], set_tilt.inputs["Curve"])

    spline_param_main = nodes.new("GeometryNodeSplineParameter")
    store_u = nodes.new("GeometryNodeStoreNamedAttribute")
    store_u.inputs["Name"].default_value = "U_Factor"
    store_u.data_type = 'FLOAT'
    store_u.domain = 'POINT'
    links.new(set_tilt.outputs["Curve"], store_u.inputs["Geometry"])
    links.new(spline_param_main.outputs["Factor"], store_u.inputs["Value"])

    half_width_neg = nodes.new("ShaderNodeMath")
    half_width_neg.operation = "MULTIPLY"
    half_width_neg.inputs[1].default_value = -0.5
    links.new(in_node.outputs["Scroll Width"], half_width_neg.inputs[0])

    half_width_pos = nodes.new("ShaderNodeMath")
    half_width_pos.operation = "MULTIPLY"
    half_width_pos.inputs[1].default_value = 0.5
    links.new(in_node.outputs["Scroll Width"], half_width_pos.inputs[0])

    combine_start = nodes.new("ShaderNodeCombineXYZ")
    links.new(half_width_neg.outputs["Value"], combine_start.inputs["X"])

    combine_end = nodes.new("ShaderNodeCombineXYZ")
    links.new(half_width_pos.outputs["Value"], combine_end.inputs["X"])

    profile_line = nodes.new("GeometryNodeCurvePrimitiveLine")
    links.new(combine_start.outputs["Vector"], profile_line.inputs["Start"])
    links.new(combine_end.outputs["Vector"], profile_line.inputs["End"])

    spline_param_prof = nodes.new("GeometryNodeSplineParameter")
    store_v = nodes.new("GeometryNodeStoreNamedAttribute")
    store_v.inputs["Name"].default_value = "V_Factor"
    store_v.data_type = 'FLOAT'
    store_v.domain = 'POINT'
    links.new(profile_line.outputs["Curve"], store_v.inputs["Geometry"])
    links.new(spline_param_prof.outputs["Factor"], store_v.inputs["Value"])

    curve_to_mesh = nodes.new("GeometryNodeCurveToMesh")
    links.new(store_u.outputs["Geometry"], curve_to_mesh.inputs["Curve"])
    links.new(store_v.outputs["Geometry"], curve_to_mesh.inputs["Profile Curve"])

    img_aspect = 1.0
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        if img.size[1] > 0:
            img_aspect = img.size[0] / img.size[1]

    h_times_aspect = nodes.new("ShaderNodeMath")
    h_times_aspect.operation = "MULTIPLY"
    h_times_aspect.inputs[1].default_value = img_aspect
    links.new(in_node.outputs["Scroll Width"], h_times_aspect.inputs[0])

    tile_count = nodes.new("ShaderNodeMath")
    tile_count.operation = "DIVIDE"
    links.new(dist_node.outputs["Value"], tile_count.inputs[0])
    links.new(h_times_aspect.outputs["Value"], tile_count.inputs[1])

    read_u = nodes.new("GeometryNodeInputNamedAttribute")
    read_u.inputs["Name"].default_value = "U_Factor"
    read_u.data_type = 'FLOAT'

    scaled_u = nodes.new("ShaderNodeMath")
    scaled_u.operation = "MULTIPLY"
    links.new(read_u.outputs["Attribute"], scaled_u.inputs[0])
    links.new(tile_count.outputs["Value"], scaled_u.inputs[1])

    read_v = nodes.new("GeometryNodeInputNamedAttribute")
    read_v.inputs["Name"].default_value = "V_Factor"
    read_v.data_type = 'FLOAT'

    flip_v = nodes.new("ShaderNodeMath")
    flip_v.operation = "SUBTRACT"
    flip_v.inputs[0].default_value = 1.0
    links.new(read_v.outputs["Attribute"], flip_v.inputs[1])

    combine_uv = nodes.new("ShaderNodeCombineXYZ")
    links.new(scaled_u.outputs["Value"], combine_uv.inputs["X"])
    links.new(flip_v.outputs["Value"], combine_uv.inputs["Y"])

    store_uv = nodes.new("GeometryNodeStoreNamedAttribute")
    store_uv.inputs["Name"].default_value = "UVMap"
    store_uv.data_type = 'FLOAT2'
    store_uv.domain = 'CORNER'
    links.new(curve_to_mesh.outputs["Mesh"], store_uv.inputs["Geometry"])
    links.new(combine_uv.outputs["Vector"], store_uv.inputs["Value"])

    set_material = nodes.new("GeometryNodeSetMaterial")
    scroll_mat, _ = create_image_scroll_material(image_path)
    set_material.inputs["Material"].default_value = scroll_mat
    links.new(store_uv.outputs["Geometry"], set_material.inputs["Geometry"])

    links.new(set_material.outputs["Geometry"], out_node.inputs["Geometry"])

    return gn_group, scroll_mat


def setup_geometry_node_scroll(target_mesh, unfurl_start, unfurl_end, image_path=""):
    mesh_data = bpy.data.meshes.new("Scroll_Procedural_Mesh")
    scroll_obj = bpy.data.objects.new("Scroll_GN_Object", mesh_data)
    bpy.context.scene.collection.objects.link(scroll_obj)

    mod = scroll_obj.modifiers.new("Scroll_GN", type="NODES")
    gn_tree, scroll_mat = create_pure_geometry_node_scroll(image_path=image_path)
    mod.node_group = gn_tree

    scroll_obj.data.materials.append(scroll_mat)

    r_hand_name, l_hand_name = find_hand_vertex_groups(target_mesh)

    unfurl_id = None
    for item in gn_tree.interface.items_tree:
        if getattr(item, "in_out", None) == "INPUT":
            if item.name == "Character Mesh":
                mod[item.identifier] = target_mesh
            elif item.name == "Right Hand Group":
                mod[item.identifier] = r_hand_name
            elif item.name == "Left Hand Group":
                mod[item.identifier] = l_hand_name
            elif item.name == "Forward Offset":
                mod[item.identifier] = 0.60
            elif item.name == "Scroll Width":
                mod[item.identifier] = 0.35
            elif item.name == "Unfurl":
                unfurl_id = item.identifier

    scroll_obj.hide_viewport = True
    scroll_obj.hide_render = True
    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=1)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=1)

    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=unfurl_start - 1)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=unfurl_start - 1)

    scroll_obj.hide_viewport = False
    scroll_obj.hide_render = False
    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=unfurl_start)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=unfurl_start)

    for fc in get_object_fcurves(scroll_obj):
        if fc.data_path in ["hide_viewport", "hide_render"]:
            for kp in fc.keyframe_points:
                kp.interpolation = 'CONSTANT'

    if unfurl_id:
        mod[unfurl_id] = 0.001
        mod.keyframe_insert(data_path=f'["{unfurl_id}"]', frame=unfurl_start)

        mod[unfurl_id] = 1.0
        mod.keyframe_insert(data_path=f'["{unfurl_id}"]', frame=unfurl_end)

    return scroll_obj

# ==============================================================================
# MAIN CONTROLLER
# ==============================================================================

def import_and_extract_action(fbx_path, action_name):
    existing_objects = set(bpy.data.objects)
    bpy.ops.import_scene.fbx(filepath=fbx_path)
    bpy.context.view_layer.update()

    imported_objects = set(bpy.data.objects) - existing_objects
    extracted_action = None
    for obj in imported_objects:
        if obj.animation_data and obj.animation_data.action:
            extracted_action = obj.animation_data.action
            extracted_action.name = action_name
            extracted_action.use_fake_user = True
            break
    return extracted_action, imported_objects


def purge_imported_objects(objects_to_remove):
    for obj in objects_to_remove:
        if obj.name in bpy.data.objects:
            if obj.animation_data:
                obj.animation_data_clear()
            for coll in list(obj.users_collection):
                coll.objects.unlink(obj)
    bpy.context.view_layer.update()


def setup_world_hdri_lighting(sky_hdri_path, strength=2.5):
    scene = bpy.context.scene
    if not scene.world:
        scene.world = bpy.data.worlds.new("World")

    scene.world.use_nodes = True
    nodes = scene.world.node_tree.nodes
    links = scene.world.node_tree.links
    nodes.clear()

    out_node = nodes.new("ShaderNodeOutputWorld")
    bg_node = nodes.new("ShaderNodeBackground")
    bg_node.inputs["Strength"].default_value = strength

    links.new(bg_node.outputs["Background"], out_node.inputs["Surface"])

    if os.path.exists(sky_hdri_path):
        tex_env = nodes.new("ShaderNodeTexEnvironment")
        tex_env.image = bpy.data.images.load(sky_hdri_path)
        links.new(tex_env.outputs["Color"], bg_node.inputs["Color"])


def main():
    cleanup_previous_run()

    walk_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Mutant Walking.fbx"
    turn_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Right Turn 90.fbx"
    unfurl_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Mutant Idle.fbx"

    sky_hdri_path = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/horn-koppe_spring_4k.exr"
    output_blend = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/output_scene.blend"

    image_path = "/Users/dobbinp1/Downloads/sea_PanoramaIMGP5963.jpg"
    grass_dir = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Grass004_4K-JPG"

    # --------------------------------------------------------------------------
    # COLOR MANAGEMENT & WORLD LIGHTING FIXES
    # --------------------------------------------------------------------------
    scene = bpy.context.scene
    scene.view_settings.exposure = 1.0        # Lift overall scene exposure
    scene.view_settings.view_transform = 'AgX' # 'AgX' or 'Filmic' for balanced shadows
    scene.view_settings.look = 'AgX - Medium High Contrast'

    # Set world HDRI for scene illumination
    setup_world_hdri_lighting(sky_hdri_path, strength=1.0)

    # Setup geometry node grass ground and dome
    ground_obj = setup_grass_ground_geometry_nodes(grass_dir, sky_hdri_path, ground_z_offset=-0.2)

    CYCLES = 4
    CROSSFADE = 15
    TARGET_DISTANCE = 4.0

    walk_action, walk_objs = import_and_extract_action(walk_fbx, "walk")
    target_rig = next((obj for obj in walk_objs if obj.type == "ARMATURE"), None)
    target_mesh = next((obj for obj in walk_objs if obj.type == "MESH"), None)

    if not target_rig or not target_mesh:
        raise RuntimeError("Armature or Character Mesh not found in walk FBX.")

    travel_axis_idx, motion_sign = detect_natural_forward_axis(walk_action, target_rig)
    initial_heading = math.radians(90.0) if travel_axis_idx == 0 else 0.0
    target_rig.rotation_euler[2] = initial_heading

    action_len = max(30, int(walk_action.frame_range[1] - walk_action.frame_range[0]))
    anim_data = target_rig.animation_data_create()

    walk_track = anim_data.nla_tracks.new()
    walk_track.name = "Walk_Track"

    current_frame = 1
    for i in range(CYCLES):
        strip = walk_track.strips.new(name=f"Walk_{i+1}", start=int(current_frame), action=walk_action)
        strip.blend_type = "REPLACE"
        current_frame += action_len

    walk_end_frame = int(current_frame)

    turn_action, turn_objs = import_and_extract_action(turn_fbx, "rightturn")
    purge_imported_objects(turn_objs)

    turn_track = anim_data.nla_tracks.new()
    turn_track.name = "RightTurn_Track"

    turn_start = int(walk_end_frame - CROSSFADE)
    turn_strip = turn_track.strips.new(name="RightTurn", start=turn_start, action=turn_action)
    turn_strip.blend_type = "REPLACE"
    turn_strip.blend_in = CROSSFADE
    turn_strip.use_auto_blend = True
    turn_strip.extrapolation = "NOTHING"
    turn_end_frame = int(turn_strip.frame_end)

    bpy.context.scene["turn_start_frame"] = turn_start

    unfurl_action, unfurl_objs = import_and_extract_action(unfurl_fbx, "unfurl")
    purge_imported_objects(unfurl_objs)

    unfurl_track = anim_data.nla_tracks.new()
    unfurl_track.name = "Unfurl_Track"

    unfurl_start = int(turn_end_frame - CROSSFADE)
    unfurl_strip = unfurl_track.strips.new(name="Unfurl", start=unfurl_start, action=unfurl_action)
    unfurl_strip.blend_type = "REPLACE"
    unfurl_strip.blend_in = CROSSFADE
    unfurl_strip.use_auto_blend = True
    unfurl_strip.extrapolation = "HOLD"
    unfurl_end_frame = int(unfurl_strip.frame_end)

    root_action = bpy.data.actions.new(name="World_Root_Motion")
    target_rig.animation_data.action = root_action

    if hasattr(root_action, "fcurves") and hasattr(root_action.fcurves, "new"):
        loc_fcurve = root_action.fcurves.new(data_path="location", index=travel_axis_idx)
        rot_fcurve = root_action.fcurves.new(data_path="rotation_euler", index=2)
    else:
        loc_fcurve = root_action.fcurve_ensure_for_datablock(datablock=target_rig, data_path="location", index=travel_axis_idx)
        rot_fcurve = root_action.fcurve_ensure_for_datablock(datablock=target_rig, data_path="rotation_euler", index=2)

    final_target_value = TARGET_DISTANCE * motion_sign

    k_loc1 = loc_fcurve.keyframe_points.insert(frame=1.0, value=0.0)
    k_loc2 = loc_fcurve.keyframe_points.insert(frame=float(turn_start), value=final_target_value)
    k_loc3 = loc_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=final_target_value)

    k_loc1.interpolation = "LINEAR"
    k_loc2.interpolation = "CONSTANT"
    k_loc3.interpolation = "CONSTANT"

    target_heading = initial_heading - math.radians(90.0)

    k_rot1 = rot_fcurve.keyframe_points.insert(frame=1.0, value=initial_heading)
    k_rot2 = rot_fcurve.keyframe_points.insert(frame=float(turn_start), value=initial_heading)
    k_rot3 = rot_fcurve.keyframe_points.insert(frame=float(turn_end_frame), value=target_heading)
    k_rot4 = rot_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=target_heading)

    k_rot1.interpolation = "CONSTANT"
    k_rot2.interpolation = "BEZIER"
    k_rot3.interpolation = "CONSTANT"
    k_rot4.interpolation = "CONSTANT"

    root_track = anim_data.nla_tracks.new()
    root_track.name = "World_Root_Track"
    root_strip = root_track.strips.new(name="Root_Motion_Strip", start=1, action=root_action)
    root_strip.blend_type = "REPLACE"
    root_strip.extrapolation = "HOLD"

    anim_data.action = None

    setup_geometry_node_scroll(
        target_mesh=target_mesh,
        unfurl_start=unfurl_start,
        unfurl_end=unfurl_end_frame,
        image_path=image_path,
    )

    cam_obj, focus_target = setup_modern_reel_camera(
        target_rig=target_rig,
        target_mesh=target_mesh,
        turn_start=turn_start,
        turn_end_frame=turn_end_frame,
        unfurl_end_frame=unfurl_end_frame
    )

    bpy.context.scene.frame_start = 1
    bpy.context.scene.frame_end = unfurl_end_frame

    frame_scroll_image(
        cam_obj=cam_obj,
        focus_target=focus_target,
        target_mesh=target_mesh,
        image_path=image_path,
        scroll_width=0.35,
        zoom_duration=45
    )

    ensure_world_sphere_size(cam_obj, ground_obj, safety_multiplier=2.5)
    register_audio_handler()

    bpy.ops.wm.save_as_mainfile(filepath=output_blend)
    print(f"\n[SUCCESS] Integrated parented rig tracking & saved scene to: {output_blend}")


if __name__ == "__main__":
    main()
```

```
import math
import mathutils
import os
import sys
import glob
import bpy
import aud
from bpy_extras import anim_utils
from bpy_extras.object_utils import world_to_camera_view
from bpy.app.handlers import persistent
from mathutils import Vector

# ==============================================================================
# CONFIGURATION
# ==============================================================================
FOOTSTEP_SOUND_PATH = "/Users/dobbinp1/Downloads/stomp.wav"
GROUND_THRESHOLD = 0.24

# Setup Audio Device
if "_audio_store" not in sys.modules:
    import types
    sys.modules["_audio_store"] = types.ModuleType("_audio_store")

audio_store = sys.modules["_audio_store"]
if not hasattr(audio_store, "device"):
    audio_store.device = aud.Device()

AUDIO_DEVICE = audio_store.device
SOUND_FACTORY = None
AUDIO_HANDLES = []

if os.path.exists(FOOTSTEP_SOUND_PATH):
    SOUND_FACTORY = aud.Sound(FOOTSTEP_SOUND_PATH).cache()
    print("[AUDIO] Loaded sound file successfully.")
else:
    print(f"[AUDIO ERROR] File not found at: {FOOTSTEP_SOUND_PATH}")

_left_foot_was_down = False
_right_foot_was_down = False
_last_frame = -1


def play_footstep():
    global AUDIO_HANDLES
    if SOUND_FACTORY:
        handle = AUDIO_DEVICE.play(SOUND_FACTORY)
        AUDIO_HANDLES.append(handle)
        if len(AUDIO_HANDLES) > 10:
            AUDIO_HANDLES.pop(0)


def find_foot_bone(pose_bones, side="Left"):
    target_keywords = [
        f"mixamorig:{side}foot",
        f"{side.lower()}foot",
        f"foot.{side[0].lower()}",
        f"foot_{side[0].lower()}"
    ]
    for bone in pose_bones:
        b_name = bone.name.lower()
        if any(kw in b_name for kw in target_keywords):
            return bone
    return None


@persistent
def footstep_frame_handler(scene):
    global _left_foot_was_down, _right_foot_was_down, _last_frame

    try:
        current_frame = scene.frame_current
        turn_start = scene.get("turn_start_frame", None)

        if turn_start is not None and current_frame >= turn_start:
            _left_foot_was_down = False
            _right_foot_was_down = False
            return

        if current_frame <= _last_frame or abs(current_frame - _last_frame) > 10:
            _left_foot_was_down = False
            _right_foot_was_down = False
        _last_frame = current_frame

        rig = next((o for o in scene.objects if o.type == 'ARMATURE'), None)
        if not rig:
            return

        depsgraph = bpy.context.evaluated_depsgraph_get()
        eval_rig = rig.evaluated_get(depsgraph)

        left_bone = find_foot_bone(eval_rig.pose.bones, "Left")
        right_bone = find_foot_bone(eval_rig.pose.bones, "Right")

        if not left_bone or not right_bone:
            return

        l_z = (eval_rig.matrix_world @ left_bone.matrix).translation.z
        r_z = (eval_rig.matrix_world @ right_bone.matrix).translation.z

        l_is_down = l_z <= GROUND_THRESHOLD
        r_is_down = r_z <= GROUND_THRESHOLD

        if l_is_down and not _left_foot_was_down:
            play_footstep()

        if r_is_down and not _right_foot_was_down:
            play_footstep()

        _left_foot_was_down = l_is_down
        _right_foot_was_down = r_is_down

    except Exception:
        import traceback
        traceback.print_exc()


def register_audio_handler():
    for h in list(bpy.app.handlers.frame_change_post):
        if getattr(h, "__name__", "") == "footstep_frame_handler":
            bpy.app.handlers.frame_change_post.remove(h)

    bpy.app.handlers.frame_change_post.append(footstep_frame_handler)


register_audio_handler()

# ==============================================================================
# CLEANUP & HELPERS
# ==============================================================================

def cleanup_previous_run():
    target_names = [
        "Master_NLA_Control",
        "Scroll_GN_Object",
        "Scroll_Start_Pin",
        "Scroll_End_Pin",
        "Scroll_Master_Container",
        "HDRI_Ground_Plane",
        "Ground_Mesh_Object",
        "Camera_Focus_Target",
        "Camera_Anchor",
        "Vertical_Reel_Camera"
    ]
    for obj in list(bpy.data.objects):
        if obj.name in target_names or (obj.name.startswith("mixamorig") and obj.type not in {'ARMATURE', 'MESH'}):
            if obj.animation_data:
                obj.animation_data_clear()
            bpy.data.objects.remove(obj, do_unlink=True)

    for action in list(bpy.data.actions):
        if action.users == 0 or action.use_fake_user:
            action.use_fake_user = False
            bpy.data.actions.remove(action)

    for ng in list(bpy.data.node_groups):
        if ng.name in ["Scroll_Unfurl_GN", "Scroll_Pure_GN", "HDRI_Ground_GN", "Grass_Ground_GN"]:
            bpy.data.node_groups.remove(ng)

    for coll_name in ["Scroll_Master_Container", "Grass_Artifacts_Collection"]:
        if coll_name in bpy.data.collections:
            coll = bpy.data.collections[coll_name]
            for obj in list(coll.objects):
                bpy.data.objects.remove(obj, do_unlink=True)
            bpy.data.collections.remove(coll)

    bpy.context.view_layer.update()


def get_slot_fcurves(action, anim_target_object):
    fcurves = []
    if not action:
        return fcurves

    anim_data = getattr(anim_target_object, "animation_data", None)
    if anim_data and getattr(anim_data, "action_slot", None):
        slot = anim_data.action_slot
        channelbag = anim_utils.action_get_channelbag_for_slot(action, slot)
        if channelbag and hasattr(channelbag, "fcurves"):
            return list(channelbag.fcurves)

    if hasattr(action, "slots"):
        for slot in action.slots:
            for cb in getattr(slot, "channelbags", []):
                fcurves.extend(getattr(cb, "fcurves", []))
        if fcurves:
            return fcurves

    if hasattr(action, "fcurves"):
        return list(action.fcurves)

    return fcurves


def get_object_fcurves(obj):
    if not obj or not obj.animation_data or not obj.animation_data.action:
        return []
    return get_slot_fcurves(obj.animation_data.action, obj)


def detect_natural_forward_axis(action, rig_object):
    fcurves = get_slot_fcurves(action, rig_object)
    max_delta = 0.0
    best_axis = 1
    net_delta = 0.0

    start_frame, end_frame = action.frame_range

    for fc in fcurves:
        if fc.data_path.endswith("location") and fc.array_index in (0, 1, 2):
            delta = fc.evaluate(end_frame) - fc.evaluate(start_frame)
            if abs(delta) > abs(max_delta):
                max_delta = delta
                best_axis = fc.array_index
                net_delta = delta

    direction_sign = 1.0 if net_delta >= 0 else -1.0
    return best_axis, direction_sign


def find_hand_vertex_groups(target_mesh):
    r_name, l_name = "mixamorig:RightHand", "mixamorig:LeftHand"
    if target_mesh and hasattr(target_mesh, "vertex_groups"):
        for vg in target_mesh.vertex_groups:
            vg_lower = vg.name.lower()
            if any(k in vg_lower for k in ["righthand", "hand_r", "hand.r"]):
                r_name = vg.name
            elif any(k in vg_lower for k in ["lefthand", "hand_l", "hand.l"]):
                l_name = vg.name
    return r_name, l_name

# ==============================================================================
# RIG-PARENTED CAMERA RIG SETUP (BEST PRACTICE)
# ==============================================================================

import bpy
import math
from mathutils import Vector

def setup_modern_reel_camera(target_rig, target_mesh, turn_start, turn_end_frame, unfurl_end_frame):
    scene = bpy.context.scene
    scene.render.resolution_x = 1080
    scene.render.resolution_y = 1920
    scene.render.resolution_percentage = 100

    # 1. Setup Camera and Focus Target
    focus_target = bpy.data.objects.get("Camera_Focus_Target") or bpy.data.objects.new("Camera_Focus_Target", None)
    if focus_target.name not in scene.collection.objects:
        scene.collection.objects.link(focus_target)

    cam_data = bpy.data.cameras.get("Vertical_Reel_Cam_Data") or bpy.data.cameras.new("Vertical_Reel_Cam_Data")
    cam_data.lens = 45.0
    cam_data.sensor_fit = 'VERTICAL'

    cam_obj = bpy.data.objects.get("Vertical_Reel_Camera") or bpy.data.objects.new("Vertical_Reel_Camera", cam_data)
    if cam_obj.name not in scene.collection.objects:
        scene.collection.objects.link(cam_obj)

    scene.camera = cam_obj
    cam_obj.parent = None

    cam_obj.constraints.clear()
    track = cam_obj.constraints.new(type='TRACK_TO')
    track.target = focus_target
    track.track_axis = 'TRACK_NEGATIVE_Z'
    track.up_axis = 'UP_Y'

    if cam_obj.animation_data:
        cam_obj.animation_data_clear()
    if focus_target.animation_data:
        focus_target.animation_data_clear()

    # 2. Calculate Maximum Deformed Extent Across Sample Frames
    max_radius = 0.0
    sample_frames = list(range(1, unfurl_end_frame + 1, 5))
    if unfurl_end_frame not in sample_frames:
        sample_frames.append(unfurl_end_frame)

    depsgraph = bpy.context.evaluated_depsgraph_get()

    for f in sample_frames:
        scene.frame_set(f)
        depsgraph.update()

        # Evaluate the mesh WITH armature modifiers active
        eval_mesh_obj = target_mesh.evaluated_get(depsgraph)
        me = eval_mesh_obj.to_mesh()  # Gets actual deformed geometry
        matrix = eval_mesh_obj.matrix_world

        # Find world-space vertices for radius calculation
        world_verts = [matrix @ v.co for v in me.vertices]
        eval_mesh_obj.to_mesh_clear()  # Free memory

        if not world_verts:
            continue

        # Find center of world-space bounds
        min_x = min(v.x for v in world_verts)
        max_x = max(v.x for v in world_verts)
        min_y = min(v.y for v in world_verts)
        max_y = max(v.y for v in world_verts)
        min_z = min(v.z for v in world_verts)
        max_z = max(v.z for v in world_verts)

        center = Vector(((min_x + max_x) * 0.5, (min_y + max_y) * 0.5, (min_z + max_z) * 0.5))

        # Radius to furthest vertex from frame center
        for v in world_verts:
            dist = (v - center).length
            if dist > max_radius:
                max_radius = dist

    # Safety margin (15% padding)
    padded_radius = max_radius * 1.15

    # 3. Trigonometric Distance for Planar FOV (Vertical 9:16 Aspect Ratio)
    sensor_width = cam_data.sensor_width
    sensor_height = cam_data.sensor_height
    focal_length = cam_data.lens

    fov_v = 2.0 * math.atan((sensor_height * 0.5) / focal_length)
    aspect_ratio = scene.render.resolution_x / scene.render.resolution_y
    fov_h = 2.0 * math.atan(math.tan(fov_v * 0.5) * aspect_ratio)

    # Tighter FOV in 9:16 portrait mode is Horizontal (fov_h)
    min_fov = min(fov_v, fov_h)

    # Correct perspective camera distance formula
    required_dist = padded_radius / math.tan(min_fov * 0.5)

    root_bone_name = "mixamorig:Hips" if "mixamorig:Hips" in target_rig.pose.bones else target_rig.pose.bones[0].name

    # 4. Fast Keyframe Loop (Uses Bounding Box Corners for Speed)
    for f in range(1, unfurl_end_frame + 1):
        scene.frame_set(f)
        depsgraph.update()

        eval_mesh_obj = target_mesh.evaluated_get(depsgraph)

        # Fast Z-bound calculation using 8 corners of evaluated bounding box
        bbox_world = [eval_mesh_obj.matrix_world @ Vector(corner) for corner in eval_mesh_obj.bound_box]
        min_z = min(v.z for v in bbox_world)
        max_z = max(v.z for v in bbox_world)
        target_mid_z = min_z + ((max_z - min_z) * 0.50)

        eval_rig = target_rig.evaluated_get(depsgraph)
        pbone = eval_rig.pose.bones[root_bone_name]
        bone_world_matrix = eval_rig.matrix_world @ pbone.matrix

        char_ground_loc = Vector((bone_world_matrix.translation.x, bone_world_matrix.translation.y, min_z))

        focus_target.location = Vector((char_ground_loc.x, char_ground_loc.y, target_mid_z))
        focus_target.keyframe_insert(data_path="location", frame=f)

        rig_matrix = eval_rig.matrix_world
        forward_vec = (rig_matrix.to_3x3() @ Vector((0, 1, 0))).xy.normalized()
        if forward_vec.length == 0:
            forward_vec = Vector((0, -1))

        base_angle = math.atan2(forward_vec.y, forward_vec.x)

        if f <= turn_start:
            angle_offset = math.radians(-45.0)
        elif f >= turn_end_frame:
            angle_offset = 0.0
        else:
            t = (f - turn_start) / (turn_end_frame - turn_start)
            angle_offset = math.radians(-45.0) * (1.0 - t)

        cam_angle = base_angle + angle_offset

        cam_x = char_ground_loc.x - (required_dist * math.cos(cam_angle))
        cam_y = char_ground_loc.y - (required_dist * math.sin(cam_angle))

        cam_obj.location = Vector((cam_x, cam_y, target_mid_z))
        cam_obj.keyframe_insert(data_path="location", frame=f)

    # 5. Set Linear Interpolation across Blender 3.x and 4.x+
    for obj in [cam_obj, focus_target]:
        if obj and obj.animation_data and obj.animation_data.action:
            action = obj.animation_data.action
            fcurves = getattr(action, "fcurves", None)
            if fcurves is None and hasattr(action, "bindings"):
                fcurves = [fc for b in action.bindings for fc in b.fcurves if hasattr(b, "fcurves")]
            if fcurves:
                for fc in fcurves:
                    for kp in fc.keyframe_points:
                        kp.interpolation = 'LINEAR'

    # 6. Set NLA Extrapolation to HOLD so character remains still post frame 225
    for obj in [target_rig, target_mesh]:
        if obj and obj.animation_data and obj.animation_data.nla_tracks:
            for track in obj.animation_data.nla_tracks:
                for strip in track.strips:
                    strip.extrapolation = 'HOLD'

    scene.frame_set(1)
    depsgraph.update()

    return cam_obj, focus_target


def frame_scroll_image(cam_obj, focus_target, target_mesh, image_path, scroll_width=0.35, zoom_duration=45):
    scene = bpy.context.scene
    unfurl_end_frame = scene.frame_end
    zoom_end_frame = unfurl_end_frame + zoom_duration

    scene.frame_set(unfurl_end_frame)
    bpy.context.view_layer.update()

    img_aspect = 1.0
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        if img.size[1] > 0:
            img_aspect = img.size[0] / img.size[1]

    tile_height = scroll_width * img_aspect
    cam_data = cam_obj.data
    aspect_ratio = scene.render.resolution_x / scene.render.resolution_y
    sensor_width = cam_data.sensor_height * aspect_ratio

    fov_h_half = math.atan((sensor_width / 2.0) / cam_data.lens)
    target_framing_distance = (tile_height / 2.0) / math.tan(fov_h_half)

    mesh_matrix = target_mesh.matrix_world
    depsgraph = bpy.context.evaluated_depsgraph_get()
    eval_mesh = target_mesh.evaluated_get(depsgraph)

    r_hand_name, l_hand_name = find_hand_vertex_groups(target_mesh)
    vg_r = eval_mesh.vertex_groups.get(r_hand_name)
    vg_l = eval_mesh.vertex_groups.get(l_hand_name)

    r_verts = [mesh_matrix @ v.co for v in eval_mesh.data.vertices if any(g.group == vg_r.index for g in v.groups)] if vg_r else []
    l_verts = [mesh_matrix @ v.co for v in eval_mesh.data.vertices if any(g.group == vg_l.index for g in v.groups)] if vg_l else []

    target_height = target_mesh.dimensions.z if target_mesh.dimensions.z > 0 else 1.8
    torso_fallback = mesh_matrix.translation + Vector((0, 0, target_height * 0.55))

    r_hand_pos = sum(r_verts, Vector((0, 0, 0))) / len(r_verts) if r_verts else torso_fallback
    l_hand_pos = sum(l_verts, Vector((0, 0, 0))) / len(l_verts) if l_verts else torso_fallback

    scroll_midpoint = (r_hand_pos + l_hand_pos) * 0.5

    cam_pos = cam_obj.matrix_world.translation.copy()
    cam_dir = cam_pos - scroll_midpoint
    cam_dir.z = 0.0

    scroll_normal = cam_dir.normalized() if cam_dir.length > 0.001 else Vector((0, -1, 0))

    tile_center = scroll_midpoint + (scroll_normal * 0.10)
    end_pos = tile_center + (scroll_normal * target_framing_distance)

    # Detach to free world space for precise final zoom pass
    cam_obj.parent = None
    focus_target.parent = None

    current_world_focus = focus_target.matrix_world.translation.copy()
    focus_target.matrix_world.translation = current_world_focus

    focus_target.location = current_world_focus
    focus_target.keyframe_insert(data_path="location", frame=unfurl_end_frame)
    focus_target.location = tile_center
    focus_target.keyframe_insert(data_path="location", frame=zoom_end_frame)

    cam_obj.location = cam_pos
    cam_obj.keyframe_insert(data_path="location", frame=unfurl_end_frame)
    cam_obj.location = end_pos
    cam_obj.keyframe_insert(data_path="location", frame=zoom_end_frame)

    for obj in [cam_obj, focus_target]:
        for fc in get_object_fcurves(obj):
            for kp in fc.keyframe_points:
                if kp.co.x >= unfurl_end_frame:
                    kp.interpolation = 'BEZIER'
                    kp.handle_left_type = 'AUTO_CLAMPED'
                    kp.handle_right_type = 'AUTO_CLAMPED'

    scene.frame_end = zoom_end_frame
    return zoom_end_frame

# ==============================================================================
# GEOMETRY NODES GROUND, GRASS & SKY DOME SETUP
# ==============================================================================

def ensure_world_sphere_size(cam_obj, ground_obj, safety_multiplier=2.5):
    if not cam_obj or not ground_obj:
        return

    scene = bpy.context.scene
    max_cam_dist = 0.0

    for frame in range(scene.frame_start, scene.frame_end + 1, 5):
        scene.frame_set(frame)
        bpy.context.view_layer.update()
        max_cam_dist = max(max_cam_dist, cam_obj.matrix_world.translation.length)

    world_radius = max(25.0, max_cam_dist * safety_multiplier)

    mod = ground_obj.modifiers.get("Grass_Ground_GN")
    if mod and mod.node_group:
        for node in mod.node_group.nodes:
            if node.bl_idname == "GeometryNodeMeshIcoSphere":
                node.inputs["Radius"].default_value = world_radius

    if ground_obj.type == 'MESH':
        bpy.context.view_layer.objects.active = ground_obj
        bpy.ops.object.mode_set(mode='OBJECT')
        ground_scale = world_radius / 12.5
        ground_obj.scale = (ground_scale, ground_scale, 1.0)
        bpy.ops.object.transform_apply(location=False, rotation=False, scale=True)

    cam_obj.data.clip_end = max(cam_obj.data.clip_end, world_radius * 2.5)
    scene.frame_set(1)


def create_grass_material(grass_dir):
    mat = bpy.data.materials.new(name="Grass_PBR_Material")
    mat.use_nodes = True
    nodes, links = mat.node_tree.nodes, mat.node_tree.links
    nodes.clear()

    out_node = nodes.new("ShaderNodeOutputMaterial")
    bsdf = nodes.new("ShaderNodeBsdfPrincipled")
    links.new(bsdf.outputs["BSDF"], out_node.inputs["Surface"])

    col_files = glob.glob(os.path.join(grass_dir, "*Color*.jpg")) + glob.glob(os.path.join(grass_dir, "*JPG*.jpg"))
    if col_files:
        tex_col = nodes.new("ShaderNodeTexImage")
        tex_col.image = bpy.data.images.load(col_files[0])
        links.new(tex_col.outputs["Color"], bsdf.inputs["Base Color"])

    rough_files = glob.glob(os.path.join(grass_dir, "*Roughness*.jpg"))
    if rough_files:
        tex_rough = nodes.new("ShaderNodeTexImage")
        tex_rough.image = bpy.data.images.load(rough_files[0])
        tex_rough.image.colorspace_settings.name = 'Non-Color'
        links.new(tex_rough.outputs["Color"], bsdf.inputs["Roughness"])

    norm_files = glob.glob(os.path.join(grass_dir, "*Normal*.jpg"))
    if norm_files:
        tex_norm = nodes.new("ShaderNodeTexImage")
        tex_norm.image = bpy.data.images.load(norm_files[0])
        tex_norm.image.colorspace_settings.name = 'Non-Color'
        normal_map_node = nodes.new("ShaderNodeNormalMap")
        links.new(tex_norm.outputs["Color"], normal_map_node.inputs["Color"])
        links.new(normal_map_node.outputs["Normal"], bsdf.inputs["Normal"])

    return mat


def create_sky_material(sky_hdri_path):
    mat = bpy.data.materials.new(name="Sky_Dome_Material")
    mat.use_nodes = True
    nodes, links = mat.node_tree.nodes, mat.node_tree.links
    nodes.clear()

    out_node = nodes.new("ShaderNodeOutputMaterial")
    emission = nodes.new("ShaderNodeEmission")
    emission.inputs["Strength"].default_value = 1.0

    transparent = nodes.new("ShaderNodeBsdfTransparent")
    mix_shader = nodes.new("ShaderNodeMixShader")
    light_path = nodes.new("ShaderNodeLightPath")

    # Is Camera Ray -> Fac (Camera sees Emission, all other rays see Transparent)
    links.new(light_path.outputs["Is Camera Ray"], mix_shader.inputs["Fac"])
    links.new(transparent.outputs["BSDF"], mix_shader.inputs[1])
    links.new(emission.outputs["Emission"], mix_shader.inputs[2])
    links.new(mix_shader.outputs["Shader"], out_node.inputs["Surface"])

    if os.path.exists(sky_hdri_path):
        tex_env = nodes.new("ShaderNodeTexEnvironment")
        tex_env.image = bpy.data.images.load(sky_hdri_path)
        links.new(tex_env.outputs["Color"], emission.inputs["Color"])

    return mat


def setup_grass_ground_geometry_nodes(grass_dir, sky_hdri_path, ground_z_offset=-0.2):
    grass_coll = bpy.data.collections.get("Grass_Artifacts_Collection") or bpy.data.collections.new("Grass_Artifacts_Collection")
    if grass_coll.name not in bpy.context.scene.collection.children:
        bpy.context.scene.collection.children.link(grass_coll)

    grass_mat = create_grass_material(grass_dir)
    sky_mat = create_sky_material(sky_hdri_path)

    mesh_files = glob.glob(os.path.join(grass_dir, "*.obj")) + glob.glob(os.path.join(grass_dir, "*.fbx"))

    if mesh_files:
        for fpath in mesh_files:
            if fpath.endswith(".obj"):
                if hasattr(bpy.ops.wm, "obj_import"):
                    bpy.ops.wm.obj_import(filepath=fpath)
                else:
                    bpy.ops.import_scene.obj(filepath=fpath)
            elif fpath.endswith(".fbx"):
                bpy.ops.import_scene.fbx(filepath=fpath)

            for o in bpy.context.selected_objects:
                grass_coll.objects.link(o)
                if o.name in bpy.context.scene.collection.objects:
                    bpy.context.scene.collection.objects.unlink(o)

        for obj in list(grass_coll.objects):
            if obj.dimensions.z > 0.5:
                grass_coll.objects.unlink(obj)
                bpy.data.objects.remove(obj, do_unlink=True)
    else:
        for i in range(3):
            m_data = bpy.data.meshes.new(f"Grass_Blade_Mesh_{i}")
            g_obj = bpy.data.objects.new(f"Grass_Blade_Obj_{i}", m_data)
            grass_coll.objects.link(g_obj)

            verts = [(0, 0, 0), (0.05, 0, 0), (0.04, 0, 0.15), (0.01, 0, 0.25), (0, 0, 0.3)]
            faces = [(0, 1, 2), (1, 2, 3), (2, 3, 4)]
            m_data.from_pydata(verts, [], faces)
            m_data.update()

            g_obj.data.materials.append(grass_mat)
            g_obj.location = (100, 100, 0)

    bpy.ops.mesh.primitive_circle_add(radius=12.5, vertices=64, fill_type='NGON', location=(0, 0, ground_z_offset))
    ground_obj = bpy.context.active_object
    ground_obj.name = "Ground_Mesh_Object"

    vg = ground_obj.vertex_groups.new(name="Grass_Weight_Group")
    vg.add([v.index for v in ground_obj.data.vertices], 1.0, 'REPLACE')
    ground_obj.data.materials.append(grass_mat)
    ground_obj.data.materials.append(sky_mat)

    gn_group = bpy.data.node_groups.new("Grass_Ground_GN", type="GeometryNodeTree")
    gn_group.interface.new_socket(name="Geometry", in_out="INPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Geometry", in_out="OUTPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Density Mask", in_out="INPUT", socket_type="NodeSocketFloat")

    nodes, links = gn_group.nodes, gn_group.links
    out_node, in_node = nodes.new("NodeGroupOutput"), nodes.new("NodeGroupInput")

    distribute_node = nodes.new("GeometryNodeDistributePointsOnFaces")
    distribute_node.distribute_method = 'POISSON'
    distribute_node.inputs["Distance Min"].default_value = 0.02
    distribute_node.inputs["Density Max"].default_value = 300.0
    links.new(in_node.outputs["Geometry"], distribute_node.inputs["Mesh"])

    noise_node = nodes.new("ShaderNodeTexNoise")
    noise_node.inputs["Scale"].default_value = 2.5
    noise_node.inputs["Detail"].default_value = 2.0

    color_ramp = nodes.new("ShaderNodeValToRGB")
    color_ramp.color_ramp.elements[0].position = 0.20
    color_ramp.color_ramp.elements[1].position = 0.80
    links.new(noise_node.outputs["Fac"], color_ramp.inputs["Fac"])

    math_weight = nodes.new("ShaderNodeMath")
    math_weight.operation = "MULTIPLY"
    links.new(color_ramp.outputs["Color"], math_weight.inputs[0])
    links.new(in_node.outputs["Density Mask"], math_weight.inputs[1])
    links.new(math_weight.outputs["Value"], distribute_node.inputs["Density Factor"])

    instance_node = nodes.new("GeometryNodeInstanceOnPoints")
    instance_node.inputs["Pick Instance"].default_value = True
    links.new(distribute_node.outputs["Points"], instance_node.inputs["Points"])

    coll_info = nodes.new("GeometryNodeCollectionInfo")
    coll_info.inputs["Collection"].default_value = grass_coll
    coll_info.inputs["Separate Children"].default_value = True
    coll_info.inputs["Reset Children"].default_value = True
    links.new(coll_info.outputs["Instances"], instance_node.inputs["Instance"])

    rand_scale = nodes.new("FunctionNodeRandomValue")
    rand_scale.inputs["Min"].default_value = 0.25
    rand_scale.inputs["Max"].default_value = 0.50

    combine_scale = nodes.new("ShaderNodeCombineXYZ")
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["X"])
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["Y"])
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["Z"])
    links.new(combine_scale.outputs["Vector"], instance_node.inputs["Scale"])

    rand_rot_z = nodes.new("FunctionNodeRandomValue")
    rand_rot_z.inputs["Min"].default_value = 0.0
    rand_rot_z.inputs["Max"].default_value = math.radians(360.0)

    combine_rot = nodes.new("ShaderNodeCombineXYZ")
    links.new(rand_rot_z.outputs["Value"], combine_rot.inputs["Z"])
    links.new(combine_rot.outputs["Vector"], instance_node.inputs["Rotation"])

    sky_sphere = nodes.new("GeometryNodeMeshIcoSphere")
    sky_sphere.inputs["Radius"].default_value = 12.5
    sky_sphere.inputs["Subdivisions"].default_value = 4

    flip_faces = nodes.new("GeometryNodeFlipFaces")
    links.new(sky_sphere.outputs["Mesh"], flip_faces.inputs["Mesh"])

    transform_sky = nodes.new("GeometryNodeTransform")
    transform_sky.inputs["Translation"].default_value = (0.0, 0.0, ground_z_offset)
    links.new(flip_faces.outputs["Mesh"], transform_sky.inputs["Geometry"])

    set_sky_mat = nodes.new("GeometryNodeSetMaterial")
    set_sky_mat.inputs["Material"].default_value = sky_mat
    links.new(transform_sky.outputs["Geometry"], set_sky_mat.inputs["Geometry"])

    join_geo = nodes.new("GeometryNodeJoinGeometry")
    links.new(in_node.outputs["Geometry"], join_geo.inputs["Geometry"])
    links.new(instance_node.outputs["Instances"], join_geo.inputs["Geometry"])
    links.new(set_sky_mat.outputs["Geometry"], join_geo.inputs["Geometry"])

    links.new(join_geo.outputs["Geometry"], out_node.inputs["Geometry"])

    mod = ground_obj.modifiers.new(name="Grass_Ground_GN", type='NODES')
    mod.node_group = gn_group

    for item in gn_group.interface.items_tree:
        if getattr(item, "in_out", None) == "INPUT" and item.name == "Density Mask":
            mod[item.identifier + "_attribute_name"] = "Grass_Weight_Group"
            mod[item.identifier + "_use_attribute"] = True

    return ground_obj

# ==============================================================================
# SCROLL MATERIAL & GEOMETRY NODES SETUP
# ==============================================================================

def create_image_scroll_material(image_path):
    mat_name = "Scroll_Texture_Mat"
    if mat := bpy.data.materials.get(mat_name):
        bpy.data.materials.remove(mat)

    mat = bpy.data.materials.new(name=mat_name)
    mat.use_nodes = True
    nodes, links = mat.node_tree.nodes, mat.node_tree.links
    nodes.clear()

    output_node = nodes.new("ShaderNodeOutputMaterial")
    bsdf = nodes.new("ShaderNodeBsdfPrincipled")
    bsdf.inputs["Roughness"].default_value = 0.5

    tex_img = nodes.new("ShaderNodeTexImage")
    if os.path.exists(image_path):
        tex_img.image = bpy.data.images.load(image_path)
        tex_img.extension = 'REPEAT'

    uv_map_node = nodes.new("ShaderNodeUVMap")
    uv_map_node.uv_map = "UVMap"

    links.new(uv_map_node.outputs["UV"], tex_img.inputs["Vector"])
    links.new(tex_img.outputs["Color"], bsdf.inputs["Base Color"])
    links.new(bsdf.outputs["BSDF"], output_node.inputs["Surface"])

    return mat, tex_img.image if os.path.exists(image_path) else None


def create_pure_geometry_node_scroll(image_path=""):
    gn_group = bpy.data.node_groups.new("Scroll_Pure_GN", type="GeometryNodeTree")

    gn_group.interface.new_socket(name="Geometry", in_out="OUTPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Character Mesh", in_out="INPUT", socket_type="NodeSocketObject")

    sock_r_vg = gn_group.interface.new_socket(name="Right Hand Group", in_out="INPUT", socket_type="NodeSocketString")
    sock_r_vg.default_value = "mixamorig:RightHand"

    sock_l_vg = gn_group.interface.new_socket(name="Left Hand Group", in_out="INPUT", socket_type="NodeSocketString")
    sock_l_vg.default_value = "mixamorig:LeftHand"

    sock_offset = gn_group.interface.new_socket(name="Forward Offset", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_offset.default_value = 0.60

    sock_unfurl = gn_group.interface.new_socket(name="Unfurl", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_unfurl.default_value = 0.0

    sock_width = gn_group.interface.new_socket(name="Scroll Width", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_width.default_value = 0.35

    nodes, links = gn_group.nodes, gn_group.links
    in_node, out_node = nodes.new("NodeGroupInput"), nodes.new("NodeGroupOutput")

    char_info = nodes.new("GeometryNodeObjectInfo")
    char_info.transform_space = "RELATIVE"
    links.new(in_node.outputs["Character Mesh"], char_info.inputs["Object"])

    pos_node = nodes.new("GeometryNodeInputPosition")

    named_r = nodes.new("GeometryNodeInputNamedAttribute")
    named_r.data_type = "FLOAT"
    links.new(in_node.outputs["Right Hand Group"], named_r.inputs["Name"])

    cmp_r = nodes.new("FunctionNodeCompare")
    cmp_r.data_type = "FLOAT"
    cmp_r.operation = "GREATER_THAN"
    cmp_r.inputs[1].default_value = 0.001
    links.new(named_r.outputs["Attribute"], cmp_r.inputs[0])

    stat_r = nodes.new("GeometryNodeAttributeStatistic")
    stat_r.data_type = "FLOAT_VECTOR"
    links.new(char_info.outputs["Geometry"], stat_r.inputs["Geometry"])
    links.new(cmp_r.outputs["Result"], stat_r.inputs["Selection"])
    links.new(pos_node.outputs["Position"], stat_r.inputs["Attribute"])

    named_l = nodes.new("GeometryNodeInputNamedAttribute")
    named_l.data_type = "FLOAT"
    links.new(in_node.outputs["Left Hand Group"], named_l.inputs["Name"])

    cmp_l = nodes.new("FunctionNodeCompare")
    cmp_l.data_type = "FLOAT"
    cmp_l.operation = "GREATER_THAN"
    cmp_l.inputs[1].default_value = 0.001
    links.new(named_l.outputs["Attribute"], cmp_l.inputs[0])

    stat_l = nodes.new("GeometryNodeAttributeStatistic")
    stat_l.data_type = "FLOAT_VECTOR"
    links.new(char_info.outputs["Geometry"], stat_l.inputs["Geometry"])
    links.new(cmp_l.outputs["Result"], stat_l.inputs["Selection"])
    links.new(pos_node.outputs["Position"], stat_l.inputs["Attribute"])

    neg_offset = nodes.new("ShaderNodeMath")
    neg_offset.operation = "MULTIPLY"
    neg_offset.inputs[1].default_value = -1.0
    links.new(in_node.outputs["Forward Offset"], neg_offset.inputs[0])

    forward_vec = nodes.new("ShaderNodeCombineXYZ")
    forward_vec.inputs["X"].default_value = 0.0
    links.new(neg_offset.outputs["Value"], forward_vec.inputs["Y"])
    forward_vec.inputs["Z"].default_value = 0.0

    start_pos = nodes.new("ShaderNodeVectorMath")
    start_pos.operation = "ADD"
    links.new(stat_r.outputs["Mean"], start_pos.inputs[0])
    links.new(forward_vec.outputs["Vector"], start_pos.inputs[1])

    target_pos = nodes.new("ShaderNodeVectorMath")
    target_pos.operation = "ADD"
    links.new(stat_l.outputs["Mean"], target_pos.inputs[0])
    links.new(forward_vec.outputs["Vector"], target_pos.inputs[1])

    delta_vec = nodes.new("ShaderNodeVectorMath")
    delta_vec.operation = "SUBTRACT"
    links.new(target_pos.outputs["Vector"], delta_vec.inputs[0])
    links.new(start_pos.outputs["Vector"], delta_vec.inputs[1])

    dist_node = nodes.new("ShaderNodeVectorMath")
    dist_node.operation = "DISTANCE"
    links.new(start_pos.outputs["Vector"], dist_node.inputs[0])

    scale_vec = nodes.new("ShaderNodeVectorMath")
    scale_vec.operation = "SCALE"
    links.new(delta_vec.outputs["Vector"], scale_vec.inputs["Vector"])
    links.new(in_node.outputs["Unfurl"], scale_vec.inputs["Scale"])

    end_pos = nodes.new("ShaderNodeVectorMath")
    end_pos.operation = "ADD"
    links.new(start_pos.outputs["Vector"], end_pos.inputs[0])
    links.new(scale_vec.outputs["Vector"], end_pos.inputs[1])
    links.new(end_pos.outputs["Vector"], dist_node.inputs[1])

    curve_line = nodes.new("GeometryNodeCurvePrimitiveLine")
    links.new(start_pos.outputs["Vector"], curve_line.inputs["Start"])
    links.new(end_pos.outputs["Vector"], curve_line.inputs["End"])

    set_tilt = nodes.new("GeometryNodeSetCurveTilt")
    set_tilt.inputs["Tilt"].default_value = 1.5708
    links.new(curve_line.outputs["Curve"], set_tilt.inputs["Curve"])

    spline_param_main = nodes.new("GeometryNodeSplineParameter")
    store_u = nodes.new("GeometryNodeStoreNamedAttribute")
    store_u.inputs["Name"].default_value = "U_Factor"
    store_u.data_type = 'FLOAT'
    store_u.domain = 'POINT'
    links.new(set_tilt.outputs["Curve"], store_u.inputs["Geometry"])
    links.new(spline_param_main.outputs["Factor"], store_u.inputs["Value"])

    half_width_neg = nodes.new("ShaderNodeMath")
    half_width_neg.operation = "MULTIPLY"
    half_width_neg.inputs[1].default_value = -0.5
    links.new(in_node.outputs["Scroll Width"], half_width_neg.inputs[0])

    half_width_pos = nodes.new("ShaderNodeMath")
    half_width_pos.operation = "MULTIPLY"
    half_width_pos.inputs[1].default_value = 0.5
    links.new(in_node.outputs["Scroll Width"], half_width_pos.inputs[0])

    combine_start = nodes.new("ShaderNodeCombineXYZ")
    links.new(half_width_neg.outputs["Value"], combine_start.inputs["X"])

    combine_end = nodes.new("ShaderNodeCombineXYZ")
    links.new(half_width_pos.outputs["Value"], combine_end.inputs["X"])

    profile_line = nodes.new("GeometryNodeCurvePrimitiveLine")
    links.new(combine_start.outputs["Vector"], profile_line.inputs["Start"])
    links.new(combine_end.outputs["Vector"], profile_line.inputs["End"])

    spline_param_prof = nodes.new("GeometryNodeSplineParameter")
    store_v = nodes.new("GeometryNodeStoreNamedAttribute")
    store_v.inputs["Name"].default_value = "V_Factor"
    store_v.data_type = 'FLOAT'
    store_v.domain = 'POINT'
    links.new(profile_line.outputs["Curve"], store_v.inputs["Geometry"])
    links.new(spline_param_prof.outputs["Factor"], store_v.inputs["Value"])

    curve_to_mesh = nodes.new("GeometryNodeCurveToMesh")
    links.new(store_u.outputs["Geometry"], curve_to_mesh.inputs["Curve"])
    links.new(store_v.outputs["Geometry"], curve_to_mesh.inputs["Profile Curve"])

    img_aspect = 1.0
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        if img.size[1] > 0:
            img_aspect = img.size[0] / img.size[1]

    h_times_aspect = nodes.new("ShaderNodeMath")
    h_times_aspect.operation = "MULTIPLY"
    h_times_aspect.inputs[1].default_value = img_aspect
    links.new(in_node.outputs["Scroll Width"], h_times_aspect.inputs[0])

    tile_count = nodes.new("ShaderNodeMath")
    tile_count.operation = "DIVIDE"
    links.new(dist_node.outputs["Value"], tile_count.inputs[0])
    links.new(h_times_aspect.outputs["Value"], tile_count.inputs[1])

    read_u = nodes.new("GeometryNodeInputNamedAttribute")
    read_u.inputs["Name"].default_value = "U_Factor"
    read_u.data_type = 'FLOAT'

    scaled_u = nodes.new("ShaderNodeMath")
    scaled_u.operation = "MULTIPLY"
    links.new(read_u.outputs["Attribute"], scaled_u.inputs[0])
    links.new(tile_count.outputs["Value"], scaled_u.inputs[1])

    read_v = nodes.new("GeometryNodeInputNamedAttribute")
    read_v.inputs["Name"].default_value = "V_Factor"
    read_v.data_type = 'FLOAT'

    flip_v = nodes.new("ShaderNodeMath")
    flip_v.operation = "SUBTRACT"
    flip_v.inputs[0].default_value = 1.0
    links.new(read_v.outputs["Attribute"], flip_v.inputs[1])

    combine_uv = nodes.new("ShaderNodeCombineXYZ")
    links.new(scaled_u.outputs["Value"], combine_uv.inputs["X"])
    links.new(flip_v.outputs["Value"], combine_uv.inputs["Y"])

    store_uv = nodes.new("GeometryNodeStoreNamedAttribute")
    store_uv.inputs["Name"].default_value = "UVMap"
    store_uv.data_type = 'FLOAT2'
    store_uv.domain = 'CORNER'
    links.new(curve_to_mesh.outputs["Mesh"], store_uv.inputs["Geometry"])
    links.new(combine_uv.outputs["Vector"], store_uv.inputs["Value"])

    set_material = nodes.new("GeometryNodeSetMaterial")
    scroll_mat, _ = create_image_scroll_material(image_path)
    set_material.inputs["Material"].default_value = scroll_mat
    links.new(store_uv.outputs["Geometry"], set_material.inputs["Geometry"])

    links.new(set_material.outputs["Geometry"], out_node.inputs["Geometry"])

    return gn_group, scroll_mat


def setup_geometry_node_scroll(target_mesh, unfurl_start, unfurl_end, image_path=""):
    mesh_data = bpy.data.meshes.new("Scroll_Procedural_Mesh")
    scroll_obj = bpy.data.objects.new("Scroll_GN_Object", mesh_data)
    bpy.context.scene.collection.objects.link(scroll_obj)

    mod = scroll_obj.modifiers.new("Scroll_GN", type="NODES")
    gn_tree, scroll_mat = create_pure_geometry_node_scroll(image_path=image_path)
    mod.node_group = gn_tree

    scroll_obj.data.materials.append(scroll_mat)

    r_hand_name, l_hand_name = find_hand_vertex_groups(target_mesh)

    unfurl_id = None
    for item in gn_tree.interface.items_tree:
        if getattr(item, "in_out", None) == "INPUT":
            if item.name == "Character Mesh":
                mod[item.identifier] = target_mesh
            elif item.name == "Right Hand Group":
                mod[item.identifier] = r_hand_name
            elif item.name == "Left Hand Group":
                mod[item.identifier] = l_hand_name
            elif item.name == "Forward Offset":
                mod[item.identifier] = 0.60
            elif item.name == "Scroll Width":
                mod[item.identifier] = 0.35
            elif item.name == "Unfurl":
                unfurl_id = item.identifier

    scroll_obj.hide_viewport = True
    scroll_obj.hide_render = True
    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=1)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=1)

    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=unfurl_start - 1)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=unfurl_start - 1)

    scroll_obj.hide_viewport = False
    scroll_obj.hide_render = False
    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=unfurl_start)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=unfurl_start)

    for fc in get_object_fcurves(scroll_obj):
        if fc.data_path in ["hide_viewport", "hide_render"]:
            for kp in fc.keyframe_points:
                kp.interpolation = 'CONSTANT'

    if unfurl_id:
        mod[unfurl_id] = 0.001
        mod.keyframe_insert(data_path=f'["{unfurl_id}"]', frame=unfurl_start)

        mod[unfurl_id] = 1.0
        mod.keyframe_insert(data_path=f'["{unfurl_id}"]', frame=unfurl_end)

    return scroll_obj

# ==============================================================================
# MAIN CONTROLLER
# ==============================================================================

def import_and_extract_action(fbx_path, action_name):
    existing_objects = set(bpy.data.objects)
    bpy.ops.import_scene.fbx(filepath=fbx_path)
    bpy.context.view_layer.update()

    imported_objects = set(bpy.data.objects) - existing_objects
    extracted_action = None
    for obj in imported_objects:
        if obj.animation_data and obj.animation_data.action:
            extracted_action = obj.animation_data.action
            extracted_action.name = action_name
            extracted_action.use_fake_user = True
            break
    return extracted_action, imported_objects


def purge_imported_objects(objects_to_remove):
    for obj in objects_to_remove:
        if obj.name in bpy.data.objects:
            if obj.animation_data:
                obj.animation_data_clear()
            for coll in list(obj.users_collection):
                coll.objects.unlink(obj)
    bpy.context.view_layer.update()



def setup_world_hdri_lighting(sky_hdri_path, strength=2.5):
    scene = bpy.context.scene
    if not scene.world:
        scene.world = bpy.data.worlds.new("World")

    scene.world.use_nodes = True
    nodes = scene.world.node_tree.nodes
    links = scene.world.node_tree.links
    nodes.clear()

    out_node = nodes.new("ShaderNodeOutputWorld")
    bg_node = nodes.new("ShaderNodeBackground")
    bg_node.inputs["Strength"].default_value = strength

    links.new(bg_node.outputs["Background"], out_node.inputs["Surface"])

    if os.path.exists(sky_hdri_path):
        tex_env = nodes.new("ShaderNodeTexEnvironment")
        tex_env.image = bpy.data.images.load(sky_hdri_path)
        links.new(tex_env.outputs["Color"], bg_node.inputs["Color"])

def main():
    cleanup_previous_run()

    walk_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Mutant Walking.fbx"
    turn_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Right Turn 90.fbx"
    unfurl_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Mutant Idle.fbx"

    sky_hdri_path = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/horn-koppe_spring_4k.exr"
    output_blend = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/output_scene.blend"

    image_path = "/Users/dobbinp1/Downloads/IMGP6691_psedit.jpg"
    image_path = "/Users/dobbinp1/Downloads/sea_PanoramaIMGP5963.jpg"
    grass_dir = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Grass004_4K-JPG"

    # --------------------------------------------------------------------------
    # COLOR MANAGEMENT & WORLD LIGHTING FIXES
    # --------------------------------------------------------------------------
    scene = bpy.context.scene
    scene.view_settings.exposure = 1.0        # Lift overall scene exposure (Default 0.0)
    scene.view_settings.view_transform = 'AgX' # 'AgX' or 'Filmic' for balanced shadows
    scene.view_settings.look = 'AgX - Medium High Contrast'

    # Set world HDRI for scene illumination
    setup_world_hdri_lighting(sky_hdri_path, strength=1.0)

    # Setup geometry node grass ground and dome
    ground_obj = setup_grass_ground_geometry_nodes(grass_dir, sky_hdri_path, ground_z_offset=-0.2)

    CYCLES = 4
    CROSSFADE = 15
    TARGET_DISTANCE = 4.0

    walk_action, walk_objs = import_and_extract_action(walk_fbx, "walk")
    target_rig = next((obj for obj in walk_objs if obj.type == "ARMATURE"), None)
    target_mesh = next((obj for obj in walk_objs if obj.type == "MESH"), None)

    if not target_rig or not target_mesh:
        raise RuntimeError("Armature or Character Mesh not found in walk FBX.")

    travel_axis_idx, motion_sign = detect_natural_forward_axis(walk_action, target_rig)
    initial_heading = math.radians(90.0) if travel_axis_idx == 0 else 0.0
    target_rig.rotation_euler[2] = initial_heading

    action_len = max(30, int(walk_action.frame_range[1] - walk_action.frame_range[0]))
    anim_data = target_rig.animation_data_create()

    walk_track = anim_data.nla_tracks.new()
    walk_track.name = "Walk_Track"

    current_frame = 1
    for i in range(CYCLES):
        strip = walk_track.strips.new(name=f"Walk_{i+1}", start=int(current_frame), action=walk_action)
        strip.blend_type = "REPLACE"
        current_frame += action_len

    walk_end_frame = int(current_frame)

    turn_action, turn_objs = import_and_extract_action(turn_fbx, "rightturn")
    purge_imported_objects(turn_objs)

    turn_track = anim_data.nla_tracks.new()
    turn_track.name = "RightTurn_Track"

    turn_start = int(walk_end_frame - CROSSFADE)
    turn_strip = turn_track.strips.new(name="RightTurn", start=turn_start, action=turn_action)
    turn_strip.blend_type = "REPLACE"
    turn_strip.blend_in = CROSSFADE
    turn_strip.use_auto_blend = True
    turn_strip.extrapolation = "NOTHING"  # Best Practice: Avoid locked-out lower tracks
    turn_end_frame = int(turn_strip.frame_end)

    bpy.context.scene["turn_start_frame"] = turn_start

    unfurl_action, unfurl_objs = import_and_extract_action(unfurl_fbx, "unfurl")
    purge_imported_objects(unfurl_objs)

    unfurl_track = anim_data.nla_tracks.new()
    unfurl_track.name = "Unfurl_Track"

    unfurl_start = int(turn_end_frame - CROSSFADE)
    unfurl_strip = unfurl_track.strips.new(name="Unfurl", start=unfurl_start, action=unfurl_action)
    unfurl_strip.blend_type = "REPLACE"
    unfurl_strip.blend_in = CROSSFADE
    unfurl_strip.use_auto_blend = True
    unfurl_strip.extrapolation = "HOLD"  # Best Practice: Avoid locked-out lower tracks
    unfurl_end_frame = int(unfurl_strip.frame_end)

    root_action = bpy.data.actions.new(name="World_Root_Motion")
    target_rig.animation_data.action = root_action

    if hasattr(root_action, "fcurves") and hasattr(root_action.fcurves, "new"):
        loc_fcurve = root_action.fcurves.new(data_path="location", index=travel_axis_idx)
        rot_fcurve = root_action.fcurves.new(data_path="rotation_euler", index=2)
    else:
        loc_fcurve = root_action.fcurve_ensure_for_datablock(datablock=target_rig, data_path="location", index=travel_axis_idx)
        rot_fcurve = root_action.fcurve_ensure_for_datablock(datablock=target_rig, data_path="rotation_euler", index=2)

    final_target_value = TARGET_DISTANCE * motion_sign

    k_loc1 = loc_fcurve.keyframe_points.insert(frame=1.0, value=0.0)
    k_loc2 = loc_fcurve.keyframe_points.insert(frame=float(turn_start), value=final_target_value)
    k_loc3 = loc_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=final_target_value)

    k_loc1.interpolation = "LINEAR"
    k_loc2.interpolation = "CONSTANT"
    k_loc3.interpolation = "CONSTANT"

    target_heading = initial_heading - math.radians(90.0)

    k_rot1 = rot_fcurve.keyframe_points.insert(frame=1.0, value=initial_heading)
    k_rot2 = rot_fcurve.keyframe_points.insert(frame=float(turn_start), value=initial_heading)
    k_rot3 = rot_fcurve.keyframe_points.insert(frame=float(turn_end_frame), value=target_heading)
    k_rot4 = rot_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=target_heading)

    k_rot1.interpolation = "CONSTANT"
    k_rot2.interpolation = "BEZIER"
    k_rot3.interpolation = "CONSTANT"
    k_rot4.interpolation = "CONSTANT"

    root_track = anim_data.nla_tracks.new()
    root_track.name = "World_Root_Track"
    root_strip = root_track.strips.new(name="Root_Motion_Strip", start=1, action=root_action)
    root_strip.blend_type = "REPLACE"
    root_strip.extrapolation = "HOLD"

    anim_data.action = None

    setup_geometry_node_scroll(
        target_mesh=target_mesh,
        unfurl_start=unfurl_start,
        unfurl_end=unfurl_end_frame,
        image_path=image_path,
    )

    cam_obj, focus_target = setup_modern_reel_camera(
        target_rig=target_rig,
        target_mesh=target_mesh,
        turn_start=turn_start,
        turn_end_frame=turn_end_frame,
        unfurl_end_frame=unfurl_end_frame
    )

    bpy.context.scene.frame_start = 1
    bpy.context.scene.frame_end = unfurl_end_frame

    frame_scroll_image(
        cam_obj=cam_obj,
        focus_target=focus_target,
        target_mesh=target_mesh,
        image_path=image_path,
        scroll_width=0.35,
        zoom_duration=45
    )

    ensure_world_sphere_size(cam_obj, ground_obj, safety_multiplier=2.5)
    register_audio_handler()

    bpy.ops.wm.save_as_mainfile(filepath=output_blend)
    print(f"\n[SUCCESS] Integrated parented rig tracking & saved scene to: {output_blend}")


if __name__ == "__main__":
    main()
```

```
import math
import mathutils
import os
import sys
import glob
import bpy
import aud
from bpy_extras import anim_utils
from bpy_extras.object_utils import world_to_camera_view
from bpy.app.handlers import persistent
from mathutils import Vector

# ==============================================================================
# CONFIGURATION
# ==============================================================================
FOOTSTEP_SOUND_PATH = "/Users/dobbinp1/Downloads/stomp.wav"
GROUND_THRESHOLD = 0.24  # Raised slightly for testing

# Setup Audio Device
if "_audio_store" not in sys.modules:
    import types
    sys.modules["_audio_store"] = types.ModuleType("_audio_store")

audio_store = sys.modules["_audio_store"]
if not hasattr(audio_store, "device"):
    audio_store.device = aud.Device()

AUDIO_DEVICE = audio_store.device
SOUND_FACTORY = None
AUDIO_HANDLES = []

if os.path.exists(FOOTSTEP_SOUND_PATH):
    SOUND_FACTORY = aud.Sound(FOOTSTEP_SOUND_PATH).cache()
    print(f"[AUDIO] Loaded sound file successfully.")
else:
    print(f"[AUDIO ERROR] File not found at: {FOOTSTEP_SOUND_PATH}")

_left_foot_was_down = False
_right_foot_was_down = False
_last_frame = -1


def play_footstep():
    global AUDIO_HANDLES
    if SOUND_FACTORY:
        print("  >>> PLAYING FOOTSTEP SOUND <<<")
        handle = AUDIO_DEVICE.play(SOUND_FACTORY)
        AUDIO_HANDLES.append(handle)
        if len(AUDIO_HANDLES) > 10:
            AUDIO_HANDLES.pop(0)


def find_foot_bone(pose_bones, side="Left"):
    target_keywords = [
        f"mixamorig:{side}foot",
        f"{side.lower()}foot",
        f"foot.{side[0].lower()}",
        f"foot_{side[0].lower()}"
    ]
    for bone in pose_bones:
        b_name = bone.name.lower()
        if any(kw in b_name for kw in target_keywords):
            return bone
    return None


@persistent
def footstep_frame_handler(scene):
    global _left_foot_was_down, _right_foot_was_down, _last_frame

    try:
        current_frame = scene.frame_current
        turn_start = scene.get("turn_start_frame", None)

        # Stop footsteps once turning/idling starts
        if turn_start is not None and current_frame >= turn_start:
            _left_foot_was_down = False
            _right_foot_was_down = False
            return

        if current_frame <= _last_frame or abs(current_frame - _last_frame) > 10:
            _left_foot_was_down = False
            _right_foot_was_down = False
        _last_frame = current_frame

        rig = next((o for o in scene.objects if o.type == 'ARMATURE'), None)
        if not rig:
            return

        depsgraph = bpy.context.evaluated_depsgraph_get()
        eval_rig = rig.evaluated_get(depsgraph)

        left_bone = find_foot_bone(eval_rig.pose.bones, "Left")
        right_bone = find_foot_bone(eval_rig.pose.bones, "Right")

        if not left_bone or not right_bone:
            return

        l_z = (eval_rig.matrix_world @ left_bone.matrix).translation.z
        r_z = (eval_rig.matrix_world @ right_bone.matrix).translation.z

        l_is_down = l_z <= GROUND_THRESHOLD
        r_is_down = r_z <= GROUND_THRESHOLD

        if l_is_down and not _left_foot_was_down:
            play_footstep()

        if r_is_down and not _right_foot_was_down:
            play_footstep()

        _left_foot_was_down = l_is_down
        _right_foot_was_down = r_is_down

    except Exception:
        import traceback
        traceback.print_exc()


def register_audio_handler():
    # Clear existing handlers
    for h in list(bpy.app.handlers.frame_change_post):
        if getattr(h, "__name__", "") == "footstep_frame_handler":
            bpy.app.handlers.frame_change_post.remove(h)

    bpy.app.handlers.frame_change_post.append(footstep_frame_handler)
    print("[SUCCESS] Handler registered to frame_change_post.")


# Register audio handler on load
register_audio_handler()


# ==============================================================================
# CLEANUP & HELPERS
# ==============================================================================

def cleanup_previous_run():
    """Clears old scroll objects, ground, node groups, camera anchors, collections, and orphaned data."""
    target_names = [
        "Master_NLA_Control",
        "Scroll_GN_Object",
        "Scroll_Start_Pin",
        "Scroll_End_Pin",
        "Scroll_Master_Container",
        "HDRI_Ground_Plane",
        "Ground_Mesh_Object",
        "Camera_Focus_Target",
        "Vertical_Reel_Camera"
    ]
    for obj in list(bpy.data.objects):
        if obj.name in target_names or (obj.name.startswith("mixamorig") and obj.type != "ARMATURE" and obj.type != "MESH"):
            if obj.animation_data:
                obj.animation_data_clear()
            bpy.data.objects.remove(obj, do_unlink=True)

    for action in list(bpy.data.actions):
        if action.users == 0 or action.use_fake_user:
            action.use_fake_user = False
            bpy.data.actions.remove(action)

    for ng in list(bpy.data.node_groups):
        if ng.name in ["Scroll_Unfurl_GN", "Scroll_Pure_GN", "HDRI_Ground_GN", "Grass_Ground_GN"]:
            bpy.data.node_groups.remove(ng)

    for coll_name in ["Scroll_Master_Container", "Grass_Artifacts_Collection"]:
        if coll_name in bpy.data.collections:
            coll = bpy.data.collections[coll_name]
            for obj in list(coll.objects):
                bpy.data.objects.remove(obj, do_unlink=True)
            bpy.data.collections.remove(coll)

    bpy.context.view_layer.update()


def get_slot_fcurves(action, anim_target_object):
    fcurves = []
    if not action:
        return fcurves

    anim_data = getattr(anim_target_object, "animation_data", None)
    if anim_data and getattr(anim_data, "action_slot", None):
        slot = anim_data.action_slot
        channelbag = anim_utils.action_get_channelbag_for_slot(action, slot)
        if channelbag and hasattr(channelbag, "fcurves"):
            return list(channelbag.fcurves)

    if hasattr(action, "slots"):
        for slot in action.slots:
            for cb in getattr(slot, "channelbags", []):
                fcurves.extend(getattr(cb, "fcurves", []))
        if fcurves:
            return fcurves

    if hasattr(action, "fcurves"):
        return list(action.fcurves)

    return fcurves


def get_object_fcurves(obj):
    if not obj or not obj.animation_data or not obj.animation_data.action:
        return []
    action = obj.animation_data.action
    return get_slot_fcurves(action, obj)


def detect_natural_forward_axis(action, rig_object):
    fcurves = get_slot_fcurves(action, rig_object)
    max_delta = 0.0
    best_axis = 1
    net_delta = 0.0

    start_frame = action.frame_range[0]
    end_frame = action.frame_range[1]

    for fc in fcurves:
        if fc.data_path.endswith("location") and fc.array_index in [0, 1, 2]:
            start_val = fc.evaluate(start_frame)
            end_val = fc.evaluate(end_frame)
            delta = end_val - start_val
            if abs(delta) > abs(max_delta):
                max_delta = delta
                best_axis = fc.array_index
                net_delta = delta

    direction_sign = 1.0 if net_delta >= 0 else -1.0
    return best_axis, direction_sign


def find_hand_vertex_groups(target_mesh):
    r_name = "mixamorig:RightHand"
    l_name = "mixamorig:LeftHand"

    if target_mesh and hasattr(target_mesh, "vertex_groups"):
        for vg in target_mesh.vertex_groups:
            vg_lower = vg.name.lower()
            if "righthand" in vg_lower or "hand_r" in vg_lower or "hand.r" in vg_lower:
                r_name = vg.name
            elif "lefthand" in vg_lower or "hand_l" in vg_lower or "hand.l" in vg_lower:
                l_name = vg.name

    return r_name, l_name


# ==============================================================================
# PARENTED VERTICAL CAMERA SETUP (OPTION A)
# ==============================================================================

def setup_modern_reel_camera(target_rig, target_mesh, turn_start, turn_end_frame, unfurl_end_frame):
    scene = bpy.context.scene
    scene.render.resolution_x = 1080
    scene.render.resolution_y = 1920
    scene.render.resolution_percentage = 100

    # 1. Force evaluation of Frame 1
    scene.frame_set(1)
    bpy.context.view_layer.update()
    dg = bpy.context.evaluated_depsgraph_get()

    # 2. Get World-Space Bounding Box of Evaluated Mesh
    eval_mesh_obj = target_mesh.evaluated_get(dg)
    bbox_world = [eval_mesh_obj.matrix_world @ Vector(corner) for corner in eval_mesh_obj.bound_box]

    min_z = min(v.z for v in bbox_world)
    max_z = max(v.z for v in bbox_world)
    target_mid_z = min_z + ((max_z - min_z) * 0.55)

    root_bone_name = "mixamorig:Hips" if "mixamorig:Hips" in target_rig.pose.bones else target_rig.pose.bones[0].name

    # 3. Setup Unparented Objects
    focus_target = bpy.data.objects.get("Camera_Focus_Target") or bpy.data.objects.new("Camera_Focus_Target", None)
    if focus_target.name not in scene.collection.objects:
        scene.collection.objects.link(focus_target)

    cam_data = bpy.data.cameras.get("Vertical_Reel_Cam_Data") or bpy.data.cameras.new("Vertical_Reel_Cam_Data")
    cam_data.lens = 45.0
    cam_data.sensor_fit = 'VERTICAL'

    cam_obj = bpy.data.objects.get("Vertical_Reel_Camera") or bpy.data.objects.new("Vertical_Reel_Camera", cam_data)
    if cam_obj.name not in scene.collection.objects:
        scene.collection.objects.link(cam_obj)

    scene.camera = cam_obj
    cam_obj.parent = None

    cam_obj.constraints.clear()
    track = cam_obj.constraints.new(type='TRACK_TO')
    track.target = focus_target
    track.track_axis = 'TRACK_NEGATIVE_Z'
    track.up_axis = 'UP_Y'

    if cam_obj.animation_data:
        cam_obj.animation_data_clear()

    cam_dist = 4.5

    # 4. Keyframe Loop
    for f in range(1, unfurl_end_frame + 1):
        scene.frame_set(f)
        bpy.context.view_layer.update()
        frame_dg = bpy.context.evaluated_depsgraph_get()

        eval_rig = target_rig.evaluated_get(frame_dg)
        pbone = eval_rig.pose.bones[root_bone_name]

        bone_world_matrix = eval_rig.matrix_world @ pbone.matrix
        char_ground_loc = Vector((bone_world_matrix.translation.x, bone_world_matrix.translation.y, min_z))

        # Set focus target to true character mid-height
        focus_target.location = Vector((char_ground_loc.x, char_ground_loc.y, target_mid_z))
        focus_target.keyframe_insert(data_path="location", frame=f)

        # Determine horizontal facing direction from armature world matrix + rig orientation
        # (Avoids reading Hip joint bone roll)
        rig_matrix = eval_rig.matrix_world
        forward_vec = (rig_matrix.to_3x3() @ Vector((0, 1, 0))).xy.normalized()

        if forward_vec.length == 0:
            forward_vec = Vector((0, -1))

        # Base heading angle in world space
        base_angle = math.atan2(forward_vec.y, forward_vec.x)

        # Smooth angle offset: -45 deg for walk, 0 deg (front-facing) for unfurl
        if f <= turn_start:
            angle_offset = math.radians(-45.0)
        elif f >= turn_end_frame:
            angle_offset = 0.0
        else:
            t = (f - turn_start) / (turn_end_frame - turn_start)
            angle_offset = math.radians(-45.0) * (1.0 - t)

        cam_angle = base_angle + angle_offset

        # Position camera strictly on horizontal XY circle relative to character location
        cam_x = char_ground_loc.x - (cam_dist * math.cos(cam_angle))
        cam_y = char_ground_loc.y - (cam_dist * math.sin(cam_angle))

        cam_obj.location = Vector((cam_x, cam_y, target_mid_z))
        cam_obj.keyframe_insert(data_path="location", frame=f)

    # Set linear keyframes (compatible with Blender 3.x AND 4.x+)
    for obj in [cam_obj, focus_target]:
        if obj.animation_data and obj.animation_data.action:
            action = obj.animation_data.action
            # Blender 4.x+ Slot/Binding system check vs Legacy Blender 3.x
            fcurves = getattr(action, "fcurves", None)
            if fcurves is None and hasattr(action, "bindings"):
                # Blender 4.0+ stores fcurves inside action bindings or slots
                fcurves = [fc for binding in action.bindings for fc in binding.fcurves]

            if fcurves:
                for fc in fcurves:
                    for kp in fc.keyframe_points:
                        kp.interpolation = 'LINEAR'

    scene.frame_set(1)
    bpy.context.view_layer.update()

    return cam_obj, focus_target


# ==============================================================================
# GEOMETRY NODES GROUND, GRASS & SKY DOME SETUP
# ==============================================================================

def ensure_world_sphere_size(cam_obj, ground_obj, safety_multiplier=2.5):
    if not cam_obj or not ground_obj:
        return

    scene = bpy.context.scene
    max_cam_dist = 0.0

    for frame in range(scene.frame_start, scene.frame_end + 1, 5):
        scene.frame_set(frame)
        bpy.context.view_layer.update()

        cam_dist = cam_obj.matrix_world.translation.length
        if cam_dist > max_cam_dist:
            max_cam_dist = cam_dist

    world_radius = max(25.0, max_cam_dist * safety_multiplier)

    mod = ground_obj.modifiers.get("Grass_Ground_GN")
    if mod and mod.node_group:
        for node in mod.node_group.nodes:
            if node.bl_idname == "GeometryNodeMeshIcoSphere":
                node.inputs["Radius"].default_value = world_radius

    if ground_obj.type == 'MESH':
        bpy.context.view_layer.objects.active = ground_obj
        bpy.ops.object.mode_set(mode='OBJECT')

        ground_scale = world_radius / 12.5
        ground_obj.scale = (ground_scale, ground_scale, 1.0)
        bpy.ops.object.transform_apply(location=False, rotation=False, scale=True)

    cam_obj.data.clip_end = max(cam_obj.data.clip_end, world_radius * 2.5)

    scene.frame_set(1)
    bpy.context.view_layer.update()


def create_grass_material(grass_dir):
    mat = bpy.data.materials.new(name="Grass_PBR_Material")
    mat.use_nodes = True
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    out_node = nodes.new("ShaderNodeOutputMaterial")
    bsdf = nodes.new("ShaderNodeBsdfPrincipled")
    links.new(bsdf.outputs["BSDF"], out_node.inputs["Surface"])

    col_files = glob.glob(os.path.join(grass_dir, "*Color*.jpg")) + glob.glob(os.path.join(grass_dir, "*JPG*.jpg"))
    if col_files:
        tex_col = nodes.new("ShaderNodeTexImage")
        tex_col.image = bpy.data.images.load(col_files[0])
        links.new(tex_col.outputs["Color"], bsdf.inputs["Base Color"])

    rough_files = glob.glob(os.path.join(grass_dir, "*Roughness*.jpg"))
    if rough_files:
        tex_rough = nodes.new("ShaderNodeTexImage")
        tex_rough.image = bpy.data.images.load(rough_files[0])
        tex_rough.image.colorspace_settings.name = 'Non-Color'
        links.new(tex_rough.outputs["Color"], bsdf.inputs["Roughness"])

    norm_files = glob.glob(os.path.join(grass_dir, "*Normal*.jpg"))
    if norm_files:
        tex_norm = nodes.new("ShaderNodeTexImage")
        tex_norm.image = bpy.data.images.load(norm_files[0])
        tex_norm.image.colorspace_settings.name = 'Non-Color'

        normal_map_node = nodes.new("ShaderNodeNormalMap")
        links.new(tex_norm.outputs["Color"], normal_map_node.inputs["Color"])
        links.new(normal_map_node.outputs["Normal"], bsdf.inputs["Normal"])

    return mat


def create_sky_material(sky_hdri_path):
    mat = bpy.data.materials.new(name="Sky_Dome_Material")
    mat.use_nodes = True
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    out_node = nodes.new("ShaderNodeOutputMaterial")
    emission = nodes.new("ShaderNodeEmission")
    emission.inputs["Strength"].default_value = 1.0

    links.new(emission.outputs["Emission"], out_node.inputs["Surface"])

    if os.path.exists(sky_hdri_path):
        tex_env = nodes.new("ShaderNodeTexEnvironment")
        tex_env.image = bpy.data.images.load(sky_hdri_path)
        links.new(tex_env.outputs["Color"], emission.inputs["Color"])

    return mat


def setup_grass_ground_geometry_nodes(grass_dir, sky_hdri_path, ground_z_offset=-0.2):
    grass_coll = bpy.data.collections.get("Grass_Artifacts_Collection")
    if not grass_coll:
        grass_coll = bpy.data.collections.new("Grass_Artifacts_Collection")
        bpy.context.scene.collection.children.link(grass_coll)

    grass_mat = create_grass_material(grass_dir)
    sky_mat = create_sky_material(sky_hdri_path)

    mesh_files = glob.glob(os.path.join(grass_dir, "*.obj")) + glob.glob(os.path.join(grass_dir, "*.fbx"))

    if mesh_files:
        for fpath in mesh_files:
            if fpath.endswith(".obj"):
                if hasattr(bpy.ops.wm, "obj_import"):
                    bpy.ops.wm.obj_import(filepath=fpath)
                else:
                    bpy.ops.import_scene.obj(filepath=fpath)
            elif fpath.endswith(".fbx"):
                bpy.ops.import_scene.fbx(filepath=fpath)

            for o in bpy.context.selected_objects:
                grass_coll.objects.link(o)
                if o.name in bpy.context.scene.collection.objects:
                    bpy.context.scene.collection.objects.unlink(o)

        for obj in list(grass_coll.objects):
            if obj.dimensions.z > 0.5:
                grass_coll.objects.unlink(obj)
                bpy.data.objects.remove(obj, do_unlink=True)
    else:
        for i in range(3):
            m_data = bpy.data.meshes.new(f"Grass_Blade_Mesh_{i}")
            g_obj = bpy.data.objects.new(f"Grass_Blade_Obj_{i}", m_data)
            grass_coll.objects.link(g_obj)

            verts = [(0, 0, 0), (0.05, 0, 0), (0.04, 0, 0.15), (0.01, 0, 0.25), (0, 0, 0.3)]
            faces = [(0, 1, 2), (1, 2, 3), (2, 3, 4)]
            m_data.from_pydata(verts, [], faces)
            m_data.update()

            if g_obj.data.materials:
                g_obj.data.materials[0] = grass_mat
            else:
                g_obj.data.materials.append(grass_mat)

            g_obj.location = (100, 100, 0)

    bpy.ops.mesh.primitive_circle_add(radius=12.5, vertices=64, fill_type='NGON', location=(0, 0, ground_z_offset))
    ground_obj = bpy.context.active_object
    ground_obj.name = "Ground_Mesh_Object"

    vg = ground_obj.vertex_groups.new(name="Grass_Weight_Group")
    all_vert_indices = [v.index for v in ground_obj.data.vertices]
    vg.add(all_vert_indices, 1.0, 'REPLACE')
    ground_obj.data.materials.append(grass_mat)
    ground_obj.data.materials.append(sky_mat)

    gn_group = bpy.data.node_groups.new("Grass_Ground_GN", type="GeometryNodeTree")
    gn_group.interface.new_socket(name="Geometry", in_out="INPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Geometry", in_out="OUTPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Density Mask", in_out="INPUT", socket_type="NodeSocketFloat")

    nodes = gn_group.nodes
    links = gn_group.links
    out_node = nodes.new("NodeGroupOutput")
    in_node = nodes.new("NodeGroupInput")

    distribute_node = nodes.new("GeometryNodeDistributePointsOnFaces")
    distribute_node.distribute_method = 'POISSON'
    distribute_node.inputs["Distance Min"].default_value = 0.02
    distribute_node.inputs["Density Max"].default_value = 300.0
    links.new(in_node.outputs["Geometry"], distribute_node.inputs["Mesh"])

    noise_node = nodes.new("ShaderNodeTexNoise")
    noise_node.inputs["Scale"].default_value = 2.5
    noise_node.inputs["Detail"].default_value = 2.0

    color_ramp = nodes.new("ShaderNodeValToRGB")
    color_ramp.color_ramp.elements[0].position = 0.20
    color_ramp.color_ramp.elements[1].position = 0.80
    links.new(noise_node.outputs["Fac"], color_ramp.inputs["Fac"])

    math_weight = nodes.new("ShaderNodeMath")
    math_weight.operation = "MULTIPLY"
    links.new(color_ramp.outputs["Color"], math_weight.inputs[0])
    links.new(in_node.outputs["Density Mask"], math_weight.inputs[1])

    links.new(math_weight.outputs["Value"], distribute_node.inputs["Density Factor"])

    instance_node = nodes.new("GeometryNodeInstanceOnPoints")
    instance_node.inputs["Pick Instance"].default_value = True
    links.new(distribute_node.outputs["Points"], instance_node.inputs["Points"])

    coll_info = nodes.new("GeometryNodeCollectionInfo")
    coll_info.inputs["Collection"].default_value = grass_coll
    coll_info.inputs["Separate Children"].default_value = True
    coll_info.inputs["Reset Children"].default_value = True
    links.new(coll_info.outputs["Instances"], instance_node.inputs["Instance"])

    rand_scale = nodes.new("FunctionNodeRandomValue")
    rand_scale.data_type = 'FLOAT'
    rand_scale.inputs["Min"].default_value = 0.25
    rand_scale.inputs["Max"].default_value = 0.50

    combine_scale = nodes.new("ShaderNodeCombineXYZ")
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["X"])
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["Y"])
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["Z"])
    links.new(combine_scale.outputs["Vector"], instance_node.inputs["Scale"])

    rand_rot_z = nodes.new("FunctionNodeRandomValue")
    rand_rot_z.data_type = 'FLOAT'
    rand_rot_z.inputs["Min"].default_value = 0.0
    rand_rot_z.inputs["Max"].default_value = math.radians(360.0)

    combine_rot = nodes.new("ShaderNodeCombineXYZ")
    links.new(rand_rot_z.outputs["Value"], combine_rot.inputs["Z"])
    links.new(combine_rot.outputs["Vector"], instance_node.inputs["Rotation"])

    sky_sphere = nodes.new("GeometryNodeMeshIcoSphere")
    sky_sphere.inputs["Radius"].default_value = 12.5
    sky_sphere.inputs["Subdivisions"].default_value = 4

    flip_faces = nodes.new("GeometryNodeFlipFaces")
    links.new(sky_sphere.outputs["Mesh"], flip_faces.inputs["Mesh"])

    transform_sky = nodes.new("GeometryNodeTransform")
    transform_sky.inputs["Translation"].default_value = (0.0, 0.0, ground_z_offset)
    links.new(flip_faces.outputs["Mesh"], transform_sky.inputs["Geometry"])

    set_sky_mat = nodes.new("GeometryNodeSetMaterial")
    set_sky_mat.inputs["Material"].default_value = sky_mat
    links.new(transform_sky.outputs["Geometry"], set_sky_mat.inputs["Geometry"])

    join_geo = nodes.new("GeometryNodeJoinGeometry")
    links.new(in_node.outputs["Geometry"], join_geo.inputs["Geometry"])
    links.new(instance_node.outputs["Instances"], join_geo.inputs["Geometry"])
    links.new(set_sky_mat.outputs["Geometry"], join_geo.inputs["Geometry"])

    links.new(join_geo.outputs["Geometry"], out_node.inputs["Geometry"])

    mod = ground_obj.modifiers.new(name="Grass_Ground_GN", type='NODES')
    mod.node_group = gn_group

    for item in gn_group.interface.items_tree:
        if getattr(item, "in_out", None) == "INPUT" and item.name == "Density Mask":
            mod[item.identifier + "_attribute_name"] = "Grass_Weight_Group"
            mod[item.identifier + "_use_attribute"] = True

    return ground_obj


# ==============================================================================
# SCROLL MATERIAL & GEOMETRY NODES SETUP
# ==============================================================================

def create_image_scroll_material(image_path):
    mat_name = "Scroll_Texture_Mat"
    mat = bpy.data.materials.get(mat_name)
    if mat:
        bpy.data.materials.remove(mat)

    mat = bpy.data.materials.new(name=mat_name)
    mat.use_nodes = True
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    output_node = nodes.new("ShaderNodeOutputMaterial")
    bsdf = nodes.new("ShaderNodeBsdfPrincipled")
    bsdf.inputs["Roughness"].default_value = 0.5

    tex_img = nodes.new("ShaderNodeTexImage")
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        tex_img.image = img
        tex_img.extension = 'REPEAT'

    uv_map_node = nodes.new("ShaderNodeUVMap")
    uv_map_node.uv_map = "UVMap"

    links.new(uv_map_node.outputs["UV"], tex_img.inputs["Vector"])
    links.new(tex_img.outputs["Color"], bsdf.inputs["Base Color"])
    links.new(bsdf.outputs["BSDF"], output_node.inputs["Surface"])

    return mat, tex_img.image if os.path.exists(image_path) else None


def create_pure_geometry_node_scroll(image_path=""):
    gn_group = bpy.data.node_groups.new("Scroll_Pure_GN", type="GeometryNodeTree")

    gn_group.interface.new_socket(name="Geometry", in_out="OUTPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Character Mesh", in_out="INPUT", socket_type="NodeSocketObject")

    sock_r_vg = gn_group.interface.new_socket(name="Right Hand Group", in_out="INPUT", socket_type="NodeSocketString")
    sock_r_vg.default_value = "mixamorig:RightHand"

    sock_l_vg = gn_group.interface.new_socket(name="Left Hand Group", in_out="INPUT", socket_type="NodeSocketString")
    sock_l_vg.default_value = "mixamorig:LeftHand"

    sock_offset = gn_group.interface.new_socket(name="Forward Offset", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_offset.default_value = 0.60

    sock_unfurl = gn_group.interface.new_socket(name="Unfurl", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_unfurl.default_value = 0.0

    sock_width = gn_group.interface.new_socket(name="Scroll Width", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_width.default_value = 0.35

    nodes = gn_group.nodes
    links = gn_group.links

    in_node = nodes.new("NodeGroupInput")
    out_node = nodes.new("NodeGroupOutput")

    char_info = nodes.new("GeometryNodeObjectInfo")
    char_info.transform_space = "RELATIVE"
    links.new(in_node.outputs["Character Mesh"], char_info.inputs["Object"])

    pos_node = nodes.new("GeometryNodeInputPosition")

    named_r = nodes.new("GeometryNodeInputNamedAttribute")
    named_r.data_type = "FLOAT"
    links.new(in_node.outputs["Right Hand Group"], named_r.inputs["Name"])

    cmp_r = nodes.new("FunctionNodeCompare")
    cmp_r.data_type = "FLOAT"
    cmp_r.operation = "GREATER_THAN"
    cmp_r.inputs[1].default_value = 0.001
    links.new(named_r.outputs["Attribute"], cmp_r.inputs[0])

    stat_r = nodes.new("GeometryNodeAttributeStatistic")
    stat_r.data_type = "FLOAT_VECTOR"
    links.new(char_info.outputs["Geometry"], stat_r.inputs["Geometry"])
    links.new(cmp_r.outputs["Result"], stat_r.inputs["Selection"])
    links.new(pos_node.outputs["Position"], stat_r.inputs["Attribute"])

    named_l = nodes.new("GeometryNodeInputNamedAttribute")
    named_l.data_type = "FLOAT"
    links.new(in_node.outputs["Left Hand Group"], named_l.inputs["Name"])

    cmp_l = nodes.new("FunctionNodeCompare")
    cmp_l.data_type = "FLOAT"
    cmp_l.operation = "GREATER_THAN"
    cmp_l.inputs[1].default_value = 0.001
    links.new(named_l.outputs["Attribute"], cmp_l.inputs[0])

    stat_l = nodes.new("GeometryNodeAttributeStatistic")
    stat_l.data_type = "FLOAT_VECTOR"
    links.new(char_info.outputs["Geometry"], stat_l.inputs["Geometry"])
    links.new(cmp_l.outputs["Result"], stat_l.inputs["Selection"])
    links.new(pos_node.outputs["Position"], stat_l.inputs["Attribute"])

    neg_offset = nodes.new("ShaderNodeMath")
    neg_offset.operation = "MULTIPLY"
    neg_offset.inputs[1].default_value = -1.0
    links.new(in_node.outputs["Forward Offset"], neg_offset.inputs[0])

    forward_vec = nodes.new("ShaderNodeCombineXYZ")
    forward_vec.inputs["X"].default_value = 0.0
    links.new(neg_offset.outputs["Value"], forward_vec.inputs["Y"])
    forward_vec.inputs["Z"].default_value = 0.0

    start_pos = nodes.new("ShaderNodeVectorMath")
    start_pos.operation = "ADD"
    links.new(stat_r.outputs["Mean"], start_pos.inputs[0])
    links.new(forward_vec.outputs["Vector"], start_pos.inputs[1])

    target_pos = nodes.new("ShaderNodeVectorMath")
    target_pos.operation = "ADD"
    links.new(stat_l.outputs["Mean"], target_pos.inputs[0])
    links.new(forward_vec.outputs["Vector"], target_pos.inputs[1])

    delta_vec = nodes.new("ShaderNodeVectorMath")
    delta_vec.operation = "SUBTRACT"
    links.new(target_pos.outputs["Vector"], delta_vec.inputs[0])
    links.new(start_pos.outputs["Vector"], delta_vec.inputs[1])

    dist_node = nodes.new("ShaderNodeVectorMath")
    dist_node.operation = "DISTANCE"
    links.new(start_pos.outputs["Vector"], dist_node.inputs[0])

    scale_vec = nodes.new("ShaderNodeVectorMath")
    scale_vec.operation = "SCALE"
    links.new(delta_vec.outputs["Vector"], scale_vec.inputs["Vector"])
    links.new(in_node.outputs["Unfurl"], scale_vec.inputs["Scale"])

    end_pos = nodes.new("ShaderNodeVectorMath")
    end_pos.operation = "ADD"
    links.new(start_pos.outputs["Vector"], end_pos.inputs[0])
    links.new(scale_vec.outputs["Vector"], end_pos.inputs[1])

    links.new(end_pos.outputs["Vector"], dist_node.inputs[1])

    curve_line = nodes.new("GeometryNodeCurvePrimitiveLine")
    links.new(start_pos.outputs["Vector"], curve_line.inputs["Start"])
    links.new(end_pos.outputs["Vector"], curve_line.inputs["End"])

    set_tilt = nodes.new("GeometryNodeSetCurveTilt")
    set_tilt.inputs["Tilt"].default_value = 1.5708
    links.new(curve_line.outputs["Curve"], set_tilt.inputs["Curve"])

    spline_param_main = nodes.new("GeometryNodeSplineParameter")
    store_u = nodes.new("GeometryNodeStoreNamedAttribute")
    store_u.inputs["Name"].default_value = "U_Factor"
    store_u.data_type = 'FLOAT'
    store_u.domain = 'POINT'
    links.new(set_tilt.outputs["Curve"], store_u.inputs["Geometry"])
    links.new(spline_param_main.outputs["Factor"], store_u.inputs["Value"])

    half_width_neg = nodes.new("ShaderNodeMath")
    half_width_neg.operation = "MULTIPLY"
    half_width_neg.inputs[1].default_value = -0.5
    links.new(in_node.outputs["Scroll Width"], half_width_neg.inputs[0])

    half_width_pos = nodes.new("ShaderNodeMath")
    half_width_pos.operation = "MULTIPLY"
    half_width_pos.inputs[1].default_value = 0.5
    links.new(in_node.outputs["Scroll Width"], half_width_pos.inputs[0])

    combine_start = nodes.new("ShaderNodeCombineXYZ")
    links.new(half_width_neg.outputs["Value"], combine_start.inputs["X"])

    combine_end = nodes.new("ShaderNodeCombineXYZ")
    links.new(half_width_pos.outputs["Value"], combine_end.inputs["X"])

    profile_line = nodes.new("GeometryNodeCurvePrimitiveLine")
    links.new(combine_start.outputs["Vector"], profile_line.inputs["Start"])
    links.new(combine_end.outputs["Vector"], profile_line.inputs["End"])

    spline_param_prof = nodes.new("GeometryNodeSplineParameter")
    store_v = nodes.new("GeometryNodeStoreNamedAttribute")
    store_v.inputs["Name"].default_value = "V_Factor"
    store_v.data_type = 'FLOAT'
    store_v.domain = 'POINT'
    links.new(profile_line.outputs["Curve"], store_v.inputs["Geometry"])
    links.new(spline_param_prof.outputs["Factor"], store_v.inputs["Value"])

    curve_to_mesh = nodes.new("GeometryNodeCurveToMesh")
    links.new(store_u.outputs["Geometry"], curve_to_mesh.inputs["Curve"])
    links.new(store_v.outputs["Geometry"], curve_to_mesh.inputs["Profile Curve"])

    img_aspect = 1.0
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        if img.size[1] > 0:
            img_aspect = img.size[0] / img.size[1]

    h_times_aspect = nodes.new("ShaderNodeMath")
    h_times_aspect.operation = "MULTIPLY"
    h_times_aspect.inputs[1].default_value = img_aspect
    links.new(in_node.outputs["Scroll Width"], h_times_aspect.inputs[0])

    tile_count = nodes.new("ShaderNodeMath")
    tile_count.operation = "DIVIDE"
    links.new(dist_node.outputs["Value"], tile_count.inputs[0])
    links.new(h_times_aspect.outputs["Value"], tile_count.inputs[1])

    read_u = nodes.new("GeometryNodeInputNamedAttribute")
    read_u.inputs["Name"].default_value = "U_Factor"
    read_u.data_type = 'FLOAT'

    scaled_u = nodes.new("ShaderNodeMath")
    scaled_u.operation = "MULTIPLY"
    links.new(read_u.outputs["Attribute"], scaled_u.inputs[0])
    links.new(tile_count.outputs["Value"], scaled_u.inputs[1])

    read_v = nodes.new("GeometryNodeInputNamedAttribute")
    read_v.inputs["Name"].default_value = "V_Factor"
    read_v.data_type = 'FLOAT'

    flip_v = nodes.new("ShaderNodeMath")
    flip_v.operation = "SUBTRACT"
    flip_v.inputs[0].default_value = 1.0
    links.new(read_v.outputs["Attribute"], flip_v.inputs[1])

    combine_uv = nodes.new("ShaderNodeCombineXYZ")
    links.new(scaled_u.outputs["Value"], combine_uv.inputs["X"])
    links.new(flip_v.outputs["Value"], combine_uv.inputs["Y"])

    store_uv = nodes.new("GeometryNodeStoreNamedAttribute")
    store_uv.inputs["Name"].default_value = "UVMap"
    store_uv.data_type = 'FLOAT2'
    store_uv.domain = 'CORNER'
    links.new(curve_to_mesh.outputs["Mesh"], store_uv.inputs["Geometry"])
    links.new(combine_uv.outputs["Vector"], store_uv.inputs["Value"])

    set_material = nodes.new("GeometryNodeSetMaterial")
    scroll_mat, _ = create_image_scroll_material(image_path)
    set_material.inputs["Material"].default_value = scroll_mat
    links.new(store_uv.outputs["Geometry"], set_material.inputs["Geometry"])

    links.new(set_material.outputs["Geometry"], out_node.inputs["Geometry"])

    return gn_group, scroll_mat


def setup_geometry_node_scroll(target_mesh, unfurl_start, unfurl_end, image_path=""):
    mesh_data = bpy.data.meshes.new("Scroll_Procedural_Mesh")
    scroll_obj = bpy.data.objects.new("Scroll_GN_Object", mesh_data)
    bpy.context.scene.collection.objects.link(scroll_obj)

    mod = scroll_obj.modifiers.new("Scroll_GN", type="NODES")
    gn_tree, scroll_mat = create_pure_geometry_node_scroll(image_path=image_path)
    mod.node_group = gn_tree

    scroll_obj.data.materials.append(scroll_mat)

    r_hand_name, l_hand_name = find_hand_vertex_groups(target_mesh)

    unfurl_id = None
    for item in gn_tree.interface.items_tree:
        if getattr(item, "in_out", None) == "INPUT":
            if item.name == "Character Mesh":
                mod[item.identifier] = target_mesh
            elif item.name == "Right Hand Group":
                mod[item.identifier] = r_hand_name
            elif item.name == "Left Hand Group":
                mod[item.identifier] = l_hand_name
            elif item.name == "Forward Offset":
                mod[item.identifier] = 0.60
            elif item.name == "Scroll Width":
                mod[item.identifier] = 0.35
            elif item.name == "Unfurl":
                unfurl_id = item.identifier

    for m in target_mesh.modifiers:
        if m.type == 'ARMATURE':
            m.show_viewport = True
            m.show_render = True
            m.show_in_editmode = True

    scroll_obj.hide_viewport = True
    scroll_obj.hide_render = True
    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=1)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=1)

    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=unfurl_start - 1)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=unfurl_start - 1)

    scroll_obj.hide_viewport = False
    scroll_obj.hide_render = False
    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=unfurl_start)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=unfurl_start)

    for fc in get_object_fcurves(scroll_obj):
        if fc.data_path in ["hide_viewport", "hide_render"]:
            for kp in fc.keyframe_points:
                kp.interpolation = 'CONSTANT'

    if unfurl_id:
        mod[unfurl_id] = 0.001
        mod.keyframe_insert(data_path=f'["{unfurl_id}"]', frame=unfurl_start)

        mod[unfurl_id] = 1.0
        mod.keyframe_insert(data_path=f'["{unfurl_id}"]', frame=unfurl_end)

    return scroll_obj


# ==============================================================================
# MAIN CONTROLLER
# ==============================================================================

def import_and_extract_action(fbx_path, action_name):
    existing_objects = set(bpy.data.objects)
    bpy.ops.import_scene.fbx(filepath=fbx_path)
    bpy.context.view_layer.update()

    imported_objects = set(bpy.data.objects) - existing_objects
    extracted_action = None
    for obj in imported_objects:
        if obj.animation_data and obj.animation_data.action:
            extracted_action = obj.animation_data.action
            extracted_action.name = action_name
            extracted_action.use_fake_user = True
            break
    return extracted_action, imported_objects


def purge_imported_objects(objects_to_remove):
    for obj in objects_to_remove:
        if obj.name in bpy.data.objects:
            if obj.animation_data:
                obj.animation_data_clear()
            for coll in list(obj.users_collection):
                coll.objects.unlink(obj)
    bpy.context.view_layer.update()


def main():
    cleanup_previous_run()

    walk_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Mutant Walking.fbx"
    turn_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Right Turn 90.fbx"
    unfurl_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Mutant Idle.fbx"

    sky_hdri_path = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/horn-koppe_spring_4k.exr"
    output_blend = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/output_scene.blend"

    image_path = "/Users/dobbinp1/Downloads/IMGP6691_psedit.jpg"
    grass_dir = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Grass004_4K-JPG"

    ground_obj = setup_grass_ground_geometry_nodes(grass_dir, sky_hdri_path, ground_z_offset=-0.2)

    CYCLES = 4
    CROSSFADE = 15
    TARGET_DISTANCE = 4.0

    walk_action, walk_objs = import_and_extract_action(walk_fbx, "walk")
    target_rig = next((obj for obj in walk_objs if obj.type == "ARMATURE"), None)
    target_mesh = next((obj for obj in walk_objs if obj.type == "MESH"), None)

    if not target_rig or not target_mesh:
        raise RuntimeError("Armature or Character Mesh not found in walk FBX.")

    travel_axis_idx, motion_sign = detect_natural_forward_axis(walk_action, target_rig)
    initial_heading = math.radians(90.0) if travel_axis_idx == 0 else 0.0
    target_rig.rotation_euler[2] = initial_heading

    action_len = int(walk_action.frame_range[1] - walk_action.frame_range[0])
    if action_len <= 0:
        action_len = 30

    anim_data = target_rig.animation_data_create()

    walk_track = anim_data.nla_tracks.new()
    walk_track.name = "Walk_Track"

    current_frame = 1
    for i in range(CYCLES):
        strip = walk_track.strips.new(name=f"Walk_{i+1}", start=int(current_frame), action=walk_action)
        strip.blend_type = "REPLACE"
        current_frame += action_len

    walk_end_frame = int(current_frame)

    turn_action, turn_objs = import_and_extract_action(turn_fbx, "rightturn")
    purge_imported_objects(turn_objs)

    turn_track = anim_data.nla_tracks.new()
    turn_track.name = "RightTurn_Track"

    turn_start = int(walk_end_frame - CROSSFADE)
    turn_strip = turn_track.strips.new(name="RightTurn", start=turn_start, action=turn_action)
    turn_strip.blend_type = "REPLACE"
    turn_strip.blend_in = CROSSFADE
    turn_strip.use_auto_blend = True
    turn_strip.extrapolation = "HOLD"
    turn_end_frame = int(turn_strip.frame_end)

    # Pass turn_start frame to scene data so handler stops sound when character stops walking
    bpy.context.scene["turn_start_frame"] = turn_start

    unfurl_action, unfurl_objs = import_and_extract_action(unfurl_fbx, "unfurl")
    purge_imported_objects(unfurl_objs)

    unfurl_track = anim_data.nla_tracks.new()
    unfurl_track.name = "Unfurl_Track"

    unfurl_start = int(turn_end_frame - CROSSFADE)
    unfurl_strip = unfurl_track.strips.new(name="Unfurl", start=unfurl_start, action=unfurl_action)
    unfurl_strip.blend_type = "REPLACE"
    unfurl_strip.blend_in = CROSSFADE
    unfurl_strip.use_auto_blend = True
    unfurl_strip.extrapolation = "HOLD"
    unfurl_end_frame = int(unfurl_strip.frame_end)

    root_action = bpy.data.actions.new(name="World_Root_Motion")

    # Ensure animation_data exists and assign action before accessing channelbags/fcurves
    if not target_rig.animation_data:
        target_rig.animation_data_create()
    target_rig.animation_data.action = root_action

    # Modernized F-Curve creation logic for single multi-channel action compatibility
    if hasattr(root_action, "fcurves") and hasattr(root_action.fcurves, "new"):
        loc_fcurve = root_action.fcurves.new(data_path="location", index=travel_axis_idx)
        rot_fcurve = root_action.fcurves.new(data_path="rotation_euler", index=2)
    else:
        loc_fcurve = root_action.fcurve_ensure_for_datablock(
            datablock=target_rig, data_path="location", index=travel_axis_idx
        )
        rot_fcurve = root_action.fcurve_ensure_for_datablock(
            datablock=target_rig, data_path="rotation_euler", index=2
        )

    final_target_value = TARGET_DISTANCE * motion_sign

    k_loc1 = loc_fcurve.keyframe_points.insert(frame=1.0, value=0.0)
    k_loc2 = loc_fcurve.keyframe_points.insert(frame=float(turn_start), value=final_target_value)
    k_loc3 = loc_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=final_target_value)

    k_loc1.interpolation = "LINEAR"
    k_loc2.interpolation = "CONSTANT"
    k_loc3.interpolation = "CONSTANT"

    target_heading = initial_heading - math.radians(90.0)

    k_rot1 = rot_fcurve.keyframe_points.insert(frame=1.0, value=initial_heading)
    k_rot2 = rot_fcurve.keyframe_points.insert(frame=float(turn_start), value=initial_heading)
    k_rot3 = rot_fcurve.keyframe_points.insert(frame=float(turn_end_frame), value=target_heading)
    k_rot4 = rot_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=target_heading)

    k_rot1.interpolation = "CONSTANT"
    k_rot2.interpolation = "BEZIER"
    k_rot3.interpolation = "CONSTANT"
    k_rot4.interpolation = "CONSTANT"

    root_track = anim_data.nla_tracks.new()
    root_track.name = "World_Root_Track"
    root_strip = root_track.strips.new(name="Root_Motion_Strip", start=1, action=root_action)
    root_strip.blend_type = "REPLACE"
    root_strip.extrapolation = "HOLD"

    anim_data.action = None

    setup_geometry_node_scroll(
        target_mesh=target_mesh,
        unfurl_start=unfurl_start,
        unfurl_end=unfurl_end_frame,
        image_path=image_path,
    )

    cam_obj, focus_target = setup_modern_reel_camera(
        target_rig=target_rig,
        target_mesh=target_mesh,
        turn_start=turn_start,
        turn_end_frame=turn_end_frame,
        unfurl_end_frame=unfurl_end_frame
    )

    bpy.context.scene.frame_start = 1
    bpy.context.scene.frame_end = unfurl_end_frame

    # Animate camera zoom to frame the scroll tile perfectly
    final_frame = frame_scroll_image(
        cam_obj=cam_obj,
        focus_target=focus_target,
        target_mesh=target_mesh,
        image_path=image_path,
        scroll_width=0.35,
        zoom_duration=45
    )

    ensure_world_sphere_size(cam_obj, ground_obj, safety_multiplier=2.5)

    # Register real-time footstep audio handler
    register_audio_handler()

    bpy.ops.wm.save_as_mainfile(filepath=output_blend)
    print(f"\n[SUCCESS] Integrated parented target tracking & saved scene to: {output_blend}")


if __name__ == "__main__":
    main()
```

```
import math
import mathutils
import os
import sys
import glob
import bpy
import aud
from bpy_extras import anim_utils
from bpy_extras.object_utils import world_to_camera_view
from bpy.app.handlers import persistent
from mathutils import Vector

# ==============================================================================
# CONFIGURATION
# ==============================================================================
FOOTSTEP_SOUND_PATH = "/Users/dobbinp1/Downloads/stomp.wav"
GROUND_THRESHOLD = 0.24  # Raised slightly for testing

# Setup Audio Device
if "_audio_store" not in sys.modules:
    import types
    sys.modules["_audio_store"] = types.ModuleType("_audio_store")

audio_store = sys.modules["_audio_store"]
if not hasattr(audio_store, "device"):
    audio_store.device = aud.Device()

AUDIO_DEVICE = audio_store.device
SOUND_FACTORY = None
AUDIO_HANDLES = []

if os.path.exists(FOOTSTEP_SOUND_PATH):
    SOUND_FACTORY = aud.Sound(FOOTSTEP_SOUND_PATH).cache()
    print(f"[AUDIO] Loaded sound file successfully.")
else:
    print(f"[AUDIO ERROR] File not found at: {FOOTSTEP_SOUND_PATH}")

_left_foot_was_down = False
_right_foot_was_down = False
_last_frame = -1


def play_footstep():
    global AUDIO_HANDLES
    if SOUND_FACTORY:
        print("  >>> PLAYING FOOTSTEP SOUND <<<")
        handle = AUDIO_DEVICE.play(SOUND_FACTORY)
        AUDIO_HANDLES.append(handle)
        if len(AUDIO_HANDLES) > 10:
            AUDIO_HANDLES.pop(0)


def find_foot_bone(pose_bones, side="Left"):
    target_keywords = [
        f"mixamorig:{side}foot",
        f"{side.lower()}foot",
        f"foot.{side[0].lower()}",
        f"foot_{side[0].lower()}"
    ]
    for bone in pose_bones:
        b_name = bone.name.lower()
        if any(kw in b_name for kw in target_keywords):
            return bone
    return None


@persistent
def footstep_frame_handler(scene):
    global _left_foot_was_down, _right_foot_was_down, _last_frame

    try:
        current_frame = scene.frame_current
        turn_start = scene.get("turn_start_frame", None)

        # Stop footsteps once turning/idling starts
        if turn_start is not None and current_frame >= turn_start:
            _left_foot_was_down = False
            _right_foot_was_down = False
            return

        if current_frame <= _last_frame or abs(current_frame - _last_frame) > 10:
            _left_foot_was_down = False
            _right_foot_was_down = False
        _last_frame = current_frame

        rig = next((o for o in scene.objects if o.type == 'ARMATURE'), None)
        if not rig:
            return

        depsgraph = bpy.context.evaluated_depsgraph_get()
        eval_rig = rig.evaluated_get(depsgraph)

        left_bone = find_foot_bone(eval_rig.pose.bones, "Left")
        right_bone = find_foot_bone(eval_rig.pose.bones, "Right")

        if not left_bone or not right_bone:
            return

        l_z = (eval_rig.matrix_world @ left_bone.matrix).translation.z
        r_z = (eval_rig.matrix_world @ right_bone.matrix).translation.z

        l_is_down = l_z <= GROUND_THRESHOLD
        r_is_down = r_z <= GROUND_THRESHOLD

        if l_is_down and not _left_foot_was_down:
            play_footstep()

        if r_is_down and not _right_foot_was_down:
            play_footstep()

        _left_foot_was_down = l_is_down
        _right_foot_was_down = r_is_down

    except Exception:
        import traceback
        traceback.print_exc()


def register_audio_handler():
    # Clear existing handlers
    for h in list(bpy.app.handlers.frame_change_post):
        if getattr(h, "__name__", "") == "footstep_frame_handler":
            bpy.app.handlers.frame_change_post.remove(h)

    bpy.app.handlers.frame_change_post.append(footstep_frame_handler)
    print("[SUCCESS] Handler registered to frame_change_post.")


# Register audio handler on load
register_audio_handler()


# ==============================================================================
# CLEANUP & HELPERS
# ==============================================================================

def cleanup_previous_run():
    """Clears old scroll objects, ground, node groups, camera anchors, collections, and orphaned data."""
    target_names = [
        "Master_NLA_Control",
        "Scroll_GN_Object",
        "Scroll_Start_Pin",
        "Scroll_End_Pin",
        "Scroll_Master_Container",
        "HDRI_Ground_Plane",
        "Ground_Mesh_Object",
        "Camera_Focus_Target",
        "Vertical_Reel_Camera"
    ]
    for obj in list(bpy.data.objects):
        if obj.name in target_names or (obj.name.startswith("mixamorig") and obj.type != "ARMATURE" and obj.type != "MESH"):
            if obj.animation_data:
                obj.animation_data_clear()
            bpy.data.objects.remove(obj, do_unlink=True)

    for action in list(bpy.data.actions):
        if action.users == 0 or action.use_fake_user:
            action.use_fake_user = False
            bpy.data.actions.remove(action)

    for ng in list(bpy.data.node_groups):
        if ng.name in ["Scroll_Unfurl_GN", "Scroll_Pure_GN", "HDRI_Ground_GN", "Grass_Ground_GN"]:
            bpy.data.node_groups.remove(ng)

    for coll_name in ["Scroll_Master_Container", "Grass_Artifacts_Collection"]:
        if coll_name in bpy.data.collections:
            coll = bpy.data.collections[coll_name]
            for obj in list(coll.objects):
                bpy.data.objects.remove(obj, do_unlink=True)
            bpy.data.collections.remove(coll)

    bpy.context.view_layer.update()


def get_slot_fcurves(action, anim_target_object):
    fcurves = []
    if not action:
        return fcurves

    anim_data = getattr(anim_target_object, "animation_data", None)
    if anim_data and getattr(anim_data, "action_slot", None):
        slot = anim_data.action_slot
        channelbag = anim_utils.action_get_channelbag_for_slot(action, slot)
        if channelbag and hasattr(channelbag, "fcurves"):
            return list(channelbag.fcurves)

    if hasattr(action, "slots"):
        for slot in action.slots:
            for cb in getattr(slot, "channelbags", []):
                fcurves.extend(getattr(cb, "fcurves", []))
        if fcurves:
            return fcurves

    if hasattr(action, "fcurves"):
        return list(action.fcurves)

    return fcurves


def get_object_fcurves(obj):
    if not obj or not obj.animation_data or not obj.animation_data.action:
        return []
    action = obj.animation_data.action
    return get_slot_fcurves(action, obj)


def detect_natural_forward_axis(action, rig_object):
    fcurves = get_slot_fcurves(action, rig_object)
    max_delta = 0.0
    best_axis = 1
    net_delta = 0.0

    start_frame = action.frame_range[0]
    end_frame = action.frame_range[1]

    for fc in fcurves:
        if fc.data_path.endswith("location") and fc.array_index in [0, 1, 2]:
            start_val = fc.evaluate(start_frame)
            end_val = fc.evaluate(end_frame)
            delta = end_val - start_val
            if abs(delta) > abs(max_delta):
                max_delta = delta
                best_axis = fc.array_index
                net_delta = delta

    direction_sign = 1.0 if net_delta >= 0 else -1.0
    return best_axis, direction_sign


def find_hand_vertex_groups(target_mesh):
    r_name = "mixamorig:RightHand"
    l_name = "mixamorig:LeftHand"

    if target_mesh and hasattr(target_mesh, "vertex_groups"):
        for vg in target_mesh.vertex_groups:
            vg_lower = vg.name.lower()
            if "righthand" in vg_lower or "hand_r" in vg_lower or "hand.r" in vg_lower:
                r_name = vg.name
            elif "lefthand" in vg_lower or "hand_l" in vg_lower or "hand.l" in vg_lower:
                l_name = vg.name

    return r_name, l_name


# ==============================================================================
# PARENTED VERTICAL CAMERA SETUP (OPTION A)
# ==============================================================================

def setup_modern_reel_camera(target_rig, target_mesh, turn_start, turn_end_frame, unfurl_end_frame):
    scene = bpy.context.scene
    scene.render.resolution_x = 1080
    scene.render.resolution_y = 1920
    scene.render.resolution_percentage = 100

    target_height = target_mesh.dimensions.z if target_mesh.dimensions.z > 0 else 1.8
    midpoint_z_offset = target_height * 0.55

    # 1. Setup parented camera focus target directly on target_mesh
    focus_target = bpy.data.objects.get("Camera_Focus_Target")
    if not focus_target:
        focus_target = bpy.data.objects.new("Camera_Focus_Target", None)
        focus_target.empty_display_type = 'PLAIN_AXES'
        focus_target.empty_display_size = 0.25
        scene.collection.objects.link(focus_target)

    focus_target.constraints.clear()
    focus_target.parent = target_mesh
    focus_target.matrix_parent_inverse = target_mesh.matrix_world.inverted()
    focus_target.location = (0.0, 0.0, midpoint_z_offset)

    # 2. Camera Setup
    cam_data = bpy.data.cameras.get("Vertical_Reel_Cam_Data")
    if not cam_data:
        cam_data = bpy.data.cameras.new("Vertical_Reel_Cam_Data")
    cam_data.lens = 45.0
    cam_data.sensor_fit = 'VERTICAL'
    cam_data.clip_end = 100.0

    cam_obj = bpy.data.objects.get("Vertical_Reel_Camera")
    if not cam_obj:
        cam_obj = bpy.data.objects.new("Vertical_Reel_Camera", cam_data)
        scene.collection.objects.link(cam_obj)

    scene.camera = cam_obj
    cam_obj.parent = None

    track = cam_obj.constraints.get("Track To")
    if not track:
        track = cam_obj.constraints.new(type='TRACK_TO')
    track.target = focus_target
    track.track_axis = 'TRACK_NEGATIVE_Z'
    track.up_axis = 'UP_Y'

    def check_mesh_in_frame(frame_num, cam_location, margin=0.08):
        cam_obj.location = cam_location
        scene.frame_set(int(frame_num))
        bpy.context.view_layer.update()

        depsgraph = bpy.context.evaluated_depsgraph_get()
        eval_mesh_obj = target_mesh.evaluated_get(depsgraph)
        mesh_data = eval_mesh_obj.data

        vert_step = max(1, len(mesh_data.vertices) // 300)
        world_matrix = target_mesh.matrix_world

        min_x, max_x = 1.0, 0.0
        min_y, max_y = 1.0, 0.0

        for i in range(0, len(mesh_data.vertices), vert_step):
            v_world = world_matrix @ mesh_data.vertices[i].co
            co_2d = world_to_camera_view(scene, cam_obj, v_world)

            if co_2d.x < min_x: min_x = co_2d.x
            if co_2d.x > max_x: max_x = co_2d.x
            if co_2d.y < min_y: min_y = co_2d.y
            if co_2d.y > max_y: max_y = co_2d.y

        in_bounds = (min_x >= margin and max_x <= (1.0 - margin) and
                     min_y >= margin and max_y <= (1.0 - margin))

        return in_bounds, min_x, max_x, min_y, max_y

    def find_fit_distance(frame_num, angle_rad, base_height_offset, initial_dist=5.0, margin=0.08):
        scene.frame_set(int(frame_num))
        bpy.context.view_layer.update()

        center_pos = focus_target.matrix_world.translation.copy()
        heading = target_rig.matrix_world.to_euler('XYZ').z
        view_angle = heading + angle_rad

        dist = initial_dist
        for _ in range(30):
            test_x = center_pos.x + dist * math.sin(view_angle)
            test_y = center_pos.y - dist * math.cos(view_angle)
            test_z = center_pos.z + base_height_offset

            fit, min_x, max_x, min_y, max_y = check_mesh_in_frame(frame_num, (test_x, test_y, test_z), margin=margin)
            if fit:
                break
            dist += 0.4

        return dist, center_pos, view_angle

    # --- Framing Calculation for Walking Phase ---
    max_walk_dist = 5.0
    walk_frames_to_sample = range(1, turn_start + 1, 4)

    for f in walk_frames_to_sample:
        req_d, _, _ = find_fit_distance(f, angle_rad=math.radians(-45.0), base_height_offset=0.0, initial_dist=max_walk_dist, margin=0.10)
        if req_d > max_walk_dist:
            max_walk_dist = req_d

    # Lock Frame 1 framing explicitly
    scene.frame_set(1)
    bpy.context.view_layer.update()
    c1 = focus_target.matrix_world.translation.copy()
    h1 = target_rig.matrix_world.to_euler('XYZ').z + math.radians(-45.0)
    w_x1 = c1.x + max_walk_dist * math.sin(h1)
    w_y1 = c1.y - max_walk_dist * math.cos(h1)
    w_z1 = c1.z

    # Turn start position framing
    scene.frame_set(turn_start)
    bpy.context.view_layer.update()
    cts = focus_target.matrix_world.translation.copy()
    hts = target_rig.matrix_world.to_euler('XYZ').z + math.radians(-45.0)
    w_xts = cts.x + max_walk_dist * math.sin(hts)
    w_yts = cts.y - max_walk_dist * math.cos(hts)
    w_zts = cts.z

    # --- Framing Calculation for Unfurl Phase ---
    unfurl_frames_to_sample = range(turn_end_frame, unfurl_end_frame + 1, 6)
    max_unfurl_dist = max_walk_dist

    for f in unfurl_frames_to_sample:
        req_d, _, _ = find_fit_distance(f, angle_rad=math.radians(0.0), base_height_offset=0.0, initial_dist=max_unfurl_dist, margin=0.10)
        if req_d > max_unfurl_dist:
            max_unfurl_dist = req_d

    scene.frame_set(turn_end_frame)
    bpy.context.view_layer.update()
    cte = focus_target.matrix_world.translation.copy()
    hte = target_rig.matrix_world.to_euler('XYZ').z + math.radians(0.0)
    f_x = cte.x + max_unfurl_dist * math.sin(hte)
    f_y = cte.y - max_unfurl_dist * math.cos(hte)
    f_z = cte.z

    if cam_obj.animation_data:
        cam_obj.animation_data_clear()

    # Keyframe 1: Monster visible and centered from frame 1
    cam_obj.location = (w_x1, w_y1, w_z1)
    cam_obj.keyframe_insert(data_path="location", frame=1)

    # Keyframe 2: Maintain tracking offset through turn start
    cam_obj.location = (w_xts, w_yts, w_zts)
    cam_obj.keyframe_insert(data_path="location", frame=turn_start)

    # Keyframe 3: Settle position post-turn
    cam_obj.location = (f_x, f_y, f_z)
    cam_obj.keyframe_insert(data_path="location", frame=turn_end_frame)

    # Keyframe 4: Hold framing through scroll unfurl
    cam_obj.location = (f_x, f_y, f_z)
    cam_obj.keyframe_insert(data_path="location", frame=unfurl_end_frame)

    for fc in get_object_fcurves(cam_obj):
        for kp in fc.keyframe_points:
            kp.interpolation = 'BEZIER'

    scene.frame_set(1)
    return cam_obj, focus_target


def frame_scroll_image(cam_obj, focus_target, target_mesh, image_path, scroll_width=0.35, zoom_duration=45):
    scene = bpy.context.scene
    unfurl_end_frame = scene.frame_end
    zoom_end_frame = unfurl_end_frame + zoom_duration

    scene.frame_set(unfurl_end_frame)
    bpy.context.view_layer.update()

    # 1. Aspect Ratio & Framing Distance Calculation
    img_aspect = 1.0
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        if img.size[1] > 0:
            img_aspect = img.size[0] / img.size[1]

    tile_height = scroll_width * img_aspect
    cam_data = cam_obj.data
    aspect_ratio = scene.render.resolution_x / scene.render.resolution_y
    sensor_width = cam_data.sensor_height * aspect_ratio

    fov_h_half = math.atan((sensor_width / 2.0) / cam_data.lens)
    target_framing_distance = (tile_height / 2.0) / math.tan(fov_h_half)

    # 2. Hand Position & Center Evaluation
    mesh_matrix = target_mesh.matrix_world
    depsgraph = bpy.context.evaluated_depsgraph_get()
    eval_mesh = target_mesh.evaluated_get(depsgraph)

    r_hand_name, l_hand_name = find_hand_vertex_groups(target_mesh)
    vg_r = eval_mesh.vertex_groups.get(r_hand_name)
    vg_l = eval_mesh.vertex_groups.get(l_hand_name)

    r_verts = [mesh_matrix @ v.co for v in eval_mesh.data.vertices if any(g.group == vg_r.index for g in v.groups)] if vg_r else []
    l_verts = [mesh_matrix @ v.co for v in eval_mesh.data.vertices if any(g.group == vg_l.index for g in v.groups)] if vg_l else []

    target_height = target_mesh.dimensions.z if target_mesh.dimensions.z > 0 else 1.8
    torso_fallback = mesh_matrix.translation + mathutils.Vector((0, 0, target_height * 0.55))

    r_hand_pos = sum(r_verts, mathutils.Vector((0, 0, 0))) / len(r_verts) if r_verts else torso_fallback
    l_hand_pos = sum(l_verts, mathutils.Vector((0, 0, 0))) / len(l_verts) if l_verts else torso_fallback

    scroll_midpoint = (r_hand_pos + l_hand_pos) * 0.5

    # 3. Guaranteed Outward Normal (Pointing TOWARD current camera)
    cam_pos = cam_obj.matrix_world.translation.copy()
    cam_dir = cam_pos - scroll_midpoint
    cam_dir.z = 0.0  # Horizontal plane projection

    if cam_dir.length > 0.001:
        scroll_normal = cam_dir.normalized()
    else:
        scroll_normal = mathutils.Vector((0, -1, 0))

    tile_center = scroll_midpoint + (scroll_normal * 0.10)
    end_pos = tile_center + (scroll_normal * target_framing_distance)

    # 4. OPTION A: Unparent Focus Target & Keyframe Zoom Smoothly
    current_world_focus = focus_target.matrix_world.translation.copy()

    focus_target.parent = None
    focus_target.matrix_world.translation = current_world_focus

    # Keyframe focus target sliding from current position to tile_center
    focus_target.location = current_world_focus
    focus_target.keyframe_insert(data_path="location", frame=unfurl_end_frame)
    focus_target.location = tile_center
    focus_target.keyframe_insert(data_path="location", frame=zoom_end_frame)

    # Keyframe camera moving from current position to end_pos
    cam_obj.location = cam_pos
    cam_obj.keyframe_insert(data_path="location", frame=unfurl_end_frame)
    cam_obj.location = end_pos
    cam_obj.keyframe_insert(data_path="location", frame=zoom_end_frame)

    # Clamp F-Curve handles to prevent curve overshoot
    for obj in [cam_obj, focus_target]:
        for fc in get_object_fcurves(obj):
            for kp in fc.keyframe_points:
                if kp.co.x >= unfurl_end_frame:
                    kp.interpolation = 'BEZIER'
                    kp.handle_left_type = 'AUTO_CLAMPED'
                    kp.handle_right_type = 'AUTO_CLAMPED'

    scene.frame_end = zoom_end_frame
    return zoom_end_frame


# ==============================================================================
# GEOMETRY NODES GROUND, GRASS & SKY DOME SETUP
# ==============================================================================

def ensure_world_sphere_size(cam_obj, ground_obj, safety_multiplier=2.5):
    if not cam_obj or not ground_obj:
        return

    scene = bpy.context.scene
    max_cam_dist = 0.0

    for frame in range(scene.frame_start, scene.frame_end + 1, 5):
        scene.frame_set(frame)
        bpy.context.view_layer.update()

        cam_dist = cam_obj.matrix_world.translation.length
        if cam_dist > max_cam_dist:
            max_cam_dist = cam_dist

    world_radius = max(25.0, max_cam_dist * safety_multiplier)

    mod = ground_obj.modifiers.get("Grass_Ground_GN")
    if mod and mod.node_group:
        for node in mod.node_group.nodes:
            if node.bl_idname == "GeometryNodeMeshIcoSphere":
                node.inputs["Radius"].default_value = world_radius

    if ground_obj.type == 'MESH':
        bpy.context.view_layer.objects.active = ground_obj
        bpy.ops.object.mode_set(mode='OBJECT')

        ground_scale = world_radius / 12.5
        ground_obj.scale = (ground_scale, ground_scale, 1.0)
        bpy.ops.object.transform_apply(location=False, rotation=False, scale=True)

    cam_obj.data.clip_end = max(cam_obj.data.clip_end, world_radius * 2.5)

    scene.frame_set(1)
    bpy.context.view_layer.update()


def create_grass_material(grass_dir):
    mat = bpy.data.materials.new(name="Grass_PBR_Material")
    mat.use_nodes = True
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    out_node = nodes.new("ShaderNodeOutputMaterial")
    bsdf = nodes.new("ShaderNodeBsdfPrincipled")
    links.new(bsdf.outputs["BSDF"], out_node.inputs["Surface"])

    col_files = glob.glob(os.path.join(grass_dir, "*Color*.jpg")) + glob.glob(os.path.join(grass_dir, "*JPG*.jpg"))
    if col_files:
        tex_col = nodes.new("ShaderNodeTexImage")
        tex_col.image = bpy.data.images.load(col_files[0])
        links.new(tex_col.outputs["Color"], bsdf.inputs["Base Color"])

    rough_files = glob.glob(os.path.join(grass_dir, "*Roughness*.jpg"))
    if rough_files:
        tex_rough = nodes.new("ShaderNodeTexImage")
        tex_rough.image = bpy.data.images.load(rough_files[0])
        tex_rough.image.colorspace_settings.name = 'Non-Color'
        links.new(tex_rough.outputs["Color"], bsdf.inputs["Roughness"])

    norm_files = glob.glob(os.path.join(grass_dir, "*Normal*.jpg"))
    if norm_files:
        tex_norm = nodes.new("ShaderNodeTexImage")
        tex_norm.image = bpy.data.images.load(norm_files[0])
        tex_norm.image.colorspace_settings.name = 'Non-Color'

        normal_map_node = nodes.new("ShaderNodeNormalMap")
        links.new(tex_norm.outputs["Color"], normal_map_node.inputs["Color"])
        links.new(normal_map_node.outputs["Normal"], bsdf.inputs["Normal"])

    return mat


def create_sky_material(sky_hdri_path):
    mat = bpy.data.materials.new(name="Sky_Dome_Material")
    mat.use_nodes = True
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    out_node = nodes.new("ShaderNodeOutputMaterial")
    emission = nodes.new("ShaderNodeEmission")
    emission.inputs["Strength"].default_value = 1.0

    links.new(emission.outputs["Emission"], out_node.inputs["Surface"])

    if os.path.exists(sky_hdri_path):
        tex_env = nodes.new("ShaderNodeTexEnvironment")
        tex_env.image = bpy.data.images.load(sky_hdri_path)
        links.new(tex_env.outputs["Color"], emission.inputs["Color"])

    return mat


def setup_grass_ground_geometry_nodes(grass_dir, sky_hdri_path, ground_z_offset=-0.2):
    grass_coll = bpy.data.collections.get("Grass_Artifacts_Collection")
    if not grass_coll:
        grass_coll = bpy.data.collections.new("Grass_Artifacts_Collection")
        bpy.context.scene.collection.children.link(grass_coll)

    grass_mat = create_grass_material(grass_dir)
    sky_mat = create_sky_material(sky_hdri_path)

    mesh_files = glob.glob(os.path.join(grass_dir, "*.obj")) + glob.glob(os.path.join(grass_dir, "*.fbx"))

    if mesh_files:
        for fpath in mesh_files:
            if fpath.endswith(".obj"):
                if hasattr(bpy.ops.wm, "obj_import"):
                    bpy.ops.wm.obj_import(filepath=fpath)
                else:
                    bpy.ops.import_scene.obj(filepath=fpath)
            elif fpath.endswith(".fbx"):
                bpy.ops.import_scene.fbx(filepath=fpath)

            for o in bpy.context.selected_objects:
                grass_coll.objects.link(o)
                if o.name in bpy.context.scene.collection.objects:
                    bpy.context.scene.collection.objects.unlink(o)

        for obj in list(grass_coll.objects):
            if obj.dimensions.z > 0.5:
                grass_coll.objects.unlink(obj)
                bpy.data.objects.remove(obj, do_unlink=True)
    else:
        for i in range(3):
            m_data = bpy.data.meshes.new(f"Grass_Blade_Mesh_{i}")
            g_obj = bpy.data.objects.new(f"Grass_Blade_Obj_{i}", m_data)
            grass_coll.objects.link(g_obj)

            verts = [(0, 0, 0), (0.05, 0, 0), (0.04, 0, 0.15), (0.01, 0, 0.25), (0, 0, 0.3)]
            faces = [(0, 1, 2), (1, 2, 3), (2, 3, 4)]
            m_data.from_pydata(verts, [], faces)
            m_data.update()

            if g_obj.data.materials:
                g_obj.data.materials[0] = grass_mat
            else:
                g_obj.data.materials.append(grass_mat)

            g_obj.location = (100, 100, 0)

    bpy.ops.mesh.primitive_circle_add(radius=12.5, vertices=64, fill_type='NGON', location=(0, 0, ground_z_offset))
    ground_obj = bpy.context.active_object
    ground_obj.name = "Ground_Mesh_Object"

    vg = ground_obj.vertex_groups.new(name="Grass_Weight_Group")
    all_vert_indices = [v.index for v in ground_obj.data.vertices]
    vg.add(all_vert_indices, 1.0, 'REPLACE')
    ground_obj.data.materials.append(grass_mat)
    ground_obj.data.materials.append(sky_mat)

    gn_group = bpy.data.node_groups.new("Grass_Ground_GN", type="GeometryNodeTree")
    gn_group.interface.new_socket(name="Geometry", in_out="INPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Geometry", in_out="OUTPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Density Mask", in_out="INPUT", socket_type="NodeSocketFloat")

    nodes = gn_group.nodes
    links = gn_group.links
    out_node = nodes.new("NodeGroupOutput")
    in_node = nodes.new("NodeGroupInput")

    distribute_node = nodes.new("GeometryNodeDistributePointsOnFaces")
    distribute_node.distribute_method = 'POISSON'
    distribute_node.inputs["Distance Min"].default_value = 0.02
    distribute_node.inputs["Density Max"].default_value = 300.0
    links.new(in_node.outputs["Geometry"], distribute_node.inputs["Mesh"])

    noise_node = nodes.new("ShaderNodeTexNoise")
    noise_node.inputs["Scale"].default_value = 2.5
    noise_node.inputs["Detail"].default_value = 2.0

    color_ramp = nodes.new("ShaderNodeValToRGB")
    color_ramp.color_ramp.elements[0].position = 0.20
    color_ramp.color_ramp.elements[1].position = 0.80
    links.new(noise_node.outputs["Fac"], color_ramp.inputs["Fac"])

    math_weight = nodes.new("ShaderNodeMath")
    math_weight.operation = "MULTIPLY"
    links.new(color_ramp.outputs["Color"], math_weight.inputs[0])
    links.new(in_node.outputs["Density Mask"], math_weight.inputs[1])

    links.new(math_weight.outputs["Value"], distribute_node.inputs["Density Factor"])

    instance_node = nodes.new("GeometryNodeInstanceOnPoints")
    instance_node.inputs["Pick Instance"].default_value = True
    links.new(distribute_node.outputs["Points"], instance_node.inputs["Points"])

    coll_info = nodes.new("GeometryNodeCollectionInfo")
    coll_info.inputs["Collection"].default_value = grass_coll
    coll_info.inputs["Separate Children"].default_value = True
    coll_info.inputs["Reset Children"].default_value = True
    links.new(coll_info.outputs["Instances"], instance_node.inputs["Instance"])

    rand_scale = nodes.new("FunctionNodeRandomValue")
    rand_scale.data_type = 'FLOAT'
    rand_scale.inputs["Min"].default_value = 0.25
    rand_scale.inputs["Max"].default_value = 0.50

    combine_scale = nodes.new("ShaderNodeCombineXYZ")
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["X"])
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["Y"])
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["Z"])
    links.new(combine_scale.outputs["Vector"], instance_node.inputs["Scale"])

    rand_rot_z = nodes.new("FunctionNodeRandomValue")
    rand_rot_z.data_type = 'FLOAT'
    rand_rot_z.inputs["Min"].default_value = 0.0
    rand_rot_z.inputs["Max"].default_value = math.radians(360.0)

    combine_rot = nodes.new("ShaderNodeCombineXYZ")
    links.new(rand_rot_z.outputs["Value"], combine_rot.inputs["Z"])
    links.new(combine_rot.outputs["Vector"], instance_node.inputs["Rotation"])

    sky_sphere = nodes.new("GeometryNodeMeshIcoSphere")
    sky_sphere.inputs["Radius"].default_value = 12.5
    sky_sphere.inputs["Subdivisions"].default_value = 4

    flip_faces = nodes.new("GeometryNodeFlipFaces")
    links.new(sky_sphere.outputs["Mesh"], flip_faces.inputs["Mesh"])

    transform_sky = nodes.new("GeometryNodeTransform")
    transform_sky.inputs["Translation"].default_value = (0.0, 0.0, ground_z_offset)
    links.new(flip_faces.outputs["Mesh"], transform_sky.inputs["Geometry"])

    set_sky_mat = nodes.new("GeometryNodeSetMaterial")
    set_sky_mat.inputs["Material"].default_value = sky_mat
    links.new(transform_sky.outputs["Geometry"], set_sky_mat.inputs["Geometry"])

    join_geo = nodes.new("GeometryNodeJoinGeometry")
    links.new(in_node.outputs["Geometry"], join_geo.inputs["Geometry"])
    links.new(instance_node.outputs["Instances"], join_geo.inputs["Geometry"])
    links.new(set_sky_mat.outputs["Geometry"], join_geo.inputs["Geometry"])

    links.new(join_geo.outputs["Geometry"], out_node.inputs["Geometry"])

    mod = ground_obj.modifiers.new(name="Grass_Ground_GN", type='NODES')
    mod.node_group = gn_group

    for item in gn_group.interface.items_tree:
        if getattr(item, "in_out", None) == "INPUT" and item.name == "Density Mask":
            mod[item.identifier + "_attribute_name"] = "Grass_Weight_Group"
            mod[item.identifier + "_use_attribute"] = True

    return ground_obj


# ==============================================================================
# SCROLL MATERIAL & GEOMETRY NODES SETUP
# ==============================================================================

def create_image_scroll_material(image_path):
    mat_name = "Scroll_Texture_Mat"
    mat = bpy.data.materials.get(mat_name)
    if mat:
        bpy.data.materials.remove(mat)

    mat = bpy.data.materials.new(name=mat_name)
    mat.use_nodes = True
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    output_node = nodes.new("ShaderNodeOutputMaterial")
    bsdf = nodes.new("ShaderNodeBsdfPrincipled")
    bsdf.inputs["Roughness"].default_value = 0.5

    tex_img = nodes.new("ShaderNodeTexImage")
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        tex_img.image = img
        tex_img.extension = 'REPEAT'

    uv_map_node = nodes.new("ShaderNodeUVMap")
    uv_map_node.uv_map = "UVMap"

    links.new(uv_map_node.outputs["UV"], tex_img.inputs["Vector"])
    links.new(tex_img.outputs["Color"], bsdf.inputs["Base Color"])
    links.new(bsdf.outputs["BSDF"], output_node.inputs["Surface"])

    return mat, tex_img.image if os.path.exists(image_path) else None


def create_pure_geometry_node_scroll(image_path=""):
    gn_group = bpy.data.node_groups.new("Scroll_Pure_GN", type="GeometryNodeTree")

    gn_group.interface.new_socket(name="Geometry", in_out="OUTPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Character Mesh", in_out="INPUT", socket_type="NodeSocketObject")

    sock_r_vg = gn_group.interface.new_socket(name="Right Hand Group", in_out="INPUT", socket_type="NodeSocketString")
    sock_r_vg.default_value = "mixamorig:RightHand"

    sock_l_vg = gn_group.interface.new_socket(name="Left Hand Group", in_out="INPUT", socket_type="NodeSocketString")
    sock_l_vg.default_value = "mixamorig:LeftHand"

    sock_offset = gn_group.interface.new_socket(name="Forward Offset", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_offset.default_value = 0.60

    sock_unfurl = gn_group.interface.new_socket(name="Unfurl", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_unfurl.default_value = 0.0

    sock_width = gn_group.interface.new_socket(name="Scroll Width", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_width.default_value = 0.35

    nodes = gn_group.nodes
    links = gn_group.links

    in_node = nodes.new("NodeGroupInput")
    out_node = nodes.new("NodeGroupOutput")

    char_info = nodes.new("GeometryNodeObjectInfo")
    char_info.transform_space = "RELATIVE"
    links.new(in_node.outputs["Character Mesh"], char_info.inputs["Object"])

    pos_node = nodes.new("GeometryNodeInputPosition")

    named_r = nodes.new("GeometryNodeInputNamedAttribute")
    named_r.data_type = "FLOAT"
    links.new(in_node.outputs["Right Hand Group"], named_r.inputs["Name"])

    cmp_r = nodes.new("FunctionNodeCompare")
    cmp_r.data_type = "FLOAT"
    cmp_r.operation = "GREATER_THAN"
    cmp_r.inputs[1].default_value = 0.001
    links.new(named_r.outputs["Attribute"], cmp_r.inputs[0])

    stat_r = nodes.new("GeometryNodeAttributeStatistic")
    stat_r.data_type = "FLOAT_VECTOR"
    links.new(char_info.outputs["Geometry"], stat_r.inputs["Geometry"])
    links.new(cmp_r.outputs["Result"], stat_r.inputs["Selection"])
    links.new(pos_node.outputs["Position"], stat_r.inputs["Attribute"])

    named_l = nodes.new("GeometryNodeInputNamedAttribute")
    named_l.data_type = "FLOAT"
    links.new(in_node.outputs["Left Hand Group"], named_l.inputs["Name"])

    cmp_l = nodes.new("FunctionNodeCompare")
    cmp_l.data_type = "FLOAT"
    cmp_l.operation = "GREATER_THAN"
    cmp_l.inputs[1].default_value = 0.001
    links.new(named_l.outputs["Attribute"], cmp_l.inputs[0])

    stat_l = nodes.new("GeometryNodeAttributeStatistic")
    stat_l.data_type = "FLOAT_VECTOR"
    links.new(char_info.outputs["Geometry"], stat_l.inputs["Geometry"])
    links.new(cmp_l.outputs["Result"], stat_l.inputs["Selection"])
    links.new(pos_node.outputs["Position"], stat_l.inputs["Attribute"])

    neg_offset = nodes.new("ShaderNodeMath")
    neg_offset.operation = "MULTIPLY"
    neg_offset.inputs[1].default_value = -1.0
    links.new(in_node.outputs["Forward Offset"], neg_offset.inputs[0])

    forward_vec = nodes.new("ShaderNodeCombineXYZ")
    forward_vec.inputs["X"].default_value = 0.0
    links.new(neg_offset.outputs["Value"], forward_vec.inputs["Y"])
    forward_vec.inputs["Z"].default_value = 0.0

    start_pos = nodes.new("ShaderNodeVectorMath")
    start_pos.operation = "ADD"
    links.new(stat_r.outputs["Mean"], start_pos.inputs[0])
    links.new(forward_vec.outputs["Vector"], start_pos.inputs[1])

    target_pos = nodes.new("ShaderNodeVectorMath")
    target_pos.operation = "ADD"
    links.new(stat_l.outputs["Mean"], target_pos.inputs[0])
    links.new(forward_vec.outputs["Vector"], target_pos.inputs[1])

    delta_vec = nodes.new("ShaderNodeVectorMath")
    delta_vec.operation = "SUBTRACT"
    links.new(target_pos.outputs["Vector"], delta_vec.inputs[0])
    links.new(start_pos.outputs["Vector"], delta_vec.inputs[1])

    dist_node = nodes.new("ShaderNodeVectorMath")
    dist_node.operation = "DISTANCE"
    links.new(start_pos.outputs["Vector"], dist_node.inputs[0])

    scale_vec = nodes.new("ShaderNodeVectorMath")
    scale_vec.operation = "SCALE"
    links.new(delta_vec.outputs["Vector"], scale_vec.inputs["Vector"])
    links.new(in_node.outputs["Unfurl"], scale_vec.inputs["Scale"])

    end_pos = nodes.new("ShaderNodeVectorMath")
    end_pos.operation = "ADD"
    links.new(start_pos.outputs["Vector"], end_pos.inputs[0])
    links.new(scale_vec.outputs["Vector"], end_pos.inputs[1])

    links.new(end_pos.outputs["Vector"], dist_node.inputs[1])

    curve_line = nodes.new("GeometryNodeCurvePrimitiveLine")
    links.new(start_pos.outputs["Vector"], curve_line.inputs["Start"])
    links.new(end_pos.outputs["Vector"], curve_line.inputs["End"])

    set_tilt = nodes.new("GeometryNodeSetCurveTilt")
    set_tilt.inputs["Tilt"].default_value = 1.5708
    links.new(curve_line.outputs["Curve"], set_tilt.inputs["Curve"])

    spline_param_main = nodes.new("GeometryNodeSplineParameter")
    store_u = nodes.new("GeometryNodeStoreNamedAttribute")
    store_u.inputs["Name"].default_value = "U_Factor"
    store_u.data_type = 'FLOAT'
    store_u.domain = 'POINT'
    links.new(set_tilt.outputs["Curve"], store_u.inputs["Geometry"])
    links.new(spline_param_main.outputs["Factor"], store_u.inputs["Value"])

    half_width_neg = nodes.new("ShaderNodeMath")
    half_width_neg.operation = "MULTIPLY"
    half_width_neg.inputs[1].default_value = -0.5
    links.new(in_node.outputs["Scroll Width"], half_width_neg.inputs[0])

    half_width_pos = nodes.new("ShaderNodeMath")
    half_width_pos.operation = "MULTIPLY"
    half_width_pos.inputs[1].default_value = 0.5
    links.new(in_node.outputs["Scroll Width"], half_width_pos.inputs[0])

    combine_start = nodes.new("ShaderNodeCombineXYZ")
    links.new(half_width_neg.outputs["Value"], combine_start.inputs["X"])

    combine_end = nodes.new("ShaderNodeCombineXYZ")
    links.new(half_width_pos.outputs["Value"], combine_end.inputs["X"])

    profile_line = nodes.new("GeometryNodeCurvePrimitiveLine")
    links.new(combine_start.outputs["Vector"], profile_line.inputs["Start"])
    links.new(combine_end.outputs["Vector"], profile_line.inputs["End"])

    spline_param_prof = nodes.new("GeometryNodeSplineParameter")
    store_v = nodes.new("GeometryNodeStoreNamedAttribute")
    store_v.inputs["Name"].default_value = "V_Factor"
    store_v.data_type = 'FLOAT'
    store_v.domain = 'POINT'
    links.new(profile_line.outputs["Curve"], store_v.inputs["Geometry"])
    links.new(spline_param_prof.outputs["Factor"], store_v.inputs["Value"])

    curve_to_mesh = nodes.new("GeometryNodeCurveToMesh")
    links.new(store_u.outputs["Geometry"], curve_to_mesh.inputs["Curve"])
    links.new(store_v.outputs["Geometry"], curve_to_mesh.inputs["Profile Curve"])

    img_aspect = 1.0
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        if img.size[1] > 0:
            img_aspect = img.size[0] / img.size[1]

    h_times_aspect = nodes.new("ShaderNodeMath")
    h_times_aspect.operation = "MULTIPLY"
    h_times_aspect.inputs[1].default_value = img_aspect
    links.new(in_node.outputs["Scroll Width"], h_times_aspect.inputs[0])

    tile_count = nodes.new("ShaderNodeMath")
    tile_count.operation = "DIVIDE"
    links.new(dist_node.outputs["Value"], tile_count.inputs[0])
    links.new(h_times_aspect.outputs["Value"], tile_count.inputs[1])

    read_u = nodes.new("GeometryNodeInputNamedAttribute")
    read_u.inputs["Name"].default_value = "U_Factor"
    read_u.data_type = 'FLOAT'

    scaled_u = nodes.new("ShaderNodeMath")
    scaled_u.operation = "MULTIPLY"
    links.new(read_u.outputs["Attribute"], scaled_u.inputs[0])
    links.new(tile_count.outputs["Value"], scaled_u.inputs[1])

    read_v = nodes.new("GeometryNodeInputNamedAttribute")
    read_v.inputs["Name"].default_value = "V_Factor"
    read_v.data_type = 'FLOAT'

    flip_v = nodes.new("ShaderNodeMath")
    flip_v.operation = "SUBTRACT"
    flip_v.inputs[0].default_value = 1.0
    links.new(read_v.outputs["Attribute"], flip_v.inputs[1])

    combine_uv = nodes.new("ShaderNodeCombineXYZ")
    links.new(scaled_u.outputs["Value"], combine_uv.inputs["X"])
    links.new(flip_v.outputs["Value"], combine_uv.inputs["Y"])

    store_uv = nodes.new("GeometryNodeStoreNamedAttribute")
    store_uv.inputs["Name"].default_value = "UVMap"
    store_uv.data_type = 'FLOAT2'
    store_uv.domain = 'CORNER'
    links.new(curve_to_mesh.outputs["Mesh"], store_uv.inputs["Geometry"])
    links.new(combine_uv.outputs["Vector"], store_uv.inputs["Value"])

    set_material = nodes.new("GeometryNodeSetMaterial")
    scroll_mat, _ = create_image_scroll_material(image_path)
    set_material.inputs["Material"].default_value = scroll_mat
    links.new(store_uv.outputs["Geometry"], set_material.inputs["Geometry"])

    links.new(set_material.outputs["Geometry"], out_node.inputs["Geometry"])

    return gn_group, scroll_mat


def setup_geometry_node_scroll(target_mesh, unfurl_start, unfurl_end, image_path=""):
    mesh_data = bpy.data.meshes.new("Scroll_Procedural_Mesh")
    scroll_obj = bpy.data.objects.new("Scroll_GN_Object", mesh_data)
    bpy.context.scene.collection.objects.link(scroll_obj)

    mod = scroll_obj.modifiers.new("Scroll_GN", type="NODES")
    gn_tree, scroll_mat = create_pure_geometry_node_scroll(image_path=image_path)
    mod.node_group = gn_tree

    scroll_obj.data.materials.append(scroll_mat)

    r_hand_name, l_hand_name = find_hand_vertex_groups(target_mesh)

    unfurl_id = None
    for item in gn_tree.interface.items_tree:
        if getattr(item, "in_out", None) == "INPUT":
            if item.name == "Character Mesh":
                mod[item.identifier] = target_mesh
            elif item.name == "Right Hand Group":
                mod[item.identifier] = r_hand_name
            elif item.name == "Left Hand Group":
                mod[item.identifier] = l_hand_name
            elif item.name == "Forward Offset":
                mod[item.identifier] = 0.60
            elif item.name == "Scroll Width":
                mod[item.identifier] = 0.35
            elif item.name == "Unfurl":
                unfurl_id = item.identifier

    for m in target_mesh.modifiers:
        if m.type == 'ARMATURE':
            m.show_viewport = True
            m.show_render = True
            m.show_in_editmode = True

    scroll_obj.hide_viewport = True
    scroll_obj.hide_render = True
    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=1)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=1)

    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=unfurl_start - 1)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=unfurl_start - 1)

    scroll_obj.hide_viewport = False
    scroll_obj.hide_render = False
    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=unfurl_start)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=unfurl_start)

    for fc in get_object_fcurves(scroll_obj):
        if fc.data_path in ["hide_viewport", "hide_render"]:
            for kp in fc.keyframe_points:
                kp.interpolation = 'CONSTANT'

    if unfurl_id:
        mod[unfurl_id] = 0.001
        mod.keyframe_insert(data_path=f'["{unfurl_id}"]', frame=unfurl_start)

        mod[unfurl_id] = 1.0
        mod.keyframe_insert(data_path=f'["{unfurl_id}"]', frame=unfurl_end)

    return scroll_obj


# ==============================================================================
# MAIN CONTROLLER
# ==============================================================================

def import_and_extract_action(fbx_path, action_name):
    existing_objects = set(bpy.data.objects)
    bpy.ops.import_scene.fbx(filepath=fbx_path)
    bpy.context.view_layer.update()

    imported_objects = set(bpy.data.objects) - existing_objects
    extracted_action = None
    for obj in imported_objects:
        if obj.animation_data and obj.animation_data.action:
            extracted_action = obj.animation_data.action
            extracted_action.name = action_name
            extracted_action.use_fake_user = True
            break
    return extracted_action, imported_objects


def purge_imported_objects(objects_to_remove):
    for obj in objects_to_remove:
        if obj.name in bpy.data.objects:
            if obj.animation_data:
                obj.animation_data_clear()
            for coll in list(obj.users_collection):
                coll.objects.unlink(obj)
    bpy.context.view_layer.update()


def main():
    cleanup_previous_run()

    walk_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Mutant Walking.fbx"
    turn_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Right Turn 90.fbx"
    unfurl_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Mutant Idle.fbx"

    sky_hdri_path = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/horn-koppe_spring_4k.exr"
    output_blend = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/output_scene.blend"

    image_path = "/Users/dobbinp1/Downloads/IMGP6691_psedit.jpg"
    grass_dir = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Grass004_4K-JPG"

    ground_obj = setup_grass_ground_geometry_nodes(grass_dir, sky_hdri_path, ground_z_offset=-0.2)

    CYCLES = 4
    CROSSFADE = 15
    TARGET_DISTANCE = 4.0



    walk_action, walk_objs = import_and_extract_action(walk_fbx, "walk")
    target_rig = next((obj for obj in walk_objs if obj.type == "ARMATURE"), None)
    target_mesh = next((obj for obj in walk_objs if obj.type == "MESH"), None)




    if not target_rig or not target_mesh:
        raise RuntimeError("Armature or Character Mesh not found in walk FBX.")

    travel_axis_idx, motion_sign = detect_natural_forward_axis(walk_action, target_rig)
    initial_heading = math.radians(90.0) if travel_axis_idx == 0 else 0.0
    target_rig.rotation_euler[2] = initial_heading

    action_len = int(walk_action.frame_range[1] - walk_action.frame_range[0])
    if action_len <= 0:
        action_len = 30

    anim_data = target_rig.animation_data_create()

    walk_track = anim_data.nla_tracks.new()
    walk_track.name = "Walk_Track"

    current_frame = 1
    for i in range(CYCLES):
        strip = walk_track.strips.new(name=f"Walk_{i+1}", start=int(current_frame), action=walk_action)
        strip.blend_type = "REPLACE"
        current_frame += action_len

    walk_end_frame = int(current_frame)

    turn_action, turn_objs = import_and_extract_action(turn_fbx, "rightturn")
    purge_imported_objects(turn_objs)

    turn_track = anim_data.nla_tracks.new()
    turn_track.name = "RightTurn_Track"

    turn_start = int(walk_end_frame - CROSSFADE)
    turn_strip = turn_track.strips.new(name="RightTurn", start=turn_start, action=turn_action)
    turn_strip.blend_type = "REPLACE"
    turn_strip.blend_in = CROSSFADE
    turn_strip.use_auto_blend = True
    turn_strip.extrapolation = "HOLD"
    turn_end_frame = int(turn_strip.frame_end)

    # Pass turn_start frame to scene data so handler stops sound when character stops walking
    bpy.context.scene["turn_start_frame"] = turn_start

    unfurl_action, unfurl_objs = import_and_extract_action(unfurl_fbx, "unfurl")
    purge_imported_objects(unfurl_objs)

    unfurl_track = anim_data.nla_tracks.new()
    unfurl_track.name = "Unfurl_Track"

    unfurl_start = int(turn_end_frame - CROSSFADE)
    unfurl_strip = unfurl_track.strips.new(name="Unfurl", start=unfurl_start, action=unfurl_action)
    unfurl_strip.blend_type = "REPLACE"
    unfurl_strip.blend_in = CROSSFADE
    unfurl_strip.use_auto_blend = True
    unfurl_strip.extrapolation = "HOLD"
    unfurl_end_frame = int(unfurl_strip.frame_end)

    root_action = bpy.data.actions.new(name="World_Root_Motion")





        # Ensure the animation data container exists on the target rig
    if not target_rig.animation_data:
        target_rig.animation_data_create()

    # Assign the action to the rig before accessing slot-based F-Curves
    target_rig.animation_data.action = root_action

    # Now create or retrieve the F-Curves
    if hasattr(root_action, "fcurves") and hasattr(root_action.fcurves, "new"):
        loc_fcurve = root_action.fcurves.new(data_path="location", index=travel_axis_idx)
        rot_fcurve = root_action.fcurves.new(data_path="rotation_euler", index=2)
    else:
        loc_fcurve = root_action.fcurve_ensure_for_datablock(
            datablock=target_rig, data_path="location", index=travel_axis_idx
        )
        rot_fcurve = root_action.fcurve_ensure_for_datablock(
            datablock=target_rig, data_path="rotation_euler", index=2
        )












    # Modernized F-Curve creation logic for single multi-channel action compatibility
    if hasattr(root_action, "fcurves") and hasattr(root_action.fcurves, "new"):
        loc_fcurve = root_action.fcurves.new(data_path="location", index=travel_axis_idx)
        rot_fcurve = root_action.fcurves.new(data_path="rotation_euler", index=2)
    else:
        loc_fcurve = root_action.fcurve_ensure_for_datablock(datablock=target_rig, data_path="location", index=travel_axis_idx)
        rot_fcurve = root_action.fcurve_ensure_for_datablock(datablock=target_rig, data_path="rotation_euler", index=2)

    final_target_value = TARGET_DISTANCE * motion_sign

    k_loc1 = loc_fcurve.keyframe_points.insert(frame=1.0, value=0.0)
    k_loc2 = loc_fcurve.keyframe_points.insert(frame=float(turn_start), value=final_target_value)
    k_loc3 = loc_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=final_target_value)

    k_loc1.interpolation = "LINEAR"
    k_loc2.interpolation = "CONSTANT"
    k_loc3.interpolation = "CONSTANT"

    target_heading = initial_heading - math.radians(90.0)

    k_rot1 = rot_fcurve.keyframe_points.insert(frame=1.0, value=initial_heading)
    k_rot2 = rot_fcurve.keyframe_points.insert(frame=float(turn_start), value=initial_heading)
    k_rot3 = rot_fcurve.keyframe_points.insert(frame=float(turn_end_frame), value=target_heading)
    k_rot4 = rot_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=target_heading)

    k_rot1.interpolation = "CONSTANT"
    k_rot2.interpolation = "BEZIER"
    k_rot3.interpolation = "CONSTANT"
    k_rot4.interpolation = "CONSTANT"

    root_track = anim_data.nla_tracks.new()
    root_track.name = "World_Root_Track"
    root_strip = root_track.strips.new(name="Root_Motion_Strip", start=1, action=root_action)
    root_strip.blend_type = "REPLACE"
    root_strip.extrapolation = "HOLD"

    anim_data.action = None

    setup_geometry_node_scroll(
        target_mesh=target_mesh,
        unfurl_start=unfurl_start,
        unfurl_end=unfurl_end_frame,
        image_path=image_path,
    )

    cam_obj, focus_target = setup_modern_reel_camera(
        target_rig=target_rig,
        target_mesh=target_mesh,
        turn_start=turn_start,
        turn_end_frame=turn_end_frame,
        unfurl_end_frame=unfurl_end_frame
    )

    bpy.context.scene.frame_start = 1
    bpy.context.scene.frame_end = unfurl_end_frame

    # Animate camera zoom to frame the scroll tile perfectly
    final_frame = frame_scroll_image(
        cam_obj=cam_obj,
        focus_target=focus_target,
        target_mesh=target_mesh,
        image_path=image_path,
        scroll_width=0.35,
        zoom_duration=45
    )

    ensure_world_sphere_size(cam_obj, ground_obj, safety_multiplier=2.5)

    # Register real-time footstep audio handler
    register_audio_handler()

    bpy.ops.wm.save_as_mainfile(filepath=output_blend)
    print(f"\n[SUCCESS] Integrated parented target tracking & saved scene to: {output_blend}")


if __name__ == "__main__":
    main()
```

```
import math
import mathutils
import os
import sys
import glob
import bpy
import aud
from bpy_extras import anim_utils
from bpy_extras.object_utils import world_to_camera_view
from bpy.app.handlers import persistent
from mathutils import Vector

# ==============================================================================
# CONFIGURATION
# ==============================================================================
FOOTSTEP_SOUND_PATH = "/Users/dobbinp1/Downloads/stomp.wav"
GROUND_THRESHOLD = 0.24  # Raised slightly for testing

# Setup Audio Device
if "_audio_store" not in sys.modules:
    import types
    sys.modules["_audio_store"] = types.ModuleType("_audio_store")

audio_store = sys.modules["_audio_store"]
if not hasattr(audio_store, "device"):
    audio_store.device = aud.Device()

AUDIO_DEVICE = audio_store.device
SOUND_FACTORY = None
AUDIO_HANDLES = []

if os.path.exists(FOOTSTEP_SOUND_PATH):
    SOUND_FACTORY = aud.Sound(FOOTSTEP_SOUND_PATH).cache()
    print(f"[AUDIO] Loaded sound file successfully.")
else:
    print(f"[AUDIO ERROR] File not found at: {FOOTSTEP_SOUND_PATH}")

_left_foot_was_down = False
_right_foot_was_down = False
_last_frame = -1


def play_footstep():
    global AUDIO_HANDLES
    if SOUND_FACTORY:
        print("  >>> PLAYING FOOTSTEP SOUND <<<")
        handle = AUDIO_DEVICE.play(SOUND_FACTORY)
        AUDIO_HANDLES.append(handle)
        if len(AUDIO_HANDLES) > 10:
            AUDIO_HANDLES.pop(0)


def find_foot_bone(pose_bones, side="Left"):
    target_keywords = [
        f"mixamorig:{side}foot",
        f"{side.lower()}foot",
        f"foot.{side[0].lower()}",
        f"foot_{side[0].lower()}"
    ]
    for bone in pose_bones:
        b_name = bone.name.lower()
        if any(kw in b_name for kw in target_keywords):
            return bone
    return None


@persistent
def footstep_frame_handler(scene):
    global _left_foot_was_down, _right_foot_was_down, _last_frame

    try:
        current_frame = scene.frame_current
        turn_start = scene.get("turn_start_frame", None)

        # Stop footsteps once turning/idling starts
        if turn_start is not None and current_frame >= turn_start:
            _left_foot_was_down = False
            _right_foot_was_down = False
            return

        if current_frame <= _last_frame or abs(current_frame - _last_frame) > 10:
            _left_foot_was_down = False
            _right_foot_was_down = False
        _last_frame = current_frame

        rig = next((o for o in scene.objects if o.type == 'ARMATURE'), None)
        if not rig:
            return

        depsgraph = bpy.context.evaluated_depsgraph_get()
        eval_rig = rig.evaluated_get(depsgraph)

        left_bone = find_foot_bone(eval_rig.pose.bones, "Left")
        right_bone = find_foot_bone(eval_rig.pose.bones, "Right")

        if not left_bone or not right_bone:
            return

        l_z = (eval_rig.matrix_world @ left_bone.matrix).translation.z
        r_z = (eval_rig.matrix_world @ right_bone.matrix).translation.z

        l_is_down = l_z <= GROUND_THRESHOLD
        r_is_down = r_z <= GROUND_THRESHOLD

        if l_is_down and not _left_foot_was_down:
            play_footstep()

        if r_is_down and not _right_foot_was_down:
            play_footstep()

        _left_foot_was_down = l_is_down
        _right_foot_was_down = r_is_down

    except Exception:
        import traceback
        traceback.print_exc()


def register_audio_handler():
    # Clear existing handlers
    for h in list(bpy.app.handlers.frame_change_post):
        if getattr(h, "__name__", "") == "footstep_frame_handler":
            bpy.app.handlers.frame_change_post.remove(h)

    bpy.app.handlers.frame_change_post.append(footstep_frame_handler)
    print("[SUCCESS] Handler registered to frame_change_post.")


# Register audio handler on load
register_audio_handler()


# ==============================================================================
# CLEANUP & HELPERS
# ==============================================================================

def cleanup_previous_run():
    """Clears old scroll objects, ground, node groups, camera anchors, collections, and orphaned data."""
    target_names = [
        "Master_NLA_Control",
        "Scroll_GN_Object",
        "Scroll_Start_Pin",
        "Scroll_End_Pin",
        "Scroll_Master_Container",
        "HDRI_Ground_Plane",
        "Ground_Mesh_Object",
        "Camera_Focus_Target",
        "Vertical_Reel_Camera"
    ]
    for obj in list(bpy.data.objects):
        if obj.name in target_names or (obj.name.startswith("mixamorig") and obj.type != "ARMATURE" and obj.type != "MESH"):
            if obj.animation_data:
                obj.animation_data_clear()
            bpy.data.objects.remove(obj, do_unlink=True)

    for action in list(bpy.data.actions):
        if action.users == 0 or action.use_fake_user:
            action.use_fake_user = False
            bpy.data.actions.remove(action)

    for ng in list(bpy.data.node_groups):
        if ng.name in ["Scroll_Unfurl_GN", "Scroll_Pure_GN", "HDRI_Ground_GN", "Grass_Ground_GN"]:
            bpy.data.node_groups.remove(ng)

    for coll_name in ["Scroll_Master_Container", "Grass_Artifacts_Collection"]:
        if coll_name in bpy.data.collections:
            coll = bpy.data.collections[coll_name]
            for obj in list(coll.objects):
                bpy.data.objects.remove(obj, do_unlink=True)
            bpy.data.collections.remove(coll)

    bpy.context.view_layer.update()


def get_slot_fcurves(action, anim_target_object):
    fcurves = []
    if not action:
        return fcurves

    anim_data = getattr(anim_target_object, "animation_data", None)
    if anim_data and getattr(anim_data, "action_slot", None):
        slot = anim_data.action_slot
        channelbag = anim_utils.action_get_channelbag_for_slot(action, slot)
        if channelbag and hasattr(channelbag, "fcurves"):
            return list(channelbag.fcurves)

    if hasattr(action, "slots"):
        for slot in action.slots:
            for cb in getattr(slot, "channelbags", []):
                fcurves.extend(getattr(cb, "fcurves", []))
        if fcurves:
            return fcurves

    if hasattr(action, "fcurves"):
        return list(action.fcurves)

    return fcurves


def get_object_fcurves(obj):
    if not obj or not obj.animation_data or not obj.animation_data.action:
        return []
    action = obj.animation_data.action
    return get_slot_fcurves(action, obj)


def detect_natural_forward_axis(action, rig_object):
    fcurves = get_slot_fcurves(action, rig_object)
    max_delta = 0.0
    best_axis = 1
    net_delta = 0.0

    start_frame = action.frame_range[0]
    end_frame = action.frame_range[1]

    for fc in fcurves:
        if fc.data_path.endswith("location") and fc.array_index in [0, 1, 2]:
            start_val = fc.evaluate(start_frame)
            end_val = fc.evaluate(end_frame)
            delta = end_val - start_val
            if abs(delta) > abs(max_delta):
                max_delta = delta
                best_axis = fc.array_index
                net_delta = delta

    direction_sign = 1.0 if net_delta >= 0 else -1.0
    return best_axis, direction_sign


def find_hand_vertex_groups(target_mesh):
    r_name = "mixamorig:RightHand"
    l_name = "mixamorig:LeftHand"

    if target_mesh and hasattr(target_mesh, "vertex_groups"):
        for vg in target_mesh.vertex_groups:
            vg_lower = vg.name.lower()
            if "righthand" in vg_lower or "hand_r" in vg_lower or "hand.r" in vg_lower:
                r_name = vg.name
            elif "lefthand" in vg_lower or "hand_l" in vg_lower or "hand.l" in vg_lower:
                l_name = vg.name

    return r_name, l_name


# ==============================================================================
# PARENTED VERTICAL CAMERA SETUP (OPTION A)
# ==============================================================================

def setup_modern_reel_camera(target_rig, target_mesh, turn_start, turn_end_frame, unfurl_end_frame):
    scene = bpy.context.scene
    scene.render.resolution_x = 1080
    scene.render.resolution_y = 1920
    scene.render.resolution_percentage = 100

    target_height = target_mesh.dimensions.z if target_mesh.dimensions.z > 0 else 1.8
    midpoint_z_offset = target_height * 0.55

    # 1. Setup parented camera focus target directly on target_mesh
    focus_target = bpy.data.objects.get("Camera_Focus_Target")
    if not focus_target:
        focus_target = bpy.data.objects.new("Camera_Focus_Target", None)
        focus_target.empty_display_type = 'PLAIN_AXES'
        focus_target.empty_display_size = 0.25
        scene.collection.objects.link(focus_target)

    focus_target.constraints.clear()
    focus_target.parent = target_mesh
    focus_target.matrix_parent_inverse = target_mesh.matrix_world.inverted()
    focus_target.location = (0.0, 0.0, midpoint_z_offset)

    # 2. Camera Setup
    cam_data = bpy.data.cameras.get("Vertical_Reel_Cam_Data")
    if not cam_data:
        cam_data = bpy.data.cameras.new("Vertical_Reel_Cam_Data")
    cam_data.lens = 45.0
    cam_data.sensor_fit = 'VERTICAL'
    cam_data.clip_end = 100.0

    cam_obj = bpy.data.objects.get("Vertical_Reel_Camera")
    if not cam_obj:
        cam_obj = bpy.data.objects.new("Vertical_Reel_Camera", cam_data)
        scene.collection.objects.link(cam_obj)

    scene.camera = cam_obj
    cam_obj.parent = None

    track = cam_obj.constraints.get("Track To")
    if not track:
        track = cam_obj.constraints.new(type='TRACK_TO')
    track.target = focus_target
    track.track_axis = 'TRACK_NEGATIVE_Z'
    track.up_axis = 'UP_Y'

    def check_mesh_in_frame(frame_num, cam_location, margin=0.08):
        cam_obj.location = cam_location
        scene.frame_set(int(frame_num))
        bpy.context.view_layer.update()

        depsgraph = bpy.context.evaluated_depsgraph_get()
        eval_mesh_obj = target_mesh.evaluated_get(depsgraph)
        mesh_data = eval_mesh_obj.data

        vert_step = max(1, len(mesh_data.vertices) // 300)
        world_matrix = target_mesh.matrix_world

        min_x, max_x = 1.0, 0.0
        min_y, max_y = 1.0, 0.0

        for i in range(0, len(mesh_data.vertices), vert_step):
            v_world = world_matrix @ mesh_data.vertices[i].co
            co_2d = world_to_camera_view(scene, cam_obj, v_world)

            if co_2d.x < min_x: min_x = co_2d.x
            if co_2d.x > max_x: max_x = co_2d.x
            if co_2d.y < min_y: min_y = co_2d.y
            if co_2d.y > max_y: max_y = co_2d.y

        in_bounds = (min_x >= margin and max_x <= (1.0 - margin) and
                     min_y >= margin and max_y <= (1.0 - margin))

        return in_bounds, min_x, max_x, min_y, max_y

    def find_fit_distance(frame_num, angle_rad, base_height_offset, initial_dist=5.0, margin=0.08):
        scene.frame_set(int(frame_num))
        bpy.context.view_layer.update()

        center_pos = focus_target.matrix_world.translation.copy()
        heading = target_rig.matrix_world.to_euler('XYZ').z
        view_angle = heading + angle_rad

        dist = initial_dist
        for _ in range(30):
            test_x = center_pos.x + dist * math.sin(view_angle)
            test_y = center_pos.y - dist * math.cos(view_angle)
            test_z = center_pos.z + base_height_offset

            fit, min_x, max_x, min_y, max_y = check_mesh_in_frame(frame_num, (test_x, test_y, test_z), margin=margin)
            if fit:
                break
            dist += 0.4

        return dist, center_pos, view_angle

    # --- Framing Calculation for Walking Phase ---
    max_walk_dist = 5.0
    walk_frames_to_sample = range(1, turn_start + 1, 4)

    for f in walk_frames_to_sample:
        req_d, _, _ = find_fit_distance(f, angle_rad=math.radians(-45.0), base_height_offset=0.0, initial_dist=max_walk_dist, margin=0.10)
        if req_d > max_walk_dist:
            max_walk_dist = req_d

    # Lock Frame 1 framing explicitly
    scene.frame_set(1)
    bpy.context.view_layer.update()
    c1 = focus_target.matrix_world.translation.copy()
    h1 = target_rig.matrix_world.to_euler('XYZ').z + math.radians(-45.0)
    w_x1 = c1.x + max_walk_dist * math.sin(h1)
    w_y1 = c1.y - max_walk_dist * math.cos(h1)
    w_z1 = c1.z

    # Turn start position framing
    scene.frame_set(turn_start)
    bpy.context.view_layer.update()
    cts = focus_target.matrix_world.translation.copy()
    hts = target_rig.matrix_world.to_euler('XYZ').z + math.radians(-45.0)
    w_xts = cts.x + max_walk_dist * math.sin(hts)
    w_yts = cts.y - max_walk_dist * math.cos(hts)
    w_zts = cts.z

    # --- Framing Calculation for Unfurl Phase ---
    unfurl_frames_to_sample = range(turn_end_frame, unfurl_end_frame + 1, 6)
    max_unfurl_dist = max_walk_dist

    for f in unfurl_frames_to_sample:
        req_d, _, _ = find_fit_distance(f, angle_rad=math.radians(0.0), base_height_offset=0.0, initial_dist=max_unfurl_dist, margin=0.10)
        if req_d > max_unfurl_dist:
            max_unfurl_dist = req_d

    scene.frame_set(turn_end_frame)
    bpy.context.view_layer.update()
    cte = focus_target.matrix_world.translation.copy()
    hte = target_rig.matrix_world.to_euler('XYZ').z + math.radians(0.0)
    f_x = cte.x + max_unfurl_dist * math.sin(hte)
    f_y = cte.y - max_unfurl_dist * math.cos(hte)
    f_z = cte.z

    if cam_obj.animation_data:
        cam_obj.animation_data_clear()

    # Keyframe 1: Monster visible and centered from frame 1
    cam_obj.location = (w_x1, w_y1, w_z1)
    cam_obj.keyframe_insert(data_path="location", frame=1)

    # Keyframe 2: Maintain tracking offset through turn start
    cam_obj.location = (w_xts, w_yts, w_zts)
    cam_obj.keyframe_insert(data_path="location", frame=turn_start)

    # Keyframe 3: Settle position post-turn
    cam_obj.location = (f_x, f_y, f_z)
    cam_obj.keyframe_insert(data_path="location", frame=turn_end_frame)

    # Keyframe 4: Hold framing through scroll unfurl
    cam_obj.location = (f_x, f_y, f_z)
    cam_obj.keyframe_insert(data_path="location", frame=unfurl_end_frame)

    for fc in get_object_fcurves(cam_obj):
        for kp in fc.keyframe_points:
            kp.interpolation = 'BEZIER'

    scene.frame_set(1)
    return cam_obj, focus_target


def frame_scroll_image(cam_obj, focus_target, target_mesh, image_path, scroll_width=0.35, zoom_duration=45):
    scene = bpy.context.scene
    unfurl_end_frame = scene.frame_end
    zoom_end_frame = unfurl_end_frame + zoom_duration

    scene.frame_set(unfurl_end_frame)
    bpy.context.view_layer.update()

    # 1. Aspect Ratio & Framing Distance Calculation
    img_aspect = 1.0
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        if img.size[1] > 0:
            img_aspect = img.size[0] / img.size[1]

    tile_height = scroll_width * img_aspect
    cam_data = cam_obj.data
    aspect_ratio = scene.render.resolution_x / scene.render.resolution_y
    sensor_width = cam_data.sensor_height * aspect_ratio

    fov_h_half = math.atan((sensor_width / 2.0) / cam_data.lens)
    target_framing_distance = (tile_height / 2.0) / math.tan(fov_h_half)

    # 2. Hand Position & Center Evaluation
    mesh_matrix = target_mesh.matrix_world
    depsgraph = bpy.context.evaluated_depsgraph_get()
    eval_mesh = target_mesh.evaluated_get(depsgraph)

    r_hand_name, l_hand_name = find_hand_vertex_groups(target_mesh)
    vg_r = eval_mesh.vertex_groups.get(r_hand_name)
    vg_l = eval_mesh.vertex_groups.get(l_hand_name)

    r_verts = [mesh_matrix @ v.co for v in eval_mesh.data.vertices if any(g.group == vg_r.index for g in v.groups)] if vg_r else []
    l_verts = [mesh_matrix @ v.co for v in eval_mesh.data.vertices if any(g.group == vg_l.index for g in v.groups)] if vg_l else []

    target_height = target_mesh.dimensions.z if target_mesh.dimensions.z > 0 else 1.8
    torso_fallback = mesh_matrix.translation + mathutils.Vector((0, 0, target_height * 0.55))

    r_hand_pos = sum(r_verts, mathutils.Vector((0, 0, 0))) / len(r_verts) if r_verts else torso_fallback
    l_hand_pos = sum(l_verts, mathutils.Vector((0, 0, 0))) / len(l_verts) if l_verts else torso_fallback

    scroll_midpoint = (r_hand_pos + l_hand_pos) * 0.5

    # 3. Guaranteed Outward Normal (Pointing TOWARD current camera)
    cam_pos = cam_obj.matrix_world.translation.copy()
    cam_dir = cam_pos - scroll_midpoint
    cam_dir.z = 0.0  # Horizontal plane projection

    if cam_dir.length > 0.001:
        scroll_normal = cam_dir.normalized()
    else:
        scroll_normal = mathutils.Vector((0, -1, 0))

    tile_center = scroll_midpoint + (scroll_normal * 0.10)
    end_pos = tile_center + (scroll_normal * target_framing_distance)

    # 4. OPTION A: Unparent Focus Target & Keyframe Zoom Smoothly
    current_world_focus = focus_target.matrix_world.translation.copy()

    focus_target.parent = None
    focus_target.matrix_world.translation = current_world_focus

    # Keyframe focus target sliding from current position to tile_center
    focus_target.location = current_world_focus
    focus_target.keyframe_insert(data_path="location", frame=unfurl_end_frame)
    focus_target.location = tile_center
    focus_target.keyframe_insert(data_path="location", frame=zoom_end_frame)

    # Keyframe camera moving from current position to end_pos
    cam_obj.location = cam_pos
    cam_obj.keyframe_insert(data_path="location", frame=unfurl_end_frame)
    cam_obj.location = end_pos
    cam_obj.keyframe_insert(data_path="location", frame=zoom_end_frame)

    # Clamp F-Curve handles to prevent curve overshoot
    for obj in [cam_obj, focus_target]:
        for fc in get_object_fcurves(obj):
            for kp in fc.keyframe_points:
                if kp.co.x >= unfurl_end_frame:
                    kp.interpolation = 'BEZIER'
                    kp.handle_left_type = 'AUTO_CLAMPED'
                    kp.handle_right_type = 'AUTO_CLAMPED'

    scene.frame_end = zoom_end_frame
    return zoom_end_frame


# ==============================================================================
# GEOMETRY NODES GROUND, GRASS & SKY DOME SETUP
# ==============================================================================

def ensure_world_sphere_size(cam_obj, ground_obj, safety_multiplier=2.5):
    if not cam_obj or not ground_obj:
        return

    scene = bpy.context.scene
    max_cam_dist = 0.0

    for frame in range(scene.frame_start, scene.frame_end + 1, 5):
        scene.frame_set(frame)
        bpy.context.view_layer.update()

        cam_dist = cam_obj.matrix_world.translation.length
        if cam_dist > max_cam_dist:
            max_cam_dist = cam_dist

    world_radius = max(25.0, max_cam_dist * safety_multiplier)

    mod = ground_obj.modifiers.get("Grass_Ground_GN")
    if mod and mod.node_group:
        for node in mod.node_group.nodes:
            if node.bl_idname == "GeometryNodeMeshIcoSphere":
                node.inputs["Radius"].default_value = world_radius

    if ground_obj.type == 'MESH':
        bpy.context.view_layer.objects.active = ground_obj
        bpy.ops.object.mode_set(mode='OBJECT')

        ground_scale = world_radius / 12.5
        ground_obj.scale = (ground_scale, ground_scale, 1.0)
        bpy.ops.object.transform_apply(location=False, rotation=False, scale=True)

    cam_obj.data.clip_end = max(cam_obj.data.clip_end, world_radius * 2.5)

    scene.frame_set(1)
    bpy.context.view_layer.update()


def create_grass_material(grass_dir):
    mat = bpy.data.materials.new(name="Grass_PBR_Material")
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    out_node = nodes.new("ShaderNodeOutputMaterial")
    bsdf = nodes.new("ShaderNodeBsdfPrincipled")
    links.new(bsdf.outputs["BSDF"], out_node.inputs["Surface"])

    col_files = glob.glob(os.path.join(grass_dir, "*Color*.jpg")) + glob.glob(os.path.join(grass_dir, "*JPG*.jpg"))
    if col_files:
        tex_col = nodes.new("ShaderNodeTexImage")
        tex_col.image = bpy.data.images.load(col_files[0])
        links.new(tex_col.outputs["Color"], bsdf.inputs["Base Color"])

    rough_files = glob.glob(os.path.join(grass_dir, "*Roughness*.jpg"))
    if rough_files:
        tex_rough = nodes.new("ShaderNodeTexImage")
        tex_rough.image = bpy.data.images.load(rough_files[0])
        tex_rough.image.colorspace_settings.name = 'Non-Color'
        links.new(tex_rough.outputs["Color"], bsdf.inputs["Roughness"])

    norm_files = glob.glob(os.path.join(grass_dir, "*Normal*.jpg"))
    if norm_files:
        tex_norm = nodes.new("ShaderNodeTexImage")
        tex_norm.image = bpy.data.images.load(norm_files[0])
        tex_norm.image.colorspace_settings.name = 'Non-Color'

        normal_map_node = nodes.new("ShaderNodeNormalMap")
        links.new(tex_norm.outputs["Color"], normal_map_node.inputs["Color"])
        links.new(normal_map_node.outputs["Normal"], bsdf.inputs["Normal"])

    return mat


def create_sky_material(sky_hdri_path):
    mat = bpy.data.materials.new(name="Sky_Dome_Material")
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    out_node = nodes.new("ShaderNodeOutputMaterial")
    emission = nodes.new("ShaderNodeEmission")
    emission.inputs["Strength"].default_value = 1.0

    links.new(emission.outputs["Emission"], out_node.inputs["Surface"])

    if os.path.exists(sky_hdri_path):
        tex_env = nodes.new("ShaderNodeTexEnvironment")
        tex_env.image = bpy.data.images.load(sky_hdri_path)
        links.new(tex_env.outputs["Color"], emission.inputs["Color"])

    return mat


def setup_grass_ground_geometry_nodes(grass_dir, sky_hdri_path, ground_z_offset=-0.2):
    grass_coll = bpy.data.collections.get("Grass_Artifacts_Collection")
    if not grass_coll:
        grass_coll = bpy.data.collections.new("Grass_Artifacts_Collection")
        bpy.context.scene.collection.children.link(grass_coll)

    grass_mat = create_grass_material(grass_dir)
    sky_mat = create_sky_material(sky_hdri_path)

    mesh_files = glob.glob(os.path.join(grass_dir, "*.obj")) + glob.glob(os.path.join(grass_dir, "*.fbx"))

    if mesh_files:
        for fpath in mesh_files:
            if fpath.endswith(".obj"):
                bpy.ops.wm.obj_import(filepath=fpath)
            elif fpath.endswith(".fbx"):
                bpy.ops.import_scene.fbx(filepath=fpath)

            for o in bpy.context.selected_objects:
                grass_coll.objects.link(o)
                if o.name in bpy.context.scene.collection.objects:
                    bpy.context.scene.collection.objects.unlink(o)

        for obj in list(grass_coll.objects):
            if obj.dimensions.z > 0.5:
                grass_coll.objects.unlink(obj)
                bpy.data.objects.remove(obj, do_unlink=True)
    else:
        for i in range(3):
            m_data = bpy.data.meshes.new(f"Grass_Blade_Mesh_{i}")
            g_obj = bpy.data.objects.new(f"Grass_Blade_Obj_{i}", m_data)
            grass_coll.objects.link(g_obj)

            verts = [(0, 0, 0), (0.05, 0, 0), (0.04, 0, 0.15), (0.01, 0, 0.25), (0, 0, 0.3)]
            faces = [(0, 1, 2), (1, 2, 3), (2, 3, 4)]
            m_data.from_pydata(verts, [], faces)
            m_data.update()

            if g_obj.data.materials:
                g_obj.data.materials[0] = grass_mat
            else:
                g_obj.data.materials.append(grass_mat)

            g_obj.location = (100, 100, 0)

    bpy.ops.mesh.primitive_circle_add(radius=12.5, vertices=64, fill_type='NGON', location=(0, 0, ground_z_offset))
    ground_obj = bpy.context.active_object
    ground_obj.name = "Ground_Mesh_Object"

    vg = ground_obj.vertex_groups.new(name="Grass_Weight_Group")
    all_vert_indices = [v.index for v in ground_obj.data.vertices]
    vg.add(all_vert_indices, 1.0, 'REPLACE')
    ground_obj.data.materials.append(grass_mat)
    ground_obj.data.materials.append(sky_mat)

    gn_group = bpy.data.node_groups.new("Grass_Ground_GN", type="GeometryNodeTree")
    gn_group.interface.new_socket(name="Geometry", in_out="INPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Geometry", in_out="OUTPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Density Mask", in_out="INPUT", socket_type="NodeSocketFloat")

    nodes = gn_group.nodes
    links = gn_group.links
    out_node = nodes.new("NodeGroupOutput")
    in_node = nodes.new("NodeGroupInput")

    distribute_node = nodes.new("GeometryNodeDistributePointsOnFaces")
    distribute_node.distribute_method = 'POISSON'
    distribute_node.inputs["Distance Min"].default_value = 0.02
    distribute_node.inputs["Density Max"].default_value = 300.0
    links.new(in_node.outputs["Geometry"], distribute_node.inputs["Mesh"])

    noise_node = nodes.new("ShaderNodeTexNoise")
    noise_node.inputs["Scale"].default_value = 2.5
    noise_node.inputs["Detail"].default_value = 2.0

    color_ramp = nodes.new("ShaderNodeValToRGB")
    color_ramp.color_ramp.elements[0].position = 0.20
    color_ramp.color_ramp.elements[1].position = 0.80
    links.new(noise_node.outputs["Fac"], color_ramp.inputs["Fac"])

    math_weight = nodes.new("ShaderNodeMath")
    math_weight.operation = "MULTIPLY"
    links.new(color_ramp.outputs["Color"], math_weight.inputs[0])
    links.new(in_node.outputs["Density Mask"], math_weight.inputs[1])

    links.new(math_weight.outputs["Value"], distribute_node.inputs["Density Factor"])

    instance_node = nodes.new("GeometryNodeInstanceOnPoints")
    instance_node.inputs["Pick Instance"].default_value = True
    links.new(distribute_node.outputs["Points"], instance_node.inputs["Points"])

    coll_info = nodes.new("GeometryNodeCollectionInfo")
    coll_info.inputs["Collection"].default_value = grass_coll
    coll_info.inputs["Separate Children"].default_value = True
    coll_info.inputs["Reset Children"].default_value = True
    links.new(coll_info.outputs["Instances"], instance_node.inputs["Instance"])

    rand_scale = nodes.new("FunctionNodeRandomValue")
    rand_scale.data_type = 'FLOAT'
    rand_scale.inputs["Min"].default_value = 0.25
    rand_scale.inputs["Max"].default_value = 0.50

    combine_scale = nodes.new("ShaderNodeCombineXYZ")
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["X"])
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["Y"])
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["Z"])
    links.new(combine_scale.outputs["Vector"], instance_node.inputs["Scale"])

    rand_rot_z = nodes.new("FunctionNodeRandomValue")
    rand_rot_z.data_type = 'FLOAT'
    rand_rot_z.inputs["Min"].default_value = 0.0
    rand_rot_z.inputs["Max"].default_value = math.radians(360.0)

    combine_rot = nodes.new("ShaderNodeCombineXYZ")
    links.new(rand_rot_z.outputs["Value"], combine_rot.inputs["Z"])
    links.new(combine_rot.outputs["Vector"], instance_node.inputs["Rotation"])

    sky_sphere = nodes.new("GeometryNodeMeshIcoSphere")
    sky_sphere.inputs["Radius"].default_value = 12.5
    sky_sphere.inputs["Subdivisions"].default_value = 4

    flip_faces = nodes.new("GeometryNodeFlipFaces")
    links.new(sky_sphere.outputs["Mesh"], flip_faces.inputs["Mesh"])

    transform_sky = nodes.new("GeometryNodeTransform")
    transform_sky.inputs["Translation"].default_value = (0.0, 0.0, ground_z_offset)
    links.new(flip_faces.outputs["Mesh"], transform_sky.inputs["Geometry"])

    set_sky_mat = nodes.new("GeometryNodeSetMaterial")
    set_sky_mat.inputs["Material"].default_value = sky_mat
    links.new(transform_sky.outputs["Geometry"], set_sky_mat.inputs["Geometry"])

    join_geo = nodes.new("GeometryNodeJoinGeometry")
    links.new(in_node.outputs["Geometry"], join_geo.inputs["Geometry"])
    links.new(instance_node.outputs["Instances"], join_geo.inputs["Geometry"])
    links.new(set_sky_mat.outputs["Geometry"], join_geo.inputs["Geometry"])

    links.new(join_geo.outputs["Geometry"], out_node.inputs["Geometry"])

    mod = ground_obj.modifiers.new(name="Grass_Ground_GN", type='NODES')
    mod.node_group = gn_group

    for item in gn_group.interface.items_tree:
        if getattr(item, "in_out", None) == "INPUT" and item.name == "Density Mask":
            mod[item.identifier + "_attribute_name"] = "Grass_Weight_Group"
            mod[item.identifier + "_use_attribute"] = True

    return ground_obj


# ==============================================================================
# SCROLL MATERIAL & GEOMETRY NODES SETUP
# ==============================================================================

def create_image_scroll_material(image_path):
    mat_name = "Scroll_Texture_Mat"
    mat = bpy.data.materials.get(mat_name)
    if mat:
        bpy.data.materials.remove(mat)

    mat = bpy.data.materials.new(name=mat_name)
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    output_node = nodes.new("ShaderNodeOutputMaterial")
    bsdf = nodes.new("ShaderNodeBsdfPrincipled")
    bsdf.inputs["Roughness"].default_value = 0.5

    tex_img = nodes.new("ShaderNodeTexImage")
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        tex_img.image = img
        tex_img.extension = 'REPEAT'

    uv_map_node = nodes.new("ShaderNodeUVMap")
    uv_map_node.uv_map = "UVMap"

    links.new(uv_map_node.outputs["UV"], tex_img.inputs["Vector"])
    links.new(tex_img.outputs["Color"], bsdf.inputs["Base Color"])
    links.new(bsdf.outputs["BSDF"], output_node.inputs["Surface"])

    return mat, tex_img.image if os.path.exists(image_path) else None


def create_pure_geometry_node_scroll(image_path=""):
    gn_group = bpy.data.node_groups.new("Scroll_Pure_GN", type="GeometryNodeTree")

    gn_group.interface.new_socket(name="Geometry", in_out="OUTPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Character Mesh", in_out="INPUT", socket_type="NodeSocketObject")

    sock_r_vg = gn_group.interface.new_socket(name="Right Hand Group", in_out="INPUT", socket_type="NodeSocketString")
    sock_r_vg.default_value = "mixamorig:RightHand"

    sock_l_vg = gn_group.interface.new_socket(name="Left Hand Group", in_out="INPUT", socket_type="NodeSocketString")
    sock_l_vg.default_value = "mixamorig:LeftHand"

    sock_offset = gn_group.interface.new_socket(name="Forward Offset", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_offset.default_value = 0.60

    sock_unfurl = gn_group.interface.new_socket(name="Unfurl", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_unfurl.default_value = 0.0

    sock_width = gn_group.interface.new_socket(name="Scroll Width", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_width.default_value = 0.35

    nodes = gn_group.nodes
    links = gn_group.links

    in_node = nodes.new("NodeGroupInput")
    out_node = nodes.new("NodeGroupOutput")

    char_info = nodes.new("GeometryNodeObjectInfo")
    char_info.transform_space = "RELATIVE"
    links.new(in_node.outputs["Character Mesh"], char_info.inputs["Object"])

    pos_node = nodes.new("GeometryNodeInputPosition")

    named_r = nodes.new("GeometryNodeInputNamedAttribute")
    named_r.data_type = "FLOAT"
    links.new(in_node.outputs["Right Hand Group"], named_r.inputs["Name"])

    cmp_r = nodes.new("FunctionNodeCompare")
    cmp_r.data_type = "FLOAT"
    cmp_r.operation = "GREATER_THAN"
    cmp_r.inputs[1].default_value = 0.001
    links.new(named_r.outputs["Attribute"], cmp_r.inputs[0])

    stat_r = nodes.new("GeometryNodeAttributeStatistic")
    stat_r.data_type = "FLOAT_VECTOR"
    links.new(char_info.outputs["Geometry"], stat_r.inputs["Geometry"])
    links.new(cmp_r.outputs["Result"], stat_r.inputs["Selection"])
    links.new(pos_node.outputs["Position"], stat_r.inputs["Attribute"])

    named_l = nodes.new("GeometryNodeInputNamedAttribute")
    named_l.data_type = "FLOAT"
    links.new(in_node.outputs["Left Hand Group"], named_l.inputs["Name"])

    cmp_l = nodes.new("FunctionNodeCompare")
    cmp_l.data_type = "FLOAT"
    cmp_l.operation = "GREATER_THAN"
    cmp_l.inputs[1].default_value = 0.001
    links.new(named_l.outputs["Attribute"], cmp_l.inputs[0])

    stat_l = nodes.new("GeometryNodeAttributeStatistic")
    stat_l.data_type = "FLOAT_VECTOR"
    links.new(char_info.outputs["Geometry"], stat_l.inputs["Geometry"])
    links.new(cmp_l.outputs["Result"], stat_l.inputs["Selection"])
    links.new(pos_node.outputs["Position"], stat_l.inputs["Attribute"])

    neg_offset = nodes.new("ShaderNodeMath")
    neg_offset.operation = "MULTIPLY"
    neg_offset.inputs[1].default_value = -1.0
    links.new(in_node.outputs["Forward Offset"], neg_offset.inputs[0])

    forward_vec = nodes.new("ShaderNodeCombineXYZ")
    forward_vec.inputs["X"].default_value = 0.0
    links.new(neg_offset.outputs["Value"], forward_vec.inputs["Y"])
    forward_vec.inputs["Z"].default_value = 0.0

    start_pos = nodes.new("ShaderNodeVectorMath")
    start_pos.operation = "ADD"
    links.new(stat_r.outputs["Mean"], start_pos.inputs[0])
    links.new(forward_vec.outputs["Vector"], start_pos.inputs[1])

    target_pos = nodes.new("ShaderNodeVectorMath")
    target_pos.operation = "ADD"
    links.new(stat_l.outputs["Mean"], target_pos.inputs[0])
    links.new(forward_vec.outputs["Vector"], target_pos.inputs[1])

    delta_vec = nodes.new("ShaderNodeVectorMath")
    delta_vec.operation = "SUBTRACT"
    links.new(target_pos.outputs["Vector"], delta_vec.inputs[0])
    links.new(start_pos.outputs["Vector"], delta_vec.inputs[1])

    dist_node = nodes.new("ShaderNodeVectorMath")
    dist_node.operation = "DISTANCE"
    links.new(start_pos.outputs["Vector"], dist_node.inputs[0])

    scale_vec = nodes.new("ShaderNodeVectorMath")
    scale_vec.operation = "SCALE"
    links.new(delta_vec.outputs["Vector"], scale_vec.inputs["Vector"])
    links.new(in_node.outputs["Unfurl"], scale_vec.inputs["Scale"])

    end_pos = nodes.new("ShaderNodeVectorMath")
    end_pos.operation = "ADD"
    links.new(start_pos.outputs["Vector"], end_pos.inputs[0])
    links.new(scale_vec.outputs["Vector"], end_pos.inputs[1])

    links.new(end_pos.outputs["Vector"], dist_node.inputs[1])

    curve_line = nodes.new("GeometryNodeCurvePrimitiveLine")
    links.new(start_pos.outputs["Vector"], curve_line.inputs["Start"])
    links.new(end_pos.outputs["Vector"], curve_line.inputs["End"])

    set_tilt = nodes.new("GeometryNodeSetCurveTilt")
    set_tilt.inputs["Tilt"].default_value = 1.5708
    links.new(curve_line.outputs["Curve"], set_tilt.inputs["Curve"])

    spline_param_main = nodes.new("GeometryNodeSplineParameter")
    store_u = nodes.new("GeometryNodeStoreNamedAttribute")
    store_u.inputs["Name"].default_value = "U_Factor"
    store_u.data_type = 'FLOAT'
    store_u.domain = 'POINT'
    links.new(set_tilt.outputs["Curve"], store_u.inputs["Geometry"])
    links.new(spline_param_main.outputs["Factor"], store_u.inputs["Value"])

    half_width_neg = nodes.new("ShaderNodeMath")
    half_width_neg.operation = "MULTIPLY"
    half_width_neg.inputs[1].default_value = -0.5
    links.new(in_node.outputs["Scroll Width"], half_width_neg.inputs[0])

    half_width_pos = nodes.new("ShaderNodeMath")
    half_width_pos.operation = "MULTIPLY"
    half_width_pos.inputs[1].default_value = 0.5
    links.new(in_node.outputs["Scroll Width"], half_width_pos.inputs[0])

    combine_start = nodes.new("ShaderNodeCombineXYZ")
    links.new(half_width_neg.outputs["Value"], combine_start.inputs["X"])

    combine_end = nodes.new("ShaderNodeCombineXYZ")
    links.new(half_width_pos.outputs["Value"], combine_end.inputs["X"])

    profile_line = nodes.new("GeometryNodeCurvePrimitiveLine")
    links.new(combine_start.outputs["Vector"], profile_line.inputs["Start"])
    links.new(combine_end.outputs["Vector"], profile_line.inputs["End"])

    spline_param_prof = nodes.new("GeometryNodeSplineParameter")
    store_v = nodes.new("GeometryNodeStoreNamedAttribute")
    store_v.inputs["Name"].default_value = "V_Factor"
    store_v.data_type = 'FLOAT'
    store_v.domain = 'POINT'
    links.new(profile_line.outputs["Curve"], store_v.inputs["Geometry"])
    links.new(spline_param_prof.outputs["Factor"], store_v.inputs["Value"])

    curve_to_mesh = nodes.new("GeometryNodeCurveToMesh")
    links.new(store_u.outputs["Geometry"], curve_to_mesh.inputs["Curve"])
    links.new(store_v.outputs["Geometry"], curve_to_mesh.inputs["Profile Curve"])

    img_aspect = 1.0
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        if img.size[1] > 0:
            img_aspect = img.size[0] / img.size[1]

    h_times_aspect = nodes.new("ShaderNodeMath")
    h_times_aspect.operation = "MULTIPLY"
    h_times_aspect.inputs[1].default_value = img_aspect
    links.new(in_node.outputs["Scroll Width"], h_times_aspect.inputs[0])

    tile_count = nodes.new("ShaderNodeMath")
    tile_count.operation = "DIVIDE"
    links.new(dist_node.outputs["Value"], tile_count.inputs[0])
    links.new(h_times_aspect.outputs["Value"], tile_count.inputs[1])

    read_u = nodes.new("GeometryNodeInputNamedAttribute")
    read_u.inputs["Name"].default_value = "U_Factor"
    read_u.data_type = 'FLOAT'

    scaled_u = nodes.new("ShaderNodeMath")
    scaled_u.operation = "MULTIPLY"
    links.new(read_u.outputs["Attribute"], scaled_u.inputs[0])
    links.new(tile_count.outputs["Value"], scaled_u.inputs[1])

    read_v = nodes.new("GeometryNodeInputNamedAttribute")
    read_v.inputs["Name"].default_value = "V_Factor"
    read_v.data_type = 'FLOAT'

    flip_v = nodes.new("ShaderNodeMath")
    flip_v.operation = "SUBTRACT"
    flip_v.inputs[0].default_value = 1.0
    links.new(read_v.outputs["Attribute"], flip_v.inputs[1])

    combine_uv = nodes.new("ShaderNodeCombineXYZ")
    links.new(scaled_u.outputs["Value"], combine_uv.inputs["X"])
    links.new(flip_v.outputs["Value"], combine_uv.inputs["Y"])

    store_uv = nodes.new("GeometryNodeStoreNamedAttribute")
    store_uv.inputs["Name"].default_value = "UVMap"
    store_uv.data_type = 'FLOAT2'
    store_uv.domain = 'CORNER'
    links.new(curve_to_mesh.outputs["Mesh"], store_uv.inputs["Geometry"])
    links.new(combine_uv.outputs["Vector"], store_uv.inputs["Value"])

    set_material = nodes.new("GeometryNodeSetMaterial")
    scroll_mat, _ = create_image_scroll_material(image_path)
    set_material.inputs["Material"].default_value = scroll_mat
    links.new(store_uv.outputs["Geometry"], set_material.inputs["Geometry"])

    links.new(set_material.outputs["Geometry"], out_node.inputs["Geometry"])

    return gn_group, scroll_mat


def setup_geometry_node_scroll(target_mesh, unfurl_start, unfurl_end, image_path=""):
    mesh_data = bpy.data.meshes.new("Scroll_Procedural_Mesh")
    scroll_obj = bpy.data.objects.new("Scroll_GN_Object", mesh_data)
    bpy.context.scene.collection.objects.link(scroll_obj)

    mod = scroll_obj.modifiers.new("Scroll_GN", type="NODES")
    gn_tree, scroll_mat = create_pure_geometry_node_scroll(image_path=image_path)
    mod.node_group = gn_tree

    scroll_obj.data.materials.append(scroll_mat)

    r_hand_name, l_hand_name = find_hand_vertex_groups(target_mesh)

    unfurl_id = None
    for item in gn_tree.interface.items_tree:
        if getattr(item, "in_out", None) == "INPUT":
            if item.name == "Character Mesh":
                mod[item.identifier] = target_mesh
            elif item.name == "Right Hand Group":
                mod[item.identifier] = r_hand_name
            elif item.name == "Left Hand Group":
                mod[item.identifier] = l_hand_name
            elif item.name == "Forward Offset":
                mod[item.identifier] = 0.60
            elif item.name == "Scroll Width":
                mod[item.identifier] = 0.35
            elif item.name == "Unfurl":
                unfurl_id = item.identifier

    for m in target_mesh.modifiers:
        if m.type == 'ARMATURE':
            m.show_viewport = True
            m.show_render = True
            m.show_in_editmode = True

    scroll_obj.hide_viewport = True
    scroll_obj.hide_render = True
    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=1)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=1)

    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=unfurl_start - 1)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=unfurl_start - 1)

    scroll_obj.hide_viewport = False
    scroll_obj.hide_render = False
    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=unfurl_start)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=unfurl_start)

    for fc in get_object_fcurves(scroll_obj):
        if fc.data_path in ["hide_viewport", "hide_render"]:
            for kp in fc.keyframe_points:
                kp.interpolation = 'CONSTANT'

    if unfurl_id:
        mod[unfurl_id] = 0.001
        mod.keyframe_insert(data_path=f'["{unfurl_id}"]', frame=unfurl_start)

        mod[unfurl_id] = 1.0
        mod.keyframe_insert(data_path=f'["{unfurl_id}"]', frame=unfurl_end)

    return scroll_obj


# ==============================================================================
# MAIN CONTROLLER
# ==============================================================================

def import_and_extract_action(fbx_path, action_name):
    existing_objects = set(bpy.data.objects)
    bpy.ops.import_scene.fbx(filepath=fbx_path)
    bpy.context.view_layer.update()

    imported_objects = set(bpy.data.objects) - existing_objects
    extracted_action = None
    for obj in imported_objects:
        if obj.animation_data and obj.animation_data.action:
            extracted_action = obj.animation_data.action
            extracted_action.name = action_name
            extracted_action.use_fake_user = True
            break
    return extracted_action, imported_objects


def purge_imported_objects(objects_to_remove):
    for obj in objects_to_remove:
        if obj.name in bpy.data.objects:
            if obj.animation_data:
                obj.animation_data_clear()
            for coll in list(obj.users_collection):
                coll.objects.unlink(obj)
    bpy.context.view_layer.update()


def main():
    cleanup_previous_run()

    walk_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Mutant Walking.fbx"
    turn_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Right Turn 90.fbx"
    unfurl_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Mutant Idle.fbx"

    sky_hdri_path = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/horn-koppe_spring_4k.exr"
    output_blend = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/output_scene.blend"

    image_path = "/Users/dobbinp1/Downloads/IMGP6691_psedit.jpg"
    grass_dir = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Grass004_4K-JPG"

    ground_obj = setup_grass_ground_geometry_nodes(grass_dir, sky_hdri_path, ground_z_offset=-0.2)

    CYCLES = 4
    CROSSFADE = 15
    TARGET_DISTANCE = 4.0

    walk_action, walk_objs = import_and_extract_action(walk_fbx, "walk")
    target_rig = next((obj for obj in walk_objs if obj.type == "ARMATURE"), None)
    target_mesh = next((obj for obj in walk_objs if obj.type == "MESH"), None)

    if not target_rig or not target_mesh:
        raise RuntimeError("Armature or Character Mesh not found in walk FBX.")

    travel_axis_idx, motion_sign = detect_natural_forward_axis(walk_action, target_rig)
    initial_heading = math.radians(90.0) if travel_axis_idx == 0 else 0.0
    target_rig.rotation_euler[2] = initial_heading

    action_len = int(walk_action.frame_range[1] - walk_action.frame_range[0])
    if action_len <= 0:
        action_len = 30

    anim_data = target_rig.animation_data_create()

    walk_track = anim_data.nla_tracks.new()
    walk_track.name = "Walk_Track"

    current_frame = 1
    for i in range(CYCLES):
        strip = walk_track.strips.new(name=f"Walk_{i+1}", start=int(current_frame), action=walk_action)
        strip.blend_type = "REPLACE"
        current_frame += action_len

    walk_end_frame = int(current_frame)

    turn_action, turn_objs = import_and_extract_action(turn_fbx, "rightturn")
    purge_imported_objects(turn_objs)

    turn_track = anim_data.nla_tracks.new()
    turn_track.name = "RightTurn_Track"

    turn_start = int(walk_end_frame - CROSSFADE)
    turn_strip = turn_track.strips.new(name="RightTurn", start=turn_start, action=turn_action)
    turn_strip.blend_type = "REPLACE"
    turn_strip.blend_in = CROSSFADE
    turn_strip.use_auto_blend = True
    turn_strip.extrapolation = "HOLD"
    turn_end_frame = int(turn_strip.frame_end)

    # Pass turn_start frame to scene data so handler stops sound when character stops walking
    bpy.context.scene["turn_start_frame"] = turn_start

    unfurl_action, unfurl_objs = import_and_extract_action(unfurl_fbx, "unfurl")
    purge_imported_objects(unfurl_objs)

    unfurl_track = anim_data.nla_tracks.new()
    unfurl_track.name = "Unfurl_Track"

    unfurl_start = int(turn_end_frame - CROSSFADE)
    unfurl_strip = unfurl_track.strips.new(name="Unfurl", start=unfurl_start, action=unfurl_action)
    unfurl_strip.blend_type = "REPLACE"
    unfurl_strip.blend_in = CROSSFADE
    unfurl_strip.use_auto_blend = True
    unfurl_strip.extrapolation = "HOLD"
    unfurl_end_frame = int(unfurl_strip.frame_end)

    root_action = bpy.data.actions.new(name="World_Root_Motion")
    target_rig.animation_data.action = root_action

    loc_fcurve = root_action.fcurve_ensure_for_datablock(datablock=target_rig, data_path="location", index=travel_axis_idx)
    final_target_value = TARGET_DISTANCE * motion_sign

    k_loc1 = loc_fcurve.keyframe_points.insert(frame=1.0, value=0.0)
    k_loc2 = loc_fcurve.keyframe_points.insert(frame=float(turn_start), value=final_target_value)
    k_loc3 = loc_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=final_target_value)

    k_loc1.interpolation = "LINEAR"
    k_loc2.interpolation = "CONSTANT"
    k_loc3.interpolation = "CONSTANT"

    rot_fcurve = root_action.fcurve_ensure_for_datablock(datablock=target_rig, data_path="rotation_euler", index=2)
    target_heading = initial_heading - math.radians(90.0)

    k_rot1 = rot_fcurve.keyframe_points.insert(frame=1.0, value=initial_heading)
    k_rot2 = rot_fcurve.keyframe_points.insert(frame=float(turn_start), value=initial_heading)
    k_rot3 = rot_fcurve.keyframe_points.insert(frame=float(turn_end_frame), value=target_heading)
    k_rot4 = rot_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=target_heading)

    k_rot1.interpolation = "CONSTANT"
    k_rot2.interpolation = "BEZIER"
    k_rot3.interpolation = "CONSTANT"
    k_rot4.interpolation = "CONSTANT"

    root_track = anim_data.nla_tracks.new()
    root_track.name = "World_Root_Track"
    root_strip = root_track.strips.new(name="Root_Motion_Strip", start=1, action=root_action)
    root_strip.blend_type = "REPLACE"
    root_strip.extrapolation = "HOLD"

    anim_data.action = None

    setup_geometry_node_scroll(
        target_mesh=target_mesh,
        unfurl_start=unfurl_start,
        unfurl_end=unfurl_end_frame,
        image_path=image_path,
    )

    cam_obj, focus_target = setup_modern_reel_camera(
        target_rig=target_rig,
        target_mesh=target_mesh,
        turn_start=turn_start,
        turn_end_frame=turn_end_frame,
        unfurl_end_frame=unfurl_end_frame
    )

    bpy.context.scene.frame_start = 1
    bpy.context.scene.frame_end = unfurl_end_frame

    # Animate camera zoom to frame the scroll tile perfectly
    final_frame = frame_scroll_image(
        cam_obj=cam_obj,
        focus_target=focus_target,
        target_mesh=target_mesh,
        image_path=image_path,
        scroll_width=0.35,
        zoom_duration=45
    )

    ensure_world_sphere_size(cam_obj, ground_obj, safety_multiplier=2.5)

    # Register real-time footstep audio handler
    register_audio_handler()

    bpy.ops.wm.save_as_mainfile(filepath=output_blend)
    print(f"\n[SUCCESS] Integrated parented target tracking & saved scene to: {output_blend}")


if __name__ == "__main__":
    main()
```

```
import math
import mathutils
import os
import sys
import glob
import bpy
import aud
from bpy_extras import anim_utils
from bpy_extras.object_utils import world_to_camera_view
from bpy.app.handlers import persistent
from mathutils import Vector

# ==============================================================================
# CONFIGURATION
# ==============================================================================
FOOTSTEP_SOUND_PATH = "/Users/dobbinp1/Downloads/stomp.wav"
GROUND_THRESHOLD = 0.24  # Raised slightly for testing

# Setup Audio Device
if "_audio_store" not in sys.modules:
    import types
    sys.modules["_audio_store"] = types.ModuleType("_audio_store")

audio_store = sys.modules["_audio_store"]
if not hasattr(audio_store, "device"):
    audio_store.device = aud.Device()

AUDIO_DEVICE = audio_store.device
SOUND_FACTORY = None
AUDIO_HANDLES = []

if os.path.exists(FOOTSTEP_SOUND_PATH):
    SOUND_FACTORY = aud.Sound(FOOTSTEP_SOUND_PATH).cache()
    print(f"[AUDIO] Loaded sound file successfully.")
else:
    print(f"[AUDIO ERROR] File not found at: {FOOTSTEP_SOUND_PATH}")

_left_foot_was_down = False
_right_foot_was_down = False
_last_frame = -1


def play_footstep():
    global AUDIO_HANDLES
    if SOUND_FACTORY:
        print("  >>> PLAYING FOOTSTEP SOUND <<<")
        handle = AUDIO_DEVICE.play(SOUND_FACTORY)
        AUDIO_HANDLES.append(handle)
        if len(AUDIO_HANDLES) > 10:
            AUDIO_HANDLES.pop(0)


def find_foot_bone(pose_bones, side="Left"):
    target_keywords = [
        f"mixamorig:{side}foot",
        f"{side.lower()}foot",
        f"foot.{side[0].lower()}",
        f"foot_{side[0].lower()}"
    ]
    for bone in pose_bones:
        b_name = bone.name.lower()
        if any(kw in b_name for kw in target_keywords):
            return bone
    return None


# Set threshold above the lowest observed Z (~0.227)
GROUND_THRESHOLD = 0.240

@persistent
def footstep_frame_handler(scene):
    global _left_foot_was_down, _right_foot_was_down, _last_frame

    try:
        current_frame = scene.frame_current
        turn_start = scene.get("turn_start_frame", None)

        # Stop footsteps once turning/idling starts
        if turn_start is not None and current_frame >= turn_start:
            _left_foot_was_down = False
            _right_foot_was_down = False
            return

        if current_frame <= _last_frame or abs(current_frame - _last_frame) > 10:
            _left_foot_was_down = False
            _right_foot_was_down = False
        _last_frame = current_frame

        rig = next((o for o in scene.objects if o.type == 'ARMATURE'), None)
        if not rig:
            return

        depsgraph = bpy.context.evaluated_depsgraph_get()
        eval_rig = rig.evaluated_get(depsgraph)

        left_bone = find_foot_bone(eval_rig.pose.bones, "Left")
        right_bone = find_foot_bone(eval_rig.pose.bones, "Right")

        if not left_bone or not right_bone:
            return

        l_z = (eval_rig.matrix_world @ left_bone.matrix).translation.z
        r_z = (eval_rig.matrix_world @ right_bone.matrix).translation.z

        l_is_down = l_z <= GROUND_THRESHOLD
        r_is_down = r_z <= GROUND_THRESHOLD

        if l_is_down and not _left_foot_was_down:
            play_footstep()

        if r_is_down and not _right_foot_was_down:
            play_footstep()

        _left_foot_was_down = l_is_down
        _right_foot_was_down = r_is_down

    except Exception:
        import traceback
        traceback.print_exc()


def register_audio_handler():
    # Clear existing handlers
    for h in list(bpy.app.handlers.frame_change_post):
        if getattr(h, "__name__", "") == "footstep_frame_handler":
            bpy.app.handlers.frame_change_post.remove(h)

    bpy.app.handlers.frame_change_post.append(footstep_frame_handler)
    print("[SUCCESS] Handler registered to frame_change_post.")


# 1. Register handler
register_audio_handler()


# ==============================================================================
# CLEANUP & HELPERS
# ==============================================================================

def cleanup_previous_run():
    """Clears old scroll objects, ground, node groups, camera anchors, collections, and orphaned data."""
    target_names = [
        "Master_NLA_Control",
        "Scroll_GN_Object",
        "Scroll_Start_Pin",
        "Scroll_End_Pin",
        "Scroll_Master_Container",
        "HDRI_Ground_Plane",
        "Ground_Mesh_Object",
        "Camera_Focus_Target",
        "Vertical_Reel_Camera"
    ]
    for obj in list(bpy.data.objects):
        if obj.name in target_names or (obj.name.startswith("mixamorig") and obj.type != "ARMATURE" and obj.type != "MESH"):
            if obj.animation_data:
                obj.animation_data_clear()
            bpy.data.objects.remove(obj, do_unlink=True)

    for action in list(bpy.data.actions):
        if action.users == 0 or action.use_fake_user:
            action.use_fake_user = False
            bpy.data.actions.remove(action)

    for ng in list(bpy.data.node_groups):
        if ng.name in ["Scroll_Unfurl_GN", "Scroll_Pure_GN", "HDRI_Ground_GN", "Grass_Ground_GN"]:
            bpy.data.node_groups.remove(ng)

    for coll_name in ["Scroll_Master_Container", "Grass_Artifacts_Collection"]:
        if coll_name in bpy.data.collections:
            coll = bpy.data.collections[coll_name]
            for obj in list(coll.objects):
                bpy.data.objects.remove(obj, do_unlink=True)
            bpy.data.collections.remove(coll)

    bpy.context.view_layer.update()


def get_slot_fcurves(action, anim_target_object):
    fcurves = []
    if not action:
        return fcurves

    anim_data = getattr(anim_target_object, "animation_data", None)
    if anim_data and getattr(anim_data, "action_slot", None):
        slot = anim_data.action_slot
        channelbag = anim_utils.action_get_channelbag_for_slot(action, slot)
        if channelbag and hasattr(channelbag, "fcurves"):
            return list(channelbag.fcurves)

    if hasattr(action, "slots"):
        for slot in action.slots:
            for cb in getattr(slot, "channelbags", []):
                fcurves.extend(getattr(cb, "fcurves", []))
        if fcurves:
            return fcurves

    if hasattr(action, "fcurves"):
        return list(action.fcurves)

    return fcurves


def get_object_fcurves(obj):
    if not obj or not obj.animation_data or not obj.animation_data.action:
        return []
    action = obj.animation_data.action
    return get_slot_fcurves(action, obj)


def detect_natural_forward_axis(action, rig_object):
    fcurves = get_slot_fcurves(action, rig_object)
    max_delta = 0.0
    best_axis = 1
    net_delta = 0.0

    start_frame = action.frame_range[0]
    end_frame = action.frame_range[1]

    for fc in fcurves:
        if fc.data_path.endswith("location") and fc.array_index in [0, 1, 2]:
            start_val = fc.evaluate(start_frame)
            end_val = fc.evaluate(end_frame)
            delta = end_val - start_val
            if abs(delta) > abs(max_delta):
                max_delta = delta
                best_axis = fc.array_index
                net_delta = delta

    direction_sign = 1.0 if net_delta >= 0 else -1.0
    return best_axis, direction_sign


def find_hand_vertex_groups(target_mesh):
    r_name = "mixamorig:RightHand"
    l_name = "mixamorig:LeftHand"

    if target_mesh and hasattr(target_mesh, "vertex_groups"):
        for vg in target_mesh.vertex_groups:
            vg_lower = vg.name.lower()
            if "righthand" in vg_lower or "hand_r" in vg_lower or "hand.r" in vg_lower:
                r_name = vg.name
            elif "lefthand" in vg_lower or "hand_l" in vg_lower or "hand.l" in vg_lower:
                l_name = vg.name

    return r_name, l_name


# ==============================================================================
# UNPARENTED MATRIX-EVALUATED VERTICAL CAMERA SETUP
# ==============================================================================

def setup_modern_reel_camera(target_rig, target_mesh, turn_start, turn_end_frame, unfurl_end_frame):
    scene = bpy.context.scene
    scene.render.resolution_x = 1080
    scene.render.resolution_y = 1920
    scene.render.resolution_percentage = 100

    target_height = target_mesh.dimensions.z if target_mesh.dimensions.z > 0 else 1.8
    midpoint_z_offset = target_height * 0.50

    focus_target = bpy.data.objects.get("Camera_Focus_Target")
    if not focus_target:
        focus_target = bpy.data.objects.new("Camera_Focus_Target", None)
        focus_target.empty_display_type = 'PLAIN_AXES'
        focus_target.empty_display_size = 0.25
        scene.collection.objects.link(focus_target)

    focus_target.parent = None

    copy_loc = focus_target.constraints.get("Copy Location")
    if not copy_loc:
        copy_loc = focus_target.constraints.new(type='COPY_LOCATION')
    copy_loc.target = target_mesh
    copy_loc.use_offset = True
    focus_target.location.z = midpoint_z_offset

    cam_data = bpy.data.cameras.get("Vertical_Reel_Cam_Data")
    if not cam_data:
        cam_data = bpy.data.cameras.new("Vertical_Reel_Cam_Data")
    cam_data.lens = 45.0
    cam_data.sensor_fit = 'VERTICAL'
    cam_data.clip_end = 100.0

    cam_obj = bpy.data.objects.get("Vertical_Reel_Camera")
    if not cam_obj:
        cam_obj = bpy.data.objects.new("Vertical_Reel_Camera", cam_data)
        scene.collection.objects.link(cam_obj)

    scene.camera = cam_obj
    cam_obj.parent = None

    track = cam_obj.constraints.get("Track To")
    if not track:
        track = cam_obj.constraints.new(type='TRACK_TO')
    track.target = focus_target
    track.track_axis = 'TRACK_NEGATIVE_Z'
    track.up_axis = 'UP_Y'

    def check_mesh_in_frame(frame_num, cam_location, margin=0.08):
        cam_obj.location = cam_location
        scene.frame_set(int(frame_num))
        bpy.context.view_layer.update()

        depsgraph = bpy.context.evaluated_depsgraph_get()
        eval_mesh_obj = target_mesh.evaluated_get(depsgraph)
        mesh_data = eval_mesh_obj.data

        vert_step = max(1, len(mesh_data.vertices) // 300)
        world_matrix = target_mesh.matrix_world

        min_x, max_x = 1.0, 0.0
        min_y, max_y = 1.0, 0.0

        for i in range(0, len(mesh_data.vertices), vert_step):
            v_world = world_matrix @ mesh_data.vertices[i].co
            co_2d = world_to_camera_view(scene, cam_obj, v_world)

            if co_2d.x < min_x: min_x = co_2d.x
            if co_2d.x > max_x: max_x = co_2d.x
            if co_2d.y < min_y: min_y = co_2d.y
            if co_2d.y > max_y: max_y = co_2d.y

        in_bounds = (min_x >= margin and max_x <= (1.0 - margin) and
                     min_y >= margin and max_y <= (1.0 - margin))

        return in_bounds, min_x, max_x, min_y, max_y

    def find_fit_distance(frame_num, angle_rad, base_height_offset, initial_dist=6.0, margin=0.08):
        scene.frame_set(int(frame_num))
        bpy.context.view_layer.update()

        center_pos = focus_target.matrix_world.translation.copy()
        heading = target_rig.matrix_world.to_euler('XYZ').z
        view_angle = heading + angle_rad

        dist = initial_dist
        for _ in range(30):
            test_x = center_pos.x + dist * math.sin(view_angle)
            test_y = center_pos.y - dist * math.cos(view_angle)
            test_z = center_pos.z + base_height_offset

            fit, min_x, max_x, min_y, max_y = check_mesh_in_frame(frame_num, (test_x, test_y, test_z), margin=margin)
            if fit:
                break
            dist += 0.4

        return dist, center_pos, view_angle

    max_walk_dist = 6.0
    walk_frames_to_sample = range(1, turn_start + 1, 6)

    for f in walk_frames_to_sample:
        req_d, _, _ = find_fit_distance(f, angle_rad=math.radians(-45.0), base_height_offset=0.0, initial_dist=max_walk_dist, margin=0.10)
        if req_d > max_walk_dist:
            max_walk_dist = req_d

    scene.frame_set(1)
    bpy.context.view_layer.update()
    c1 = focus_target.matrix_world.translation.copy()
    h1 = target_rig.matrix_world.to_euler('XYZ').z + math.radians(-45.0)
    w_x1 = c1.x + max_walk_dist * math.sin(h1)
    w_y1 = c1.y - max_walk_dist * math.cos(h1)
    w_z1 = c1.z

    scene.frame_set(turn_start)
    bpy.context.view_layer.update()
    cts = focus_target.matrix_world.translation.copy()
    hts = target_rig.matrix_world.to_euler('XYZ').z + math.radians(-45.0)
    w_xts = cts.x + max_walk_dist * math.sin(hts)
    w_yts = cts.y - max_walk_dist * math.cos(hts)
    w_zts = cts.z

    unfurl_frames_to_sample = range(turn_end_frame, unfurl_end_frame + 1, 6)
    max_unfurl_dist = max_walk_dist

    for f in unfurl_frames_to_sample:
        req_d, _, _ = find_fit_distance(f, angle_rad=math.radians(0.0), base_height_offset=0.0, initial_dist=max_unfurl_dist, margin=0.10)
        if req_d > max_unfurl_dist:
            max_unfurl_dist = req_d

    scene.frame_set(turn_end_frame)
    bpy.context.view_layer.update()
    cte = focus_target.matrix_world.translation.copy()
    hte = target_rig.matrix_world.to_euler('XYZ').z + math.radians(0.0)
    f_x = cte.x + max_unfurl_dist * math.sin(hte)
    f_y = cte.y - max_unfurl_dist * math.cos(hte)
    f_z = cte.z

    if cam_obj.animation_data:
        cam_obj.animation_data_clear()

    cam_obj.location = (w_x1, w_y1, w_z1)
    cam_obj.keyframe_insert(data_path="location", frame=1)

    cam_obj.location = (w_xts, w_yts, w_zts)
    cam_obj.keyframe_insert(data_path="location", frame=turn_start)

    cam_obj.location = (f_x, f_y, f_z)
    cam_obj.keyframe_insert(data_path="location", frame=turn_end_frame)

    cam_obj.location = (f_x, f_y, f_z)
    cam_obj.keyframe_insert(data_path="location", frame=unfurl_end_frame)

    for fc in get_object_fcurves(cam_obj):
        for kp in fc.keyframe_points:
            kp.interpolation = 'BEZIER'

    scene.frame_set(1)
    return cam_obj, focus_target


def frame_scroll_image(cam_obj, focus_target, target_mesh, image_path, scroll_width=0.35, zoom_duration=45):
    scene = bpy.context.scene
    unfurl_end_frame = scene.frame_end
    zoom_end_frame = unfurl_end_frame + zoom_duration

    scene.frame_set(unfurl_end_frame)
    bpy.context.view_layer.update()

    # 1. Aspect Ratio & Framing Distance Calculation
    img_aspect = 1.0
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        if img.size[1] > 0:
            img_aspect = img.size[0] / img.size[1]

    tile_height = scroll_width * img_aspect
    cam_data = cam_obj.data
    aspect_ratio = scene.render.resolution_x / scene.render.resolution_y
    sensor_width = cam_data.sensor_height * aspect_ratio

    fov_h_half = math.atan((sensor_width / 2.0) / cam_data.lens)
    target_framing_distance = (tile_height / 2.0) / math.tan(fov_h_half)

    # 2. Hand Position & Center Evaluation
    mesh_matrix = target_mesh.matrix_world
    depsgraph = bpy.context.evaluated_depsgraph_get()
    eval_mesh = target_mesh.evaluated_get(depsgraph)

    r_hand_name, l_hand_name = find_hand_vertex_groups(target_mesh)
    vg_r = eval_mesh.vertex_groups.get(r_hand_name)
    vg_l = eval_mesh.vertex_groups.get(l_hand_name)

    r_verts = [mesh_matrix @ v.co for v in eval_mesh.data.vertices if any(g.group == vg_r.index for g in v.groups)] if vg_r else []
    l_verts = [mesh_matrix @ v.co for v in eval_mesh.data.vertices if any(g.group == vg_l.index for g in v.groups)] if vg_l else []

    target_height = target_mesh.dimensions.z if target_mesh.dimensions.z > 0 else 1.8
    torso_fallback = mesh_matrix.translation + mathutils.Vector((0, 0, target_height * 0.50))

    r_hand_pos = sum(r_verts, mathutils.Vector((0, 0, 0))) / len(r_verts) if r_verts else torso_fallback
    l_hand_pos = sum(l_verts, mathutils.Vector((0, 0, 0))) / len(l_verts) if l_verts else torso_fallback

    scroll_midpoint = (r_hand_pos + l_hand_pos) * 0.5

    # 3. Guaranteed Outward Normal (Pointing TOWARD current camera)
    cam_pos = cam_obj.matrix_world.translation.copy()
    cam_dir = cam_pos - scroll_midpoint
    cam_dir.z = 0.0  # Horizontal plane projection

    if cam_dir.length > 0.001:
        scroll_normal = cam_dir.normalized()
    else:
        scroll_normal = mathutils.Vector((0, -1, 0))

    # Tile center offset slightly in front of scroll surface
    tile_center = scroll_midpoint + (scroll_normal * 0.10)

    # End position pushes OUTWARD from tile center toward camera
    end_pos = tile_center + (scroll_normal * target_framing_distance)

    # 4. Keyframing Target & Camera without Overshoot
    start_cam_pos = cam_pos
    current_focus_pos = focus_target.matrix_world.translation.copy()

    # Disable Copy Location on focus target so it can be keyframed manually
    copy_loc = focus_target.constraints.get("Copy Location")
    if copy_loc:
        copy_loc.keyframe_insert(data_path="influence", frame=unfurl_end_frame)
        copy_loc.influence = 0.0
        copy_loc.keyframe_insert(data_path="influence", frame=unfurl_end_frame + 1)

    bpy.context.view_layer.update()

    # Keyframe focus target sliding to tile_center
    focus_target.location = current_focus_pos
    focus_target.keyframe_insert(data_path="location", frame=unfurl_end_frame)
    focus_target.location = tile_center
    focus_target.keyframe_insert(data_path="location", frame=zoom_end_frame)

    # Keyframe camera moving from start location to end_pos
    cam_obj.location = start_cam_pos
    cam_obj.keyframe_insert(data_path="location", frame=unfurl_end_frame)
    cam_obj.location = end_pos
    cam_obj.keyframe_insert(data_path="location", frame=zoom_end_frame)

    # Clamp F-Curve interpolation handles to prevent clipping through geometry
    for obj in [cam_obj, focus_target]:
        for fc in get_object_fcurves(obj):
            for kp in fc.keyframe_points:
                if kp.co.x >= unfurl_end_frame:
                    kp.interpolation = 'BEZIER'
                    kp.handle_left_type = 'AUTO_CLAMPED'
                    kp.handle_right_type = 'AUTO_CLAMPED'

    scene.frame_end = zoom_end_frame
    return zoom_end_frame


# ==============================================================================
# GEOMETRY NODES GROUND, GRASS & SKY DOME SETUP
# ==============================================================================

def ensure_world_sphere_size(cam_obj, ground_obj, safety_multiplier=2.5):
    if not cam_obj or not ground_obj:
        return

    scene = bpy.context.scene
    max_cam_dist = 0.0

    for frame in range(scene.frame_start, scene.frame_end + 1, 5):
        scene.frame_set(frame)
        bpy.context.view_layer.update()

        cam_dist = cam_obj.matrix_world.translation.length
        if cam_dist > max_cam_dist:
            max_cam_dist = cam_dist

    world_radius = max(25.0, max_cam_dist * safety_multiplier)

    mod = ground_obj.modifiers.get("Grass_Ground_GN")
    if mod and mod.node_group:
        for node in mod.node_group.nodes:
            if node.bl_idname == "GeometryNodeMeshIcoSphere":
                node.inputs["Radius"].default_value = world_radius

    if ground_obj.type == 'MESH':
        bpy.context.view_layer.objects.active = ground_obj
        bpy.ops.object.mode_set(mode='OBJECT')

        ground_scale = world_radius / 12.5
        ground_obj.scale = (ground_scale, ground_scale, 1.0)
        bpy.ops.object.transform_apply(location=False, rotation=False, scale=True)

    cam_obj.data.clip_end = max(cam_obj.data.clip_end, world_radius * 2.5)

    scene.frame_set(1)
    bpy.context.view_layer.update()


def create_grass_material(grass_dir):
    mat = bpy.data.materials.new(name="Grass_PBR_Material")
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    out_node = nodes.new("ShaderNodeOutputMaterial")
    bsdf = nodes.new("ShaderNodeBsdfPrincipled")
    links.new(bsdf.outputs["BSDF"], out_node.inputs["Surface"])

    col_files = glob.glob(os.path.join(grass_dir, "*Color*.jpg")) + glob.glob(os.path.join(grass_dir, "*JPG*.jpg"))
    if col_files:
        tex_col = nodes.new("ShaderNodeTexImage")
        tex_col.image = bpy.data.images.load(col_files[0])
        links.new(tex_col.outputs["Color"], bsdf.inputs["Base Color"])

    rough_files = glob.glob(os.path.join(grass_dir, "*Roughness*.jpg"))
    if rough_files:
        tex_rough = nodes.new("ShaderNodeTexImage")
        tex_rough.image = bpy.data.images.load(rough_files[0])
        tex_rough.image.colorspace_settings.name = 'Non-Color'
        links.new(tex_rough.outputs["Color"], bsdf.inputs["Roughness"])

    norm_files = glob.glob(os.path.join(grass_dir, "*Normal*.jpg"))
    if norm_files:
        tex_norm = nodes.new("ShaderNodeTexImage")
        tex_norm.image = bpy.data.images.load(norm_files[0])
        tex_norm.image.colorspace_settings.name = 'Non-Color'

        normal_map_node = nodes.new("ShaderNodeNormalMap")
        links.new(tex_norm.outputs["Color"], normal_map_node.inputs["Color"])
        links.new(normal_map_node.outputs["Normal"], bsdf.inputs["Normal"])

    return mat


def create_sky_material(sky_hdri_path):
    mat = bpy.data.materials.new(name="Sky_Dome_Material")
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    out_node = nodes.new("ShaderNodeOutputMaterial")
    emission = nodes.new("ShaderNodeEmission")
    emission.inputs["Strength"].default_value = 1.0

    links.new(emission.outputs["Emission"], out_node.inputs["Surface"])

    if os.path.exists(sky_hdri_path):
        tex_env = nodes.new("ShaderNodeTexEnvironment")
        tex_env.image = bpy.data.images.load(sky_hdri_path)
        links.new(tex_env.outputs["Color"], emission.inputs["Color"])

    return mat


def setup_grass_ground_geometry_nodes(grass_dir, sky_hdri_path, ground_z_offset=-0.2):
    grass_coll = bpy.data.collections.get("Grass_Artifacts_Collection")
    if not grass_coll:
        grass_coll = bpy.data.collections.new("Grass_Artifacts_Collection")
        bpy.context.scene.collection.children.link(grass_coll)

    grass_mat = create_grass_material(grass_dir)
    sky_mat = create_sky_material(sky_hdri_path)

    mesh_files = glob.glob(os.path.join(grass_dir, "*.obj")) + glob.glob(os.path.join(grass_dir, "*.fbx"))

    if mesh_files:
        for fpath in mesh_files:
            if fpath.endswith(".obj"):
                bpy.ops.wm.obj_import(filepath=fpath)
            elif fpath.endswith(".fbx"):
                bpy.ops.import_scene.fbx(filepath=fpath)

            for o in bpy.context.selected_objects:
                grass_coll.objects.link(o)
                if o.name in bpy.context.scene.collection.objects:
                    bpy.context.scene.collection.objects.unlink(o)

        for obj in list(grass_coll.objects):
            if obj.dimensions.z > 0.5:
                grass_coll.objects.unlink(obj)
                bpy.data.objects.remove(obj, do_unlink=True)
    else:
        for i in range(3):
            m_data = bpy.data.meshes.new(f"Grass_Blade_Mesh_{i}")
            g_obj = bpy.data.objects.new(f"Grass_Blade_Obj_{i}", m_data)
            grass_coll.objects.link(g_obj)

            verts = [(0, 0, 0), (0.05, 0, 0), (0.04, 0, 0.15), (0.01, 0, 0.25), (0, 0, 0.3)]
            faces = [(0, 1, 2), (1, 2, 3), (2, 3, 4)]
            m_data.from_pydata(verts, [], faces)
            m_data.update()

            if g_obj.data.materials:
                g_obj.data.materials[0] = grass_mat
            else:
                g_obj.data.materials.append(grass_mat)

            g_obj.location = (100, 100, 0)

    bpy.ops.mesh.primitive_circle_add(radius=12.5, vertices=64, fill_type='NGON', location=(0, 0, ground_z_offset))
    ground_obj = bpy.context.active_object
    ground_obj.name = "Ground_Mesh_Object"

    vg = ground_obj.vertex_groups.new(name="Grass_Weight_Group")
    all_vert_indices = [v.index for v in ground_obj.data.vertices]
    vg.add(all_vert_indices, 1.0, 'REPLACE')
    ground_obj.data.materials.append(grass_mat)
    ground_obj.data.materials.append(sky_mat)

    gn_group = bpy.data.node_groups.new("Grass_Ground_GN", type="GeometryNodeTree")
    gn_group.interface.new_socket(name="Geometry", in_out="INPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Geometry", in_out="OUTPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Density Mask", in_out="INPUT", socket_type="NodeSocketFloat")

    nodes = gn_group.nodes
    links = gn_group.links
    out_node = nodes.new("NodeGroupOutput")
    in_node = nodes.new("NodeGroupInput")

    distribute_node = nodes.new("GeometryNodeDistributePointsOnFaces")
    distribute_node.distribute_method = 'POISSON'
    distribute_node.inputs["Distance Min"].default_value = 0.02
    distribute_node.inputs["Density Max"].default_value = 300.0
    links.new(in_node.outputs["Geometry"], distribute_node.inputs["Mesh"])

    noise_node = nodes.new("ShaderNodeTexNoise")
    noise_node.inputs["Scale"].default_value = 2.5
    noise_node.inputs["Detail"].default_value = 2.0

    color_ramp = nodes.new("ShaderNodeValToRGB")
    color_ramp.color_ramp.elements[0].position = 0.20
    color_ramp.color_ramp.elements[1].position = 0.80
    links.new(noise_node.outputs["Fac"], color_ramp.inputs["Fac"])

    math_weight = nodes.new("ShaderNodeMath")
    math_weight.operation = "MULTIPLY"
    links.new(color_ramp.outputs["Color"], math_weight.inputs[0])
    links.new(in_node.outputs["Density Mask"], math_weight.inputs[1])

    links.new(math_weight.outputs["Value"], distribute_node.inputs["Density Factor"])

    instance_node = nodes.new("GeometryNodeInstanceOnPoints")
    instance_node.inputs["Pick Instance"].default_value = True
    links.new(distribute_node.outputs["Points"], instance_node.inputs["Points"])

    coll_info = nodes.new("GeometryNodeCollectionInfo")
    coll_info.inputs["Collection"].default_value = grass_coll
    coll_info.inputs["Separate Children"].default_value = True
    coll_info.inputs["Reset Children"].default_value = True
    links.new(coll_info.outputs["Instances"], instance_node.inputs["Instance"])

    rand_scale = nodes.new("FunctionNodeRandomValue")
    rand_scale.data_type = 'FLOAT'
    rand_scale.inputs["Min"].default_value = 0.25
    rand_scale.inputs["Max"].default_value = 0.50

    combine_scale = nodes.new("ShaderNodeCombineXYZ")
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["X"])
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["Y"])
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["Z"])
    links.new(combine_scale.outputs["Vector"], instance_node.inputs["Scale"])

    rand_rot_z = nodes.new("FunctionNodeRandomValue")
    rand_rot_z.data_type = 'FLOAT'
    rand_rot_z.inputs["Min"].default_value = 0.0
    rand_rot_z.inputs["Max"].default_value = math.radians(360.0)

    combine_rot = nodes.new("ShaderNodeCombineXYZ")
    links.new(rand_rot_z.outputs["Value"], combine_rot.inputs["Z"])
    links.new(combine_rot.outputs["Vector"], instance_node.inputs["Rotation"])

    sky_sphere = nodes.new("GeometryNodeMeshIcoSphere")
    sky_sphere.inputs["Radius"].default_value = 12.5
    sky_sphere.inputs["Subdivisions"].default_value = 4

    flip_faces = nodes.new("GeometryNodeFlipFaces")
    links.new(sky_sphere.outputs["Mesh"], flip_faces.inputs["Mesh"])

    transform_sky = nodes.new("GeometryNodeTransform")
    transform_sky.inputs["Translation"].default_value = (0.0, 0.0, ground_z_offset)
    links.new(flip_faces.outputs["Mesh"], transform_sky.inputs["Geometry"])

    set_sky_mat = nodes.new("GeometryNodeSetMaterial")
    set_sky_mat.inputs["Material"].default_value = sky_mat
    links.new(transform_sky.outputs["Geometry"], set_sky_mat.inputs["Geometry"])

    join_geo = nodes.new("GeometryNodeJoinGeometry")
    links.new(in_node.outputs["Geometry"], join_geo.inputs["Geometry"])
    links.new(instance_node.outputs["Instances"], join_geo.inputs["Geometry"])
    links.new(set_sky_mat.outputs["Geometry"], join_geo.inputs["Geometry"])

    links.new(join_geo.outputs["Geometry"], out_node.inputs["Geometry"])

    mod = ground_obj.modifiers.new(name="Grass_Ground_GN", type='NODES')
    mod.node_group = gn_group

    for item in gn_group.interface.items_tree:
        if getattr(item, "in_out", None) == "INPUT" and item.name == "Density Mask":
            mod[item.identifier + "_attribute_name"] = "Grass_Weight_Group"
            mod[item.identifier + "_use_attribute"] = True

    return ground_obj


# ==============================================================================
# SCROLL MATERIAL & GEOMETRY NODES SETUP
# ==============================================================================

def create_image_scroll_material(image_path):
    mat_name = "Scroll_Texture_Mat"
    mat = bpy.data.materials.get(mat_name)
    if mat:
        bpy.data.materials.remove(mat)

    mat = bpy.data.materials.new(name=mat_name)
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    output_node = nodes.new("ShaderNodeOutputMaterial")
    bsdf = nodes.new("ShaderNodeBsdfPrincipled")
    bsdf.inputs["Roughness"].default_value = 0.5

    tex_img = nodes.new("ShaderNodeTexImage")
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        tex_img.image = img
        tex_img.extension = 'REPEAT'

    uv_map_node = nodes.new("ShaderNodeUVMap")
    uv_map_node.uv_map = "UVMap"

    links.new(uv_map_node.outputs["UV"], tex_img.inputs["Vector"])
    links.new(tex_img.outputs["Color"], bsdf.inputs["Base Color"])
    links.new(bsdf.outputs["BSDF"], output_node.inputs["Surface"])

    return mat, tex_img.image if os.path.exists(image_path) else None


def create_pure_geometry_node_scroll(image_path=""):
    gn_group = bpy.data.node_groups.new("Scroll_Pure_GN", type="GeometryNodeTree")

    gn_group.interface.new_socket(name="Geometry", in_out="OUTPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Character Mesh", in_out="INPUT", socket_type="NodeSocketObject")

    sock_r_vg = gn_group.interface.new_socket(name="Right Hand Group", in_out="INPUT", socket_type="NodeSocketString")
    sock_r_vg.default_value = "mixamorig:RightHand"

    sock_l_vg = gn_group.interface.new_socket(name="Left Hand Group", in_out="INPUT", socket_type="NodeSocketString")
    sock_l_vg.default_value = "mixamorig:LeftHand"

    sock_offset = gn_group.interface.new_socket(name="Forward Offset", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_offset.default_value = 0.60

    sock_unfurl = gn_group.interface.new_socket(name="Unfurl", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_unfurl.default_value = 0.0

    sock_width = gn_group.interface.new_socket(name="Scroll Width", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_width.default_value = 0.35

    nodes = gn_group.nodes
    links = gn_group.links

    in_node = nodes.new("NodeGroupInput")
    out_node = nodes.new("NodeGroupOutput")

    char_info = nodes.new("GeometryNodeObjectInfo")
    char_info.transform_space = "RELATIVE"
    links.new(in_node.outputs["Character Mesh"], char_info.inputs["Object"])

    pos_node = nodes.new("GeometryNodeInputPosition")

    named_r = nodes.new("GeometryNodeInputNamedAttribute")
    named_r.data_type = "FLOAT"
    links.new(in_node.outputs["Right Hand Group"], named_r.inputs["Name"])

    cmp_r = nodes.new("FunctionNodeCompare")
    cmp_r.data_type = "FLOAT"
    cmp_r.operation = "GREATER_THAN"
    cmp_r.inputs[1].default_value = 0.001
    links.new(named_r.outputs["Attribute"], cmp_r.inputs[0])

    stat_r = nodes.new("GeometryNodeAttributeStatistic")
    stat_r.data_type = "FLOAT_VECTOR"
    links.new(char_info.outputs["Geometry"], stat_r.inputs["Geometry"])
    links.new(cmp_r.outputs["Result"], stat_r.inputs["Selection"])
    links.new(pos_node.outputs["Position"], stat_r.inputs["Attribute"])

    named_l = nodes.new("GeometryNodeInputNamedAttribute")
    named_l.data_type = "FLOAT"
    links.new(in_node.outputs["Left Hand Group"], named_l.inputs["Name"])

    cmp_l = nodes.new("FunctionNodeCompare")
    cmp_l.data_type = "FLOAT"
    cmp_l.operation = "GREATER_THAN"
    cmp_l.inputs[1].default_value = 0.001
    links.new(named_l.outputs["Attribute"], cmp_l.inputs[0])

    stat_l = nodes.new("GeometryNodeAttributeStatistic")
    stat_l.data_type = "FLOAT_VECTOR"
    links.new(char_info.outputs["Geometry"], stat_l.inputs["Geometry"])
    links.new(cmp_l.outputs["Result"], stat_l.inputs["Selection"])
    links.new(pos_node.outputs["Position"], stat_l.inputs["Attribute"])

    neg_offset = nodes.new("ShaderNodeMath")
    neg_offset.operation = "MULTIPLY"
    neg_offset.inputs[1].default_value = -1.0
    links.new(in_node.outputs["Forward Offset"], neg_offset.inputs[0])

    forward_vec = nodes.new("ShaderNodeCombineXYZ")
    forward_vec.inputs["X"].default_value = 0.0
    links.new(neg_offset.outputs["Value"], forward_vec.inputs["Y"])
    forward_vec.inputs["Z"].default_value = 0.0

    start_pos = nodes.new("ShaderNodeVectorMath")
    start_pos.operation = "ADD"
    links.new(stat_r.outputs["Mean"], start_pos.inputs[0])
    links.new(forward_vec.outputs["Vector"], start_pos.inputs[1])

    target_pos = nodes.new("ShaderNodeVectorMath")
    target_pos.operation = "ADD"
    links.new(stat_l.outputs["Mean"], target_pos.inputs[0])
    links.new(forward_vec.outputs["Vector"], target_pos.inputs[1])

    delta_vec = nodes.new("ShaderNodeVectorMath")
    delta_vec.operation = "SUBTRACT"
    links.new(target_pos.outputs["Vector"], delta_vec.inputs[0])
    links.new(start_pos.outputs["Vector"], delta_vec.inputs[1])

    dist_node = nodes.new("ShaderNodeVectorMath")
    dist_node.operation = "DISTANCE"
    links.new(start_pos.outputs["Vector"], dist_node.inputs[0])

    scale_vec = nodes.new("ShaderNodeVectorMath")
    scale_vec.operation = "SCALE"
    links.new(delta_vec.outputs["Vector"], scale_vec.inputs["Vector"])
    links.new(in_node.outputs["Unfurl"], scale_vec.inputs["Scale"])

    end_pos = nodes.new("ShaderNodeVectorMath")
    end_pos.operation = "ADD"
    links.new(start_pos.outputs["Vector"], end_pos.inputs[0])
    links.new(scale_vec.outputs["Vector"], end_pos.inputs[1])

    links.new(end_pos.outputs["Vector"], dist_node.inputs[1])

    curve_line = nodes.new("GeometryNodeCurvePrimitiveLine")
    links.new(start_pos.outputs["Vector"], curve_line.inputs["Start"])
    links.new(end_pos.outputs["Vector"], curve_line.inputs["End"])

    set_tilt = nodes.new("GeometryNodeSetCurveTilt")
    set_tilt.inputs["Tilt"].default_value = 1.5708
    links.new(curve_line.outputs["Curve"], set_tilt.inputs["Curve"])

    spline_param_main = nodes.new("GeometryNodeSplineParameter")
    store_u = nodes.new("GeometryNodeStoreNamedAttribute")
    store_u.inputs["Name"].default_value = "U_Factor"
    store_u.data_type = 'FLOAT'
    store_u.domain = 'POINT'
    links.new(set_tilt.outputs["Curve"], store_u.inputs["Geometry"])
    links.new(spline_param_main.outputs["Factor"], store_u.inputs["Value"])

    half_width_neg = nodes.new("ShaderNodeMath")
    half_width_neg.operation = "MULTIPLY"
    half_width_neg.inputs[1].default_value = -0.5
    links.new(in_node.outputs["Scroll Width"], half_width_neg.inputs[0])

    half_width_pos = nodes.new("ShaderNodeMath")
    half_width_pos.operation = "MULTIPLY"
    half_width_pos.inputs[1].default_value = 0.5
    links.new(in_node.outputs["Scroll Width"], half_width_pos.inputs[0])

    combine_start = nodes.new("ShaderNodeCombineXYZ")
    links.new(half_width_neg.outputs["Value"], combine_start.inputs["X"])

    combine_end = nodes.new("ShaderNodeCombineXYZ")
    links.new(half_width_pos.outputs["Value"], combine_end.inputs["X"])

    profile_line = nodes.new("GeometryNodeCurvePrimitiveLine")
    links.new(combine_start.outputs["Vector"], profile_line.inputs["Start"])
    links.new(combine_end.outputs["Vector"], profile_line.inputs["End"])

    spline_param_prof = nodes.new("GeometryNodeSplineParameter")
    store_v = nodes.new("GeometryNodeStoreNamedAttribute")
    store_v.inputs["Name"].default_value = "V_Factor"
    store_v.data_type = 'FLOAT'
    store_v.domain = 'POINT'
    links.new(profile_line.outputs["Curve"], store_v.inputs["Geometry"])
    links.new(spline_param_prof.outputs["Factor"], store_v.inputs["Value"])

    curve_to_mesh = nodes.new("GeometryNodeCurveToMesh")
    links.new(store_u.outputs["Geometry"], curve_to_mesh.inputs["Curve"])
    links.new(store_v.outputs["Geometry"], curve_to_mesh.inputs["Profile Curve"])

    img_aspect = 1.0
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        if img.size[1] > 0:
            img_aspect = img.size[0] / img.size[1]

    h_times_aspect = nodes.new("ShaderNodeMath")
    h_times_aspect.operation = "MULTIPLY"
    h_times_aspect.inputs[1].default_value = img_aspect
    links.new(in_node.outputs["Scroll Width"], h_times_aspect.inputs[0])

    tile_count = nodes.new("ShaderNodeMath")
    tile_count.operation = "DIVIDE"
    links.new(dist_node.outputs["Value"], tile_count.inputs[0])
    links.new(h_times_aspect.outputs["Value"], tile_count.inputs[1])

    read_u = nodes.new("GeometryNodeInputNamedAttribute")
    read_u.inputs["Name"].default_value = "U_Factor"
    read_u.data_type = 'FLOAT'

    scaled_u = nodes.new("ShaderNodeMath")
    scaled_u.operation = "MULTIPLY"
    links.new(read_u.outputs["Attribute"], scaled_u.inputs[0])
    links.new(tile_count.outputs["Value"], scaled_u.inputs[1])

    read_v = nodes.new("GeometryNodeInputNamedAttribute")
    read_v.inputs["Name"].default_value = "V_Factor"
    read_v.data_type = 'FLOAT'

    flip_v = nodes.new("ShaderNodeMath")
    flip_v.operation = "SUBTRACT"
    flip_v.inputs[0].default_value = 1.0
    links.new(read_v.outputs["Attribute"], flip_v.inputs[1])

    combine_uv = nodes.new("ShaderNodeCombineXYZ")
    links.new(scaled_u.outputs["Value"], combine_uv.inputs["X"])
    links.new(flip_v.outputs["Value"], combine_uv.inputs["Y"])

    store_uv = nodes.new("GeometryNodeStoreNamedAttribute")
    store_uv.inputs["Name"].default_value = "UVMap"
    store_uv.data_type = 'FLOAT2'
    store_uv.domain = 'CORNER'
    links.new(curve_to_mesh.outputs["Mesh"], store_uv.inputs["Geometry"])
    links.new(combine_uv.outputs["Vector"], store_uv.inputs["Value"])

    set_material = nodes.new("GeometryNodeSetMaterial")
    scroll_mat, _ = create_image_scroll_material(image_path)
    set_material.inputs["Material"].default_value = scroll_mat
    links.new(store_uv.outputs["Geometry"], set_material.inputs["Geometry"])

    links.new(set_material.outputs["Geometry"], out_node.inputs["Geometry"])

    return gn_group, scroll_mat


def setup_geometry_node_scroll(target_mesh, unfurl_start, unfurl_end, image_path=""):
    mesh_data = bpy.data.meshes.new("Scroll_Procedural_Mesh")
    scroll_obj = bpy.data.objects.new("Scroll_GN_Object", mesh_data)
    bpy.context.scene.collection.objects.link(scroll_obj)

    mod = scroll_obj.modifiers.new("Scroll_GN", type="NODES")
    gn_tree, scroll_mat = create_pure_geometry_node_scroll(image_path=image_path)
    mod.node_group = gn_tree

    scroll_obj.data.materials.append(scroll_mat)

    r_hand_name, l_hand_name = find_hand_vertex_groups(target_mesh)

    unfurl_id = None
    for item in gn_tree.interface.items_tree:
        if getattr(item, "in_out", None) == "INPUT":
            if item.name == "Character Mesh":
                mod[item.identifier] = target_mesh
            elif item.name == "Right Hand Group":
                mod[item.identifier] = r_hand_name
            elif item.name == "Left Hand Group":
                mod[item.identifier] = l_hand_name
            elif item.name == "Forward Offset":
                mod[item.identifier] = 0.60
            elif item.name == "Scroll Width":
                mod[item.identifier] = 0.35
            elif item.name == "Unfurl":
                unfurl_id = item.identifier

    for m in target_mesh.modifiers:
        if m.type == 'ARMATURE':
            m.show_viewport = True
            m.show_render = True
            m.show_in_editmode = True

    scroll_obj.hide_viewport = True
    scroll_obj.hide_render = True
    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=1)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=1)

    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=unfurl_start - 1)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=unfurl_start - 1)

    scroll_obj.hide_viewport = False
    scroll_obj.hide_render = False
    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=unfurl_start)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=unfurl_start)

    for fc in get_object_fcurves(scroll_obj):
        if fc.data_path in ["hide_viewport", "hide_render"]:
            for kp in fc.keyframe_points:
                kp.interpolation = 'CONSTANT'

    if unfurl_id:
        mod[unfurl_id] = 0.001
        mod.keyframe_insert(data_path=f'["{unfurl_id}"]', frame=unfurl_start)

        mod[unfurl_id] = 1.0
        mod.keyframe_insert(data_path=f'["{unfurl_id}"]', frame=unfurl_end)

    return scroll_obj


# ==============================================================================
# MAIN CONTROLLER
# ==============================================================================

def import_and_extract_action(fbx_path, action_name):
    existing_objects = set(bpy.data.objects)
    bpy.ops.import_scene.fbx(filepath=fbx_path)
    bpy.context.view_layer.update()

    imported_objects = set(bpy.data.objects) - existing_objects
    extracted_action = None
    for obj in imported_objects:
        if obj.animation_data and obj.animation_data.action:
            extracted_action = obj.animation_data.action
            extracted_action.name = action_name
            extracted_action.use_fake_user = True
            break
    return extracted_action, imported_objects


def purge_imported_objects(objects_to_remove):
    for obj in objects_to_remove:
        if obj.name in bpy.data.objects:
            if obj.animation_data:
                obj.animation_data_clear()
            for coll in list(obj.users_collection):
                coll.objects.unlink(obj)
    bpy.context.view_layer.update()


def main():
    cleanup_previous_run()

    walk_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Mutant Walking.fbx"
    turn_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Right Turn 90.fbx"
    unfurl_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Mutant Idle.fbx"

    sky_hdri_path = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/horn-koppe_spring_4k.exr"
    output_blend = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/output_scene.blend"

    image_path = "/Users/dobbinp1/Downloads/IMGP6691_psedit.jpg"
    grass_dir = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Grass004_4K-JPG"

    ground_obj = setup_grass_ground_geometry_nodes(grass_dir, sky_hdri_path, ground_z_offset=-0.2)

    CYCLES = 4
    CROSSFADE = 15
    TARGET_DISTANCE = 4.0

    walk_action, walk_objs = import_and_extract_action(walk_fbx, "walk")
    target_rig = next((obj for obj in walk_objs if obj.type == "ARMATURE"), None)
    target_mesh = next((obj for obj in walk_objs if obj.type == "MESH"), None)

    if not target_rig or not target_mesh:
        raise RuntimeError("Armature or Character Mesh not found in walk FBX.")

    travel_axis_idx, motion_sign = detect_natural_forward_axis(walk_action, target_rig)
    initial_heading = math.radians(90.0) if travel_axis_idx == 0 else 0.0
    target_rig.rotation_euler[2] = initial_heading

    action_len = int(walk_action.frame_range[1] - walk_action.frame_range[0])
    if action_len <= 0:
        action_len = 30

    anim_data = target_rig.animation_data_create()

    walk_track = anim_data.nla_tracks.new()
    walk_track.name = "Walk_Track"

    current_frame = 1
    for i in range(CYCLES):
        strip = walk_track.strips.new(name=f"Walk_{i+1}", start=int(current_frame), action=walk_action)
        strip.blend_type = "REPLACE"
        current_frame += action_len

    walk_end_frame = int(current_frame)

    turn_action, turn_objs = import_and_extract_action(turn_fbx, "rightturn")
    purge_imported_objects(turn_objs)

    turn_track = anim_data.nla_tracks.new()
    turn_track.name = "RightTurn_Track"

    turn_start = int(walk_end_frame - CROSSFADE)
    turn_strip = turn_track.strips.new(name="RightTurn", start=turn_start, action=turn_action)
    turn_strip.blend_type = "REPLACE"
    turn_strip.blend_in = CROSSFADE
    turn_strip.use_auto_blend = True
    turn_strip.extrapolation = "HOLD"
    turn_end_frame = int(turn_strip.frame_end)

    # Pass turn_start frame to scene data so handler stops sound when character stops walking
    bpy.context.scene["turn_start_frame"] = turn_start

    unfurl_action, unfurl_objs = import_and_extract_action(unfurl_fbx, "unfurl")
    purge_imported_objects(unfurl_objs)

    unfurl_track = anim_data.nla_tracks.new()
    unfurl_track.name = "Unfurl_Track"

    unfurl_start = int(turn_end_frame - CROSSFADE)
    unfurl_strip = unfurl_track.strips.new(name="Unfurl", start=unfurl_start, action=unfurl_action)
    unfurl_strip.blend_type = "REPLACE"
    unfurl_strip.blend_in = CROSSFADE
    unfurl_strip.use_auto_blend = True
    unfurl_strip.extrapolation = "HOLD"
    unfurl_end_frame = int(unfurl_strip.frame_end)

    root_action = bpy.data.actions.new(name="World_Root_Motion")
    target_rig.animation_data.action = root_action

    loc_fcurve = root_action.fcurve_ensure_for_datablock(datablock=target_rig, data_path="location", index=travel_axis_idx)
    final_target_value = TARGET_DISTANCE * motion_sign

    k_loc1 = loc_fcurve.keyframe_points.insert(frame=1.0, value=0.0)
    k_loc2 = loc_fcurve.keyframe_points.insert(frame=float(turn_start), value=final_target_value)
    k_loc3 = loc_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=final_target_value)

    k_loc1.interpolation = "LINEAR"
    k_loc2.interpolation = "CONSTANT"
    k_loc3.interpolation = "CONSTANT"

    rot_fcurve = root_action.fcurve_ensure_for_datablock(datablock=target_rig, data_path="rotation_euler", index=2)
    target_heading = initial_heading - math.radians(90.0)

    k_rot1 = rot_fcurve.keyframe_points.insert(frame=1.0, value=initial_heading)
    k_rot2 = rot_fcurve.keyframe_points.insert(frame=float(turn_start), value=initial_heading)
    k_rot3 = rot_fcurve.keyframe_points.insert(frame=float(turn_end_frame), value=target_heading)
    k_rot4 = rot_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=target_heading)

    k_rot1.interpolation = "CONSTANT"
    k_rot2.interpolation = "BEZIER"
    k_rot3.interpolation = "CONSTANT"
    k_rot4.interpolation = "CONSTANT"

    root_track = anim_data.nla_tracks.new()
    root_track.name = "World_Root_Track"
    root_strip = root_track.strips.new(name="Root_Motion_Strip", start=1, action=root_action)
    root_strip.blend_type = "REPLACE"
    root_strip.extrapolation = "HOLD"

    anim_data.action = None

    setup_geometry_node_scroll(
        target_mesh=target_mesh,
        unfurl_start=unfurl_start,
        unfurl_end=unfurl_end_frame,
        image_path=image_path,
    )

    cam_obj, focus_target = setup_modern_reel_camera(
        target_rig=target_rig,
        target_mesh=target_mesh,
        turn_start=turn_start,
        turn_end_frame=turn_end_frame,
        unfurl_end_frame=unfurl_end_frame
    )

    bpy.context.scene.frame_start = 1
    bpy.context.scene.frame_end = unfurl_end_frame

    # Animate camera zoom to frame the scroll tile perfectly
    final_frame = frame_scroll_image(
        cam_obj=cam_obj,
        focus_target=focus_target,
        target_mesh=target_mesh,
        image_path=image_path,
        scroll_width=0.35,
        zoom_duration=45
    )

    ensure_world_sphere_size(cam_obj, ground_obj, safety_multiplier=2.5)

    # Register real-time footstep audio handler
    register_audio_handler()

    bpy.ops.wm.save_as_mainfile(filepath=output_blend)
    print(f"\n[SUCCESS] Integrated real-time audio handler & saved scene to: {output_blend}")


if __name__ == "__main__":
    main()
```

```
import math
import mathutils
import os
import sys
import glob
import bpy
import aud
from bpy_extras import anim_utils
from bpy_extras.object_utils import world_to_camera_view
from bpy.app.handlers import persistent

# ==============================================================================
# CONFIGURATION
# ==============================================================================
FOOTSTEP_SOUND_PATH = "/Users/dobbinp1/Downloads/stomp.wav"
GROUND_THRESHOLD = 0.24  # Raised slightly for testing

# Setup Audio Device
if "_audio_store" not in sys.modules:
    import types
    sys.modules["_audio_store"] = types.ModuleType("_audio_store")

audio_store = sys.modules["_audio_store"]
if not hasattr(audio_store, "device"):
    audio_store.device = aud.Device()

AUDIO_DEVICE = audio_store.device
SOUND_FACTORY = None
AUDIO_HANDLES = []

if os.path.exists(FOOTSTEP_SOUND_PATH):
    SOUND_FACTORY = aud.Sound(FOOTSTEP_SOUND_PATH).cache()
    print(f"[AUDIO] Loaded sound file successfully.")
else:
    print(f"[AUDIO ERROR] File not found at: {FOOTSTEP_SOUND_PATH}")

_left_foot_was_down = False
_right_foot_was_down = False
_last_frame = -1


def play_footstep():
    global AUDIO_HANDLES
    if SOUND_FACTORY:
        print("  >>> PLAYING FOOTSTEP SOUND <<<")
        handle = AUDIO_DEVICE.play(SOUND_FACTORY)
        AUDIO_HANDLES.append(handle)
        if len(AUDIO_HANDLES) > 10:
            AUDIO_HANDLES.pop(0)


def find_foot_bone(pose_bones, side="Left"):
    target_keywords = [
        f"mixamorig:{side}foot",
        f"{side.lower()}foot",
        f"foot.{side[0].lower()}",
        f"foot_{side[0].lower()}"
    ]
    for bone in pose_bones:
        b_name = bone.name.lower()
        if any(kw in b_name for kw in target_keywords):
            return bone
    return None


# Set threshold above the lowest observed Z (~0.227)
GROUND_THRESHOLD = 0.240

@persistent
def footstep_frame_handler(scene):
    global _left_foot_was_down, _right_foot_was_down, _last_frame

    try:
        current_frame = scene.frame_current

        # Reset states on scrub or frame jump back
        if current_frame <= _last_frame or abs(current_frame - _last_frame) > 10:
            _left_foot_was_down = False
            _right_foot_was_down = False
        _last_frame = current_frame

        # Find Armature
        rig = next((o for o in scene.objects if o.type == 'ARMATURE'), None)
        if not rig:
            return

        depsgraph = bpy.context.evaluated_depsgraph_get()
        eval_rig = rig.evaluated_get(depsgraph)

        left_bone = find_foot_bone(eval_rig.pose.bones, "Left")
        right_bone = find_foot_bone(eval_rig.pose.bones, "Right")

        if not left_bone or not right_bone:
            return

        l_z = (eval_rig.matrix_world @ left_bone.matrix).translation.z
        r_z = (eval_rig.matrix_world @ right_bone.matrix).translation.z

        l_is_down = l_z <= GROUND_THRESHOLD
        r_is_down = r_z <= GROUND_THRESHOLD

        # State transition triggers
        if l_is_down and not _left_foot_was_down:
            print(f"*** STEP LEFT TRIGGERED (Frame {current_frame} | Z: {l_z:.4f}) ***")
            play_footstep()

        if r_is_down and not _right_foot_was_down:
            print(f"*** STEP RIGHT TRIGGERED (Frame {current_frame} | Z: {r_z:.4f}) ***")
            play_footstep()

        _left_foot_was_down = l_is_down
        _right_foot_was_down = r_is_down

    except Exception as e:
        import traceback
        print("\n[HANDLER ERROR]")
        traceback.print_exc()


def register_audio_handler():
    # Clear existing handlers
    for h in list(bpy.app.handlers.frame_change_post):
        if getattr(h, "__name__", "") == "footstep_frame_handler":
            bpy.app.handlers.frame_change_post.remove(h)

    bpy.app.handlers.frame_change_post.append(footstep_frame_handler)
    print("[SUCCESS] Handler registered to frame_change_post.")


# 1. Register handler
register_audio_handler()


# ==============================================================================
# CLEANUP & HELPERS
# ==============================================================================

def cleanup_previous_run():
    """Clears old scroll objects, ground, node groups, camera anchors, collections, and orphaned data."""
    #unregister_audio_handler()

    target_names = [
        "Master_NLA_Control",
        "Scroll_GN_Object",
        "Scroll_Start_Pin",
        "Scroll_End_Pin",
        "Scroll_Master_Container",
        "HDRI_Ground_Plane",
        "Ground_Mesh_Object",
        "Camera_Focus_Target",
        "Vertical_Reel_Camera"
    ]
    for obj in list(bpy.data.objects):
        if obj.name in target_names or (obj.name.startswith("mixamorig") and obj.type != "ARMATURE" and obj.type != "MESH"):
            if obj.animation_data:
                obj.animation_data_clear()
            bpy.data.objects.remove(obj, do_unlink=True)

    for action in list(bpy.data.actions):
        if action.users == 0 or action.use_fake_user:
            action.use_fake_user = False
            bpy.data.actions.remove(action)

    for ng in list(bpy.data.node_groups):
        if ng.name in ["Scroll_Unfurl_GN", "Scroll_Pure_GN", "HDRI_Ground_GN", "Grass_Ground_GN"]:
            bpy.data.node_groups.remove(ng)

    for coll_name in ["Scroll_Master_Container", "Grass_Artifacts_Collection"]:
        if coll_name in bpy.data.collections:
            coll = bpy.data.collections[coll_name]
            for obj in list(coll.objects):
                bpy.data.objects.remove(obj, do_unlink=True)
            bpy.data.collections.remove(coll)

    bpy.context.view_layer.update()


def get_slot_fcurves(action, anim_target_object):
    fcurves = []
    if not action:
        return fcurves

    anim_data = getattr(anim_target_object, "animation_data", None)
    if anim_data and getattr(anim_data, "action_slot", None):
        slot = anim_data.action_slot
        channelbag = anim_utils.action_get_channelbag_for_slot(action, slot)
        if channelbag and hasattr(channelbag, "fcurves"):
            return list(channelbag.fcurves)

    if hasattr(action, "slots"):
        for slot in action.slots:
            for cb in getattr(slot, "channelbags", []):
                fcurves.extend(getattr(cb, "fcurves", []))
        if fcurves:
            return fcurves

    if hasattr(action, "fcurves"):
        return list(action.fcurves)

    return fcurves


def get_object_fcurves(obj):
    if not obj or not obj.animation_data or not obj.animation_data.action:
        return []
    action = obj.animation_data.action
    return get_slot_fcurves(action, obj)


def detect_natural_forward_axis(action, rig_object):
    fcurves = get_slot_fcurves(action, rig_object)
    max_delta = 0.0
    best_axis = 1
    net_delta = 0.0

    start_frame = action.frame_range[0]
    end_frame = action.frame_range[1]

    for fc in fcurves:
        if fc.data_path.endswith("location") and fc.array_index in [0, 1, 2]:
            start_val = fc.evaluate(start_frame)
            end_val = fc.evaluate(end_frame)
            delta = end_val - start_val
            if abs(delta) > abs(max_delta):
                max_delta = delta
                best_axis = fc.array_index
                net_delta = delta

    direction_sign = 1.0 if net_delta >= 0 else -1.0
    return best_axis, direction_sign


def find_hand_vertex_groups(target_mesh):
    r_name = "mixamorig:RightHand"
    l_name = "mixamorig:LeftHand"

    if target_mesh and hasattr(target_mesh, "vertex_groups"):
        for vg in target_mesh.vertex_groups:
            vg_lower = vg.name.lower()
            if "righthand" in vg_lower or "hand_r" in vg_lower or "hand.r" in vg_lower:
                r_name = vg.name
            elif "lefthand" in vg_lower or "hand_l" in vg_lower or "hand.l" in vg_lower:
                l_name = vg.name

    return r_name, l_name


# ==============================================================================
# UNPARENTED MATRIX-EVALUATED VERTICAL CAMERA SETUP
# ==============================================================================

def setup_modern_reel_camera(target_rig, target_mesh, turn_start, turn_end_frame, unfurl_end_frame):
    scene = bpy.context.scene
    scene.render.resolution_x = 1080
    scene.render.resolution_y = 1920
    scene.render.resolution_percentage = 100

    target_height = target_mesh.dimensions.z if target_mesh.dimensions.z > 0 else 1.8
    midpoint_z_offset = target_height * 0.50

    focus_target = bpy.data.objects.get("Camera_Focus_Target")
    if not focus_target:
        focus_target = bpy.data.objects.new("Camera_Focus_Target", None)
        focus_target.empty_display_type = 'PLAIN_AXES'
        focus_target.empty_display_size = 0.25
        scene.collection.objects.link(focus_target)

    focus_target.parent = None

    copy_loc = focus_target.constraints.get("Copy Location")
    if not copy_loc:
        copy_loc = focus_target.constraints.new(type='COPY_LOCATION')
    copy_loc.target = target_mesh
    copy_loc.use_offset = True
    focus_target.location.z = midpoint_z_offset

    cam_data = bpy.data.cameras.get("Vertical_Reel_Cam_Data")
    if not cam_data:
        cam_data = bpy.data.cameras.new("Vertical_Reel_Cam_Data")
    cam_data.lens = 45.0
    cam_data.sensor_fit = 'VERTICAL'
    cam_data.clip_end = 100.0

    cam_obj = bpy.data.objects.get("Vertical_Reel_Camera")
    if not cam_obj:
        cam_obj = bpy.data.objects.new("Vertical_Reel_Camera", cam_data)
        scene.collection.objects.link(cam_obj)

    scene.camera = cam_obj
    cam_obj.parent = None

    track = cam_obj.constraints.get("Track To")
    if not track:
        track = cam_obj.constraints.new(type='TRACK_TO')
    track.target = focus_target
    track.track_axis = 'TRACK_NEGATIVE_Z'
    track.up_axis = 'UP_Y'

    def check_mesh_in_frame(frame_num, cam_location, margin=0.08):
        cam_obj.location = cam_location
        scene.frame_set(int(frame_num))
        bpy.context.view_layer.update()

        depsgraph = bpy.context.evaluated_depsgraph_get()
        eval_mesh_obj = target_mesh.evaluated_get(depsgraph)
        mesh_data = eval_mesh_obj.data

        vert_step = max(1, len(mesh_data.vertices) // 300)
        world_matrix = target_mesh.matrix_world

        min_x, max_x = 1.0, 0.0
        min_y, max_y = 1.0, 0.0

        for i in range(0, len(mesh_data.vertices), vert_step):
            v_world = world_matrix @ mesh_data.vertices[i].co
            co_2d = world_to_camera_view(scene, cam_obj, v_world)

            if co_2d.x < min_x: min_x = co_2d.x
            if co_2d.x > max_x: max_x = co_2d.x
            if co_2d.y < min_y: min_y = co_2d.y
            if co_2d.y > max_y: max_y = co_2d.y

        in_bounds = (min_x >= margin and max_x <= (1.0 - margin) and
                     min_y >= margin and max_y <= (1.0 - margin))

        return in_bounds, min_x, max_x, min_y, max_y

    def find_fit_distance(frame_num, angle_rad, base_height_offset, initial_dist=6.0, margin=0.08):
        scene.frame_set(int(frame_num))
        bpy.context.view_layer.update()

        center_pos = focus_target.matrix_world.translation.copy()
        heading = target_rig.matrix_world.to_euler('XYZ').z
        view_angle = heading + angle_rad

        dist = initial_dist
        for _ in range(30):
            test_x = center_pos.x + dist * math.sin(view_angle)
            test_y = center_pos.y - dist * math.cos(view_angle)
            test_z = center_pos.z + base_height_offset

            fit, min_x, max_x, min_y, max_y = check_mesh_in_frame(frame_num, (test_x, test_y, test_z), margin=margin)
            if fit:
                break
            dist += 0.4

        return dist, center_pos, view_angle

    max_walk_dist = 6.0
    walk_frames_to_sample = range(1, turn_start + 1, 6)

    for f in walk_frames_to_sample:
        req_d, _, _ = find_fit_distance(f, angle_rad=math.radians(-45.0), base_height_offset=0.0, initial_dist=max_walk_dist, margin=0.10)
        if req_d > max_walk_dist:
            max_walk_dist = req_d

    scene.frame_set(1)
    bpy.context.view_layer.update()
    c1 = focus_target.matrix_world.translation.copy()
    h1 = target_rig.matrix_world.to_euler('XYZ').z + math.radians(-45.0)
    w_x1 = c1.x + max_walk_dist * math.sin(h1)
    w_y1 = c1.y - max_walk_dist * math.cos(h1)
    w_z1 = c1.z

    scene.frame_set(turn_start)
    bpy.context.view_layer.update()
    cts = focus_target.matrix_world.translation.copy()
    hts = target_rig.matrix_world.to_euler('XYZ').z + math.radians(-45.0)
    w_xts = cts.x + max_walk_dist * math.sin(hts)
    w_yts = cts.y - max_walk_dist * math.cos(hts)
    w_zts = cts.z

    unfurl_frames_to_sample = range(turn_end_frame, unfurl_end_frame + 1, 6)
    max_unfurl_dist = max_walk_dist

    for f in unfurl_frames_to_sample:
        req_d, _, _ = find_fit_distance(f, angle_rad=math.radians(0.0), base_height_offset=0.0, initial_dist=max_unfurl_dist, margin=0.10)
        if req_d > max_unfurl_dist:
            max_unfurl_dist = req_d

    scene.frame_set(turn_end_frame)
    bpy.context.view_layer.update()
    cte = focus_target.matrix_world.translation.copy()
    hte = target_rig.matrix_world.to_euler('XYZ').z + math.radians(0.0)
    f_x = cte.x + max_unfurl_dist * math.sin(hte)
    f_y = cte.y - max_unfurl_dist * math.cos(hte)
    f_z = cte.z

    if cam_obj.animation_data:
        cam_obj.animation_data_clear()

    cam_obj.location = (w_x1, w_y1, w_z1)
    cam_obj.keyframe_insert(data_path="location", frame=1)

    cam_obj.location = (w_xts, w_yts, w_zts)
    cam_obj.keyframe_insert(data_path="location", frame=turn_start)

    cam_obj.location = (f_x, f_y, f_z)
    cam_obj.keyframe_insert(data_path="location", frame=turn_end_frame)

    cam_obj.location = (f_x, f_y, f_z)
    cam_obj.keyframe_insert(data_path="location", frame=unfurl_end_frame)

    for fc in get_object_fcurves(cam_obj):
        for kp in fc.keyframe_points:
            kp.interpolation = 'BEZIER'

    scene.frame_set(1)
    return cam_obj, focus_target


# ==============================================================================
# GEOMETRY NODES GROUND, GRASS & SKY DOME SETUP
# ==============================================================================

def ensure_world_sphere_size(cam_obj, ground_obj, safety_multiplier=2.5):
    if not cam_obj or not ground_obj:
        return

    scene = bpy.context.scene
    max_cam_dist = 0.0

    for frame in range(scene.frame_start, scene.frame_end + 1, 5):
        scene.frame_set(frame)
        bpy.context.view_layer.update()

        cam_dist = cam_obj.matrix_world.translation.length
        if cam_dist > max_cam_dist:
            max_cam_dist = cam_dist

    world_radius = max(25.0, max_cam_dist * safety_multiplier)

    mod = ground_obj.modifiers.get("Grass_Ground_GN")
    if mod and mod.node_group:
        for node in mod.node_group.nodes:
            if node.bl_idname == "GeometryNodeMeshIcoSphere":
                node.inputs["Radius"].default_value = world_radius

    if ground_obj.type == 'MESH':
        bpy.context.view_layer.objects.active = ground_obj
        bpy.ops.object.mode_set(mode='OBJECT')

        ground_scale = world_radius / 12.5
        ground_obj.scale = (ground_scale, ground_scale, 1.0)
        bpy.ops.object.transform_apply(location=False, rotation=False, scale=True)

    cam_obj.data.clip_end = max(cam_obj.data.clip_end, world_radius * 2.5)

    scene.frame_set(1)
    bpy.context.view_layer.update()


def create_grass_material(grass_dir):
    mat = bpy.data.materials.new(name="Grass_PBR_Material")
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    out_node = nodes.new("ShaderNodeOutputMaterial")
    bsdf = nodes.new("ShaderNodeBsdfPrincipled")
    links.new(bsdf.outputs["BSDF"], out_node.inputs["Surface"])

    col_files = glob.glob(os.path.join(grass_dir, "*Color*.jpg")) + glob.glob(os.path.join(grass_dir, "*JPG*.jpg"))
    if col_files:
        tex_col = nodes.new("ShaderNodeTexImage")
        tex_col.image = bpy.data.images.load(col_files[0])
        links.new(tex_col.outputs["Color"], bsdf.inputs["Base Color"])

    rough_files = glob.glob(os.path.join(grass_dir, "*Roughness*.jpg"))
    if rough_files:
        tex_rough = nodes.new("ShaderNodeTexImage")
        tex_rough.image = bpy.data.images.load(rough_files[0])
        tex_rough.image.colorspace_settings.name = 'Non-Color'
        links.new(tex_rough.outputs["Color"], bsdf.inputs["Roughness"])

    norm_files = glob.glob(os.path.join(grass_dir, "*Normal*.jpg"))
    if norm_files:
        tex_norm = nodes.new("ShaderNodeTexImage")
        tex_norm.image = bpy.data.images.load(norm_files[0])
        tex_norm.image.colorspace_settings.name = 'Non-Color'

        normal_map_node = nodes.new("ShaderNodeNormalMap")
        links.new(tex_norm.outputs["Color"], normal_map_node.inputs["Color"])
        links.new(normal_map_node.outputs["Normal"], bsdf.inputs["Normal"])

    return mat


def create_sky_material(sky_hdri_path):
    mat = bpy.data.materials.new(name="Sky_Dome_Material")
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    out_node = nodes.new("ShaderNodeOutputMaterial")
    emission = nodes.new("ShaderNodeEmission")
    emission.inputs["Strength"].default_value = 1.0

    links.new(emission.outputs["Emission"], out_node.inputs["Surface"])

    if os.path.exists(sky_hdri_path):
        tex_env = nodes.new("ShaderNodeTexEnvironment")
        tex_env.image = bpy.data.images.load(sky_hdri_path)
        links.new(tex_env.outputs["Color"], emission.inputs["Color"])

    return mat


def setup_grass_ground_geometry_nodes(grass_dir, sky_hdri_path, ground_z_offset=-0.2):
    grass_coll = bpy.data.collections.get("Grass_Artifacts_Collection")
    if not grass_coll:
        grass_coll = bpy.data.collections.new("Grass_Artifacts_Collection")
        bpy.context.scene.collection.children.link(grass_coll)

    grass_mat = create_grass_material(grass_dir)
    sky_mat = create_sky_material(sky_hdri_path)

    mesh_files = glob.glob(os.path.join(grass_dir, "*.obj")) + glob.glob(os.path.join(grass_dir, "*.fbx"))

    if mesh_files:
        for fpath in mesh_files:
            if fpath.endswith(".obj"):
                bpy.ops.wm.obj_import(filepath=fpath)
            elif fpath.endswith(".fbx"):
                bpy.ops.import_scene.fbx(filepath=fpath)

            for o in bpy.context.selected_objects:
                grass_coll.objects.link(o)
                if o.name in bpy.context.scene.collection.objects:
                    bpy.context.scene.collection.objects.unlink(o)

        for obj in list(grass_coll.objects):
            if obj.dimensions.z > 0.5:
                grass_coll.objects.unlink(obj)
                bpy.data.objects.remove(obj, do_unlink=True)
    else:
        for i in range(3):
            m_data = bpy.data.meshes.new(f"Grass_Blade_Mesh_{i}")
            g_obj = bpy.data.objects.new(f"Grass_Blade_Obj_{i}", m_data)
            grass_coll.objects.link(g_obj)

            verts = [(0, 0, 0), (0.05, 0, 0), (0.04, 0, 0.15), (0.01, 0, 0.25), (0, 0, 0.3)]
            faces = [(0, 1, 2), (1, 2, 3), (2, 3, 4)]
            m_data.from_pydata(verts, [], faces)
            m_data.update()

            if g_obj.data.materials:
                g_obj.data.materials[0] = grass_mat
            else:
                g_obj.data.materials.append(grass_mat)

            g_obj.location = (100, 100, 0)

    bpy.ops.mesh.primitive_circle_add(radius=12.5, vertices=64, fill_type='NGON', location=(0, 0, ground_z_offset))
    ground_obj = bpy.context.active_object
    ground_obj.name = "Ground_Mesh_Object"

    vg = ground_obj.vertex_groups.new(name="Grass_Weight_Group")
    all_vert_indices = [v.index for v in ground_obj.data.vertices]
    vg.add(all_vert_indices, 1.0, 'REPLACE')
    ground_obj.data.materials.append(grass_mat)
    ground_obj.data.materials.append(sky_mat)

    gn_group = bpy.data.node_groups.new("Grass_Ground_GN", type="GeometryNodeTree")
    gn_group.interface.new_socket(name="Geometry", in_out="INPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Geometry", in_out="OUTPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Density Mask", in_out="INPUT", socket_type="NodeSocketFloat")

    nodes = gn_group.nodes
    links = gn_group.links
    out_node = nodes.new("NodeGroupOutput")
    in_node = nodes.new("NodeGroupInput")

    distribute_node = nodes.new("GeometryNodeDistributePointsOnFaces")
    distribute_node.distribute_method = 'POISSON'
    distribute_node.inputs["Distance Min"].default_value = 0.02
    distribute_node.inputs["Density Max"].default_value = 300.0
    links.new(in_node.outputs["Geometry"], distribute_node.inputs["Mesh"])

    noise_node = nodes.new("ShaderNodeTexNoise")
    noise_node.inputs["Scale"].default_value = 2.5
    noise_node.inputs["Detail"].default_value = 2.0

    color_ramp = nodes.new("ShaderNodeValToRGB")
    color_ramp.color_ramp.elements[0].position = 0.20
    color_ramp.color_ramp.elements[1].position = 0.80
    links.new(noise_node.outputs["Fac"], color_ramp.inputs["Fac"])

    math_weight = nodes.new("ShaderNodeMath")
    math_weight.operation = "MULTIPLY"
    links.new(color_ramp.outputs["Color"], math_weight.inputs[0])
    links.new(in_node.outputs["Density Mask"], math_weight.inputs[1])

    links.new(math_weight.outputs["Value"], distribute_node.inputs["Density Factor"])

    instance_node = nodes.new("GeometryNodeInstanceOnPoints")
    instance_node.inputs["Pick Instance"].default_value = True
    links.new(distribute_node.outputs["Points"], instance_node.inputs["Points"])

    coll_info = nodes.new("GeometryNodeCollectionInfo")
    coll_info.inputs["Collection"].default_value = grass_coll
    coll_info.inputs["Separate Children"].default_value = True
    coll_info.inputs["Reset Children"].default_value = True
    links.new(coll_info.outputs["Instances"], instance_node.inputs["Instance"])

    rand_scale = nodes.new("FunctionNodeRandomValue")
    rand_scale.data_type = 'FLOAT'
    rand_scale.inputs["Min"].default_value = 0.25
    rand_scale.inputs["Max"].default_value = 0.50

    combine_scale = nodes.new("ShaderNodeCombineXYZ")
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["X"])
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["Y"])
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["Z"])
    links.new(combine_scale.outputs["Vector"], instance_node.inputs["Scale"])

    rand_rot_z = nodes.new("FunctionNodeRandomValue")
    rand_rot_z.data_type = 'FLOAT'
    rand_rot_z.inputs["Min"].default_value = 0.0
    rand_rot_z.inputs["Max"].default_value = math.radians(360.0)

    combine_rot = nodes.new("ShaderNodeCombineXYZ")
    links.new(rand_rot_z.outputs["Value"], combine_rot.inputs["Z"])
    links.new(combine_rot.outputs["Vector"], instance_node.inputs["Rotation"])

    sky_sphere = nodes.new("GeometryNodeMeshIcoSphere")
    sky_sphere.inputs["Radius"].default_value = 12.5
    sky_sphere.inputs["Subdivisions"].default_value = 4

    flip_faces = nodes.new("GeometryNodeFlipFaces")
    links.new(sky_sphere.outputs["Mesh"], flip_faces.inputs["Mesh"])

    transform_sky = nodes.new("GeometryNodeTransform")
    transform_sky.inputs["Translation"].default_value = (0.0, 0.0, ground_z_offset)
    links.new(flip_faces.outputs["Mesh"], transform_sky.inputs["Geometry"])

    set_sky_mat = nodes.new("GeometryNodeSetMaterial")
    set_sky_mat.inputs["Material"].default_value = sky_mat
    links.new(transform_sky.outputs["Geometry"], set_sky_mat.inputs["Geometry"])

    join_geo = nodes.new("GeometryNodeJoinGeometry")
    links.new(in_node.outputs["Geometry"], join_geo.inputs["Geometry"])
    links.new(instance_node.outputs["Instances"], join_geo.inputs["Geometry"])
    links.new(set_sky_mat.outputs["Geometry"], join_geo.inputs["Geometry"])

    links.new(join_geo.outputs["Geometry"], out_node.inputs["Geometry"])

    mod = ground_obj.modifiers.new(name="Grass_Ground_GN", type='NODES')
    mod.node_group = gn_group

    for item in gn_group.interface.items_tree:
        if getattr(item, "in_out", None) == "INPUT" and item.name == "Density Mask":
            mod[item.identifier + "_attribute_name"] = "Grass_Weight_Group"
            mod[item.identifier + "_use_attribute"] = True

    return ground_obj


# ==============================================================================
# SCROLL MATERIAL & GEOMETRY NODES SETUP
# ==============================================================================

def create_image_scroll_material(image_path):
    mat_name = "Scroll_Texture_Mat"
    mat = bpy.data.materials.get(mat_name)
    if mat:
        bpy.data.materials.remove(mat)

    mat = bpy.data.materials.new(name=mat_name)
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    output_node = nodes.new("ShaderNodeOutputMaterial")
    bsdf = nodes.new("ShaderNodeBsdfPrincipled")
    bsdf.inputs["Roughness"].default_value = 0.5

    tex_img = nodes.new("ShaderNodeTexImage")
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        tex_img.image = img
        tex_img.extension = 'REPEAT'

    uv_map_node = nodes.new("ShaderNodeUVMap")
    uv_map_node.uv_map = "UVMap"

    links.new(uv_map_node.outputs["UV"], tex_img.inputs["Vector"])
    links.new(tex_img.outputs["Color"], bsdf.inputs["Base Color"])
    links.new(bsdf.outputs["BSDF"], output_node.inputs["Surface"])

    return mat, tex_img.image if os.path.exists(image_path) else None


def create_pure_geometry_node_scroll(image_path=""):
    gn_group = bpy.data.node_groups.new("Scroll_Pure_GN", type="GeometryNodeTree")

    gn_group.interface.new_socket(name="Geometry", in_out="OUTPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Character Mesh", in_out="INPUT", socket_type="NodeSocketObject")

    sock_r_vg = gn_group.interface.new_socket(name="Right Hand Group", in_out="INPUT", socket_type="NodeSocketString")
    sock_r_vg.default_value = "mixamorig:RightHand"

    sock_l_vg = gn_group.interface.new_socket(name="Left Hand Group", in_out="INPUT", socket_type="NodeSocketString")
    sock_l_vg.default_value = "mixamorig:LeftHand"

    sock_offset = gn_group.interface.new_socket(name="Forward Offset", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_offset.default_value = 0.60

    sock_unfurl = gn_group.interface.new_socket(name="Unfurl", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_unfurl.default_value = 0.0

    sock_width = gn_group.interface.new_socket(name="Scroll Width", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_width.default_value = 0.35

    nodes = gn_group.nodes
    links = gn_group.links

    in_node = nodes.new("NodeGroupInput")
    out_node = nodes.new("NodeGroupOutput")

    char_info = nodes.new("GeometryNodeObjectInfo")
    char_info.transform_space = "RELATIVE"
    links.new(in_node.outputs["Character Mesh"], char_info.inputs["Object"])

    pos_node = nodes.new("GeometryNodeInputPosition")

    named_r = nodes.new("GeometryNodeInputNamedAttribute")
    named_r.data_type = "FLOAT"
    links.new(in_node.outputs["Right Hand Group"], named_r.inputs["Name"])

    cmp_r = nodes.new("FunctionNodeCompare")
    cmp_r.data_type = "FLOAT"
    cmp_r.operation = "GREATER_THAN"
    cmp_r.inputs[1].default_value = 0.001
    links.new(named_r.outputs["Attribute"], cmp_r.inputs[0])

    stat_r = nodes.new("GeometryNodeAttributeStatistic")
    stat_r.data_type = "FLOAT_VECTOR"
    links.new(char_info.outputs["Geometry"], stat_r.inputs["Geometry"])
    links.new(cmp_r.outputs["Result"], stat_r.inputs["Selection"])
    links.new(pos_node.outputs["Position"], stat_r.inputs["Attribute"])

    named_l = nodes.new("GeometryNodeInputNamedAttribute")
    named_l.data_type = "FLOAT"
    links.new(in_node.outputs["Left Hand Group"], named_l.inputs["Name"])

    cmp_l = nodes.new("FunctionNodeCompare")
    cmp_l.data_type = "FLOAT"
    cmp_l.operation = "GREATER_THAN"
    cmp_l.inputs[1].default_value = 0.001
    links.new(named_l.outputs["Attribute"], cmp_l.inputs[0])

    stat_l = nodes.new("GeometryNodeAttributeStatistic")
    stat_l.data_type = "FLOAT_VECTOR"
    links.new(char_info.outputs["Geometry"], stat_l.inputs["Geometry"])
    links.new(cmp_l.outputs["Result"], stat_l.inputs["Selection"])
    links.new(pos_node.outputs["Position"], stat_l.inputs["Attribute"])

    neg_offset = nodes.new("ShaderNodeMath")
    neg_offset.operation = "MULTIPLY"
    neg_offset.inputs[1].default_value = -1.0
    links.new(in_node.outputs["Forward Offset"], neg_offset.inputs[0])

    forward_vec = nodes.new("ShaderNodeCombineXYZ")
    forward_vec.inputs["X"].default_value = 0.0
    links.new(neg_offset.outputs["Value"], forward_vec.inputs["Y"])
    forward_vec.inputs["Z"].default_value = 0.0

    start_pos = nodes.new("ShaderNodeVectorMath")
    start_pos.operation = "ADD"
    links.new(stat_r.outputs["Mean"], start_pos.inputs[0])
    links.new(forward_vec.outputs["Vector"], start_pos.inputs[1])

    target_pos = nodes.new("ShaderNodeVectorMath")
    target_pos.operation = "ADD"
    links.new(stat_l.outputs["Mean"], target_pos.inputs[0])
    links.new(forward_vec.outputs["Vector"], target_pos.inputs[1])

    delta_vec = nodes.new("ShaderNodeVectorMath")
    delta_vec.operation = "SUBTRACT"
    links.new(target_pos.outputs["Vector"], delta_vec.inputs[0])
    links.new(start_pos.outputs["Vector"], delta_vec.inputs[1])

    dist_node = nodes.new("ShaderNodeVectorMath")
    dist_node.operation = "DISTANCE"
    links.new(start_pos.outputs["Vector"], dist_node.inputs[0])

    scale_vec = nodes.new("ShaderNodeVectorMath")
    scale_vec.operation = "SCALE"
    links.new(delta_vec.outputs["Vector"], scale_vec.inputs["Vector"])
    links.new(in_node.outputs["Unfurl"], scale_vec.inputs["Scale"])

    end_pos = nodes.new("ShaderNodeVectorMath")
    end_pos.operation = "ADD"
    links.new(start_pos.outputs["Vector"], end_pos.inputs[0])
    links.new(scale_vec.outputs["Vector"], end_pos.inputs[1])

    links.new(end_pos.outputs["Vector"], dist_node.inputs[1])

    curve_line = nodes.new("GeometryNodeCurvePrimitiveLine")
    links.new(start_pos.outputs["Vector"], curve_line.inputs["Start"])
    links.new(end_pos.outputs["Vector"], curve_line.inputs["End"])

    set_tilt = nodes.new("GeometryNodeSetCurveTilt")
    set_tilt.inputs["Tilt"].default_value = 1.5708
    links.new(curve_line.outputs["Curve"], set_tilt.inputs["Curve"])

    spline_param_main = nodes.new("GeometryNodeSplineParameter")
    store_u = nodes.new("GeometryNodeStoreNamedAttribute")
    store_u.inputs["Name"].default_value = "U_Factor"
    store_u.data_type = 'FLOAT'
    store_u.domain = 'POINT'
    links.new(set_tilt.outputs["Curve"], store_u.inputs["Geometry"])
    links.new(spline_param_main.outputs["Factor"], store_u.inputs["Value"])

    half_width_neg = nodes.new("ShaderNodeMath")
    half_width_neg.operation = "MULTIPLY"
    half_width_neg.inputs[1].default_value = -0.5
    links.new(in_node.outputs["Scroll Width"], half_width_neg.inputs[0])

    half_width_pos = nodes.new("ShaderNodeMath")
    half_width_pos.operation = "MULTIPLY"
    half_width_pos.inputs[1].default_value = 0.5
    links.new(in_node.outputs["Scroll Width"], half_width_pos.inputs[0])

    combine_start = nodes.new("ShaderNodeCombineXYZ")
    links.new(half_width_neg.outputs["Value"], combine_start.inputs["X"])

    combine_end = nodes.new("ShaderNodeCombineXYZ")
    links.new(half_width_pos.outputs["Value"], combine_end.inputs["X"])

    profile_line = nodes.new("GeometryNodeCurvePrimitiveLine")
    links.new(combine_start.outputs["Vector"], profile_line.inputs["Start"])
    links.new(combine_end.outputs["Vector"], profile_line.inputs["End"])

    spline_param_prof = nodes.new("GeometryNodeSplineParameter")
    store_v = nodes.new("GeometryNodeStoreNamedAttribute")
    store_v.inputs["Name"].default_value = "V_Factor"
    store_v.data_type = 'FLOAT'
    store_v.domain = 'POINT'
    links.new(profile_line.outputs["Curve"], store_v.inputs["Geometry"])
    links.new(spline_param_prof.outputs["Factor"], store_v.inputs["Value"])

    curve_to_mesh = nodes.new("GeometryNodeCurveToMesh")
    links.new(store_u.outputs["Geometry"], curve_to_mesh.inputs["Curve"])
    links.new(store_v.outputs["Geometry"], curve_to_mesh.inputs["Profile Curve"])

    img_aspect = 1.0
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        if img.size[1] > 0:
            img_aspect = img.size[0] / img.size[1]

    h_times_aspect = nodes.new("ShaderNodeMath")
    h_times_aspect.operation = "MULTIPLY"
    h_times_aspect.inputs[1].default_value = img_aspect
    links.new(in_node.outputs["Scroll Width"], h_times_aspect.inputs[0])

    tile_count = nodes.new("ShaderNodeMath")
    tile_count.operation = "DIVIDE"
    links.new(dist_node.outputs["Value"], tile_count.inputs[0])
    links.new(h_times_aspect.outputs["Value"], tile_count.inputs[1])

    read_u = nodes.new("GeometryNodeInputNamedAttribute")
    read_u.inputs["Name"].default_value = "U_Factor"
    read_u.data_type = 'FLOAT'

    scaled_u = nodes.new("ShaderNodeMath")
    scaled_u.operation = "MULTIPLY"
    links.new(read_u.outputs["Attribute"], scaled_u.inputs[0])
    links.new(tile_count.outputs["Value"], scaled_u.inputs[1])

    read_v = nodes.new("GeometryNodeInputNamedAttribute")
    read_v.inputs["Name"].default_value = "V_Factor"
    read_v.data_type = 'FLOAT'

    flip_v = nodes.new("ShaderNodeMath")
    flip_v.operation = "SUBTRACT"
    flip_v.inputs[0].default_value = 1.0
    links.new(read_v.outputs["Attribute"], flip_v.inputs[1])

    combine_uv = nodes.new("ShaderNodeCombineXYZ")
    links.new(scaled_u.outputs["Value"], combine_uv.inputs["X"])
    links.new(flip_v.outputs["Value"], combine_uv.inputs["Y"])

    store_uv = nodes.new("GeometryNodeStoreNamedAttribute")
    store_uv.inputs["Name"].default_value = "UVMap"
    store_uv.data_type = 'FLOAT2'
    store_uv.domain = 'CORNER'
    links.new(curve_to_mesh.outputs["Mesh"], store_uv.inputs["Geometry"])
    links.new(combine_uv.outputs["Vector"], store_uv.inputs["Value"])

    set_material = nodes.new("GeometryNodeSetMaterial")
    scroll_mat, _ = create_image_scroll_material(image_path)
    set_material.inputs["Material"].default_value = scroll_mat
    links.new(store_uv.outputs["Geometry"], set_material.inputs["Geometry"])

    links.new(set_material.outputs["Geometry"], out_node.inputs["Geometry"])

    return gn_group, scroll_mat


def setup_geometry_node_scroll(target_mesh, unfurl_start, unfurl_end, image_path=""):
    mesh_data = bpy.data.meshes.new("Scroll_Procedural_Mesh")
    scroll_obj = bpy.data.objects.new("Scroll_GN_Object", mesh_data)
    bpy.context.scene.collection.objects.link(scroll_obj)

    mod = scroll_obj.modifiers.new("Scroll_GN", type="NODES")
    gn_tree, scroll_mat = create_pure_geometry_node_scroll(image_path=image_path)
    mod.node_group = gn_tree

    scroll_obj.data.materials.append(scroll_mat)

    r_hand_name, l_hand_name = find_hand_vertex_groups(target_mesh)

    unfurl_id = None
    for item in gn_tree.interface.items_tree:
        if getattr(item, "in_out", None) == "INPUT":
            if item.name == "Character Mesh":
                mod[item.identifier] = target_mesh
            elif item.name == "Right Hand Group":
                mod[item.identifier] = r_hand_name
            elif item.name == "Left Hand Group":
                mod[item.identifier] = l_hand_name
            elif item.name == "Forward Offset":
                mod[item.identifier] = 0.60
            elif item.name == "Scroll Width":
                mod[item.identifier] = 0.35
            elif item.name == "Unfurl":
                unfurl_id = item.identifier

    for m in target_mesh.modifiers:
        if m.type == 'ARMATURE':
            m.show_viewport = True
            m.show_render = True
            m.show_in_editmode = True

    scroll_obj.hide_viewport = True
    scroll_obj.hide_render = True
    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=1)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=1)

    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=unfurl_start - 1)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=unfurl_start - 1)

    scroll_obj.hide_viewport = False
    scroll_obj.hide_render = False
    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=unfurl_start)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=unfurl_start)

    for fc in get_object_fcurves(scroll_obj):
        if fc.data_path in ["hide_viewport", "hide_render"]:
            for kp in fc.keyframe_points:
                kp.interpolation = 'CONSTANT'

    if unfurl_id:
        mod[unfurl_id] = 0.001
        mod.keyframe_insert(data_path=f'["{unfurl_id}"]', frame=unfurl_start)

        mod[unfurl_id] = 1.0
        mod.keyframe_insert(data_path=f'["{unfurl_id}"]', frame=unfurl_end)

    return scroll_obj


# ==============================================================================
# MAIN CONTROLLER
# ==============================================================================

def import_and_extract_action(fbx_path, action_name):
    existing_objects = set(bpy.data.objects)
    bpy.ops.import_scene.fbx(filepath=fbx_path)
    bpy.context.view_layer.update()

    imported_objects = set(bpy.data.objects) - existing_objects
    extracted_action = None
    for obj in imported_objects:
        if obj.animation_data and obj.animation_data.action:
            extracted_action = obj.animation_data.action
            extracted_action.name = action_name
            extracted_action.use_fake_user = True
            break
    return extracted_action, imported_objects


def purge_imported_objects(objects_to_remove):
    for obj in objects_to_remove:
        if obj.name in bpy.data.objects:
            if obj.animation_data:
                obj.animation_data_clear()
            for coll in list(obj.users_collection):
                coll.objects.unlink(obj)
    bpy.context.view_layer.update()


def main():
    cleanup_previous_run()

    walk_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Mutant Walking.fbx"
    turn_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Right Turn 90.fbx"
    unfurl_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Mutant Idle.fbx"

    sky_hdri_path = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/horn-koppe_spring_4k.exr"
    output_blend = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/output_scene.blend"

    image_path = "/Users/dobbinp1/Downloads/IMGP6691_psedit.jpg"
    grass_dir = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Grass004_4K-JPG"

    ground_obj = setup_grass_ground_geometry_nodes(grass_dir, sky_hdri_path, ground_z_offset=-0.2)

    CYCLES = 4
    CROSSFADE = 15
    TARGET_DISTANCE = 4.0

    walk_action, walk_objs = import_and_extract_action(walk_fbx, "walk")
    target_rig = next((obj for obj in walk_objs if obj.type == "ARMATURE"), None)
    target_mesh = next((obj for obj in walk_objs if obj.type == "MESH"), None)

    if not target_rig or not target_mesh:
        raise RuntimeError("Armature or Character Mesh not found in walk FBX.")

    travel_axis_idx, motion_sign = detect_natural_forward_axis(walk_action, target_rig)
    initial_heading = math.radians(90.0) if travel_axis_idx == 0 else 0.0
    target_rig.rotation_euler[2] = initial_heading

    action_len = int(walk_action.frame_range[1] - walk_action.frame_range[0])
    if action_len <= 0:
        action_len = 30

    anim_data = target_rig.animation_data_create()

    walk_track = anim_data.nla_tracks.new()
    walk_track.name = "Walk_Track"

    current_frame = 1
    for i in range(CYCLES):
        strip = walk_track.strips.new(name=f"Walk_{i+1}", start=int(current_frame), action=walk_action)
        strip.blend_type = "REPLACE"
        current_frame += action_len

    walk_end_frame = int(current_frame)

    turn_action, turn_objs = import_and_extract_action(turn_fbx, "rightturn")
    purge_imported_objects(turn_objs)

    turn_track = anim_data.nla_tracks.new()
    turn_track.name = "RightTurn_Track"

    turn_start = int(walk_end_frame - CROSSFADE)
    turn_strip = turn_track.strips.new(name="RightTurn", start=turn_start, action=turn_action)
    turn_strip.blend_type = "REPLACE"
    turn_strip.blend_in = CROSSFADE
    turn_strip.use_auto_blend = True
    turn_strip.extrapolation = "HOLD"
    turn_end_frame = int(turn_strip.frame_end)

    # Pass turn_start frame to scene data so handler stops sound when character stops walking
    bpy.context.scene["turn_start_frame"] = turn_start

    unfurl_action, unfurl_objs = import_and_extract_action(unfurl_fbx, "unfurl")
    purge_imported_objects(unfurl_objs)

    unfurl_track = anim_data.nla_tracks.new()
    unfurl_track.name = "Unfurl_Track"

    unfurl_start = int(turn_end_frame - CROSSFADE)
    unfurl_strip = unfurl_track.strips.new(name="Unfurl", start=unfurl_start, action=unfurl_action)
    unfurl_strip.blend_type = "REPLACE"
    unfurl_strip.blend_in = CROSSFADE
    unfurl_strip.use_auto_blend = True
    unfurl_strip.extrapolation = "HOLD"
    unfurl_end_frame = int(unfurl_strip.frame_end)

    root_action = bpy.data.actions.new(name="World_Root_Motion")
    target_rig.animation_data.action = root_action

    loc_fcurve = root_action.fcurve_ensure_for_datablock(datablock=target_rig, data_path="location", index=travel_axis_idx)
    final_target_value = TARGET_DISTANCE * motion_sign

    k_loc1 = loc_fcurve.keyframe_points.insert(frame=1.0, value=0.0)
    k_loc2 = loc_fcurve.keyframe_points.insert(frame=float(turn_start), value=final_target_value)
    k_loc3 = loc_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=final_target_value)

    k_loc1.interpolation = "LINEAR"
    k_loc2.interpolation = "CONSTANT"
    k_loc3.interpolation = "CONSTANT"

    rot_fcurve = root_action.fcurve_ensure_for_datablock(datablock=target_rig, data_path="rotation_euler", index=2)
    target_heading = initial_heading - math.radians(90.0)

    k_rot1 = rot_fcurve.keyframe_points.insert(frame=1.0, value=initial_heading)
    k_rot2 = rot_fcurve.keyframe_points.insert(frame=float(turn_start), value=initial_heading)
    k_rot3 = rot_fcurve.keyframe_points.insert(frame=float(turn_end_frame), value=target_heading)
    k_rot4 = rot_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=target_heading)

    k_rot1.interpolation = "CONSTANT"
    k_rot2.interpolation = "BEZIER"
    k_rot3.interpolation = "CONSTANT"
    k_rot4.interpolation = "CONSTANT"

    root_track = anim_data.nla_tracks.new()
    root_track.name = "World_Root_Track"
    root_strip = root_track.strips.new(name="Root_Motion_Strip", start=1, action=root_action)
    root_strip.blend_type = "REPLACE"
    root_strip.extrapolation = "HOLD"

    anim_data.action = None

    setup_geometry_node_scroll(
        target_mesh=target_mesh,
        unfurl_start=unfurl_start,
        unfurl_end=unfurl_end_frame,
        image_path=image_path,
    )

    cam_obj, _ = setup_modern_reel_camera(
        target_rig=target_rig,
        target_mesh=target_mesh,
        turn_start=turn_start,
        turn_end_frame=turn_end_frame,
        unfurl_end_frame=unfurl_end_frame
    )

    bpy.context.scene.frame_start = 1
    bpy.context.scene.frame_end = unfurl_end_frame

    ensure_world_sphere_size(cam_obj, ground_obj, safety_multiplier=2.5)

    # Register real-time footstep audio handler
    register_audio_handler()

    bpy.ops.wm.save_as_mainfile(filepath=output_blend)
    print(f"\n[SUCCESS] Integrated real-time audio handler & saved scene to: {output_blend}")


if __name__ == "__main__":
    main()
```

```
import math
import mathutils
import os
import sys
import glob
import bpy
import aud
from bpy_extras import anim_utils
from bpy_extras.object_utils import world_to_camera_view
from bpy.app.handlers import persistent

import os
import sys
import aud
import bpy
from bpy.app.handlers import persistent

# ==============================================================================
# CONFIGURATION
# ==============================================================================
# Update with your actual .wav file path
FOOTSTEP_SOUND_PATH = "/Users/dobbinp1/Downloads/stomp.wav"
GROUND_THRESHOLD = 0.24  # Z-height threshold for foot contact


# ==============================================================================
# PERSISTENT AUDIO DEVICE SETUP
# ==============================================================================
if "_audio_store" not in sys.modules:
    import types
    sys.modules["_audio_store"] = types.ModuleType("_audio_store")

audio_store = sys.modules["_audio_store"]

if not hasattr(audio_store, "device"):
    audio_store.device = aud.Device()

AUDIO_DEVICE = audio_store.device
SOUND_FACTORY = None

if os.path.exists(FOOTSTEP_SOUND_PATH):
    SOUND_FACTORY = aud.Sound(FOOTSTEP_SOUND_PATH).cache()
    print(f"[AUDIO OK] Sound loaded from: {FOOTSTEP_SOUND_PATH}")
else:
    print(f"[AUDIO WARNING] File not found: {FOOTSTEP_SOUND_PATH}")


# State tracking across frames
_left_foot_was_down = False
_right_foot_was_down = False
_last_frame = -1


# ==============================================================================
# HELPER FUNCTIONS & HANDLER
# ==============================================================================
def play_footstep():
    if SOUND_FACTORY:
        print("  >>> PLAYING FOOTSTEP SOUND")
        AUDIO_DEVICE.play(SOUND_FACTORY)


def find_foot_bone(pose_bones, side="Left"):
    target_keywords = [
        f"mixamorig:{side}foot",
        f"{side.lower()}foot",
        f"foot.{side[0].lower()}",
        f"foot_{side[0].lower()}"
    ]
    for bone in pose_bones:
        b_name = bone.name.lower()
        if any(kw in b_name for kw in target_keywords):
            return bone
    return None


@persistent
def footstep_frame_handler(scene):
    global _left_foot_was_down, _right_foot_was_down, _last_frame

    try:
        current_frame = scene.frame_current

        # Reset states on scrub or frame jump back
        if current_frame <= _last_frame or abs(current_frame - _last_frame) > 10:
            _left_foot_was_down = False
            _right_foot_was_down = False
        _last_frame = current_frame

        # Find Armature
        rig = next((o for o in scene.objects if o.type == 'ARMATURE'), None)
        if not rig:
            return

        depsgraph = bpy.context.evaluated_depsgraph_get()
        eval_rig = rig.evaluated_get(depsgraph)

        left_bone = find_foot_bone(eval_rig.pose.bones, "Left")
        right_bone = find_foot_bone(eval_rig.pose.bones, "Right")

        if not left_bone or not right_bone:
            return

        l_z = (eval_rig.matrix_world @ left_bone.matrix).translation.z
        r_z = (eval_rig.matrix_world @ right_bone.matrix).translation.z

        l_is_down = l_z <= GROUND_THRESHOLD
        r_is_down = r_z <= GROUND_THRESHOLD

        # State transition triggers
        if l_is_down and not _left_foot_was_down:
            print(f"*** STEP LEFT TRIGGERED (Frame {current_frame} | Z: {l_z:.4f}) ***")
            play_footstep()

        if r_is_down and not _right_foot_was_down:
            print(f"*** STEP RIGHT TRIGGERED (Frame {current_frame} | Z: {r_z:.4f}) ***")
            play_footstep()

        _left_foot_was_down = l_is_down
        _right_foot_was_down = r_is_down

    except Exception as e:
        import traceback
        print("\n[HANDLER ERROR]")
        traceback.print_exc()



def unregister_audio_handler():
    """Safely removes the footstep handler from Blender's frame_change_post handlers."""
    if hasattr(bpy.app, "handlers"):
        for h in list(bpy.app.handlers.frame_change_post):
            if getattr(h, "__name__", "") == "footstep_frame_handler":
                bpy.app.handlers.frame_change_post.remove(h)
        for h in list(bpy.app.handlers.frame_change_pre):
            if getattr(h, "__name__", "") == "footstep_frame_handler":
                bpy.app.handlers.frame_change_pre.remove(h)

# ==============================================================================
# REGISTER HANDLER
# ==============================================================================
def register_audio_handler():
    unregister_audio_handler()
    bpy.app.handlers.frame_change_post.append(footstep_frame_handler)
    print("[SUCCESS] Footstep handler registered.")



register()

# Optional: Un-comment the line below to immediately play viewport timeline upon script execution:
# bpy.ops.screen.animation_play()


# ==============================================================================
# CLEANUP & HELPERS
# ==============================================================================

def cleanup_previous_run():
    """Clears old scroll objects, ground, node groups, camera anchors, collections, and orphaned data."""
    unregister_audio_handler()

    target_names = [
        "Master_NLA_Control",
        "Scroll_GN_Object",
        "Scroll_Start_Pin",
        "Scroll_End_Pin",
        "Scroll_Master_Container",
        "HDRI_Ground_Plane",
        "Ground_Mesh_Object",
        "Camera_Focus_Target",
        "Vertical_Reel_Camera"
    ]
    for obj in list(bpy.data.objects):
        if obj.name in target_names or (obj.name.startswith("mixamorig") and obj.type != "ARMATURE" and obj.type != "MESH"):
            if obj.animation_data:
                obj.animation_data_clear()
            bpy.data.objects.remove(obj, do_unlink=True)

    for action in list(bpy.data.actions):
        if action.users == 0 or action.use_fake_user:
            action.use_fake_user = False
            bpy.data.actions.remove(action)

    for ng in list(bpy.data.node_groups):
        if ng.name in ["Scroll_Unfurl_GN", "Scroll_Pure_GN", "HDRI_Ground_GN", "Grass_Ground_GN"]:
            bpy.data.node_groups.remove(ng)

    for coll_name in ["Scroll_Master_Container", "Grass_Artifacts_Collection"]:
        if coll_name in bpy.data.collections:
            coll = bpy.data.collections[coll_name]
            for obj in list(coll.objects):
                bpy.data.objects.remove(obj, do_unlink=True)
            bpy.data.collections.remove(coll)

    bpy.context.view_layer.update()


def get_slot_fcurves(action, anim_target_object):
    fcurves = []
    if not action:
        return fcurves

    anim_data = getattr(anim_target_object, "animation_data", None)
    if anim_data and getattr(anim_data, "action_slot", None):
        slot = anim_data.action_slot
        channelbag = anim_utils.action_get_channelbag_for_slot(action, slot)
        if channelbag and hasattr(channelbag, "fcurves"):
            return list(channelbag.fcurves)

    if hasattr(action, "slots"):
        for slot in action.slots:
            for cb in getattr(slot, "channelbags", []):
                fcurves.extend(getattr(cb, "fcurves", []))
        if fcurves:
            return fcurves

    if hasattr(action, "fcurves"):
        return list(action.fcurves)

    return fcurves


def get_object_fcurves(obj):
    if not obj or not obj.animation_data or not obj.animation_data.action:
        return []
    action = obj.animation_data.action
    return get_slot_fcurves(action, obj)


def detect_natural_forward_axis(action, rig_object):
    fcurves = get_slot_fcurves(action, rig_object)
    max_delta = 0.0
    best_axis = 1
    net_delta = 0.0

    start_frame = action.frame_range[0]
    end_frame = action.frame_range[1]

    for fc in fcurves:
        if fc.data_path.endswith("location") and fc.array_index in [0, 1, 2]:
            start_val = fc.evaluate(start_frame)
            end_val = fc.evaluate(end_frame)
            delta = end_val - start_val
            if abs(delta) > abs(max_delta):
                max_delta = delta
                best_axis = fc.array_index
                net_delta = delta

    direction_sign = 1.0 if net_delta >= 0 else -1.0
    return best_axis, direction_sign


def find_hand_vertex_groups(target_mesh):
    r_name = "mixamorig:RightHand"
    l_name = "mixamorig:LeftHand"

    if target_mesh and hasattr(target_mesh, "vertex_groups"):
        for vg in target_mesh.vertex_groups:
            vg_lower = vg.name.lower()
            if "righthand" in vg_lower or "hand_r" in vg_lower or "hand.r" in vg_lower:
                r_name = vg.name
            elif "lefthand" in vg_lower or "hand_l" in vg_lower or "hand.l" in vg_lower:
                l_name = vg.name

    return r_name, l_name


# ==============================================================================
# UNPARENTED MATRIX-EVALUATED VERTICAL CAMERA SETUP
# ==============================================================================

def setup_modern_reel_camera(target_rig, target_mesh, turn_start, turn_end_frame, unfurl_end_frame):
    """
    Creates a 9:16 Instagram Reel Camera:
    - Uses exact 2D camera viewport projection (world_to_camera_view) to test
      every vertex across the animation, ensuring 100% of the mesh fits inside
      the camera frame during both walking and unfurl phases.
    """
    scene = bpy.context.scene
    scene.render.resolution_x = 1080
    scene.render.resolution_y = 1920
    scene.render.resolution_percentage = 100

    target_height = target_mesh.dimensions.z if target_mesh.dimensions.z > 0 else 1.8
    midpoint_z_offset = target_height * 0.50

    focus_target = bpy.data.objects.get("Camera_Focus_Target")
    if not focus_target:
        focus_target = bpy.data.objects.new("Camera_Focus_Target", None)
        focus_target.empty_display_type = 'PLAIN_AXES'
        focus_target.empty_display_size = 0.25
        scene.collection.objects.link(focus_target)

    focus_target.parent = None

    copy_loc = focus_target.constraints.get("Copy Location")
    if not copy_loc:
        copy_loc = focus_target.constraints.new(type='COPY_LOCATION')
    copy_loc.target = target_mesh
    copy_loc.use_offset = True
    focus_target.location.z = midpoint_z_offset

    cam_data = bpy.data.cameras.get("Vertical_Reel_Cam_Data")
    if not cam_data:
        cam_data = bpy.data.cameras.new("Vertical_Reel_Cam_Data")
    cam_data.lens = 45.0
    cam_data.sensor_fit = 'VERTICAL'
    cam_data.clip_end = 100.0

    cam_obj = bpy.data.objects.get("Vertical_Reel_Camera")
    if not cam_obj:
        cam_obj = bpy.data.objects.new("Vertical_Reel_Camera", cam_data)
        scene.collection.objects.link(cam_obj)

    scene.camera = cam_obj
    cam_obj.parent = None

    track = cam_obj.constraints.get("Track To")
    if not track:
        track = cam_obj.constraints.new(type='TRACK_TO')
    track.target = focus_target
    track.track_axis = 'TRACK_NEGATIVE_Z'
    track.up_axis = 'UP_Y'

    def check_mesh_in_frame(frame_num, cam_location, margin=0.08):
        cam_obj.location = cam_location
        scene.frame_set(int(frame_num))
        bpy.context.view_layer.update()

        depsgraph = bpy.context.evaluated_depsgraph_get()
        eval_mesh_obj = target_mesh.evaluated_get(depsgraph)
        mesh_data = eval_mesh_obj.data

        vert_step = max(1, len(mesh_data.vertices) // 300)
        world_matrix = target_mesh.matrix_world

        min_x, max_x = 1.0, 0.0
        min_y, max_y = 1.0, 0.0

        for i in range(0, len(mesh_data.vertices), vert_step):
            v_world = world_matrix @ mesh_data.vertices[i].co
            co_2d = world_to_camera_view(scene, cam_obj, v_world)

            if co_2d.x < min_x: min_x = co_2d.x
            if co_2d.x > max_x: max_x = co_2d.x
            if co_2d.y < min_y: min_y = co_2d.y
            if co_2d.y > max_y: max_y = co_2d.y

        in_bounds = (min_x >= margin and max_x <= (1.0 - margin) and
                     min_y >= margin and max_y <= (1.0 - margin))

        return in_bounds, min_x, max_x, min_y, max_y

    def find_fit_distance(frame_num, angle_rad, base_height_offset, initial_dist=6.0, margin=0.08):
        scene.frame_set(int(frame_num))
        bpy.context.view_layer.update()

        center_pos = focus_target.matrix_world.translation.copy()
        heading = target_rig.matrix_world.to_euler('XYZ').z
        view_angle = heading + angle_rad

        dist = initial_dist
        for _ in range(30):
            test_x = center_pos.x + dist * math.sin(view_angle)
            test_y = center_pos.y - dist * math.cos(view_angle)
            test_z = center_pos.z + base_height_offset

            fit, min_x, max_x, min_y, max_y = check_mesh_in_frame(frame_num, (test_x, test_y, test_z), margin=margin)
            if fit:
                break
            dist += 0.4

        return dist, center_pos, view_angle

    max_walk_dist = 6.0
    walk_frames_to_sample = range(1, turn_start + 1, 6)

    for f in walk_frames_to_sample:
        req_d, _, _ = find_fit_distance(f, angle_rad=math.radians(-45.0), base_height_offset=0.0, initial_dist=max_walk_dist, margin=0.10)
        if req_d > max_walk_dist:
            max_walk_dist = req_d

    scene.frame_set(1)
    bpy.context.view_layer.update()
    c1 = focus_target.matrix_world.translation.copy()
    h1 = target_rig.matrix_world.to_euler('XYZ').z + math.radians(-45.0)
    w_x1 = c1.x + max_walk_dist * math.sin(h1)
    w_y1 = c1.y - max_walk_dist * math.cos(h1)
    w_z1 = c1.z

    scene.frame_set(turn_start)
    bpy.context.view_layer.update()
    cts = focus_target.matrix_world.translation.copy()
    hts = target_rig.matrix_world.to_euler('XYZ').z + math.radians(-45.0)
    w_xts = cts.x + max_walk_dist * math.sin(hts)
    w_yts = cts.y - max_walk_dist * math.cos(hts)
    w_zts = cts.z

    unfurl_frames_to_sample = range(turn_end_frame, unfurl_end_frame + 1, 6)
    max_unfurl_dist = max_walk_dist

    for f in unfurl_frames_to_sample:
        req_d, _, _ = find_fit_distance(f, angle_rad=math.radians(0.0), base_height_offset=0.0, initial_dist=max_unfurl_dist, margin=0.10)
        if req_d > max_unfurl_dist:
            max_unfurl_dist = req_d

    scene.frame_set(turn_end_frame)
    bpy.context.view_layer.update()
    cte = focus_target.matrix_world.translation.copy()
    hte = target_rig.matrix_world.to_euler('XYZ').z + math.radians(0.0)
    f_x = cte.x + max_unfurl_dist * math.sin(hte)
    f_y = cte.y - max_unfurl_dist * math.cos(hte)
    f_z = cte.z

    if cam_obj.animation_data:
        cam_obj.animation_data_clear()

    cam_obj.location = (w_x1, w_y1, w_z1)
    cam_obj.keyframe_insert(data_path="location", frame=1)

    cam_obj.location = (w_xts, w_yts, w_zts)
    cam_obj.keyframe_insert(data_path="location", frame=turn_start)

    cam_obj.location = (f_x, f_y, f_z)
    cam_obj.keyframe_insert(data_path="location", frame=turn_end_frame)

    cam_obj.location = (f_x, f_y, f_z)
    cam_obj.keyframe_insert(data_path="location", frame=unfurl_end_frame)

    for fc in get_object_fcurves(cam_obj):
        for kp in fc.keyframe_points:
            kp.interpolation = 'BEZIER'

    scene.frame_set(1)
    return cam_obj, focus_target


# ==============================================================================
# GEOMETRY NODES GROUND, GRASS & SKY DOME SETUP
# ==============================================================================

def ensure_world_sphere_size(cam_obj, ground_obj, safety_multiplier=2.5):
    if not cam_obj or not ground_obj:
        return

    scene = bpy.context.scene
    max_cam_dist = 0.0

    for frame in range(scene.frame_start, scene.frame_end + 1, 5):
        scene.frame_set(frame)
        bpy.context.view_layer.update()

        cam_dist = cam_obj.matrix_world.translation.length
        if cam_dist > max_cam_dist:
            max_cam_dist = cam_dist

    world_radius = max(25.0, max_cam_dist * safety_multiplier)

    mod = ground_obj.modifiers.get("Grass_Ground_GN")
    if mod and mod.node_group:
        for node in mod.node_group.nodes:
            if node.bl_idname == "GeometryNodeMeshIcoSphere":
                node.inputs["Radius"].default_value = world_radius

    if ground_obj.type == 'MESH':
        bpy.context.view_layer.objects.active = ground_obj
        bpy.ops.object.mode_set(mode='OBJECT')

        ground_scale = world_radius / 12.5
        ground_obj.scale = (ground_scale, ground_scale, 1.0)
        bpy.ops.object.transform_apply(location=False, rotation=False, scale=True)

    cam_obj.data.clip_end = max(cam_obj.data.clip_end, world_radius * 2.5)

    scene.frame_set(1)
    bpy.context.view_layer.update()


def create_grass_material(grass_dir):
    mat = bpy.data.materials.new(name="Grass_PBR_Material")
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    out_node = nodes.new("ShaderNodeOutputMaterial")
    bsdf = nodes.new("ShaderNodeBsdfPrincipled")
    links.new(bsdf.outputs["BSDF"], out_node.inputs["Surface"])

    col_files = glob.glob(os.path.join(grass_dir, "*Color*.jpg")) + glob.glob(os.path.join(grass_dir, "*JPG*.jpg"))
    if col_files:
        tex_col = nodes.new("ShaderNodeTexImage")
        tex_col.image = bpy.data.images.load(col_files[0])
        links.new(tex_col.outputs["Color"], bsdf.inputs["Base Color"])

    rough_files = glob.glob(os.path.join(grass_dir, "*Roughness*.jpg"))
    if rough_files:
        tex_rough = nodes.new("ShaderNodeTexImage")
        tex_rough.image = bpy.data.images.load(rough_files[0])
        tex_rough.image.colorspace_settings.name = 'Non-Color'
        links.new(tex_rough.outputs["Color"], bsdf.inputs["Roughness"])

    norm_files = glob.glob(os.path.join(grass_dir, "*Normal*.jpg"))
    if norm_files:
        tex_norm = nodes.new("ShaderNodeTexImage")
        tex_norm.image = bpy.data.images.load(norm_files[0])
        tex_norm.image.colorspace_settings.name = 'Non-Color'

        normal_map_node = nodes.new("ShaderNodeNormalMap")
        links.new(tex_norm.outputs["Color"], normal_map_node.inputs["Color"])
        links.new(normal_map_node.outputs["Normal"], bsdf.inputs["Normal"])

    return mat


def create_sky_material(sky_hdri_path):
    mat = bpy.data.materials.new(name="Sky_Dome_Material")
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    out_node = nodes.new("ShaderNodeOutputMaterial")
    emission = nodes.new("ShaderNodeEmission")
    emission.inputs["Strength"].default_value = 1.0

    links.new(emission.outputs["Emission"], out_node.inputs["Surface"])

    if os.path.exists(sky_hdri_path):
        tex_env = nodes.new("ShaderNodeTexEnvironment")
        tex_env.image = bpy.data.images.load(sky_hdri_path)
        links.new(tex_env.outputs["Color"], emission.inputs["Color"])

    return mat


def setup_grass_ground_geometry_nodes(grass_dir, sky_hdri_path, ground_z_offset=-0.2):
    grass_coll = bpy.data.collections.get("Grass_Artifacts_Collection")
    if not grass_coll:
        grass_coll = bpy.data.collections.new("Grass_Artifacts_Collection")
        bpy.context.scene.collection.children.link(grass_coll)

    grass_mat = create_grass_material(grass_dir)
    sky_mat = create_sky_material(sky_hdri_path)

    mesh_files = glob.glob(os.path.join(grass_dir, "*.obj")) + glob.glob(os.path.join(grass_dir, "*.fbx"))

    if mesh_files:
        for fpath in mesh_files:
            if fpath.endswith(".obj"):
                bpy.ops.wm.obj_import(filepath=fpath)
            elif fpath.endswith(".fbx"):
                bpy.ops.import_scene.fbx(filepath=fpath)

            for o in bpy.context.selected_objects:
                grass_coll.objects.link(o)
                if o.name in bpy.context.scene.collection.objects:
                    bpy.context.scene.collection.objects.unlink(o)

        for obj in list(grass_coll.objects):
            if obj.dimensions.z > 0.5:
                grass_coll.objects.unlink(obj)
                bpy.data.objects.remove(obj, do_unlink=True)
    else:
        for i in range(3):
            m_data = bpy.data.meshes.new(f"Grass_Blade_Mesh_{i}")
            g_obj = bpy.data.objects.new(f"Grass_Blade_Obj_{i}", m_data)
            grass_coll.objects.link(g_obj)

            verts = [(0, 0, 0), (0.05, 0, 0), (0.04, 0, 0.15), (0.01, 0, 0.25), (0, 0, 0.3)]
            faces = [(0, 1, 2), (1, 2, 3), (2, 3, 4)]
            m_data.from_pydata(verts, [], faces)
            m_data.update()

            if g_obj.data.materials:
                g_obj.data.materials[0] = grass_mat
            else:
                g_obj.data.materials.append(grass_mat)

            g_obj.location = (100, 100, 0)

    bpy.ops.mesh.primitive_circle_add(radius=12.5, vertices=64, fill_type='NGON', location=(0, 0, ground_z_offset))
    ground_obj = bpy.context.active_object
    ground_obj.name = "Ground_Mesh_Object"

    vg = ground_obj.vertex_groups.new(name="Grass_Weight_Group")
    all_vert_indices = [v.index for v in ground_obj.data.vertices]
    vg.add(all_vert_indices, 1.0, 'REPLACE')
    ground_obj.data.materials.append(grass_mat)
    ground_obj.data.materials.append(sky_mat)

    gn_group = bpy.data.node_groups.new("Grass_Ground_GN", type="GeometryNodeTree")
    gn_group.interface.new_socket(name="Geometry", in_out="INPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Geometry", in_out="OUTPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Density Mask", in_out="INPUT", socket_type="NodeSocketFloat")

    nodes = gn_group.nodes
    links = gn_group.links
    out_node = nodes.new("NodeGroupOutput")
    in_node = nodes.new("NodeGroupInput")

    distribute_node = nodes.new("GeometryNodeDistributePointsOnFaces")
    distribute_node.distribute_method = 'POISSON'
    distribute_node.inputs["Distance Min"].default_value = 0.02
    distribute_node.inputs["Density Max"].default_value = 300.0
    links.new(in_node.outputs["Geometry"], distribute_node.inputs["Mesh"])

    noise_node = nodes.new("ShaderNodeTexNoise")
    noise_node.inputs["Scale"].default_value = 2.5
    noise_node.inputs["Detail"].default_value = 2.0

    color_ramp = nodes.new("ShaderNodeValToRGB")
    color_ramp.color_ramp.elements[0].position = 0.20
    color_ramp.color_ramp.elements[1].position = 0.80
    links.new(noise_node.outputs["Fac"], color_ramp.inputs["Fac"])

    math_weight = nodes.new("ShaderNodeMath")
    math_weight.operation = "MULTIPLY"
    links.new(color_ramp.outputs["Color"], math_weight.inputs[0])
    links.new(in_node.outputs["Density Mask"], math_weight.inputs[1])

    links.new(math_weight.outputs["Value"], distribute_node.inputs["Density Factor"])

    instance_node = nodes.new("GeometryNodeInstanceOnPoints")
    instance_node.inputs["Pick Instance"].default_value = True
    links.new(distribute_node.outputs["Points"], instance_node.inputs["Points"])

    coll_info = nodes.new("GeometryNodeCollectionInfo")
    coll_info.inputs["Collection"].default_value = grass_coll
    coll_info.inputs["Separate Children"].default_value = True
    coll_info.inputs["Reset Children"].default_value = True
    links.new(coll_info.outputs["Instances"], instance_node.inputs["Instance"])

    rand_scale = nodes.new("FunctionNodeRandomValue")
    rand_scale.data_type = 'FLOAT'
    rand_scale.inputs["Min"].default_value = 0.25
    rand_scale.inputs["Max"].default_value = 0.50

    combine_scale = nodes.new("ShaderNodeCombineXYZ")
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["X"])
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["Y"])
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["Z"])
    links.new(combine_scale.outputs["Vector"], instance_node.inputs["Scale"])

    rand_rot_z = nodes.new("FunctionNodeRandomValue")
    rand_rot_z.data_type = 'FLOAT'
    rand_rot_z.inputs["Min"].default_value = 0.0
    rand_rot_z.inputs["Max"].default_value = math.radians(360.0)

    combine_rot = nodes.new("ShaderNodeCombineXYZ")
    links.new(rand_rot_z.outputs["Value"], combine_rot.inputs["Z"])
    links.new(combine_rot.outputs["Vector"], instance_node.inputs["Rotation"])

    sky_sphere = nodes.new("GeometryNodeMeshIcoSphere")
    sky_sphere.inputs["Radius"].default_value = 12.5
    sky_sphere.inputs["Subdivisions"].default_value = 4

    flip_faces = nodes.new("GeometryNodeFlipFaces")
    links.new(sky_sphere.outputs["Mesh"], flip_faces.inputs["Mesh"])

    transform_sky = nodes.new("GeometryNodeTransform")
    transform_sky.inputs["Translation"].default_value = (0.0, 0.0, ground_z_offset)
    links.new(flip_faces.outputs["Mesh"], transform_sky.inputs["Geometry"])

    set_sky_mat = nodes.new("GeometryNodeSetMaterial")
    set_sky_mat.inputs["Material"].default_value = sky_mat
    links.new(transform_sky.outputs["Geometry"], set_sky_mat.inputs["Geometry"])

    join_geo = nodes.new("GeometryNodeJoinGeometry")
    links.new(in_node.outputs["Geometry"], join_geo.inputs["Geometry"])
    links.new(instance_node.outputs["Instances"], join_geo.inputs["Geometry"])
    links.new(set_sky_mat.outputs["Geometry"], join_geo.inputs["Geometry"])

    links.new(join_geo.outputs["Geometry"], out_node.inputs["Geometry"])

    mod = ground_obj.modifiers.new(name="Grass_Ground_GN", type='NODES')
    mod.node_group = gn_group

    for item in gn_group.interface.items_tree:
        if getattr(item, "in_out", None) == "INPUT" and item.name == "Density Mask":
            mod[item.identifier + "_attribute_name"] = "Grass_Weight_Group"
            mod[item.identifier + "_use_attribute"] = True

    return ground_obj


# ==============================================================================
# SCROLL MATERIAL & GEOMETRY NODES SETUP
# ==============================================================================

def create_image_scroll_material(image_path):
    mat_name = "Scroll_Texture_Mat"
    mat = bpy.data.materials.get(mat_name)
    if mat:
        bpy.data.materials.remove(mat)

    mat = bpy.data.materials.new(name=mat_name)
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    output_node = nodes.new("ShaderNodeOutputMaterial")
    bsdf = nodes.new("ShaderNodeBsdfPrincipled")
    bsdf.inputs["Roughness"].default_value = 0.5

    tex_img = nodes.new("ShaderNodeTexImage")
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        tex_img.image = img
        tex_img.extension = 'REPEAT'

    uv_map_node = nodes.new("ShaderNodeUVMap")
    uv_map_node.uv_map = "UVMap"

    links.new(uv_map_node.outputs["UV"], tex_img.inputs["Vector"])
    links.new(tex_img.outputs["Color"], bsdf.inputs["Base Color"])
    links.new(bsdf.outputs["BSDF"], output_node.inputs["Surface"])

    return mat, tex_img.image if os.path.exists(image_path) else None


def create_pure_geometry_node_scroll(image_path=""):
    gn_group = bpy.data.node_groups.new("Scroll_Pure_GN", type="GeometryNodeTree")

    gn_group.interface.new_socket(name="Geometry", in_out="OUTPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Character Mesh", in_out="INPUT", socket_type="NodeSocketObject")

    sock_r_vg = gn_group.interface.new_socket(name="Right Hand Group", in_out="INPUT", socket_type="NodeSocketString")
    sock_r_vg.default_value = "mixamorig:RightHand"

    sock_l_vg = gn_group.interface.new_socket(name="Left Hand Group", in_out="INPUT", socket_type="NodeSocketString")
    sock_l_vg.default_value = "mixamorig:LeftHand"

    sock_offset = gn_group.interface.new_socket(name="Forward Offset", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_offset.default_value = 0.60

    sock_unfurl = gn_group.interface.new_socket(name="Unfurl", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_unfurl.default_value = 0.0

    sock_width = gn_group.interface.new_socket(name="Scroll Width", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_width.default_value = 0.35

    nodes = gn_group.nodes
    links = gn_group.links

    in_node = nodes.new("NodeGroupInput")
    out_node = nodes.new("NodeGroupOutput")

    char_info = nodes.new("GeometryNodeObjectInfo")
    char_info.transform_space = "RELATIVE"
    links.new(in_node.outputs["Character Mesh"], char_info.inputs["Object"])

    pos_node = nodes.new("GeometryNodeInputPosition")

    named_r = nodes.new("GeometryNodeInputNamedAttribute")
    named_r.data_type = "FLOAT"
    links.new(in_node.outputs["Right Hand Group"], named_r.inputs["Name"])

    cmp_r = nodes.new("FunctionNodeCompare")
    cmp_r.data_type = "FLOAT"
    cmp_r.operation = "GREATER_THAN"
    cmp_r.inputs[1].default_value = 0.001
    links.new(named_r.outputs["Attribute"], cmp_r.inputs[0])

    stat_r = nodes.new("GeometryNodeAttributeStatistic")
    stat_r.data_type = "FLOAT_VECTOR"
    links.new(char_info.outputs["Geometry"], stat_r.inputs["Geometry"])
    links.new(cmp_r.outputs["Result"], stat_r.inputs["Selection"])
    links.new(pos_node.outputs["Position"], stat_r.inputs["Attribute"])

    named_l = nodes.new("GeometryNodeInputNamedAttribute")
    named_l.data_type = "FLOAT"
    links.new(in_node.outputs["Left Hand Group"], named_l.inputs["Name"])

    cmp_l = nodes.new("FunctionNodeCompare")
    cmp_l.data_type = "FLOAT"
    cmp_l.operation = "GREATER_THAN"
    cmp_l.inputs[1].default_value = 0.001
    links.new(named_l.outputs["Attribute"], cmp_l.inputs[0])

    stat_l = nodes.new("GeometryNodeAttributeStatistic")
    stat_l.data_type = "FLOAT_VECTOR"
    links.new(char_info.outputs["Geometry"], stat_l.inputs["Geometry"])
    links.new(cmp_l.outputs["Result"], stat_l.inputs["Selection"])
    links.new(pos_node.outputs["Position"], stat_l.inputs["Attribute"])

    neg_offset = nodes.new("ShaderNodeMath")
    neg_offset.operation = "MULTIPLY"
    neg_offset.inputs[1].default_value = -1.0
    links.new(in_node.outputs["Forward Offset"], neg_offset.inputs[0])

    forward_vec = nodes.new("ShaderNodeCombineXYZ")
    forward_vec.inputs["X"].default_value = 0.0
    links.new(neg_offset.outputs["Value"], forward_vec.inputs["Y"])
    forward_vec.inputs["Z"].default_value = 0.0

    start_pos = nodes.new("ShaderNodeVectorMath")
    start_pos.operation = "ADD"
    links.new(stat_r.outputs["Mean"], start_pos.inputs[0])
    links.new(forward_vec.outputs["Vector"], start_pos.inputs[1])

    target_pos = nodes.new("ShaderNodeVectorMath")
    target_pos.operation = "ADD"
    links.new(stat_l.outputs["Mean"], target_pos.inputs[0])
    links.new(forward_vec.outputs["Vector"], target_pos.inputs[1])

    delta_vec = nodes.new("ShaderNodeVectorMath")
    delta_vec.operation = "SUBTRACT"
    links.new(target_pos.outputs["Vector"], delta_vec.inputs[0])
    links.new(start_pos.outputs["Vector"], delta_vec.inputs[1])

    dist_node = nodes.new("ShaderNodeVectorMath")
    dist_node.operation = "DISTANCE"
    links.new(start_pos.outputs["Vector"], dist_node.inputs[0])

    scale_vec = nodes.new("ShaderNodeVectorMath")
    scale_vec.operation = "SCALE"
    links.new(delta_vec.outputs["Vector"], scale_vec.inputs["Vector"])
    links.new(in_node.outputs["Unfurl"], scale_vec.inputs["Scale"])

    end_pos = nodes.new("ShaderNodeVectorMath")
    end_pos.operation = "ADD"
    links.new(start_pos.outputs["Vector"], end_pos.inputs[0])
    links.new(scale_vec.outputs["Vector"], end_pos.inputs[1])

    links.new(end_pos.outputs["Vector"], dist_node.inputs[1])

    curve_line = nodes.new("GeometryNodeCurvePrimitiveLine")
    links.new(start_pos.outputs["Vector"], curve_line.inputs["Start"])
    links.new(end_pos.outputs["Vector"], curve_line.inputs["End"])

    set_tilt = nodes.new("GeometryNodeSetCurveTilt")
    set_tilt.inputs["Tilt"].default_value = 1.5708
    links.new(curve_line.outputs["Curve"], set_tilt.inputs["Curve"])

    spline_param_main = nodes.new("GeometryNodeSplineParameter")
    store_u = nodes.new("GeometryNodeStoreNamedAttribute")
    store_u.inputs["Name"].default_value = "U_Factor"
    store_u.data_type = 'FLOAT'
    store_u.domain = 'POINT'
    links.new(set_tilt.outputs["Curve"], store_u.inputs["Geometry"])
    links.new(spline_param_main.outputs["Factor"], store_u.inputs["Value"])

    half_width_neg = nodes.new("ShaderNodeMath")
    half_width_neg.operation = "MULTIPLY"
    half_width_neg.inputs[1].default_value = -0.5
    links.new(in_node.outputs["Scroll Width"], half_width_neg.inputs[0])

    half_width_pos = nodes.new("ShaderNodeMath")
    half_width_pos.operation = "MULTIPLY"
    half_width_pos.inputs[1].default_value = 0.5
    links.new(in_node.outputs["Scroll Width"], half_width_pos.inputs[0])

    combine_start = nodes.new("ShaderNodeCombineXYZ")
    links.new(half_width_neg.outputs["Value"], combine_start.inputs["X"])

    combine_end = nodes.new("ShaderNodeCombineXYZ")
    links.new(half_width_pos.outputs["Value"], combine_end.inputs["X"])

    profile_line = nodes.new("GeometryNodeCurvePrimitiveLine")
    links.new(combine_start.outputs["Vector"], profile_line.inputs["Start"])
    links.new(combine_end.outputs["Vector"], profile_line.inputs["End"])

    spline_param_prof = nodes.new("GeometryNodeSplineParameter")
    store_v = nodes.new("GeometryNodeStoreNamedAttribute")
    store_v.inputs["Name"].default_value = "V_Factor"
    store_v.data_type = 'FLOAT'
    store_v.domain = 'POINT'
    links.new(profile_line.outputs["Curve"], store_v.inputs["Geometry"])
    links.new(spline_param_prof.outputs["Factor"], store_v.inputs["Value"])

    curve_to_mesh = nodes.new("GeometryNodeCurveToMesh")
    links.new(store_u.outputs["Geometry"], curve_to_mesh.inputs["Curve"])
    links.new(store_v.outputs["Geometry"], curve_to_mesh.inputs["Profile Curve"])

    img_aspect = 1.0
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        if img.size[1] > 0:
            img_aspect = img.size[0] / img.size[1]

    h_times_aspect = nodes.new("ShaderNodeMath")
    h_times_aspect.operation = "MULTIPLY"
    h_times_aspect.inputs[1].default_value = img_aspect
    links.new(in_node.outputs["Scroll Width"], h_times_aspect.inputs[0])

    tile_count = nodes.new("ShaderNodeMath")
    tile_count.operation = "DIVIDE"
    links.new(dist_node.outputs["Value"], tile_count.inputs[0])
    links.new(h_times_aspect.outputs["Value"], tile_count.inputs[1])

    read_u = nodes.new("GeometryNodeInputNamedAttribute")
    read_u.inputs["Name"].default_value = "U_Factor"
    read_u.data_type = 'FLOAT'

    scaled_u = nodes.new("ShaderNodeMath")
    scaled_u.operation = "MULTIPLY"
    links.new(read_u.outputs["Attribute"], scaled_u.inputs[0])
    links.new(tile_count.outputs["Value"], scaled_u.inputs[1])

    read_v = nodes.new("GeometryNodeInputNamedAttribute")
    read_v.inputs["Name"].default_value = "V_Factor"
    read_v.data_type = 'FLOAT'

    flip_v = nodes.new("ShaderNodeMath")
    flip_v.operation = "SUBTRACT"
    flip_v.inputs[0].default_value = 1.0
    links.new(read_v.outputs["Attribute"], flip_v.inputs[1])

    combine_uv = nodes.new("ShaderNodeCombineXYZ")
    links.new(scaled_u.outputs["Value"], combine_uv.inputs["X"])
    links.new(flip_v.outputs["Value"], combine_uv.inputs["Y"])

    store_uv = nodes.new("GeometryNodeStoreNamedAttribute")
    store_uv.inputs["Name"].default_value = "UVMap"
    store_uv.data_type = 'FLOAT2'
    store_uv.domain = 'CORNER'
    links.new(curve_to_mesh.outputs["Mesh"], store_uv.inputs["Geometry"])
    links.new(combine_uv.outputs["Vector"], store_uv.inputs["Value"])

    set_material = nodes.new("GeometryNodeSetMaterial")
    scroll_mat, _ = create_image_scroll_material(image_path)
    set_material.inputs["Material"].default_value = scroll_mat
    links.new(store_uv.outputs["Geometry"], set_material.inputs["Geometry"])

    links.new(set_material.outputs["Geometry"], out_node.inputs["Geometry"])

    return gn_group, scroll_mat


def setup_geometry_node_scroll(target_mesh, unfurl_start, unfurl_end, image_path=""):
    mesh_data = bpy.data.meshes.new("Scroll_Procedural_Mesh")
    scroll_obj = bpy.data.objects.new("Scroll_GN_Object", mesh_data)
    bpy.context.scene.collection.objects.link(scroll_obj)

    mod = scroll_obj.modifiers.new("Scroll_GN", type="NODES")
    gn_tree, scroll_mat = create_pure_geometry_node_scroll(image_path=image_path)
    mod.node_group = gn_tree

    scroll_obj.data.materials.append(scroll_mat)

    r_hand_name, l_hand_name = find_hand_vertex_groups(target_mesh)

    unfurl_id = None
    for item in gn_tree.interface.items_tree:
        if getattr(item, "in_out", None) == "INPUT":
            if item.name == "Character Mesh":
                mod[item.identifier] = target_mesh
            elif item.name == "Right Hand Group":
                mod[item.identifier] = r_hand_name
            elif item.name == "Left Hand Group":
                mod[item.identifier] = l_hand_name
            elif item.name == "Forward Offset":
                mod[item.identifier] = 0.60
            elif item.name == "Scroll Width":
                mod[item.identifier] = 0.35
            elif item.name == "Unfurl":
                unfurl_id = item.identifier

    for m in target_mesh.modifiers:
        if m.type == 'ARMATURE':
            m.show_viewport = True
            m.show_render = True
            m.show_in_editmode = True

    scroll_obj.hide_viewport = True
    scroll_obj.hide_render = True
    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=1)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=1)

    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=unfurl_start - 1)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=unfurl_start - 1)

    scroll_obj.hide_viewport = False
    scroll_obj.hide_render = False
    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=unfurl_start)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=unfurl_start)

    for fc in get_object_fcurves(scroll_obj):
        if fc.data_path in ["hide_viewport", "hide_render"]:
            for kp in fc.keyframe_points:
                kp.interpolation = 'CONSTANT'

    if unfurl_id:
        mod[unfurl_id] = 0.001
        mod.keyframe_insert(data_path=f'["{unfurl_id}"]', frame=unfurl_start)

        mod[unfurl_id] = 1.0
        mod.keyframe_insert(data_path=f'["{unfurl_id}"]', frame=unfurl_end)

    return scroll_obj


# ==============================================================================
# MAIN CONTROLLER
# ==============================================================================

def import_and_extract_action(fbx_path, action_name):
    existing_objects = set(bpy.data.objects)
    bpy.ops.import_scene.fbx(filepath=fbx_path)
    bpy.context.view_layer.update()

    imported_objects = set(bpy.data.objects) - existing_objects
    extracted_action = None
    for obj in imported_objects:
        if obj.animation_data and obj.animation_data.action:
            extracted_action = obj.animation_data.action
            extracted_action.name = action_name
            extracted_action.use_fake_user = True
            break
    return extracted_action, imported_objects


def purge_imported_objects(objects_to_remove):
    for obj in objects_to_remove:
        if obj.name in bpy.data.objects:
            if obj.animation_data:
                obj.animation_data_clear()
            for coll in list(obj.users_collection):
                coll.objects.unlink(obj)
    bpy.context.view_layer.update()


def main():
    cleanup_previous_run()

    walk_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Mutant Walking.fbx"
    turn_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Right Turn 90.fbx"
    unfurl_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Mutant Idle.fbx"

    sky_hdri_path = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/horn-koppe_spring_4k.exr"
    output_blend = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/output_scene.blend"

    image_path = "/Users/dobbinp1/Downloads/IMGP6691_psedit.jpg"
    grass_dir = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Grass004_4K-JPG"

    ground_obj = setup_grass_ground_geometry_nodes(grass_dir, sky_hdri_path, ground_z_offset=-0.2)

    CYCLES = 4
    CROSSFADE = 15
    TARGET_DISTANCE = 4.0

    walk_action, walk_objs = import_and_extract_action(walk_fbx, "walk")
    target_rig = next((obj for obj in walk_objs if obj.type == "ARMATURE"), None)
    target_mesh = next((obj for obj in walk_objs if obj.type == "MESH"), None)

    if not target_rig or not target_mesh:
        raise RuntimeError("Armature or Character Mesh not found in walk FBX.")

    travel_axis_idx, motion_sign = detect_natural_forward_axis(walk_action, target_rig)
    initial_heading = math.radians(90.0) if travel_axis_idx == 0 else 0.0
    target_rig.rotation_euler[2] = initial_heading

    action_len = int(walk_action.frame_range[1] - walk_action.frame_range[0])
    if action_len <= 0:
        action_len = 30

    anim_data = target_rig.animation_data_create()

    walk_track = anim_data.nla_tracks.new()
    walk_track.name = "Walk_Track"

    current_frame = 1
    for i in range(CYCLES):
        strip = walk_track.strips.new(name=f"Walk_{i+1}", start=int(current_frame), action=walk_action)
        strip.blend_type = "REPLACE"
        current_frame += action_len

    walk_end_frame = int(current_frame)

    turn_action, turn_objs = import_and_extract_action(turn_fbx, "rightturn")
    purge_imported_objects(turn_objs)

    turn_track = anim_data.nla_tracks.new()
    turn_track.name = "RightTurn_Track"

    turn_start = int(walk_end_frame - CROSSFADE)
    turn_strip = turn_track.strips.new(name="RightTurn", start=turn_start, action=turn_action)
    turn_strip.blend_type = "REPLACE"
    turn_strip.blend_in = CROSSFADE
    turn_strip.use_auto_blend = True
    turn_strip.extrapolation = "HOLD"
    turn_end_frame = int(turn_strip.frame_end)

    # Pass turn_start frame to scene data so handler stops sound when character stops walking
    bpy.context.scene["turn_start_frame"] = turn_start

    unfurl_action, unfurl_objs = import_and_extract_action(unfurl_fbx, "unfurl")
    purge_imported_objects(unfurl_objs)

    unfurl_track = anim_data.nla_tracks.new()
    unfurl_track.name = "Unfurl_Track"

    unfurl_start = int(turn_end_frame - CROSSFADE)
    unfurl_strip = unfurl_track.strips.new(name="Unfurl", start=unfurl_start, action=unfurl_action)
    unfurl_strip.blend_type = "REPLACE"
    unfurl_strip.blend_in = CROSSFADE
    unfurl_strip.use_auto_blend = True
    unfurl_strip.extrapolation = "HOLD"
    unfurl_end_frame = int(unfurl_strip.frame_end)

    root_action = bpy.data.actions.new(name="World_Root_Motion")
    target_rig.animation_data.action = root_action

    loc_fcurve = root_action.fcurve_ensure_for_datablock(datablock=target_rig, data_path="location", index=travel_axis_idx)
    final_target_value = TARGET_DISTANCE * motion_sign

    k_loc1 = loc_fcurve.keyframe_points.insert(frame=1.0, value=0.0)
    k_loc2 = loc_fcurve.keyframe_points.insert(frame=float(turn_start), value=final_target_value)
    k_loc3 = loc_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=final_target_value)

    k_loc1.interpolation = "LINEAR"
    k_loc2.interpolation = "CONSTANT"
    k_loc3.interpolation = "CONSTANT"

    rot_fcurve = root_action.fcurve_ensure_for_datablock(datablock=target_rig, data_path="rotation_euler", index=2)
    target_heading = initial_heading - math.radians(90.0)

    k_rot1 = rot_fcurve.keyframe_points.insert(frame=1.0, value=initial_heading)
    k_rot2 = rot_fcurve.keyframe_points.insert(frame=float(turn_start), value=initial_heading)
    k_rot3 = rot_fcurve.keyframe_points.insert(frame=float(turn_end_frame), value=target_heading)
    k_rot4 = rot_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=target_heading)

    k_rot1.interpolation = "CONSTANT"
    k_rot2.interpolation = "BEZIER"
    k_rot3.interpolation = "CONSTANT"
    k_rot4.interpolation = "CONSTANT"

    root_track = anim_data.nla_tracks.new()
    root_track.name = "World_Root_Track"
    root_strip = root_track.strips.new(name="Root_Motion_Strip", start=1, action=root_action)
    root_strip.blend_type = "REPLACE"
    root_strip.extrapolation = "HOLD"

    anim_data.action = None

    setup_geometry_node_scroll(
        target_mesh=target_mesh,
        unfurl_start=unfurl_start,
        unfurl_end=unfurl_end_frame,
        image_path=image_path,
    )

    cam_obj, _ = setup_modern_reel_camera(
        target_rig=target_rig,
        target_mesh=target_mesh,
        turn_start=turn_start,
        turn_end_frame=turn_end_frame,
        unfurl_end_frame=unfurl_end_frame
    )

    bpy.context.scene.frame_start = 1
    bpy.context.scene.frame_end = unfurl_end_frame

    ensure_world_sphere_size(cam_obj, ground_obj, safety_multiplier=2.5)

    # Register real-time footstep audio handler
    register_audio_handler()

    bpy.ops.wm.save_as_mainfile(filepath=output_blend)
    print(f"\n[Success] Dynamic world sizing, camera, & audio handler integrated. Saved to: {output_blend}")


main()
```

```
import math
import mathutils
import os
import glob
import bpy
import aud
from bpy_extras import anim_utils
from bpy_extras.object_utils import world_to_camera_view

# ==============================================================================
# AUDIO SYSTEM SETUP & HANDLERS
# ==============================================================================

FOOTSTEP_SOUND_PATH = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/footstep.wav"

AUDIO_DEVICE = aud.Device()
SOUND_FACTORY = None

# Track foot state to detect initial impact frame (downward edge detection)
_left_foot_was_down = False
_right_foot_was_down = False

if os.path.exists(FOOTSTEP_SOUND_PATH):
    SOUND_FACTORY = aud.Sound(FOOTSTEP_SOUND_PATH)
else:
    print(f"[Warning] Audio file not found at: {FOOTSTEP_SOUND_PATH}")


def play_footstep():
    """Plays the loaded footstep audio buffer."""
    if SOUND_FACTORY:
        print("[DEBUG AUDIO] >>> Triggering play_footstep()!")
        AUDIO_DEVICE.play(SOUND_FACTORY)
    else:
        print("[DEBUG AUDIO] play_footstep() called, but SOUND_FACTORY is None!")


def find_foot_bone(pose_bones, side="Left"):
    """Dynamically finds Mixamo or standard foot bones regardless of prefix numerical suffixes."""
    target_keywords = [f"mixamorig:{side}foot", f"{side.lower()}foot", f"foot.{side[0].lower()}", f"foot_{side[0].lower()}"]
    for bone in pose_bones:
        b_name = bone.name.lower()
        if any(kw in b_name for kw in target_keywords):
            return bone
    return None


def footstep_frame_handler(scene):
    global _left_foot_was_down, _right_foot_was_down

    try:
        current_frame = scene.frame_current

        # 1. IMMEDIATE PRINT TEST
        print(f"[DEBUG EXEC] Handler running on Frame: {current_frame}")

        turn_start = scene.get("turn_start_frame", 9999)

        if current_frame > turn_start + 15:
            return

        # 2. LOCATE ARMATURE
        rig = None
        for obj in scene.objects:
            if obj.type == 'ARMATURE' and "mixamorig" in obj.name.lower():
                rig = obj
                break

        if not rig:
            print("[DEBUG ERR] Rig not found!")
            return

        # 3. EVALUATE POSE & BONES
        depsgraph = bpy.context.evaluated_depsgraph_get()
        eval_rig = rig.evaluated_get(depsgraph)

        left_foot = find_foot_bone(eval_rig.pose.bones, "Left")
        right_foot = find_foot_bone(eval_rig.pose.bones, "Right")

        if not left_foot or not right_foot:
            print(f"[DEBUG ERR] Missing foot bone. Left: {left_foot}, Right: {right_foot}")
            return

        # 4. CHECK HEIGHTS
        l_z = (eval_rig.matrix_world @ left_foot.matrix).translation.z
        r_z = (eval_rig.matrix_world @ right_foot.matrix).translation.z

        ground_threshold = 0.10

        l_is_down = l_z <= ground_threshold
        r_is_down = r_z <= ground_threshold

        print(f"[DEBUG POS] Frame {current_frame} | L_Z: {l_z:.3f} | R_Z: {r_z:.3f}")

        # 5. TRIGGER AUDIO
        if l_is_down and not _left_foot_was_down:
            print(f"*** STEP TRIGGERED (LEFT) at Frame {current_frame} ***")
            play_footstep()
        elif r_is_down and not _right_foot_was_down:
            print(f"*** STEP TRIGGERED (RIGHT) at Frame {current_frame} ***")
            play_footstep()

        _left_foot_was_down = l_is_down
        _right_foot_was_down = r_is_down

    except Exception as e:
        import traceback
        print(f"\n[HANDLER CRASH] An error occurred inside the audio handler:")
        traceback.print_exc()


def register_audio_handler():
    unregister_audio_handler()
    bpy.app.handlers.frame_change_post.append(footstep_frame_handler)
    bpy.app.handlers.frame_change_pre.append(footstep_frame_handler)


def unregister_audio_handler():
    for h in list(bpy.app.handlers.frame_change_post):
        if getattr(h, "__name__", "") == "footstep_frame_handler":
            bpy.app.handlers.frame_change_post.remove(h)
    for h in list(bpy.app.handlers.frame_change_pre):
        if getattr(h, "__name__", "") == "footstep_frame_handler":
            bpy.app.handlers.frame_change_pre.remove(h)


register_audio_handler()


# ==============================================================================
# CLEANUP & HELPERS
# ==============================================================================

def cleanup_previous_run():
    """Clears old scroll objects, ground, node groups, camera anchors, collections, and orphaned data."""
    unregister_audio_handler()

    target_names = [
        "Master_NLA_Control",
        "Scroll_GN_Object",
        "Scroll_Start_Pin",
        "Scroll_End_Pin",
        "Scroll_Master_Container",
        "HDRI_Ground_Plane",
        "Ground_Mesh_Object",
        "Camera_Focus_Target",
        "Vertical_Reel_Camera"
    ]
    for obj in list(bpy.data.objects):
        if obj.name in target_names or (obj.name.startswith("mixamorig") and obj.type != "ARMATURE" and obj.type != "MESH"):
            if obj.animation_data:
                obj.animation_data_clear()
            bpy.data.objects.remove(obj, do_unlink=True)

    for action in list(bpy.data.actions):
        if action.users == 0 or action.use_fake_user:
            action.use_fake_user = False
            bpy.data.actions.remove(action)

    for ng in list(bpy.data.node_groups):
        if ng.name in ["Scroll_Unfurl_GN", "Scroll_Pure_GN", "HDRI_Ground_GN", "Grass_Ground_GN"]:
            bpy.data.node_groups.remove(ng)

    for coll_name in ["Scroll_Master_Container", "Grass_Artifacts_Collection"]:
        if coll_name in bpy.data.collections:
            coll = bpy.data.collections[coll_name]
            for obj in list(coll.objects):
                bpy.data.objects.remove(obj, do_unlink=True)
            bpy.data.collections.remove(coll)

    bpy.context.view_layer.update()


def get_slot_fcurves(action, anim_target_object):
    fcurves = []
    if not action:
        return fcurves

    anim_data = getattr(anim_target_object, "animation_data", None)
    if anim_data and getattr(anim_data, "action_slot", None):
        slot = anim_data.action_slot
        channelbag = anim_utils.action_get_channelbag_for_slot(action, slot)
        if channelbag and hasattr(channelbag, "fcurves"):
            return list(channelbag.fcurves)

    if hasattr(action, "slots"):
        for slot in action.slots:
            for cb in getattr(slot, "channelbags", []):
                fcurves.extend(getattr(cb, "fcurves", []))
        if fcurves:
            return fcurves

    if hasattr(action, "fcurves"):
        return list(action.fcurves)

    return fcurves


def get_object_fcurves(obj):
    if not obj or not obj.animation_data or not obj.animation_data.action:
        return []
    action = obj.animation_data.action
    return get_slot_fcurves(action, obj)


def detect_natural_forward_axis(action, rig_object):
    fcurves = get_slot_fcurves(action, rig_object)
    max_delta = 0.0
    best_axis = 1
    net_delta = 0.0

    start_frame = action.frame_range[0]
    end_frame = action.frame_range[1]

    for fc in fcurves:
        if fc.data_path.endswith("location") and fc.array_index in [0, 1, 2]:
            start_val = fc.evaluate(start_frame)
            end_val = fc.evaluate(end_frame)
            delta = end_val - start_val
            if abs(delta) > abs(max_delta):
                max_delta = delta
                best_axis = fc.array_index
                net_delta = delta

    direction_sign = 1.0 if net_delta >= 0 else -1.0
    return best_axis, direction_sign


def find_hand_vertex_groups(target_mesh):
    r_name = "mixamorig:RightHand"
    l_name = "mixamorig:LeftHand"

    if target_mesh and hasattr(target_mesh, "vertex_groups"):
        for vg in target_mesh.vertex_groups:
            vg_lower = vg.name.lower()
            if "righthand" in vg_lower or "hand_r" in vg_lower or "hand.r" in vg_lower:
                r_name = vg.name
            elif "lefthand" in vg_lower or "hand_l" in vg_lower or "hand.l" in vg_lower:
                l_name = vg.name

    return r_name, l_name


def get_or_create_fcurve(action, data_path, index=0, slot=None):
    """Retrieves or creates an FCurve across both classic and modern (Blender 4.0+) Action APIs."""
    if hasattr(action, "slots"):
        if not action.slots:
            slot = action.slots.new(name="Default_Slot")
        elif slot is None:
            slot = action.slots[0]

        channelbag = anim_utils.action_get_channelbag_for_slot(action, slot)
        if channelbag and hasattr(channelbag, "fcurves"):
            fc = channelbag.fcurves.find(data_path, index=index)
            if not fc:
                fc = channelbag.fcurves.new(data_path=data_path, index=index)
            return fc

    if hasattr(action, "fcurves"):
        fc = action.fcurves.find(data_path, index=index)
        if not fc:
            fc = action.fcurves.new(data_path=data_path, index=index)
        return fc

    raise AttributeError("Unable to access or create FCurves on Action object.")


# ==============================================================================
# UNPARENTED MATRIX-EVALUATED VERTICAL CAMERA SETUP
# ==============================================================================

def setup_modern_reel_camera(target_rig, target_mesh, turn_start, turn_end_frame, unfurl_end_frame):
    scene = bpy.context.scene
    scene.render.resolution_x = 1080
    scene.render.resolution_y = 1920
    scene.render.resolution_percentage = 100

    target_height = target_mesh.dimensions.z if target_mesh.dimensions.z > 0 else 1.8
    midpoint_z_offset = target_height * 0.50

    focus_target = bpy.data.objects.get("Camera_Focus_Target")
    if not focus_target:
        focus_target = bpy.data.objects.new("Camera_Focus_Target", None)
        focus_target.empty_display_type = 'PLAIN_AXES'
        focus_target.empty_display_size = 0.25
        scene.collection.objects.link(focus_target)

    focus_target.parent = None

    copy_loc = focus_target.constraints.get("Copy Location")
    if not copy_loc:
        copy_loc = focus_target.constraints.new(type='COPY_LOCATION')
    copy_loc.target = target_mesh
    copy_loc.use_offset = True
    focus_target.location.z = midpoint_z_offset

    cam_data = bpy.data.cameras.get("Vertical_Reel_Cam_Data")
    if not cam_data:
        cam_data = bpy.data.cameras.new("Vertical_Reel_Cam_Data")
    cam_data.lens = 45.0
    cam_data.sensor_fit = 'VERTICAL'
    cam_data.clip_end = 100.0

    cam_obj = bpy.data.objects.get("Vertical_Reel_Camera")
    if not cam_obj:
        cam_obj = bpy.data.objects.new("Vertical_Reel_Camera", cam_data)
        scene.collection.objects.link(cam_obj)

    scene.camera = cam_obj
    cam_obj.parent = None

    track = cam_obj.constraints.get("Track To")
    if not track:
        track = cam_obj.constraints.new(type='TRACK_TO')
    track.target = focus_target
    track.track_axis = 'TRACK_NEGATIVE_Z'
    track.up_axis = 'UP_Y'

    def check_mesh_in_frame(frame_num, cam_location, margin=0.08):
        cam_obj.location = cam_location
        scene.frame_set(int(frame_num))
        bpy.context.view_layer.update()

        depsgraph = bpy.context.evaluated_depsgraph_get()
        eval_mesh_obj = target_mesh.evaluated_get(depsgraph)
        mesh_data = eval_mesh_obj.data

        vert_step = max(1, len(mesh_data.vertices) // 300)
        world_matrix = target_mesh.matrix_world

        min_x, max_x = 1.0, 0.0
        min_y, max_y = 1.0, 0.0

        for i in range(0, len(mesh_data.vertices), vert_step):
            v_world = world_matrix @ mesh_data.vertices[i].co
            co_2d = world_to_camera_view(scene, cam_obj, v_world)

            if co_2d.x < min_x: min_x = co_2d.x
            if co_2d.x > max_x: max_x = co_2d.x
            if co_2d.y < min_y: min_y = co_2d.y
            if co_2d.y > max_y: max_y = co_2d.y

        in_bounds = (min_x >= margin and max_x <= (1.0 - margin) and
                     min_y >= margin and max_y <= (1.0 - margin))

        return in_bounds, min_x, max_x, min_y, max_y

    def find_fit_distance(frame_num, angle_rad, base_height_offset, initial_dist=6.0, margin=0.08):
        scene.frame_set(int(frame_num))
        bpy.context.view_layer.update()

        center_pos = focus_target.matrix_world.translation.copy()
        heading = target_rig.matrix_world.to_euler('XYZ').z
        view_angle = heading + angle_rad

        dist = initial_dist
        for _ in range(30):
            test_x = center_pos.x + dist * math.sin(view_angle)
            test_y = center_pos.y - dist * math.cos(view_angle)
            test_z = center_pos.z + base_height_offset

            fit, min_x, max_x, min_y, max_y = check_mesh_in_frame(frame_num, (test_x, test_y, test_z), margin=margin)
            if fit:
                break
            dist += 0.4

        return dist, center_pos, view_angle

    max_walk_dist = 6.0
    walk_frames_to_sample = range(1, turn_start + 1, 6)

    for f in walk_frames_to_sample:
        req_d, _, _ = find_fit_distance(f, angle_rad=math.radians(-45.0), base_height_offset=0.0, initial_dist=max_walk_dist, margin=0.10)
        if req_d > max_walk_dist:
            max_walk_dist = req_d

    scene.frame_set(1)
    bpy.context.view_layer.update()
    c1 = focus_target.matrix_world.translation.copy()
    h1 = target_rig.matrix_world.to_euler('XYZ').z + math.radians(-45.0)
    w_x1 = c1.x + max_walk_dist * math.sin(h1)
    w_y1 = c1.y - max_walk_dist * math.cos(h1)
    w_z1 = c1.z

    scene.frame_set(turn_start)
    bpy.context.view_layer.update()
    cts = focus_target.matrix_world.translation.copy()
    hts = target_rig.matrix_world.to_euler('XYZ').z + math.radians(-45.0)
    w_xts = cts.x + max_walk_dist * math.sin(hts)
    w_yts = cts.y - max_walk_dist * math.cos(hts)
    w_zts = cts.z

    unfurl_frames_to_sample = range(turn_end_frame, unfurl_end_frame + 1, 6)
    max_unfurl_dist = max_walk_dist

    for f in unfurl_frames_to_sample:
        req_d, _, _ = find_fit_distance(f, angle_rad=math.radians(0.0), base_height_offset=0.0, initial_dist=max_unfurl_dist, margin=0.10)
        if req_d > max_unfurl_dist:
            max_unfurl_dist = req_d

    scene.frame_set(turn_end_frame)
    bpy.context.view_layer.update()
    cte = focus_target.matrix_world.translation.copy()
    hte = target_rig.matrix_world.to_euler('XYZ').z + math.radians(0.0)
    f_x = cte.x + max_unfurl_dist * math.sin(hte)
    f_y = cte.y - max_unfurl_dist * math.cos(hte)
    f_z = cte.z

    if cam_obj.animation_data:
        cam_obj.animation_data_clear()

    cam_obj.location = (w_x1, w_y1, w_z1)
    cam_obj.keyframe_insert(data_path="location", frame=1)

    cam_obj.location = (w_xts, w_yts, w_zts)
    cam_obj.keyframe_insert(data_path="location", frame=turn_start)

    cam_obj.location = (f_x, f_y, f_z)
    cam_obj.keyframe_insert(data_path="location", frame=turn_end_frame)

    cam_obj.location = (f_x, f_y, f_z)
    cam_obj.keyframe_insert(data_path="location", frame=unfurl_end_frame)

    for fc in get_object_fcurves(cam_obj):
        for kp in fc.keyframe_points:
            kp.interpolation = 'BEZIER'

    scene.frame_set(1)
    return cam_obj, focus_target


# ==============================================================================
# GEOMETRY NODES GROUND, GRASS & SKY DOME SETUP
# ==============================================================================

def ensure_world_sphere_size(cam_obj, ground_obj, safety_multiplier=2.5):
    if not cam_obj or not ground_obj:
        return

    scene = bpy.context.scene
    max_cam_dist = 0.0

    for frame in range(scene.frame_start, scene.frame_end + 1, 5):
        scene.frame_set(frame)
        bpy.context.view_layer.update()

        cam_dist = cam_obj.matrix_world.translation.length
        if cam_dist > max_cam_dist:
            max_cam_dist = cam_dist

    world_radius = max(25.0, max_cam_dist * safety_multiplier)

    mod = ground_obj.modifiers.get("Grass_Ground_GN")
    if mod and mod.node_group:
        for node in mod.node_group.nodes:
            if node.bl_idname == "GeometryNodeMeshIcoSphere":
                node.inputs["Radius"].default_value = world_radius

    if ground_obj.type == 'MESH':
        bpy.context.view_layer.objects.active = ground_obj
        bpy.ops.object.mode_set(mode='OBJECT')

        ground_scale = world_radius / 12.5
        ground_obj.scale = (ground_scale, ground_scale, 1.0)
        bpy.ops.object.transform_apply(location=False, rotation=False, scale=True)

    cam_obj.data.clip_end = max(cam_obj.data.clip_end, world_radius * 2.5)

    scene.frame_set(1)
    bpy.context.view_layer.update()


def create_grass_material(grass_dir):
    mat = bpy.data.materials.new(name="Grass_PBR_Material")
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    out_node = nodes.new("ShaderNodeOutputMaterial")
    bsdf = nodes.new("ShaderNodeBsdfPrincipled")
    links.new(bsdf.outputs["BSDF"], out_node.inputs["Surface"])

    col_files = glob.glob(os.path.join(grass_dir, "*Color*.jpg")) + glob.glob(os.path.join(grass_dir, "*JPG*.jpg"))
    if col_files:
        tex_col = nodes.new("ShaderNodeTexImage")
        tex_col.image = bpy.data.images.load(col_files[0])
        links.new(tex_col.outputs["Color"], bsdf.inputs["Base Color"])

    rough_files = glob.glob(os.path.join(grass_dir, "*Roughness*.jpg"))
    if rough_files:
        tex_rough = nodes.new("ShaderNodeTexImage")
        tex_rough.image = bpy.data.images.load(rough_files[0])
        tex_rough.image.colorspace_settings.name = 'Non-Color'
        links.new(tex_rough.outputs["Color"], bsdf.inputs["Roughness"])

    norm_files = glob.glob(os.path.join(grass_dir, "*Normal*.jpg"))
    if norm_files:
        tex_norm = nodes.new("ShaderNodeTexImage")
        tex_norm.image = bpy.data.images.load(norm_files[0])
        tex_norm.image.colorspace_settings.name = 'Non-Color'

        normal_map_node = nodes.new("ShaderNodeNormalMap")
        links.new(tex_norm.outputs["Color"], normal_map_node.inputs["Color"])
        links.new(normal_map_node.outputs["Normal"], bsdf.inputs["Normal"])

    return mat


def create_sky_material(sky_hdri_path):
    mat = bpy.data.materials.new(name="Sky_Dome_Material")
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    out_node = nodes.new("ShaderNodeOutputMaterial")
    emission = nodes.new("ShaderNodeEmission")
    emission.inputs["Strength"].default_value = 1.0

    links.new(emission.outputs["Emission"], out_node.inputs["Surface"])

    if os.path.exists(sky_hdri_path):
        tex_env = nodes.new("ShaderNodeTexEnvironment")
        tex_env.image = bpy.data.images.load(sky_hdri_path)
        links.new(tex_env.outputs["Color"], emission.inputs["Color"])

    return mat


def setup_grass_ground_geometry_nodes(grass_dir, sky_hdri_path, ground_z_offset=-0.2):
    grass_coll = bpy.data.collections.get("Grass_Artifacts_Collection")
    if not grass_coll:
        grass_coll = bpy.data.collections.new("Grass_Artifacts_Collection")
        bpy.context.scene.collection.children.link(grass_coll)

    grass_mat = create_grass_material(grass_dir)
    sky_mat = create_sky_material(sky_hdri_path)

    mesh_files = glob.glob(os.path.join(grass_dir, "*.obj")) + glob.glob(os.path.join(grass_dir, "*.fbx"))

    if mesh_files:
        for fpath in mesh_files:
            if fpath.endswith(".obj"):
                bpy.ops.wm.obj_import(filepath=fpath)
            elif fpath.endswith(".fbx"):
                bpy.ops.import_scene.fbx(filepath=fpath)

            for o in bpy.context.selected_objects:
                grass_coll.objects.link(o)
                if o.name in bpy.context.scene.collection.objects:
                    bpy.context.scene.collection.objects.unlink(o)

        for obj in list(grass_coll.objects):
            if obj.dimensions.z > 0.5:
                grass_coll.objects.unlink(obj)
                bpy.data.objects.remove(obj, do_unlink=True)
    else:
        for i in range(3):
            m_data = bpy.data.meshes.new(f"Grass_Blade_Mesh_{i}")
            g_obj = bpy.data.objects.new(f"Grass_Blade_Obj_{i}", m_data)
            grass_coll.objects.link(g_obj)

            verts = [(0, 0, 0), (0.05, 0, 0), (0.04, 0, 0.15), (0.01, 0, 0.25), (0, 0, 0.3)]
            faces = [(0, 1, 2), (1, 2, 3), (2, 3, 4)]
            m_data.from_pydata(verts, [], faces)
            m_data.update()

            if g_obj.data.materials:
                g_obj.data.materials[0] = grass_mat
            else:
                g_obj.data.materials.append(grass_mat)

            g_obj.location = (100, 100, 0)

    bpy.ops.mesh.primitive_circle_add(radius=12.5, vertices=64, fill_type='NGON', location=(0, 0, ground_z_offset))
    ground_obj = bpy.context.active_object
    ground_obj.name = "Ground_Mesh_Object"

    vg = ground_obj.vertex_groups.new(name="Grass_Weight_Group")
    all_vert_indices = [v.index for v in ground_obj.data.vertices]
    vg.add(all_vert_indices, 1.0, 'REPLACE')
    ground_obj.data.materials.append(grass_mat)
    ground_obj.data.materials.append(sky_mat)

    gn_group = bpy.data.node_groups.new("Grass_Ground_GN", type="GeometryNodeTree")
    gn_group.interface.new_socket(name="Geometry", in_out="OUTPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Geometry", in_out="INPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Density Mask", in_out="INPUT", socket_type="NodeSocketFloat")

    nodes = gn_group.nodes
    links = gn_group.links
    out_node = nodes.new("NodeGroupOutput")
    in_node = nodes.new("NodeGroupInput")

    distribute_node = nodes.new("GeometryNodeDistributePointsOnFaces")
    distribute_node.distribute_method = 'POISSON'
    distribute_node.inputs["Distance Min"].default_value = 0.02
    distribute_node.inputs["Density Max"].default_value = 300.0
    links.new(in_node.outputs["Geometry"], distribute_node.inputs["Mesh"])

    noise_node = nodes.new("ShaderNodeTexNoise")
    noise_node.inputs["Scale"].default_value = 2.5
    noise_node.inputs["Detail"].default_value = 2.0

    color_ramp = nodes.new("ShaderNodeValToRGB")
    color_ramp.color_ramp.elements[0].position = 0.20
    color_ramp.color_ramp.elements[1].position = 0.80
    links.new(noise_node.outputs["Fac"], color_ramp.inputs["Fac"])

    math_weight = nodes.new("ShaderNodeMath")
    math_weight.operation = "MULTIPLY"
    links.new(color_ramp.outputs["Color"], math_weight.inputs[0])
    links.new(in_node.outputs["Density Mask"], math_weight.inputs[1])

    links.new(math_weight.outputs["Value"], distribute_node.inputs["Density Factor"])

    instance_node = nodes.new("GeometryNodeInstanceOnPoints")
    instance_node.inputs["Pick Instance"].default_value = True
    links.new(distribute_node.outputs["Points"], instance_node.inputs["Points"])

    coll_info = nodes.new("GeometryNodeCollectionInfo")
    coll_info.inputs["Collection"].default_value = grass_coll
    coll_info.inputs["Separate Children"].default_value = True
    coll_info.inputs["Reset Children"].default_value = True
    links.new(coll_info.outputs["Instances"], instance_node.inputs["Instance"])

    rand_scale = nodes.new("FunctionNodeRandomValue")
    rand_scale.data_type = 'FLOAT'
    rand_scale.inputs["Min"].default_value = 0.25
    rand_scale.inputs["Max"].default_value = 0.50

    combine_scale = nodes.new("ShaderNodeCombineXYZ")
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["X"])
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["Y"])
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["Z"])
    links.new(combine_scale.outputs["Vector"], instance_node.inputs["Scale"])

    rand_rot_z = nodes.new("FunctionNodeRandomValue")
    rand_rot_z.data_type = 'FLOAT'
    rand_rot_z.inputs["Min"].default_value = 0.0
    rand_rot_z.inputs["Max"].default_value = math.radians(360.0)

    combine_rot = nodes.new("ShaderNodeCombineXYZ")
    links.new(rand_rot_z.outputs["Value"], combine_rot.inputs["Z"])
    links.new(combine_rot.outputs["Vector"], instance_node.inputs["Rotation"])

    sky_sphere = nodes.new("GeometryNodeMeshIcoSphere")
    sky_sphere.inputs["Radius"].default_value = 12.5
    sky_sphere.inputs["Subdivisions"].default_value = 4

    flip_faces = nodes.new("GeometryNodeFlipFaces")
    links.new(sky_sphere.outputs["Mesh"], flip_faces.inputs["Mesh"])

    transform_sky = nodes.new("GeometryNodeTransform")
    transform_sky.inputs["Translation"].default_value = (0.0, 0.0, ground_z_offset)
    links.new(flip_faces.outputs["Mesh"], transform_sky.inputs["Geometry"])

    set_sky_mat = nodes.new("GeometryNodeSetMaterial")
    set_sky_mat.inputs["Material"].default_value = sky_mat
    links.new(transform_sky.outputs["Geometry"], set_sky_mat.inputs["Geometry"])

    join_geo = nodes.new("GeometryNodeJoinGeometry")
    links.new(in_node.outputs["Geometry"], join_geo.inputs["Geometry"])
    links.new(instance_node.outputs["Instances"], join_geo.inputs["Geometry"])
    links.new(set_sky_mat.outputs["Geometry"], join_geo.inputs["Geometry"])

    links.new(join_geo.outputs["Geometry"], out_node.inputs["Geometry"])

    mod = ground_obj.modifiers.new(name="Grass_Ground_GN", type='NODES')
    mod.node_group = gn_group

    for item in gn_group.interface.items_tree:
        if getattr(item, "in_out", None) == "INPUT" and item.name == "Density Mask":
            mod[item.identifier + "_attribute_name"] = "Grass_Weight_Group"
            mod[item.identifier + "_use_attribute"] = True

    return ground_obj


# ==============================================================================
# SCROLL MATERIAL & GEOMETRY NODES SETUP
# ==============================================================================

def create_image_scroll_material(image_path):
    mat_name = "Scroll_Texture_Mat"
    mat = bpy.data.materials.get(mat_name)
    if mat:
        bpy.data.materials.remove(mat)

    mat = bpy.data.materials.new(name=mat_name)
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    output_node = nodes.new("ShaderNodeOutputMaterial")
    bsdf = nodes.new("ShaderNodeBsdfPrincipled")
    bsdf.inputs["Roughness"].default_value = 0.5

    tex_img = nodes.new("ShaderNodeTexImage")
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        tex_img.image = img
        tex_img.extension = 'REPEAT'

    uv_map_node = nodes.new("ShaderNodeUVMap")
    uv_map_node.uv_map = "UVMap"

    links.new(uv_map_node.outputs["UV"], tex_img.inputs["Vector"])
    links.new(tex_img.outputs["Color"], bsdf.inputs["Base Color"])
    links.new(bsdf.outputs["BSDF"], output_node.inputs["Surface"])

    return mat, tex_img.image if os.path.exists(image_path) else None


def create_pure_geometry_node_scroll(image_path=""):
    gn_group = bpy.data.node_groups.new("Scroll_Pure_GN", type="GeometryNodeTree")

    gn_group.interface.new_socket(name="Geometry", in_out="OUTPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Character Mesh", in_out="INPUT", socket_type="NodeSocketObject")

    sock_r_vg = gn_group.interface.new_socket(name="Right Hand Group", in_out="INPUT", socket_type="NodeSocketString")
    sock_r_vg.default_value = "mixamorig:RightHand"

    sock_l_vg = gn_group.interface.new_socket(name="Left Hand Group", in_out="INPUT", socket_type="NodeSocketString")
    sock_l_vg.default_value = "mixamorig:LeftHand"

    sock_offset = gn_group.interface.new_socket(name="Forward Offset", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_offset.default_value = 0.60

    sock_unfurl = gn_group.interface.new_socket(name="Unfurl", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_unfurl.default_value = 0.0

    sock_width = gn_group.interface.new_socket(name="Scroll Width", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_width.default_value = 0.35

    nodes = gn_group.nodes
    links = gn_group.links

    in_node = nodes.new("NodeGroupInput")
    out_node = nodes.new("NodeGroupOutput")

    char_info = nodes.new("GeometryNodeObjectInfo")
    char_info.transform_space = "RELATIVE"
    links.new(in_node.outputs["Character Mesh"], char_info.inputs["Object"])

    pos_node = nodes.new("GeometryNodeInputPosition")

    named_r = nodes.new("GeometryNodeInputNamedAttribute")
    named_r.data_type = "FLOAT"
    links.new(in_node.outputs["Right Hand Group"], named_r.inputs["Name"])

    cmp_r = nodes.new("FunctionNodeCompare")
    cmp_r.data_type = "FLOAT"
    cmp_r.operation = "GREATER_THAN"
    cmp_r.inputs[1].default_value = 0.001
    links.new(named_r.outputs["Attribute"], cmp_r.inputs[0])

    stat_r = nodes.new("GeometryNodeAttributeStatistic")
    stat_r.data_type = "FLOAT_VECTOR"
    links.new(char_info.outputs["Geometry"], stat_r.inputs["Geometry"])
    links.new(cmp_r.outputs["Result"], stat_r.inputs["Selection"])
    links.new(pos_node.outputs["Position"], stat_r.inputs["Attribute"])

    named_l = nodes.new("GeometryNodeInputNamedAttribute")
    named_l.data_type = "FLOAT"
    links.new(in_node.outputs["Left Hand Group"], named_l.inputs["Name"])

    cmp_l = nodes.new("FunctionNodeCompare")
    cmp_l.data_type = "FLOAT"
    cmp_l.operation = "GREATER_THAN"
    cmp_l.inputs[1].default_value = 0.001
    links.new(named_l.outputs["Attribute"], cmp_l.inputs[0])

    stat_l = nodes.new("GeometryNodeAttributeStatistic")
    stat_l.data_type = "FLOAT_VECTOR"
    links.new(char_info.outputs["Geometry"], stat_l.inputs["Geometry"])
    links.new(cmp_l.outputs["Result"], stat_l.inputs["Selection"])
    links.new(pos_node.outputs["Position"], stat_l.inputs["Attribute"])

    neg_offset = nodes.new("ShaderNodeMath")
    neg_offset.operation = "MULTIPLY"
    neg_offset.inputs[1].default_value = -1.0
    links.new(in_node.outputs["Forward Offset"], neg_offset.inputs[0])

    forward_vec = nodes.new("ShaderNodeCombineXYZ")
    forward_vec.inputs["X"].default_value = 0.0
    links.new(neg_offset.outputs["Value"], forward_vec.inputs["Y"])
    forward_vec.inputs["Z"].default_value = 0.0

    start_pos = nodes.new("ShaderNodeVectorMath")
    start_pos.operation = "ADD"
    links.new(stat_r.outputs["Mean"], start_pos.inputs[0])
    links.new(forward_vec.outputs["Vector"], start_pos.inputs[1])

    target_pos = nodes.new("ShaderNodeVectorMath")
    target_pos.operation = "ADD"
    links.new(stat_l.outputs["Mean"], target_pos.inputs[0])
    links.new(forward_vec.outputs["Vector"], target_pos.inputs[1])

    delta_vec = nodes.new("ShaderNodeVectorMath")
    delta_vec.operation = "SUBTRACT"
    links.new(target_pos.outputs["Vector"], delta_vec.inputs[0])
    links.new(start_pos.outputs["Vector"], delta_vec.inputs[1])

    dist_node = nodes.new("ShaderNodeVectorMath")
    dist_node.operation = "DISTANCE"
    links.new(start_pos.outputs["Vector"], dist_node.inputs[0])

    scale_vec = nodes.new("ShaderNodeVectorMath")
    scale_vec.operation = "SCALE"
    links.new(delta_vec.outputs["Vector"], scale_vec.inputs["Vector"])
    links.new(in_node.outputs["Unfurl"], scale_vec.inputs["Scale"])

    end_pos = nodes.new("ShaderNodeVectorMath")
    end_pos.operation = "ADD"
    links.new(start_pos.outputs["Vector"], end_pos.inputs[0])
    links.new(scale_vec.outputs["Vector"], end_pos.inputs[1])

    links.new(end_pos.outputs["Vector"], dist_node.inputs[1])

    curve_line = nodes.new("GeometryNodeCurvePrimitiveLine")
    links.new(start_pos.outputs["Vector"], curve_line.inputs["Start"])
    links.new(end_pos.outputs["Vector"], curve_line.inputs["End"])

    set_tilt = nodes.new("GeometryNodeSetCurveTilt")
    set_tilt.inputs["Tilt"].default_value = 1.5708
    links.new(curve_line.outputs["Curve"], set_tilt.inputs["Curve"])

    spline_param_main = nodes.new("GeometryNodeSplineParameter")
    store_u = nodes.new("GeometryNodeStoreNamedAttribute")
    store_u.inputs["Name"].default_value = "U_Factor"
    store_u.data_type = 'FLOAT'
    store_u.domain = 'POINT'
    links.new(set_tilt.outputs["Curve"], store_u.inputs["Geometry"])
    links.new(spline_param_main.outputs["Factor"], store_u.inputs["Value"])

    half_width_neg = nodes.new("ShaderNodeMath")
    half_width_neg.operation = "MULTIPLY"
    half_width_neg.inputs[1].default_value = -0.5
    links.new(in_node.outputs["Scroll Width"], half_width_neg.inputs[0])

    half_width_pos = nodes.new("ShaderNodeMath")
    half_width_pos.operation = "MULTIPLY"
    half_width_pos.inputs[1].default_value = 0.5
    links.new(in_node.outputs["Scroll Width"], half_width_pos.inputs[0])

    combine_start = nodes.new("ShaderNodeCombineXYZ")
    links.new(half_width_neg.outputs["Value"], combine_start.inputs["X"])

    combine_end = nodes.new("ShaderNodeCombineXYZ")
    links.new(half_width_pos.outputs["Value"], combine_end.inputs["X"])

    profile_line = nodes.new("GeometryNodeCurvePrimitiveLine")
    links.new(combine_start.outputs["Vector"], profile_line.inputs["Start"])
    links.new(combine_end.outputs["Vector"], profile_line.inputs["End"])

    spline_param_prof = nodes.new("GeometryNodeSplineParameter")
    store_v = nodes.new("GeometryNodeStoreNamedAttribute")
    store_v.inputs["Name"].default_value = "V_Factor"
    store_v.data_type = 'FLOAT'
    store_v.domain = 'POINT'
    links.new(profile_line.outputs["Curve"], store_v.inputs["Geometry"])
    links.new(spline_param_prof.outputs["Factor"], store_v.inputs["Value"])

    curve_to_mesh = nodes.new("GeometryNodeCurveToMesh")
    links.new(store_u.outputs["Geometry"], curve_to_mesh.inputs["Curve"])
    links.new(store_v.outputs["Geometry"], curve_to_mesh.inputs["Profile Curve"])

    img_aspect = 1.0
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        if img.size[1] > 0:
            img_aspect = img.size[0] / img.size[1]

    h_times_aspect = nodes.new("ShaderNodeMath")
    h_times_aspect.operation = "MULTIPLY"
    h_times_aspect.inputs[1].default_value = img_aspect
    links.new(in_node.outputs["Scroll Width"], h_times_aspect.inputs[0])

    tile_count = nodes.new("ShaderNodeMath")
    tile_count.operation = "DIVIDE"
    links.new(dist_node.outputs["Value"], tile_count.inputs[0])
    links.new(h_times_aspect.outputs["Value"], tile_count.inputs[1])

    read_u = nodes.new("GeometryNodeInputNamedAttribute")
    read_u.inputs["Name"].default_value = "U_Factor"
    read_u.data_type = 'FLOAT'

    scaled_u = nodes.new("ShaderNodeMath")
    scaled_u.operation = "MULTIPLY"
    links.new(read_u.outputs["Attribute"], scaled_u.inputs[0])
    links.new(tile_count.outputs["Value"], scaled_u.inputs[1])

    read_v = nodes.new("GeometryNodeInputNamedAttribute")
    read_v.inputs["Name"].default_value = "V_Factor"
    read_v.data_type = 'FLOAT'

    flip_v = nodes.new("ShaderNodeMath")
    flip_v.operation = "SUBTRACT"
    flip_v.inputs[0].default_value = 1.0
    links.new(read_v.outputs["Attribute"], flip_v.inputs[1])

    combine_uv = nodes.new("ShaderNodeCombineXYZ")
    links.new(scaled_u.outputs["Value"], combine_uv.inputs["X"])
    links.new(flip_v.outputs["Value"], combine_uv.inputs["Y"])

    store_uv = nodes.new("GeometryNodeStoreNamedAttribute")
    store_uv.inputs["Name"].default_value = "UVMap"
    store_uv.data_type = 'FLOAT2'
    store_uv.domain = 'CORNER'
    links.new(curve_to_mesh.outputs["Mesh"], store_uv.inputs["Geometry"])
    links.new(combine_uv.outputs["Vector"], store_uv.inputs["Value"])

    set_material = nodes.new("GeometryNodeSetMaterial")
    scroll_mat, _ = create_image_scroll_material(image_path)
    set_material.inputs["Material"].default_value = scroll_mat
    links.new(store_uv.outputs["Geometry"], set_material.inputs["Geometry"])

    links.new(set_material.outputs["Geometry"], out_node.inputs["Geometry"])

    return gn_group, scroll_mat


def setup_geometry_node_scroll(target_mesh, unfurl_start, unfurl_end, image_path=""):
    mesh_data = bpy.data.meshes.new("Scroll_Procedural_Mesh")
    scroll_obj = bpy.data.objects.new("Scroll_GN_Object", mesh_data)
    bpy.context.scene.collection.objects.link(scroll_obj)

    mod = scroll_obj.modifiers.new("Scroll_GN", type="NODES")
    gn_tree, scroll_mat = create_pure_geometry_node_scroll(image_path=image_path)
    mod.node_group = gn_tree

    scroll_obj.data.materials.append(scroll_mat)

    r_hand_name, l_hand_name = find_hand_vertex_groups(target_mesh)

    unfurl_id = None
    for item in gn_tree.interface.items_tree:
        if getattr(item, "in_out", None) == "INPUT":
            if item.name == "Character Mesh":
                mod[item.identifier] = target_mesh
            elif item.name == "Right Hand Group":
                mod[item.identifier] = r_hand_name
            elif item.name == "Left Hand Group":
                mod[item.identifier] = l_hand_name
            elif item.name == "Forward Offset":
                mod[item.identifier] = 0.60
            elif item.name == "Scroll Width":
                mod[item.identifier] = 0.35
            elif item.name == "Unfurl":
                unfurl_id = item.identifier

    for m in target_mesh.modifiers:
        if m.type == 'ARMATURE':
            m.show_viewport = True
            m.show_render = True
            m.show_in_editmode = True

    scroll_obj.hide_viewport = True
    scroll_obj.hide_render = True
    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=1)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=1)

    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=unfurl_start - 1)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=unfurl_start - 1)

    scroll_obj.hide_viewport = False
    scroll_obj.hide_render = False
    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=unfurl_start)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=unfurl_start)

    for fc in get_object_fcurves(scroll_obj):
        if fc.data_path in ["hide_viewport", "hide_render"]:
            for kp in fc.keyframe_points:
                kp.interpolation = 'CONSTANT'

    if unfurl_id:
        mod[unfurl_id] = 0.001
        mod.keyframe_insert(data_path=f'["{unfurl_id}"]', frame=unfurl_start)

        mod[unfurl_id] = 1.0
        mod.keyframe_insert(data_path=f'["{unfurl_id}"]', frame=unfurl_end)

    return scroll_obj


# ==============================================================================
# MAIN CONTROLLER
# ==============================================================================

def import_and_extract_action(fbx_path, action_name):
    existing_objects = set(bpy.data.objects)
    bpy.ops.import_scene.fbx(filepath=fbx_path)
    bpy.context.view_layer.update()

    imported_objects = set(bpy.data.objects) - existing_objects
    extracted_action = None
    for obj in imported_objects:
        if obj.animation_data and obj.animation_data.action:
            extracted_action = obj.animation_data.action
            extracted_action.name = action_name
            extracted_action.use_fake_user = True
            break
    return extracted_action, imported_objects


def purge_imported_objects(objects_to_remove):
    for obj in objects_to_remove:
        if obj.name in bpy.data.objects:
            if obj.animation_data:
                obj.animation_data_clear()
            for coll in list(obj.users_collection):
                coll.objects.unlink(obj)
    bpy.context.view_layer.update()


def main():
    cleanup_previous_run()

    walk_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Mutant Walking.fbx"
    turn_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Right Turn 90.fbx"
    unfurl_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Mutant Idle.fbx"

    sky_hdri_path = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/horn-koppe_spring_4k.exr"
    output_blend = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/output_scene.blend"

    image_path = "/Users/dobbinp1/Downloads/IMGP6691_psedit.jpg"
    grass_dir = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Grass004_4K-JPG"

    ground_obj = setup_grass_ground_geometry_nodes(grass_dir, sky_hdri_path, ground_z_offset=-0.2)

    CYCLES = 4
    CROSSFADE = 15
    TARGET_DISTANCE = 4.0

    walk_action, walk_objs = import_and_extract_action(walk_fbx, "walk")
    target_rig = next((obj for obj in walk_objs if obj.type == "ARMATURE"), None)
    target_mesh = next((obj for obj in walk_objs if obj.type == "MESH"), None)

    if not target_rig or not target_mesh:
        raise RuntimeError("Armature or Character Mesh not found in walk FBX.")

    travel_axis_idx, motion_sign = detect_natural_forward_axis(walk_action, target_rig)
    initial_heading = math.radians(90.0) if travel_axis_idx == 0 else 0.0
    target_rig.rotation_euler[2] = initial_heading

    action_len = int(walk_action.frame_range[1] - walk_action.frame_range[0])
    if action_len <= 0:
        action_len = 30

    anim_data = target_rig.animation_data_create()

    walk_track = anim_data.nla_tracks.new()
    walk_track.name = "Walk_Track"

    current_frame = 1
    for i in range(CYCLES):
        strip = walk_track.strips.new(name=f"Walk_{i+1}", start=int(current_frame), action=walk_action)
        strip.blend_type = "REPLACE"
        current_frame += action_len

    walk_end_frame = int(current_frame)

    turn_action, turn_objs = import_and_extract_action(turn_fbx, "rightturn")
    purge_imported_objects(turn_objs)

    turn_track = anim_data.nla_tracks.new()
    turn_track.name = "RightTurn_Track"

    turn_start = int(walk_end_frame - CROSSFADE)
    turn_strip = turn_track.strips.new(name="RightTurn", start=turn_start, action=turn_action)
    turn_strip.blend_type = "REPLACE"
    turn_strip.blend_in = CROSSFADE
    turn_strip.use_auto_blend = True
    turn_strip.extrapolation = "HOLD"
    turn_end_frame = int(turn_strip.frame_end)

    bpy.context.scene["turn_start_frame"] = turn_start

    unfurl_action, unfurl_objs = import_and_extract_action(unfurl_fbx, "unfurl")
    purge_imported_objects(unfurl_objs)

    unfurl_track = anim_data.nla_tracks.new()
    unfurl_track.name = "Unfurl_Track"

    unfurl_start = int(turn_end_frame - CROSSFADE)
    unfurl_strip = unfurl_track.strips.new(name="Unfurl", start=unfurl_start, action=unfurl_action)
    unfurl_strip.blend_type = "REPLACE"
    unfurl_strip.blend_in = CROSSFADE
    unfurl_strip.use_auto_blend = True
    unfurl_strip.extrapolation = "HOLD"
    unfurl_end_frame = int(unfurl_strip.frame_end)

    root_action = bpy.data.actions.new(name="World_Root_Motion")
    target_rig.animation_data.action = root_action

    loc_fcurve = get_or_create_fcurve(root_action, "location", index=travel_axis_idx)
    final_target_value = TARGET_DISTANCE * motion_sign

    k_loc1 = loc_fcurve.keyframe_points.insert(frame=1.0, value=0.0)
    k_loc2 = loc_fcurve.keyframe_points.insert(frame=float(turn_start), value=final_target_value)
    k_loc3 = loc_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=final_target_value)

    k_loc1.interpolation = "LINEAR"
    k_loc2.interpolation = "CONSTANT"
    k_loc3.interpolation = "CONSTANT"

    rot_fcurve = get_or_create_fcurve(root_action, "rotation_euler", index=2)
    target_heading = initial_heading - math.radians(90.0)

    k_rot1 = rot_fcurve.keyframe_points.insert(frame=1.0, value=initial_heading)
    k_rot2 = rot_fcurve.keyframe_points.insert(frame=float(turn_start), value=initial_heading)
    k_rot3 = rot_fcurve.keyframe_points.insert(frame=float(turn_end_frame), value=target_heading)
    k_rot4 = rot_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=target_heading)

    k_rot1.interpolation = "CONSTANT"
    k_rot2.interpolation = "BEZIER"
    k_rot3.interpolation = "CONSTANT"
    k_rot4.interpolation = "CONSTANT"

    root_track = anim_data.nla_tracks.new()
    root_track.name = "World_Root_Track"
    root_strip = root_track.strips.new(name="Root_Motion_Strip", start=1, action=root_action)
    root_strip.blend_type = "REPLACE"
    root_strip.extrapolation = "HOLD"

    anim_data.action = None

    setup_geometry_node_scroll(
        target_mesh=target_mesh,
        unfurl_start=unfurl_start,
        unfurl_end=unfurl_end_frame,
        image_path=image_path,
    )

    cam_obj, _ = setup_modern_reel_camera(
        target_rig=target_rig,
        target_mesh=target_mesh,
        turn_start=turn_start,
        turn_end_frame=turn_end_frame,
        unfurl_end_frame=unfurl_end_frame
    )

    bpy.context.scene.frame_start = 1
    bpy.context.scene.frame_end = unfurl_end_frame

    ensure_world_sphere_size(cam_obj, ground_obj, safety_multiplier=2.5)

    register_audio_handler()

    bpy.ops.wm.save_as_mainfile(filepath=output_blend)
    print(f"\n[Success] Dynamic world sizing, camera, & audio handler integrated. Saved to: {output_blend}")


if __name__ == "__main__":
    main()
```

```
import math
import mathutils
import os
import glob
import bpy
import aud
from bpy_extras import anim_utils
from bpy_extras.object_utils import world_to_camera_view

# ==============================================================================
# AUDIO SYSTEM SETUP & HANDLERS
# ==============================================================================

FOOTSTEP_SOUND_PATH = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/footstep.wav"

AUDIO_DEVICE = aud.Device()
SOUND_FACTORY = None

# Track foot state to detect initial impact frame (downward edge detection)
_left_foot_was_down = False
_right_foot_was_down = False

if os.path.exists(FOOTSTEP_SOUND_PATH):
    SOUND_FACTORY = aud.Sound(FOOTSTEP_SOUND_PATH)
else:
    print(f"[Warning] Audio file not found at: {FOOTSTEP_SOUND_PATH}")


def play_footstep():
    """Plays the loaded footstep audio buffer."""
    if SOUND_FACTORY:
        print("[DEBUG AUDIO] >>> Triggering play_footstep()!")
        AUDIO_DEVICE.play(SOUND_FACTORY)
    else:
        print("[DEBUG AUDIO] play_footstep() called, but SOUND_FACTORY is None!")


def find_foot_bone(pose_bones, side="Left"):
    """Dynamically finds Mixamo or standard foot bones regardless of prefix numerical suffixes."""
    target_keywords = [f"mixamorig:{side}foot", f"{side.lower()}foot", f"foot.{side[0].lower()}", f"foot_{side[0].lower()}"]
    for bone in pose_bones:
        b_name = bone.name.lower()
        if any(kw in b_name for kw in target_keywords):
            return bone
    return None


def footstep_frame_handler(scene):
    global _left_foot_was_down, _right_foot_was_down

    try:
        current_frame = scene.frame_current

        # 1. IMMEDIATE PRINT TEST
        print(f"[DEBUG EXEC] Handler running on Frame: {current_frame}")

        turn_start = scene.get("turn_start_frame", 9999)

        if current_frame > turn_start + 15:
            return

        # 2. LOCATE ARMATURE
        rig = None
        for obj in scene.objects:
            if obj.type == 'ARMATURE' and "mixamorig" in obj.name.lower():
                rig = obj
                break

        if not rig:
            print("[DEBUG ERR] Rig not found!")
            return

        # 3. EVALUATE POSE & BONES
        depsgraph = bpy.context.evaluated_depsgraph_get()
        eval_rig = rig.evaluated_get(depsgraph)

        left_foot = find_foot_bone(eval_rig.pose.bones, "Left")
        right_foot = find_foot_bone(eval_rig.pose.bones, "Right")

        if not left_foot or not right_foot:
            print(f"[DEBUG ERR] Missing foot bone. Left: {left_foot}, Right: {right_foot}")
            return

        # 4. CHECK HEIGHTS
        l_z = (eval_rig.matrix_world @ left_foot.matrix).translation.z
        r_z = (eval_rig.matrix_world @ right_foot.matrix).translation.z

        ground_threshold = 0.10

        l_is_down = l_z <= ground_threshold
        r_is_down = r_z <= ground_threshold

        print(f"[DEBUG POS] Frame {current_frame} | L_Z: {l_z:.3f} | R_Z: {r_z:.3f}")

        # 5. TRIGGER AUDIO
        if l_is_down and not _left_foot_was_down:
            print(f"*** STEP TRIGGERED (LEFT) at Frame {current_frame} ***")
            play_footstep()
        elif r_is_down and not _right_foot_was_down:
            print(f"*** STEP TRIGGERED (RIGHT) at Frame {current_frame} ***")
            play_footstep()

        _left_foot_was_down = l_is_down
        _right_foot_was_down = r_is_down

    except Exception as e:
        import traceback
        print(f"\n[HANDLER CRASH] An error occurred inside the audio handler:")
        traceback.print_exc()



# --- DEFINE YOUR VARIABLES AND FUNCTIONS HERE ---

def register_audio_handler():
    unregister_audio_handler()
    bpy.app.handlers.frame_change_post.append(footstep_frame_handler)
    bpy.app.handlers.frame_change_pre.append(footstep_frame_handler) # Backup for viewport updates

def unregister_audio_handler():
    for h in list(bpy.app.handlers.frame_change_post):
        if h.__name__ == "footstep_frame_handler":
            bpy.app.handlers.frame_change_post.remove(h)
    for h in list(bpy.app.handlers.frame_change_pre):
        if h.__name__ == "footstep_frame_handler":
            bpy.app.handlers.frame_change_pre.remove(h)

# Ensure registration is executed at script runtime
register_audio_handler()

# Verify registration immediately
print("Current active frame_change_post handlers:", [h.__name__ for h in bpy.app.handlers.frame_change_post])


# ==============================================================================
# CLEANUP & HELPERS
# ==============================================================================

def cleanup_previous_run():
    """Clears old scroll objects, ground, node groups, camera anchors, collections, and orphaned data."""
    unregister_audio_handler()

    target_names = [
        "Master_NLA_Control",
        "Scroll_GN_Object",
        "Scroll_Start_Pin",
        "Scroll_End_Pin",
        "Scroll_Master_Container",
        "HDRI_Ground_Plane",
        "Ground_Mesh_Object",
        "Camera_Focus_Target",
        "Vertical_Reel_Camera"
    ]
    for obj in list(bpy.data.objects):
        if obj.name in target_names or (obj.name.startswith("mixamorig") and obj.type != "ARMATURE" and obj.type != "MESH"):
            if obj.animation_data:
                obj.animation_data_clear()
            bpy.data.objects.remove(obj, do_unlink=True)

    for action in list(bpy.data.actions):
        if action.users == 0 or action.use_fake_user:
            action.use_fake_user = False
            bpy.data.actions.remove(action)

    for ng in list(bpy.data.node_groups):
        if ng.name in ["Scroll_Unfurl_GN", "Scroll_Pure_GN", "HDRI_Ground_GN", "Grass_Ground_GN"]:
            bpy.data.node_groups.remove(ng)

    for coll_name in ["Scroll_Master_Container", "Grass_Artifacts_Collection"]:
        if coll_name in bpy.data.collections:
            coll = bpy.data.collections[coll_name]
            for obj in list(coll.objects):
                bpy.data.objects.remove(obj, do_unlink=True)
            bpy.data.collections.remove(coll)

    bpy.context.view_layer.update()


def get_slot_fcurves(action, anim_target_object):
    fcurves = []
    if not action:
        return fcurves

    anim_data = getattr(anim_target_object, "animation_data", None)
    if anim_data and getattr(anim_data, "action_slot", None):
        slot = anim_data.action_slot
        channelbag = anim_utils.action_get_channelbag_for_slot(action, slot)
        if channelbag and hasattr(channelbag, "fcurves"):
            return list(channelbag.fcurves)

    if hasattr(action, "slots"):
        for slot in action.slots:
            for cb in getattr(slot, "channelbags", []):
                fcurves.extend(getattr(cb, "fcurves", []))
        if fcurves:
            return fcurves

    if hasattr(action, "fcurves"):
        return list(action.fcurves)

    return fcurves


def get_object_fcurves(obj):
    if not obj or not obj.animation_data or not obj.animation_data.action:
        return []
    action = obj.animation_data.action
    return get_slot_fcurves(action, obj)


def detect_natural_forward_axis(action, rig_object):
    fcurves = get_slot_fcurves(action, rig_object)
    max_delta = 0.0
    best_axis = 1
    net_delta = 0.0

    start_frame = action.frame_range[0]
    end_frame = action.frame_range[1]

    for fc in fcurves:
        if fc.data_path.endswith("location") and fc.array_index in [0, 1, 2]:
            start_val = fc.evaluate(start_frame)
            end_val = fc.evaluate(end_frame)
            delta = end_val - start_val
            if abs(delta) > abs(max_delta):
                max_delta = delta
                best_axis = fc.array_index
                net_delta = delta

    direction_sign = 1.0 if net_delta >= 0 else -1.0
    return best_axis, direction_sign


def find_hand_vertex_groups(target_mesh):
    r_name = "mixamorig:RightHand"
    l_name = "mixamorig:LeftHand"

    if target_mesh and hasattr(target_mesh, "vertex_groups"):
        for vg in target_mesh.vertex_groups:
            vg_lower = vg.name.lower()
            if "righthand" in vg_lower or "hand_r" in vg_lower or "hand.r" in vg_lower:
                r_name = vg.name
            elif "lefthand" in vg_lower or "hand_l" in vg_lower or "hand.l" in vg_lower:
                l_name = vg.name

    return r_name, l_name


# ==============================================================================
# UNPARENTED MATRIX-EVALUATED VERTICAL CAMERA SETUP
# ==============================================================================

def setup_modern_reel_camera(target_rig, target_mesh, turn_start, turn_end_frame, unfurl_end_frame):
    """
    Creates a 9:16 Instagram Reel Camera:
    - Uses exact 2D camera viewport projection (world_to_camera_view) to test
      every vertex across the animation, ensuring 100% of the mesh fits inside
      the camera frame during both walking and unfurl phases.
    """
    scene = bpy.context.scene
    scene.render.resolution_x = 1080
    scene.render.resolution_y = 1920
    scene.render.resolution_percentage = 100

    target_height = target_mesh.dimensions.z if target_mesh.dimensions.z > 0 else 1.8
    midpoint_z_offset = target_height * 0.50

    focus_target = bpy.data.objects.get("Camera_Focus_Target")
    if not focus_target:
        focus_target = bpy.data.objects.new("Camera_Focus_Target", None)
        focus_target.empty_display_type = 'PLAIN_AXES'
        focus_target.empty_display_size = 0.25
        scene.collection.objects.link(focus_target)

    focus_target.parent = None

    copy_loc = focus_target.constraints.get("Copy Location")
    if not copy_loc:
        copy_loc = focus_target.constraints.new(type='COPY_LOCATION')
    copy_loc.target = target_mesh
    copy_loc.use_offset = True
    focus_target.location.z = midpoint_z_offset

    cam_data = bpy.data.cameras.get("Vertical_Reel_Cam_Data")
    if not cam_data:
        cam_data = bpy.data.cameras.new("Vertical_Reel_Cam_Data")
    cam_data.lens = 45.0
    cam_data.sensor_fit = 'VERTICAL'
    cam_data.clip_end = 100.0

    cam_obj = bpy.data.objects.get("Vertical_Reel_Camera")
    if not cam_obj:
        cam_obj = bpy.data.objects.new("Vertical_Reel_Camera", cam_data)
        scene.collection.objects.link(cam_obj)

    scene.camera = cam_obj
    cam_obj.parent = None

    track = cam_obj.constraints.get("Track To")
    if not track:
        track = cam_obj.constraints.new(type='TRACK_TO')
    track.target = focus_target
    track.track_axis = 'TRACK_NEGATIVE_Z'
    track.up_axis = 'UP_Y'

    def check_mesh_in_frame(frame_num, cam_location, margin=0.08):
        cam_obj.location = cam_location
        scene.frame_set(int(frame_num))
        bpy.context.view_layer.update()

        depsgraph = bpy.context.evaluated_depsgraph_get()
        eval_mesh_obj = target_mesh.evaluated_get(depsgraph)
        mesh_data = eval_mesh_obj.data

        vert_step = max(1, len(mesh_data.vertices) // 300)
        world_matrix = target_mesh.matrix_world

        min_x, max_x = 1.0, 0.0
        min_y, max_y = 1.0, 0.0

        for i in range(0, len(mesh_data.vertices), vert_step):
            v_world = world_matrix @ mesh_data.vertices[i].co
            co_2d = world_to_camera_view(scene, cam_obj, v_world)

            if co_2d.x < min_x: min_x = co_2d.x
            if co_2d.x > max_x: max_x = co_2d.x
            if co_2d.y < min_y: min_y = co_2d.y
            if co_2d.y > max_y: max_y = co_2d.y

        in_bounds = (min_x >= margin and max_x <= (1.0 - margin) and
                     min_y >= margin and max_y <= (1.0 - margin))

        return in_bounds, min_x, max_x, min_y, max_y

    def find_fit_distance(frame_num, angle_rad, base_height_offset, initial_dist=6.0, margin=0.08):
        scene.frame_set(int(frame_num))
        bpy.context.view_layer.update()

        center_pos = focus_target.matrix_world.translation.copy()
        heading = target_rig.matrix_world.to_euler('XYZ').z
        view_angle = heading + angle_rad

        dist = initial_dist
        for _ in range(30):
            test_x = center_pos.x + dist * math.sin(view_angle)
            test_y = center_pos.y - dist * math.cos(view_angle)
            test_z = center_pos.z + base_height_offset

            fit, min_x, max_x, min_y, max_y = check_mesh_in_frame(frame_num, (test_x, test_y, test_z), margin=margin)
            if fit:
                break
            dist += 0.4

        return dist, center_pos, view_angle

    max_walk_dist = 6.0
    walk_frames_to_sample = range(1, turn_start + 1, 6)

    for f in walk_frames_to_sample:
        req_d, _, _ = find_fit_distance(f, angle_rad=math.radians(-45.0), base_height_offset=0.0, initial_dist=max_walk_dist, margin=0.10)
        if req_d > max_walk_dist:
            max_walk_dist = req_d

    scene.frame_set(1)
    bpy.context.view_layer.update()
    c1 = focus_target.matrix_world.translation.copy()
    h1 = target_rig.matrix_world.to_euler('XYZ').z + math.radians(-45.0)
    w_x1 = c1.x + max_walk_dist * math.sin(h1)
    w_y1 = c1.y - max_walk_dist * math.cos(h1)
    w_z1 = c1.z

    scene.frame_set(turn_start)
    bpy.context.view_layer.update()
    cts = focus_target.matrix_world.translation.copy()
    hts = target_rig.matrix_world.to_euler('XYZ').z + math.radians(-45.0)
    w_xts = cts.x + max_walk_dist * math.sin(hts)
    w_yts = cts.y - max_walk_dist * math.cos(hts)
    w_zts = cts.z

    unfurl_frames_to_sample = range(turn_end_frame, unfurl_end_frame + 1, 6)
    max_unfurl_dist = max_walk_dist

    for f in unfurl_frames_to_sample:
        req_d, _, _ = find_fit_distance(f, angle_rad=math.radians(0.0), base_height_offset=0.0, initial_dist=max_unfurl_dist, margin=0.10)
        if req_d > max_unfurl_dist:
            max_unfurl_dist = req_d

    scene.frame_set(turn_end_frame)
    bpy.context.view_layer.update()
    cte = focus_target.matrix_world.translation.copy()
    hte = target_rig.matrix_world.to_euler('XYZ').z + math.radians(0.0)
    f_x = cte.x + max_unfurl_dist * math.sin(hte)
    f_y = cte.y - max_unfurl_dist * math.cos(hte)
    f_z = cte.z

    if cam_obj.animation_data:
        cam_obj.animation_data_clear()

    cam_obj.location = (w_x1, w_y1, w_z1)
    cam_obj.keyframe_insert(data_path="location", frame=1)

    cam_obj.location = (w_xts, w_yts, w_zts)
    cam_obj.keyframe_insert(data_path="location", frame=turn_start)

    cam_obj.location = (f_x, f_y, f_z)
    cam_obj.keyframe_insert(data_path="location", frame=turn_end_frame)

    cam_obj.location = (f_x, f_y, f_z)
    cam_obj.keyframe_insert(data_path="location", frame=unfurl_end_frame)

    for fc in get_object_fcurves(cam_obj):
        for kp in fc.keyframe_points:
            kp.interpolation = 'BEZIER'

    scene.frame_set(1)
    return cam_obj, focus_target


# ==============================================================================
# GEOMETRY NODES GROUND, GRASS & SKY DOME SETUP
# ==============================================================================

def ensure_world_sphere_size(cam_obj, ground_obj, safety_multiplier=2.5):
    if not cam_obj or not ground_obj:
        return

    scene = bpy.context.scene
    max_cam_dist = 0.0

    for frame in range(scene.frame_start, scene.frame_end + 1, 5):
        scene.frame_set(frame)
        bpy.context.view_layer.update()

        cam_dist = cam_obj.matrix_world.translation.length
        if cam_dist > max_cam_dist:
            max_cam_dist = cam_dist

    world_radius = max(25.0, max_cam_dist * safety_multiplier)

    mod = ground_obj.modifiers.get("Grass_Ground_GN")
    if mod and mod.node_group:
        for node in mod.node_group.nodes:
            if node.bl_idname == "GeometryNodeMeshIcoSphere":
                node.inputs["Radius"].default_value = world_radius

    if ground_obj.type == 'MESH':
        bpy.context.view_layer.objects.active = ground_obj
        bpy.ops.object.mode_set(mode='OBJECT')

        ground_scale = world_radius / 12.5
        ground_obj.scale = (ground_scale, ground_scale, 1.0)
        bpy.ops.object.transform_apply(location=False, rotation=False, scale=True)

    cam_obj.data.clip_end = max(cam_obj.data.clip_end, world_radius * 2.5)

    scene.frame_set(1)
    bpy.context.view_layer.update()


def create_grass_material(grass_dir):
    mat = bpy.data.materials.new(name="Grass_PBR_Material")
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    out_node = nodes.new("ShaderNodeOutputMaterial")
    bsdf = nodes.new("ShaderNodeBsdfPrincipled")
    links.new(bsdf.outputs["BSDF"], out_node.inputs["Surface"])

    col_files = glob.glob(os.path.join(grass_dir, "*Color*.jpg")) + glob.glob(os.path.join(grass_dir, "*JPG*.jpg"))
    if col_files:
        tex_col = nodes.new("ShaderNodeTexImage")
        tex_col.image = bpy.data.images.load(col_files[0])
        links.new(tex_col.outputs["Color"], bsdf.inputs["Base Color"])

    rough_files = glob.glob(os.path.join(grass_dir, "*Roughness*.jpg"))
    if rough_files:
        tex_rough = nodes.new("ShaderNodeTexImage")
        tex_rough.image = bpy.data.images.load(rough_files[0])
        tex_rough.image.colorspace_settings.name = 'Non-Color'
        links.new(tex_rough.outputs["Color"], bsdf.inputs["Roughness"])

    norm_files = glob.glob(os.path.join(grass_dir, "*Normal*.jpg"))
    if norm_files:
        tex_norm = nodes.new("ShaderNodeTexImage")
        tex_norm.image = bpy.data.images.load(norm_files[0])
        tex_norm.image.colorspace_settings.name = 'Non-Color'

        normal_map_node = nodes.new("ShaderNodeNormalMap")
        links.new(tex_norm.outputs["Color"], normal_map_node.inputs["Color"])
        links.new(normal_map_node.outputs["Normal"], bsdf.inputs["Normal"])

    return mat


def create_sky_material(sky_hdri_path):
    mat = bpy.data.materials.new(name="Sky_Dome_Material")
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    out_node = nodes.new("ShaderNodeOutputMaterial")
    emission = nodes.new("ShaderNodeEmission")
    emission.inputs["Strength"].default_value = 1.0

    links.new(emission.outputs["Emission"], out_node.inputs["Surface"])

    if os.path.exists(sky_hdri_path):
        tex_env = nodes.new("ShaderNodeTexEnvironment")
        tex_env.image = bpy.data.images.load(sky_hdri_path)
        links.new(tex_env.outputs["Color"], emission.inputs["Color"])

    return mat


def setup_grass_ground_geometry_nodes(grass_dir, sky_hdri_path, ground_z_offset=-0.2):
    grass_coll = bpy.data.collections.get("Grass_Artifacts_Collection")
    if not grass_coll:
        grass_coll = bpy.data.collections.new("Grass_Artifacts_Collection")
        bpy.context.scene.collection.children.link(grass_coll)

    grass_mat = create_grass_material(grass_dir)
    sky_mat = create_sky_material(sky_hdri_path)

    mesh_files = glob.glob(os.path.join(grass_dir, "*.obj")) + glob.glob(os.path.join(grass_dir, "*.fbx"))

    if mesh_files:
        for fpath in mesh_files:
            if fpath.endswith(".obj"):
                bpy.ops.wm.obj_import(filepath=fpath)
            elif fpath.endswith(".fbx"):
                bpy.ops.import_scene.fbx(filepath=fpath)

            for o in bpy.context.selected_objects:
                grass_coll.objects.link(o)
                if o.name in bpy.context.scene.collection.objects:
                    bpy.context.scene.collection.objects.unlink(o)

        for obj in list(grass_coll.objects):
            if obj.dimensions.z > 0.5:
                grass_coll.objects.unlink(obj)
                bpy.data.objects.remove(obj, do_unlink=True)
    else:
        for i in range(3):
            m_data = bpy.data.meshes.new(f"Grass_Blade_Mesh_{i}")
            g_obj = bpy.data.objects.new(f"Grass_Blade_Obj_{i}", m_data)
            grass_coll.objects.link(g_obj)

            verts = [(0, 0, 0), (0.05, 0, 0), (0.04, 0, 0.15), (0.01, 0, 0.25), (0, 0, 0.3)]
            faces = [(0, 1, 2), (1, 2, 3), (2, 3, 4)]
            m_data.from_pydata(verts, [], faces)
            m_data.update()

            if g_obj.data.materials:
                g_obj.data.materials[0] = grass_mat
            else:
                g_obj.data.materials.append(grass_mat)

            g_obj.location = (100, 100, 0)

    bpy.ops.mesh.primitive_circle_add(radius=12.5, vertices=64, fill_type='NGON', location=(0, 0, ground_z_offset))
    ground_obj = bpy.context.active_object
    ground_obj.name = "Ground_Mesh_Object"

    vg = ground_obj.vertex_groups.new(name="Grass_Weight_Group")
    all_vert_indices = [v.index for v in ground_obj.data.vertices]
    vg.add(all_vert_indices, 1.0, 'REPLACE')
    ground_obj.data.materials.append(grass_mat)
    ground_obj.data.materials.append(sky_mat)

    gn_group = bpy.data.node_groups.new("Grass_Ground_GN", type="GeometryNodeTree")
    gn_group.interface.new_socket(name="Geometry", in_out="INPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Geometry", in_out="OUTPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Density Mask", in_out="INPUT", socket_type="NodeSocketFloat")

    nodes = gn_group.nodes
    links = gn_group.links
    out_node = nodes.new("NodeGroupOutput")
    in_node = nodes.new("NodeGroupInput")

    distribute_node = nodes.new("GeometryNodeDistributePointsOnFaces")
    distribute_node.distribute_method = 'POISSON'
    distribute_node.inputs["Distance Min"].default_value = 0.02
    distribute_node.inputs["Density Max"].default_value = 300.0
    links.new(in_node.outputs["Geometry"], distribute_node.inputs["Mesh"])

    noise_node = nodes.new("ShaderNodeTexNoise")
    noise_node.inputs["Scale"].default_value = 2.5
    noise_node.inputs["Detail"].default_value = 2.0

    color_ramp = nodes.new("ShaderNodeValToRGB")
    color_ramp.color_ramp.elements[0].position = 0.20
    color_ramp.color_ramp.elements[1].position = 0.80
    links.new(noise_node.outputs["Fac"], color_ramp.inputs["Fac"])

    math_weight = nodes.new("ShaderNodeMath")
    math_weight.operation = "MULTIPLY"
    links.new(color_ramp.outputs["Color"], math_weight.inputs[0])
    links.new(in_node.outputs["Density Mask"], math_weight.inputs[1])

    links.new(math_weight.outputs["Value"], distribute_node.inputs["Density Factor"])

    instance_node = nodes.new("GeometryNodeInstanceOnPoints")
    instance_node.inputs["Pick Instance"].default_value = True
    links.new(distribute_node.outputs["Points"], instance_node.inputs["Points"])

    coll_info = nodes.new("GeometryNodeCollectionInfo")
    coll_info.inputs["Collection"].default_value = grass_coll
    coll_info.inputs["Separate Children"].default_value = True
    coll_info.inputs["Reset Children"].default_value = True
    links.new(coll_info.outputs["Instances"], instance_node.inputs["Instance"])

    rand_scale = nodes.new("FunctionNodeRandomValue")
    rand_scale.data_type = 'FLOAT'
    rand_scale.inputs["Min"].default_value = 0.25
    rand_scale.inputs["Max"].default_value = 0.50

    combine_scale = nodes.new("ShaderNodeCombineXYZ")
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["X"])
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["Y"])
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["Z"])
    links.new(combine_scale.outputs["Vector"], instance_node.inputs["Scale"])

    rand_rot_z = nodes.new("FunctionNodeRandomValue")
    rand_rot_z.data_type = 'FLOAT'
    rand_rot_z.inputs["Min"].default_value = 0.0
    rand_rot_z.inputs["Max"].default_value = math.radians(360.0)

    combine_rot = nodes.new("ShaderNodeCombineXYZ")
    links.new(rand_rot_z.outputs["Value"], combine_rot.inputs["Z"])
    links.new(combine_rot.outputs["Vector"], instance_node.inputs["Rotation"])

    sky_sphere = nodes.new("GeometryNodeMeshIcoSphere")
    sky_sphere.inputs["Radius"].default_value = 12.5
    sky_sphere.inputs["Subdivisions"].default_value = 4

    flip_faces = nodes.new("GeometryNodeFlipFaces")
    links.new(sky_sphere.outputs["Mesh"], flip_faces.inputs["Mesh"])

    transform_sky = nodes.new("GeometryNodeTransform")
    transform_sky.inputs["Translation"].default_value = (0.0, 0.0, ground_z_offset)
    links.new(flip_faces.outputs["Mesh"], transform_sky.inputs["Geometry"])

    set_sky_mat = nodes.new("GeometryNodeSetMaterial")
    set_sky_mat.inputs["Material"].default_value = sky_mat
    links.new(transform_sky.outputs["Geometry"], set_sky_mat.inputs["Geometry"])

    join_geo = nodes.new("GeometryNodeJoinGeometry")
    links.new(in_node.outputs["Geometry"], join_geo.inputs["Geometry"])
    links.new(instance_node.outputs["Instances"], join_geo.inputs["Geometry"])
    links.new(set_sky_mat.outputs["Geometry"], join_geo.inputs["Geometry"])

    links.new(join_geo.outputs["Geometry"], out_node.inputs["Geometry"])

    mod = ground_obj.modifiers.new(name="Grass_Ground_GN", type='NODES')
    mod.node_group = gn_group

    for item in gn_group.interface.items_tree:
        if getattr(item, "in_out", None) == "INPUT" and item.name == "Density Mask":
            mod[item.identifier + "_attribute_name"] = "Grass_Weight_Group"
            mod[item.identifier + "_use_attribute"] = True

    return ground_obj


# ==============================================================================
# SCROLL MATERIAL & GEOMETRY NODES SETUP
# ==============================================================================

def create_image_scroll_material(image_path):
    mat_name = "Scroll_Texture_Mat"
    mat = bpy.data.materials.get(mat_name)
    if mat:
        bpy.data.materials.remove(mat)

    mat = bpy.data.materials.new(name=mat_name)
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    output_node = nodes.new("ShaderNodeOutputMaterial")
    bsdf = nodes.new("ShaderNodeBsdfPrincipled")
    bsdf.inputs["Roughness"].default_value = 0.5

    tex_img = nodes.new("ShaderNodeTexImage")
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        tex_img.image = img
        tex_img.extension = 'REPEAT'

    uv_map_node = nodes.new("ShaderNodeUVMap")
    uv_map_node.uv_map = "UVMap"

    links.new(uv_map_node.outputs["UV"], tex_img.inputs["Vector"])
    links.new(tex_img.outputs["Color"], bsdf.inputs["Base Color"])
    links.new(bsdf.outputs["BSDF"], output_node.inputs["Surface"])

    return mat, tex_img.image if os.path.exists(image_path) else None


def create_pure_geometry_node_scroll(image_path=""):
    gn_group = bpy.data.node_groups.new("Scroll_Pure_GN", type="GeometryNodeTree")

    gn_group.interface.new_socket(name="Geometry", in_out="OUTPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Character Mesh", in_out="INPUT", socket_type="NodeSocketObject")

    sock_r_vg = gn_group.interface.new_socket(name="Right Hand Group", in_out="INPUT", socket_type="NodeSocketString")
    sock_r_vg.default_value = "mixamorig:RightHand"

    sock_l_vg = gn_group.interface.new_socket(name="Left Hand Group", in_out="INPUT", socket_type="NodeSocketString")
    sock_l_vg.default_value = "mixamorig:LeftHand"

    sock_offset = gn_group.interface.new_socket(name="Forward Offset", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_offset.default_value = 0.60

    sock_unfurl = gn_group.interface.new_socket(name="Unfurl", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_unfurl.default_value = 0.0

    sock_width = gn_group.interface.new_socket(name="Scroll Width", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_width.default_value = 0.35

    nodes = gn_group.nodes
    links = gn_group.links

    in_node = nodes.new("NodeGroupInput")
    out_node = nodes.new("NodeGroupOutput")

    char_info = nodes.new("GeometryNodeObjectInfo")
    char_info.transform_space = "RELATIVE"
    links.new(in_node.outputs["Character Mesh"], char_info.inputs["Object"])

    pos_node = nodes.new("GeometryNodeInputPosition")

    named_r = nodes.new("GeometryNodeInputNamedAttribute")
    named_r.data_type = "FLOAT"
    links.new(in_node.outputs["Right Hand Group"], named_r.inputs["Name"])

    cmp_r = nodes.new("FunctionNodeCompare")
    cmp_r.data_type = "FLOAT"
    cmp_r.operation = "GREATER_THAN"
    cmp_r.inputs[1].default_value = 0.001
    links.new(named_r.outputs["Attribute"], cmp_r.inputs[0])

    stat_r = nodes.new("GeometryNodeAttributeStatistic")
    stat_r.data_type = "FLOAT_VECTOR"
    links.new(char_info.outputs["Geometry"], stat_r.inputs["Geometry"])
    links.new(cmp_r.outputs["Result"], stat_r.inputs["Selection"])
    links.new(pos_node.outputs["Position"], stat_r.inputs["Attribute"])

    named_l = nodes.new("GeometryNodeInputNamedAttribute")
    named_l.data_type = "FLOAT"
    links.new(in_node.outputs["Left Hand Group"], named_l.inputs["Name"])

    cmp_l = nodes.new("FunctionNodeCompare")
    cmp_l.data_type = "FLOAT"
    cmp_l.operation = "GREATER_THAN"
    cmp_l.inputs[1].default_value = 0.001
    links.new(named_l.outputs["Attribute"], cmp_l.inputs[0])

    stat_l = nodes.new("GeometryNodeAttributeStatistic")
    stat_l.data_type = "FLOAT_VECTOR"
    links.new(char_info.outputs["Geometry"], stat_l.inputs["Geometry"])
    links.new(cmp_l.outputs["Result"], stat_l.inputs["Selection"])
    links.new(pos_node.outputs["Position"], stat_l.inputs["Attribute"])

    neg_offset = nodes.new("ShaderNodeMath")
    neg_offset.operation = "MULTIPLY"
    neg_offset.inputs[1].default_value = -1.0
    links.new(in_node.outputs["Forward Offset"], neg_offset.inputs[0])

    forward_vec = nodes.new("ShaderNodeCombineXYZ")
    forward_vec.inputs["X"].default_value = 0.0
    links.new(neg_offset.outputs["Value"], forward_vec.inputs["Y"])
    forward_vec.inputs["Z"].default_value = 0.0

    start_pos = nodes.new("ShaderNodeVectorMath")
    start_pos.operation = "ADD"
    links.new(stat_r.outputs["Mean"], start_pos.inputs[0])
    links.new(forward_vec.outputs["Vector"], start_pos.inputs[1])

    target_pos = nodes.new("ShaderNodeVectorMath")
    target_pos.operation = "ADD"
    links.new(stat_l.outputs["Mean"], target_pos.inputs[0])
    links.new(forward_vec.outputs["Vector"], target_pos.inputs[1])

    delta_vec = nodes.new("ShaderNodeVectorMath")
    delta_vec.operation = "SUBTRACT"
    links.new(target_pos.outputs["Vector"], delta_vec.inputs[0])
    links.new(start_pos.outputs["Vector"], delta_vec.inputs[1])

    dist_node = nodes.new("ShaderNodeVectorMath")
    dist_node.operation = "DISTANCE"
    links.new(start_pos.outputs["Vector"], dist_node.inputs[0])

    scale_vec = nodes.new("ShaderNodeVectorMath")
    scale_vec.operation = "SCALE"
    links.new(delta_vec.outputs["Vector"], scale_vec.inputs["Vector"])
    links.new(in_node.outputs["Unfurl"], scale_vec.inputs["Scale"])

    end_pos = nodes.new("ShaderNodeVectorMath")
    end_pos.operation = "ADD"
    links.new(start_pos.outputs["Vector"], end_pos.inputs[0])
    links.new(scale_vec.outputs["Vector"], end_pos.inputs[1])

    links.new(end_pos.outputs["Vector"], dist_node.inputs[1])

    curve_line = nodes.new("GeometryNodeCurvePrimitiveLine")
    links.new(start_pos.outputs["Vector"], curve_line.inputs["Start"])
    links.new(end_pos.outputs["Vector"], curve_line.inputs["End"])

    set_tilt = nodes.new("GeometryNodeSetCurveTilt")
    set_tilt.inputs["Tilt"].default_value = 1.5708
    links.new(curve_line.outputs["Curve"], set_tilt.inputs["Curve"])

    spline_param_main = nodes.new("GeometryNodeSplineParameter")
    store_u = nodes.new("GeometryNodeStoreNamedAttribute")
    store_u.inputs["Name"].default_value = "U_Factor"
    store_u.data_type = 'FLOAT'
    store_u.domain = 'POINT'
    links.new(set_tilt.outputs["Curve"], store_u.inputs["Geometry"])
    links.new(spline_param_main.outputs["Factor"], store_u.inputs["Value"])

    half_width_neg = nodes.new("ShaderNodeMath")
    half_width_neg.operation = "MULTIPLY"
    half_width_neg.inputs[1].default_value = -0.5
    links.new(in_node.outputs["Scroll Width"], half_width_neg.inputs[0])

    half_width_pos = nodes.new("ShaderNodeMath")
    half_width_pos.operation = "MULTIPLY"
    half_width_pos.inputs[1].default_value = 0.5
    links.new(in_node.outputs["Scroll Width"], half_width_pos.inputs[0])

    combine_start = nodes.new("ShaderNodeCombineXYZ")
    links.new(half_width_neg.outputs["Value"], combine_start.inputs["X"])

    combine_end = nodes.new("ShaderNodeCombineXYZ")
    links.new(half_width_pos.outputs["Value"], combine_end.inputs["X"])

    profile_line = nodes.new("GeometryNodeCurvePrimitiveLine")
    links.new(combine_start.outputs["Vector"], profile_line.inputs["Start"])
    links.new(combine_end.outputs["Vector"], profile_line.inputs["End"])

    spline_param_prof = nodes.new("GeometryNodeSplineParameter")
    store_v = nodes.new("GeometryNodeStoreNamedAttribute")
    store_v.inputs["Name"].default_value = "V_Factor"
    store_v.data_type = 'FLOAT'
    store_v.domain = 'POINT'
    links.new(profile_line.outputs["Curve"], store_v.inputs["Geometry"])
    links.new(spline_param_prof.outputs["Factor"], store_v.inputs["Value"])

    curve_to_mesh = nodes.new("GeometryNodeCurveToMesh")
    links.new(store_u.outputs["Geometry"], curve_to_mesh.inputs["Curve"])
    links.new(store_v.outputs["Geometry"], curve_to_mesh.inputs["Profile Curve"])

    img_aspect = 1.0
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        if img.size[1] > 0:
            img_aspect = img.size[0] / img.size[1]

    h_times_aspect = nodes.new("ShaderNodeMath")
    h_times_aspect.operation = "MULTIPLY"
    h_times_aspect.inputs[1].default_value = img_aspect
    links.new(in_node.outputs["Scroll Width"], h_times_aspect.inputs[0])

    tile_count = nodes.new("ShaderNodeMath")
    tile_count.operation = "DIVIDE"
    links.new(dist_node.outputs["Value"], tile_count.inputs[0])
    links.new(h_times_aspect.outputs["Value"], tile_count.inputs[1])

    read_u = nodes.new("GeometryNodeInputNamedAttribute")
    read_u.inputs["Name"].default_value = "U_Factor"
    read_u.data_type = 'FLOAT'

    scaled_u = nodes.new("ShaderNodeMath")
    scaled_u.operation = "MULTIPLY"
    links.new(read_u.outputs["Attribute"], scaled_u.inputs[0])
    links.new(tile_count.outputs["Value"], scaled_u.inputs[1])

    read_v = nodes.new("GeometryNodeInputNamedAttribute")
    read_v.inputs["Name"].default_value = "V_Factor"
    read_v.data_type = 'FLOAT'

    flip_v = nodes.new("ShaderNodeMath")
    flip_v.operation = "SUBTRACT"
    flip_v.inputs[0].default_value = 1.0
    links.new(read_v.outputs["Attribute"], flip_v.inputs[1])

    combine_uv = nodes.new("ShaderNodeCombineXYZ")
    links.new(scaled_u.outputs["Value"], combine_uv.inputs["X"])
    links.new(flip_v.outputs["Value"], combine_uv.inputs["Y"])

    store_uv = nodes.new("GeometryNodeStoreNamedAttribute")
    store_uv.inputs["Name"].default_value = "UVMap"
    store_uv.data_type = 'FLOAT2'
    store_uv.domain = 'CORNER'
    links.new(curve_to_mesh.outputs["Mesh"], store_uv.inputs["Geometry"])
    links.new(combine_uv.outputs["Vector"], store_uv.inputs["Value"])

    set_material = nodes.new("GeometryNodeSetMaterial")
    scroll_mat, _ = create_image_scroll_material(image_path)
    set_material.inputs["Material"].default_value = scroll_mat
    links.new(store_uv.outputs["Geometry"], set_material.inputs["Geometry"])

    links.new(set_material.outputs["Geometry"], out_node.inputs["Geometry"])

    return gn_group, scroll_mat


def setup_geometry_node_scroll(target_mesh, unfurl_start, unfurl_end, image_path=""):
    mesh_data = bpy.data.meshes.new("Scroll_Procedural_Mesh")
    scroll_obj = bpy.data.objects.new("Scroll_GN_Object", mesh_data)
    bpy.context.scene.collection.objects.link(scroll_obj)

    mod = scroll_obj.modifiers.new("Scroll_GN", type="NODES")
    gn_tree, scroll_mat = create_pure_geometry_node_scroll(image_path=image_path)
    mod.node_group = gn_tree

    scroll_obj.data.materials.append(scroll_mat)

    r_hand_name, l_hand_name = find_hand_vertex_groups(target_mesh)

    unfurl_id = None
    for item in gn_tree.interface.items_tree:
        if getattr(item, "in_out", None) == "INPUT":
            if item.name == "Character Mesh":
                mod[item.identifier] = target_mesh
            elif item.name == "Right Hand Group":
                mod[item.identifier] = r_hand_name
            elif item.name == "Left Hand Group":
                mod[item.identifier] = l_hand_name
            elif item.name == "Forward Offset":
                mod[item.identifier] = 0.60
            elif item.name == "Scroll Width":
                mod[item.identifier] = 0.35
            elif item.name == "Unfurl":
                unfurl_id = item.identifier

    for m in target_mesh.modifiers:
        if m.type == 'ARMATURE':
            m.show_viewport = True
            m.show_render = True
            m.show_in_editmode = True

    scroll_obj.hide_viewport = True
    scroll_obj.hide_render = True
    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=1)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=1)

    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=unfurl_start - 1)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=unfurl_start - 1)

    scroll_obj.hide_viewport = False
    scroll_obj.hide_render = False
    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=unfurl_start)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=unfurl_start)

    for fc in get_object_fcurves(scroll_obj):
        if fc.data_path in ["hide_viewport", "hide_render"]:
            for kp in fc.keyframe_points:
                kp.interpolation = 'CONSTANT'

    if unfurl_id:
        mod[unfurl_id] = 0.001
        mod.keyframe_insert(data_path=f'["{unfurl_id}"]', frame=unfurl_start)

        mod[unfurl_id] = 1.0
        mod.keyframe_insert(data_path=f'["{unfurl_id}"]', frame=unfurl_end)

    return scroll_obj


# ==============================================================================
# MAIN CONTROLLER
# ==============================================================================

def import_and_extract_action(fbx_path, action_name):
    existing_objects = set(bpy.data.objects)
    bpy.ops.import_scene.fbx(filepath=fbx_path)
    bpy.context.view_layer.update()

    imported_objects = set(bpy.data.objects) - existing_objects
    extracted_action = None
    for obj in imported_objects:
        if obj.animation_data and obj.animation_data.action:
            extracted_action = obj.animation_data.action
            extracted_action.name = action_name
            extracted_action.use_fake_user = True
            break
    return extracted_action, imported_objects


def purge_imported_objects(objects_to_remove):
    for obj in objects_to_remove:
        if obj.name in bpy.data.objects:
            if obj.animation_data:
                obj.animation_data_clear()
            for coll in list(obj.users_collection):
                coll.objects.unlink(obj)
    bpy.context.view_layer.update()


def main():
    cleanup_previous_run()

    walk_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Mutant Walking.fbx"
    turn_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Right Turn 90.fbx"
    unfurl_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Mutant Idle.fbx"

    sky_hdri_path = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/horn-koppe_spring_4k.exr"
    output_blend = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/output_scene.blend"

    image_path = "/Users/dobbinp1/Downloads/IMGP6691_psedit.jpg"
    grass_dir = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Grass004_4K-JPG"

    ground_obj = setup_grass_ground_geometry_nodes(grass_dir, sky_hdri_path, ground_z_offset=-0.2)

    CYCLES = 4
    CROSSFADE = 15
    TARGET_DISTANCE = 4.0

    walk_action, walk_objs = import_and_extract_action(walk_fbx, "walk")
    target_rig = next((obj for obj in walk_objs if obj.type == "ARMATURE"), None)
    target_mesh = next((obj for obj in walk_objs if obj.type == "MESH"), None)

    if not target_rig or not target_mesh:
        raise RuntimeError("Armature or Character Mesh not found in walk FBX.")

    travel_axis_idx, motion_sign = detect_natural_forward_axis(walk_action, target_rig)
    initial_heading = math.radians(90.0) if travel_axis_idx == 0 else 0.0
    target_rig.rotation_euler[2] = initial_heading

    action_len = int(walk_action.frame_range[1] - walk_action.frame_range[0])
    if action_len <= 0:
        action_len = 30

    anim_data = target_rig.animation_data_create()

    walk_track = anim_data.nla_tracks.new()
    walk_track.name = "Walk_Track"

    current_frame = 1
    for i in range(CYCLES):
        strip = walk_track.strips.new(name=f"Walk_{i+1}", start=int(current_frame), action=walk_action)
        strip.blend_type = "REPLACE"
        current_frame += action_len

    walk_end_frame = int(current_frame)

    turn_action, turn_objs = import_and_extract_action(turn_fbx, "rightturn")
    purge_imported_objects(turn_objs)

    turn_track = anim_data.nla_tracks.new()
    turn_track.name = "RightTurn_Track"

    turn_start = int(walk_end_frame - CROSSFADE)
    turn_strip = turn_track.strips.new(name="RightTurn", start=turn_start, action=turn_action)
    turn_strip.blend_type = "REPLACE"
    turn_strip.blend_in = CROSSFADE
    turn_strip.use_auto_blend = True
    turn_strip.extrapolation = "HOLD"
    turn_end_frame = int(turn_strip.frame_end)

    # Pass turn_start frame to scene data so handler stops sound when character stops walking
    bpy.context.scene["turn_start_frame"] = turn_start

    unfurl_action, unfurl_objs = import_and_extract_action(unfurl_fbx, "unfurl")
    purge_imported_objects(unfurl_objs)

    unfurl_track = anim_data.nla_tracks.new()
    unfurl_track.name = "Unfurl_Track"

    unfurl_start = int(turn_end_frame - CROSSFADE)
    unfurl_strip = unfurl_track.strips.new(name="Unfurl", start=unfurl_start, action=unfurl_action)
    unfurl_strip.blend_type = "REPLACE"
    unfurl_strip.blend_in = CROSSFADE
    unfurl_strip.use_auto_blend = True
    unfurl_strip.extrapolation = "HOLD"
    unfurl_end_frame = int(unfurl_strip.frame_end)

    root_action = bpy.data.actions.new(name="World_Root_Motion")
    target_rig.animation_data.action = root_action

    loc_fcurve = root_action.fcurve_ensure_for_datablock(datablock=target_rig, data_path="location", index=travel_axis_idx)
    final_target_value = TARGET_DISTANCE * motion_sign

    k_loc1 = loc_fcurve.keyframe_points.insert(frame=1.0, value=0.0)
    k_loc2 = loc_fcurve.keyframe_points.insert(frame=float(turn_start), value=final_target_value)
    k_loc3 = loc_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=final_target_value)

    k_loc1.interpolation = "LINEAR"
    k_loc2.interpolation = "CONSTANT"
    k_loc3.interpolation = "CONSTANT"

    rot_fcurve = root_action.fcurve_ensure_for_datablock(datablock=target_rig, data_path="rotation_euler", index=2)
    target_heading = initial_heading - math.radians(90.0)

    k_rot1 = rot_fcurve.keyframe_points.insert(frame=1.0, value=initial_heading)
    k_rot2 = rot_fcurve.keyframe_points.insert(frame=float(turn_start), value=initial_heading)
    k_rot3 = rot_fcurve.keyframe_points.insert(frame=float(turn_end_frame), value=target_heading)
    k_rot4 = rot_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=target_heading)

    k_rot1.interpolation = "CONSTANT"
    k_rot2.interpolation = "BEZIER"
    k_rot3.interpolation = "CONSTANT"
    k_rot4.interpolation = "CONSTANT"

    root_track = anim_data.nla_tracks.new()
    root_track.name = "World_Root_Track"
    root_strip = root_track.strips.new(name="Root_Motion_Strip", start=1, action=root_action)
    root_strip.blend_type = "REPLACE"
    root_strip.extrapolation = "HOLD"

    anim_data.action = None

    setup_geometry_node_scroll(
        target_mesh=target_mesh,
        unfurl_start=unfurl_start,
        unfurl_end=unfurl_end_frame,
        image_path=image_path,
    )

    cam_obj, _ = setup_modern_reel_camera(
        target_rig=target_rig,
        target_mesh=target_mesh,
        turn_start=turn_start,
        turn_end_frame=turn_end_frame,
        unfurl_end_frame=unfurl_end_frame
    )

    bpy.context.scene.frame_start = 1
    bpy.context.scene.frame_end = unfurl_end_frame

    ensure_world_sphere_size(cam_obj, ground_obj, safety_multiplier=2.5)

    # Register real-time footstep audio handler
    register_audio_handler()

    bpy.ops.wm.save_as_mainfile(filepath=output_blend)
    print(f"\n[Success] Dynamic world sizing, camera, & audio handler integrated. Saved to: {output_blend}")


main()
```

```
import math
import mathutils
import os
import glob
import bpy
from bpy_extras import anim_utils
from bpy_extras.object_utils import world_to_camera_view

# ==============================================================================
# CLEANUP & HELPERS
# ==============================================================================

def cleanup_previous_run():
    """Clears old scroll objects, ground, node groups, camera anchors, collections, and orphaned data."""
    target_names = [
        "Master_NLA_Control",
        "Scroll_GN_Object",
        "Scroll_Start_Pin",
        "Scroll_End_Pin",
        "Scroll_Master_Container",
        "HDRI_Ground_Plane",
        "Ground_Mesh_Object",
        "Camera_Focus_Target",
        "Vertical_Reel_Camera"
    ]
    for obj in list(bpy.data.objects):
        if obj.name in target_names or (obj.name.startswith("mixamorig") and obj.type != "ARMATURE" and obj.type != "MESH"):
            if obj.animation_data:
                obj.animation_data_clear()
            bpy.data.objects.remove(obj, do_unlink=True)

    for action in list(bpy.data.actions):
        if action.users == 0 or action.use_fake_user:
            action.use_fake_user = False
            bpy.data.actions.remove(action)

    for ng in list(bpy.data.node_groups):
        if ng.name in ["Scroll_Unfurl_GN", "Scroll_Pure_GN", "HDRI_Ground_GN", "Grass_Ground_GN"]:
            bpy.data.node_groups.remove(ng)

    for coll_name in ["Scroll_Master_Container", "Grass_Artifacts_Collection"]:
        if coll_name in bpy.data.collections:
            coll = bpy.data.collections[coll_name]
            for obj in list(coll.objects):
                bpy.data.objects.remove(obj, do_unlink=True)
            bpy.data.collections.remove(coll)

    bpy.context.view_layer.update()


def get_slot_fcurves(action, anim_target_object):
    fcurves = []
    if not action:
        return fcurves

    anim_data = getattr(anim_target_object, "animation_data", None)
    if anim_data and getattr(anim_data, "action_slot", None):
        slot = anim_data.action_slot
        channelbag = anim_utils.action_get_channelbag_for_slot(action, slot)
        if channelbag and hasattr(channelbag, "fcurves"):
            return list(channelbag.fcurves)

    if hasattr(action, "slots"):
        for slot in action.slots:
            for cb in getattr(slot, "channelbags", []):
                fcurves.extend(getattr(cb, "fcurves", []))
        if fcurves:
            return fcurves

    if hasattr(action, "fcurves"):
        return list(action.fcurves)

    return fcurves


def get_object_fcurves(obj):
    if not obj or not obj.animation_data or not obj.animation_data.action:
        return []
    action = obj.animation_data.action
    return get_slot_fcurves(action, obj)


def detect_natural_forward_axis(action, rig_object):
    fcurves = get_slot_fcurves(action, rig_object)
    max_delta = 0.0
    best_axis = 1
    net_delta = 0.0

    start_frame = action.frame_range[0]
    end_frame = action.frame_range[1]

    for fc in fcurves:
        if fc.data_path.endswith("location") and fc.array_index in [0, 1, 2]:
            start_val = fc.evaluate(start_frame)
            end_val = fc.evaluate(end_frame)
            delta = end_val - start_val
            if abs(delta) > abs(max_delta):
                max_delta = delta
                best_axis = fc.array_index
                net_delta = delta

    direction_sign = 1.0 if net_delta >= 0 else -1.0
    return best_axis, direction_sign


def find_hand_vertex_groups(target_mesh):
    r_name = "mixamorig:RightHand"
    l_name = "mixamorig:LeftHand"

    if target_mesh and hasattr(target_mesh, "vertex_groups"):
        for vg in target_mesh.vertex_groups:
            vg_lower = vg.name.lower()
            if "righthand" in vg_lower or "hand_r" in vg_lower or "hand.r" in vg_lower:
                r_name = vg.name
            elif "lefthand" in vg_lower or "hand_l" in vg_lower or "hand.l" in vg_lower:
                l_name = vg.name

    return r_name, l_name


# ==============================================================================
# UNPARENTED MATRIX-EVALUATED VERTICAL CAMERA SETUP
# ==============================================================================

def setup_modern_reel_camera(target_rig, target_mesh, turn_start, turn_end_frame, unfurl_end_frame):
    """
    Creates a 9:16 Instagram Reel Camera:
    - Uses exact 2D camera viewport projection (world_to_camera_view) to test
      every vertex across the animation, ensuring 100% of the mesh fits inside
      the camera frame during both walking and unfurl phases.
    """
    scene = bpy.context.scene
    scene.render.resolution_x = 1080
    scene.render.resolution_y = 1920
    scene.render.resolution_percentage = 100

    target_height = target_mesh.dimensions.z if target_mesh.dimensions.z > 0 else 1.8
    midpoint_z_offset = target_height * 0.50  # Targets exact vertical middle of the mesh

    # 1. Focus Target Null
    focus_target = bpy.data.objects.get("Camera_Focus_Target")
    if not focus_target:
        focus_target = bpy.data.objects.new("Camera_Focus_Target", None)
        focus_target.empty_display_type = 'PLAIN_AXES'
        focus_target.empty_display_size = 0.25
        scene.collection.objects.link(focus_target)

    focus_target.parent = None

    copy_loc = focus_target.constraints.get("Copy Location")
    if not copy_loc:
        copy_loc = focus_target.constraints.new(type='COPY_LOCATION')
    copy_loc.target = target_mesh
    copy_loc.use_offset = True
    focus_target.location.z = midpoint_z_offset

    # 2. Camera Setup
    cam_data = bpy.data.cameras.get("Vertical_Reel_Cam_Data")
    if not cam_data:
        cam_data = bpy.data.cameras.new("Vertical_Reel_Cam_Data")
    cam_data.lens = 45.0
    cam_data.sensor_fit = 'VERTICAL'
    cam_data.clip_end = 100.0

    cam_obj = bpy.data.objects.get("Vertical_Reel_Camera")
    if not cam_obj:
        cam_obj = bpy.data.objects.new("Vertical_Reel_Camera", cam_data)
        scene.collection.objects.link(cam_obj)

    scene.camera = cam_obj
    cam_obj.parent = None

    # 3. Track To Constraint
    track = cam_obj.constraints.get("Track To")
    if not track:
        track = cam_obj.constraints.new(type='TRACK_TO')
    track.target = focus_target
    track.track_axis = 'TRACK_NEGATIVE_Z'
    track.up_axis = 'UP_Y'

    # Helper: Test if all mesh vertices fall within viewport margins [margin, 1.0 - margin]
    def check_mesh_in_frame(frame_num, cam_location, margin=0.08):
        cam_obj.location = cam_location
        scene.frame_set(int(frame_num))
        bpy.context.view_layer.update()

        depsgraph = bpy.context.evaluated_depsgraph_get()
        eval_mesh_obj = target_mesh.evaluated_get(depsgraph)
        mesh_data = eval_mesh_obj.data

        # Sample mesh vertices (step sampling for performance)
        vert_step = max(1, len(mesh_data.vertices) // 300)
        world_matrix = target_mesh.matrix_world

        min_x, max_x = 1.0, 0.0
        min_y, max_y = 1.0, 0.0

        for i in range(0, len(mesh_data.vertices), vert_step):
            v_world = world_matrix @ mesh_data.vertices[i].co
            co_2d = world_to_camera_view(scene, cam_obj, v_world)

            if co_2d.x < min_x: min_x = co_2d.x
            if co_2d.x > max_x: max_x = co_2d.x
            if co_2d.y < min_y: min_y = co_2d.y
            if co_2d.y > max_y: max_y = co_2d.y

        # Check if bounds stay within safe viewport margins
        in_bounds = (min_x >= margin and max_x <= (1.0 - margin) and
                     min_y >= margin and max_y <= (1.0 - margin))

        return in_bounds, min_x, max_x, min_y, max_y

    # Helper: Find minimum camera distance needed to fit mesh at a given angle/position
    def find_fit_distance(frame_num, angle_rad, base_height_offset, initial_dist=6.0, margin=0.08):
        scene.frame_set(int(frame_num))
        bpy.context.view_layer.update()

        center_pos = focus_target.matrix_world.translation.copy()
        heading = target_rig.matrix_world.to_euler('XYZ').z
        view_angle = heading + angle_rad

        dist = initial_dist
        # Iteratively step back until all vertices are inside camera view
        for _ in range(30):
            test_x = center_pos.x + dist * math.sin(view_angle)
            test_y = center_pos.y - dist * math.cos(view_angle)
            test_z = center_pos.z + base_height_offset

            fit, min_x, max_x, min_y, max_y = check_mesh_in_frame(frame_num, (test_x, test_y, test_z), margin=margin)
            if fit:
                break
            dist += 0.4  # Step back further

        return dist, center_pos, view_angle

    # 4. Solve Distance for Walk Phase (Sampling key frames across walk cycle)
    max_walk_dist = 6.0
    walk_frames_to_sample = range(1, turn_start + 1, 6)

    for f in walk_frames_to_sample:
        req_d, _, _ = find_fit_distance(f, angle_rad=math.radians(-45.0), base_height_offset=0.0, initial_dist=max_walk_dist, margin=0.10)
        if req_d > max_walk_dist:
            max_walk_dist = req_d

    # Frame 1: Walk Start
    scene.frame_set(1)
    bpy.context.view_layer.update()
    c1 = focus_target.matrix_world.translation.copy()
    h1 = target_rig.matrix_world.to_euler('XYZ').z + math.radians(-45.0)
    w_x1 = c1.x + max_walk_dist * math.sin(h1)
    w_y1 = c1.y - max_walk_dist * math.cos(h1)
    w_z1 = c1.z

    # Frame turn_start: Walk End / Turn Start
    scene.frame_set(turn_start)
    bpy.context.view_layer.update()
    cts = focus_target.matrix_world.translation.copy()
    hts = target_rig.matrix_world.to_euler('XYZ').z + math.radians(-45.0)
    w_xts = cts.x + max_walk_dist * math.sin(hts)
    w_yts = cts.y - max_walk_dist * math.cos(hts)
    w_zts = cts.z

    # 5. Solve Distance for Unfurl Phase
    unfurl_frames_to_sample = range(turn_end_frame, unfurl_end_frame + 1, 6)
    max_unfurl_dist = max_walk_dist

    for f in unfurl_frames_to_sample:
        req_d, _, _ = find_fit_distance(f, angle_rad=math.radians(0.0), base_height_offset=0.0, initial_dist=max_unfurl_dist, margin=0.10)
        if req_d > max_unfurl_dist:
            max_unfurl_dist = req_d

    scene.frame_set(turn_end_frame)
    bpy.context.view_layer.update()
    cte = focus_target.matrix_world.translation.copy()
    hte = target_rig.matrix_world.to_euler('XYZ').z + math.radians(0.0)
    f_x = cte.x + max_unfurl_dist * math.sin(hte)
    f_y = cte.y - max_unfurl_dist * math.cos(hte)
    f_z = cte.z

    # Apply Keyframes
    if cam_obj.animation_data:
        cam_obj.animation_data_clear()

    cam_obj.location = (w_x1, w_y1, w_z1)
    cam_obj.keyframe_insert(data_path="location", frame=1)

    cam_obj.location = (w_xts, w_yts, w_zts)
    cam_obj.keyframe_insert(data_path="location", frame=turn_start)

    cam_obj.location = (f_x, f_y, f_z)
    cam_obj.keyframe_insert(data_path="location", frame=turn_end_frame)

    cam_obj.location = (f_x, f_y, f_z)
    cam_obj.keyframe_insert(data_path="location", frame=unfurl_end_frame)

    # Smooth Bezier handle transitions
    for fc in get_object_fcurves(cam_obj):
        for kp in fc.keyframe_points:
            kp.interpolation = 'BEZIER'

    scene.frame_set(1)
    return cam_obj, focus_target


# ==============================================================================
# GEOMETRY NODES GROUND, GRASS & SKY DOME SETUP
# ==============================================================================

def ensure_world_sphere_size(cam_obj, ground_obj, safety_multiplier=2.5):
    """
    Dynamically measures camera trajectory across all frames and resizes
    the sky sphere, ground mesh, and camera far clipping plane accordingly.
    """
    if not cam_obj or not ground_obj:
        return

    scene = bpy.context.scene
    max_cam_dist = 0.0

    # Sample camera distance relative to world origin across the animation range
    for frame in range(scene.frame_start, scene.frame_end + 1, 5):
        scene.frame_set(frame)
        bpy.context.view_layer.update()

        cam_dist = cam_obj.matrix_world.translation.length
        if cam_dist > max_cam_dist:
            max_cam_dist = cam_dist

    # Calculate target radius with safety factor
    world_radius = max(25.0, max_cam_dist * safety_multiplier)

    # 1. Update Sky Dome Radius in Geometry Nodes Tree
    mod = ground_obj.modifiers.get("Grass_Ground_GN")
    if mod and mod.node_group:
        for node in mod.node_group.nodes:
            if node.bl_idname == "GeometryNodeMeshIcoSphere":
                node.inputs["Radius"].default_value = world_radius

    # 2. Update Ground Circle Mesh Dimensions to match new world scale
    if ground_obj.type == 'MESH':
        bpy.context.view_layer.objects.active = ground_obj
        bpy.ops.object.mode_set(mode='OBJECT')

        # Scale ground circle diameter to match sky dome diameter
        ground_scale = world_radius / 12.5
        ground_obj.scale = (ground_scale, ground_scale, 1.0)
        bpy.ops.object.transform_apply(location=False, rotation=False, scale=True)

    # 3. Extend Camera Clip End so sky dome is never clipped out of view
    cam_obj.data.clip_end = max(cam_obj.data.clip_end, world_radius * 2.5)

    scene.frame_set(1)
    bpy.context.view_layer.update()


def create_grass_material(grass_dir):
    mat = bpy.data.materials.new(name="Grass_PBR_Material")
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    out_node = nodes.new("ShaderNodeOutputMaterial")
    bsdf = nodes.new("ShaderNodeBsdfPrincipled")
    links.new(bsdf.outputs["BSDF"], out_node.inputs["Surface"])

    col_files = glob.glob(os.path.join(grass_dir, "*Color*.jpg")) + glob.glob(os.path.join(grass_dir, "*JPG*.jpg"))
    if col_files:
        tex_col = nodes.new("ShaderNodeTexImage")
        tex_col.image = bpy.data.images.load(col_files[0])
        links.new(tex_col.outputs["Color"], bsdf.inputs["Base Color"])

    rough_files = glob.glob(os.path.join(grass_dir, "*Roughness*.jpg"))
    if rough_files:
        tex_rough = nodes.new("ShaderNodeTexImage")
        tex_rough.image = bpy.data.images.load(rough_files[0])
        tex_rough.image.colorspace_settings.name = 'Non-Color'
        links.new(tex_rough.outputs["Color"], bsdf.inputs["Roughness"])

    norm_files = glob.glob(os.path.join(grass_dir, "*Normal*.jpg"))
    if norm_files:
        tex_norm = nodes.new("ShaderNodeTexImage")
        tex_norm.image = bpy.data.images.load(norm_files[0])
        tex_norm.image.colorspace_settings.name = 'Non-Color'

        normal_map_node = nodes.new("ShaderNodeNormalMap")
        links.new(tex_norm.outputs["Color"], normal_map_node.inputs["Color"])
        links.new(normal_map_node.outputs["Normal"], bsdf.inputs["Normal"])

    return mat


def create_sky_material(sky_hdri_path):
    mat = bpy.data.materials.new(name="Sky_Dome_Material")
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    out_node = nodes.new("ShaderNodeOutputMaterial")
    emission = nodes.new("ShaderNodeEmission")
    emission.inputs["Strength"].default_value = 1.0

    links.new(emission.outputs["Emission"], out_node.inputs["Surface"])

    if os.path.exists(sky_hdri_path):
        tex_env = nodes.new("ShaderNodeTexEnvironment")
        tex_env.image = bpy.data.images.load(sky_hdri_path)
        links.new(tex_env.outputs["Color"], emission.inputs["Color"])

    return mat


def setup_grass_ground_geometry_nodes(grass_dir, sky_hdri_path, ground_z_offset=-0.2):
    grass_coll = bpy.data.collections.get("Grass_Artifacts_Collection")
    if not grass_coll:
        grass_coll = bpy.data.collections.new("Grass_Artifacts_Collection")
        bpy.context.scene.collection.children.link(grass_coll)

    grass_mat = create_grass_material(grass_dir)
    sky_mat = create_sky_material(sky_hdri_path)

    mesh_files = glob.glob(os.path.join(grass_dir, "*.obj")) + glob.glob(os.path.join(grass_dir, "*.fbx"))

    if mesh_files:
        for fpath in mesh_files:
            if fpath.endswith(".obj"):
                bpy.ops.wm.obj_import(filepath=fpath)
            elif fpath.endswith(".fbx"):
                bpy.ops.import_scene.fbx(filepath=fpath)

            for o in bpy.context.selected_objects:
                grass_coll.objects.link(o)
                if o.name in bpy.context.scene.collection.objects:
                    bpy.context.scene.collection.objects.unlink(o)

        for obj in list(grass_coll.objects):
            if obj.dimensions.z > 0.5:
                grass_coll.objects.unlink(obj)
                bpy.data.objects.remove(obj, do_unlink=True)
    else:
        for i in range(3):
            m_data = bpy.data.meshes.new(f"Grass_Blade_Mesh_{i}")
            g_obj = bpy.data.objects.new(f"Grass_Blade_Obj_{i}", m_data)
            grass_coll.objects.link(g_obj)

            verts = [(0, 0, 0), (0.05, 0, 0), (0.04, 0, 0.15), (0.01, 0, 0.25), (0, 0, 0.3)]
            faces = [(0, 1, 2), (1, 2, 3), (2, 3, 4)]
            m_data.from_pydata(verts, [], faces)
            m_data.update()

            if g_obj.data.materials:
                g_obj.data.materials[0] = grass_mat
            else:
                g_obj.data.materials.append(grass_mat)

            g_obj.location = (100, 100, 0)

    bpy.ops.mesh.primitive_circle_add(radius=12.5, vertices=64, fill_type='NGON', location=(0, 0, ground_z_offset))
    ground_obj = bpy.context.active_object
    ground_obj.name = "Ground_Mesh_Object"

    vg = ground_obj.vertex_groups.new(name="Grass_Weight_Group")
    all_vert_indices = [v.index for v in ground_obj.data.vertices]
    vg.add(all_vert_indices, 1.0, 'REPLACE')
    ground_obj.data.materials.append(grass_mat)
    ground_obj.data.materials.append(sky_mat)

    gn_group = bpy.data.node_groups.new("Grass_Ground_GN", type="GeometryNodeTree")
    gn_group.interface.new_socket(name="Geometry", in_out="INPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Geometry", in_out="OUTPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Density Mask", in_out="INPUT", socket_type="NodeSocketFloat")

    nodes = gn_group.nodes
    links = gn_group.links
    out_node = nodes.new("NodeGroupOutput")
    in_node = nodes.new("NodeGroupInput")

    distribute_node = nodes.new("GeometryNodeDistributePointsOnFaces")
    distribute_node.distribute_method = 'POISSON'
    distribute_node.inputs["Distance Min"].default_value = 0.02
    distribute_node.inputs["Density Max"].default_value = 300.0
    links.new(in_node.outputs["Geometry"], distribute_node.inputs["Mesh"])

    noise_node = nodes.new("ShaderNodeTexNoise")
    noise_node.inputs["Scale"].default_value = 2.5
    noise_node.inputs["Detail"].default_value = 2.0

    color_ramp = nodes.new("ShaderNodeValToRGB")
    color_ramp.color_ramp.elements[0].position = 0.20
    color_ramp.color_ramp.elements[1].position = 0.80
    links.new(noise_node.outputs["Fac"], color_ramp.inputs["Fac"])

    math_weight = nodes.new("ShaderNodeMath")
    math_weight.operation = "MULTIPLY"
    links.new(color_ramp.outputs["Color"], math_weight.inputs[0])
    links.new(in_node.outputs["Density Mask"], math_weight.inputs[1])

    links.new(math_weight.outputs["Value"], distribute_node.inputs["Density Factor"])

    instance_node = nodes.new("GeometryNodeInstanceOnPoints")
    instance_node.inputs["Pick Instance"].default_value = True
    links.new(distribute_node.outputs["Points"], instance_node.inputs["Points"])

    coll_info = nodes.new("GeometryNodeCollectionInfo")
    coll_info.inputs["Collection"].default_value = grass_coll
    coll_info.inputs["Separate Children"].default_value = True
    coll_info.inputs["Reset Children"].default_value = True
    links.new(coll_info.outputs["Instances"], instance_node.inputs["Instance"])

    rand_scale = nodes.new("FunctionNodeRandomValue")
    rand_scale.data_type = 'FLOAT'
    rand_scale.inputs["Min"].default_value = 0.25
    rand_scale.inputs["Max"].default_value = 0.50

    combine_scale = nodes.new("ShaderNodeCombineXYZ")
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["X"])
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["Y"])
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["Z"])
    links.new(combine_scale.outputs["Vector"], instance_node.inputs["Scale"])

    rand_rot_z = nodes.new("FunctionNodeRandomValue")
    rand_rot_z.data_type = 'FLOAT'
    rand_rot_z.inputs["Min"].default_value = 0.0
    rand_rot_z.inputs["Max"].default_value = math.radians(360.0)

    combine_rot = nodes.new("ShaderNodeCombineXYZ")
    links.new(rand_rot_z.outputs["Value"], combine_rot.inputs["Z"])
    links.new(combine_rot.outputs["Vector"], instance_node.inputs["Rotation"])

    sky_sphere = nodes.new("GeometryNodeMeshIcoSphere")
    sky_sphere.inputs["Radius"].default_value = 12.5
    sky_sphere.inputs["Subdivisions"].default_value = 4

    flip_faces = nodes.new("GeometryNodeFlipFaces")
    links.new(sky_sphere.outputs["Mesh"], flip_faces.inputs["Mesh"])

    transform_sky = nodes.new("GeometryNodeTransform")
    transform_sky.inputs["Translation"].default_value = (0.0, 0.0, ground_z_offset)
    links.new(flip_faces.outputs["Mesh"], transform_sky.inputs["Geometry"])

    set_sky_mat = nodes.new("GeometryNodeSetMaterial")
    set_sky_mat.inputs["Material"].default_value = sky_mat
    links.new(transform_sky.outputs["Geometry"], set_sky_mat.inputs["Geometry"])

    join_geo = nodes.new("GeometryNodeJoinGeometry")
    links.new(in_node.outputs["Geometry"], join_geo.inputs["Geometry"])
    links.new(instance_node.outputs["Instances"], join_geo.inputs["Geometry"])
    links.new(set_sky_mat.outputs["Geometry"], join_geo.inputs["Geometry"])

    links.new(join_geo.outputs["Geometry"], out_node.inputs["Geometry"])

    mod = ground_obj.modifiers.new(name="Grass_Ground_GN", type='NODES')
    mod.node_group = gn_group

    for item in gn_group.interface.items_tree:
        if getattr(item, "in_out", None) == "INPUT" and item.name == "Density Mask":
            mod[item.identifier + "_attribute_name"] = "Grass_Weight_Group"
            mod[item.identifier + "_use_attribute"] = True

    return ground_obj


# ==============================================================================
# SCROLL MATERIAL & GEOMETRY NODES SETUP
# ==============================================================================

def create_image_scroll_material(image_path):
    mat_name = "Scroll_Texture_Mat"
    mat = bpy.data.materials.get(mat_name)
    if mat:
        bpy.data.materials.remove(mat)

    mat = bpy.data.materials.new(name=mat_name)
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    output_node = nodes.new("ShaderNodeOutputMaterial")
    bsdf = nodes.new("ShaderNodeBsdfPrincipled")
    bsdf.inputs["Roughness"].default_value = 0.5

    tex_img = nodes.new("ShaderNodeTexImage")
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        tex_img.image = img
        tex_img.extension = 'REPEAT'

    uv_map_node = nodes.new("ShaderNodeUVMap")
    uv_map_node.uv_map = "UVMap"

    links.new(uv_map_node.outputs["UV"], tex_img.inputs["Vector"])
    links.new(tex_img.outputs["Color"], bsdf.inputs["Base Color"])
    links.new(bsdf.outputs["BSDF"], output_node.inputs["Surface"])

    return mat, tex_img.image if os.path.exists(image_path) else None


def create_pure_geometry_node_scroll(image_path=""):
    gn_group = bpy.data.node_groups.new("Scroll_Pure_GN", type="GeometryNodeTree")

    gn_group.interface.new_socket(name="Geometry", in_out="OUTPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Character Mesh", in_out="INPUT", socket_type="NodeSocketObject")

    sock_r_vg = gn_group.interface.new_socket(name="Right Hand Group", in_out="INPUT", socket_type="NodeSocketString")
    sock_r_vg.default_value = "mixamorig:RightHand"

    sock_l_vg = gn_group.interface.new_socket(name="Left Hand Group", in_out="INPUT", socket_type="NodeSocketString")
    sock_l_vg.default_value = "mixamorig:LeftHand"

    sock_offset = gn_group.interface.new_socket(name="Forward Offset", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_offset.default_value = 0.60

    sock_unfurl = gn_group.interface.new_socket(name="Unfurl", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_unfurl.default_value = 0.0

    sock_width = gn_group.interface.new_socket(name="Scroll Width", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_width.default_value = 0.35

    nodes = gn_group.nodes
    links = gn_group.links

    in_node = nodes.new("NodeGroupInput")
    out_node = nodes.new("NodeGroupOutput")

    char_info = nodes.new("GeometryNodeObjectInfo")
    char_info.transform_space = "RELATIVE"
    links.new(in_node.outputs["Character Mesh"], char_info.inputs["Object"])

    pos_node = nodes.new("GeometryNodeInputPosition")

    named_r = nodes.new("GeometryNodeInputNamedAttribute")
    named_r.data_type = "FLOAT"
    links.new(in_node.outputs["Right Hand Group"], named_r.inputs["Name"])

    cmp_r = nodes.new("FunctionNodeCompare")
    cmp_r.data_type = "FLOAT"
    cmp_r.operation = "GREATER_THAN"
    cmp_r.inputs[1].default_value = 0.001
    links.new(named_r.outputs["Attribute"], cmp_r.inputs[0])

    stat_r = nodes.new("GeometryNodeAttributeStatistic")
    stat_r.data_type = "FLOAT_VECTOR"
    links.new(char_info.outputs["Geometry"], stat_r.inputs["Geometry"])
    links.new(cmp_r.outputs["Result"], stat_r.inputs["Selection"])
    links.new(pos_node.outputs["Position"], stat_r.inputs["Attribute"])

    named_l = nodes.new("GeometryNodeInputNamedAttribute")
    named_l.data_type = "FLOAT"
    links.new(in_node.outputs["Left Hand Group"], named_l.inputs["Name"])

    cmp_l = nodes.new("FunctionNodeCompare")
    cmp_l.data_type = "FLOAT"
    cmp_l.operation = "GREATER_THAN"
    cmp_l.inputs[1].default_value = 0.001
    links.new(named_l.outputs["Attribute"], cmp_l.inputs[0])

    stat_l = nodes.new("GeometryNodeAttributeStatistic")
    stat_l.data_type = "FLOAT_VECTOR"
    links.new(char_info.outputs["Geometry"], stat_l.inputs["Geometry"])
    links.new(cmp_l.outputs["Result"], stat_l.inputs["Selection"])
    links.new(pos_node.outputs["Position"], stat_l.inputs["Attribute"])

    neg_offset = nodes.new("ShaderNodeMath")
    neg_offset.operation = "MULTIPLY"
    neg_offset.inputs[1].default_value = -1.0
    links.new(in_node.outputs["Forward Offset"], neg_offset.inputs[0])

    forward_vec = nodes.new("ShaderNodeCombineXYZ")
    forward_vec.inputs["X"].default_value = 0.0
    links.new(neg_offset.outputs["Value"], forward_vec.inputs["Y"])
    forward_vec.inputs["Z"].default_value = 0.0

    start_pos = nodes.new("ShaderNodeVectorMath")
    start_pos.operation = "ADD"
    links.new(stat_r.outputs["Mean"], start_pos.inputs[0])
    links.new(forward_vec.outputs["Vector"], start_pos.inputs[1])

    target_pos = nodes.new("ShaderNodeVectorMath")
    target_pos.operation = "ADD"
    links.new(stat_l.outputs["Mean"], target_pos.inputs[0])
    links.new(forward_vec.outputs["Vector"], target_pos.inputs[1])

    delta_vec = nodes.new("ShaderNodeVectorMath")
    delta_vec.operation = "SUBTRACT"
    links.new(target_pos.outputs["Vector"], delta_vec.inputs[0])
    links.new(start_pos.outputs["Vector"], delta_vec.inputs[1])

    dist_node = nodes.new("ShaderNodeVectorMath")
    dist_node.operation = "DISTANCE"
    links.new(start_pos.outputs["Vector"], dist_node.inputs[0])

    scale_vec = nodes.new("ShaderNodeVectorMath")
    scale_vec.operation = "SCALE"
    links.new(delta_vec.outputs["Vector"], scale_vec.inputs["Vector"])
    links.new(in_node.outputs["Unfurl"], scale_vec.inputs["Scale"])

    end_pos = nodes.new("ShaderNodeVectorMath")
    end_pos.operation = "ADD"
    links.new(start_pos.outputs["Vector"], end_pos.inputs[0])
    links.new(scale_vec.outputs["Vector"], end_pos.inputs[1])

    links.new(end_pos.outputs["Vector"], dist_node.inputs[1])

    curve_line = nodes.new("GeometryNodeCurvePrimitiveLine")
    links.new(start_pos.outputs["Vector"], curve_line.inputs["Start"])
    links.new(end_pos.outputs["Vector"], curve_line.inputs["End"])

    set_tilt = nodes.new("GeometryNodeSetCurveTilt")
    set_tilt.inputs["Tilt"].default_value = 1.5708
    links.new(curve_line.outputs["Curve"], set_tilt.inputs["Curve"])

    spline_param_main = nodes.new("GeometryNodeSplineParameter")
    store_u = nodes.new("GeometryNodeStoreNamedAttribute")
    store_u.inputs["Name"].default_value = "U_Factor"
    store_u.data_type = 'FLOAT'
    store_u.domain = 'POINT'
    links.new(set_tilt.outputs["Curve"], store_u.inputs["Geometry"])
    links.new(spline_param_main.outputs["Factor"], store_u.inputs["Value"])

    half_width_neg = nodes.new("ShaderNodeMath")
    half_width_neg.operation = "MULTIPLY"
    half_width_neg.inputs[1].default_value = -0.5
    links.new(in_node.outputs["Scroll Width"], half_width_neg.inputs[0])

    half_width_pos = nodes.new("ShaderNodeMath")
    half_width_pos.operation = "MULTIPLY"
    half_width_pos.inputs[1].default_value = 0.5
    links.new(in_node.outputs["Scroll Width"], half_width_pos.inputs[0])

    combine_start = nodes.new("ShaderNodeCombineXYZ")
    links.new(half_width_neg.outputs["Value"], combine_start.inputs["X"])

    combine_end = nodes.new("ShaderNodeCombineXYZ")
    links.new(half_width_pos.outputs["Value"], combine_end.inputs["X"])

    profile_line = nodes.new("GeometryNodeCurvePrimitiveLine")
    links.new(combine_start.outputs["Vector"], profile_line.inputs["Start"])
    links.new(combine_end.outputs["Vector"], profile_line.inputs["End"])

    spline_param_prof = nodes.new("GeometryNodeSplineParameter")
    store_v = nodes.new("GeometryNodeStoreNamedAttribute")
    store_v.inputs["Name"].default_value = "V_Factor"
    store_v.data_type = 'FLOAT'
    store_v.domain = 'POINT'
    links.new(profile_line.outputs["Curve"], store_v.inputs["Geometry"])
    links.new(spline_param_prof.outputs["Factor"], store_v.inputs["Value"])

    curve_to_mesh = nodes.new("GeometryNodeCurveToMesh")
    links.new(store_u.outputs["Geometry"], curve_to_mesh.inputs["Curve"])
    links.new(store_v.outputs["Geometry"], curve_to_mesh.inputs["Profile Curve"])

    img_aspect = 1.0
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        if img.size[1] > 0:
            img_aspect = img.size[0] / img.size[1]

    h_times_aspect = nodes.new("ShaderNodeMath")
    h_times_aspect.operation = "MULTIPLY"
    h_times_aspect.inputs[1].default_value = img_aspect
    links.new(in_node.outputs["Scroll Width"], h_times_aspect.inputs[0])

    tile_count = nodes.new("ShaderNodeMath")
    tile_count.operation = "DIVIDE"
    links.new(dist_node.outputs["Value"], tile_count.inputs[0])
    links.new(h_times_aspect.outputs["Value"], tile_count.inputs[1])

    read_u = nodes.new("GeometryNodeInputNamedAttribute")
    read_u.inputs["Name"].default_value = "U_Factor"
    read_u.data_type = 'FLOAT'

    scaled_u = nodes.new("ShaderNodeMath")
    scaled_u.operation = "MULTIPLY"
    links.new(read_u.outputs["Attribute"], scaled_u.inputs[0])
    links.new(tile_count.outputs["Value"], scaled_u.inputs[1])

    read_v = nodes.new("GeometryNodeInputNamedAttribute")
    read_v.inputs["Name"].default_value = "V_Factor"
    read_v.data_type = 'FLOAT'

    flip_v = nodes.new("ShaderNodeMath")
    flip_v.operation = "SUBTRACT"
    flip_v.inputs[0].default_value = 1.0
    links.new(read_v.outputs["Attribute"], flip_v.inputs[1])

    combine_uv = nodes.new("ShaderNodeCombineXYZ")
    links.new(scaled_u.outputs["Value"], combine_uv.inputs["X"])
    links.new(flip_v.outputs["Value"], combine_uv.inputs["Y"])

    store_uv = nodes.new("GeometryNodeStoreNamedAttribute")
    store_uv.inputs["Name"].default_value = "UVMap"
    store_uv.data_type = 'FLOAT2'
    store_uv.domain = 'CORNER'
    links.new(curve_to_mesh.outputs["Mesh"], store_uv.inputs["Geometry"])
    links.new(combine_uv.outputs["Vector"], store_uv.inputs["Value"])

    set_material = nodes.new("GeometryNodeSetMaterial")
    scroll_mat, _ = create_image_scroll_material(image_path)
    set_material.inputs["Material"].default_value = scroll_mat
    links.new(store_uv.outputs["Geometry"], set_material.inputs["Geometry"])

    links.new(set_material.outputs["Geometry"], out_node.inputs["Geometry"])

    return gn_group, scroll_mat


def setup_geometry_node_scroll(target_mesh, unfurl_start, unfurl_end, image_path=""):
    mesh_data = bpy.data.meshes.new("Scroll_Procedural_Mesh")
    scroll_obj = bpy.data.objects.new("Scroll_GN_Object", mesh_data)
    bpy.context.scene.collection.objects.link(scroll_obj)

    mod = scroll_obj.modifiers.new("Scroll_GN", type="NODES")
    gn_tree, scroll_mat = create_pure_geometry_node_scroll(image_path=image_path)
    mod.node_group = gn_tree

    scroll_obj.data.materials.append(scroll_mat)

    r_hand_name, l_hand_name = find_hand_vertex_groups(target_mesh)

    unfurl_id = None
    for item in gn_tree.interface.items_tree:
        if getattr(item, "in_out", None) == "INPUT":
            if item.name == "Character Mesh":
                mod[item.identifier] = target_mesh
            elif item.name == "Right Hand Group":
                mod[item.identifier] = r_hand_name
            elif item.name == "Left Hand Group":
                mod[item.identifier] = l_hand_name
            elif item.name == "Forward Offset":
                mod[item.identifier] = 0.60
            elif item.name == "Scroll Width":
                mod[item.identifier] = 0.35
            elif item.name == "Unfurl":
                unfurl_id = item.identifier

    for m in target_mesh.modifiers:
        if m.type == 'ARMATURE':
            m.show_viewport = True
            m.show_render = True
            m.show_in_editmode = True

    scroll_obj.hide_viewport = True
    scroll_obj.hide_render = True
    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=1)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=1)

    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=unfurl_start - 1)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=unfurl_start - 1)

    scroll_obj.hide_viewport = False
    scroll_obj.hide_render = False
    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=unfurl_start)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=unfurl_start)

    for fc in get_object_fcurves(scroll_obj):
        if fc.data_path in ["hide_viewport", "hide_render"]:
            for kp in fc.keyframe_points:
                kp.interpolation = 'CONSTANT'

    if unfurl_id:
        mod[unfurl_id] = 0.001
        mod.keyframe_insert(data_path=f'["{unfurl_id}"]', frame=unfurl_start)

        mod[unfurl_id] = 1.0
        mod.keyframe_insert(data_path=f'["{unfurl_id}"]', frame=unfurl_end)

    return scroll_obj


# ==============================================================================
# MAIN CONTROLLER
# ==============================================================================

def import_and_extract_action(fbx_path, action_name):
    existing_objects = set(bpy.data.objects)
    bpy.ops.import_scene.fbx(filepath=fbx_path)
    bpy.context.view_layer.update()

    imported_objects = set(bpy.data.objects) - existing_objects
    extracted_action = None
    for obj in imported_objects:
        if obj.animation_data and obj.animation_data.action:
            extracted_action = obj.animation_data.action
            extracted_action.name = action_name
            extracted_action.use_fake_user = True
            break
    return extracted_action, imported_objects


def purge_imported_objects(objects_to_remove):
    for obj in objects_to_remove:
        if obj.name in bpy.data.objects:
            if obj.animation_data:
                obj.animation_data_clear()
            for coll in list(obj.users_collection):
                coll.objects.unlink(obj)
    bpy.context.view_layer.update()


def main():
    cleanup_previous_run()

    walk_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Mutant Walking.fbx"
    turn_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Right Turn 90.fbx"
    unfurl_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Mutant Idle.fbx"

    sky_hdri_path = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/horn-koppe_spring_4k.exr"
    output_blend = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/output_scene.blend"

    image_path = "/Users/dobbinp1/Downloads/IMGP6691_psedit.jpg"
    grass_dir = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Grass004_4K-JPG"

    ground_obj = setup_grass_ground_geometry_nodes(grass_dir, sky_hdri_path, ground_z_offset=-0.2)

    CYCLES = 4
    CROSSFADE = 15
    TARGET_DISTANCE = 4.0

    walk_action, walk_objs = import_and_extract_action(walk_fbx, "walk")
    target_rig = next((obj for obj in walk_objs if obj.type == "ARMATURE"), None)
    target_mesh = next((obj for obj in walk_objs if obj.type == "MESH"), None)

    if not target_rig or not target_mesh:
        raise RuntimeError("Armature or Character Mesh not found in walk FBX.")

    travel_axis_idx, motion_sign = detect_natural_forward_axis(walk_action, target_rig)
    initial_heading = math.radians(90.0) if travel_axis_idx == 0 else 0.0
    target_rig.rotation_euler[2] = initial_heading

    action_len = int(walk_action.frame_range[1] - walk_action.frame_range[0])
    if action_len <= 0:
        action_len = 30

    anim_data = target_rig.animation_data_create()

    walk_track = anim_data.nla_tracks.new()
    walk_track.name = "Walk_Track"

    current_frame = 1
    for i in range(CYCLES):
        strip = walk_track.strips.new(name=f"Walk_{i+1}", start=int(current_frame), action=walk_action)
        strip.blend_type = "REPLACE"
        current_frame += action_len

    walk_end_frame = int(current_frame)

    turn_action, turn_objs = import_and_extract_action(turn_fbx, "rightturn")
    purge_imported_objects(turn_objs)

    turn_track = anim_data.nla_tracks.new()
    turn_track.name = "RightTurn_Track"

    turn_start = int(walk_end_frame - CROSSFADE)
    turn_strip = turn_track.strips.new(name="RightTurn", start=turn_start, action=turn_action)
    turn_strip.blend_type = "REPLACE"
    turn_strip.blend_in = CROSSFADE
    turn_strip.use_auto_blend = True
    turn_strip.extrapolation = "HOLD"
    turn_end_frame = int(turn_strip.frame_end)

    unfurl_action, unfurl_objs = import_and_extract_action(unfurl_fbx, "unfurl")
    purge_imported_objects(unfurl_objs)

    unfurl_track = anim_data.nla_tracks.new()
    unfurl_track.name = "Unfurl_Track"

    unfurl_start = int(turn_end_frame - CROSSFADE)
    unfurl_strip = unfurl_track.strips.new(name="Unfurl", start=unfurl_start, action=unfurl_action)
    unfurl_strip.blend_type = "REPLACE"
    unfurl_strip.blend_in = CROSSFADE
    unfurl_strip.use_auto_blend = True
    unfurl_strip.extrapolation = "HOLD"
    unfurl_end_frame = int(unfurl_strip.frame_end)

    root_action = bpy.data.actions.new(name="World_Root_Motion")
    target_rig.animation_data.action = root_action

    loc_fcurve = root_action.fcurve_ensure_for_datablock(datablock=target_rig, data_path="location", index=travel_axis_idx)
    final_target_value = TARGET_DISTANCE * motion_sign

    k_loc1 = loc_fcurve.keyframe_points.insert(frame=1.0, value=0.0)
    k_loc2 = loc_fcurve.keyframe_points.insert(frame=float(turn_start), value=final_target_value)
    k_loc3 = loc_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=final_target_value)

    k_loc1.interpolation = "LINEAR"
    k_loc2.interpolation = "CONSTANT"
    k_loc3.interpolation = "CONSTANT"

    rot_fcurve = root_action.fcurve_ensure_for_datablock(datablock=target_rig, data_path="rotation_euler", index=2)
    target_heading = initial_heading - math.radians(90.0)

    k_rot1 = rot_fcurve.keyframe_points.insert(frame=1.0, value=initial_heading)
    k_rot2 = rot_fcurve.keyframe_points.insert(frame=float(turn_start), value=initial_heading)
    k_rot3 = rot_fcurve.keyframe_points.insert(frame=float(turn_end_frame), value=target_heading)
    k_rot4 = rot_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=target_heading)

    k_rot1.interpolation = "CONSTANT"
    k_rot2.interpolation = "BEZIER"
    k_rot3.interpolation = "CONSTANT"
    k_rot4.interpolation = "CONSTANT"

    root_track = anim_data.nla_tracks.new()
    root_track.name = "World_Root_Track"
    root_strip = root_track.strips.new(name="Root_Motion_Strip", start=1, action=root_action)
    root_strip.blend_type = "REPLACE"
    root_strip.extrapolation = "HOLD"

    anim_data.action = None

    setup_geometry_node_scroll(
        target_mesh=target_mesh,
        unfurl_start=unfurl_start,
        unfurl_end=unfurl_end_frame,
        image_path=image_path,
    )

    # Modern Instagram Reel Vertical Camera Setup
    cam_obj, _ = setup_modern_reel_camera(
        target_rig=target_rig,
        target_mesh=target_mesh,
        turn_start=turn_start,
        turn_end_frame=turn_end_frame,
        unfurl_end_frame=unfurl_end_frame
    )

    bpy.context.scene.frame_start = 1
    bpy.context.scene.frame_end = unfurl_end_frame

    # Dynamically scale world sphere, ground plane, grass scatter area, and camera far clip plane
    ensure_world_sphere_size(cam_obj, ground_obj, safety_multiplier=2.5)

    bpy.ops.wm.save_as_mainfile(filepath=output_blend)
    print(f"\n[Success] Dynamic world sizing & camera integrated. Saved to: {output_blend}")


main()
```

```
import math
import mathutils
import os
import glob
import bpy
from bpy_extras import anim_utils
from bpy_extras.object_utils import world_to_camera_view

# ==============================================================================
# CLEANUP & HELPERS
# ==============================================================================

def cleanup_previous_run():
    """Clears old scroll objects, ground, node groups, camera anchors, collections, and orphaned data."""
    target_names = [
        "Master_NLA_Control",
        "Scroll_GN_Object",
        "Scroll_Start_Pin",
        "Scroll_End_Pin",
        "Scroll_Master_Container",
        "HDRI_Ground_Plane",
        "Ground_Mesh_Object",
        "Camera_Focus_Target",
        "Vertical_Reel_Camera"
    ]
    for obj in list(bpy.data.objects):
        if obj.name in target_names or (obj.name.startswith("mixamorig") and obj.type != "ARMATURE" and obj.type != "MESH"):
            if obj.animation_data:
                obj.animation_data_clear()
            bpy.data.objects.remove(obj, do_unlink=True)

    for action in list(bpy.data.actions):
        if action.users == 0 or action.use_fake_user:
            action.use_fake_user = False
            bpy.data.actions.remove(action)

    for ng in list(bpy.data.node_groups):
        if ng.name in ["Scroll_Unfurl_GN", "Scroll_Pure_GN", "HDRI_Ground_GN", "Grass_Ground_GN"]:
            bpy.data.node_groups.remove(ng)

    for coll_name in ["Scroll_Master_Container", "Grass_Artifacts_Collection"]:
        if coll_name in bpy.data.collections:
            coll = bpy.data.collections[coll_name]
            for obj in list(coll.objects):
                bpy.data.objects.remove(obj, do_unlink=True)
            bpy.data.collections.remove(coll)

    bpy.context.view_layer.update()


def get_slot_fcurves(action, anim_target_object):
    fcurves = []
    if not action:
        return fcurves

    anim_data = getattr(anim_target_object, "animation_data", None)
    if anim_data and getattr(anim_data, "action_slot", None):
        slot = anim_data.action_slot
        channelbag = anim_utils.action_get_channelbag_for_slot(action, slot)
        if channelbag and hasattr(channelbag, "fcurves"):
            return list(channelbag.fcurves)

    if hasattr(action, "slots"):
        for slot in action.slots:
            for cb in getattr(slot, "channelbags", []):
                fcurves.extend(getattr(cb, "fcurves", []))
        if fcurves:
            return fcurves

    if hasattr(action, "fcurves"):
        return list(action.fcurves)

    return fcurves


def get_object_fcurves(obj):
    if not obj or not obj.animation_data or not obj.animation_data.action:
        return []
    action = obj.animation_data.action
    return get_slot_fcurves(action, obj)


def detect_natural_forward_axis(action, rig_object):
    fcurves = get_slot_fcurves(action, rig_object)
    max_delta = 0.0
    best_axis = 1
    net_delta = 0.0

    start_frame = action.frame_range[0]
    end_frame = action.frame_range[1]

    for fc in fcurves:
        if fc.data_path.endswith("location") and fc.array_index in [0, 1, 2]:
            start_val = fc.evaluate(start_frame)
            end_val = fc.evaluate(end_frame)
            delta = end_val - start_val
            if abs(delta) > abs(max_delta):
                max_delta = delta
                best_axis = fc.array_index
                net_delta = delta

    direction_sign = 1.0 if net_delta >= 0 else -1.0
    return best_axis, direction_sign


def find_hand_vertex_groups(target_mesh):
    r_name = "mixamorig:RightHand"
    l_name = "mixamorig:LeftHand"

    if target_mesh and hasattr(target_mesh, "vertex_groups"):
        for vg in target_mesh.vertex_groups:
            vg_lower = vg.name.lower()
            if "righthand" in vg_lower or "hand_r" in vg_lower or "hand.r" in vg_lower:
                r_name = vg.name
            elif "lefthand" in vg_lower or "hand_l" in vg_lower or "hand.l" in vg_lower:
                l_name = vg.name

    return r_name, l_name


# ==============================================================================
# UNPARENTED MATRIX-EVALUATED VERTICAL CAMERA SETUP
# ==============================================================================

def setup_modern_reel_camera(target_rig, target_mesh, turn_start, turn_end_frame, unfurl_end_frame):
    """
    Creates a 9:16 Instagram Reel Camera:
    - Uses exact 2D camera viewport projection (world_to_camera_view) to test
      every vertex across the animation, ensuring 100% of the mesh fits inside
      the camera frame during both walking and unfurl phases.
    """
    scene = bpy.context.scene
    scene.render.resolution_x = 1080
    scene.render.resolution_y = 1920
    scene.render.resolution_percentage = 100

    # 1. Focus Target Null
    focus_target = bpy.data.objects.get("Camera_Focus_Target")
    if not focus_target:
        focus_target = bpy.data.objects.new("Camera_Focus_Target", None)
        focus_target.empty_display_type = 'PLAIN_AXES'
        focus_target.empty_display_size = 0.25
        scene.collection.objects.link(focus_target)

    focus_target.parent = None

    copy_loc = focus_target.constraints.get("Copy Location")
    if not copy_loc:
        copy_loc = focus_target.constraints.new(type='COPY_LOCATION')
    copy_loc.target = target_mesh

    # 2. Camera Setup
    cam_data = bpy.data.cameras.get("Vertical_Reel_Cam_Data")
    if not cam_data:
        cam_data = bpy.data.cameras.new("Vertical_Reel_Cam_Data")
    cam_data.lens = 45.0
    cam_data.sensor_fit = 'VERTICAL'
    cam_data.clip_end = 100.0

    cam_obj = bpy.data.objects.get("Vertical_Reel_Camera")
    if not cam_obj:
        cam_obj = bpy.data.objects.new("Vertical_Reel_Camera", cam_data)
        scene.collection.objects.link(cam_obj)

    scene.camera = cam_obj
    cam_obj.parent = None

    # 3. Track To Constraint
    track = cam_obj.constraints.get("Track To")
    if not track:
        track = cam_obj.constraints.new(type='TRACK_TO')
    track.target = focus_target
    track.track_axis = 'TRACK_NEGATIVE_Z'
    track.up_axis = 'UP_Y'

    # Helper: Test if all mesh vertices fall within viewport margins [margin, 1.0 - margin]
    def check_mesh_in_frame(frame_num, cam_location, margin=0.08):
        cam_obj.location = cam_location
        scene.frame_set(int(frame_num))
        bpy.context.view_layer.update()

        depsgraph = bpy.context.evaluated_depsgraph_get()
        eval_mesh_obj = target_mesh.evaluated_get(depsgraph)
        mesh_data = eval_mesh_obj.data

        # Sample mesh vertices (step sampling for performance)
        vert_step = max(1, len(mesh_data.vertices) // 300)
        world_matrix = target_mesh.matrix_world

        min_x, max_x = 1.0, 0.0
        min_y, max_y = 1.0, 0.0

        for i in range(0, len(mesh_data.vertices), vert_step):
            v_world = world_matrix @ mesh_data.vertices[i].co
            co_2d = world_to_camera_view(scene, cam_obj, v_world)

            if co_2d.x < min_x: min_x = co_2d.x
            if co_2d.x > max_x: max_x = co_2d.x
            if co_2d.y < min_y: min_y = co_2d.y
            if co_2d.y > max_y: max_y = co_2d.y

        # Check if bounds stay within safe viewport margins
        in_bounds = (min_x >= margin and max_x <= (1.0 - margin) and
                     min_y >= margin and max_y <= (1.0 - margin))

        return in_bounds, min_x, max_x, min_y, max_y

    # Helper: Find minimum camera distance needed to fit mesh at a given angle/position
    def find_fit_distance(frame_num, angle_rad, base_height_offset, initial_dist=6.0, margin=0.08):
        scene.frame_set(int(frame_num))
        bpy.context.view_layer.update()

        center_pos = focus_target.matrix_world.translation.copy()
        heading = target_rig.matrix_world.to_euler('XYZ').z
        view_angle = heading + angle_rad

        dist = initial_dist
        # Iteratively step back until all vertices are inside camera view
        for _ in range(30):
            test_x = center_pos.x + dist * math.sin(view_angle)
            test_y = center_pos.y - dist * math.cos(view_angle)
            test_z = center_pos.z + base_height_offset

            fit, min_x, max_x, min_y, max_y = check_mesh_in_frame(frame_num, (test_x, test_y, test_z), margin=margin)
            if fit:
                break
            dist += 0.4  # Step back further

        return dist, center_pos, view_angle

    target_height = target_mesh.dimensions.z if target_mesh.dimensions.z > 0 else 1.8
    chest_z_offset = target_height * 0.50

    # 4. Solve Distance for Walk Phase (Sampling key frames across walk cycle)
    max_walk_dist = 6.0
    walk_frames_to_sample = range(1, turn_start + 1, 6)

    for f in walk_frames_to_sample:
        req_d, _, _ = find_fit_distance(f, angle_rad=math.radians(-45.0), base_height_offset=chest_z_offset, initial_dist=max_walk_dist, margin=0.10)
        if req_d > max_walk_dist:
            max_walk_dist = req_d

    # Frame 1: Walk Start
    scene.frame_set(1)
    bpy.context.view_layer.update()
    c1 = focus_target.matrix_world.translation.copy()
    h1 = target_rig.matrix_world.to_euler('XYZ').z + math.radians(-45.0)
    w_x1 = c1.x + max_walk_dist * math.sin(h1)
    w_y1 = c1.y - max_walk_dist * math.cos(h1)
    w_z1 = c1.z + chest_z_offset

    # Frame turn_start: Walk End / Turn Start
    scene.frame_set(turn_start)
    bpy.context.view_layer.update()
    cts = focus_target.matrix_world.translation.copy()
    hts = target_rig.matrix_world.to_euler('XYZ').z + math.radians(-45.0)
    w_xts = cts.x + max_walk_dist * math.sin(hts)
    w_yts = cts.y - max_walk_dist * math.cos(hts)
    w_zts = cts.z + chest_z_offset

    # 5. Solve Distance for Unfurl Phase
    unfurl_frames_to_sample = range(turn_end_frame, unfurl_end_frame + 1, 6)
    max_unfurl_dist = max_walk_dist

    for f in unfurl_frames_to_sample:
        req_d, _, _ = find_fit_distance(f, angle_rad=math.radians(0.0), base_height_offset=chest_z_offset, initial_dist=max_unfurl_dist, margin=0.10)
        if req_d > max_unfurl_dist:
            max_unfurl_dist = req_d

    scene.frame_set(turn_end_frame)
    bpy.context.view_layer.update()
    cte = focus_target.matrix_world.translation.copy()
    hte = target_rig.matrix_world.to_euler('XYZ').z + math.radians(0.0)
    f_x = cte.x + max_unfurl_dist * math.sin(hte)
    f_y = cte.y - max_unfurl_dist * math.cos(hte)
    f_z = cte.z + chest_z_offset

    # Apply Keyframes
    if cam_obj.animation_data:
        cam_obj.animation_data_clear()

    cam_obj.location = (w_x1, w_y1, w_z1)
    cam_obj.keyframe_insert(data_path="location", frame=1)

    cam_obj.location = (w_xts, w_yts, w_zts)
    cam_obj.keyframe_insert(data_path="location", frame=turn_start)

    cam_obj.location = (f_x, f_y, f_z)
    cam_obj.keyframe_insert(data_path="location", frame=turn_end_frame)

    cam_obj.location = (f_x, f_y, f_z)
    cam_obj.keyframe_insert(data_path="location", frame=unfurl_end_frame)

    # Smooth Bezier handle transitions
    for fc in get_object_fcurves(cam_obj):
        for kp in fc.keyframe_points:
            kp.interpolation = 'BEZIER'

    scene.frame_set(1)
    return cam_obj, focus_target


# ==============================================================================
# GEOMETRY NODES GROUND, GRASS & SKY DOME SETUP
# ==============================================================================

def ensure_world_sphere_size(cam_obj, ground_obj, safety_multiplier=2.5):
    """
    Dynamically measures camera trajectory across all frames and resizes
    the sky sphere, ground mesh, and camera far clipping plane accordingly.
    """
    if not cam_obj or not ground_obj:
        return

    scene = bpy.context.scene
    max_cam_dist = 0.0

    # Sample camera distance relative to world origin across the animation range
    for frame in range(scene.frame_start, scene.frame_end + 1, 5):
        scene.frame_set(frame)
        bpy.context.view_layer.update()

        cam_dist = cam_obj.matrix_world.translation.length
        if cam_dist > max_cam_dist:
            max_cam_dist = cam_dist

    # Calculate target radius with safety factor
    world_radius = max(25.0, max_cam_dist * safety_multiplier)

    # 1. Update Sky Dome Radius in Geometry Nodes Tree
    mod = ground_obj.modifiers.get("Grass_Ground_GN")
    if mod and mod.node_group:
        for node in mod.node_group.nodes:
            if node.bl_idname == "GeometryNodeMeshIcoSphere":
                node.inputs["Radius"].default_value = world_radius

    # 2. Update Ground Circle Mesh Dimensions to match new world scale
    if ground_obj.type == 'MESH':
        bpy.context.view_layer.objects.active = ground_obj
        bpy.ops.object.mode_set(mode='OBJECT')

        # Scale ground circle diameter to match sky dome diameter
        ground_scale = world_radius / 12.5
        ground_obj.scale = (ground_scale, ground_scale, 1.0)
        bpy.ops.object.transform_apply(location=False, rotation=False, scale=True)

    # 3. Extend Camera Clip End so sky dome is never clipped out of view
    cam_obj.data.clip_end = max(cam_obj.data.clip_end, world_radius * 2.5)

    scene.frame_set(1)
    bpy.context.view_layer.update()


def create_grass_material(grass_dir):
    mat = bpy.data.materials.new(name="Grass_PBR_Material")
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    out_node = nodes.new("ShaderNodeOutputMaterial")
    bsdf = nodes.new("ShaderNodeBsdfPrincipled")
    links.new(bsdf.outputs["BSDF"], out_node.inputs["Surface"])

    col_files = glob.glob(os.path.join(grass_dir, "*Color*.jpg")) + glob.glob(os.path.join(grass_dir, "*JPG*.jpg"))
    if col_files:
        tex_col = nodes.new("ShaderNodeTexImage")
        tex_col.image = bpy.data.images.load(col_files[0])
        links.new(tex_col.outputs["Color"], bsdf.inputs["Base Color"])

    rough_files = glob.glob(os.path.join(grass_dir, "*Roughness*.jpg"))
    if rough_files:
        tex_rough = nodes.new("ShaderNodeTexImage")
        tex_rough.image = bpy.data.images.load(rough_files[0])
        tex_rough.image.colorspace_settings.name = 'Non-Color'
        links.new(tex_rough.outputs["Color"], bsdf.inputs["Roughness"])

    norm_files = glob.glob(os.path.join(grass_dir, "*Normal*.jpg"))
    if norm_files:
        tex_norm = nodes.new("ShaderNodeTexImage")
        tex_norm.image = bpy.data.images.load(norm_files[0])
        tex_norm.image.colorspace_settings.name = 'Non-Color'

        normal_map_node = nodes.new("ShaderNodeNormalMap")
        links.new(tex_norm.outputs["Color"], normal_map_node.inputs["Color"])
        links.new(normal_map_node.outputs["Normal"], bsdf.inputs["Normal"])

    return mat


def create_sky_material(sky_hdri_path):
    mat = bpy.data.materials.new(name="Sky_Dome_Material")
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    out_node = nodes.new("ShaderNodeOutputMaterial")
    emission = nodes.new("ShaderNodeEmission")
    emission.inputs["Strength"].default_value = 1.0

    links.new(emission.outputs["Emission"], out_node.inputs["Surface"])

    if os.path.exists(sky_hdri_path):
        tex_env = nodes.new("ShaderNodeTexEnvironment")
        tex_env.image = bpy.data.images.load(sky_hdri_path)
        links.new(tex_env.outputs["Color"], emission.inputs["Color"])

    return mat


def setup_grass_ground_geometry_nodes(grass_dir, sky_hdri_path, ground_z_offset=-0.2):
    grass_coll = bpy.data.collections.get("Grass_Artifacts_Collection")
    if not grass_coll:
        grass_coll = bpy.data.collections.new("Grass_Artifacts_Collection")
        bpy.context.scene.collection.children.link(grass_coll)

    grass_mat = create_grass_material(grass_dir)
    sky_mat = create_sky_material(sky_hdri_path)

    mesh_files = glob.glob(os.path.join(grass_dir, "*.obj")) + glob.glob(os.path.join(grass_dir, "*.fbx"))

    if mesh_files:
        for fpath in mesh_files:
            if fpath.endswith(".obj"):
                bpy.ops.wm.obj_import(filepath=fpath)
            elif fpath.endswith(".fbx"):
                bpy.ops.import_scene.fbx(filepath=fpath)

            for o in bpy.context.selected_objects:
                grass_coll.objects.link(o)
                if o.name in bpy.context.scene.collection.objects:
                    bpy.context.scene.collection.objects.unlink(o)

        for obj in list(grass_coll.objects):
            if obj.dimensions.z > 0.5:
                grass_coll.objects.unlink(obj)
                bpy.data.objects.remove(obj, do_unlink=True)
    else:
        for i in range(3):
            m_data = bpy.data.meshes.new(f"Grass_Blade_Mesh_{i}")
            g_obj = bpy.data.objects.new(f"Grass_Blade_Obj_{i}", m_data)
            grass_coll.objects.link(g_obj)

            verts = [(0, 0, 0), (0.05, 0, 0), (0.04, 0, 0.15), (0.01, 0, 0.25), (0, 0, 0.3)]
            faces = [(0, 1, 2), (1, 2, 3), (2, 3, 4)]
            m_data.from_pydata(verts, [], faces)
            m_data.update()

            if g_obj.data.materials:
                g_obj.data.materials[0] = grass_mat
            else:
                g_obj.data.materials.append(grass_mat)

            g_obj.location = (100, 100, 0)

    bpy.ops.mesh.primitive_circle_add(radius=12.5, vertices=64, fill_type='NGON', location=(0, 0, ground_z_offset))
    ground_obj = bpy.context.active_object
    ground_obj.name = "Ground_Mesh_Object"

    vg = ground_obj.vertex_groups.new(name="Grass_Weight_Group")
    all_vert_indices = [v.index for v in ground_obj.data.vertices]
    vg.add(all_vert_indices, 1.0, 'REPLACE')
    ground_obj.data.materials.append(grass_mat)
    ground_obj.data.materials.append(sky_mat)

    gn_group = bpy.data.node_groups.new("Grass_Ground_GN", type="GeometryNodeTree")
    gn_group.interface.new_socket(name="Geometry", in_out="INPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Geometry", in_out="OUTPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Density Mask", in_out="INPUT", socket_type="NodeSocketFloat")

    nodes = gn_group.nodes
    links = gn_group.links
    out_node = nodes.new("NodeGroupOutput")
    in_node = nodes.new("NodeGroupInput")

    distribute_node = nodes.new("GeometryNodeDistributePointsOnFaces")
    distribute_node.distribute_method = 'POISSON'
    distribute_node.inputs["Distance Min"].default_value = 0.02
    distribute_node.inputs["Density Max"].default_value = 300.0
    links.new(in_node.outputs["Geometry"], distribute_node.inputs["Mesh"])

    noise_node = nodes.new("ShaderNodeTexNoise")
    noise_node.inputs["Scale"].default_value = 2.5
    noise_node.inputs["Detail"].default_value = 2.0

    color_ramp = nodes.new("ShaderNodeValToRGB")
    color_ramp.color_ramp.elements[0].position = 0.20
    color_ramp.color_ramp.elements[1].position = 0.80
    links.new(noise_node.outputs["Fac"], color_ramp.inputs["Fac"])

    math_weight = nodes.new("ShaderNodeMath")
    math_weight.operation = "MULTIPLY"
    links.new(color_ramp.outputs["Color"], math_weight.inputs[0])
    links.new(in_node.outputs["Density Mask"], math_weight.inputs[1])

    links.new(math_weight.outputs["Value"], distribute_node.inputs["Density Factor"])

    instance_node = nodes.new("GeometryNodeInstanceOnPoints")
    instance_node.inputs["Pick Instance"].default_value = True
    links.new(distribute_node.outputs["Points"], instance_node.inputs["Points"])

    coll_info = nodes.new("GeometryNodeCollectionInfo")
    coll_info.inputs["Collection"].default_value = grass_coll
    coll_info.inputs["Separate Children"].default_value = True
    coll_info.inputs["Reset Children"].default_value = True
    links.new(coll_info.outputs["Instances"], instance_node.inputs["Instance"])

    rand_scale = nodes.new("FunctionNodeRandomValue")
    rand_scale.data_type = 'FLOAT'
    rand_scale.inputs["Min"].default_value = 0.25
    rand_scale.inputs["Max"].default_value = 0.50

    combine_scale = nodes.new("ShaderNodeCombineXYZ")
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["X"])
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["Y"])
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["Z"])
    links.new(combine_scale.outputs["Vector"], instance_node.inputs["Scale"])

    rand_rot_z = nodes.new("FunctionNodeRandomValue")
    rand_rot_z.data_type = 'FLOAT'
    rand_rot_z.inputs["Min"].default_value = 0.0
    rand_rot_z.inputs["Max"].default_value = math.radians(360.0)

    combine_rot = nodes.new("ShaderNodeCombineXYZ")
    links.new(rand_rot_z.outputs["Value"], combine_rot.inputs["Z"])
    links.new(combine_rot.outputs["Vector"], instance_node.inputs["Rotation"])

    sky_sphere = nodes.new("GeometryNodeMeshIcoSphere")
    sky_sphere.inputs["Radius"].default_value = 12.5
    sky_sphere.inputs["Subdivisions"].default_value = 4

    flip_faces = nodes.new("GeometryNodeFlipFaces")
    links.new(sky_sphere.outputs["Mesh"], flip_faces.inputs["Mesh"])

    transform_sky = nodes.new("GeometryNodeTransform")
    transform_sky.inputs["Translation"].default_value = (0.0, 0.0, ground_z_offset)
    links.new(flip_faces.outputs["Mesh"], transform_sky.inputs["Geometry"])

    set_sky_mat = nodes.new("GeometryNodeSetMaterial")
    set_sky_mat.inputs["Material"].default_value = sky_mat
    links.new(transform_sky.outputs["Geometry"], set_sky_mat.inputs["Geometry"])

    join_geo = nodes.new("GeometryNodeJoinGeometry")
    links.new(in_node.outputs["Geometry"], join_geo.inputs["Geometry"])
    links.new(instance_node.outputs["Instances"], join_geo.inputs["Geometry"])
    links.new(set_sky_mat.outputs["Geometry"], join_geo.inputs["Geometry"])

    links.new(join_geo.outputs["Geometry"], out_node.inputs["Geometry"])

    mod = ground_obj.modifiers.new(name="Grass_Ground_GN", type='NODES')
    mod.node_group = gn_group

    for item in gn_group.interface.items_tree:
        if getattr(item, "in_out", None) == "INPUT" and item.name == "Density Mask":
            mod[item.identifier + "_attribute_name"] = "Grass_Weight_Group"
            mod[item.identifier + "_use_attribute"] = True

    return ground_obj


# ==============================================================================
# SCROLL MATERIAL & GEOMETRY NODES SETUP
# ==============================================================================

def create_image_scroll_material(image_path):
    mat_name = "Scroll_Texture_Mat"
    mat = bpy.data.materials.get(mat_name)
    if mat:
        bpy.data.materials.remove(mat)

    mat = bpy.data.materials.new(name=mat_name)
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    output_node = nodes.new("ShaderNodeOutputMaterial")
    bsdf = nodes.new("ShaderNodeBsdfPrincipled")
    bsdf.inputs["Roughness"].default_value = 0.5

    tex_img = nodes.new("ShaderNodeTexImage")
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        tex_img.image = img
        tex_img.extension = 'REPEAT'

    uv_map_node = nodes.new("ShaderNodeUVMap")
    uv_map_node.uv_map = "UVMap"

    links.new(uv_map_node.outputs["UV"], tex_img.inputs["Vector"])
    links.new(tex_img.outputs["Color"], bsdf.inputs["Base Color"])
    links.new(bsdf.outputs["BSDF"], output_node.inputs["Surface"])

    return mat, tex_img.image if os.path.exists(image_path) else None


def create_pure_geometry_node_scroll(image_path=""):
    gn_group = bpy.data.node_groups.new("Scroll_Pure_GN", type="GeometryNodeTree")

    gn_group.interface.new_socket(name="Geometry", in_out="OUTPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Character Mesh", in_out="INPUT", socket_type="NodeSocketObject")

    sock_r_vg = gn_group.interface.new_socket(name="Right Hand Group", in_out="INPUT", socket_type="NodeSocketString")
    sock_r_vg.default_value = "mixamorig:RightHand"

    sock_l_vg = gn_group.interface.new_socket(name="Left Hand Group", in_out="INPUT", socket_type="NodeSocketString")
    sock_l_vg.default_value = "mixamorig:LeftHand"

    sock_offset = gn_group.interface.new_socket(name="Forward Offset", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_offset.default_value = 0.60

    sock_unfurl = gn_group.interface.new_socket(name="Unfurl", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_unfurl.default_value = 0.0

    sock_width = gn_group.interface.new_socket(name="Scroll Width", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_width.default_value = 0.35

    nodes = gn_group.nodes
    links = gn_group.links

    in_node = nodes.new("NodeGroupInput")
    out_node = nodes.new("NodeGroupOutput")

    char_info = nodes.new("GeometryNodeObjectInfo")
    char_info.transform_space = "RELATIVE"
    links.new(in_node.outputs["Character Mesh"], char_info.inputs["Object"])

    pos_node = nodes.new("GeometryNodeInputPosition")

    named_r = nodes.new("GeometryNodeInputNamedAttribute")
    named_r.data_type = "FLOAT"
    links.new(in_node.outputs["Right Hand Group"], named_r.inputs["Name"])

    cmp_r = nodes.new("FunctionNodeCompare")
    cmp_r.data_type = "FLOAT"
    cmp_r.operation = "GREATER_THAN"
    cmp_r.inputs[1].default_value = 0.001
    links.new(named_r.outputs["Attribute"], cmp_r.inputs[0])

    stat_r = nodes.new("GeometryNodeAttributeStatistic")
    stat_r.data_type = "FLOAT_VECTOR"
    links.new(char_info.outputs["Geometry"], stat_r.inputs["Geometry"])
    links.new(cmp_r.outputs["Result"], stat_r.inputs["Selection"])
    links.new(pos_node.outputs["Position"], stat_r.inputs["Attribute"])

    named_l = nodes.new("GeometryNodeInputNamedAttribute")
    named_l.data_type = "FLOAT"
    links.new(in_node.outputs["Left Hand Group"], named_l.inputs["Name"])

    cmp_l = nodes.new("FunctionNodeCompare")
    cmp_l.data_type = "FLOAT"
    cmp_l.operation = "GREATER_THAN"
    cmp_l.inputs[1].default_value = 0.001
    links.new(named_l.outputs["Attribute"], cmp_l.inputs[0])

    stat_l = nodes.new("GeometryNodeAttributeStatistic")
    stat_l.data_type = "FLOAT_VECTOR"
    links.new(char_info.outputs["Geometry"], stat_l.inputs["Geometry"])
    links.new(cmp_l.outputs["Result"], stat_l.inputs["Selection"])
    links.new(pos_node.outputs["Position"], stat_l.inputs["Attribute"])

    neg_offset = nodes.new("ShaderNodeMath")
    neg_offset.operation = "MULTIPLY"
    neg_offset.inputs[1].default_value = -1.0
    links.new(in_node.outputs["Forward Offset"], neg_offset.inputs[0])

    forward_vec = nodes.new("ShaderNodeCombineXYZ")
    forward_vec.inputs["X"].default_value = 0.0
    links.new(neg_offset.outputs["Value"], forward_vec.inputs["Y"])
    forward_vec.inputs["Z"].default_value = 0.0

    start_pos = nodes.new("ShaderNodeVectorMath")
    start_pos.operation = "ADD"
    links.new(stat_r.outputs["Mean"], start_pos.inputs[0])
    links.new(forward_vec.outputs["Vector"], start_pos.inputs[1])

    target_pos = nodes.new("ShaderNodeVectorMath")
    target_pos.operation = "ADD"
    links.new(stat_l.outputs["Mean"], target_pos.inputs[0])
    links.new(forward_vec.outputs["Vector"], target_pos.inputs[1])

    delta_vec = nodes.new("ShaderNodeVectorMath")
    delta_vec.operation = "SUBTRACT"
    links.new(target_pos.outputs["Vector"], delta_vec.inputs[0])
    links.new(start_pos.outputs["Vector"], delta_vec.inputs[1])

    dist_node = nodes.new("ShaderNodeVectorMath")
    dist_node.operation = "DISTANCE"
    links.new(start_pos.outputs["Vector"], dist_node.inputs[0])

    scale_vec = nodes.new("ShaderNodeVectorMath")
    scale_vec.operation = "SCALE"
    links.new(delta_vec.outputs["Vector"], scale_vec.inputs["Vector"])
    links.new(in_node.outputs["Unfurl"], scale_vec.inputs["Scale"])

    end_pos = nodes.new("ShaderNodeVectorMath")
    end_pos.operation = "ADD"
    links.new(start_pos.outputs["Vector"], end_pos.inputs[0])
    links.new(scale_vec.outputs["Vector"], end_pos.inputs[1])

    links.new(end_pos.outputs["Vector"], dist_node.inputs[1])

    curve_line = nodes.new("GeometryNodeCurvePrimitiveLine")
    links.new(start_pos.outputs["Vector"], curve_line.inputs["Start"])
    links.new(end_pos.outputs["Vector"], curve_line.inputs["End"])

    set_tilt = nodes.new("GeometryNodeSetCurveTilt")
    set_tilt.inputs["Tilt"].default_value = 1.5708
    links.new(curve_line.outputs["Curve"], set_tilt.inputs["Curve"])

    spline_param_main = nodes.new("GeometryNodeSplineParameter")
    store_u = nodes.new("GeometryNodeStoreNamedAttribute")
    store_u.inputs["Name"].default_value = "U_Factor"
    store_u.data_type = 'FLOAT'
    store_u.domain = 'POINT'
    links.new(set_tilt.outputs["Curve"], store_u.inputs["Geometry"])
    links.new(spline_param_main.outputs["Factor"], store_u.inputs["Value"])

    half_width_neg = nodes.new("ShaderNodeMath")
    half_width_neg.operation = "MULTIPLY"
    half_width_neg.inputs[1].default_value = -0.5
    links.new(in_node.outputs["Scroll Width"], half_width_neg.inputs[0])

    half_width_pos = nodes.new("ShaderNodeMath")
    half_width_pos.operation = "MULTIPLY"
    half_width_pos.inputs[1].default_value = 0.5
    links.new(in_node.outputs["Scroll Width"], half_width_pos.inputs[0])

    combine_start = nodes.new("ShaderNodeCombineXYZ")
    links.new(half_width_neg.outputs["Value"], combine_start.inputs["X"])

    combine_end = nodes.new("ShaderNodeCombineXYZ")
    links.new(half_width_pos.outputs["Value"], combine_end.inputs["X"])

    profile_line = nodes.new("GeometryNodeCurvePrimitiveLine")
    links.new(combine_start.outputs["Vector"], profile_line.inputs["Start"])
    links.new(combine_end.outputs["Vector"], profile_line.inputs["End"])

    spline_param_prof = nodes.new("GeometryNodeSplineParameter")
    store_v = nodes.new("GeometryNodeStoreNamedAttribute")
    store_v.inputs["Name"].default_value = "V_Factor"
    store_v.data_type = 'FLOAT'
    store_v.domain = 'POINT'
    links.new(profile_line.outputs["Curve"], store_v.inputs["Geometry"])
    links.new(spline_param_prof.outputs["Factor"], store_v.inputs["Value"])

    curve_to_mesh = nodes.new("GeometryNodeCurveToMesh")
    links.new(store_u.outputs["Geometry"], curve_to_mesh.inputs["Curve"])
    links.new(store_v.outputs["Geometry"], curve_to_mesh.inputs["Profile Curve"])

    img_aspect = 1.0
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        if img.size[1] > 0:
            img_aspect = img.size[0] / img.size[1]

    h_times_aspect = nodes.new("ShaderNodeMath")
    h_times_aspect.operation = "MULTIPLY"
    h_times_aspect.inputs[1].default_value = img_aspect
    links.new(in_node.outputs["Scroll Width"], h_times_aspect.inputs[0])

    tile_count = nodes.new("ShaderNodeMath")
    tile_count.operation = "DIVIDE"
    links.new(dist_node.outputs["Value"], tile_count.inputs[0])
    links.new(h_times_aspect.outputs["Value"], tile_count.inputs[1])

    read_u = nodes.new("GeometryNodeInputNamedAttribute")
    read_u.inputs["Name"].default_value = "U_Factor"
    read_u.data_type = 'FLOAT'

    scaled_u = nodes.new("ShaderNodeMath")
    scaled_u.operation = "MULTIPLY"
    links.new(read_u.outputs["Attribute"], scaled_u.inputs[0])
    links.new(tile_count.outputs["Value"], scaled_u.inputs[1])

    read_v = nodes.new("GeometryNodeInputNamedAttribute")
    read_v.inputs["Name"].default_value = "V_Factor"
    read_v.data_type = 'FLOAT'

    flip_v = nodes.new("ShaderNodeMath")
    flip_v.operation = "SUBTRACT"
    flip_v.inputs[0].default_value = 1.0
    links.new(read_v.outputs["Attribute"], flip_v.inputs[1])

    combine_uv = nodes.new("ShaderNodeCombineXYZ")
    links.new(scaled_u.outputs["Value"], combine_uv.inputs["X"])
    links.new(flip_v.outputs["Value"], combine_uv.inputs["Y"])

    store_uv = nodes.new("GeometryNodeStoreNamedAttribute")
    store_uv.inputs["Name"].default_value = "UVMap"
    store_uv.data_type = 'FLOAT2'
    store_uv.domain = 'CORNER'
    links.new(curve_to_mesh.outputs["Mesh"], store_uv.inputs["Geometry"])
    links.new(combine_uv.outputs["Vector"], store_uv.inputs["Value"])

    set_material = nodes.new("GeometryNodeSetMaterial")
    scroll_mat, _ = create_image_scroll_material(image_path)
    set_material.inputs["Material"].default_value = scroll_mat
    links.new(store_uv.outputs["Geometry"], set_material.inputs["Geometry"])

    links.new(set_material.outputs["Geometry"], out_node.inputs["Geometry"])

    return gn_group, scroll_mat


def setup_geometry_node_scroll(target_mesh, unfurl_start, unfurl_end, image_path=""):
    mesh_data = bpy.data.meshes.new("Scroll_Procedural_Mesh")
    scroll_obj = bpy.data.objects.new("Scroll_GN_Object", mesh_data)
    bpy.context.scene.collection.objects.link(scroll_obj)

    mod = scroll_obj.modifiers.new("Scroll_GN", type="NODES")
    gn_tree, scroll_mat = create_pure_geometry_node_scroll(image_path=image_path)
    mod.node_group = gn_tree

    scroll_obj.data.materials.append(scroll_mat)

    r_hand_name, l_hand_name = find_hand_vertex_groups(target_mesh)

    unfurl_id = None
    for item in gn_tree.interface.items_tree:
        if getattr(item, "in_out", None) == "INPUT":
            if item.name == "Character Mesh":
                mod[item.identifier] = target_mesh
            elif item.name == "Right Hand Group":
                mod[item.identifier] = r_hand_name
            elif item.name == "Left Hand Group":
                mod[item.identifier] = l_hand_name
            elif item.name == "Forward Offset":
                mod[item.identifier] = 0.60
            elif item.name == "Scroll Width":
                mod[item.identifier] = 0.35
            elif item.name == "Unfurl":
                unfurl_id = item.identifier

    for m in target_mesh.modifiers:
        if m.type == 'ARMATURE':
            m.show_viewport = True
            m.show_render = True
            m.show_in_editmode = True

    scroll_obj.hide_viewport = True
    scroll_obj.hide_render = True
    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=1)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=1)

    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=unfurl_start - 1)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=unfurl_start - 1)

    scroll_obj.hide_viewport = False
    scroll_obj.hide_render = False
    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=unfurl_start)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=unfurl_start)

    for fc in get_object_fcurves(scroll_obj):
        if fc.data_path in ["hide_viewport", "hide_render"]:
            for kp in fc.keyframe_points:
                kp.interpolation = 'CONSTANT'

    if unfurl_id:
        mod[unfurl_id] = 0.001
        mod.keyframe_insert(data_path=f'["{unfurl_id}"]', frame=unfurl_start)

        mod[unfurl_id] = 1.0
        mod.keyframe_insert(data_path=f'["{unfurl_id}"]', frame=unfurl_end)

    return scroll_obj


# ==============================================================================
# MAIN CONTROLLER
# ==============================================================================

def import_and_extract_action(fbx_path, action_name):
    existing_objects = set(bpy.data.objects)
    bpy.ops.import_scene.fbx(filepath=fbx_path)
    bpy.context.view_layer.update()

    imported_objects = set(bpy.data.objects) - existing_objects
    extracted_action = None
    for obj in imported_objects:
        if obj.animation_data and obj.animation_data.action:
            extracted_action = obj.animation_data.action
            extracted_action.name = action_name
            extracted_action.use_fake_user = True
            break
    return extracted_action, imported_objects


def purge_imported_objects(objects_to_remove):
    for obj in objects_to_remove:
        if obj.name in bpy.data.objects:
            if obj.animation_data:
                obj.animation_data_clear()
            for coll in list(obj.users_collection):
                coll.objects.unlink(obj)
    bpy.context.view_layer.update()


def main():
    cleanup_previous_run()

    walk_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Mutant Walking.fbx"
    turn_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Right Turn 90.fbx"
    unfurl_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Mutant Idle.fbx"

    sky_hdri_path = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/horn-koppe_spring_4k.exr"
    output_blend = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/output_scene.blend"

    image_path = "/Users/dobbinp1/Downloads/IMGP6691_psedit.jpg"
    grass_dir = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Grass004_4K-JPG"

    ground_obj = setup_grass_ground_geometry_nodes(grass_dir, sky_hdri_path, ground_z_offset=-0.2)

    CYCLES = 4
    CROSSFADE = 15
    TARGET_DISTANCE = 4.0

    walk_action, walk_objs = import_and_extract_action(walk_fbx, "walk")
    target_rig = next((obj for obj in walk_objs if obj.type == "ARMATURE"), None)
    target_mesh = next((obj for obj in walk_objs if obj.type == "MESH"), None)

    if not target_rig or not target_mesh:
        raise RuntimeError("Armature or Character Mesh not found in walk FBX.")

    travel_axis_idx, motion_sign = detect_natural_forward_axis(walk_action, target_rig)
    initial_heading = math.radians(90.0) if travel_axis_idx == 0 else 0.0
    target_rig.rotation_euler[2] = initial_heading

    action_len = int(walk_action.frame_range[1] - walk_action.frame_range[0])
    if action_len <= 0:
        action_len = 30

    anim_data = target_rig.animation_data_create()

    walk_track = anim_data.nla_tracks.new()
    walk_track.name = "Walk_Track"

    current_frame = 1
    for i in range(CYCLES):
        strip = walk_track.strips.new(name=f"Walk_{i+1}", start=int(current_frame), action=walk_action)
        strip.blend_type = "REPLACE"
        current_frame += action_len

    walk_end_frame = int(current_frame)

    turn_action, turn_objs = import_and_extract_action(turn_fbx, "rightturn")
    purge_imported_objects(turn_objs)

    turn_track = anim_data.nla_tracks.new()
    turn_track.name = "RightTurn_Track"

    turn_start = int(walk_end_frame - CROSSFADE)
    turn_strip = turn_track.strips.new(name="RightTurn", start=turn_start, action=turn_action)
    turn_strip.blend_type = "REPLACE"
    turn_strip.blend_in = CROSSFADE
    turn_strip.use_auto_blend = True
    turn_strip.extrapolation = "HOLD"
    turn_end_frame = int(turn_strip.frame_end)

    unfurl_action, unfurl_objs = import_and_extract_action(unfurl_fbx, "unfurl")
    purge_imported_objects(unfurl_objs)

    unfurl_track = anim_data.nla_tracks.new()
    unfurl_track.name = "Unfurl_Track"

    unfurl_start = int(turn_end_frame - CROSSFADE)
    unfurl_strip = unfurl_track.strips.new(name="Unfurl", start=unfurl_start, action=unfurl_action)
    unfurl_strip.blend_type = "REPLACE"
    unfurl_strip.blend_in = CROSSFADE
    unfurl_strip.use_auto_blend = True
    unfurl_strip.extrapolation = "HOLD"
    unfurl_end_frame = int(unfurl_strip.frame_end)

    root_action = bpy.data.actions.new(name="World_Root_Motion")
    target_rig.animation_data.action = root_action

    loc_fcurve = root_action.fcurve_ensure_for_datablock(datablock=target_rig, data_path="location", index=travel_axis_idx)
    final_target_value = TARGET_DISTANCE * motion_sign

    k_loc1 = loc_fcurve.keyframe_points.insert(frame=1.0, value=0.0)
    k_loc2 = loc_fcurve.keyframe_points.insert(frame=float(turn_start), value=final_target_value)
    k_loc3 = loc_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=final_target_value)

    k_loc1.interpolation = "LINEAR"
    k_loc2.interpolation = "CONSTANT"
    k_loc3.interpolation = "CONSTANT"

    rot_fcurve = root_action.fcurve_ensure_for_datablock(datablock=target_rig, data_path="rotation_euler", index=2)
    target_heading = initial_heading - math.radians(90.0)

    k_rot1 = rot_fcurve.keyframe_points.insert(frame=1.0, value=initial_heading)
    k_rot2 = rot_fcurve.keyframe_points.insert(frame=float(turn_start), value=initial_heading)
    k_rot3 = rot_fcurve.keyframe_points.insert(frame=float(turn_end_frame), value=target_heading)
    k_rot4 = rot_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=target_heading)

    k_rot1.interpolation = "CONSTANT"
    k_rot2.interpolation = "BEZIER"
    k_rot3.interpolation = "CONSTANT"
    k_rot4.interpolation = "CONSTANT"

    root_track = anim_data.nla_tracks.new()
    root_track.name = "World_Root_Track"
    root_strip = root_track.strips.new(name="Root_Motion_Strip", start=1, action=root_action)
    root_strip.blend_type = "REPLACE"
    root_strip.extrapolation = "HOLD"

    anim_data.action = None

    setup_geometry_node_scroll(
        target_mesh=target_mesh,
        unfurl_start=unfurl_start,
        unfurl_end=unfurl_end_frame,
        image_path=image_path,
    )

    # Modern Instagram Reel Vertical Camera Setup
    cam_obj, _ = setup_modern_reel_camera(
        target_rig=target_rig,
        target_mesh=target_mesh,
        turn_start=turn_start,
        turn_end_frame=turn_end_frame,
        unfurl_end_frame=unfurl_end_frame
    )

    bpy.context.scene.frame_start = 1
    bpy.context.scene.frame_end = unfurl_end_frame

    # Dynamically scale world sphere, ground plane, grass scatter area, and camera far clip plane
    ensure_world_sphere_size(cam_obj, ground_obj, safety_multiplier=2.5)

    bpy.ops.wm.save_as_mainfile(filepath=output_blend)
    print(f"\n[Success] Dynamic world sizing & camera integrated. Saved to: {output_blend}")


main()
```

```
import math
import mathutils
import os
import glob
import bpy
from bpy_extras import anim_utils
from bpy_extras.object_utils import world_to_camera_view

# ==============================================================================
# CLEANUP & HELPERS
# ==============================================================================

def cleanup_previous_run():
    """Clears old scroll objects, ground, node groups, camera anchors, collections, and orphaned data."""
    target_names = [
        "Master_NLA_Control",
        "Scroll_GN_Object",
        "Scroll_Start_Pin",
        "Scroll_End_Pin",
        "Scroll_Master_Container",
        "HDRI_Ground_Plane",
        "Ground_Mesh_Object",
        "Camera_Focus_Target",
        "Vertical_Reel_Camera"
    ]
    for obj in list(bpy.data.objects):
        if obj.name in target_names or (obj.name.startswith("mixamorig") and obj.type != "ARMATURE" and obj.type != "MESH"):
            if obj.animation_data:
                obj.animation_data_clear()
            bpy.data.objects.remove(obj, do_unlink=True)

    for action in list(bpy.data.actions):
        if action.users == 0 or action.use_fake_user:
            action.use_fake_user = False
            bpy.data.actions.remove(action)

    for ng in list(bpy.data.node_groups):
        if ng.name in ["Scroll_Unfurl_GN", "Scroll_Pure_GN", "HDRI_Ground_GN", "Grass_Ground_GN"]:
            bpy.data.node_groups.remove(ng)

    for coll_name in ["Scroll_Master_Container", "Grass_Artifacts_Collection"]:
        if coll_name in bpy.data.collections:
            coll = bpy.data.collections[coll_name]
            for obj in list(coll.objects):
                bpy.data.objects.remove(obj, do_unlink=True)
            bpy.data.collections.remove(coll)

    bpy.context.view_layer.update()


def get_slot_fcurves(action, anim_target_object):
    fcurves = []
    if not action:
        return fcurves

    anim_data = getattr(anim_target_object, "animation_data", None)
    if anim_data and getattr(anim_data, "action_slot", None):
        slot = anim_data.action_slot
        channelbag = anim_utils.action_get_channelbag_for_slot(action, slot)
        if channelbag and hasattr(channelbag, "fcurves"):
            return list(channelbag.fcurves)

    if hasattr(action, "slots"):
        for slot in action.slots:
            for cb in getattr(slot, "channelbags", []):
                fcurves.extend(getattr(cb, "fcurves", []))
        if fcurves:
            return fcurves

    if hasattr(action, "fcurves"):
        return list(action.fcurves)

    return fcurves


def get_object_fcurves(obj):
    if not obj or not obj.animation_data or not obj.animation_data.action:
        return []
    action = obj.animation_data.action
    return get_slot_fcurves(action, obj)


def detect_natural_forward_axis(action, rig_object):
    fcurves = get_slot_fcurves(action, rig_object)
    max_delta = 0.0
    best_axis = 1
    net_delta = 0.0

    start_frame = action.frame_range[0]
    end_frame = action.frame_range[1]

    for fc in fcurves:
        if fc.data_path.endswith("location") and fc.array_index in [0, 1, 2]:
            start_val = fc.evaluate(start_frame)
            end_val = fc.evaluate(end_frame)
            delta = end_val - start_val
            if abs(delta) > abs(max_delta):
                max_delta = delta
                best_axis = fc.array_index
                net_delta = delta

    direction_sign = 1.0 if net_delta >= 0 else -1.0
    return best_axis, direction_sign


def find_hand_vertex_groups(target_mesh):
    r_name = "mixamorig:RightHand"
    l_name = "mixamorig:LeftHand"

    if target_mesh and hasattr(target_mesh, "vertex_groups"):
        for vg in target_mesh.vertex_groups:
            vg_lower = vg.name.lower()
            if "righthand" in vg_lower or "hand_r" in vg_lower or "hand.r" in vg_lower:
                r_name = vg.name
            elif "lefthand" in vg_lower or "hand_l" in vg_lower or "hand.l" in vg_lower:
                l_name = vg.name

    return r_name, l_name


# ==============================================================================
# UNPARENTED MATRIX-EVALUATED VERTICAL CAMERA SETUP
# ==============================================================================

def setup_modern_reel_camera(target_rig, target_mesh, turn_start, turn_end_frame, unfurl_end_frame):
    """
    Creates a 9:16 Instagram Reel Camera:
    - Uses exact 2D camera viewport projection (world_to_camera_view) to test
      every vertex across the animation, ensuring 100% of the mesh fits inside
      the camera frame during both walking and unfurl phases.
    """
    scene = bpy.context.scene
    scene.render.resolution_x = 1080
    scene.render.resolution_y = 1920
    scene.render.resolution_percentage = 100

    # 1. Focus Target Null
    focus_target = bpy.data.objects.get("Camera_Focus_Target")
    if not focus_target:
        focus_target = bpy.data.objects.new("Camera_Focus_Target", None)
        focus_target.empty_display_type = 'PLAIN_AXES'
        focus_target.empty_display_size = 0.25
        scene.collection.objects.link(focus_target)

    focus_target.parent = None

    copy_loc = focus_target.constraints.get("Copy Location")
    if not copy_loc:
        copy_loc = focus_target.constraints.new(type='COPY_LOCATION')
    copy_loc.target = target_mesh

    # 2. Camera Setup
    cam_data = bpy.data.cameras.get("Vertical_Reel_Cam_Data")
    if not cam_data:
        cam_data = bpy.data.cameras.new("Vertical_Reel_Cam_Data")
    cam_data.lens = 45.0
    cam_data.sensor_fit = 'VERTICAL'
    cam_data.clip_end = 100.0

    cam_obj = bpy.data.objects.get("Vertical_Reel_Camera")
    if not cam_obj:
        cam_obj = bpy.data.objects.new("Vertical_Reel_Camera", cam_data)
        scene.collection.objects.link(cam_obj)

    scene.camera = cam_obj
    cam_obj.parent = None

    # 3. Track To Constraint
    track = cam_obj.constraints.get("Track To")
    if not track:
        track = cam_obj.constraints.new(type='TRACK_TO')
    track.target = focus_target
    track.track_axis = 'TRACK_NEGATIVE_Z'
    track.up_axis = 'UP_Y'

    # Helper: Test if all mesh vertices fall within viewport margins [margin, 1.0 - margin]
    def check_mesh_in_frame(frame_num, cam_location, margin=0.08):
        cam_obj.location = cam_location
        scene.frame_set(int(frame_num))
        bpy.context.view_layer.update()

        depsgraph = bpy.context.evaluated_depsgraph_get()
        eval_mesh_obj = target_mesh.evaluated_get(depsgraph)
        mesh_data = eval_mesh_obj.data

        # Sample mesh vertices (step sampling for performance)
        vert_step = max(1, len(mesh_data.vertices) // 300)
        world_matrix = target_mesh.matrix_world

        min_x, max_x = 1.0, 0.0
        min_y, max_y = 1.0, 0.0

        for i in range(0, len(mesh_data.vertices), vert_step):
            v_world = world_matrix @ mesh_data.vertices[i].co
            co_2d = world_to_camera_view(scene, cam_obj, v_world)

            if co_2d.x < min_x: min_x = co_2d.x
            if co_2d.x > max_x: max_x = co_2d.x
            if co_2d.y < min_y: min_y = co_2d.y
            if co_2d.y > max_y: max_y = co_2d.y

        # Check if bounds stay within safe viewport margins
        in_bounds = (min_x >= margin and max_x <= (1.0 - margin) and
                     min_y >= margin and max_y <= (1.0 - margin))

        return in_bounds, min_x, max_x, min_y, max_y

    # Helper: Find minimum camera distance needed to fit mesh at a given angle/position
    def find_fit_distance(frame_num, angle_rad, base_height_offset, initial_dist=6.0, margin=0.08):
        scene.frame_set(int(frame_num))
        bpy.context.view_layer.update()

        center_pos = focus_target.matrix_world.translation.copy()
        heading = target_rig.matrix_world.to_euler('XYZ').z
        view_angle = heading + angle_rad

        dist = initial_dist
        # Iteratively step back until all vertices are inside camera view
        for _ in range(30):
            test_x = center_pos.x + dist * math.sin(view_angle)
            test_y = center_pos.y - dist * math.cos(view_angle)
            test_z = center_pos.z + base_height_offset

            fit, min_x, max_x, min_y, max_y = check_mesh_in_frame(frame_num, (test_x, test_y, test_z), margin=margin)
            if fit:
                break
            dist += 0.4  # Step back further

        return dist, center_pos, view_angle

    target_height = target_mesh.dimensions.z if target_mesh.dimensions.z > 0 else 1.8
    chest_z_offset = target_height * 0.50

    # 4. Solve Distance for Walk Phase (Sampling key frames across walk cycle)
    max_walk_dist = 6.0
    walk_frames_to_sample = range(1, turn_start + 1, 6)

    for f in walk_frames_to_sample:
        req_d, _, _ = find_fit_distance(f, angle_rad=math.radians(-45.0), base_height_offset=chest_z_offset, initial_dist=max_walk_dist, margin=0.10)
        if req_d > max_walk_dist:
            max_walk_dist = req_d

    # Frame 1: Walk Start
    scene.frame_set(1)
    bpy.context.view_layer.update()
    c1 = focus_target.matrix_world.translation.copy()
    h1 = target_rig.matrix_world.to_euler('XYZ').z + math.radians(-45.0)
    w_x1 = c1.x + max_walk_dist * math.sin(h1)
    w_y1 = c1.y - max_walk_dist * math.cos(h1)
    w_z1 = c1.z + chest_z_offset

    # Frame turn_start: Walk End / Turn Start
    scene.frame_set(turn_start)
    bpy.context.view_layer.update()
    cts = focus_target.matrix_world.translation.copy()
    hts = target_rig.matrix_world.to_euler('XYZ').z + math.radians(-45.0)
    w_xts = cts.x + max_walk_dist * math.sin(hts)
    w_yts = cts.y - max_walk_dist * math.cos(hts)
    w_zts = cts.z + chest_z_offset

    # 5. Solve Distance for Unfurl Phase
    unfurl_frames_to_sample = range(turn_end_frame, unfurl_end_frame + 1, 6)
    max_unfurl_dist = max_walk_dist

    for f in unfurl_frames_to_sample:
        req_d, _, _ = find_fit_distance(f, angle_rad=math.radians(0.0), base_height_offset=chest_z_offset, initial_dist=max_unfurl_dist, margin=0.10)
        if req_d > max_unfurl_dist:
            max_unfurl_dist = req_d

    scene.frame_set(turn_end_frame)
    bpy.context.view_layer.update()
    cte = focus_target.matrix_world.translation.copy()
    hte = target_rig.matrix_world.to_euler('XYZ').z + math.radians(0.0)
    f_x = cte.x + max_unfurl_dist * math.sin(hte)
    f_y = cte.y - max_unfurl_dist * math.cos(hte)
    f_z = cte.z + chest_z_offset

    # Apply Keyframes
    if cam_obj.animation_data:
        cam_obj.animation_data_clear()

    cam_obj.location = (w_x1, w_y1, w_z1)
    cam_obj.keyframe_insert(data_path="location", frame=1)

    cam_obj.location = (w_xts, w_yts, w_zts)
    cam_obj.keyframe_insert(data_path="location", frame=turn_start)

    cam_obj.location = (f_x, f_y, f_z)
    cam_obj.keyframe_insert(data_path="location", frame=turn_end_frame)

    cam_obj.location = (f_x, f_y, f_z)
    cam_obj.keyframe_insert(data_path="location", frame=unfurl_end_frame)

    # Smooth Bezier handle transitions
    for fc in get_object_fcurves(cam_obj):
        for kp in fc.keyframe_points:
            kp.interpolation = 'BEZIER'

    scene.frame_set(1)
    return cam_obj, focus_target


# ==============================================================================
# GEOMETRY NODES GROUND, GRASS & SKY DOME SETUP
# ==============================================================================

def create_grass_material(grass_dir):
    mat = bpy.data.materials.new(name="Grass_PBR_Material")
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    out_node = nodes.new("ShaderNodeOutputMaterial")
    bsdf = nodes.new("ShaderNodeBsdfPrincipled")
    links.new(bsdf.outputs["BSDF"], out_node.inputs["Surface"])

    col_files = glob.glob(os.path.join(grass_dir, "*Color*.jpg")) + glob.glob(os.path.join(grass_dir, "*JPG*.jpg"))
    if col_files:
        tex_col = nodes.new("ShaderNodeTexImage")
        tex_col.image = bpy.data.images.load(col_files[0])
        links.new(tex_col.outputs["Color"], bsdf.inputs["Base Color"])

    rough_files = glob.glob(os.path.join(grass_dir, "*Roughness*.jpg"))
    if rough_files:
        tex_rough = nodes.new("ShaderNodeTexImage")
        tex_rough.image = bpy.data.images.load(rough_files[0])
        tex_rough.image.colorspace_settings.name = 'Non-Color'
        links.new(tex_rough.outputs["Color"], bsdf.inputs["Roughness"])

    norm_files = glob.glob(os.path.join(grass_dir, "*Normal*.jpg"))
    if norm_files:
        tex_norm = nodes.new("ShaderNodeTexImage")
        tex_norm.image = bpy.data.images.load(norm_files[0])
        tex_norm.image.colorspace_settings.name = 'Non-Color'

        normal_map_node = nodes.new("ShaderNodeNormalMap")
        links.new(tex_norm.outputs["Color"], normal_map_node.inputs["Color"])
        links.new(normal_map_node.outputs["Normal"], bsdf.inputs["Normal"])

    return mat


def create_sky_material(sky_hdri_path):
    mat = bpy.data.materials.new(name="Sky_Dome_Material")
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    out_node = nodes.new("ShaderNodeOutputMaterial")
    emission = nodes.new("ShaderNodeEmission")
    emission.inputs["Strength"].default_value = 1.0

    links.new(emission.outputs["Emission"], out_node.inputs["Surface"])

    if os.path.exists(sky_hdri_path):
        tex_env = nodes.new("ShaderNodeTexEnvironment")
        tex_env.image = bpy.data.images.load(sky_hdri_path)
        links.new(tex_env.outputs["Color"], emission.inputs["Color"])

    return mat


def setup_grass_ground_geometry_nodes(grass_dir, sky_hdri_path, ground_z_offset=-0.2):
    grass_coll = bpy.data.collections.get("Grass_Artifacts_Collection")
    if not grass_coll:
        grass_coll = bpy.data.collections.new("Grass_Artifacts_Collection")
        bpy.context.scene.collection.children.link(grass_coll)

    grass_mat = create_grass_material(grass_dir)
    sky_mat = create_sky_material(sky_hdri_path)

    mesh_files = glob.glob(os.path.join(grass_dir, "*.obj")) + glob.glob(os.path.join(grass_dir, "*.fbx"))

    if mesh_files:
        for fpath in mesh_files:
            if fpath.endswith(".obj"):
                bpy.ops.wm.obj_import(filepath=fpath)
            elif fpath.endswith(".fbx"):
                bpy.ops.import_scene.fbx(filepath=fpath)

            for o in bpy.context.selected_objects:
                grass_coll.objects.link(o)
                if o.name in bpy.context.scene.collection.objects:
                    bpy.context.scene.collection.objects.unlink(o)

        for obj in list(grass_coll.objects):
            if obj.dimensions.z > 0.5:
                grass_coll.objects.unlink(obj)
                bpy.data.objects.remove(obj, do_unlink=True)
    else:
        for i in range(3):
            m_data = bpy.data.meshes.new(f"Grass_Blade_Mesh_{i}")
            g_obj = bpy.data.objects.new(f"Grass_Blade_Obj_{i}", m_data)
            grass_coll.objects.link(g_obj)

            verts = [(0, 0, 0), (0.05, 0, 0), (0.04, 0, 0.15), (0.01, 0, 0.25), (0, 0, 0.3)]
            faces = [(0, 1, 2), (1, 2, 3), (2, 3, 4)]
            m_data.from_pydata(verts, [], faces)
            m_data.update()

            if g_obj.data.materials:
                g_obj.data.materials[0] = grass_mat
            else:
                g_obj.data.materials.append(grass_mat)

            g_obj.location = (100, 100, 0)

    bpy.ops.mesh.primitive_circle_add(radius=12.5, vertices=64, fill_type='NGON', location=(0, 0, ground_z_offset))
    ground_obj = bpy.context.active_object
    ground_obj.name = "Ground_Mesh_Object"

    vg = ground_obj.vertex_groups.new(name="Grass_Weight_Group")
    all_vert_indices = [v.index for v in ground_obj.data.vertices]
    vg.add(all_vert_indices, 1.0, 'REPLACE')
    ground_obj.data.materials.append(grass_mat)
    ground_obj.data.materials.append(sky_mat)

    gn_group = bpy.data.node_groups.new("Grass_Ground_GN", type="GeometryNodeTree")
    gn_group.interface.new_socket(name="Geometry", in_out="INPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Geometry", in_out="OUTPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Density Mask", in_out="INPUT", socket_type="NodeSocketFloat")

    nodes = gn_group.nodes
    links = gn_group.links
    out_node = nodes.new("NodeGroupOutput")
    in_node = nodes.new("NodeGroupInput")

    distribute_node = nodes.new("GeometryNodeDistributePointsOnFaces")
    distribute_node.distribute_method = 'POISSON'
    distribute_node.inputs["Distance Min"].default_value = 0.02
    distribute_node.inputs["Density Max"].default_value = 300.0
    links.new(in_node.outputs["Geometry"], distribute_node.inputs["Mesh"])

    noise_node = nodes.new("ShaderNodeTexNoise")
    noise_node.inputs["Scale"].default_value = 2.5
    noise_node.inputs["Detail"].default_value = 2.0

    color_ramp = nodes.new("ShaderNodeValToRGB")
    color_ramp.color_ramp.elements[0].position = 0.20
    color_ramp.color_ramp.elements[1].position = 0.80
    links.new(noise_node.outputs["Fac"], color_ramp.inputs["Fac"])

    math_weight = nodes.new("ShaderNodeMath")
    math_weight.operation = "MULTIPLY"
    links.new(color_ramp.outputs["Color"], math_weight.inputs[0])
    links.new(in_node.outputs["Density Mask"], math_weight.inputs[1])

    links.new(math_weight.outputs["Value"], distribute_node.inputs["Density Factor"])

    instance_node = nodes.new("GeometryNodeInstanceOnPoints")
    instance_node.inputs["Pick Instance"].default_value = True
    links.new(distribute_node.outputs["Points"], instance_node.inputs["Points"])

    coll_info = nodes.new("GeometryNodeCollectionInfo")
    coll_info.inputs["Collection"].default_value = grass_coll
    coll_info.inputs["Separate Children"].default_value = True
    coll_info.inputs["Reset Children"].default_value = True
    links.new(coll_info.outputs["Instances"], instance_node.inputs["Instance"])

    rand_scale = nodes.new("FunctionNodeRandomValue")
    rand_scale.data_type = 'FLOAT'
    rand_scale.inputs["Min"].default_value = 0.25
    rand_scale.inputs["Max"].default_value = 0.50

    combine_scale = nodes.new("ShaderNodeCombineXYZ")
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["X"])
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["Y"])
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["Z"])
    links.new(combine_scale.outputs["Vector"], instance_node.inputs["Scale"])

    rand_rot_z = nodes.new("FunctionNodeRandomValue")
    rand_rot_z.data_type = 'FLOAT'
    rand_rot_z.inputs["Min"].default_value = 0.0
    rand_rot_z.inputs["Max"].default_value = math.radians(360.0)

    combine_rot = nodes.new("ShaderNodeCombineXYZ")
    links.new(rand_rot_z.outputs["Value"], combine_rot.inputs["Z"])
    links.new(combine_rot.outputs["Vector"], instance_node.inputs["Rotation"])

    sky_sphere = nodes.new("GeometryNodeMeshIcoSphere")
    sky_sphere.inputs["Radius"].default_value = 12.5
    sky_sphere.inputs["Subdivisions"].default_value = 4

    flip_faces = nodes.new("GeometryNodeFlipFaces")
    links.new(sky_sphere.outputs["Mesh"], flip_faces.inputs["Mesh"])

    transform_sky = nodes.new("GeometryNodeTransform")
    transform_sky.inputs["Translation"].default_value = (0.0, 0.0, ground_z_offset)
    links.new(flip_faces.outputs["Mesh"], transform_sky.inputs["Geometry"])

    set_sky_mat = nodes.new("GeometryNodeSetMaterial")
    set_sky_mat.inputs["Material"].default_value = sky_mat
    links.new(transform_sky.outputs["Geometry"], set_sky_mat.inputs["Geometry"])

    join_geo = nodes.new("GeometryNodeJoinGeometry")
    links.new(in_node.outputs["Geometry"], join_geo.inputs["Geometry"])
    links.new(instance_node.outputs["Instances"], join_geo.inputs["Geometry"])
    links.new(set_sky_mat.outputs["Geometry"], join_geo.inputs["Geometry"])

    links.new(join_geo.outputs["Geometry"], out_node.inputs["Geometry"])

    mod = ground_obj.modifiers.new(name="Grass_Ground_GN", type='NODES')
    mod.node_group = gn_group

    for item in gn_group.interface.items_tree:
        if getattr(item, "in_out", None) == "INPUT" and item.name == "Density Mask":
            mod[item.identifier + "_attribute_name"] = "Grass_Weight_Group"
            mod[item.identifier + "_use_attribute"] = True

    return ground_obj


# ==============================================================================
# SCROLL MATERIAL & GEOMETRY NODES SETUP
# ==============================================================================

def create_image_scroll_material(image_path):
    mat_name = "Scroll_Texture_Mat"
    mat = bpy.data.materials.get(mat_name)
    if mat:
        bpy.data.materials.remove(mat)

    mat = bpy.data.materials.new(name=mat_name)
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    output_node = nodes.new("ShaderNodeOutputMaterial")
    bsdf = nodes.new("ShaderNodeBsdfPrincipled")
    bsdf.inputs["Roughness"].default_value = 0.5

    tex_img = nodes.new("ShaderNodeTexImage")
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        tex_img.image = img
        tex_img.extension = 'REPEAT'

    uv_map_node = nodes.new("ShaderNodeUVMap")
    uv_map_node.uv_map = "UVMap"

    links.new(uv_map_node.outputs["UV"], tex_img.inputs["Vector"])
    links.new(tex_img.outputs["Color"], bsdf.inputs["Base Color"])
    links.new(bsdf.outputs["BSDF"], output_node.inputs["Surface"])

    return mat, tex_img.image if os.path.exists(image_path) else None


def create_pure_geometry_node_scroll(image_path=""):
    gn_group = bpy.data.node_groups.new("Scroll_Pure_GN", type="GeometryNodeTree")

    gn_group.interface.new_socket(name="Geometry", in_out="OUTPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Character Mesh", in_out="INPUT", socket_type="NodeSocketObject")

    sock_r_vg = gn_group.interface.new_socket(name="Right Hand Group", in_out="INPUT", socket_type="NodeSocketString")
    sock_r_vg.default_value = "mixamorig:RightHand"

    sock_l_vg = gn_group.interface.new_socket(name="Left Hand Group", in_out="INPUT", socket_type="NodeSocketString")
    sock_l_vg.default_value = "mixamorig:LeftHand"

    sock_offset = gn_group.interface.new_socket(name="Forward Offset", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_offset.default_value = 0.60

    sock_unfurl = gn_group.interface.new_socket(name="Unfurl", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_unfurl.default_value = 0.0

    sock_width = gn_group.interface.new_socket(name="Scroll Width", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_width.default_value = 0.35

    nodes = gn_group.nodes
    links = gn_group.links

    in_node = nodes.new("NodeGroupInput")
    out_node = nodes.new("NodeGroupOutput")

    char_info = nodes.new("GeometryNodeObjectInfo")
    char_info.transform_space = "RELATIVE"
    links.new(in_node.outputs["Character Mesh"], char_info.inputs["Object"])

    pos_node = nodes.new("GeometryNodeInputPosition")

    named_r = nodes.new("GeometryNodeInputNamedAttribute")
    named_r.data_type = "FLOAT"
    links.new(in_node.outputs["Right Hand Group"], named_r.inputs["Name"])

    cmp_r = nodes.new("FunctionNodeCompare")
    cmp_r.data_type = "FLOAT"
    cmp_r.operation = "GREATER_THAN"
    cmp_r.inputs[1].default_value = 0.001
    links.new(named_r.outputs["Attribute"], cmp_r.inputs[0])

    stat_r = nodes.new("GeometryNodeAttributeStatistic")
    stat_r.data_type = "FLOAT_VECTOR"
    links.new(char_info.outputs["Geometry"], stat_r.inputs["Geometry"])
    links.new(cmp_r.outputs["Result"], stat_r.inputs["Selection"])
    links.new(pos_node.outputs["Position"], stat_r.inputs["Attribute"])

    named_l = nodes.new("GeometryNodeInputNamedAttribute")
    named_l.data_type = "FLOAT"
    links.new(in_node.outputs["Left Hand Group"], named_l.inputs["Name"])

    cmp_l = nodes.new("FunctionNodeCompare")
    cmp_l.data_type = "FLOAT"
    cmp_l.operation = "GREATER_THAN"
    cmp_l.inputs[1].default_value = 0.001
    links.new(named_l.outputs["Attribute"], cmp_l.inputs[0])

    stat_l = nodes.new("GeometryNodeAttributeStatistic")
    stat_l.data_type = "FLOAT_VECTOR"
    links.new(char_info.outputs["Geometry"], stat_l.inputs["Geometry"])
    links.new(cmp_l.outputs["Result"], stat_l.inputs["Selection"])
    links.new(pos_node.outputs["Position"], stat_l.inputs["Attribute"])

    neg_offset = nodes.new("ShaderNodeMath")
    neg_offset.operation = "MULTIPLY"
    neg_offset.inputs[1].default_value = -1.0
    links.new(in_node.outputs["Forward Offset"], neg_offset.inputs[0])

    forward_vec = nodes.new("ShaderNodeCombineXYZ")
    forward_vec.inputs["X"].default_value = 0.0
    links.new(neg_offset.outputs["Value"], forward_vec.inputs["Y"])
    forward_vec.inputs["Z"].default_value = 0.0

    start_pos = nodes.new("ShaderNodeVectorMath")
    start_pos.operation = "ADD"
    links.new(stat_r.outputs["Mean"], start_pos.inputs[0])
    links.new(forward_vec.outputs["Vector"], start_pos.inputs[1])

    target_pos = nodes.new("ShaderNodeVectorMath")
    target_pos.operation = "ADD"
    links.new(stat_l.outputs["Mean"], target_pos.inputs[0])
    links.new(forward_vec.outputs["Vector"], target_pos.inputs[1])

    delta_vec = nodes.new("ShaderNodeVectorMath")
    delta_vec.operation = "SUBTRACT"
    links.new(target_pos.outputs["Vector"], delta_vec.inputs[0])
    links.new(start_pos.outputs["Vector"], delta_vec.inputs[1])

    dist_node = nodes.new("ShaderNodeVectorMath")
    dist_node.operation = "DISTANCE"
    links.new(start_pos.outputs["Vector"], dist_node.inputs[0])

    scale_vec = nodes.new("ShaderNodeVectorMath")
    scale_vec.operation = "SCALE"
    links.new(delta_vec.outputs["Vector"], scale_vec.inputs["Vector"])
    links.new(in_node.outputs["Unfurl"], scale_vec.inputs["Scale"])

    end_pos = nodes.new("ShaderNodeVectorMath")
    end_pos.operation = "ADD"
    links.new(start_pos.outputs["Vector"], end_pos.inputs[0])
    links.new(scale_vec.outputs["Vector"], end_pos.inputs[1])

    links.new(end_pos.outputs["Vector"], dist_node.inputs[1])

    curve_line = nodes.new("GeometryNodeCurvePrimitiveLine")
    links.new(start_pos.outputs["Vector"], curve_line.inputs["Start"])
    links.new(end_pos.outputs["Vector"], curve_line.inputs["End"])

    set_tilt = nodes.new("GeometryNodeSetCurveTilt")
    set_tilt.inputs["Tilt"].default_value = 1.5708
    links.new(curve_line.outputs["Curve"], set_tilt.inputs["Curve"])

    spline_param_main = nodes.new("GeometryNodeSplineParameter")
    store_u = nodes.new("GeometryNodeStoreNamedAttribute")
    store_u.inputs["Name"].default_value = "U_Factor"
    store_u.data_type = 'FLOAT'
    store_u.domain = 'POINT'
    links.new(set_tilt.outputs["Curve"], store_u.inputs["Geometry"])
    links.new(spline_param_main.outputs["Factor"], store_u.inputs["Value"])

    half_width_neg = nodes.new("ShaderNodeMath")
    half_width_neg.operation = "MULTIPLY"
    half_width_neg.inputs[1].default_value = -0.5
    links.new(in_node.outputs["Scroll Width"], half_width_neg.inputs[0])

    half_width_pos = nodes.new("ShaderNodeMath")
    half_width_pos.operation = "MULTIPLY"
    half_width_pos.inputs[1].default_value = 0.5
    links.new(in_node.outputs["Scroll Width"], half_width_pos.inputs[0])

    combine_start = nodes.new("ShaderNodeCombineXYZ")
    links.new(half_width_neg.outputs["Value"], combine_start.inputs["X"])

    combine_end = nodes.new("ShaderNodeCombineXYZ")
    links.new(half_width_pos.outputs["Value"], combine_end.inputs["X"])

    profile_line = nodes.new("GeometryNodeCurvePrimitiveLine")
    links.new(combine_start.outputs["Vector"], profile_line.inputs["Start"])
    links.new(combine_end.outputs["Vector"], profile_line.inputs["End"])

    spline_param_prof = nodes.new("GeometryNodeSplineParameter")
    store_v = nodes.new("GeometryNodeStoreNamedAttribute")
    store_v.inputs["Name"].default_value = "V_Factor"
    store_v.data_type = 'FLOAT'
    store_v.domain = 'POINT'
    links.new(profile_line.outputs["Curve"], store_v.inputs["Geometry"])
    links.new(spline_param_prof.outputs["Factor"], store_v.inputs["Value"])

    curve_to_mesh = nodes.new("GeometryNodeCurveToMesh")
    links.new(store_u.outputs["Geometry"], curve_to_mesh.inputs["Curve"])
    links.new(store_v.outputs["Geometry"], curve_to_mesh.inputs["Profile Curve"])

    img_aspect = 1.0
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        if img.size[1] > 0:
            img_aspect = img.size[0] / img.size[1]

    h_times_aspect = nodes.new("ShaderNodeMath")
    h_times_aspect.operation = "MULTIPLY"
    h_times_aspect.inputs[1].default_value = img_aspect
    links.new(in_node.outputs["Scroll Width"], h_times_aspect.inputs[0])

    tile_count = nodes.new("ShaderNodeMath")
    tile_count.operation = "DIVIDE"
    links.new(dist_node.outputs["Value"], tile_count.inputs[0])
    links.new(h_times_aspect.outputs["Value"], tile_count.inputs[1])

    read_u = nodes.new("GeometryNodeInputNamedAttribute")
    read_u.inputs["Name"].default_value = "U_Factor"
    read_u.data_type = 'FLOAT'

    scaled_u = nodes.new("ShaderNodeMath")
    scaled_u.operation = "MULTIPLY"
    links.new(read_u.outputs["Attribute"], scaled_u.inputs[0])
    links.new(tile_count.outputs["Value"], scaled_u.inputs[1])

    read_v = nodes.new("GeometryNodeInputNamedAttribute")
    read_v.inputs["Name"].default_value = "V_Factor"
    read_v.data_type = 'FLOAT'

    flip_v = nodes.new("ShaderNodeMath")
    flip_v.operation = "SUBTRACT"
    flip_v.inputs[0].default_value = 1.0
    links.new(read_v.outputs["Attribute"], flip_v.inputs[1])

    combine_uv = nodes.new("ShaderNodeCombineXYZ")
    links.new(scaled_u.outputs["Value"], combine_uv.inputs["X"])
    links.new(flip_v.outputs["Value"], combine_uv.inputs["Y"])

    store_uv = nodes.new("GeometryNodeStoreNamedAttribute")
    store_uv.inputs["Name"].default_value = "UVMap"
    store_uv.data_type = 'FLOAT2'
    store_uv.domain = 'CORNER'
    links.new(curve_to_mesh.outputs["Mesh"], store_uv.inputs["Geometry"])
    links.new(combine_uv.outputs["Vector"], store_uv.inputs["Value"])

    set_material = nodes.new("GeometryNodeSetMaterial")
    scroll_mat, _ = create_image_scroll_material(image_path)
    set_material.inputs["Material"].default_value = scroll_mat
    links.new(store_uv.outputs["Geometry"], set_material.inputs["Geometry"])

    links.new(set_material.outputs["Geometry"], out_node.inputs["Geometry"])

    return gn_group, scroll_mat


def setup_geometry_node_scroll(target_mesh, unfurl_start, unfurl_end, image_path=""):
    mesh_data = bpy.data.meshes.new("Scroll_Procedural_Mesh")
    scroll_obj = bpy.data.objects.new("Scroll_GN_Object", mesh_data)
    bpy.context.scene.collection.objects.link(scroll_obj)

    mod = scroll_obj.modifiers.new("Scroll_GN", type="NODES")
    gn_tree, scroll_mat = create_pure_geometry_node_scroll(image_path=image_path)
    mod.node_group = gn_tree

    scroll_obj.data.materials.append(scroll_mat)

    r_hand_name, l_hand_name = find_hand_vertex_groups(target_mesh)

    unfurl_id = None
    for item in gn_tree.interface.items_tree:
        if getattr(item, "in_out", None) == "INPUT":
            if item.name == "Character Mesh":
                mod[item.identifier] = target_mesh
            elif item.name == "Right Hand Group":
                mod[item.identifier] = r_hand_name
            elif item.name == "Left Hand Group":
                mod[item.identifier] = l_hand_name
            elif item.name == "Forward Offset":
                mod[item.identifier] = 0.60
            elif item.name == "Scroll Width":
                mod[item.identifier] = 0.35
            elif item.name == "Unfurl":
                unfurl_id = item.identifier

    for m in target_mesh.modifiers:
        if m.type == 'ARMATURE':
            m.show_viewport = True
            m.show_render = True
            m.show_in_editmode = True

    scroll_obj.hide_viewport = True
    scroll_obj.hide_render = True
    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=1)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=1)

    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=unfurl_start - 1)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=unfurl_start - 1)

    scroll_obj.hide_viewport = False
    scroll_obj.hide_render = False
    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=unfurl_start)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=unfurl_start)

    for fc in get_object_fcurves(scroll_obj):
        if fc.data_path in ["hide_viewport", "hide_render"]:
            for kp in fc.keyframe_points:
                kp.interpolation = 'CONSTANT'

    if unfurl_id:
        mod[unfurl_id] = 0.001
        mod.keyframe_insert(data_path=f'["{unfurl_id}"]', frame=unfurl_start)

        mod[unfurl_id] = 1.0
        mod.keyframe_insert(data_path=f'["{unfurl_id}"]', frame=unfurl_end)

    return scroll_obj


# ==============================================================================
# MAIN CONTROLLER
# ==============================================================================

def import_and_extract_action(fbx_path, action_name):
    existing_objects = set(bpy.data.objects)
    bpy.ops.import_scene.fbx(filepath=fbx_path)
    bpy.context.view_layer.update()

    imported_objects = set(bpy.data.objects) - existing_objects
    extracted_action = None
    for obj in imported_objects:
        if obj.animation_data and obj.animation_data.action:
            extracted_action = obj.animation_data.action
            extracted_action.name = action_name
            extracted_action.use_fake_user = True
            break
    return extracted_action, imported_objects


def purge_imported_objects(objects_to_remove):
    for obj in objects_to_remove:
        if obj.name in bpy.data.objects:
            if obj.animation_data:
                obj.animation_data_clear()
            for coll in list(obj.users_collection):
                coll.objects.unlink(obj)
    bpy.context.view_layer.update()


def main():
    cleanup_previous_run()

    walk_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Mutant Walking.fbx"
    turn_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Right Turn 90.fbx"
    unfurl_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Mutant Idle.fbx"

    sky_hdri_path = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/horn-koppe_spring_4k.exr"
    output_blend = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/output_scene.blend"

    image_path = "/Users/dobbinp1/Downloads/IMGP6691_psedit.jpg"
    grass_dir = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Grass004_4K-JPG"

    setup_grass_ground_geometry_nodes(grass_dir, sky_hdri_path, ground_z_offset=-0.2)

    CYCLES = 4
    CROSSFADE = 15
    TARGET_DISTANCE = 4.0

    walk_action, walk_objs = import_and_extract_action(walk_fbx, "walk")
    target_rig = next((obj for obj in walk_objs if obj.type == "ARMATURE"), None)
    target_mesh = next((obj for obj in walk_objs if obj.type == "MESH"), None)

    if not target_rig or not target_mesh:
        raise RuntimeError("Armature or Character Mesh not found in walk FBX.")

    travel_axis_idx, motion_sign = detect_natural_forward_axis(walk_action, target_rig)
    initial_heading = math.radians(90.0) if travel_axis_idx == 0 else 0.0
    target_rig.rotation_euler[2] = initial_heading

    action_len = int(walk_action.frame_range[1] - walk_action.frame_range[0])
    if action_len <= 0:
        action_len = 30

    anim_data = target_rig.animation_data_create()

    walk_track = anim_data.nla_tracks.new()
    walk_track.name = "Walk_Track"

    current_frame = 1
    for i in range(CYCLES):
        strip = walk_track.strips.new(name=f"Walk_{i+1}", start=int(current_frame), action=walk_action)
        strip.blend_type = "REPLACE"
        current_frame += action_len

    walk_end_frame = int(current_frame)

    turn_action, turn_objs = import_and_extract_action(turn_fbx, "rightturn")
    purge_imported_objects(turn_objs)

    turn_track = anim_data.nla_tracks.new()
    turn_track.name = "RightTurn_Track"

    turn_start = int(walk_end_frame - CROSSFADE)
    turn_strip = turn_track.strips.new(name="RightTurn", start=turn_start, action=turn_action)
    turn_strip.blend_type = "REPLACE"
    turn_strip.blend_in = CROSSFADE
    turn_strip.use_auto_blend = True
    turn_strip.extrapolation = "HOLD"
    turn_end_frame = int(turn_strip.frame_end)

    unfurl_action, unfurl_objs = import_and_extract_action(unfurl_fbx, "unfurl")
    purge_imported_objects(unfurl_objs)

    unfurl_track = anim_data.nla_tracks.new()
    unfurl_track.name = "Unfurl_Track"

    unfurl_start = int(turn_end_frame - CROSSFADE)
    unfurl_strip = unfurl_track.strips.new(name="Unfurl", start=unfurl_start, action=unfurl_action)
    unfurl_strip.blend_type = "REPLACE"
    unfurl_strip.blend_in = CROSSFADE
    unfurl_strip.use_auto_blend = True
    unfurl_strip.extrapolation = "HOLD"
    unfurl_end_frame = int(unfurl_strip.frame_end)

    root_action = bpy.data.actions.new(name="World_Root_Motion")
    target_rig.animation_data.action = root_action

    loc_fcurve = root_action.fcurve_ensure_for_datablock(datablock=target_rig, data_path="location", index=travel_axis_idx)
    final_target_value = TARGET_DISTANCE * motion_sign

    k_loc1 = loc_fcurve.keyframe_points.insert(frame=1.0, value=0.0)
    k_loc2 = loc_fcurve.keyframe_points.insert(frame=float(turn_start), value=final_target_value)
    k_loc3 = loc_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=final_target_value)

    k_loc1.interpolation = "LINEAR"
    k_loc2.interpolation = "CONSTANT"
    k_loc3.interpolation = "CONSTANT"

    rot_fcurve = root_action.fcurve_ensure_for_datablock(datablock=target_rig, data_path="rotation_euler", index=2)
    target_heading = initial_heading - math.radians(90.0)

    k_rot1 = rot_fcurve.keyframe_points.insert(frame=1.0, value=initial_heading)
    k_rot2 = rot_fcurve.keyframe_points.insert(frame=float(turn_start), value=initial_heading)
    k_rot3 = rot_fcurve.keyframe_points.insert(frame=float(turn_end_frame), value=target_heading)
    k_rot4 = rot_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=target_heading)

    k_rot1.interpolation = "CONSTANT"
    k_rot2.interpolation = "BEZIER"
    k_rot3.interpolation = "CONSTANT"
    k_rot4.interpolation = "CONSTANT"

    root_track = anim_data.nla_tracks.new()
    root_track.name = "World_Root_Track"
    root_strip = root_track.strips.new(name="Root_Motion_Strip", start=1, action=root_action)
    root_strip.blend_type = "REPLACE"
    root_strip.extrapolation = "HOLD"

    anim_data.action = None

    setup_geometry_node_scroll(
        target_mesh=target_mesh,
        unfurl_start=unfurl_start,
        unfurl_end=unfurl_end_frame,
        image_path=image_path,
    )

    # Modern Instagram Reel Vertical Camera Setup
    setup_modern_reel_camera(
        target_rig=target_rig,
        target_mesh=target_mesh,
        turn_start=turn_start,
        turn_end_frame=turn_end_frame,
        unfurl_end_frame=unfurl_end_frame
    )

    bpy.context.scene.frame_start = 1
    bpy.context.scene.frame_end = unfurl_end_frame

    bpy.ops.wm.save_as_mainfile(filepath=output_blend)
    print(f"\n[Success] Modern vertical reel camera added. Saved to: {output_blend}")


main()
```

```
import math
import os
import glob
import bpy
from bpy_extras import anim_utils

# ==============================================================================
# CLEANUP & HELPERS
# ==============================================================================

def cleanup_previous_run():
    """Clears old scroll objects, ground, node groups, camera anchors, collections, and orphaned data."""
    target_names = [
        "Master_NLA_Control",
        "Scroll_GN_Object",
        "Scroll_Start_Pin",
        "Scroll_End_Pin",
        "Scroll_Master_Container",
        "HDRI_Ground_Plane",
        "Ground_Mesh_Object",
        "Camera_Focus_Target",
        "Vertical_Reel_Camera"
    ]
    for obj in list(bpy.data.objects):
        if obj.name in target_names or (obj.name.startswith("mixamorig") and obj.type != "ARMATURE" and obj.type != "MESH"):
            if obj.animation_data:
                obj.animation_data_clear()
            bpy.data.objects.remove(obj, do_unlink=True)

    for action in list(bpy.data.actions):
        if action.users == 0 or action.use_fake_user:
            action.use_fake_user = False
            bpy.data.actions.remove(action)

    for ng in list(bpy.data.node_groups):
        if ng.name in ["Scroll_Unfurl_GN", "Scroll_Pure_GN", "HDRI_Ground_GN", "Grass_Ground_GN"]:
            bpy.data.node_groups.remove(ng)

    for coll_name in ["Scroll_Master_Container", "Grass_Artifacts_Collection"]:
        if coll_name in bpy.data.collections:
            coll = bpy.data.collections[coll_name]
            for obj in list(coll.objects):
                bpy.data.objects.remove(obj, do_unlink=True)
            bpy.data.collections.remove(coll)

    bpy.context.view_layer.update()


def get_slot_fcurves(action, anim_target_object):
    fcurves = []
    if not action:
        return fcurves

    anim_data = getattr(anim_target_object, "animation_data", None)
    if anim_data and getattr(anim_data, "action_slot", None):
        slot = anim_data.action_slot
        channelbag = anim_utils.action_get_channelbag_for_slot(action, slot)
        if channelbag and hasattr(channelbag, "fcurves"):
            return list(channelbag.fcurves)

    if hasattr(action, "slots"):
        for slot in action.slots:
            for cb in getattr(slot, "channelbags", []):
                fcurves.extend(getattr(cb, "fcurves", []))
        if fcurves:
            return fcurves

    if hasattr(action, "fcurves"):
        return list(action.fcurves)

    return fcurves


def get_object_fcurves(obj):
    if not obj or not obj.animation_data or not obj.animation_data.action:
        return []
    action = obj.animation_data.action
    return get_slot_fcurves(action, obj)


def detect_natural_forward_axis(action, rig_object):
    fcurves = get_slot_fcurves(action, rig_object)
    max_delta = 0.0
    best_axis = 1
    net_delta = 0.0

    start_frame = action.frame_range[0]
    end_frame = action.frame_range[1]

    for fc in fcurves:
        if fc.data_path.endswith("location") and fc.array_index in [0, 1, 2]:
            start_val = fc.evaluate(start_frame)
            end_val = fc.evaluate(end_frame)
            delta = end_val - start_val
            if abs(delta) > abs(max_delta):
                max_delta = delta
                best_axis = fc.array_index
                net_delta = delta

    direction_sign = 1.0 if net_delta >= 0 else -1.0
    return best_axis, direction_sign


def find_hand_vertex_groups(target_mesh):
    r_name = "mixamorig:RightHand"
    l_name = "mixamorig:LeftHand"

    if target_mesh and hasattr(target_mesh, "vertex_groups"):
        for vg in target_mesh.vertex_groups:
            vg_lower = vg.name.lower()
            if "righthand" in vg_lower or "hand_r" in vg_lower or "hand.r" in vg_lower:
                r_name = vg.name
            elif "lefthand" in vg_lower or "hand_l" in vg_lower or "hand.l" in vg_lower:
                l_name = vg.name

    return r_name, l_name


# ==============================================================================
# LOCAL VERTICAL CAMERA SETUP (45° SHOULDER VIEW)
# ==============================================================================

def setup_modern_reel_camera(target_rig, target_mesh, turn_start, turn_end_frame, unfurl_end_frame):
    """
    Creates a 9:16 Instagram Reel Camera vertical relative to the monster:
    - Camera is parented to the target rig to remain vertical relative to the monster.
    - Walk Phase: Fixed at a 45° offset relative to the front-facing shoulder view.
    - Turn Phase: Transitions locally to 0° (direct front-facing view).
    """
    scene = bpy.context.scene
    scene.render.resolution_x = 1080
    scene.render.resolution_y = 1920
    scene.render.resolution_percentage = 100

    # 1. Target Focus Null (Placed at chest height relative to monster)
    target_height = target_mesh.dimensions.z if target_mesh.dimensions.z > 0 else 1.8
    chest_z_offset = target_height * 0.60

    focus_target = bpy.data.objects.get("Camera_Focus_Target")
    if not focus_target:
        focus_target = bpy.data.objects.new("Camera_Focus_Target", None)
        focus_target.empty_display_type = 'PLAIN_AXES'
        focus_target.empty_display_size = 0.25
        scene.collection.objects.link(focus_target)

    # Parent target empty directly to character rig so local vertical Z is identical
    focus_target.parent = target_rig
    focus_target.location = (0.0, 0.0, chest_z_offset)

    # 2. Camera Setup (Vertical 9:16 Aspect)
    cam_data = bpy.data.cameras.new("Vertical_Reel_Cam_Data")
    cam_data.lens = 45.0
    cam_data.sensor_fit = 'VERTICAL'
    cam_data.clip_end = 100.0

    cam_obj = bpy.data.objects.new("Vertical_Reel_Camera", cam_data)
    scene.collection.objects.link(cam_obj)
    scene.camera = cam_obj

    # Parent camera to focus target to inherit upright orientation
    cam_obj.parent = focus_target

    # 3. Damped Track Constraint (Focuses camera sightline on chest target)
    damped_track = cam_obj.constraints.new(type='DAMPED_TRACK')
    damped_track.target = focus_target
    damped_track.track_axis = 'TRACK_NEGATIVE_Z'

    # 4. Local Offset Calculation (Monster local space: facing -Y, left is -X)
    dist = 6.0

    # Walk Phase Offset: 45° toward the shoulder
    walk_angle = math.radians(-45.0)
    walk_local_x = dist * math.sin(walk_angle)
    walk_local_y = -dist * math.cos(walk_angle)

    # Turn/Front View Offset: 0° direct front view
    front_local_x = 0.0
    front_local_y = -dist

    # 5. Local Keyframing
    if cam_obj.animation_data:
        cam_obj.animation_data_clear()

    # Walk Phase (Frame 1 to turn_start): Hold local 45° shoulder position
    cam_obj.location = (walk_local_x, walk_local_y, 0.0)
    cam_obj.keyframe_insert(data_path="location", frame=1)
    cam_obj.keyframe_insert(data_path="location", frame=turn_start)

    # Turn Phase (turn_start to turn_end_frame): Smoothly swing to direct front (0°)
    cam_obj.location = (front_local_x, front_local_y, 0.0)
    cam_obj.keyframe_insert(data_path="location", frame=turn_end_frame)
    cam_obj.keyframe_insert(data_path="location", frame=unfurl_end_frame)

    # Apply smooth interpolation curves
    for fc in get_object_fcurves(cam_obj):
        for kp in fc.keyframe_points:
            kp.interpolation = 'BEZIER'

    return cam_obj, focus_target


# ==============================================================================
# GEOMETRY NODES GROUND, GRASS & SKY DOME SETUP
# ==============================================================================

def create_grass_material(grass_dir):
    mat = bpy.data.materials.new(name="Grass_PBR_Material")
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    out_node = nodes.new("ShaderNodeOutputMaterial")
    bsdf = nodes.new("ShaderNodeBsdfPrincipled")
    links.new(bsdf.outputs["BSDF"], out_node.inputs["Surface"])

    col_files = glob.glob(os.path.join(grass_dir, "*Color*.jpg")) + glob.glob(os.path.join(grass_dir, "*JPG*.jpg"))
    if col_files:
        tex_col = nodes.new("ShaderNodeTexImage")
        tex_col.image = bpy.data.images.load(col_files[0])
        links.new(tex_col.outputs["Color"], bsdf.inputs["Base Color"])

    rough_files = glob.glob(os.path.join(grass_dir, "*Roughness*.jpg"))
    if rough_files:
        tex_rough = nodes.new("ShaderNodeTexImage")
        tex_rough.image = bpy.data.images.load(rough_files[0])
        tex_rough.image.colorspace_settings.name = 'Non-Color'
        links.new(tex_rough.outputs["Color"], bsdf.inputs["Roughness"])

    norm_files = glob.glob(os.path.join(grass_dir, "*Normal*.jpg"))
    if norm_files:
        tex_norm = nodes.new("ShaderNodeTexImage")
        tex_norm.image = bpy.data.images.load(norm_files[0])
        tex_norm.image.colorspace_settings.name = 'Non-Color'

        normal_map_node = nodes.new("ShaderNodeNormalMap")
        links.new(tex_norm.outputs["Color"], normal_map_node.inputs["Color"])
        links.new(normal_map_node.outputs["Normal"], bsdf.inputs["Normal"])

    return mat


def create_sky_material(sky_hdri_path):
    mat = bpy.data.materials.new(name="Sky_Dome_Material")
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    out_node = nodes.new("ShaderNodeOutputMaterial")
    emission = nodes.new("ShaderNodeEmission")
    emission.inputs["Strength"].default_value = 1.0

    links.new(emission.outputs["Emission"], out_node.inputs["Surface"])

    if os.path.exists(sky_hdri_path):
        tex_env = nodes.new("ShaderNodeTexEnvironment")
        tex_env.image = bpy.data.images.load(sky_hdri_path)
        links.new(tex_env.outputs["Color"], emission.inputs["Color"])

    return mat


def setup_grass_ground_geometry_nodes(grass_dir, sky_hdri_path, ground_z_offset=-0.2):
    grass_coll = bpy.data.collections.get("Grass_Artifacts_Collection")
    if not grass_coll:
        grass_coll = bpy.data.collections.new("Grass_Artifacts_Collection")
        bpy.context.scene.collection.children.link(grass_coll)

    grass_mat = create_grass_material(grass_dir)
    sky_mat = create_sky_material(sky_hdri_path)

    mesh_files = glob.glob(os.path.join(grass_dir, "*.obj")) + glob.glob(os.path.join(grass_dir, "*.fbx"))

    if mesh_files:
        for fpath in mesh_files:
            if fpath.endswith(".obj"):
                bpy.ops.wm.obj_import(filepath=fpath)
            elif fpath.endswith(".fbx"):
                bpy.ops.import_scene.fbx(filepath=fpath)

            for o in bpy.context.selected_objects:
                grass_coll.objects.link(o)
                if o.name in bpy.context.scene.collection.objects:
                    bpy.context.scene.collection.objects.unlink(o)

        for obj in list(grass_coll.objects):
            if obj.dimensions.z > 0.5:
                grass_coll.objects.unlink(obj)
                bpy.data.objects.remove(obj, do_unlink=True)
    else:
        for i in range(3):
            m_data = bpy.data.meshes.new(f"Grass_Blade_Mesh_{i}")
            g_obj = bpy.data.objects.new(f"Grass_Blade_Obj_{i}", m_data)
            grass_coll.objects.link(g_obj)

            verts = [(0, 0, 0), (0.05, 0, 0), (0.04, 0, 0.15), (0.01, 0, 0.25), (0, 0, 0.3)]
            faces = [(0, 1, 2), (1, 2, 3), (2, 3, 4)]
            m_data.from_pydata(verts, [], faces)
            m_data.update()

            if g_obj.data.materials:
                g_obj.data.materials[0] = grass_mat
            else:
                g_obj.data.materials.append(grass_mat)

            g_obj.location = (100, 100, 0)

    bpy.ops.mesh.primitive_circle_add(radius=12.5, vertices=64, fill_type='NGON', location=(0, 0, ground_z_offset))
    ground_obj = bpy.context.active_object
    ground_obj.name = "Ground_Mesh_Object"

    vg = ground_obj.vertex_groups.new(name="Grass_Weight_Group")
    all_vert_indices = [v.index for v in ground_obj.data.vertices]
    vg.add(all_vert_indices, 1.0, 'REPLACE')
    ground_obj.data.materials.append(grass_mat)
    ground_obj.data.materials.append(sky_mat)

    gn_group = bpy.data.node_groups.new("Grass_Ground_GN", type="GeometryNodeTree")
    gn_group.interface.new_socket(name="Geometry", in_out="INPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Geometry", in_out="OUTPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Density Mask", in_out="INPUT", socket_type="NodeSocketFloat")

    nodes = gn_group.nodes
    links = gn_group.links
    out_node = nodes.new("NodeGroupOutput")
    in_node = nodes.new("NodeGroupInput")

    distribute_node = nodes.new("GeometryNodeDistributePointsOnFaces")
    distribute_node.distribute_method = 'POISSON'
    distribute_node.inputs["Distance Min"].default_value = 0.02
    distribute_node.inputs["Density Max"].default_value = 300.0
    links.new(in_node.outputs["Geometry"], distribute_node.inputs["Mesh"])

    noise_node = nodes.new("ShaderNodeTexNoise")
    noise_node.inputs["Scale"].default_value = 2.5
    noise_node.inputs["Detail"].default_value = 2.0

    color_ramp = nodes.new("ShaderNodeValToRGB")
    color_ramp.color_ramp.elements[0].position = 0.20
    color_ramp.color_ramp.elements[1].position = 0.80
    links.new(noise_node.outputs["Fac"], color_ramp.inputs["Fac"])

    math_weight = nodes.new("ShaderNodeMath")
    math_weight.operation = "MULTIPLY"
    links.new(color_ramp.outputs["Color"], math_weight.inputs[0])
    links.new(in_node.outputs["Density Mask"], math_weight.inputs[1])

    links.new(math_weight.outputs["Value"], distribute_node.inputs["Density Factor"])

    instance_node = nodes.new("GeometryNodeInstanceOnPoints")
    instance_node.inputs["Pick Instance"].default_value = True
    links.new(distribute_node.outputs["Points"], instance_node.inputs["Points"])

    coll_info = nodes.new("GeometryNodeCollectionInfo")
    coll_info.inputs["Collection"].default_value = grass_coll
    coll_info.inputs["Separate Children"].default_value = True
    coll_info.inputs["Reset Children"].default_value = True
    links.new(coll_info.outputs["Instances"], instance_node.inputs["Instance"])

    rand_scale = nodes.new("FunctionNodeRandomValue")
    rand_scale.data_type = 'FLOAT'
    rand_scale.inputs["Min"].default_value = 0.25
    rand_scale.inputs["Max"].default_value = 0.50

    combine_scale = nodes.new("ShaderNodeCombineXYZ")
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["X"])
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["Y"])
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["Z"])
    links.new(combine_scale.outputs["Vector"], instance_node.inputs["Scale"])

    rand_rot_z = nodes.new("FunctionNodeRandomValue")
    rand_rot_z.data_type = 'FLOAT'
    rand_rot_z.inputs["Min"].default_value = 0.0
    rand_rot_z.inputs["Max"].default_value = math.radians(360.0)

    combine_rot = nodes.new("ShaderNodeCombineXYZ")
    links.new(rand_rot_z.outputs["Value"], combine_rot.inputs["Z"])
    links.new(combine_rot.outputs["Vector"], instance_node.inputs["Rotation"])

    sky_sphere = nodes.new("GeometryNodeMeshIcoSphere")
    sky_sphere.inputs["Radius"].default_value = 12.5
    sky_sphere.inputs["Subdivisions"].default_value = 4

    flip_faces = nodes.new("GeometryNodeFlipFaces")
    links.new(sky_sphere.outputs["Mesh"], flip_faces.inputs["Mesh"])

    transform_sky = nodes.new("GeometryNodeTransform")
    transform_sky.inputs["Translation"].default_value = (0.0, 0.0, ground_z_offset)
    links.new(flip_faces.outputs["Mesh"], transform_sky.inputs["Geometry"])

    set_sky_mat = nodes.new("GeometryNodeSetMaterial")
    set_sky_mat.inputs["Material"].default_value = sky_mat
    links.new(transform_sky.outputs["Geometry"], set_sky_mat.inputs["Geometry"])

    join_geo = nodes.new("GeometryNodeJoinGeometry")
    links.new(in_node.outputs["Geometry"], join_geo.inputs["Geometry"])
    links.new(instance_node.outputs["Instances"], join_geo.inputs["Geometry"])
    links.new(set_sky_mat.outputs["Geometry"], join_geo.inputs["Geometry"])

    links.new(join_geo.outputs["Geometry"], out_node.inputs["Geometry"])

    mod = ground_obj.modifiers.new(name="Grass_Ground_GN", type='NODES')
    mod.node_group = gn_group

    for item in gn_group.interface.items_tree:
        if getattr(item, "in_out", None) == "INPUT" and item.name == "Density Mask":
            mod[item.identifier + "_attribute_name"] = "Grass_Weight_Group"
            mod[item.identifier + "_use_attribute"] = True

    return ground_obj


# ==============================================================================
# SCROLL MATERIAL & GEOMETRY NODES SETUP
# ==============================================================================

def create_image_scroll_material(image_path):
    mat_name = "Scroll_Texture_Mat"
    mat = bpy.data.materials.get(mat_name)
    if mat:
        bpy.data.materials.remove(mat)

    mat = bpy.data.materials.new(name=mat_name)
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    output_node = nodes.new("ShaderNodeOutputMaterial")
    bsdf = nodes.new("ShaderNodeBsdfPrincipled")
    bsdf.inputs["Roughness"].default_value = 0.5

    tex_img = nodes.new("ShaderNodeTexImage")
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        tex_img.image = img
        tex_img.extension = 'REPEAT'

    uv_map_node = nodes.new("ShaderNodeUVMap")
    uv_map_node.uv_map = "UVMap"

    links.new(uv_map_node.outputs["UV"], tex_img.inputs["Vector"])
    links.new(tex_img.outputs["Color"], bsdf.inputs["Base Color"])
    links.new(bsdf.outputs["BSDF"], output_node.inputs["Surface"])

    return mat, tex_img.image if os.path.exists(image_path) else None


def create_pure_geometry_node_scroll(image_path=""):
    gn_group = bpy.data.node_groups.new("Scroll_Pure_GN", type="GeometryNodeTree")

    gn_group.interface.new_socket(name="Geometry", in_out="OUTPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Character Mesh", in_out="INPUT", socket_type="NodeSocketObject")

    sock_r_vg = gn_group.interface.new_socket(name="Right Hand Group", in_out="INPUT", socket_type="NodeSocketString")
    sock_r_vg.default_value = "mixamorig:RightHand"

    sock_l_vg = gn_group.interface.new_socket(name="Left Hand Group", in_out="INPUT", socket_type="NodeSocketString")
    sock_l_vg.default_value = "mixamorig:LeftHand"

    sock_offset = gn_group.interface.new_socket(name="Forward Offset", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_offset.default_value = 0.60

    sock_unfurl = gn_group.interface.new_socket(name="Unfurl", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_unfurl.default_value = 0.0

    sock_width = gn_group.interface.new_socket(name="Scroll Width", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_width.default_value = 0.35

    nodes = gn_group.nodes
    links = gn_group.links

    in_node = nodes.new("NodeGroupInput")
    out_node = nodes.new("NodeGroupOutput")

    char_info = nodes.new("GeometryNodeObjectInfo")
    char_info.transform_space = "RELATIVE"
    links.new(in_node.outputs["Character Mesh"], char_info.inputs["Object"])

    pos_node = nodes.new("GeometryNodeInputPosition")

    named_r = nodes.new("GeometryNodeInputNamedAttribute")
    named_r.data_type = "FLOAT"
    links.new(in_node.outputs["Right Hand Group"], named_r.inputs["Name"])

    cmp_r = nodes.new("FunctionNodeCompare")
    cmp_r.data_type = "FLOAT"
    cmp_r.operation = "GREATER_THAN"
    cmp_r.inputs[1].default_value = 0.001
    links.new(named_r.outputs["Attribute"], cmp_r.inputs[0])

    stat_r = nodes.new("GeometryNodeAttributeStatistic")
    stat_r.data_type = "FLOAT_VECTOR"
    links.new(char_info.outputs["Geometry"], stat_r.inputs["Geometry"])
    links.new(cmp_r.outputs["Result"], stat_r.inputs["Selection"])
    links.new(pos_node.outputs["Position"], stat_r.inputs["Attribute"])

    named_l = nodes.new("GeometryNodeInputNamedAttribute")
    named_l.data_type = "FLOAT"
    links.new(in_node.outputs["Left Hand Group"], named_l.inputs["Name"])

    cmp_l = nodes.new("FunctionNodeCompare")
    cmp_l.data_type = "FLOAT"
    cmp_l.operation = "GREATER_THAN"
    cmp_l.inputs[1].default_value = 0.001
    links.new(named_l.outputs["Attribute"], cmp_l.inputs[0])

    stat_l = nodes.new("GeometryNodeAttributeStatistic")
    stat_l.data_type = "FLOAT_VECTOR"
    links.new(char_info.outputs["Geometry"], stat_l.inputs["Geometry"])
    links.new(cmp_l.outputs["Result"], stat_l.inputs["Selection"])
    links.new(pos_node.outputs["Position"], stat_l.inputs["Attribute"])

    neg_offset = nodes.new("ShaderNodeMath")
    neg_offset.operation = "MULTIPLY"
    neg_offset.inputs[1].default_value = -1.0
    links.new(in_node.outputs["Forward Offset"], neg_offset.inputs[0])

    forward_vec = nodes.new("ShaderNodeCombineXYZ")
    forward_vec.inputs["X"].default_value = 0.0
    links.new(neg_offset.outputs["Value"], forward_vec.inputs["Y"])
    forward_vec.inputs["Z"].default_value = 0.0

    start_pos = nodes.new("ShaderNodeVectorMath")
    start_pos.operation = "ADD"
    links.new(stat_r.outputs["Mean"], start_pos.inputs[0])
    links.new(forward_vec.outputs["Vector"], start_pos.inputs[1])

    target_pos = nodes.new("ShaderNodeVectorMath")
    target_pos.operation = "ADD"
    links.new(stat_l.outputs["Mean"], target_pos.inputs[0])
    links.new(forward_vec.outputs["Vector"], target_pos.inputs[1])

    delta_vec = nodes.new("ShaderNodeVectorMath")
    delta_vec.operation = "SUBTRACT"
    links.new(target_pos.outputs["Vector"], delta_vec.inputs[0])
    links.new(start_pos.outputs["Vector"], delta_vec.inputs[1])

    dist_node = nodes.new("ShaderNodeVectorMath")
    dist_node.operation = "DISTANCE"
    links.new(start_pos.outputs["Vector"], dist_node.inputs[0])

    scale_vec = nodes.new("ShaderNodeVectorMath")
    scale_vec.operation = "SCALE"
    links.new(delta_vec.outputs["Vector"], scale_vec.inputs["Vector"])
    links.new(in_node.outputs["Unfurl"], scale_vec.inputs["Scale"])

    end_pos = nodes.new("ShaderNodeVectorMath")
    end_pos.operation = "ADD"
    links.new(start_pos.outputs["Vector"], end_pos.inputs[0])
    links.new(scale_vec.outputs["Vector"], end_pos.inputs[1])

    links.new(end_pos.outputs["Vector"], dist_node.inputs[1])

    curve_line = nodes.new("GeometryNodeCurvePrimitiveLine")
    links.new(start_pos.outputs["Vector"], curve_line.inputs["Start"])
    links.new(end_pos.outputs["Vector"], curve_line.inputs["End"])

    set_tilt = nodes.new("GeometryNodeSetCurveTilt")
    set_tilt.inputs["Tilt"].default_value = 1.5708
    links.new(curve_line.outputs["Curve"], set_tilt.inputs["Curve"])

    spline_param_main = nodes.new("GeometryNodeSplineParameter")
    store_u = nodes.new("GeometryNodeStoreNamedAttribute")
    store_u.inputs["Name"].default_value = "U_Factor"
    store_u.data_type = 'FLOAT'
    store_u.domain = 'POINT'
    links.new(set_tilt.outputs["Curve"], store_u.inputs["Geometry"])
    links.new(spline_param_main.outputs["Factor"], store_u.inputs["Value"])

    half_width_neg = nodes.new("ShaderNodeMath")
    half_width_neg.operation = "MULTIPLY"
    half_width_neg.inputs[1].default_value = -0.5
    links.new(in_node.outputs["Scroll Width"], half_width_neg.inputs[0])

    half_width_pos = nodes.new("ShaderNodeMath")
    half_width_pos.operation = "MULTIPLY"
    half_width_pos.inputs[1].default_value = 0.5
    links.new(in_node.outputs["Scroll Width"], half_width_pos.inputs[0])

    combine_start = nodes.new("ShaderNodeCombineXYZ")
    links.new(half_width_neg.outputs["Value"], combine_start.inputs["X"])

    combine_end = nodes.new("ShaderNodeCombineXYZ")
    links.new(half_width_pos.outputs["Value"], combine_end.inputs["X"])

    profile_line = nodes.new("GeometryNodeCurvePrimitiveLine")
    links.new(combine_start.outputs["Vector"], profile_line.inputs["Start"])
    links.new(combine_end.outputs["Vector"], profile_line.inputs["End"])

    spline_param_prof = nodes.new("GeometryNodeSplineParameter")
    store_v = nodes.new("GeometryNodeStoreNamedAttribute")
    store_v.inputs["Name"].default_value = "V_Factor"
    store_v.data_type = 'FLOAT'
    store_v.domain = 'POINT'
    links.new(profile_line.outputs["Curve"], store_v.inputs["Geometry"])
    links.new(spline_param_prof.outputs["Factor"], store_v.inputs["Value"])

    curve_to_mesh = nodes.new("GeometryNodeCurveToMesh")
    links.new(store_u.outputs["Geometry"], curve_to_mesh.inputs["Curve"])
    links.new(store_v.outputs["Geometry"], curve_to_mesh.inputs["Profile Curve"])

    img_aspect = 1.0
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        if img.size[1] > 0:
            img_aspect = img.size[0] / img.size[1]

    h_times_aspect = nodes.new("ShaderNodeMath")
    h_times_aspect.operation = "MULTIPLY"
    h_times_aspect.inputs[1].default_value = img_aspect
    links.new(in_node.outputs["Scroll Width"], h_times_aspect.inputs[0])

    tile_count = nodes.new("ShaderNodeMath")
    tile_count.operation = "DIVIDE"
    links.new(dist_node.outputs["Value"], tile_count.inputs[0])
    links.new(h_times_aspect.outputs["Value"], tile_count.inputs[1])

    read_u = nodes.new("GeometryNodeInputNamedAttribute")
    read_u.inputs["Name"].default_value = "U_Factor"
    read_u.data_type = 'FLOAT'

    scaled_u = nodes.new("ShaderNodeMath")
    scaled_u.operation = "MULTIPLY"
    links.new(read_u.outputs["Attribute"], scaled_u.inputs[0])
    links.new(tile_count.outputs["Value"], scaled_u.inputs[1])

    read_v = nodes.new("GeometryNodeInputNamedAttribute")
    read_v.inputs["Name"].default_value = "V_Factor"
    read_v.data_type = 'FLOAT'

    flip_v = nodes.new("ShaderNodeMath")
    flip_v.operation = "SUBTRACT"
    flip_v.inputs[0].default_value = 1.0
    links.new(read_v.outputs["Attribute"], flip_v.inputs[1])

    combine_uv = nodes.new("ShaderNodeCombineXYZ")
    links.new(scaled_u.outputs["Value"], combine_uv.inputs["X"])
    links.new(flip_v.outputs["Value"], combine_uv.inputs["Y"])

    store_uv = nodes.new("GeometryNodeStoreNamedAttribute")
    store_uv.inputs["Name"].default_value = "UVMap"
    store_uv.data_type = 'FLOAT2'
    store_uv.domain = 'CORNER'
    links.new(curve_to_mesh.outputs["Mesh"], store_uv.inputs["Geometry"])
    links.new(combine_uv.outputs["Vector"], store_uv.inputs["Value"])

    set_material = nodes.new("GeometryNodeSetMaterial")
    scroll_mat, _ = create_image_scroll_material(image_path)
    set_material.inputs["Material"].default_value = scroll_mat
    links.new(store_uv.outputs["Geometry"], set_material.inputs["Geometry"])

    links.new(set_material.outputs["Geometry"], out_node.inputs["Geometry"])

    return gn_group, scroll_mat


def setup_geometry_node_scroll(target_mesh, unfurl_start, unfurl_end, image_path=""):
    mesh_data = bpy.data.meshes.new("Scroll_Procedural_Mesh")
    scroll_obj = bpy.data.objects.new("Scroll_GN_Object", mesh_data)
    bpy.context.scene.collection.objects.link(scroll_obj)

    mod = scroll_obj.modifiers.new("Scroll_GN", type="NODES")
    gn_tree, scroll_mat = create_pure_geometry_node_scroll(image_path=image_path)
    mod.node_group = gn_tree

    scroll_obj.data.materials.append(scroll_mat)

    r_hand_name, l_hand_name = find_hand_vertex_groups(target_mesh)

    unfurl_id = None
    for item in gn_tree.interface.items_tree:
        if getattr(item, "in_out", None) == "INPUT":
            if item.name == "Character Mesh":
                mod[item.identifier] = target_mesh
            elif item.name == "Right Hand Group":
                mod[item.identifier] = r_hand_name
            elif item.name == "Left Hand Group":
                mod[item.identifier] = l_hand_name
            elif item.name == "Forward Offset":
                mod[item.identifier] = 0.60
            elif item.name == "Scroll Width":
                mod[item.identifier] = 0.35
            elif item.name == "Unfurl":
                unfurl_id = item.identifier

    for m in target_mesh.modifiers:
        if m.type == 'ARMATURE':
            m.show_viewport = True
            m.show_render = True
            m.show_in_editmode = True

    scroll_obj.hide_viewport = True
    scroll_obj.hide_render = True
    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=1)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=1)

    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=unfurl_start - 1)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=unfurl_start - 1)

    scroll_obj.hide_viewport = False
    scroll_obj.hide_render = False
    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=unfurl_start)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=unfurl_start)

    for fc in get_object_fcurves(scroll_obj):
        if fc.data_path in ["hide_viewport", "hide_render"]:
            for kp in fc.keyframe_points:
                kp.interpolation = 'CONSTANT'

    if unfurl_id:
        mod[unfurl_id] = 0.001
        mod.keyframe_insert(data_path=f'["{unfurl_id}"]', frame=unfurl_start)

        mod[unfurl_id] = 1.0
        mod.keyframe_insert(data_path=f'["{unfurl_id}"]', frame=unfurl_end)

    return scroll_obj


# ==============================================================================
# MAIN CONTROLLER
# ==============================================================================

def import_and_extract_action(fbx_path, action_name):
    existing_objects = set(bpy.data.objects)
    bpy.ops.import_scene.fbx(filepath=fbx_path)
    bpy.context.view_layer.update()

    imported_objects = set(bpy.data.objects) - existing_objects
    extracted_action = None
    for obj in imported_objects:
        if obj.animation_data and obj.animation_data.action:
            extracted_action = obj.animation_data.action
            extracted_action.name = action_name
            extracted_action.use_fake_user = True
            break
    return extracted_action, imported_objects


def purge_imported_objects(objects_to_remove):
    for obj in objects_to_remove:
        if obj.name in bpy.data.objects:
            if obj.animation_data:
                obj.animation_data_clear()
            for coll in list(obj.users_collection):
                coll.objects.unlink(obj)
    bpy.context.view_layer.update()


def main():
    cleanup_previous_run()

    walk_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Mutant Walking.fbx"
    turn_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Right Turn 90.fbx"
    unfurl_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Mutant Idle.fbx"

    sky_hdri_path = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/horn-koppe_spring_4k.exr"
    output_blend = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/output_scene.blend"

    image_path = "/Users/dobbinp1/Downloads/IMGP6691_psedit.jpg"
    grass_dir = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Grass004_4K-JPG"

    setup_grass_ground_geometry_nodes(grass_dir, sky_hdri_path, ground_z_offset=-0.2)

    CYCLES = 4
    CROSSFADE = 15
    TARGET_DISTANCE = 4.0

    walk_action, walk_objs = import_and_extract_action(walk_fbx, "walk")
    target_rig = next((obj for obj in walk_objs if obj.type == "ARMATURE"), None)
    target_mesh = next((obj for obj in walk_objs if obj.type == "MESH"), None)

    if not target_rig or not target_mesh:
        raise RuntimeError("Armature or Character Mesh not found in walk FBX.")

    travel_axis_idx, motion_sign = detect_natural_forward_axis(walk_action, target_rig)
    initial_heading = math.radians(90.0) if travel_axis_idx == 0 else 0.0
    target_rig.rotation_euler[2] = initial_heading

    action_len = int(walk_action.frame_range[1] - walk_action.frame_range[0])
    if action_len <= 0:
        action_len = 30

    anim_data = target_rig.animation_data_create()

    walk_track = anim_data.nla_tracks.new()
    walk_track.name = "Walk_Track"

    current_frame = 1
    for i in range(CYCLES):
        strip = walk_track.strips.new(name=f"Walk_{i+1}", start=int(current_frame), action=walk_action)
        strip.blend_type = "REPLACE"
        current_frame += action_len

    walk_end_frame = int(current_frame)

    turn_action, turn_objs = import_and_extract_action(turn_fbx, "rightturn")
    purge_imported_objects(turn_objs)

    turn_track = anim_data.nla_tracks.new()
    turn_track.name = "RightTurn_Track"

    turn_start = int(walk_end_frame - CROSSFADE)
    turn_strip = turn_track.strips.new(name="RightTurn", start=turn_start, action=turn_action)
    turn_strip.blend_type = "REPLACE"
    turn_strip.blend_in = CROSSFADE
    turn_strip.use_auto_blend = True
    turn_strip.extrapolation = "HOLD"
    turn_end_frame = int(turn_strip.frame_end)

    unfurl_action, unfurl_objs = import_and_extract_action(unfurl_fbx, "unfurl")
    purge_imported_objects(unfurl_objs)

    unfurl_track = anim_data.nla_tracks.new()
    unfurl_track.name = "Unfurl_Track"

    unfurl_start = int(turn_end_frame - CROSSFADE)
    unfurl_strip = unfurl_track.strips.new(name="Unfurl", start=unfurl_start, action=unfurl_action)
    unfurl_strip.blend_type = "REPLACE"
    unfurl_strip.blend_in = CROSSFADE
    unfurl_strip.use_auto_blend = True
    unfurl_strip.extrapolation = "HOLD"
    unfurl_end_frame = int(unfurl_strip.frame_end)

    root_action = bpy.data.actions.new(name="World_Root_Motion")
    target_rig.animation_data.action = root_action

    loc_fcurve = root_action.fcurve_ensure_for_datablock(datablock=target_rig, data_path="location", index=travel_axis_idx)
    final_target_value = TARGET_DISTANCE * motion_sign

    k_loc1 = loc_fcurve.keyframe_points.insert(frame=1.0, value=0.0)
    k_loc2 = loc_fcurve.keyframe_points.insert(frame=float(turn_start), value=final_target_value)
    k_loc3 = loc_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=final_target_value)

    k_loc1.interpolation = "LINEAR"
    k_loc2.interpolation = "CONSTANT"
    k_loc3.interpolation = "CONSTANT"

    rot_fcurve = root_action.fcurve_ensure_for_datablock(datablock=target_rig, data_path="rotation_euler", index=2)
    target_heading = initial_heading - math.radians(90.0)

    k_rot1 = rot_fcurve.keyframe_points.insert(frame=1.0, value=initial_heading)
    k_rot2 = rot_fcurve.keyframe_points.insert(frame=float(turn_start), value=initial_heading)
    k_rot3 = rot_fcurve.keyframe_points.insert(frame=float(turn_end_frame), value=target_heading)
    k_rot4 = rot_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=target_heading)

    k_rot1.interpolation = "CONSTANT"
    k_rot2.interpolation = "BEZIER"
    k_rot3.interpolation = "CONSTANT"
    k_rot4.interpolation = "CONSTANT"

    root_track = anim_data.nla_tracks.new()
    root_track.name = "World_Root_Track"
    root_strip = root_track.strips.new(name="Root_Motion_Strip", start=1, action=root_action)
    root_strip.blend_type = "REPLACE"
    root_strip.extrapolation = "HOLD"

    anim_data.action = None

    setup_geometry_node_scroll(
        target_mesh=target_mesh,
        unfurl_start=unfurl_start,
        unfurl_end=unfurl_end_frame,
        image_path=image_path,
    )

    # Modern Instagram Reel Vertical Camera Setup
    setup_modern_reel_camera(
        target_rig=target_rig,
        target_mesh=target_mesh,
        turn_start=turn_start,
        turn_end_frame=turn_end_frame,
        unfurl_end_frame=unfurl_end_frame
    )

    bpy.context.scene.frame_start = 1
    bpy.context.scene.frame_end = unfurl_end_frame

    bpy.ops.wm.save_as_mainfile(filepath=output_blend)
    print(f"\n[Success] Modern vertical reel camera added. Saved to: {output_blend}")


main()
```

```
import math
import os
import glob
import bpy
from bpy_extras import anim_utils

# ==============================================================================
# CLEANUP & HELPERS
# ==============================================================================

def cleanup_previous_run():
    """Clears old scroll objects, ground, node groups, camera anchors, collections, and orphaned data."""
    target_names = [
        "Master_NLA_Control",
        "Scroll_GN_Object",
        "Scroll_Start_Pin",
        "Scroll_End_Pin",
        "Scroll_Master_Container",
        "HDRI_Ground_Plane",
        "HDRI_Camera",
        "Ground_Mesh_Object",
        "Camera_Tracking_Anchor",
        "Tracking_Camera",
        "Camera_Focus_Target"
    ]
    for obj in list(bpy.data.objects):
        if obj.name in target_names or (obj.name.startswith("mixamorig") and obj.type != "ARMATURE" and obj.type != "MESH"):
            if obj.animation_data:
                obj.animation_data_clear()
            bpy.data.objects.remove(obj, do_unlink=True)

    for action in list(bpy.data.actions):
        if action.users == 0 or action.use_fake_user:
            action.use_fake_user = False
            bpy.data.actions.remove(action)

    for ng in list(bpy.data.node_groups):
        if ng.name in ["Scroll_Unfurl_GN", "Scroll_Pure_GN", "HDRI_Ground_GN", "Grass_Ground_GN", "Camera_Tracker_GN"]:
            bpy.data.node_groups.remove(ng)

    for coll_name in ["Scroll_Master_Container", "Grass_Artifacts_Collection"]:
        if coll_name in bpy.data.collections:
            coll = bpy.data.collections[coll_name]
            for obj in list(coll.objects):
                bpy.data.objects.remove(obj, do_unlink=True)
            bpy.data.collections.remove(coll)

    bpy.context.view_layer.update()


def configure_camera_and_viewport_horizon(far_clip=50.0):
    if bpy.context.scene.camera:
        bpy.context.scene.camera.data.clip_end = far_clip

    for area in bpy.context.screen.areas:
        if area.type == 'VIEW_3D':
            for space in area.spaces:
                if space.type == 'VIEW_3D':
                    space.clip_end = far_clip


def get_slot_fcurves(action, anim_target_object):
    fcurves = []
    if not action:
        return fcurves

    anim_data = getattr(anim_target_object, "animation_data", None)
    if anim_data and getattr(anim_data, "action_slot", None):
        slot = anim_data.action_slot
        channelbag = anim_utils.action_get_channelbag_for_slot(action, slot)
        if channelbag and hasattr(channelbag, "fcurves"):
            return list(channelbag.fcurves)

    if hasattr(action, "slots"):
        for slot in action.slots:
            for cb in getattr(slot, "channelbags", []):
                fcurves.extend(getattr(cb, "fcurves", []))
        if fcurves:
            return fcurves

    if hasattr(action, "fcurves"):
        return list(action.fcurves)

    return fcurves


def get_object_fcurves(obj):
    if not obj or not obj.animation_data or not obj.animation_data.action:
        return []
    action = obj.animation_data.action
    return get_slot_fcurves(action, obj)


def detect_natural_forward_axis(action, rig_object):
    fcurves = get_slot_fcurves(action, rig_object)
    max_delta = 0.0
    best_axis = 1
    net_delta = 0.0

    start_frame = action.frame_range[0]
    end_frame = action.frame_range[1]

    for fc in fcurves:
        if fc.data_path.endswith("location") and fc.array_index in [0, 1, 2]:
            start_val = fc.evaluate(start_frame)
            end_val = fc.evaluate(end_frame)
            delta = end_val - start_val
            if abs(delta) > abs(max_delta):
                max_delta = delta
                best_axis = fc.array_index
                net_delta = delta

    direction_sign = 1.0 if net_delta >= 0 else -1.0
    return best_axis, direction_sign


def find_hand_vertex_groups(target_mesh):
    r_name = "mixamorig:RightHand"
    l_name = "mixamorig:LeftHand"

    if target_mesh and hasattr(target_mesh, "vertex_groups"):
        for vg in target_mesh.vertex_groups:
            vg_lower = vg.name.lower()
            if "righthand" in vg_lower or "hand_r" in vg_lower or "hand.r" in vg_lower:
                r_name = vg.name
            elif "lefthand" in vg_lower or "hand_l" in vg_lower or "hand.l" in vg_lower:
                l_name = vg.name

    return r_name, l_name


# ==============================================================================
# INSTAGRAM VERTICAL REEL CAMERA SYSTEM (9:16)
# ==============================================================================

def setup_vertical_reel_camera(target_mesh, turn_start, turn_end_frame, unfurl_end_frame, focal_length=35.0):
    """
    Sets up an Instagram Vertical Reel (9:16 ratio, 1080x1920) camera.
    - Walk phase: Monster in left third, viewed at 45° angle.
    - Turn phase: Transitions smoothly from 45° to 90° (direct front), moving monster to center.
    - Unfurl phase: Stays stationary at 90°, monster centered.
    """
    # 1. Set Scene Render Resolution to Vertical Reel 9:16
    scene = bpy.context.scene
    scene.render.resolution_x = 1080
    scene.render.resolution_y = 1920
    scene.render.resolution_percentage = 100

    # 2. Instantiate Vertical Camera Data
    cam_data = bpy.data.cameras.new("Vertical_Reel_Camera_Data")
    cam_data.lens = focal_length
    cam_data.sensor_fit = 'VERTICAL'  # Ensures correct framing across 9:16 height

    cam_obj = bpy.data.objects.new("Tracking_Camera", cam_data)
    scene.collection.objects.link(cam_obj)
    scene.camera = cam_obj

    # 3. Position Constraints relative to Target Mesh
    copy_pos = cam_obj.constraints.new(type='COPY_LOCATION')
    copy_pos.target = target_mesh
    copy_pos.use_offset = True

    track_to = cam_obj.constraints.new(type='TRACK_TO')
    track_to.target = target_mesh
    track_to.track_axis = 'TRACK_NEGATIVE_Z'
    track_to.up_axis = 'UP_Y'
    track_to.use_target_z = True

    # 4. Offsets for 45° Angle and 90° Front Angle
    dist = 8.5  # Extended slightly to ensure full character height fits vertical view
    angle_45_x = dist * math.sin(math.radians(45))
    angle_45_y = -dist * math.cos(math.radians(45))

    angle_90_x = 0.0
    angle_90_y = -dist

    cam_height = 1.1

    # 5. Keyframe Animation Phases
    # Frame 1 to Turn Start: 45° Angle + Left Third Framing (shift_x = -0.18 for 9:16 vertical frame)
    cam_obj.location = (angle_45_x, angle_45_y, cam_height)
    cam_obj.keyframe_insert(data_path="location", frame=1)
    cam_obj.keyframe_insert(data_path="location", frame=turn_start)

    cam_data.shift_x = -0.18
    cam_data.shift_y = 0.10
    cam_data.keyframe_insert(data_path="shift_x", frame=1)
    cam_data.keyframe_insert(data_path="shift_x", frame=turn_start)

    # Turn Start to Turn End: Smoothly orbit to 90° Front + Dead Center Framing
    cam_obj.location = (angle_90_x, angle_90_y, cam_height)
    cam_obj.keyframe_insert(data_path="location", frame=turn_end_frame)
    cam_obj.keyframe_insert(data_path="location", frame=unfurl_end_frame)

    cam_data.shift_x = 0.0
    cam_data.keyframe_insert(data_path="shift_x", frame=turn_end_frame)
    cam_data.keyframe_insert(data_path="shift_x", frame=unfurl_end_frame)

    # Smooth Interpolation across keyframes
    for fc in get_object_fcurves(cam_obj):
        for kp in fc.keyframe_points:
            kp.interpolation = 'BEZIER'

    for fc in get_object_fcurves(cam_data):
        for kp in fc.keyframe_points:
            kp.interpolation = 'BEZIER'

    return cam_obj


# ==============================================================================
# GEOMETRY NODES GROUND, GRASS & SKY DOME SETUP
# ==============================================================================

def create_grass_material(grass_dir):
    mat = bpy.data.materials.new(name="Grass_PBR_Material")
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    out_node = nodes.new("ShaderNodeOutputMaterial")
    bsdf = nodes.new("ShaderNodeBsdfPrincipled")
    links.new(bsdf.outputs["BSDF"], out_node.inputs["Surface"])

    col_files = glob.glob(os.path.join(grass_dir, "*Color*.jpg")) + glob.glob(os.path.join(grass_dir, "*JPG*.jpg"))
    if col_files:
        tex_col = nodes.new("ShaderNodeTexImage")
        tex_col.image = bpy.data.images.load(col_files[0])
        links.new(tex_col.outputs["Color"], bsdf.inputs["Base Color"])

    rough_files = glob.glob(os.path.join(grass_dir, "*Roughness*.jpg"))
    if rough_files:
        tex_rough = nodes.new("ShaderNodeTexImage")
        tex_rough.image = bpy.data.images.load(rough_files[0])
        tex_rough.image.colorspace_settings.name = 'Non-Color'
        links.new(tex_rough.outputs["Color"], bsdf.inputs["Roughness"])

    norm_files = glob.glob(os.path.join(grass_dir, "*Normal*.jpg"))
    if norm_files:
        tex_norm = nodes.new("ShaderNodeTexImage")
        tex_norm.image = bpy.data.images.load(norm_files[0])
        tex_norm.image.colorspace_settings.name = 'Non-Color'

        normal_map_node = nodes.new("ShaderNodeNormalMap")
        links.new(tex_norm.outputs["Color"], normal_map_node.inputs["Color"])
        links.new(normal_map_node.outputs["Normal"], bsdf.inputs["Normal"])

    return mat


def create_sky_material(sky_hdri_path):
    mat = bpy.data.materials.new(name="Sky_Dome_Material")
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    out_node = nodes.new("ShaderNodeOutputMaterial")
    emission = nodes.new("ShaderNodeEmission")
    emission.inputs["Strength"].default_value = 1.0

    links.new(emission.outputs["Emission"], out_node.inputs["Surface"])

    if os.path.exists(sky_hdri_path):
        tex_env = nodes.new("ShaderNodeTexEnvironment")
        tex_env.image = bpy.data.images.load(sky_hdri_path)
        links.new(tex_env.outputs["Color"], emission.inputs["Color"])

    return mat


def setup_grass_ground_geometry_nodes(grass_dir, sky_hdri_path, ground_z_offset=-0.2):
    grass_coll = bpy.data.collections.get("Grass_Artifacts_Collection")
    if not grass_coll:
        grass_coll = bpy.data.collections.new("Grass_Artifacts_Collection")
        bpy.context.scene.collection.children.link(grass_coll)

    grass_mat = create_grass_material(grass_dir)
    sky_mat = create_sky_material(sky_hdri_path)

    mesh_files = glob.glob(os.path.join(grass_dir, "*.obj")) + glob.glob(os.path.join(grass_dir, "*.fbx"))

    if mesh_files:
        for fpath in mesh_files:
            if fpath.endswith(".obj"):
                bpy.ops.wm.obj_import(filepath=fpath)
            elif fpath.endswith(".fbx"):
                bpy.ops.import_scene.fbx(filepath=fpath)

            for o in bpy.context.selected_objects:
                grass_coll.objects.link(o)
                if o.name in bpy.context.scene.collection.objects:
                    bpy.context.scene.collection.objects.unlink(o)

        for obj in list(grass_coll.objects):
            if obj.dimensions.z > 0.5:
                grass_coll.objects.unlink(obj)
                bpy.data.objects.remove(obj, do_unlink=True)
    else:
        for i in range(3):
            m_data = bpy.data.meshes.new(f"Grass_Blade_Mesh_{i}")
            g_obj = bpy.data.objects.new(f"Grass_Blade_Obj_{i}", m_data)
            grass_coll.objects.link(g_obj)

            verts = [(0, 0, 0), (0.05, 0, 0), (0.04, 0, 0.15), (0.01, 0, 0.25), (0, 0, 0.3)]
            faces = [(0, 1, 2), (1, 2, 3), (2, 3, 4)]
            m_data.from_pydata(verts, [], faces)
            m_data.update()

            if g_obj.data.materials:
                g_obj.data.materials[0] = grass_mat
            else:
                g_obj.data.materials.append(grass_mat)

            g_obj.location = (100, 100, 0)

    bpy.ops.mesh.primitive_circle_add(radius=12.5, vertices=64, fill_type='NGON', location=(0, 0, ground_z_offset))
    ground_obj = bpy.context.active_object
    ground_obj.name = "Ground_Mesh_Object"

    vg = ground_obj.vertex_groups.new(name="Grass_Weight_Group")
    all_vert_indices = [v.index for v in ground_obj.data.vertices]
    vg.add(all_vert_indices, 1.0, 'REPLACE')
    ground_obj.data.materials.append(grass_mat)
    ground_obj.data.materials.append(sky_mat)

    gn_group = bpy.data.node_groups.new("Grass_Ground_GN", type="GeometryNodeTree")
    gn_group.interface.new_socket(name="Geometry", in_out="INPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Geometry", in_out="OUTPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Density Mask", in_out="INPUT", socket_type="NodeSocketFloat")

    nodes = gn_group.nodes
    links = gn_group.links
    out_node = nodes.new("NodeGroupOutput")
    in_node = nodes.new("NodeGroupInput")

    distribute_node = nodes.new("GeometryNodeDistributePointsOnFaces")
    distribute_node.distribute_method = 'POISSON'
    distribute_node.inputs["Distance Min"].default_value = 0.02
    distribute_node.inputs["Density Max"].default_value = 300.0
    links.new(in_node.outputs["Geometry"], distribute_node.inputs["Mesh"])

    noise_node = nodes.new("ShaderNodeTexNoise")
    noise_node.inputs["Scale"].default_value = 2.5
    noise_node.inputs["Detail"].default_value = 2.0

    color_ramp = nodes.new("ShaderNodeValToRGB")
    color_ramp.color_ramp.elements[0].position = 0.20
    color_ramp.color_ramp.elements[1].position = 0.80
    links.new(noise_node.outputs["Fac"], color_ramp.inputs["Fac"])

    math_weight = nodes.new("ShaderNodeMath")
    math_weight.operation = "MULTIPLY"
    links.new(color_ramp.outputs["Color"], math_weight.inputs[0])
    links.new(in_node.outputs["Density Mask"], math_weight.inputs[1])

    links.new(math_weight.outputs["Value"], distribute_node.inputs["Density Factor"])

    instance_node = nodes.new("GeometryNodeInstanceOnPoints")
    instance_node.inputs["Pick Instance"].default_value = True
    links.new(distribute_node.outputs["Points"], instance_node.inputs["Points"])

    coll_info = nodes.new("GeometryNodeCollectionInfo")
    coll_info.inputs["Collection"].default_value = grass_coll
    coll_info.inputs["Separate Children"].default_value = True
    coll_info.inputs["Reset Children"].default_value = True
    links.new(coll_info.outputs["Instances"], instance_node.inputs["Instance"])

    rand_scale = nodes.new("FunctionNodeRandomValue")
    rand_scale.data_type = 'FLOAT'
    rand_scale.inputs["Min"].default_value = 0.25
    rand_scale.inputs["Max"].default_value = 0.50

    combine_scale = nodes.new("ShaderNodeCombineXYZ")
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["X"])
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["Y"])
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["Z"])
    links.new(combine_scale.outputs["Vector"], instance_node.inputs["Scale"])

    rand_rot_z = nodes.new("FunctionNodeRandomValue")
    rand_rot_z.data_type = 'FLOAT'
    rand_rot_z.inputs["Min"].default_value = 0.0
    rand_rot_z.inputs["Max"].default_value = math.radians(360.0)

    combine_rot = nodes.new("ShaderNodeCombineXYZ")
    links.new(rand_rot_z.outputs["Value"], combine_rot.inputs["Z"])
    links.new(combine_rot.outputs["Vector"], instance_node.inputs["Rotation"])

    sky_sphere = nodes.new("GeometryNodeMeshIcoSphere")
    sky_sphere.inputs["Radius"].default_value = 12.5
    sky_sphere.inputs["Subdivisions"].default_value = 4

    flip_faces = nodes.new("GeometryNodeFlipFaces")
    links.new(sky_sphere.outputs["Mesh"], flip_faces.inputs["Mesh"])

    transform_sky = nodes.new("GeometryNodeTransform")
    transform_sky.inputs["Translation"].default_value = (0.0, 0.0, ground_z_offset)
    links.new(flip_faces.outputs["Mesh"], transform_sky.inputs["Geometry"])

    set_sky_mat = nodes.new("GeometryNodeSetMaterial")
    set_sky_mat.inputs["Material"].default_value = sky_mat
    links.new(transform_sky.outputs["Geometry"], set_sky_mat.inputs["Geometry"])

    join_geo = nodes.new("GeometryNodeJoinGeometry")
    links.new(in_node.outputs["Geometry"], join_geo.inputs["Geometry"])
    links.new(instance_node.outputs["Instances"], join_geo.inputs["Geometry"])
    links.new(set_sky_mat.outputs["Geometry"], join_geo.inputs["Geometry"])

    links.new(join_geo.outputs["Geometry"], out_node.inputs["Geometry"])

    mod = ground_obj.modifiers.new(name="Grass_Ground_GN", type='NODES')
    mod.node_group = gn_group

    for item in gn_group.interface.items_tree:
        if getattr(item, "in_out", None) == "INPUT" and item.name == "Density Mask":
            mod[item.identifier + "_attribute_name"] = "Grass_Weight_Group"
            mod[item.identifier + "_use_attribute"] = True

    return ground_obj


# ==============================================================================
# SCROLL MATERIAL & GEOMETRY NODES SETUP
# ==============================================================================

def create_image_scroll_material(image_path):
    mat_name = "Scroll_Texture_Mat"
    mat = bpy.data.materials.get(mat_name)
    if mat:
        bpy.data.materials.remove(mat)

    mat = bpy.data.materials.new(name=mat_name)
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    output_node = nodes.new("ShaderNodeOutputMaterial")
    bsdf = nodes.new("ShaderNodeBsdfPrincipled")
    bsdf.inputs["Roughness"].default_value = 0.5

    tex_img = nodes.new("ShaderNodeTexImage")
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        tex_img.image = img
        tex_img.extension = 'REPEAT'

    uv_map_node = nodes.new("ShaderNodeUVMap")
    uv_map_node.uv_map = "UVMap"

    links.new(uv_map_node.outputs["UV"], tex_img.inputs["Vector"])
    links.new(tex_img.outputs["Color"], bsdf.inputs["Base Color"])
    links.new(bsdf.outputs["BSDF"], output_node.inputs["Surface"])

    return mat, tex_img.image if os.path.exists(image_path) else None


def create_pure_geometry_node_scroll(image_path=""):
    gn_group = bpy.data.node_groups.new("Scroll_Pure_GN", type="GeometryNodeTree")

    gn_group.interface.new_socket(name="Geometry", in_out="OUTPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Character Mesh", in_out="INPUT", socket_type="NodeSocketObject")

    sock_r_vg = gn_group.interface.new_socket(name="Right Hand Group", in_out="INPUT", socket_type="NodeSocketString")
    sock_r_vg.default_value = "mixamorig:RightHand"

    sock_l_vg = gn_group.interface.new_socket(name="Left Hand Group", in_out="INPUT", socket_type="NodeSocketString")
    sock_l_vg.default_value = "mixamorig:LeftHand"

    sock_offset = gn_group.interface.new_socket(name="Forward Offset", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_offset.default_value = 0.60

    sock_unfurl = gn_group.interface.new_socket(name="Unfurl", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_unfurl.default_value = 0.0

    sock_width = gn_group.interface.new_socket(name="Scroll Width", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_width.default_value = 0.35

    nodes = gn_group.nodes
    links = gn_group.links

    in_node = nodes.new("NodeGroupInput")
    out_node = nodes.new("NodeGroupOutput")

    char_info = nodes.new("GeometryNodeObjectInfo")
    char_info.transform_space = "RELATIVE"
    links.new(in_node.outputs["Character Mesh"], char_info.inputs["Object"])

    pos_node = nodes.new("GeometryNodeInputPosition")

    named_r = nodes.new("GeometryNodeInputNamedAttribute")
    named_r.data_type = "FLOAT"
    links.new(in_node.outputs["Right Hand Group"], named_r.inputs["Name"])

    cmp_r = nodes.new("FunctionNodeCompare")
    cmp_r.data_type = "FLOAT"
    cmp_r.operation = "GREATER_THAN"
    cmp_r.inputs[1].default_value = 0.001
    links.new(named_r.outputs["Attribute"], cmp_r.inputs[0])

    stat_r = nodes.new("GeometryNodeAttributeStatistic")
    stat_r.data_type = "FLOAT_VECTOR"
    links.new(char_info.outputs["Geometry"], stat_r.inputs["Geometry"])
    links.new(cmp_r.outputs["Result"], stat_r.inputs["Selection"])
    links.new(pos_node.outputs["Position"], stat_r.inputs["Attribute"])

    named_l = nodes.new("GeometryNodeInputNamedAttribute")
    named_l.data_type = "FLOAT"
    links.new(in_node.outputs["Left Hand Group"], named_l.inputs["Name"])

    cmp_l = nodes.new("FunctionNodeCompare")
    cmp_l.data_type = "FLOAT"
    cmp_l.operation = "GREATER_THAN"
    cmp_l.inputs[1].default_value = 0.001
    links.new(named_l.outputs["Attribute"], cmp_l.inputs[0])

    stat_l = nodes.new("GeometryNodeAttributeStatistic")
    stat_l.data_type = "FLOAT_VECTOR"
    links.new(char_info.outputs["Geometry"], stat_l.inputs["Geometry"])
    links.new(cmp_l.outputs["Result"], stat_l.inputs["Selection"])
    links.new(pos_node.outputs["Position"], stat_l.inputs["Attribute"])

    neg_offset = nodes.new("ShaderNodeMath")
    neg_offset.operation = "MULTIPLY"
    neg_offset.inputs[1].default_value = -1.0
    links.new(in_node.outputs["Forward Offset"], neg_offset.inputs[0])

    forward_vec = nodes.new("ShaderNodeCombineXYZ")
    forward_vec.inputs["X"].default_value = 0.0
    links.new(neg_offset.outputs["Value"], forward_vec.inputs["Y"])
    forward_vec.inputs["Z"].default_value = 0.0

    start_pos = nodes.new("ShaderNodeVectorMath")
    start_pos.operation = "ADD"
    links.new(stat_r.outputs["Mean"], start_pos.inputs[0])
    links.new(forward_vec.outputs["Vector"], start_pos.inputs[1])

    target_pos = nodes.new("ShaderNodeVectorMath")
    target_pos.operation = "ADD"
    links.new(stat_l.outputs["Mean"], target_pos.inputs[0])
    links.new(forward_vec.outputs["Vector"], target_pos.inputs[1])

    delta_vec = nodes.new("ShaderNodeVectorMath")
    delta_vec.operation = "SUBTRACT"
    links.new(target_pos.outputs["Vector"], delta_vec.inputs[0])
    links.new(start_pos.outputs["Vector"], delta_vec.inputs[1])

    dist_node = nodes.new("ShaderNodeVectorMath")
    dist_node.operation = "DISTANCE"
    links.new(start_pos.outputs["Vector"], dist_node.inputs[0])

    scale_vec = nodes.new("ShaderNodeVectorMath")
    scale_vec.operation = "SCALE"
    links.new(delta_vec.outputs["Vector"], scale_vec.inputs["Vector"])
    links.new(in_node.outputs["Unfurl"], scale_vec.inputs["Scale"])

    end_pos = nodes.new("ShaderNodeVectorMath")
    end_pos.operation = "ADD"
    links.new(start_pos.outputs["Vector"], end_pos.inputs[0])
    links.new(scale_vec.outputs["Vector"], end_pos.inputs[1])

    links.new(end_pos.outputs["Vector"], dist_node.inputs[1])

    curve_line = nodes.new("GeometryNodeCurvePrimitiveLine")
    links.new(start_pos.outputs["Vector"], curve_line.inputs["Start"])
    links.new(end_pos.outputs["Vector"], curve_line.inputs["End"])

    set_tilt = nodes.new("GeometryNodeSetCurveTilt")
    set_tilt.inputs["Tilt"].default_value = 1.5708
    links.new(curve_line.outputs["Curve"], set_tilt.inputs["Curve"])

    spline_param_main = nodes.new("GeometryNodeSplineParameter")
    store_u = nodes.new("GeometryNodeStoreNamedAttribute")
    store_u.inputs["Name"].default_value = "U_Factor"
    store_u.data_type = 'FLOAT'
    store_u.domain = 'POINT'
    links.new(set_tilt.outputs["Curve"], store_u.inputs["Geometry"])
    links.new(spline_param_main.outputs["Factor"], store_u.inputs["Value"])

    half_width_neg = nodes.new("ShaderNodeMath")
    half_width_neg.operation = "MULTIPLY"
    half_width_neg.inputs[1].default_value = -0.5
    links.new(in_node.outputs["Scroll Width"], half_width_neg.inputs[0])

    half_width_pos = nodes.new("ShaderNodeMath")
    half_width_pos.operation = "MULTIPLY"
    half_width_pos.inputs[1].default_value = 0.5
    links.new(in_node.outputs["Scroll Width"], half_width_pos.inputs[0])

    combine_start = nodes.new("ShaderNodeCombineXYZ")
    links.new(half_width_neg.outputs["Value"], combine_start.inputs["X"])

    combine_end = nodes.new("ShaderNodeCombineXYZ")
    links.new(half_width_pos.outputs["Value"], combine_end.inputs["X"])

    profile_line = nodes.new("GeometryNodeCurvePrimitiveLine")
    links.new(combine_start.outputs["Vector"], profile_line.inputs["Start"])
    links.new(combine_end.outputs["Vector"], profile_line.inputs["End"])

    spline_param_prof = nodes.new("GeometryNodeSplineParameter")
    store_v = nodes.new("GeometryNodeStoreNamedAttribute")
    store_v.inputs["Name"].default_value = "V_Factor"
    store_v.data_type = 'FLOAT'
    store_v.domain = 'POINT'
    links.new(profile_line.outputs["Curve"], store_v.inputs["Geometry"])
    links.new(spline_param_prof.outputs["Factor"], store_v.inputs["Value"])

    curve_to_mesh = nodes.new("GeometryNodeCurveToMesh")
    links.new(store_u.outputs["Geometry"], curve_to_mesh.inputs["Curve"])
    links.new(store_v.outputs["Geometry"], curve_to_mesh.inputs["Profile Curve"])

    img_aspect = 1.0
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        if img.size[1] > 0:
            img_aspect = img.size[0] / img.size[1]

    h_times_aspect = nodes.new("ShaderNodeMath")
    h_times_aspect.operation = "MULTIPLY"
    h_times_aspect.inputs[1].default_value = img_aspect
    links.new(in_node.outputs["Scroll Width"], h_times_aspect.inputs[0])

    tile_count = nodes.new("ShaderNodeMath")
    tile_count.operation = "DIVIDE"
    links.new(dist_node.outputs["Value"], tile_count.inputs[0])
    links.new(h_times_aspect.outputs["Value"], tile_count.inputs[1])

    read_u = nodes.new("GeometryNodeInputNamedAttribute")
    read_u.inputs["Name"].default_value = "U_Factor"
    read_u.data_type = 'FLOAT'

    scaled_u = nodes.new("ShaderNodeMath")
    scaled_u.operation = "MULTIPLY"
    links.new(read_u.outputs["Attribute"], scaled_u.inputs[0])
    links.new(tile_count.outputs["Value"], scaled_u.inputs[1])

    read_v = nodes.new("GeometryNodeInputNamedAttribute")
    read_v.inputs["Name"].default_value = "V_Factor"
    read_v.data_type = 'FLOAT'

    flip_v = nodes.new("ShaderNodeMath")
    flip_v.operation = "SUBTRACT"
    flip_v.inputs[0].default_value = 1.0
    links.new(read_v.outputs["Attribute"], flip_v.inputs[1])

    combine_uv = nodes.new("ShaderNodeCombineXYZ")
    links.new(scaled_u.outputs["Value"], combine_uv.inputs["X"])
    links.new(flip_v.outputs["Value"], combine_uv.inputs["Y"])

    store_uv = nodes.new("GeometryNodeStoreNamedAttribute")
    store_uv.inputs["Name"].default_value = "UVMap"
    store_uv.data_type = 'FLOAT2'
    store_uv.domain = 'CORNER'
    links.new(curve_to_mesh.outputs["Mesh"], store_uv.inputs["Geometry"])
    links.new(combine_uv.outputs["Vector"], store_uv.inputs["Value"])

    set_material = nodes.new("GeometryNodeSetMaterial")
    scroll_mat, _ = create_image_scroll_material(image_path)
    set_material.inputs["Material"].default_value = scroll_mat
    links.new(store_uv.outputs["Geometry"], set_material.inputs["Geometry"])

    links.new(set_material.outputs["Geometry"], out_node.inputs["Geometry"])

    return gn_group, scroll_mat


def setup_geometry_node_scroll(target_mesh, unfurl_start, unfurl_end, image_path=""):
    mesh_data = bpy.data.meshes.new("Scroll_Procedural_Mesh")
    scroll_obj = bpy.data.objects.new("Scroll_GN_Object", mesh_data)
    bpy.context.scene.collection.objects.link(scroll_obj)

    mod = scroll_obj.modifiers.new("Scroll_GN", type="NODES")
    gn_tree, scroll_mat = create_pure_geometry_node_scroll(image_path=image_path)
    mod.node_group = gn_tree

    scroll_obj.data.materials.append(scroll_mat)

    r_hand_name, l_hand_name = find_hand_vertex_groups(target_mesh)

    unfurl_id = None
    for item in gn_tree.interface.items_tree:
        if getattr(item, "in_out", None) == "INPUT":
            if item.name == "Character Mesh":
                mod[item.identifier] = target_mesh
            elif item.name == "Right Hand Group":
                mod[item.identifier] = r_hand_name
            elif item.name == "Left Hand Group":
                mod[item.identifier] = l_hand_name
            elif item.name == "Forward Offset":
                mod[item.identifier] = 0.60
            elif item.name == "Scroll Width":
                mod[item.identifier] = 0.35
            elif item.name == "Unfurl":
                unfurl_id = item.identifier

    for m in target_mesh.modifiers:
        if m.type == 'ARMATURE':
            m.show_viewport = True
            m.show_render = True
            m.show_in_editmode = True

    scroll_obj.hide_viewport = True
    scroll_obj.hide_render = True
    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=1)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=1)

    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=unfurl_start - 1)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=unfurl_start - 1)

    scroll_obj.hide_viewport = False
    scroll_obj.hide_render = False
    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=unfurl_start)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=unfurl_start)

    for fc in get_object_fcurves(scroll_obj):
        if fc.data_path in ["hide_viewport", "hide_render"]:
            for kp in fc.keyframe_points:
                kp.interpolation = 'CONSTANT'

    if unfurl_id:
        mod[unfurl_id] = 0.001
        mod.keyframe_insert(data_path=f'["{unfurl_id}"]', frame=unfurl_start)

        mod[unfurl_id] = 1.0
        mod.keyframe_insert(data_path=f'["{unfurl_id}"]', frame=unfurl_end)

    return scroll_obj


# ==============================================================================
# MAIN CONTROLLER
# ==============================================================================

def import_and_extract_action(fbx_path, action_name):
    existing_objects = set(bpy.data.objects)
    bpy.ops.import_scene.fbx(filepath=fbx_path)
    bpy.context.view_layer.update()

    imported_objects = set(bpy.data.objects) - existing_objects
    extracted_action = None
    for obj in imported_objects:
        if obj.animation_data and obj.animation_data.action:
            extracted_action = obj.animation_data.action
            extracted_action.name = action_name
            extracted_action.use_fake_user = True
            break
    return extracted_action, imported_objects


def purge_imported_objects(objects_to_remove):
    for obj in objects_to_remove:
        if obj.name in bpy.data.objects:
            if obj.animation_data:
                obj.animation_data_clear()
            for coll in list(obj.users_collection):
                coll.objects.unlink(obj)
    bpy.context.view_layer.update()


def main():
    cleanup_previous_run()

    walk_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Mutant Walking.fbx"
    turn_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Right Turn 90.fbx"
    unfurl_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Mutant Idle.fbx"

    sky_hdri_path = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/horn-koppe_spring_4k.exr"
    output_blend = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/output_scene.blend"

    image_path = "/Users/dobbinp1/Downloads/IMGP6691_psedit.jpg"
    grass_dir = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Grass004_4K-JPG"

    setup_grass_ground_geometry_nodes(grass_dir, sky_hdri_path, ground_z_offset=-0.2)

    CYCLES = 4
    CROSSFADE = 15
    TARGET_DISTANCE = 4.0

    walk_action, walk_objs = import_and_extract_action(walk_fbx, "walk")
    target_rig = next((obj for obj in walk_objs if obj.type == "ARMATURE"), None)
    target_mesh = next((obj for obj in walk_objs if obj.type == "MESH"), None)

    if not target_rig or not target_mesh:
        raise RuntimeError("Armature or Character Mesh not found in walk FBX.")

    travel_axis_idx, motion_sign = detect_natural_forward_axis(walk_action, target_rig)
    initial_heading = math.radians(90.0) if travel_axis_idx == 0 else 0.0
    target_rig.rotation_euler[2] = initial_heading

    action_len = int(walk_action.frame_range[1] - walk_action.frame_range[0])
    if action_len <= 0:
        action_len = 30

    anim_data = target_rig.animation_data_create()

    walk_track = anim_data.nla_tracks.new()
    walk_track.name = "Walk_Track"

    current_frame = 1
    for i in range(CYCLES):
        strip = walk_track.strips.new(name=f"Walk_{i+1}", start=int(current_frame), action=walk_action)
        strip.blend_type = "REPLACE"
        current_frame += action_len

    walk_end_frame = int(current_frame)

    turn_action, turn_objs = import_and_extract_action(turn_fbx, "rightturn")
    purge_imported_objects(turn_objs)

    turn_track = anim_data.nla_tracks.new()
    turn_track.name = "RightTurn_Track"

    turn_start = int(walk_end_frame - CROSSFADE)
    turn_strip = turn_track.strips.new(name="RightTurn", start=turn_start, action=turn_action)
    turn_strip.blend_type = "REPLACE"
    turn_strip.blend_in = CROSSFADE
    turn_strip.use_auto_blend = True
    turn_strip.extrapolation = "HOLD"
    turn_end_frame = int(turn_strip.frame_end)

    unfurl_action, unfurl_objs = import_and_extract_action(unfurl_fbx, "unfurl")
    purge_imported_objects(unfurl_objs)

    unfurl_track = anim_data.nla_tracks.new()
    unfurl_track.name = "Unfurl_Track"

    unfurl_start = int(turn_end_frame - CROSSFADE)
    unfurl_strip = unfurl_track.strips.new(name="Unfurl", start=unfurl_start, action=unfurl_action)
    unfurl_strip.blend_type = "REPLACE"
    unfurl_strip.blend_in = CROSSFADE
    unfurl_strip.use_auto_blend = True
    unfurl_strip.extrapolation = "HOLD"
    unfurl_end_frame = int(unfurl_strip.frame_end)

    root_action = bpy.data.actions.new(name="World_Root_Motion")
    target_rig.animation_data.action = root_action

    loc_fcurve = root_action.fcurve_ensure_for_datablock(datablock=target_rig, data_path="location", index=travel_axis_idx)
    final_target_value = TARGET_DISTANCE * motion_sign

    k_loc1 = loc_fcurve.keyframe_points.insert(frame=1.0, value=0.0)
    k_loc2 = loc_fcurve.keyframe_points.insert(frame=float(turn_start), value=final_target_value)
    k_loc3 = loc_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=final_target_value)

    k_loc1.interpolation = "LINEAR"
    k_loc2.interpolation = "CONSTANT"
    k_loc3.interpolation = "CONSTANT"

    rot_fcurve = root_action.fcurve_ensure_for_datablock(datablock=target_rig, data_path="rotation_euler", index=2)
    target_heading = initial_heading - math.radians(90.0)

    k_rot1 = rot_fcurve.keyframe_points.insert(frame=1.0, value=initial_heading)
    k_rot2 = rot_fcurve.keyframe_points.insert(frame=float(turn_start), value=initial_heading)
    k_rot3 = rot_fcurve.keyframe_points.insert(frame=float(turn_end_frame), value=target_heading)
    k_rot4 = rot_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=target_heading)

    k_rot1.interpolation = "CONSTANT"
    k_rot2.interpolation = "BEZIER"
    k_rot3.interpolation = "CONSTANT"
    k_rot4.interpolation = "CONSTANT"

    root_track = anim_data.nla_tracks.new()
    root_track.name = "World_Root_Track"
    root_strip = root_track.strips.new(name="Root_Motion_Strip", start=1, action=root_action)
    root_strip.blend_type = "REPLACE"
    root_strip.extrapolation = "HOLD"

    anim_data.action = None

    setup_geometry_node_scroll(
        target_mesh=target_mesh,
        unfurl_start=unfurl_start,
        unfurl_end=unfurl_end_frame,
        image_path=image_path,
    )

    # Attach Vertical Reel Camera Configuration
    cam_obj = setup_vertical_reel_camera(
        target_mesh=target_mesh,
        turn_start=turn_start,
        turn_end_frame=turn_end_frame,
        unfurl_end_frame=unfurl_end_frame,
        focal_length=35.0
    )

    configure_camera_and_viewport_horizon(far_clip=50.0)

    bpy.context.scene.frame_start = 1
    bpy.context.scene.frame_end = unfurl_end_frame

    bpy.ops.wm.save_as_mainfile(filepath=output_blend)
    print(f"\n[Success] Instagram Vertical Reel Camera setup complete. Saved to: {output_blend}")


main()
```

```
import math
import os
import glob
import bpy
from bpy_extras import anim_utils

# ==============================================================================
# CLEANUP & HELPERS
# ==============================================================================

def cleanup_previous_run():
    """Clears old scroll objects, ground, node groups, camera anchors, collections, and orphaned data."""
    target_names = [
        "Master_NLA_Control",
        "Scroll_GN_Object",
        "Scroll_Start_Pin",
        "Scroll_End_Pin",
        "Scroll_Master_Container",
        "HDRI_Ground_Plane",
        "HDRI_Camera",
        "Ground_Mesh_Object",
        "Camera_Tracking_Anchor",
        "Tracking_Camera",
        "Camera_Focus_Target"
    ]
    for obj in list(bpy.data.objects):
        if obj.name in target_names or (obj.name.startswith("mixamorig") and obj.type != "ARMATURE" and obj.type != "MESH"):
            if obj.animation_data:
                obj.animation_data_clear()
            bpy.data.objects.remove(obj, do_unlink=True)

    for action in list(bpy.data.actions):
        if action.users == 0 or action.use_fake_user:
            action.use_fake_user = False
            bpy.data.actions.remove(action)

    for ng in list(bpy.data.node_groups):
        if ng.name in ["Scroll_Unfurl_GN", "Scroll_Pure_GN", "HDRI_Ground_GN", "Grass_Ground_GN", "Camera_Tracker_GN"]:
            bpy.data.node_groups.remove(ng)

    for coll_name in ["Scroll_Master_Container", "Grass_Artifacts_Collection"]:
        if coll_name in bpy.data.collections:
            coll = bpy.data.collections[coll_name]
            for obj in list(coll.objects):
                bpy.data.objects.remove(obj, do_unlink=True)
            bpy.data.collections.remove(coll)

    bpy.context.view_layer.update()


def configure_camera_and_viewport_horizon(far_clip=50.0):
    if bpy.context.scene.camera:
        bpy.context.scene.camera.data.clip_end = far_clip

    for area in bpy.context.screen.areas:
        if area.type == 'VIEW_3D':
            for space in area.spaces:
                if space.type == 'VIEW_3D':
                    space.clip_end = far_clip


def get_slot_fcurves(action, anim_target_object):
    fcurves = []
    if not action:
        return fcurves

    anim_data = getattr(anim_target_object, "animation_data", None)
    if anim_data and getattr(anim_data, "action_slot", None):
        slot = anim_data.action_slot
        channelbag = anim_utils.action_get_channelbag_for_slot(action, slot)
        if channelbag and hasattr(channelbag, "fcurves"):
            return list(channelbag.fcurves)

    if hasattr(action, "slots"):
        for slot in action.slots:
            for cb in getattr(slot, "channelbags", []):
                fcurves.extend(getattr(cb, "fcurves", []))
        if fcurves:
            return fcurves

    if hasattr(action, "fcurves"):
        return list(action.fcurves)

    return fcurves


def get_object_fcurves(obj):
    if not obj or not obj.animation_data or not obj.animation_data.action:
        return []
    action = obj.animation_data.action
    return get_slot_fcurves(action, obj)


def detect_natural_forward_axis(action, rig_object):
    fcurves = get_slot_fcurves(action, rig_object)
    max_delta = 0.0
    best_axis = 1
    net_delta = 0.0

    start_frame = action.frame_range[0]
    end_frame = action.frame_range[1]

    for fc in fcurves:
        if fc.data_path.endswith("location") and fc.array_index in [0, 1, 2]:
            start_val = fc.evaluate(start_frame)
            end_val = fc.evaluate(end_frame)
            delta = end_val - start_val
            if abs(delta) > abs(max_delta):
                max_delta = delta
                best_axis = fc.array_index
                net_delta = delta

    direction_sign = 1.0 if net_delta >= 0 else -1.0
    return best_axis, direction_sign


def find_hand_vertex_groups(target_mesh):
    r_name = "mixamorig:RightHand"
    l_name = "mixamorig:LeftHand"

    if target_mesh and hasattr(target_mesh, "vertex_groups"):
        for vg in target_mesh.vertex_groups:
            vg_lower = vg.name.lower()
            if "righthand" in vg_lower or "hand_r" in vg_lower or "hand.r" in vg_lower:
                r_name = vg.name
            elif "lefthand" in vg_lower or "hand_l" in vg_lower or "hand.l" in vg_lower:
                l_name = vg.name

    return r_name, l_name


# ==============================================================================
# CAMERA TRACKING SETUP WITH ANIMATED ANGLE & SHIFT
# ==============================================================================

def setup_camera_tracking(target_mesh, turn_start, turn_end_frame, unfurl_end_frame, focal_length=35.0):
    """
    Sets up a tracking camera that positions the monster in the first-third of
    the frame at 45° during walk, and smoothly animates to center frame at 90°
    when turning.
    """
    cam_data = bpy.data.cameras.new("Tracking_Camera_Data")
    cam_data.lens = focal_length

    cam_obj = bpy.data.objects.new("Tracking_Camera", cam_data)
    bpy.context.scene.collection.objects.link(cam_obj)
    bpy.context.scene.camera = cam_obj

    copy_pos = cam_obj.constraints.new(type='COPY_LOCATION')
    copy_pos.target = target_mesh
    copy_pos.use_offset = True

    track_to = cam_obj.constraints.new(type='TRACK_TO')
    track_to.target = target_mesh
    track_to.track_axis = 'TRACK_NEGATIVE_Z'
    track_to.up_axis = 'UP_Y'
    track_to.use_target_z = True

    # Position Offsets
    dist = 7.5
    angle_45_x = dist * math.sin(math.radians(45))
    angle_45_y = -dist * math.cos(math.radians(45))

    angle_90_x = 0.0
    angle_90_y = -dist

    cam_height = 1.0

    # Keyframe Location Offset & Camera Shift
    cam_obj.location = (angle_45_x, angle_45_y, cam_height)
    cam_obj.keyframe_insert(data_path="location", frame=1)
    cam_obj.keyframe_insert(data_path="location", frame=turn_start)

    cam_data.shift_x = -0.3
    cam_data.shift_y = 0.2
    cam_data.keyframe_insert(data_path="shift_x", frame=1)
    cam_data.keyframe_insert(data_path="shift_x", frame=turn_start)

    cam_obj.location = (angle_90_x, angle_90_y, cam_height)
    cam_obj.keyframe_insert(data_path="location", frame=turn_end_frame)
    cam_obj.keyframe_insert(data_path="location", frame=unfurl_end_frame)

    cam_data.shift_x = 0.0
    cam_data.keyframe_insert(data_path="shift_x", frame=turn_end_frame)
    cam_data.keyframe_insert(data_path="shift_x", frame=unfurl_end_frame)

    # Safely retrieve and smooth F-Curves across Blender 4.x/5.x versions
    cam_fcurves = get_object_fcurves(cam_obj)
    for fc in cam_fcurves:
        for kp in fc.keyframe_points:
            kp.interpolation = 'BEZIER'

    cam_data_fcurves = get_object_fcurves(cam_data)
    for fc in cam_data_fcurves:
        for kp in fc.keyframe_points:
            kp.interpolation = 'BEZIER'

    return cam_obj


# ==============================================================================
# GEOMETRY NODES GROUND, GRASS & SKY DOME SETUP
# ==============================================================================

def create_grass_material(grass_dir):
    mat = bpy.data.materials.new(name="Grass_PBR_Material")
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    out_node = nodes.new("ShaderNodeOutputMaterial")
    bsdf = nodes.new("ShaderNodeBsdfPrincipled")
    links.new(bsdf.outputs["BSDF"], out_node.inputs["Surface"])

    col_files = glob.glob(os.path.join(grass_dir, "*Color*.jpg")) + glob.glob(os.path.join(grass_dir, "*JPG*.jpg"))
    if col_files:
        tex_col = nodes.new("ShaderNodeTexImage")
        tex_col.image = bpy.data.images.load(col_files[0])
        links.new(tex_col.outputs["Color"], bsdf.inputs["Base Color"])

    rough_files = glob.glob(os.path.join(grass_dir, "*Roughness*.jpg"))
    if rough_files:
        tex_rough = nodes.new("ShaderNodeTexImage")
        tex_rough.image = bpy.data.images.load(rough_files[0])
        tex_rough.image.colorspace_settings.name = 'Non-Color'
        links.new(tex_rough.outputs["Color"], bsdf.inputs["Roughness"])

    norm_files = glob.glob(os.path.join(grass_dir, "*Normal*.jpg"))
    if norm_files:
        tex_norm = nodes.new("ShaderNodeTexImage")
        tex_norm.image = bpy.data.images.load(norm_files[0])
        tex_norm.image.colorspace_settings.name = 'Non-Color'

        normal_map_node = nodes.new("ShaderNodeNormalMap")
        links.new(tex_norm.outputs["Color"], normal_map_node.inputs["Color"])
        links.new(normal_map_node.outputs["Normal"], bsdf.inputs["Normal"])

    return mat


def create_sky_material(sky_hdri_path):
    mat = bpy.data.materials.new(name="Sky_Dome_Material")
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    out_node = nodes.new("ShaderNodeOutputMaterial")
    emission = nodes.new("ShaderNodeEmission")
    emission.inputs["Strength"].default_value = 1.0

    links.new(emission.outputs["Emission"], out_node.inputs["Surface"])

    if os.path.exists(sky_hdri_path):
        tex_env = nodes.new("ShaderNodeTexEnvironment")
        tex_env.image = bpy.data.images.load(sky_hdri_path)
        links.new(tex_env.outputs["Color"], emission.inputs["Color"])

    return mat


def setup_grass_ground_geometry_nodes(grass_dir, sky_hdri_path, ground_z_offset=-0.2):
    grass_coll = bpy.data.collections.get("Grass_Artifacts_Collection")
    if not grass_coll:
        grass_coll = bpy.data.collections.new("Grass_Artifacts_Collection")
        bpy.context.scene.collection.children.link(grass_coll)

    grass_mat = create_grass_material(grass_dir)
    sky_mat = create_sky_material(sky_hdri_path)

    mesh_files = glob.glob(os.path.join(grass_dir, "*.obj")) + glob.glob(os.path.join(grass_dir, "*.fbx"))

    if mesh_files:
        for fpath in mesh_files:
            if fpath.endswith(".obj"):
                bpy.ops.wm.obj_import(filepath=fpath)
            elif fpath.endswith(".fbx"):
                bpy.ops.import_scene.fbx(filepath=fpath)

            for o in bpy.context.selected_objects:
                grass_coll.objects.link(o)
                if o.name in bpy.context.scene.collection.objects:
                    bpy.context.scene.collection.objects.unlink(o)

        for obj in list(grass_coll.objects):
            if obj.dimensions.z > 0.5:
                grass_coll.objects.unlink(obj)
                bpy.data.objects.remove(obj, do_unlink=True)
    else:
        for i in range(3):
            m_data = bpy.data.meshes.new(f"Grass_Blade_Mesh_{i}")
            g_obj = bpy.data.objects.new(f"Grass_Blade_Obj_{i}", m_data)
            grass_coll.objects.link(g_obj)

            verts = [(0, 0, 0), (0.05, 0, 0), (0.04, 0, 0.15), (0.01, 0, 0.25), (0, 0, 0.3)]
            faces = [(0, 1, 2), (1, 2, 3), (2, 3, 4)]
            m_data.from_pydata(verts, [], faces)
            m_data.update()

            if g_obj.data.materials:
                g_obj.data.materials[0] = grass_mat
            else:
                g_obj.data.materials.append(grass_mat)

            g_obj.location = (100, 100, 0)

    bpy.ops.mesh.primitive_circle_add(radius=12.5, vertices=64, fill_type='NGON', location=(0, 0, ground_z_offset))
    ground_obj = bpy.context.active_object
    ground_obj.name = "Ground_Mesh_Object"

    vg = ground_obj.vertex_groups.new(name="Grass_Weight_Group")
    all_vert_indices = [v.index for v in ground_obj.data.vertices]
    vg.add(all_vert_indices, 1.0, 'REPLACE')
    ground_obj.data.materials.append(grass_mat)
    ground_obj.data.materials.append(sky_mat)

    gn_group = bpy.data.node_groups.new("Grass_Ground_GN", type="GeometryNodeTree")
    gn_group.interface.new_socket(name="Geometry", in_out="INPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Geometry", in_out="OUTPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Density Mask", in_out="INPUT", socket_type="NodeSocketFloat")

    nodes = gn_group.nodes
    links = gn_group.links
    out_node = nodes.new("NodeGroupOutput")
    in_node = nodes.new("NodeGroupInput")

    distribute_node = nodes.new("GeometryNodeDistributePointsOnFaces")
    distribute_node.distribute_method = 'POISSON'
    distribute_node.inputs["Distance Min"].default_value = 0.02
    distribute_node.inputs["Density Max"].default_value = 300.0
    links.new(in_node.outputs["Geometry"], distribute_node.inputs["Mesh"])

    noise_node = nodes.new("ShaderNodeTexNoise")
    noise_node.inputs["Scale"].default_value = 2.5
    noise_node.inputs["Detail"].default_value = 2.0

    color_ramp = nodes.new("ShaderNodeValToRGB")
    color_ramp.color_ramp.elements[0].position = 0.20
    color_ramp.color_ramp.elements[1].position = 0.80
    links.new(noise_node.outputs["Fac"], color_ramp.inputs["Fac"])

    math_weight = nodes.new("ShaderNodeMath")
    math_weight.operation = "MULTIPLY"
    links.new(color_ramp.outputs["Color"], math_weight.inputs[0])
    links.new(in_node.outputs["Density Mask"], math_weight.inputs[1])

    links.new(math_weight.outputs["Value"], distribute_node.inputs["Density Factor"])

    instance_node = nodes.new("GeometryNodeInstanceOnPoints")
    instance_node.inputs["Pick Instance"].default_value = True
    links.new(distribute_node.outputs["Points"], instance_node.inputs["Points"])

    coll_info = nodes.new("GeometryNodeCollectionInfo")
    coll_info.inputs["Collection"].default_value = grass_coll
    coll_info.inputs["Separate Children"].default_value = True
    coll_info.inputs["Reset Children"].default_value = True
    links.new(coll_info.outputs["Instances"], instance_node.inputs["Instance"])

    rand_scale = nodes.new("FunctionNodeRandomValue")
    rand_scale.data_type = 'FLOAT'
    rand_scale.inputs["Min"].default_value = 0.25
    rand_scale.inputs["Max"].default_value = 0.50

    combine_scale = nodes.new("ShaderNodeCombineXYZ")
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["X"])
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["Y"])
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["Z"])
    links.new(combine_scale.outputs["Vector"], instance_node.inputs["Scale"])

    rand_rot_z = nodes.new("FunctionNodeRandomValue")
    rand_rot_z.data_type = 'FLOAT'
    rand_rot_z.inputs["Min"].default_value = 0.0
    rand_rot_z.inputs["Max"].default_value = math.radians(360.0)

    combine_rot = nodes.new("ShaderNodeCombineXYZ")
    links.new(rand_rot_z.outputs["Value"], combine_rot.inputs["Z"])
    links.new(combine_rot.outputs["Vector"], instance_node.inputs["Rotation"])

    sky_sphere = nodes.new("GeometryNodeMeshIcoSphere")
    sky_sphere.inputs["Radius"].default_value = 12.5
    sky_sphere.inputs["Subdivisions"].default_value = 4

    flip_faces = nodes.new("GeometryNodeFlipFaces")
    links.new(sky_sphere.outputs["Mesh"], flip_faces.inputs["Mesh"])

    transform_sky = nodes.new("GeometryNodeTransform")
    transform_sky.inputs["Translation"].default_value = (0.0, 0.0, ground_z_offset)
    links.new(flip_faces.outputs["Mesh"], transform_sky.inputs["Geometry"])

    set_sky_mat = nodes.new("GeometryNodeSetMaterial")
    set_sky_mat.inputs["Material"].default_value = sky_mat
    links.new(transform_sky.outputs["Geometry"], set_sky_mat.inputs["Geometry"])

    join_geo = nodes.new("GeometryNodeJoinGeometry")
    links.new(in_node.outputs["Geometry"], join_geo.inputs["Geometry"])
    links.new(instance_node.outputs["Instances"], join_geo.inputs["Geometry"])
    links.new(set_sky_mat.outputs["Geometry"], join_geo.inputs["Geometry"])

    links.new(join_geo.outputs["Geometry"], out_node.inputs["Geometry"])

    mod = ground_obj.modifiers.new(name="Grass_Ground_GN", type='NODES')
    mod.node_group = gn_group

    for item in gn_group.interface.items_tree:
        if getattr(item, "in_out", None) == "INPUT" and item.name == "Density Mask":
            mod[item.identifier + "_attribute_name"] = "Grass_Weight_Group"
            mod[item.identifier + "_use_attribute"] = True

    return ground_obj


# ==============================================================================
# SCROLL MATERIAL & GEOMETRY NODES SETUP
# ==============================================================================

def create_image_scroll_material(image_path):
    mat_name = "Scroll_Texture_Mat"
    mat = bpy.data.materials.get(mat_name)
    if mat:
        bpy.data.materials.remove(mat)

    mat = bpy.data.materials.new(name=mat_name)
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    output_node = nodes.new("ShaderNodeOutputMaterial")
    bsdf = nodes.new("ShaderNodeBsdfPrincipled")
    bsdf.inputs["Roughness"].default_value = 0.5

    tex_img = nodes.new("ShaderNodeTexImage")
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        tex_img.image = img
        tex_img.extension = 'REPEAT'

    uv_map_node = nodes.new("ShaderNodeUVMap")
    uv_map_node.uv_map = "UVMap"

    links.new(uv_map_node.outputs["UV"], tex_img.inputs["Vector"])
    links.new(tex_img.outputs["Color"], bsdf.inputs["Base Color"])
    links.new(bsdf.outputs["BSDF"], output_node.inputs["Surface"])

    return mat, tex_img.image if os.path.exists(image_path) else None


def create_pure_geometry_node_scroll(image_path=""):
    gn_group = bpy.data.node_groups.new("Scroll_Pure_GN", type="GeometryNodeTree")

    gn_group.interface.new_socket(name="Geometry", in_out="OUTPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Character Mesh", in_out="INPUT", socket_type="NodeSocketObject")

    sock_r_vg = gn_group.interface.new_socket(name="Right Hand Group", in_out="INPUT", socket_type="NodeSocketString")
    sock_r_vg.default_value = "mixamorig:RightHand"

    sock_l_vg = gn_group.interface.new_socket(name="Left Hand Group", in_out="INPUT", socket_type="NodeSocketString")
    sock_l_vg.default_value = "mixamorig:LeftHand"

    sock_offset = gn_group.interface.new_socket(name="Forward Offset", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_offset.default_value = 0.60

    sock_unfurl = gn_group.interface.new_socket(name="Unfurl", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_unfurl.default_value = 0.0

    sock_width = gn_group.interface.new_socket(name="Scroll Width", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_width.default_value = 0.35

    nodes = gn_group.nodes
    links = gn_group.links

    in_node = nodes.new("NodeGroupInput")
    out_node = nodes.new("NodeGroupOutput")

    char_info = nodes.new("GeometryNodeObjectInfo")
    char_info.transform_space = "RELATIVE"
    links.new(in_node.outputs["Character Mesh"], char_info.inputs["Object"])

    pos_node = nodes.new("GeometryNodeInputPosition")

    named_r = nodes.new("GeometryNodeInputNamedAttribute")
    named_r.data_type = "FLOAT"
    links.new(in_node.outputs["Right Hand Group"], named_r.inputs["Name"])

    cmp_r = nodes.new("FunctionNodeCompare")
    cmp_r.data_type = "FLOAT"
    cmp_r.operation = "GREATER_THAN"
    cmp_r.inputs[1].default_value = 0.001
    links.new(named_r.outputs["Attribute"], cmp_r.inputs[0])

    stat_r = nodes.new("GeometryNodeAttributeStatistic")
    stat_r.data_type = "FLOAT_VECTOR"
    links.new(char_info.outputs["Geometry"], stat_r.inputs["Geometry"])
    links.new(cmp_r.outputs["Result"], stat_r.inputs["Selection"])
    links.new(pos_node.outputs["Position"], stat_r.inputs["Attribute"])

    named_l = nodes.new("GeometryNodeInputNamedAttribute")
    named_l.data_type = "FLOAT"
    links.new(in_node.outputs["Left Hand Group"], named_l.inputs["Name"])

    cmp_l = nodes.new("FunctionNodeCompare")
    cmp_l.data_type = "FLOAT"
    cmp_l.operation = "GREATER_THAN"
    cmp_l.inputs[1].default_value = 0.001
    links.new(named_l.outputs["Attribute"], cmp_l.inputs[0])

    stat_l = nodes.new("GeometryNodeAttributeStatistic")
    stat_l.data_type = "FLOAT_VECTOR"
    links.new(char_info.outputs["Geometry"], stat_l.inputs["Geometry"])
    links.new(cmp_l.outputs["Result"], stat_l.inputs["Selection"])
    links.new(pos_node.outputs["Position"], stat_l.inputs["Attribute"])

    neg_offset = nodes.new("ShaderNodeMath")
    neg_offset.operation = "MULTIPLY"
    neg_offset.inputs[1].default_value = -1.0
    links.new(in_node.outputs["Forward Offset"], neg_offset.inputs[0])

    forward_vec = nodes.new("ShaderNodeCombineXYZ")
    forward_vec.inputs["X"].default_value = 0.0
    links.new(neg_offset.outputs["Value"], forward_vec.inputs["Y"])
    forward_vec.inputs["Z"].default_value = 0.0

    start_pos = nodes.new("ShaderNodeVectorMath")
    start_pos.operation = "ADD"
    links.new(stat_r.outputs["Mean"], start_pos.inputs[0])
    links.new(forward_vec.outputs["Vector"], start_pos.inputs[1])

    target_pos = nodes.new("ShaderNodeVectorMath")
    target_pos.operation = "ADD"
    links.new(stat_l.outputs["Mean"], target_pos.inputs[0])
    links.new(forward_vec.outputs["Vector"], target_pos.inputs[1])

    delta_vec = nodes.new("ShaderNodeVectorMath")
    delta_vec.operation = "SUBTRACT"
    links.new(target_pos.outputs["Vector"], delta_vec.inputs[0])
    links.new(start_pos.outputs["Vector"], delta_vec.inputs[1])

    dist_node = nodes.new("ShaderNodeVectorMath")
    dist_node.operation = "DISTANCE"
    links.new(start_pos.outputs["Vector"], dist_node.inputs[0])

    scale_vec = nodes.new("ShaderNodeVectorMath")
    scale_vec.operation = "SCALE"
    links.new(delta_vec.outputs["Vector"], scale_vec.inputs["Vector"])
    links.new(in_node.outputs["Unfurl"], scale_vec.inputs["Scale"])

    end_pos = nodes.new("ShaderNodeVectorMath")
    end_pos.operation = "ADD"
    links.new(start_pos.outputs["Vector"], end_pos.inputs[0])
    links.new(scale_vec.outputs["Vector"], end_pos.inputs[1])

    links.new(end_pos.outputs["Vector"], dist_node.inputs[1])

    curve_line = nodes.new("GeometryNodeCurvePrimitiveLine")
    links.new(start_pos.outputs["Vector"], curve_line.inputs["Start"])
    links.new(end_pos.outputs["Vector"], curve_line.inputs["End"])

    set_tilt = nodes.new("GeometryNodeSetCurveTilt")
    set_tilt.inputs["Tilt"].default_value = 1.5708
    links.new(curve_line.outputs["Curve"], set_tilt.inputs["Curve"])

    spline_param_main = nodes.new("GeometryNodeSplineParameter")
    store_u = nodes.new("GeometryNodeStoreNamedAttribute")
    store_u.inputs["Name"].default_value = "U_Factor"
    store_u.data_type = 'FLOAT'
    store_u.domain = 'POINT'
    links.new(set_tilt.outputs["Curve"], store_u.inputs["Geometry"])
    links.new(spline_param_main.outputs["Factor"], store_u.inputs["Value"])

    half_width_neg = nodes.new("ShaderNodeMath")
    half_width_neg.operation = "MULTIPLY"
    half_width_neg.inputs[1].default_value = -0.5
    links.new(in_node.outputs["Scroll Width"], half_width_neg.inputs[0])

    half_width_pos = nodes.new("ShaderNodeMath")
    half_width_pos.operation = "MULTIPLY"
    half_width_pos.inputs[1].default_value = 0.5
    links.new(in_node.outputs["Scroll Width"], half_width_pos.inputs[0])

    combine_start = nodes.new("ShaderNodeCombineXYZ")
    links.new(half_width_neg.outputs["Value"], combine_start.inputs["X"])

    combine_end = nodes.new("ShaderNodeCombineXYZ")
    links.new(half_width_pos.outputs["Value"], combine_end.inputs["X"])

    profile_line = nodes.new("GeometryNodeCurvePrimitiveLine")
    links.new(combine_start.outputs["Vector"], profile_line.inputs["Start"])
    links.new(combine_end.outputs["Vector"], profile_line.inputs["End"])

    spline_param_prof = nodes.new("GeometryNodeSplineParameter")
    store_v = nodes.new("GeometryNodeStoreNamedAttribute")
    store_v.inputs["Name"].default_value = "V_Factor"
    store_v.data_type = 'FLOAT'
    store_v.domain = 'POINT'
    links.new(profile_line.outputs["Curve"], store_v.inputs["Geometry"])
    links.new(spline_param_prof.outputs["Factor"], store_v.inputs["Value"])

    curve_to_mesh = nodes.new("GeometryNodeCurveToMesh")
    links.new(store_u.outputs["Geometry"], curve_to_mesh.inputs["Curve"])
    links.new(store_v.outputs["Geometry"], curve_to_mesh.inputs["Profile Curve"])

    img_aspect = 1.0
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        if img.size[1] > 0:
            img_aspect = img.size[0] / img.size[1]

    h_times_aspect = nodes.new("ShaderNodeMath")
    h_times_aspect.operation = "MULTIPLY"
    h_times_aspect.inputs[1].default_value = img_aspect
    links.new(in_node.outputs["Scroll Width"], h_times_aspect.inputs[0])

    tile_count = nodes.new("ShaderNodeMath")
    tile_count.operation = "DIVIDE"
    links.new(dist_node.outputs["Value"], tile_count.inputs[0])
    links.new(h_times_aspect.outputs["Value"], tile_count.inputs[1])

    read_u = nodes.new("GeometryNodeInputNamedAttribute")
    read_u.inputs["Name"].default_value = "U_Factor"
    read_u.data_type = 'FLOAT'

    scaled_u = nodes.new("ShaderNodeMath")
    scaled_u.operation = "MULTIPLY"
    links.new(read_u.outputs["Attribute"], scaled_u.inputs[0])
    links.new(tile_count.outputs["Value"], scaled_u.inputs[1])

    read_v = nodes.new("GeometryNodeInputNamedAttribute")
    read_v.inputs["Name"].default_value = "V_Factor"
    read_v.data_type = 'FLOAT'

    flip_v = nodes.new("ShaderNodeMath")
    flip_v.operation = "SUBTRACT"
    flip_v.inputs[0].default_value = 1.0
    links.new(read_v.outputs["Attribute"], flip_v.inputs[1])

    combine_uv = nodes.new("ShaderNodeCombineXYZ")
    links.new(scaled_u.outputs["Value"], combine_uv.inputs["X"])
    links.new(flip_v.outputs["Value"], combine_uv.inputs["Y"])

    store_uv = nodes.new("GeometryNodeStoreNamedAttribute")
    store_uv.inputs["Name"].default_value = "UVMap"
    store_uv.data_type = 'FLOAT2'
    store_uv.domain = 'CORNER'
    links.new(curve_to_mesh.outputs["Mesh"], store_uv.inputs["Geometry"])
    links.new(combine_uv.outputs["Vector"], store_uv.inputs["Value"])

    set_material = nodes.new("GeometryNodeSetMaterial")
    scroll_mat, _ = create_image_scroll_material(image_path)
    set_material.inputs["Material"].default_value = scroll_mat
    links.new(store_uv.outputs["Geometry"], set_material.inputs["Geometry"])

    links.new(set_material.outputs["Geometry"], out_node.inputs["Geometry"])

    return gn_group, scroll_mat


def setup_geometry_node_scroll(target_mesh, unfurl_start, unfurl_end, image_path=""):
    mesh_data = bpy.data.meshes.new("Scroll_Procedural_Mesh")
    scroll_obj = bpy.data.objects.new("Scroll_GN_Object", mesh_data)
    bpy.context.scene.collection.objects.link(scroll_obj)

    mod = scroll_obj.modifiers.new("Scroll_GN", type="NODES")
    gn_tree, scroll_mat = create_pure_geometry_node_scroll(image_path=image_path)
    mod.node_group = gn_tree

    scroll_obj.data.materials.append(scroll_mat)

    r_hand_name, l_hand_name = find_hand_vertex_groups(target_mesh)

    unfurl_id = None
    for item in gn_tree.interface.items_tree:
        if getattr(item, "in_out", None) == "INPUT":
            if item.name == "Character Mesh":
                mod[item.identifier] = target_mesh
            elif item.name == "Right Hand Group":
                mod[item.identifier] = r_hand_name
            elif item.name == "Left Hand Group":
                mod[item.identifier] = l_hand_name
            elif item.name == "Forward Offset":
                mod[item.identifier] = 0.60
            elif item.name == "Scroll Width":
                mod[item.identifier] = 0.35
            elif item.name == "Unfurl":
                unfurl_id = item.identifier

    for m in target_mesh.modifiers:
        if m.type == 'ARMATURE':
            m.show_viewport = True
            m.show_render = True
            m.show_in_editmode = True

    scroll_obj.hide_viewport = True
    scroll_obj.hide_render = True
    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=1)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=1)

    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=unfurl_start - 1)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=unfurl_start - 1)

    scroll_obj.hide_viewport = False
    scroll_obj.hide_render = False
    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=unfurl_start)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=unfurl_start)

    for fc in get_object_fcurves(scroll_obj):
        if fc.data_path in ["hide_viewport", "hide_render"]:
            for kp in fc.keyframe_points:
                kp.interpolation = 'CONSTANT'

    if unfurl_id:
        mod[unfurl_id] = 0.001
        mod.keyframe_insert(data_path=f'["{unfurl_id}"]', frame=unfurl_start)

        mod[unfurl_id] = 1.0
        mod.keyframe_insert(data_path=f'["{unfurl_id}"]', frame=unfurl_end)

    return scroll_obj


# ==============================================================================
# MAIN CONTROLLER
# ==============================================================================

def import_and_extract_action(fbx_path, action_name):
    existing_objects = set(bpy.data.objects)
    bpy.ops.import_scene.fbx(filepath=fbx_path)
    bpy.context.view_layer.update()

    imported_objects = set(bpy.data.objects) - existing_objects
    extracted_action = None
    for obj in imported_objects:
        if obj.animation_data and obj.animation_data.action:
            extracted_action = obj.animation_data.action
            extracted_action.name = action_name
            extracted_action.use_fake_user = True
            break
    return extracted_action, imported_objects


def purge_imported_objects(objects_to_remove):
    for obj in objects_to_remove:
        if obj.name in bpy.data.objects:
            if obj.animation_data:
                obj.animation_data_clear()
            for coll in list(obj.users_collection):
                coll.objects.unlink(obj)
    bpy.context.view_layer.update()


def main():
    cleanup_previous_run()

    walk_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Mutant Walking.fbx"
    turn_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Right Turn 90.fbx"
    unfurl_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Mutant Idle.fbx"

    sky_hdri_path = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/horn-koppe_spring_4k.exr"
    output_blend = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/output_scene.blend"

    image_path = "/Users/dobbinp1/Downloads/IMGP6691_psedit.jpg"
    grass_dir = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Grass004_4K-JPG"

    setup_grass_ground_geometry_nodes(grass_dir, sky_hdri_path, ground_z_offset=-0.2)

    CYCLES = 4
    CROSSFADE = 15
    TARGET_DISTANCE = 4.0

    walk_action, walk_objs = import_and_extract_action(walk_fbx, "walk")
    target_rig = next((obj for obj in walk_objs if obj.type == "ARMATURE"), None)
    target_mesh = next((obj for obj in walk_objs if obj.type == "MESH"), None)

    if not target_rig or not target_mesh:
        raise RuntimeError("Armature or Character Mesh not found in walk FBX.")

    travel_axis_idx, motion_sign = detect_natural_forward_axis(walk_action, target_rig)
    initial_heading = math.radians(90.0) if travel_axis_idx == 0 else 0.0
    target_rig.rotation_euler[2] = initial_heading

    action_len = int(walk_action.frame_range[1] - walk_action.frame_range[0])
    if action_len <= 0:
        action_len = 30

    anim_data = target_rig.animation_data_create()

    walk_track = anim_data.nla_tracks.new()
    walk_track.name = "Walk_Track"

    current_frame = 1
    for i in range(CYCLES):
        strip = walk_track.strips.new(name=f"Walk_{i+1}", start=int(current_frame), action=walk_action)
        strip.blend_type = "REPLACE"
        current_frame += action_len

    walk_end_frame = int(current_frame)

    turn_action, turn_objs = import_and_extract_action(turn_fbx, "rightturn")
    purge_imported_objects(turn_objs)

    turn_track = anim_data.nla_tracks.new()
    turn_track.name = "RightTurn_Track"

    turn_start = int(walk_end_frame - CROSSFADE)
    turn_strip = turn_track.strips.new(name="RightTurn", start=turn_start, action=turn_action)
    turn_strip.blend_type = "REPLACE"
    turn_strip.blend_in = CROSSFADE
    turn_strip.use_auto_blend = True
    turn_strip.extrapolation = "HOLD"
    turn_end_frame = int(turn_strip.frame_end)

    unfurl_action, unfurl_objs = import_and_extract_action(unfurl_fbx, "unfurl")
    purge_imported_objects(unfurl_objs)

    unfurl_track = anim_data.nla_tracks.new()
    unfurl_track.name = "Unfurl_Track"

    unfurl_start = int(turn_end_frame - CROSSFADE)
    unfurl_strip = unfurl_track.strips.new(name="Unfurl", start=unfurl_start, action=unfurl_action)
    unfurl_strip.blend_type = "REPLACE"
    unfurl_strip.blend_in = CROSSFADE
    unfurl_strip.use_auto_blend = True
    unfurl_strip.extrapolation = "HOLD"
    unfurl_end_frame = int(unfurl_strip.frame_end)

    root_action = bpy.data.actions.new(name="World_Root_Motion")
    target_rig.animation_data.action = root_action

    loc_fcurve = root_action.fcurve_ensure_for_datablock(datablock=target_rig, data_path="location", index=travel_axis_idx)
    final_target_value = TARGET_DISTANCE * motion_sign

    k_loc1 = loc_fcurve.keyframe_points.insert(frame=1.0, value=0.0)
    k_loc2 = loc_fcurve.keyframe_points.insert(frame=float(turn_start), value=final_target_value)
    k_loc3 = loc_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=final_target_value)

    k_loc1.interpolation = "LINEAR"
    k_loc2.interpolation = "CONSTANT"
    k_loc3.interpolation = "CONSTANT"

    rot_fcurve = root_action.fcurve_ensure_for_datablock(datablock=target_rig, data_path="rotation_euler", index=2)
    target_heading = initial_heading - math.radians(90.0)

    k_rot1 = rot_fcurve.keyframe_points.insert(frame=1.0, value=initial_heading)
    k_rot2 = rot_fcurve.keyframe_points.insert(frame=float(turn_start), value=initial_heading)
    k_rot3 = rot_fcurve.keyframe_points.insert(frame=float(turn_end_frame), value=target_heading)
    k_rot4 = rot_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=target_heading)

    k_rot1.interpolation = "CONSTANT"
    k_rot2.interpolation = "BEZIER"
    k_rot3.interpolation = "CONSTANT"
    k_rot4.interpolation = "CONSTANT"

    root_track = anim_data.nla_tracks.new()
    root_track.name = "World_Root_Track"
    root_strip = root_track.strips.new(name="Root_Motion_Strip", start=1, action=root_action)
    root_strip.blend_type = "REPLACE"
    root_strip.extrapolation = "HOLD"

    anim_data.action = None

    setup_geometry_node_scroll(
        target_mesh=target_mesh,
        unfurl_start=unfurl_start,
        unfurl_end=unfurl_end_frame,
        image_path=image_path,
    )

    cam_obj = setup_camera_tracking(
        target_mesh=target_mesh,
        turn_start=turn_start,
        turn_end_frame=turn_end_frame,
        unfurl_end_frame=unfurl_end_frame,
        focal_length=35.0
    )

    configure_camera_and_viewport_horizon(far_clip=50.0)

    bpy.context.scene.frame_start = 1
    bpy.context.scene.frame_end = unfurl_end_frame

    bpy.ops.wm.save_as_mainfile(filepath=output_blend)
    print(f"\n[Success] Camera tracking setup complete. Saved to: {output_blend}")


main()
```

```
import math
import os
import glob
import bpy
from bpy_extras import anim_utils

# ==============================================================================
# CLEANUP & HELPERS
# ==============================================================================

def cleanup_previous_run():
    """Clears old scroll objects, ground, node groups, camera anchors, collections, and orphaned data."""
    target_names = [
        "Master_NLA_Control",
        "Scroll_GN_Object",
        "Scroll_Start_Pin",
        "Scroll_End_Pin",
        "Scroll_Master_Container",
        "HDRI_Ground_Plane",
        "HDRI_Camera",
        "Ground_Mesh_Object",
        "Camera_Tracking_Anchor",
        "Tracking_Camera",
        "Camera_Focus_Target"
    ]
    for obj in list(bpy.data.objects):
        if obj.name in target_names or (obj.name.startswith("mixamorig") and obj.type != "ARMATURE" and obj.type != "MESH"):
            if obj.animation_data:
                obj.animation_data_clear()
            bpy.data.objects.remove(obj, do_unlink=True)

    for action in list(bpy.data.actions):
        if action.users == 0 or action.use_fake_user:
            action.use_fake_user = False
            bpy.data.actions.remove(action)

    for ng in list(bpy.data.node_groups):
        if ng.name in ["Scroll_Unfurl_GN", "Scroll_Pure_GN", "HDRI_Ground_GN", "Grass_Ground_GN", "Camera_Tracker_GN"]:
            bpy.data.node_groups.remove(ng)

    for coll_name in ["Scroll_Master_Container", "Grass_Artifacts_Collection"]:
        if coll_name in bpy.data.collections:
            coll = bpy.data.collections[coll_name]
            for obj in list(coll.objects):
                bpy.data.objects.remove(obj, do_unlink=True)
            bpy.data.collections.remove(coll)

    bpy.context.view_layer.update()


def configure_camera_and_viewport_horizon(far_clip=50.0):
    if bpy.context.scene.camera:
        bpy.context.scene.camera.data.clip_end = far_clip

    for area in bpy.context.screen.areas:
        if area.type == 'VIEW_3D':
            for space in area.spaces:
                if space.type == 'VIEW_3D':
                    space.clip_end = far_clip


def get_slot_fcurves(action, anim_target_object):
    fcurves = []
    if not action:
        return fcurves

    anim_data = getattr(anim_target_object, "animation_data", None)
    if anim_data and getattr(anim_data, "action_slot", None):
        slot = anim_data.action_slot
        channelbag = anim_utils.action_get_channelbag_for_slot(action, slot)
        if channelbag and hasattr(channelbag, "fcurves"):
            return list(channelbag.fcurves)

    if hasattr(action, "slots"):
        for slot in action.slots:
            for cb in getattr(slot, "channelbags", []):
                fcurves.extend(getattr(cb, "fcurves", []))
        if fcurves:
            return fcurves

    if hasattr(action, "fcurves"):
        return list(action.fcurves)

    return fcurves


def get_object_fcurves(obj):
    if not obj or not obj.animation_data or not obj.animation_data.action:
        return []
    action = obj.animation_data.action
    return get_slot_fcurves(action, obj)


def detect_natural_forward_axis(action, rig_object):
    fcurves = get_slot_fcurves(action, rig_object)
    max_delta = 0.0
    best_axis = 1
    net_delta = 0.0

    start_frame = action.frame_range[0]
    end_frame = action.frame_range[1]

    for fc in fcurves:
        if fc.data_path.endswith("location") and fc.array_index in [0, 1, 2]:
            start_val = fc.evaluate(start_frame)
            end_val = fc.evaluate(end_frame)
            delta = end_val - start_val
            if abs(delta) > abs(max_delta):
                max_delta = delta
                best_axis = fc.array_index
                net_delta = delta

    direction_sign = 1.0 if net_delta >= 0 else -1.0
    return best_axis, direction_sign


def find_hand_vertex_groups(target_mesh):
    r_name = "mixamorig:RightHand"
    l_name = "mixamorig:LeftHand"

    if target_mesh and hasattr(target_mesh, "vertex_groups"):
        for vg in target_mesh.vertex_groups:
            vg_lower = vg.name.lower()
            if "righthand" in vg_lower or "hand_r" in vg_lower or "hand.r" in vg_lower:
                r_name = vg.name
            elif "lefthand" in vg_lower or "hand_l" in vg_lower or "hand.l" in vg_lower:
                l_name = vg.name

    return r_name, l_name


# ==============================================================================
# CAMERA TRACKING SETUP WITH ANIMATED ANGLE & SHIFT
# ==============================================================================

def setup_camera_tracking(target_mesh, turn_start, turn_end_frame, unfurl_end_frame, focal_length=35.0):
    """
    Sets up a tracking camera that positions the monster in the first-third of
    the frame at 45° during walk, and smoothly animates to center frame at 90°
    when turning.
    """
    # 1. Create Target Mesh Constraint Lock
    cam_data = bpy.data.cameras.new("Tracking_Camera_Data")
    cam_data.lens = focal_length

    cam_obj = bpy.data.objects.new("Tracking_Camera", cam_data)
    bpy.context.scene.collection.objects.link(cam_obj)
    bpy.context.scene.camera = cam_obj

    copy_pos = cam_obj.constraints.new(type='COPY_LOCATION')
    copy_pos.target = target_mesh
    copy_pos.use_offset = True

    track_to = cam_obj.constraints.new(type='TRACK_TO')
    track_to.target = target_mesh
    track_to.track_axis = 'TRACK_NEGATIVE_Z'
    track_to.up_axis = 'UP_Y'
    track_to.use_target_z = True

    # 2. Position Offsets
    # 45-degree angle offset along (X, Y, Z) diagonal relative to target:
    dist = 7.5
    angle_45_x = dist * math.sin(math.radians(45))  # ~5.30m
    angle_45_y = -dist * math.cos(math.radians(45)) # ~-5.30m

    # 90-degree side/front offset (direct on Y axis):
    angle_90_x = 0.0
    angle_90_y = -dist

    cam_height = 1.0

    # 3. Animate Location Offset & Camera Shift
    # Frame 1 to Turn Start: 45 Degree Angle + Monster in 1st Third (shift_x = -0.3)
    cam_obj.location = (angle_45_x, angle_45_y, cam_height)
    cam_obj.keyframe_insert(data_path="location", frame=1)
    cam_obj.keyframe_insert(data_path="location", frame=turn_start)

    cam_data.shift_x = -0.3  # Offsets monster into first third of the frame
    cam_data.shift_y = 0.2
    cam_data.keyframe_insert(data_path="shift_x", frame=1)
    cam_data.keyframe_insert(data_path="shift_x", frame=turn_start)

    # Turn Start to Turn End: Animate to 90 Degree Angle + Centered Monster (shift_x = 0.0)
    cam_obj.location = (angle_90_x, angle_90_y, cam_height)
    cam_obj.keyframe_insert(data_path="location", frame=turn_end_frame)
    cam_obj.keyframe_insert(data_path="location", frame=unfurl_end_frame)

    cam_data.shift_x = 0.0   # Moves monster to frame center
    cam_data.keyframe_insert(data_path="shift_x", frame=turn_end_frame)
    cam_data.keyframe_insert(data_path="shift_x", frame=unfurl_end_frame)

    # Smooth out interpolation for camera transition during turn
    if cam_obj.animation_data and cam_obj.animation_data.action:
        for fc in cam_obj.animation_data.action.fcurves:
            for kp in fc.keyframe_points:
                kp.interpolation = 'BEZIER'

    if cam_data.animation_data and cam_data.animation_data.action:
        for fc in cam_data.animation_data.action.fcurves:
            for kp in fc.keyframe_points:
                kp.interpolation = 'BEZIER'

    return cam_obj


# ==============================================================================
# GEOMETRY NODES GROUND, GRASS & SKY DOME SETUP
# ==============================================================================

def create_grass_material(grass_dir):
    mat = bpy.data.materials.new(name="Grass_PBR_Material")
    mat.use_nodes = True
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    out_node = nodes.new("ShaderNodeOutputMaterial")
    bsdf = nodes.new("ShaderNodeBsdfPrincipled")
    links.new(bsdf.outputs["BSDF"], out_node.inputs["Surface"])

    col_files = glob.glob(os.path.join(grass_dir, "*Color*.jpg")) + glob.glob(os.path.join(grass_dir, "*JPG*.jpg"))
    if col_files:
        tex_col = nodes.new("ShaderNodeTexImage")
        tex_col.image = bpy.data.images.load(col_files[0])
        links.new(tex_col.outputs["Color"], bsdf.inputs["Base Color"])

    rough_files = glob.glob(os.path.join(grass_dir, "*Roughness*.jpg"))
    if rough_files:
        tex_rough = nodes.new("ShaderNodeTexImage")
        tex_rough.image = bpy.data.images.load(rough_files[0])
        tex_rough.image.colorspace_settings.name = 'Non-Color'
        links.new(tex_rough.outputs["Color"], bsdf.inputs["Roughness"])

    norm_files = glob.glob(os.path.join(grass_dir, "*Normal*.jpg"))
    if norm_files:
        tex_norm = nodes.new("ShaderNodeTexImage")
        tex_norm.image = bpy.data.images.load(norm_files[0])
        tex_norm.image.colorspace_settings.name = 'Non-Color'

        normal_map_node = nodes.new("ShaderNodeNormalMap")
        links.new(tex_norm.outputs["Color"], normal_map_node.inputs["Color"])
        links.new(normal_map_node.outputs["Normal"], bsdf.inputs["Normal"])

    return mat


def create_sky_material(sky_hdri_path):
    mat = bpy.data.materials.new(name="Sky_Dome_Material")
    mat.use_nodes = True
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    out_node = nodes.new("ShaderNodeOutputMaterial")
    emission = nodes.new("ShaderNodeEmission")
    emission.inputs["Strength"].default_value = 1.0

    links.new(emission.outputs["Emission"], out_node.inputs["Surface"])

    if os.path.exists(sky_hdri_path):
        tex_env = nodes.new("ShaderNodeTexEnvironment")
        tex_env.image = bpy.data.images.load(sky_hdri_path)
        links.new(tex_env.outputs["Color"], emission.inputs["Color"])

    return mat


def setup_grass_ground_geometry_nodes(grass_dir, sky_hdri_path, ground_z_offset=-0.2):
    grass_coll = bpy.data.collections.get("Grass_Artifacts_Collection")
    if not grass_coll:
        grass_coll = bpy.data.collections.new("Grass_Artifacts_Collection")
        bpy.context.scene.collection.children.link(grass_coll)

    grass_mat = create_grass_material(grass_dir)
    sky_mat = create_sky_material(sky_hdri_path)

    mesh_files = glob.glob(os.path.join(grass_dir, "*.obj")) + glob.glob(os.path.join(grass_dir, "*.fbx"))

    if mesh_files:
        for fpath in mesh_files:
            if fpath.endswith(".obj"):
                bpy.ops.wm.obj_import(filepath=fpath)
            elif fpath.endswith(".fbx"):
                bpy.ops.import_scene.fbx(filepath=fpath)

            for o in bpy.context.selected_objects:
                grass_coll.objects.link(o)
                if o.name in bpy.context.scene.collection.objects:
                    bpy.context.scene.collection.objects.unlink(o)

        for obj in list(grass_coll.objects):
            if obj.dimensions.z > 0.5:
                grass_coll.objects.unlink(obj)
                bpy.data.objects.remove(obj, do_unlink=True)
    else:
        for i in range(3):
            m_data = bpy.data.meshes.new(f"Grass_Blade_Mesh_{i}")
            g_obj = bpy.data.objects.new(f"Grass_Blade_Obj_{i}", m_data)
            grass_coll.objects.link(g_obj)

            verts = [(0, 0, 0), (0.05, 0, 0), (0.04, 0, 0.15), (0.01, 0, 0.25), (0, 0, 0.3)]
            faces = [(0, 1, 2), (1, 2, 3), (2, 3, 4)]
            m_data.from_pydata(verts, [], faces)
            m_data.update()

            if g_obj.data.materials:
                g_obj.data.materials[0] = grass_mat
            else:
                g_obj.data.materials.append(grass_mat)

            g_obj.location = (100, 100, 0)

    # Circular Ground Mesh lowered by ground_z_offset
    bpy.ops.mesh.primitive_circle_add(radius=12.5, vertices=64, fill_type='NGON', location=(0, 0, ground_z_offset))
    ground_obj = bpy.context.active_object
    ground_obj.name = "Ground_Mesh_Object"

    vg = ground_obj.vertex_groups.new(name="Grass_Weight_Group")
    all_vert_indices = [v.index for v in ground_obj.data.vertices]
    vg.add(all_vert_indices, 1.0, 'REPLACE')
    ground_obj.data.materials.append(grass_mat)
    ground_obj.data.materials.append(sky_mat)

    gn_group = bpy.data.node_groups.new("Grass_Ground_GN", type="GeometryNodeTree")
    gn_group.interface.new_socket(name="Geometry", in_out="INPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Geometry", in_out="OUTPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Density Mask", in_out="INPUT", socket_type="NodeSocketFloat")

    nodes = gn_group.nodes
    links = gn_group.links
    out_node = nodes.new("NodeGroupOutput")
    in_node = nodes.new("NodeGroupInput")

    distribute_node = nodes.new("GeometryNodeDistributePointsOnFaces")
    distribute_node.distribute_method = 'POISSON'
    distribute_node.inputs["Distance Min"].default_value = 0.02
    distribute_node.inputs["Density Max"].default_value = 300.0
    links.new(in_node.outputs["Geometry"], distribute_node.inputs["Mesh"])

    noise_node = nodes.new("ShaderNodeTexNoise")
    noise_node.inputs["Scale"].default_value = 2.5
    noise_node.inputs["Detail"].default_value = 2.0

    color_ramp = nodes.new("ShaderNodeValToRGB")
    color_ramp.color_ramp.elements[0].position = 0.20
    color_ramp.color_ramp.elements[1].position = 0.80
    links.new(noise_node.outputs["Fac"], color_ramp.inputs["Fac"])

    math_weight = nodes.new("ShaderNodeMath")
    math_weight.operation = "MULTIPLY"
    links.new(color_ramp.outputs["Color"], math_weight.inputs[0])
    links.new(in_node.outputs["Density Mask"], math_weight.inputs[1])

    links.new(math_weight.outputs["Value"], distribute_node.inputs["Density Factor"])

    instance_node = nodes.new("GeometryNodeInstanceOnPoints")
    instance_node.inputs["Pick Instance"].default_value = True
    links.new(distribute_node.outputs["Points"], instance_node.inputs["Points"])

    coll_info = nodes.new("GeometryNodeCollectionInfo")
    coll_info.inputs["Collection"].default_value = grass_coll
    coll_info.inputs["Separate Children"].default_value = True
    coll_info.inputs["Reset Children"].default_value = True
    links.new(coll_info.outputs["Instances"], instance_node.inputs["Instance"])

    rand_scale = nodes.new("FunctionNodeRandomValue")
    rand_scale.data_type = 'FLOAT'
    rand_scale.inputs["Min"].default_value = 0.25
    rand_scale.inputs["Max"].default_value = 0.50

    combine_scale = nodes.new("ShaderNodeCombineXYZ")
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["X"])
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["Y"])
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["Z"])
    links.new(combine_scale.outputs["Vector"], instance_node.inputs["Scale"])

    rand_rot_z = nodes.new("FunctionNodeRandomValue")
    rand_rot_z.data_type = 'FLOAT'
    rand_rot_z.inputs["Min"].default_value = 0.0
    rand_rot_z.inputs["Max"].default_value = math.radians(360.0)

    combine_rot = nodes.new("ShaderNodeCombineXYZ")
    links.new(rand_rot_z.outputs["Value"], combine_rot.inputs["Z"])
    links.new(combine_rot.outputs["Vector"], instance_node.inputs["Rotation"])

    sky_sphere = nodes.new("GeometryNodeMeshIcoSphere")
    sky_sphere.inputs["Radius"].default_value = 12.5
    sky_sphere.inputs["Subdivisions"].default_value = 4

    flip_faces = nodes.new("GeometryNodeFlipFaces")
    links.new(sky_sphere.outputs["Mesh"], flip_faces.inputs["Mesh"])

    transform_sky = nodes.new("GeometryNodeTransform")
    transform_sky.inputs["Translation"].default_value = (0.0, 0.0, ground_z_offset)
    links.new(flip_faces.outputs["Mesh"], transform_sky.inputs["Geometry"])

    set_sky_mat = nodes.new("GeometryNodeSetMaterial")
    set_sky_mat.inputs["Material"].default_value = sky_mat
    links.new(transform_sky.outputs["Geometry"], set_sky_mat.inputs["Geometry"])

    join_geo = nodes.new("GeometryNodeJoinGeometry")
    links.new(in_node.outputs["Geometry"], join_geo.inputs["Geometry"])
    links.new(instance_node.outputs["Instances"], join_geo.inputs["Geometry"])
    links.new(set_sky_mat.outputs["Geometry"], join_geo.inputs["Geometry"])

    links.new(join_geo.outputs["Geometry"], out_node.inputs["Geometry"])

    mod = ground_obj.modifiers.new(name="Grass_Ground_GN", type='NODES')
    mod.node_group = gn_group

    for item in gn_group.interface.items_tree:
        if getattr(item, "in_out", None) == "INPUT" and item.name == "Density Mask":
            mod[item.identifier + "_attribute_name"] = "Grass_Weight_Group"
            mod[item.identifier + "_use_attribute"] = True

    return ground_obj


# ==============================================================================
# SCROLL MATERIAL & GEOMETRY NODES SETUP
# ==============================================================================

def create_image_scroll_material(image_path):
    mat_name = "Scroll_Texture_Mat"
    mat = bpy.data.materials.get(mat_name)
    if mat:
        bpy.data.materials.remove(mat)

    mat = bpy.data.materials.new(name=mat_name)
    if hasattr(mat, "use_nodes"):
        try:
            mat.use_nodes = True
        except AttributeError:
            pass

    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    output_node = nodes.new("ShaderNodeOutputMaterial")
    bsdf = nodes.new("ShaderNodeBsdfPrincipled")
    bsdf.inputs["Roughness"].default_value = 0.5

    tex_img = nodes.new("ShaderNodeTexImage")
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        tex_img.image = img
        tex_img.extension = 'REPEAT'

    uv_map_node = nodes.new("ShaderNodeUVMap")
    uv_map_node.uv_map = "UVMap"

    links.new(uv_map_node.outputs["UV"], tex_img.inputs["Vector"])
    links.new(tex_img.outputs["Color"], bsdf.inputs["Base Color"])
    links.new(bsdf.outputs["BSDF"], output_node.inputs["Surface"])

    return mat, tex_img.image if os.path.exists(image_path) else None


def create_pure_geometry_node_scroll(image_path=""):
    gn_group = bpy.data.node_groups.new("Scroll_Pure_GN", type="GeometryNodeTree")

    gn_group.interface.new_socket(name="Geometry", in_out="OUTPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Character Mesh", in_out="INPUT", socket_type="NodeSocketObject")

    sock_r_vg = gn_group.interface.new_socket(name="Right Hand Group", in_out="INPUT", socket_type="NodeSocketString")
    sock_r_vg.default_value = "mixamorig:RightHand"

    sock_l_vg = gn_group.interface.new_socket(name="Left Hand Group", in_out="INPUT", socket_type="NodeSocketString")
    sock_l_vg.default_value = "mixamorig:LeftHand"

    sock_offset = gn_group.interface.new_socket(name="Forward Offset", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_offset.default_value = 0.60

    sock_unfurl = gn_group.interface.new_socket(name="Unfurl", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_unfurl.default_value = 0.0

    sock_width = gn_group.interface.new_socket(name="Scroll Width", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_width.default_value = 0.35

    nodes = gn_group.nodes
    links = gn_group.links

    in_node = nodes.new("NodeGroupInput")
    out_node = nodes.new("NodeGroupOutput")

    char_info = nodes.new("GeometryNodeObjectInfo")
    char_info.transform_space = "RELATIVE"
    links.new(in_node.outputs["Character Mesh"], char_info.inputs["Object"])

    pos_node = nodes.new("GeometryNodeInputPosition")

    named_r = nodes.new("GeometryNodeInputNamedAttribute")
    named_r.data_type = "FLOAT"
    links.new(in_node.outputs["Right Hand Group"], named_r.inputs["Name"])

    cmp_r = nodes.new("FunctionNodeCompare")
    cmp_r.data_type = "FLOAT"
    cmp_r.operation = "GREATER_THAN"
    cmp_r.inputs[1].default_value = 0.001
    links.new(named_r.outputs["Attribute"], cmp_r.inputs[0])

    stat_r = nodes.new("GeometryNodeAttributeStatistic")
    stat_r.data_type = "FLOAT_VECTOR"
    links.new(char_info.outputs["Geometry"], stat_r.inputs["Geometry"])
    links.new(cmp_r.outputs["Result"], stat_r.inputs["Selection"])
    links.new(pos_node.outputs["Position"], stat_r.inputs["Attribute"])

    named_l = nodes.new("GeometryNodeInputNamedAttribute")
    named_l.data_type = "FLOAT"
    links.new(in_node.outputs["Left Hand Group"], named_l.inputs["Name"])

    cmp_l = nodes.new("FunctionNodeCompare")
    cmp_l.data_type = "FLOAT"
    cmp_l.operation = "GREATER_THAN"
    cmp_l.inputs[1].default_value = 0.001
    links.new(named_l.outputs["Attribute"], cmp_l.inputs[0])

    stat_l = nodes.new("GeometryNodeAttributeStatistic")
    stat_l.data_type = "FLOAT_VECTOR"
    links.new(char_info.outputs["Geometry"], stat_l.inputs["Geometry"])
    links.new(cmp_l.outputs["Result"], stat_l.inputs["Selection"])
    links.new(pos_node.outputs["Position"], stat_l.inputs["Attribute"])

    neg_offset = nodes.new("ShaderNodeMath")
    neg_offset.operation = "MULTIPLY"
    neg_offset.inputs[1].default_value = -1.0
    links.new(in_node.outputs["Forward Offset"], neg_offset.inputs[0])

    forward_vec = nodes.new("ShaderNodeCombineXYZ")
    forward_vec.inputs["X"].default_value = 0.0
    links.new(neg_offset.outputs["Value"], forward_vec.inputs["Y"])
    forward_vec.inputs["Z"].default_value = 0.0

    start_pos = nodes.new("ShaderNodeVectorMath")
    start_pos.operation = "ADD"
    links.new(stat_r.outputs["Mean"], start_pos.inputs[0])
    links.new(forward_vec.outputs["Vector"], start_pos.inputs[1])

    target_pos = nodes.new("ShaderNodeVectorMath")
    target_pos.operation = "ADD"
    links.new(stat_l.outputs["Mean"], target_pos.inputs[0])
    links.new(forward_vec.outputs["Vector"], target_pos.inputs[1])

    delta_vec = nodes.new("ShaderNodeVectorMath")
    delta_vec.operation = "SUBTRACT"
    links.new(target_pos.outputs["Vector"], delta_vec.inputs[0])
    links.new(start_pos.outputs["Vector"], delta_vec.inputs[1])

    dist_node = nodes.new("ShaderNodeVectorMath")
    dist_node.operation = "DISTANCE"
    links.new(start_pos.outputs["Vector"], dist_node.inputs[0])

    scale_vec = nodes.new("ShaderNodeVectorMath")
    scale_vec.operation = "SCALE"
    links.new(delta_vec.outputs["Vector"], scale_vec.inputs["Vector"])
    links.new(in_node.outputs["Unfurl"], scale_vec.inputs["Scale"])

    end_pos = nodes.new("ShaderNodeVectorMath")
    end_pos.operation = "ADD"
    links.new(start_pos.outputs["Vector"], end_pos.inputs[0])
    links.new(scale_vec.outputs["Vector"], end_pos.inputs[1])

    links.new(end_pos.outputs["Vector"], dist_node.inputs[1])

    curve_line = nodes.new("GeometryNodeCurvePrimitiveLine")
    links.new(start_pos.outputs["Vector"], curve_line.inputs["Start"])
    links.new(end_pos.outputs["Vector"], curve_line.inputs["End"])

    set_tilt = nodes.new("GeometryNodeSetCurveTilt")
    set_tilt.inputs["Tilt"].default_value = 1.5708
    links.new(curve_line.outputs["Curve"], set_tilt.inputs["Curve"])

    spline_param_main = nodes.new("GeometryNodeSplineParameter")
    store_u = nodes.new("GeometryNodeStoreNamedAttribute")
    store_u.inputs["Name"].default_value = "U_Factor"
    store_u.data_type = 'FLOAT'
    store_u.domain = 'POINT'
    links.new(set_tilt.outputs["Curve"], store_u.inputs["Geometry"])
    links.new(spline_param_main.outputs["Factor"], store_u.inputs["Value"])

    half_width_neg = nodes.new("ShaderNodeMath")
    half_width_neg.operation = "MULTIPLY"
    half_width_neg.inputs[1].default_value = -0.5
    links.new(in_node.outputs["Scroll Width"], half_width_neg.inputs[0])

    half_width_pos = nodes.new("ShaderNodeMath")
    half_width_pos.operation = "MULTIPLY"
    half_width_pos.inputs[1].default_value = 0.5
    links.new(in_node.outputs["Scroll Width"], half_width_pos.inputs[0])

    combine_start = nodes.new("ShaderNodeCombineXYZ")
    links.new(half_width_neg.outputs["Value"], combine_start.inputs["X"])

    combine_end = nodes.new("ShaderNodeCombineXYZ")
    links.new(half_width_pos.outputs["Value"], combine_end.inputs["X"])

    profile_line = nodes.new("GeometryNodeCurvePrimitiveLine")
    links.new(combine_start.outputs["Vector"], profile_line.inputs["Start"])
    links.new(combine_end.outputs["Vector"], profile_line.inputs["End"])

    spline_param_prof = nodes.new("GeometryNodeSplineParameter")
    store_v = nodes.new("GeometryNodeStoreNamedAttribute")
    store_v.inputs["Name"].default_value = "V_Factor"
    store_v.data_type = 'FLOAT'
    store_v.domain = 'POINT'
    links.new(profile_line.outputs["Curve"], store_v.inputs["Geometry"])
    links.new(spline_param_prof.outputs["Factor"], store_v.inputs["Value"])

    curve_to_mesh = nodes.new("GeometryNodeCurveToMesh")
    links.new(store_u.outputs["Geometry"], curve_to_mesh.inputs["Curve"])
    links.new(store_v.outputs["Geometry"], curve_to_mesh.inputs["Profile Curve"])

    img_aspect = 1.0
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        if img.size[1] > 0:
            img_aspect = img.size[0] / img.size[1]

    h_times_aspect = nodes.new("ShaderNodeMath")
    h_times_aspect.operation = "MULTIPLY"
    h_times_aspect.inputs[1].default_value = img_aspect
    links.new(in_node.outputs["Scroll Width"], h_times_aspect.inputs[0])

    tile_count = nodes.new("ShaderNodeMath")
    tile_count.operation = "DIVIDE"
    links.new(dist_node.outputs["Value"], tile_count.inputs[0])
    links.new(h_times_aspect.outputs["Value"], tile_count.inputs[1])

    read_u = nodes.new("GeometryNodeInputNamedAttribute")
    read_u.inputs["Name"].default_value = "U_Factor"
    read_u.data_type = 'FLOAT'

    scaled_u = nodes.new("ShaderNodeMath")
    scaled_u.operation = "MULTIPLY"
    links.new(read_u.outputs["Attribute"], scaled_u.inputs[0])
    links.new(tile_count.outputs["Value"], scaled_u.inputs[1])

    read_v = nodes.new("GeometryNodeInputNamedAttribute")
    read_v.inputs["Name"].default_value = "V_Factor"
    read_v.data_type = 'FLOAT'

    flip_v = nodes.new("ShaderNodeMath")
    flip_v.operation = "SUBTRACT"
    flip_v.inputs[0].default_value = 1.0
    links.new(read_v.outputs["Attribute"], flip_v.inputs[1])

    combine_uv = nodes.new("ShaderNodeCombineXYZ")
    links.new(scaled_u.outputs["Value"], combine_uv.inputs["X"])
    links.new(flip_v.outputs["Value"], combine_uv.inputs["Y"])

    store_uv = nodes.new("GeometryNodeStoreNamedAttribute")
    store_uv.inputs["Name"].default_value = "UVMap"
    store_uv.data_type = 'FLOAT2'
    store_uv.domain = 'CORNER'
    links.new(curve_to_mesh.outputs["Mesh"], store_uv.inputs["Geometry"])
    links.new(combine_uv.outputs["Vector"], store_uv.inputs["Value"])

    set_material = nodes.new("GeometryNodeSetMaterial")
    scroll_mat, _ = create_image_scroll_material(image_path)
    set_material.inputs["Material"].default_value = scroll_mat
    links.new(store_uv.outputs["Geometry"], set_material.inputs["Geometry"])

    links.new(set_material.outputs["Geometry"], out_node.inputs["Geometry"])

    return gn_group, scroll_mat


def setup_geometry_node_scroll(target_mesh, unfurl_start, unfurl_end, image_path=""):
    mesh_data = bpy.data.meshes.new("Scroll_Procedural_Mesh")
    scroll_obj = bpy.data.objects.new("Scroll_GN_Object", mesh_data)
    bpy.context.scene.collection.objects.link(scroll_obj)

    mod = scroll_obj.modifiers.new("Scroll_GN", type="NODES")
    gn_tree, scroll_mat = create_pure_geometry_node_scroll(image_path=image_path)
    mod.node_group = gn_tree

    scroll_obj.data.materials.append(scroll_mat)

    r_hand_name, l_hand_name = find_hand_vertex_groups(target_mesh)

    unfurl_id = None
    for item in gn_tree.interface.items_tree:
        if getattr(item, "in_out", None) == "INPUT":
            if item.name == "Character Mesh":
                mod[item.identifier] = target_mesh
            elif item.name == "Right Hand Group":
                mod[item.identifier] = r_hand_name
            elif item.name == "Left Hand Group":
                mod[item.identifier] = l_hand_name
            elif item.name == "Forward Offset":
                mod[item.identifier] = 0.60
            elif item.name == "Scroll Width":
                mod[item.identifier] = 0.35
            elif item.name == "Unfurl":
                unfurl_id = item.identifier

    for m in target_mesh.modifiers:
        if m.type == 'ARMATURE':
            m.show_viewport = True
            m.show_render = True
            m.show_in_editmode = True

    scroll_obj.hide_viewport = True
    scroll_obj.hide_render = True
    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=1)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=1)

    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=unfurl_start - 1)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=unfurl_start - 1)

    scroll_obj.hide_viewport = False
    scroll_obj.hide_render = False
    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=unfurl_start)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=unfurl_start)

    for fc in get_object_fcurves(scroll_obj):
        if fc.data_path in ["hide_viewport", "hide_render"]:
            for kp in fc.keyframe_points:
                kp.interpolation = 'CONSTANT'

    if unfurl_id:
        mod[unfurl_id] = 0.001
        mod.keyframe_insert(data_path=f'["{unfurl_id}"]', frame=unfurl_start)

        mod[unfurl_id] = 1.0
        mod.keyframe_insert(data_path=f'["{unfurl_id}"]', frame=unfurl_end)

    return scroll_obj


# ==============================================================================
# MAIN CONTROLLER
# ==============================================================================

def import_and_extract_action(fbx_path, action_name):
    existing_objects = set(bpy.data.objects)
    bpy.ops.import_scene.fbx(filepath=fbx_path)
    bpy.context.view_layer.update()

    imported_objects = set(bpy.data.objects) - existing_objects
    extracted_action = None
    for obj in imported_objects:
        if obj.animation_data and obj.animation_data.action:
            extracted_action = obj.animation_data.action
            extracted_action.name = action_name
            extracted_action.use_fake_user = True
            break
    return extracted_action, imported_objects


def purge_imported_objects(objects_to_remove):
    for obj in objects_to_remove:
        if obj.name in bpy.data.objects:
            if obj.animation_data:
                obj.animation_data_clear()
            for coll in list(obj.users_collection):
                coll.objects.unlink(obj)
    bpy.context.view_layer.update()


def main():
    cleanup_previous_run()

    walk_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Mutant Walking.fbx"
    turn_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Right Turn 90.fbx"
    unfurl_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Mutant Idle.fbx"

    sky_hdri_path = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/horn-koppe_spring_4k.exr"
    output_blend = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/output_scene.blend"

    image_path = "/Users/dobbinp1/Downloads/IMGP6691_psedit.jpg"
    grass_dir = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Grass004_4K-JPG"

    # Setup Circular Grass Ground lowered to Z = -0.2
    setup_grass_ground_geometry_nodes(grass_dir, sky_hdri_path, ground_z_offset=-0.2)

    CYCLES = 4
    CROSSFADE = 15
    TARGET_DISTANCE = 4.0

    walk_action, walk_objs = import_and_extract_action(walk_fbx, "walk")
    target_rig = next((obj for obj in walk_objs if obj.type == "ARMATURE"), None)
    target_mesh = next((obj for obj in walk_objs if obj.type == "MESH"), None)

    if not target_rig or not target_mesh:
        raise RuntimeError("Armature or Character Mesh not found in walk FBX.")

    travel_axis_idx, motion_sign = detect_natural_forward_axis(walk_action, target_rig)
    initial_heading = math.radians(90.0) if travel_axis_idx == 0 else 0.0
    target_rig.rotation_euler[2] = initial_heading

    action_len = int(walk_action.frame_range[1] - walk_action.frame_range[0])
    if action_len <= 0:
        action_len = 30

    anim_data = target_rig.animation_data_create()

    walk_track = anim_data.nla_tracks.new()
    walk_track.name = "Walk_Track"

    current_frame = 1
    for i in range(CYCLES):
        strip = walk_track.strips.new(name=f"Walk_{i+1}", start=int(current_frame), action=walk_action)
        strip.blend_type = "REPLACE"
        current_frame += action_len

    walk_end_frame = int(current_frame)

    turn_action, turn_objs = import_and_extract_action(turn_fbx, "rightturn")
    purge_imported_objects(turn_objs)

    turn_track = anim_data.nla_tracks.new()
    turn_track.name = "RightTurn_Track"

    turn_start = int(walk_end_frame - CROSSFADE)
    turn_strip = turn_track.strips.new(name="RightTurn", start=turn_start, action=turn_action)
    turn_strip.blend_type = "REPLACE"
    turn_strip.blend_in = CROSSFADE
    turn_strip.use_auto_blend = True
    turn_strip.extrapolation = "HOLD"
    turn_end_frame = int(turn_strip.frame_end)

    unfurl_action, unfurl_objs = import_and_extract_action(unfurl_fbx, "unfurl")
    purge_imported_objects(unfurl_objs)

    unfurl_track = anim_data.nla_tracks.new()
    unfurl_track.name = "Unfurl_Track"

    unfurl_start = int(turn_end_frame - CROSSFADE)
    unfurl_strip = unfurl_track.strips.new(name="Unfurl", start=unfurl_start, action=unfurl_action)
    unfurl_strip.blend_type = "REPLACE"
    unfurl_strip.blend_in = CROSSFADE
    unfurl_strip.use_auto_blend = True
    unfurl_strip.extrapolation = "HOLD"
    unfurl_end_frame = int(unfurl_strip.frame_end)

    root_action = bpy.data.actions.new(name="World_Root_Motion")
    target_rig.animation_data.action = root_action

    loc_fcurve = root_action.fcurve_ensure_for_datablock(datablock=target_rig, data_path="location", index=travel_axis_idx)
    final_target_value = TARGET_DISTANCE * motion_sign

    k_loc1 = loc_fcurve.keyframe_points.insert(frame=1.0, value=0.0)
    k_loc2 = loc_fcurve.keyframe_points.insert(frame=float(turn_start), value=final_target_value)
    k_loc3 = loc_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=final_target_value)

    k_loc1.interpolation = "LINEAR"
    k_loc2.interpolation = "CONSTANT"
    k_loc3.interpolation = "CONSTANT"

    rot_fcurve = root_action.fcurve_ensure_for_datablock(datablock=target_rig, data_path="rotation_euler", index=2)
    target_heading = initial_heading - math.radians(90.0)

    k_rot1 = rot_fcurve.keyframe_points.insert(frame=1.0, value=initial_heading)
    k_rot2 = rot_fcurve.keyframe_points.insert(frame=float(turn_start), value=initial_heading)
    k_rot3 = rot_fcurve.keyframe_points.insert(frame=float(turn_end_frame), value=target_heading)
    k_rot4 = rot_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=target_heading)

    k_rot1.interpolation = "CONSTANT"
    k_rot2.interpolation = "BEZIER"
    k_rot3.interpolation = "CONSTANT"
    k_rot4.interpolation = "CONSTANT"

    root_track = anim_data.nla_tracks.new()
    root_track.name = "World_Root_Track"
    root_strip = root_track.strips.new(name="Root_Motion_Strip", start=1, action=root_action)
    root_strip.blend_type = "REPLACE"
    root_strip.extrapolation = "HOLD"

    anim_data.action = None

    setup_geometry_node_scroll(
        target_mesh=target_mesh,
        unfurl_start=unfurl_start,
        unfurl_end=unfurl_end_frame,
        image_path=image_path,
    )

    # Attach Animated Camera Tracking
    cam_obj = setup_camera_tracking(
        target_mesh=target_mesh,
        turn_start=turn_start,
        turn_end_frame=turn_end_frame,
        unfurl_end_frame=unfurl_end_frame,
        focal_length=35.0
    )

    configure_camera_and_viewport_horizon(far_clip=50.0)

    bpy.context.scene.frame_start = 1
    bpy.context.scene.frame_end = unfurl_end_frame

    bpy.ops.wm.save_as_mainfile(filepath=output_blend)
    print(f"\n[Success] Camera tracking setup complete. Saved to: {output_blend}")


main()
```

```
import math
import os
import glob
import bpy
from bpy_extras import anim_utils

# ==============================================================================
# CLEANUP & HELPERS
# ==============================================================================

def cleanup_previous_run():
    """Clears old scroll objects, ground, node groups, camera anchors, collections, and orphaned data."""
    target_names = [
        "Master_NLA_Control",
        "Scroll_GN_Object",
        "Scroll_Start_Pin",
        "Scroll_End_Pin",
        "Scroll_Master_Container",
        "HDRI_Ground_Plane",
        "HDRI_Camera",
        "Ground_Mesh_Object",
        "Camera_Tracking_Anchor",
        "Tracking_Camera"
    ]
    for obj in list(bpy.data.objects):
        if obj.name in target_names or (obj.name.startswith("mixamorig") and obj.type != "ARMATURE" and obj.type != "MESH"):
            if obj.animation_data:
                obj.animation_data_clear()
            bpy.data.objects.remove(obj, do_unlink=True)

    for action in list(bpy.data.actions):
        if action.users == 0 or action.use_fake_user:
            action.use_fake_user = False
            bpy.data.actions.remove(action)

    for ng in list(bpy.data.node_groups):
        if ng.name in ["Scroll_Unfurl_GN", "Scroll_Pure_GN", "HDRI_Ground_GN", "Grass_Ground_GN", "Camera_Tracker_GN"]:
            bpy.data.node_groups.remove(ng)

    for coll_name in ["Scroll_Master_Container", "Grass_Artifacts_Collection"]:
        if coll_name in bpy.data.collections:
            coll = bpy.data.collections[coll_name]
            for obj in list(coll.objects):
                bpy.data.objects.remove(obj, do_unlink=True)
            bpy.data.collections.remove(coll)

    bpy.context.view_layer.update()


def configure_camera_and_viewport_horizon(far_clip=50.0):
    if bpy.context.scene.camera:
        bpy.context.scene.camera.data.clip_end = far_clip

    for area in bpy.context.screen.areas:
        if area.type == 'VIEW_3D':
            for space in area.spaces:
                if space.type == 'VIEW_3D':
                    space.clip_end = far_clip


def get_slot_fcurves(action, anim_target_object):
    fcurves = []
    if not action:
        return fcurves

    anim_data = getattr(anim_target_object, "animation_data", None)
    if anim_data and getattr(anim_data, "action_slot", None):
        slot = anim_data.action_slot
        channelbag = anim_utils.action_get_channelbag_for_slot(action, slot)
        if channelbag and hasattr(channelbag, "fcurves"):
            return list(channelbag.fcurves)

    if hasattr(action, "slots"):
        for slot in action.slots:
            for cb in getattr(slot, "channelbags", []):
                fcurves.extend(getattr(cb, "fcurves", []))
        if fcurves:
            return fcurves

    if hasattr(action, "fcurves"):
        return list(action.fcurves)

    return fcurves


def get_object_fcurves(obj):
    if not obj or not obj.animation_data or not obj.animation_data.action:
        return []
    action = obj.animation_data.action
    return get_slot_fcurves(action, obj)


def detect_natural_forward_axis(action, rig_object):
    fcurves = get_slot_fcurves(action, rig_object)
    max_delta = 0.0
    best_axis = 1
    net_delta = 0.0

    start_frame = action.frame_range[0]
    end_frame = action.frame_range[1]

    for fc in fcurves:
        if fc.data_path.endswith("location") and fc.array_index in [0, 1, 2]:
            start_val = fc.evaluate(start_frame)
            end_val = fc.evaluate(end_frame)
            delta = end_val - start_val
            if abs(delta) > abs(max_delta):
                max_delta = delta
                best_axis = fc.array_index
                net_delta = delta

    direction_sign = 1.0 if net_delta >= 0 else -1.0
    return best_axis, direction_sign


def find_hand_vertex_groups(target_mesh):
    r_name = "mixamorig:RightHand"
    l_name = "mixamorig:LeftHand"

    if target_mesh and hasattr(target_mesh, "vertex_groups"):
        for vg in target_mesh.vertex_groups:
            vg_lower = vg.name.lower()
            if "righthand" in vg_lower or "hand_r" in vg_lower or "hand.r" in vg_lower:
                r_name = vg.name
            elif "lefthand" in vg_lower or "hand_l" in vg_lower or "hand.l" in vg_lower:
                l_name = vg.name

    return r_name, l_name


# ==============================================================================
# GEOMETRY NODES CAMERA TRACKING SETUP
# ==============================================================================

def setup_camera_tracking(target_mesh, camera_distance=(0.0, -7.0, 3.5), focal_length=35.0):
    """
    Sets up a wide-angle tracking camera aimed at the character's midsection
    to keep the entire monster framed as it moves.
    """
    # 1. Create Focal Anchor Object at character's midsection height
    anchor_obj = bpy.data.objects.new("Camera_Focus_Target", None)
    bpy.context.scene.collection.objects.link(anchor_obj)

    # Lock anchor position to character mesh
    copy_pos_anchor = anchor_obj.constraints.new(type='COPY_LOCATION')
    copy_pos_anchor.target = target_mesh
    copy_pos_anchor.target_space = 'LOCAL'
    copy_pos_anchor.owner_space = 'LOCAL'

    # Offset target point upward (~1.5m) to center the monster vertically in frame
    track_offset = anchor_obj.constraints.new(type='LIMIT_LOCATION')
    # Alternatively, use an offset via parent/child or driver if preferred,
    # but a Track To constraint with target space/offset is cleanest:

    # 2. Create Camera
    cam_data = bpy.data.cameras.new("Tracking_Camera_Data")
    cam_data.lens = focal_length  # 35mm wide angle fits full height

    cam_obj = bpy.data.objects.new("Tracking_Camera", cam_data)
    bpy.context.scene.collection.objects.link(cam_obj)
    bpy.context.scene.camera = cam_obj

    # 3. Position Camera behind/above character
    copy_pos = cam_obj.constraints.new(type='COPY_LOCATION')
    copy_pos.target = target_mesh
    copy_pos.use_offset = True
    cam_obj.location = camera_distance  # Set further back (-7.0) and slightly higher (3.5)
    copy_pos.target_space = 'LOCAL'
    copy_pos.owner_space = 'LOCAL'


    # 4. Aim Camera at the target mesh center
    track_to = cam_obj.constraints.new(type='TRACK_TO')
    track_to.target = target_mesh
    track_to.track_axis = 'TRACK_NEGATIVE_Z'
    track_to.up_axis = 'UP_X'

    # Offset the view target up toward chest height inside constraint
    track_to.use_target_z = True

    return cam_obj


# ==============================================================================
# GEOMETRY NODES GROUND, GRASS & SKY DOME SETUP
# ==============================================================================

def create_grass_material(grass_dir):
    mat = bpy.data.materials.new(name="Grass_PBR_Material")
    mat.use_nodes = True
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    out_node = nodes.new("ShaderNodeOutputMaterial")
    bsdf = nodes.new("ShaderNodeBsdfPrincipled")
    links.new(bsdf.outputs["BSDF"], out_node.inputs["Surface"])

    col_files = glob.glob(os.path.join(grass_dir, "*Color*.jpg")) + glob.glob(os.path.join(grass_dir, "*JPG*.jpg"))
    if col_files:
        tex_col = nodes.new("ShaderNodeTexImage")
        tex_col.image = bpy.data.images.load(col_files[0])
        links.new(tex_col.outputs["Color"], bsdf.inputs["Base Color"])

    rough_files = glob.glob(os.path.join(grass_dir, "*Roughness*.jpg"))
    if rough_files:
        tex_rough = nodes.new("ShaderNodeTexImage")
        tex_rough.image = bpy.data.images.load(rough_files[0])
        tex_rough.image.colorspace_settings.name = 'Non-Color'
        links.new(tex_rough.outputs["Color"], bsdf.inputs["Roughness"])

    norm_files = glob.glob(os.path.join(grass_dir, "*Normal*.jpg"))
    if norm_files:
        tex_norm = nodes.new("ShaderNodeTexImage")
        tex_norm.image = bpy.data.images.load(norm_files[0])
        tex_norm.image.colorspace_settings.name = 'Non-Color'

        normal_map_node = nodes.new("ShaderNodeNormalMap")
        links.new(tex_norm.outputs["Color"], normal_map_node.inputs["Color"])
        links.new(normal_map_node.outputs["Normal"], bsdf.inputs["Normal"])

    return mat


def create_sky_material(sky_hdri_path):
    mat = bpy.data.materials.new(name="Sky_Dome_Material")
    mat.use_nodes = True
    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    out_node = nodes.new("ShaderNodeOutputMaterial")
    emission = nodes.new("ShaderNodeEmission")
    emission.inputs["Strength"].default_value = 1.0

    links.new(emission.outputs["Emission"], out_node.inputs["Surface"])

    if os.path.exists(sky_hdri_path):
        tex_env = nodes.new("ShaderNodeTexEnvironment")
        tex_env.image = bpy.data.images.load(sky_hdri_path)
        links.new(tex_env.outputs["Color"], emission.inputs["Color"])

    return mat


def setup_grass_ground_geometry_nodes(grass_dir, sky_hdri_path, ground_z_offset=-0.2):
    grass_coll = bpy.data.collections.get("Grass_Artifacts_Collection")
    if not grass_coll:
        grass_coll = bpy.data.collections.new("Grass_Artifacts_Collection")
        bpy.context.scene.collection.children.link(grass_coll)

    grass_mat = create_grass_material(grass_dir)
    sky_mat = create_sky_material(sky_hdri_path)

    mesh_files = glob.glob(os.path.join(grass_dir, "*.obj")) + glob.glob(os.path.join(grass_dir, "*.fbx"))

    if mesh_files:
        for fpath in mesh_files:
            if fpath.endswith(".obj"):
                bpy.ops.wm.obj_import(filepath=fpath)
            elif fpath.endswith(".fbx"):
                bpy.ops.import_scene.fbx(filepath=fpath)

            for o in bpy.context.selected_objects:
                grass_coll.objects.link(o)
                if o.name in bpy.context.scene.collection.objects:
                    bpy.context.scene.collection.objects.unlink(o)

        for obj in list(grass_coll.objects):
            if obj.dimensions.z > 0.5:
                grass_coll.objects.unlink(obj)
                bpy.data.objects.remove(obj, do_unlink=True)
    else:
        for i in range(3):
            m_data = bpy.data.meshes.new(f"Grass_Blade_Mesh_{i}")
            g_obj = bpy.data.objects.new(f"Grass_Blade_Obj_{i}", m_data)
            grass_coll.objects.link(g_obj)

            verts = [(0, 0, 0), (0.05, 0, 0), (0.04, 0, 0.15), (0.01, 0, 0.25), (0, 0, 0.3)]
            faces = [(0, 1, 2), (1, 2, 3), (2, 3, 4)]
            m_data.from_pydata(verts, [], faces)
            m_data.update()

            if g_obj.data.materials:
                g_obj.data.materials[0] = grass_mat
            else:
                g_obj.data.materials.append(grass_mat)

            g_obj.location = (100, 100, 0)

    # 1. CIRCULAR GROUND MESH LOWERED BY ground_z_offset
    bpy.ops.mesh.primitive_circle_add(radius=12.5, vertices=64, fill_type='NGON', location=(0, 0, ground_z_offset))
    ground_obj = bpy.context.active_object
    ground_obj.name = "Ground_Mesh_Object"

    vg = ground_obj.vertex_groups.new(name="Grass_Weight_Group")
    all_vert_indices = [v.index for v in ground_obj.data.vertices]
    vg.add(all_vert_indices, 1.0, 'REPLACE')
    ground_obj.data.materials.append(grass_mat)
    ground_obj.data.materials.append(sky_mat)

    # 2. GN Tree
    gn_group = bpy.data.node_groups.new("Grass_Ground_GN", type="GeometryNodeTree")
    gn_group.interface.new_socket(name="Geometry", in_out="INPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Geometry", in_out="OUTPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Density Mask", in_out="INPUT", socket_type="NodeSocketFloat")

    nodes = gn_group.nodes
    links = gn_group.links
    out_node = nodes.new("NodeGroupOutput")
    in_node = nodes.new("NodeGroupInput")

    # --- GRASS SCATTERING ---
    distribute_node = nodes.new("GeometryNodeDistributePointsOnFaces")
    distribute_node.distribute_method = 'POISSON'
    distribute_node.inputs["Distance Min"].default_value = 0.02
    distribute_node.inputs["Density Max"].default_value = 300.0
    links.new(in_node.outputs["Geometry"], distribute_node.inputs["Mesh"])

    noise_node = nodes.new("ShaderNodeTexNoise")
    noise_node.inputs["Scale"].default_value = 2.5
    noise_node.inputs["Detail"].default_value = 2.0

    color_ramp = nodes.new("ShaderNodeValToRGB")
    color_ramp.color_ramp.elements[0].position = 0.20
    color_ramp.color_ramp.elements[1].position = 0.80
    links.new(noise_node.outputs["Fac"], color_ramp.inputs["Fac"])

    math_weight = nodes.new("ShaderNodeMath")
    math_weight.operation = "MULTIPLY"
    links.new(color_ramp.outputs["Color"], math_weight.inputs[0])
    links.new(in_node.outputs["Density Mask"], math_weight.inputs[1])

    links.new(math_weight.outputs["Value"], distribute_node.inputs["Density Factor"])

    instance_node = nodes.new("GeometryNodeInstanceOnPoints")
    instance_node.inputs["Pick Instance"].default_value = True
    links.new(distribute_node.outputs["Points"], instance_node.inputs["Points"])

    coll_info = nodes.new("GeometryNodeCollectionInfo")
    coll_info.inputs["Collection"].default_value = grass_coll
    coll_info.inputs["Separate Children"].default_value = True
    coll_info.inputs["Reset Children"].default_value = True
    links.new(coll_info.outputs["Instances"], instance_node.inputs["Instance"])

    rand_scale = nodes.new("FunctionNodeRandomValue")
    rand_scale.data_type = 'FLOAT'
    rand_scale.inputs["Min"].default_value = 0.25
    rand_scale.inputs["Max"].default_value = 0.50

    combine_scale = nodes.new("ShaderNodeCombineXYZ")
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["X"])
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["Y"])
    links.new(rand_scale.outputs["Value"], combine_scale.inputs["Z"])
    links.new(combine_scale.outputs["Vector"], instance_node.inputs["Scale"])

    rand_rot_z = nodes.new("FunctionNodeRandomValue")
    rand_rot_z.data_type = 'FLOAT'
    rand_rot_z.inputs["Min"].default_value = 0.0
    rand_rot_z.inputs["Max"].default_value = math.radians(360.0)

    combine_rot = nodes.new("ShaderNodeCombineXYZ")
    links.new(rand_rot_z.outputs["Value"], combine_rot.inputs["Z"])
    links.new(combine_rot.outputs["Vector"], instance_node.inputs["Rotation"])

    # --- MATCHING SKY DOME ---
    sky_sphere = nodes.new("GeometryNodeMeshIcoSphere")
    sky_sphere.inputs["Radius"].default_value = 12.5
    sky_sphere.inputs["Subdivisions"].default_value = 4

    flip_faces = nodes.new("GeometryNodeFlipFaces")
    links.new(sky_sphere.outputs["Mesh"], flip_faces.inputs["Mesh"])

    transform_sky = nodes.new("GeometryNodeTransform")
    transform_sky.inputs["Translation"].default_value = (0.0, 0.0, ground_z_offset)
    links.new(flip_faces.outputs["Mesh"], transform_sky.inputs["Geometry"])

    set_sky_mat = nodes.new("GeometryNodeSetMaterial")
    set_sky_mat.inputs["Material"].default_value = sky_mat
    links.new(transform_sky.outputs["Geometry"], set_sky_mat.inputs["Geometry"])

    # --- JOIN GEOMETRY ---
    join_geo = nodes.new("GeometryNodeJoinGeometry")
    links.new(in_node.outputs["Geometry"], join_geo.inputs["Geometry"])
    links.new(instance_node.outputs["Instances"], join_geo.inputs["Geometry"])
    links.new(set_sky_mat.outputs["Geometry"], join_geo.inputs["Geometry"])

    links.new(join_geo.outputs["Geometry"], out_node.inputs["Geometry"])

    mod = ground_obj.modifiers.new(name="Grass_Ground_GN", type='NODES')
    mod.node_group = gn_group

    for item in gn_group.interface.items_tree:
        if getattr(item, "in_out", None) == "INPUT" and item.name == "Density Mask":
            mod[item.identifier + "_attribute_name"] = "Grass_Weight_Group"
            mod[item.identifier + "_use_attribute"] = True

    return ground_obj


# ==============================================================================
# SCROLL MATERIAL & GEOMETRY NODES SETUP
# ==============================================================================

def create_image_scroll_material(image_path):
    mat_name = "Scroll_Texture_Mat"
    mat = bpy.data.materials.get(mat_name)
    if mat:
        bpy.data.materials.remove(mat)

    mat = bpy.data.materials.new(name=mat_name)
    if hasattr(mat, "use_nodes"):
        try:
            mat.use_nodes = True
        except AttributeError:
            pass

    nodes = mat.node_tree.nodes
    links = mat.node_tree.links
    nodes.clear()

    output_node = nodes.new("ShaderNodeOutputMaterial")
    bsdf = nodes.new("ShaderNodeBsdfPrincipled")
    bsdf.inputs["Roughness"].default_value = 0.5

    tex_img = nodes.new("ShaderNodeTexImage")
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        tex_img.image = img
        tex_img.extension = 'REPEAT'

    uv_map_node = nodes.new("ShaderNodeUVMap")
    uv_map_node.uv_map = "UVMap"

    links.new(uv_map_node.outputs["UV"], tex_img.inputs["Vector"])
    links.new(tex_img.outputs["Color"], bsdf.inputs["Base Color"])
    links.new(bsdf.outputs["BSDF"], output_node.inputs["Surface"])

    return mat, tex_img.image if os.path.exists(image_path) else None


def create_pure_geometry_node_scroll(image_path=""):
    gn_group = bpy.data.node_groups.new("Scroll_Pure_GN", type="GeometryNodeTree")

    gn_group.interface.new_socket(name="Geometry", in_out="OUTPUT", socket_type="NodeSocketGeometry")
    gn_group.interface.new_socket(name="Character Mesh", in_out="INPUT", socket_type="NodeSocketObject")

    sock_r_vg = gn_group.interface.new_socket(name="Right Hand Group", in_out="INPUT", socket_type="NodeSocketString")
    sock_r_vg.default_value = "mixamorig:RightHand"

    sock_l_vg = gn_group.interface.new_socket(name="Left Hand Group", in_out="INPUT", socket_type="NodeSocketString")
    sock_l_vg.default_value = "mixamorig:LeftHand"

    sock_offset = gn_group.interface.new_socket(name="Forward Offset", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_offset.default_value = 0.60

    sock_unfurl = gn_group.interface.new_socket(name="Unfurl", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_unfurl.default_value = 0.0

    sock_width = gn_group.interface.new_socket(name="Scroll Width", in_out="INPUT", socket_type="NodeSocketFloat")
    sock_width.default_value = 0.35

    nodes = gn_group.nodes
    links = gn_group.links

    in_node = nodes.new("NodeGroupInput")
    out_node = nodes.new("NodeGroupOutput")

    char_info = nodes.new("GeometryNodeObjectInfo")
    char_info.transform_space = "RELATIVE"
    links.new(in_node.outputs["Character Mesh"], char_info.inputs["Object"])

    pos_node = nodes.new("GeometryNodeInputPosition")

    named_r = nodes.new("GeometryNodeInputNamedAttribute")
    named_r.data_type = "FLOAT"
    links.new(in_node.outputs["Right Hand Group"], named_r.inputs["Name"])

    cmp_r = nodes.new("FunctionNodeCompare")
    cmp_r.data_type = "FLOAT"
    cmp_r.operation = "GREATER_THAN"
    cmp_r.inputs[1].default_value = 0.001
    links.new(named_r.outputs["Attribute"], cmp_r.inputs[0])

    stat_r = nodes.new("GeometryNodeAttributeStatistic")
    stat_r.data_type = "FLOAT_VECTOR"
    links.new(char_info.outputs["Geometry"], stat_r.inputs["Geometry"])
    links.new(cmp_r.outputs["Result"], stat_r.inputs["Selection"])
    links.new(pos_node.outputs["Position"], stat_r.inputs["Attribute"])

    named_l = nodes.new("GeometryNodeInputNamedAttribute")
    named_l.data_type = "FLOAT"
    links.new(in_node.outputs["Left Hand Group"], named_l.inputs["Name"])

    cmp_l = nodes.new("FunctionNodeCompare")
    cmp_l.data_type = "FLOAT"
    cmp_l.operation = "GREATER_THAN"
    cmp_l.inputs[1].default_value = 0.001
    links.new(named_l.outputs["Attribute"], cmp_l.inputs[0])

    stat_l = nodes.new("GeometryNodeAttributeStatistic")
    stat_l.data_type = "FLOAT_VECTOR"
    links.new(char_info.outputs["Geometry"], stat_l.inputs["Geometry"])
    links.new(cmp_l.outputs["Result"], stat_l.inputs["Selection"])
    links.new(pos_node.outputs["Position"], stat_l.inputs["Attribute"])

    neg_offset = nodes.new("ShaderNodeMath")
    neg_offset.operation = "MULTIPLY"
    neg_offset.inputs[1].default_value = -1.0
    links.new(in_node.outputs["Forward Offset"], neg_offset.inputs[0])

    forward_vec = nodes.new("ShaderNodeCombineXYZ")
    forward_vec.inputs["X"].default_value = 0.0
    links.new(neg_offset.outputs["Value"], forward_vec.inputs["Y"])
    forward_vec.inputs["Z"].default_value = 0.0

    start_pos = nodes.new("ShaderNodeVectorMath")
    start_pos.operation = "ADD"
    links.new(stat_r.outputs["Mean"], start_pos.inputs[0])
    links.new(forward_vec.outputs["Vector"], start_pos.inputs[1])

    target_pos = nodes.new("ShaderNodeVectorMath")
    target_pos.operation = "ADD"
    links.new(stat_l.outputs["Mean"], target_pos.inputs[0])
    links.new(forward_vec.outputs["Vector"], target_pos.inputs[1])

    delta_vec = nodes.new("ShaderNodeVectorMath")
    delta_vec.operation = "SUBTRACT"
    links.new(target_pos.outputs["Vector"], delta_vec.inputs[0])
    links.new(start_pos.outputs["Vector"], delta_vec.inputs[1])

    dist_node = nodes.new("ShaderNodeVectorMath")
    dist_node.operation = "DISTANCE"
    links.new(start_pos.outputs["Vector"], dist_node.inputs[0])

    scale_vec = nodes.new("ShaderNodeVectorMath")
    scale_vec.operation = "SCALE"
    links.new(delta_vec.outputs["Vector"], scale_vec.inputs["Vector"])
    links.new(in_node.outputs["Unfurl"], scale_vec.inputs["Scale"])

    end_pos = nodes.new("ShaderNodeVectorMath")
    end_pos.operation = "ADD"
    links.new(start_pos.outputs["Vector"], end_pos.inputs[0])
    links.new(scale_vec.outputs["Vector"], end_pos.inputs[1])

    links.new(end_pos.outputs["Vector"], dist_node.inputs[1])

    curve_line = nodes.new("GeometryNodeCurvePrimitiveLine")
    links.new(start_pos.outputs["Vector"], curve_line.inputs["Start"])
    links.new(end_pos.outputs["Vector"], curve_line.inputs["End"])

    set_tilt = nodes.new("GeometryNodeSetCurveTilt")
    set_tilt.inputs["Tilt"].default_value = 1.5708
    links.new(curve_line.outputs["Curve"], set_tilt.inputs["Curve"])

    spline_param_main = nodes.new("GeometryNodeSplineParameter")
    store_u = nodes.new("GeometryNodeStoreNamedAttribute")
    store_u.inputs["Name"].default_value = "U_Factor"
    store_u.data_type = 'FLOAT'
    store_u.domain = 'POINT'
    links.new(set_tilt.outputs["Curve"], store_u.inputs["Geometry"])
    links.new(spline_param_main.outputs["Factor"], store_u.inputs["Value"])

    half_width_neg = nodes.new("ShaderNodeMath")
    half_width_neg.operation = "MULTIPLY"
    half_width_neg.inputs[1].default_value = -0.5
    links.new(in_node.outputs["Scroll Width"], half_width_neg.inputs[0])

    half_width_pos = nodes.new("ShaderNodeMath")
    half_width_pos.operation = "MULTIPLY"
    half_width_pos.inputs[1].default_value = 0.5
    links.new(in_node.outputs["Scroll Width"], half_width_pos.inputs[0])

    combine_start = nodes.new("ShaderNodeCombineXYZ")
    links.new(half_width_neg.outputs["Value"], combine_start.inputs["X"])

    combine_end = nodes.new("ShaderNodeCombineXYZ")
    links.new(half_width_pos.outputs["Value"], combine_end.inputs["X"])

    profile_line = nodes.new("GeometryNodeCurvePrimitiveLine")
    links.new(combine_start.outputs["Vector"], profile_line.inputs["Start"])
    links.new(combine_end.outputs["Vector"], profile_line.inputs["End"])

    spline_param_prof = nodes.new("GeometryNodeSplineParameter")
    store_v = nodes.new("GeometryNodeStoreNamedAttribute")
    store_v.inputs["Name"].default_value = "V_Factor"
    store_v.data_type = 'FLOAT'
    store_v.domain = 'POINT'
    links.new(profile_line.outputs["Curve"], store_v.inputs["Geometry"])
    links.new(spline_param_prof.outputs["Factor"], store_v.inputs["Value"])

    curve_to_mesh = nodes.new("GeometryNodeCurveToMesh")
    links.new(store_u.outputs["Geometry"], curve_to_mesh.inputs["Curve"])
    links.new(store_v.outputs["Geometry"], curve_to_mesh.inputs["Profile Curve"])

    img_aspect = 1.0
    if os.path.exists(image_path):
        img = bpy.data.images.load(image_path)
        if img.size[1] > 0:
            img_aspect = img.size[0] / img.size[1]

    h_times_aspect = nodes.new("ShaderNodeMath")
    h_times_aspect.operation = "MULTIPLY"
    h_times_aspect.inputs[1].default_value = img_aspect
    links.new(in_node.outputs["Scroll Width"], h_times_aspect.inputs[0])

    tile_count = nodes.new("ShaderNodeMath")
    tile_count.operation = "DIVIDE"
    links.new(dist_node.outputs["Value"], tile_count.inputs[0])
    links.new(h_times_aspect.outputs["Value"], tile_count.inputs[1])

    read_u = nodes.new("GeometryNodeInputNamedAttribute")
    read_u.inputs["Name"].default_value = "U_Factor"
    read_u.data_type = 'FLOAT'

    scaled_u = nodes.new("ShaderNodeMath")
    scaled_u.operation = "MULTIPLY"
    links.new(read_u.outputs["Attribute"], scaled_u.inputs[0])
    links.new(tile_count.outputs["Value"], scaled_u.inputs[1])

    read_v = nodes.new("GeometryNodeInputNamedAttribute")
    read_v.inputs["Name"].default_value = "V_Factor"
    read_v.data_type = 'FLOAT'

    flip_v = nodes.new("ShaderNodeMath")
    flip_v.operation = "SUBTRACT"
    flip_v.inputs[0].default_value = 1.0
    links.new(read_v.outputs["Attribute"], flip_v.inputs[1])

    combine_uv = nodes.new("ShaderNodeCombineXYZ")
    links.new(scaled_u.outputs["Value"], combine_uv.inputs["X"])
    links.new(flip_v.outputs["Value"], combine_uv.inputs["Y"])

    store_uv = nodes.new("GeometryNodeStoreNamedAttribute")
    store_uv.inputs["Name"].default_value = "UVMap"
    store_uv.data_type = 'FLOAT2'
    store_uv.domain = 'CORNER'
    links.new(curve_to_mesh.outputs["Mesh"], store_uv.inputs["Geometry"])
    links.new(combine_uv.outputs["Vector"], store_uv.inputs["Value"])

    set_material = nodes.new("GeometryNodeSetMaterial")
    scroll_mat, _ = create_image_scroll_material(image_path)
    set_material.inputs["Material"].default_value = scroll_mat
    links.new(store_uv.outputs["Geometry"], set_material.inputs["Geometry"])

    links.new(set_material.outputs["Geometry"], out_node.inputs["Geometry"])

    return gn_group, scroll_mat


def setup_geometry_node_scroll(target_mesh, unfurl_start, unfurl_end, image_path=""):
    mesh_data = bpy.data.meshes.new("Scroll_Procedural_Mesh")
    scroll_obj = bpy.data.objects.new("Scroll_GN_Object", mesh_data)
    bpy.context.scene.collection.objects.link(scroll_obj)

    mod = scroll_obj.modifiers.new("Scroll_GN", type="NODES")
    gn_tree, scroll_mat = create_pure_geometry_node_scroll(image_path=image_path)
    mod.node_group = gn_tree

    scroll_obj.data.materials.append(scroll_mat)

    r_hand_name, l_hand_name = find_hand_vertex_groups(target_mesh)

    unfurl_id = None
    for item in gn_tree.interface.items_tree:
        if getattr(item, "in_out", None) == "INPUT":
            if item.name == "Character Mesh":
                mod[item.identifier] = target_mesh
            elif item.name == "Right Hand Group":
                mod[item.identifier] = r_hand_name
            elif item.name == "Left Hand Group":
                mod[item.identifier] = l_hand_name
            elif item.name == "Forward Offset":
                mod[item.identifier] = 0.60
            elif item.name == "Scroll Width":
                mod[item.identifier] = 0.35
            elif item.name == "Unfurl":
                unfurl_id = item.identifier

    for m in target_mesh.modifiers:
        if m.type == 'ARMATURE':
            m.show_viewport = True
            m.show_render = True
            m.show_in_editmode = True

    scroll_obj.hide_viewport = True
    scroll_obj.hide_render = True
    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=1)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=1)

    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=unfurl_start - 1)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=unfurl_start - 1)

    scroll_obj.hide_viewport = False
    scroll_obj.hide_render = False
    scroll_obj.keyframe_insert(data_path="hide_viewport", frame=unfurl_start)
    scroll_obj.keyframe_insert(data_path="hide_render", frame=unfurl_start)

    for fc in get_object_fcurves(scroll_obj):
        if fc.data_path in ["hide_viewport", "hide_render"]:
            for kp in fc.keyframe_points:
                kp.interpolation = 'CONSTANT'

    if unfurl_id:
        mod[unfurl_id] = 0.001
        mod.keyframe_insert(data_path=f'["{unfurl_id}"]', frame=unfurl_start)

        mod[unfurl_id] = 1.0
        mod.keyframe_insert(data_path=f'["{unfurl_id}"]', frame=unfurl_end)

    return scroll_obj


# ==============================================================================
# MAIN CONTROLLER
# ==============================================================================

def import_and_extract_action(fbx_path, action_name):
    existing_objects = set(bpy.data.objects)
    bpy.ops.import_scene.fbx(filepath=fbx_path)
    bpy.context.view_layer.update()

    imported_objects = set(bpy.data.objects) - existing_objects
    extracted_action = None
    for obj in imported_objects:
        if obj.animation_data and obj.animation_data.action:
            extracted_action = obj.animation_data.action
            extracted_action.name = action_name
            extracted_action.use_fake_user = True
            break
    return extracted_action, imported_objects


def purge_imported_objects(objects_to_remove):
    for obj in objects_to_remove:
        if obj.name in bpy.data.objects:
            if obj.animation_data:
                obj.animation_data_clear()
            for coll in list(obj.users_collection):
                coll.objects.unlink(obj)
    bpy.context.view_layer.update()


def main():
    cleanup_previous_run()

    walk_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Mutant Walking.fbx"
    turn_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Right Turn 90.fbx"
    unfurl_fbx = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Mutant Idle.fbx"

    sky_hdri_path = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/horn-koppe_spring_4k.exr"
    output_blend = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/output_scene.blend"

    image_path = "/Users/dobbinp1/Downloads/IMGP6691_psedit.jpg"
    grass_dir = "/Volumes/macstorage/Users/dobbinp1/NLA/unfurl/Grass004_4K-JPG"

    # Setup Circular Grass Ground lowered to Z = -0.2
    setup_grass_ground_geometry_nodes(grass_dir, sky_hdri_path, ground_z_offset=-0.2)

    CYCLES = 4
    CROSSFADE = 15
    TARGET_DISTANCE = 4.0

    walk_action, walk_objs = import_and_extract_action(walk_fbx, "walk")
    target_rig = next((obj for obj in walk_objs if obj.type == "ARMATURE"), None)
    target_mesh = next((obj for obj in walk_objs if obj.type == "MESH"), None)

    if not target_rig or not target_mesh:
        raise RuntimeError("Armature or Character Mesh not found in walk FBX.")

    travel_axis_idx, motion_sign = detect_natural_forward_axis(walk_action, target_rig)
    initial_heading = math.radians(90.0) if travel_axis_idx == 0 else 0.0
    target_rig.rotation_euler[2] = initial_heading

    action_len = int(walk_action.frame_range[1] - walk_action.frame_range[0])
    if action_len <= 0:
        action_len = 30

    anim_data = target_rig.animation_data_create()

    walk_track = anim_data.nla_tracks.new()
    walk_track.name = "Walk_Track"

    current_frame = 1
    for i in range(CYCLES):
        strip = walk_track.strips.new(name=f"Walk_{i+1}", start=int(current_frame), action=walk_action)
        strip.blend_type = "REPLACE"
        current_frame += action_len

    walk_end_frame = int(current_frame)

    turn_action, turn_objs = import_and_extract_action(turn_fbx, "rightturn")
    purge_imported_objects(turn_objs)

    turn_track = anim_data.nla_tracks.new()
    turn_track.name = "RightTurn_Track"

    turn_start = int(walk_end_frame - CROSSFADE)
    turn_strip = turn_track.strips.new(name="RightTurn", start=turn_start, action=turn_action)
    turn_strip.blend_type = "REPLACE"
    turn_strip.blend_in = CROSSFADE
    turn_strip.use_auto_blend = True
    turn_strip.extrapolation = "HOLD"
    turn_end_frame = int(turn_strip.frame_end)

    unfurl_action, unfurl_objs = import_and_extract_action(unfurl_fbx, "unfurl")
    purge_imported_objects(unfurl_objs)

    unfurl_track = anim_data.nla_tracks.new()
    unfurl_track.name = "Unfurl_Track"

    unfurl_start = int(turn_end_frame - CROSSFADE)
    unfurl_strip = unfurl_track.strips.new(name="Unfurl", start=unfurl_start, action=unfurl_action)
    unfurl_strip.blend_type = "REPLACE"
    unfurl_strip.blend_in = CROSSFADE
    unfurl_strip.use_auto_blend = True
    unfurl_strip.extrapolation = "HOLD"
    unfurl_end_frame = int(unfurl_strip.frame_end)

    root_action = bpy.data.actions.new(name="World_Root_Motion")
    target_rig.animation_data.action = root_action

    loc_fcurve = root_action.fcurve_ensure_for_datablock(datablock=target_rig, data_path="location", index=travel_axis_idx)
    final_target_value = TARGET_DISTANCE * motion_sign

    k_loc1 = loc_fcurve.keyframe_points.insert(frame=1.0, value=0.0)
    k_loc2 = loc_fcurve.keyframe_points.insert(frame=float(turn_start), value=final_target_value)
    k_loc3 = loc_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=final_target_value)

    k_loc1.interpolation = "LINEAR"
    k_loc2.interpolation = "CONSTANT"
    k_loc3.interpolation = "CONSTANT"

    rot_fcurve = root_action.fcurve_ensure_for_datablock(datablock=target_rig, data_path="rotation_euler", index=2)
    target_heading = initial_heading - math.radians(90.0)

    k_rot1 = rot_fcurve.keyframe_points.insert(frame=1.0, value=initial_heading)
    k_rot2 = rot_fcurve.keyframe_points.insert(frame=float(turn_start), value=initial_heading)
    k_rot3 = rot_fcurve.keyframe_points.insert(frame=float(turn_end_frame), value=target_heading)
    k_rot4 = rot_fcurve.keyframe_points.insert(frame=float(unfurl_end_frame), value=target_heading)

    k_rot1.interpolation = "CONSTANT"
    k_rot2.interpolation = "BEZIER"
    k_rot3.interpolation = "CONSTANT"
    k_rot4.interpolation = "CONSTANT"

    root_track = anim_data.nla_tracks.new()
    root_track.name = "World_Root_Track"
    root_strip = root_track.strips.new(name="Root_Motion_Strip", start=1, action=root_action)
    root_strip.blend_type = "REPLACE"
    root_strip.extrapolation = "HOLD"

    anim_data.action = None

    setup_geometry_node_scroll(
        target_mesh=target_mesh,
        unfurl_start=unfurl_start,
        unfurl_end=unfurl_end_frame,
        image_path=image_path,
    )

    # Attach Geometry Nodes Camera Tracking
    cam_obj = setup_camera_tracking(
        target_mesh=target_mesh,
        camera_distance=(0.0, -7.5, 1 ),  # Pull back to -7.5m on Y, 3.2m on Z
        focal_length=35.0                  # 35mm wider FOV prevents clipping hands/head
    )


    cam_obj.data.shift_y = 0.2


    configure_camera_and_viewport_horizon(far_clip=50.0)

    bpy.context.scene.frame_start = 1
    bpy.context.scene.frame_end = unfurl_end_frame

    bpy.ops.wm.save_as_mainfile(filepath=output_blend)
    print(f"\n[Success] Camera tracking setup complete. Saved to: {output_blend}")


main()
```
